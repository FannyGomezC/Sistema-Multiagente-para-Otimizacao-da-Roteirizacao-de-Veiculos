# **DETALHAMENTO DO SISTEMA MULTIAGENTE**

## **SISTEMA MULTIAGENTE PARA OTIMIZAÇÃO DA ROTEIRIZAÇÃO DE VEÍCULOS EM OPERAÇÕES DE ENTREGA**

> **Do problema de roteirização a um plano de viagens auditado, melhorado e matematicamente avaliado.**

O sistema recebe os **dados da operação**, os **objetivos** e as **restrições** definidos pelo usuário e coordena agentes especializados para buscar o melhor plano de viagens possível.

---

## **O QUE O SISTEMA BUSCA?**

**MENOR DISTÂNCIA** &nbsp;&nbsp; | &nbsp;&nbsp; **MENOR TEMPO DE OPERACIONAL TOTAL** &nbsp;&nbsp; | &nbsp;&nbsp; **MENOR CUSTO** &nbsp;&nbsp; | &nbsp;&nbsp; **MENOR NÚMERO DE VEÍCULOS**

Respeitando as restrições da operação:

**Capacidade** &nbsp; • &nbsp; **Janelas de tempo** &nbsp; • &nbsp; **Jornada dos veículos** &nbsp; • &nbsp; **Multi-trip** &nbsp; • &nbsp; **Custos** &nbsp; • &nbsp; **Regras de utilização**

---

## **COMO FUNCIONA?**

### **ENTENDER O PROBLEMA → BUSCAR SOLUÇÕES → AUDITAR → MELHORAR → CERTIFICAR**

Cada agente possui uma **responsabilidade específica**, enquanto o **Orquestrador** coordena o trabalho, compartilha as informações necessárias e preserva a melhor solução aprovada.

---

## **O DIFERENCIAL**

O sistema não se limita a perguntar:

> **“Qual foi a melhor solução encontrada?”**

Ele também procura responder:

> **Essa solução é realmente válida?**  
> **Ainda é possível melhorá-la?**  
> **Quão próxima ela está da solução ótima?**

Para isso, compara a **melhor solução encontrada (UB)** com um **piso matemático válido (LB)** e utiliza essa distância para avaliar o nível de certificação alcançado.

---

## **BUSCA → AUDITORIA → MELHORIA → CERTIFICAÇÃO**

> **Uma solução encontrada não é automaticamente uma solução aprovada.**

O sistema **busca** uma solução, **audita** de forma independente o cumprimento das restrições, continua procurando **melhorias** e tenta **demonstrar matematicamente sua qualidade**.

---

> ### **O resultado final não é apenas uma rota.**
> É um **plano operacional** acompanhado de evidências sobre sua **viabilidade, qualidade e proximidade da solução ótima**.

## **ESTRUTURA MULTIAGENTE**

> **Especialização dos agentes + coordenação central + memória compartilhada**

Cada agente possui uma **responsabilidade específica** dentro do processo de otimização.

O **Orquestrador** atua como coordenador central: define o fluxo de trabalho, integra os resultados produzidos pelos agentes e preserva a **melhor solução aprovada**.

**As tarefas dos outrso agentes se dividem assim:**

### **PREPARAR → BUSCAR → AUDITAR → MELHORAR → CERTIFICAR**

Essa divisão de responsabilidades permite que os agentes trabalhem de forma especializada, mantendo a integração necessária para resolver o problema como um único sistema.

O sistema separa a construção da **solução inicial** do processo posterior de **melhoria e certificação**.

O núcleo não precisa ser executado novamente para cada rodada de continuidade. A **memória compartilhada** guarda dados, resultados e histórico. Um **checkpoint** salva informações desse estado em arquivo para permitir uma retomada posterior.

### QUAIS PLANILHAS DEVE PREPARAR?

Prepare os dados sobre **clientes, capacidades dos veículos, custos e horários de atendimento**. As duas primeiras planilhas são obrigatórias. Custos e horários dependem das opções escolhidas, conforme a tabela abaixo.





Tabela com células delimitadas: Arquivo / Quando enviar / Cabeçalhos recomendados, exatamente assim / Como preencher





**Formato mais simples: `.xlsx`.** O sistema lê a primeira aba: coloque nela a tabela, com os cabeçalhos na primeira linha. Não use títulos acima da tabela, células mescladas, subtotais, linhas de comentário ou fórmulas sem resultado salvo. Preencha os campos obrigatórios. Envie **um arquivo por solicitação**; o nome pode ser escolhido por você.

Também é possível usar **CSV em UTF-8, separado por vírgulas e com ponto decimal**. Um CSV separado por ponto e vírgula não tem o mesmo formato: salve em `.xlsx` ou exporte com as opções indicadas. Em Excel, mantenha os valores como células numéricas, mesmo que seu programa mostre vírgula decimal.

### UM EXEMPLO DE PREENCHIMENTO

Para **dois clientes**, a ficha de clientes tem **três linhas de dados**, pois o depósito não é contado como cliente:





Tabela com células delimitadas: Cliente / latitude / longitude / Demanda





São coordenadas ilustrativas. Use as coordenadas reais da operação em graus decimais: latitude entre -90 e 90 e longitude entre -180 e 180. Os identificadores devem ser números inteiros sem repetição; use `0` somente para o depósito e, preferencialmente, `1, 2, 3...` para os clientes e para os veículos em suas respectivas fichas.





Tabela com células delimitadas: id_veiculo / capacidade





Se a unidade escolhida for toneladas, **demanda e capacidade devem estar ambas em toneladas**. Não misture quilogramas com toneladas; caixas também não se convertem automaticamente em peso.





Tabela com células delimitadas: id_veiculo / custo_por_km / custo_diario





Você escolherá na conversa se o custo fixo é pago somente quando o veículo é usado ou mesmo se ele ficar parado. Esta ficha é obrigatória se você pedir custo, sozinho ou combinado com outro objetivo.





Tabela com células delimitadas: cliente / inicio_janela / fim_janela





Se todos os clientes tiverem o mesmo horário, **não envie esta última ficha**: selecione a opção de janela única e digite, por exemplo, `08:00-17:00`. Se não houver restrição de horário de atendimento, responda que não existem janelas. A duração da jornada continua sendo solicitada em minutos.

### COMO RESPONDER E CORRIGIR UMA INFORMAÇÃO

- Quantidades de clientes e veículos: números inteiros maiores que zero, sem contar o depósito como cliente.
- Jornada, velocidade e tempo mínimo ativado: números maiores que zero. Tempo por entrega pode ser zero. Use minutos para tempos e km/h para velocidade.
- Nas perguntas numéricas da conversa, `1,5` e `1.5` são aceitos. Não escreva unidades junto do número nem separadores de milhares: use `1000`, não `1.000` para indicar mil.
- Uma letra, resposta vazia ou número fora do intervalo **não faz o sistema avançar**. Ele explica o erro e repete a pergunta.
- Depois de cada valor numérico, escolha **1 para confirmar** ou **2 para corrigir**. Nos menus, digite somente uma das opções exibidas.
- No fim aparece um resumo. Você pode confirmar ou refazer a ficha inteira, reenviando as planilhas, antes de qualquer otimização. Na continuidade, o tempo adicional também é confirmado.

A conferência de formato na entrada não substitui a validação da coerência dos dados nem a auditoria do plano encontrado. Se o arquivo estiver errado, corrija-o e envie novamente.


### UNIDADE DA CARGA: USE A MESMA EM TODAS AS PLANILHAS

O nome da unidade informado pelo usuário — toneladas, caixas ou quilogramas — serve para identificar os números e apresentar o resultado. **O sistema não converte unidades nem calcula equivalências.** A restrição de capacidade usa os valores numéricos de demanda e capacidade: se o veículo comporta 10 toneladas, a demanda 4 deve significar 4 toneladas. Não misture capacidade em toneladas com demanda em quilogramas. A capacidade é um parâmetro da otimização; o nome da unidade é informativo.

### **SHAREDMEMORY — MEMÓRIA COMPARTILHADA**

Centraliza os dados e resultados utilizados pelos componentes: plano atual, auditorias, limites matemáticos e histórico. Isso reduz o risco de um agente trabalhar com informações desatualizadas.

**A memória armazena informações; não escolhe rotas nem prova o ótimo.** O orquestrador e os agentes consultam e atualizam esse estado. Um checkpoint permite salvar parte dele em arquivo para uma retomada.


````python
class SharedMemory:
    """
    Memória compartilhada central do sistema multiagente.

    Armazena:
    - dados fornecidos pelo usuário;
    - dados processados e validados;
    - função objetivo canônica;
    - resultados dos solvers;
    - incumbent global;
    - certificação de optimalidade;
    - auditoria;
    - diagnóstico;
    - explicação final.

    A SharedMemory apenas armazena estados e resultados.
    Ela não executa cálculos nem altera soluções.
    """
    VERSION = "SHARED_MEMORY_V11_CONTINUITY"

    def __init__(self):
        self.data = {
            # Entrada
            "user_answers": None,
            "df_clientes": None,
            "df_veiculos_capacidade": None,
            "df_veiculos_custo_km": None,
            "df_janela_tempo_cliente": None,

            # Pré-solver
            "validation_result": None,
            "translated_json": None,
            "operational_contract": None,
            "problem_classification": None,
            "optimization_approach": None,
            "optimization_profile": None,
            "objective_context": None,

            # Solvers / incumbent
            "solver_result": None,
            "solver_result_pre_certification": None,
            "best_incumbent": None,

            # Auditoria independente
            "independent_feasibility_report_auditor": None,
            "last_independent_feasibility_report": None,
            "independent_feasibility_report_pre_certification": None,
            "independent_feasibility_report_post_certification": None,
            "certification_fast_bounds": None,
            "audit_result": None,
            "audit_recommendation": None,

            # Certificação
            "certification_status": None,
            "certification_progress": None,
            "certification_operational_contract": None,
            "certification_preprocessing": None,
            "certification_result_provisional": None,
            "optimality_certificate": None,
            "certified_solution": None,
            "certification_result": None,
            "certification_checkpoint": None,

            # Solução oficial
            "final_solution": None,

            # Diagnóstico / relatório
            "diagnostic_result": None,
            "diagnostic_kpis": None,
            "diagnostic_bi_tables": None,
            "explanation": None,
            "final_report": None,

            # Núcleo opcional de continuidade pós-Relatório 1.
            # Estas chaves são deliberadamente separadas do pipeline principal.
            "continuity_context": None,
            "continuity_state": None,
            "continuity_baseline": None,
            "continuity_seed_solution": None,
            "continuity_candidate": None,
            "continuity_candidate_audit": None,
            "continuity_certification_provisional": None,
            "continuity_certificate": None,
            "continuity_final_solution": None,
            "continuity_final_audit": None,
            "continuity_comparison": None,
            "continuity_diagnostic": None,
            "continuity_report": None,
            "continuity_history": [],
            "continuity_baseline_valid_lower_bound": None,
        }
````

### **USERINTERFACE — ENTRADA E CONFERÊNCIA DOS DADOS**

É a porta de entrada do sistema. Ela conversa com o usuário e recebe clientes, veículos, horários, custos e preferências.

A interface confere os números imediatamente, permite corrigi-los, mostra uma prévia de cada arquivo e apresenta um resumo para confirmação antes de chamar a equipe. Consulte os formatos e exemplos de planilhas na seção 2.


````python
class UserInterface:
    def __init__(self, shared_memory):
      self.shared_memory = shared_memory
      # Mapeamento didático para facilitar a seleção do objetivo de otimização pelo usuário
      self.objective_mapping = {
          "1": "Minimizar distância",
          "2": "Minimizar tempo operacional total",
          "3": "Minimizar custo",
          "4": "Minimizar veículos",
          "5": "Multiobjetivo"
      }

      # Estilos de cores (atualmente vazios, mas mantidos para compatibilidade visual futura)
      self.C_BLUE = ''
      self.C_GREEN = ''
      self.C_BOLD = ''
      self.C_END = ''

      # Dicionário central que armazena todas as respostas coletadas durante o fluxo de perguntas
      self.answers = {}

    @staticmethod
    def _ask_choice(prompt, allowed):
        allowed = tuple(str(v) for v in allowed)
        while True:
            answer = input(prompt).strip()
            if answer in allowed:
                return answer
            print('Resposta inválida. Digite uma destas opções: ' + ', '.join(allowed) + '.')

    @staticmethod
    def _parse_number(raw, integer=False, minimum=0, maximum=None, strictly_positive=False):
        import math
        import re
        text = str(raw).strip()
        pattern = r'[0-9]+' if integer else r'[+-]?(?:[0-9]+(?:[.,][0-9]+)?|[.,][0-9]+)'
        if not re.fullmatch(pattern, text):
            raise ValueError('Digite somente um número, sem unidade ou separador de milhares.')
        value = int(text) if integer else float(text.replace(',', '.'))
        if not math.isfinite(value):
            raise ValueError('O número deve ser finito.')
        if value < minimum or (strictly_positive and value <= 0):
            raise ValueError('Use um valor maior que zero.' if strictly_positive else f'Use um valor a partir de {minimum}.')
        if maximum is not None and value > maximum:
            raise ValueError(f'Use um valor de até {maximum}.')
        return str(value)

    @classmethod
    def _ask_number(cls, prompt, integer=False, minimum=0, maximum=None, strictly_positive=False):
        while True:
            try:
                value = cls._parse_number(input(prompt), integer, minimum, maximum, strictly_positive)
            except (ValueError, OverflowError) as exc:
                print('Resposta inválida:', exc)
                continue
            print('Valor informado:', value)
            if cls._ask_choice('[1] Confirmar este valor; [2] Corrigir: ', ('1', '2')) == '1':
                return value

    @classmethod
    def _ask_window(cls, prompt):
        import datetime
        while True:
            raw = input(prompt).strip()
            try:
                left, right = [s.strip() for s in raw.split('-')]
                start = datetime.datetime.strptime(left, '%H:%M')
                end = datetime.datetime.strptime(right, '%H:%M')
                if end <= start:
                    raise ValueError()
                value = start.strftime('%H:%M') + '-' + end.strftime('%H:%M')
            except ValueError:
                print('Janela inválida. Use HH:MM-HH:MM, no mesmo dia, com o fim após o início.')
                continue
            print('Janela informada:', value)
            if cls._ask_choice('[1] Confirmar; [2] Corrigir: ', ('1', '2')) == '1':
                return value

    @classmethod
    def ask_continuity_budget(cls):
        import sys
        try:
            while True:
                print('Tempo adicional: [1] 5 min; [2] 15 min; [3] 30 min; [4] 60 min; [5] Outro; [0] Encerrar.')
                choice = cls._ask_choice('Escolha: ', ('0', '1', '2', '3', '4', '5'))
                if choice == '0':
                    return None
                if choice == '5':
                    return float(cls._ask_number('Minutos (mínimo 0,5): ', minimum=0.5, maximum=sys.float_info.max/120))*60
                minutes = {'1': 5, '2': 15, '3': 30, '4': 60}[choice]
                print(f'Orçamento escolhido: {minutes} minutos.')
                if cls._ask_choice('[1] Confirmar; [2] Escolher outro tempo: ', ('1', '2')) == '1':
                    return minutes*60
        except (EOFError, KeyboardInterrupt):
            print('\nRodada adicional não iniciada.')
            return None

    def _show_summary(self):
        a = self.answers
        print('\n' + '='*60 + '\nCONFIRA O PEDIDO ANTES DE INICIAR A BUSCA\n' + '='*60)
        pairs = [
            ('Objetivo', a.get('objective_otimizacao')),
            ('Unidade da carga', a.get('unidade_medida_carga')),
            ('Veículos disponíveis', a.get('Qual é o número de veiculos?')),
            ('Clientes (sem contar o depósito)', a.get('Quantos clientes existem?')),
            ('Tempo disponível por veículo (min)', a.get('Tempo disponível por veículo por dia (minutos)?')),
            ('Tempo mínimo por veículo (min)', a.get('tempo_minimo_minutos')),
            ('Tempo de entrega por cliente (min)', a.get('Qual é o tempo médio que cada entrega consume (em minutos)?')),
            ('Velocidade média (km/h)', a.get('Qual é a velocidade média dos veículos (em Km/h)?')),
            ('Obrigatório usar toda a frota', 'Sim' if a.get('must_use_all_vehicles') else 'Não'),
            ('Mais de uma viagem por veículo', 'Sim' if a.get('Um veiculo pode fazer mais de uma viagem por dia?') == '1' else 'Não'),
        ]
        for label, value in pairs:
            print(f'{label}: {value}')
        labels = {'DISTANCE': 'Distância', 'TIME': 'Tempo operacional total', 'COST': 'Custo', 'VEHICLES': 'Veículos'}
        if a.get('multiobjective_config', {}).get('enabled'):
            for key, value in a['multiobjective_config']['weights'].items():
                print(f'Peso de {labels[key]}: {100*value:g}%')
        if a.get('requires_cost_data'):
            print('Custo fixo:', 'só quando o veículo é usado' if a.get('fixed_cost_mode') == 'WHEN_USED' else 'mesmo sem utilizar o veículo')
        window = a.get('Qual é a janela de tempo? (Ex: 08:00-17:00)')
        print('Horários de atendimento:', window or ('por cliente, na planilha' if a.get('Existem janelas de tempo? (1 - sim, 2 - não)') == '1' else 'sem janelas de cliente'))
        for key, label in [('df_clientes_file', 'Clientes'), ('df_veiculos_capacidade_file', 'Capacidades'), ('df_veiculos_custo_km_file', 'Custos'), ('df_janela_tempo_cliente_file', 'Horários')]:
            if a.get(key):
                print(f'Arquivo de {label}: {a[key]}')

    def iniciar(self):
        while True:
            # Uma nova coleta descarta os dados anteriores, inclusive escolhas condicionais.
            self.answers = {}
            self._collect_answers()
            self._show_summary()
            answer = self._ask_choice('[1] Confirmar e iniciar a validação; [2] Refazer a ficha: ', ('1', '2'))
            if answer == '1':
                print('Pedido confirmado. O sistema ainda verificará a coerência das planilhas antes de otimizar.')
                return self.answers
            print('Vamos refazer a ficha e reenviar os arquivos. Nenhuma otimização foi iniciada.')

    def _print_welcome(self):
      # Exibe um cabeçalho visual para recepção do usuário
      print("="*60)
      print("BEM-VINDO AO SISTEMA DE OTIMIZAÇÃO DE ROTAS (VRP)")
      print("="*60)
      print("Por favor, informe os detalhes do problema que deseja otimizar.\n")

    def _ask_mandatory(self, prompt):
      # Garante que o usuário forneça uma resposta não vazia
      while True:
        res = input(prompt).strip()
        if res:
          return res
        print("⚠️ Resposta obrigatória! Por favor, insira um valor.")

    def _handle_file_upload(self, prompt_text, instructions):
        print(f'\n[UPLOAD REQUERIDO]: {prompt_text}\n{instructions}')
        print('Envie um arquivo .xlsx (primeira aba) ou .csv (UTF-8, separado por vírgulas e com ponto decimal).')
        schemas = {
            'Planilha de Capacidade dos Veículos': VEICULOS_CAPACIDADE_COLUMN_MAP,
            'Planilha de Custo por Veículo': VEICULOS_CUSTO_KM_COLUMN_MAP,
            'Planilha de Coordenadas e Demandas': CLIENTES_COLUMN_MAP,
            'Planilha de Janelas de Tempo': JANELA_TEMPO_CLIENTE_COLUMN_MAP,
        }
        while True:
            uploaded = files.upload()
            if not uploaded or len(uploaded) != 1:
                print('Envie exatamente um arquivo para esta ficha. Tente novamente.')
                continue
            filename = next(iter(uploaded))
            try:
                if filename.lower().endswith('.xlsx'):
                    frame = pd.read_excel(filename)
                elif filename.lower().endswith('.csv'):
                    frame = pd.read_csv(filename, encoding='utf-8-sig')
                else:
                    raise ValueError('Use .xlsx ou .csv; converta arquivos antigos .xls para .xlsx.')
                mapping = schemas.get(prompt_text)
                if frame.empty:
                    raise ValueError('A tabela está vazia.')
                if mapping:
                    frame = _standardize_and_map_columns(frame, mapping)
                    missing = [key for key in mapping if key not in frame.columns]
                    if missing:
                        raise ValueError('Faltam colunas: ' + ', '.join(missing) + '. Confira os cabeçalhos e o separador do CSV.')
                print(f'Arquivo lido: {filename} | {len(frame)} linhas de dados')
                print(frame.head(4).to_string(index=False))
            except Exception as exc:
                print(f'Não foi possível aceitar esta ficha: {exc} Corrija e envie novamente.')
                continue
            if self._ask_choice('[1] Confirmar este arquivo; [2] Enviar outro: ', ('1','2')) == '1':
                return filename

    def _collect_answers(self):
      # Método principal que orquestra a sequência lógica de entrada de dados
      self._print_welcome()

      print('Prepare arquivos .xlsx (primeira aba) ou CSV UTF-8 com vírgula separadora e ponto decimal.')
      print('Cabeçalhos na primeira linha; um arquivo por solicitação. Veja os exemplos antes desta célula.')
      print('Letras não são aceitas nos campos numéricos. Cada valor será confirmado antes de avançar.')

      # -- 0. Definição de Unidades de Medida ---
      # Essencial para que os relatórios finais exibam os KPIs nas unidades corretas (kg, m3, R$, etc)
      print("0. Definição de Unidades de Medida:")
      u_carga = self._ask_mandatory("Qual a unidade de medida da carga (ex: Kg, Caixas, m³, Toneladas)? ")
      self.answers['unidade_medida_carga'] = u_carga

      u_moeda = "R$"
      self.answers['unidade_monetaria'] = u_moeda
      print(f"ℹ️  O sistema utilizará '{u_moeda}' para todos os cálculos de custo.\n")

      # -- 1. Objetivo da Otimização ---
      # Define a função objetivo que o Solver tentará minimizar ou maximizar
      print("1. Qual é o objetivo da otimização?")
      print("Tempo operacional total: soma das durações de todas as viagens, incluindo deslocamento, atendimento e espera no percurso. Não é a duração do expediente nem o tempo de cálculo do programa.")
      for k, v in self.objective_mapping.items():
          print(f"  [{k}] - {v}")

      obj_choice = ""
      while obj_choice not in self.objective_mapping:
          obj_choice = self._ask_choice("Escolha uma opção (1-5): ", ("1","2","3","4","5"))
      self.answers['objective_otimizacao'] = self.objective_mapping.get(obj_choice)


      # CONFIGURAÇÃO MULTIOBJETIVO

      self.answers['multiobjective_config'] = {
          'enabled': False, 'objectives': [], 'weights': {}
      }

      # Indica explicitamente se o problema exige dados financeiros.
      # A regra é válida tanto para objetivo simples COST quanto para
      # qualquer configuração multiobjetivo que contenha COST.
      self.answers['requires_cost_data'] = False

      if obj_choice == '5':
          print('\n' + '=' * 60)
          print('CONFIGURAÇÃO DA OTIMIZAÇÃO MULTIOBJETIVO')
          print('=' * 60)
          print('\nSelecione pelo menos dois objetivos:')
          print('  [1] - Minimizar distância')
          print('  [2] - Minimizar tempo operacional total')
          print('  [3] - Minimizar custo')
          print('  [4] - Minimizar número de veículos')
          objective_options = {'1':'DISTANCE','2':'TIME','3':'COST','4':'VEHICLES'}
          while True:
              raw = input('Objetivos (ex.: 1,2,3): ').strip()
              choices = [x.strip() for x in raw.split(',') if x.strip()]
              if any(x not in objective_options for x in choices):
                  print('⚠️ Use somente 1, 2, 3 ou 4.'); continue
              selected = list(dict.fromkeys(objective_options[x] for x in choices))
              self.answers['requires_cost_data'] = ('COST' in selected)
              if len(selected) < 2:
                  print('⚠️ Selecione pelo menos dois objetivos.'); continue
              break
          labels = {'DISTANCE':'Distância','TIME':'Tempo operacional total','COST':'Custo','VEHICLES':'Número de veículos'}
          while True:
              weights_percent = {}; total = 0.0
              print('\nInforme os pesos em porcentagem. A soma deve ser 100%.')
              for objective in selected:
                  while True:
                      try:
                          value = float(self._ask_number(f"Peso - {labels[objective]} (%): ", maximum=100))
                          if value < 0: raise ValueError
                          weights_percent[objective] = value; total += value; break
                      except ValueError:
                          print('⚠️ Informe um número maior ou igual a zero.')
              if abs(total - 100.0) <= 1e-6: break
              print(f'⚠️ A soma informada foi {total:.2f}%. Tente novamente.')
          self.answers['multiobjective_config'] = {
              'enabled': True, 'objectives': selected,
              'weights': {k:v/100.0 for k,v in weights_percent.items()}
          }
          self.answers['requires_cost_data'] = ('COST' in selected)
          print('\n✅ Configuração multiobjetivo:')
          for objective in selected:
              print(f"   {labels[objective]}: {weights_percent[objective]:.2f}%")
      else:
          simple_map = {'1':'DISTANCE','2':'TIME','3':'COST','4':'VEHICLES'}
          simple_objective = simple_map[obj_choice]
          self.answers['multiobjective_config'] = {
              'enabled': False, 'objectives': [simple_objective],
              'weights': {simple_objective: 1.0}
          }
          self.answers['requires_cost_data'] = (simple_objective == 'COST')

      # -- 2. Configuração de Frota ---
      # Define o número total de recursos físicos disponíveis para a operação
      print("\n2. Número de veículos:")
      v_input = self._ask_number("Informe a quantidade (inteiro maior que zero): ", integer=True, minimum=1)
      self.answers['Qual é o número de veiculos?'] = v_input

      # -- 2.1 Obrigatoriedade de utilização da frota ---
      # Pergunta independente: o usuário decide explicitamente se toda a frota deve ser usada.
      print("\n2.1. Todos os veículos devem ser utilizados?")
      use_all_choice = ""
      while use_all_choice not in ["1", "2"]:
          use_all_choice = self._ask_choice("(1-Sim, 2-Não): ", ("1","2"))
      self.answers['Todos os veiculos devem ser utilizados?'] = "1" if use_all_choice == "1" else "2"
      self.answers['must_use_all_vehicles'] = (use_all_choice == "1")

      # -- 2.2 Tempo Mínimo de Utilização ---
      # O tempo mínimo é independente da obrigação de usar toda a frota.
      print("\n2.2. Os veículos têm um tempo mínimo de utilização?")
      t_min_choice = ""
      while t_min_choice not in ["1", "2"]:
          t_min_choice = self._ask_choice("(1-Sim, 2-Não): ", ("1","2"))
      self.answers['veiculos_tem_tempo_minimo'] = "1" if t_min_choice == "1" else "2"
      if t_min_choice == "1":
          self.answers['tempo_minimo_minutos'] = self._ask_number("Informe o tempo mínimo (em minutos): ", strictly_positive=True)
      else:
          self.answers['tempo_minimo_minutos'] = "0"

      # -- 3. Restrições Temporais da Frota ---      # --- 3. Restrições Temporais da Frota ---
      # Define o limite máximo da jornada de trabalho (ex: 480 min = 8 horas)
      print("\n3. Tempo disponível por veículo por dia (minutos):")
      tempo_disponivel_veiculo = self._ask_number("Informe o tempo disponível (minutos): ", strictly_positive=True)
      while float(tempo_disponivel_veiculo) < float(self.answers['tempo_minimo_minutos']):
          print('O tempo mínimo não pode ultrapassar o tempo disponível. Corrija um dos dois valores.')
          if self._ask_choice('[1] Corrigir o tempo disponível; [2] Corrigir o tempo mínimo: ', ('1','2')) == '1':
              tempo_disponivel_veiculo = self._ask_number('Tempo disponível (minutos): ', strictly_positive=True)
          else:
              self.answers['tempo_minimo_minutos'] = self._ask_number('Tempo mínimo (minutos): ', strictly_positive=True)
      self.answers['Tempo disponível por veículo por dia (minutos)?'] = tempo_disponivel_veiculo

      # -- 4. Configuração de Múltiplas Viagens (Multi-Trip) ---
      # Define se o solver pode planejar mais de uma saída do depósito para o mesmo veículo
      print("\n4. Um veículo pode fazer mais de uma viagem por dia?")
      multi_trip = ""
      while multi_trip not in ["1", "2"]:
          multi_trip = self._ask_choice("(1-Sim, 2-Não): ", ("1","2"))
      self.answers['Um veiculo pode fazer mais de uma viagem por dia?'] = "1" if multi_trip == "1" else "2"

      # Upload da planilha de capacidade física
      filename_cap = self._handle_file_upload(
          "Planilha de Capacidade dos Veículos",
          f"Esta planilha deve conter as colunas ID_VEICULO e CAPACIDADE (em {u_carga}). \n👉 Certifique-se que o arquivo contém exatamente {v_input} veículos cadastrados."
      )
      self.answers['df_veiculos_capacidade_file'] = filename_cap

      # -- 5. Configuração de Custos (Condicional) ---
      # CUSTO exige os dados financeiros sempre que estiver entre os
      # objetivos ativos, inclusive em problemas multiobjetivo.
      requires_cost_data = bool(
          self.answers.get(
              'requires_cost_data',
              'COST' in self.answers.get('multiobjective_config', {}).get('objectives', [])
          )
      )

      if requires_cost_data:
          self.answers['df_veiculos_custo_km_file'] = self._handle_file_upload(
              "Planilha de Custo por Veículo",
              f"Esta planilha deve informar para cada veículo as colunas ID_VEICULO, CUSTO_POR_KM e CUSTO_DIARIO (em {u_moeda})."
          )

          # -- 5.1 Natureza do Custo Fixo ---
          print("\n5.1. Como o custo fixo diário deve ser cobrado?")
          print("  [1] - Apenas se o veículo for utilizado (Ativação)")
          print("  [2] - Pago mesmo que o veículo não seja utilizado (Incondicional)")
          c_fixo_choice = ""
          while c_fixo_choice not in ["1", "2"]:
              c_fixo_choice = self._ask_choice("Escolha uma opção (1-2): ", ("1","2"))

          self.answers['tipo_custo_fixo'] = (
              "ativacao" if c_fixo_choice == "1" else "incondicional"
          )
          self.answers['fixed_cost_mode'] = (
              "WHEN_USED" if c_fixo_choice == "1" else "ALWAYS"
          )
      else:
          self.answers['df_veiculos_custo_km_file'] = None
          self.answers['tipo_custo_fixo'] = None
          self.answers['fixed_cost_mode'] = None

      # -- 6. Configuração de Clientes ---
      # Define a escala do problema em termos de pontos de entrega
      print("\n5. Número de clientes:")
      c_input = self._ask_number("Informe a quantidade (inteiro maior que zero): ", integer=True, minimum=1)
      self.answers['Quantos clientes existem?'] = c_input

      # Upload da planilha de coordenadas geográficas e demandas dos clientes
      self.answers['df_clientes_file'] = self._handle_file_upload(
          "Planilha de Coordenadas e Demandas",
          f"Esta planilha deve conter Cliente, latitude, longitude e Demanda (em {u_carga}). \n👉 Inclua o depósito com Cliente=0 e Demanda=0, além de uma linha para cada cliente. Use IDs inteiros únicos."
      )

      # -- 7. Janelas de Tempo (VRPTW) ---
      # Define restrições de horários de atendimento (SLA)
      print("\n6. Existem janelas de tempo?")
      jt_exist = ""
      while jt_exist not in ["1", "2"]:
          jt_exist = self._ask_choice("(1-Sim, 2-Não): ", ("1","2"))
      self.answers['Existem janelas de tempo? (1 - sim, 2 - não)'] = jt_exist

      if jt_exist == "1":
          jt_tipo = ""
          while jt_tipo not in ["1", "2"]:
              jt_tipo = self._ask_choice("As janelas são iguais para todos? (1-Sim, 2-Não): ", ("1","2"))
          self.answers['As janelas de tempo são iguais para todos os clientes? (1 - sim, 2 - não)'] = jt_tipo
          if jt_tipo == "1":
              self.answers['Qual é a janela de tempo? (Ex: 08:00-17:00)'] = self._ask_window("Informe a janela única (Ex: 08:00-17:00): ")
          else:
              self.answers['df_janela_tempo_cliente_file'] = self._handle_file_upload(
                  "Planilha de Janelas de Tempo",
                  "Cabeçalhos: cliente, inicio_janela, fim_janela. Uma linha por cliente (sem depósito), IDs iguais à ficha de clientes e horários HH:MM, no mesmo dia."
              )

      # -- 8. Parâmetros Operacionais Gerais ---
      # Dados necessários para cálculo de tempo de rota e produtividade no local
      print("\n7. Tempo médio de entrega (min):")
      self.answers['Qual é o tempo médio que cada entrega consume (em minutos)?'] = self._ask_number("Minutos por entrega (zero é permitido): ")
      print("8. Velocidade média (Km/h):")
      self.answers['Qual é a velocidade média dos veículos (em Km/h)?'] = self._ask_number("Km/h: ", strictly_positive=True)

      print("\nColeta concluída. Confira o resumo antes de iniciar.")
      return self.answers




# -- Mapeamentos para padronização de nomes de colunas ---
# As chaves são os nomes exatos esperados pelo ValidatorAgent.
# Os valores são listas de possíveis nomes de colunas normalizados (minúsculas, sem espaços/hífens, sem acentos) que podem ser fornecidos pelo usuário.

# Para df_clientes: espera ['Cliente', 'latitude', 'longitude', 'Demanda']
CLIENTES_COLUMN_MAP = {
    'Cliente': ['cliente', 'client', 'id_cliente', 'client_id', 'nome_cliente'],
    'latitude': ['latitude', 'lat', 'coord_y', 'lat_coord'],
    'longitude': ['longitude', 'lon', 'coord_x', 'lon_coord'],
    'Demanda': ['demanda', 'demand', 'demandas', 'qtd_demanda', 'quantidade_demanda']
}

# Para df_veiculos_capacidade: espera ['id_veiculo', 'capacidade']
VEICULOS_CAPACIDADE_COLUMN_MAP = {
    'id_veiculo': ['id_veiculo', 'vehicle_id', 'veiculo', 'id', 'id_carro', 'car_id', 'id_vh'],
    'capacidade': ['capacidade', 'capacity', 'cap', 'capacidade_total', 'capacidade_max', 'max_capacity', 'carga_maxima', 'carga_limite']
}

# Para df_veiculos_custo_veiculos: espera ['id_veiculo', 'custo_por_km']
VEICULOS_CUSTO_KM_COLUMN_MAP = {
    'id_veiculo': ['id_veiculo', 'vehicle_id', 'veiculo', 'id', 'id_carro', 'car_id', 'id_vh'],
    'custo_por_km': ['custo_por_km', 'cost_per_km', 'custo_km', 'custo', 'custo_operacional', 'valor_km'],
    'custo_diario': ['custo_diario', 'daily_cost', 'custo_fixo', 'fixed_cost'] # Adicionado mapeamento para custo diário
}

# Para df_janela_tempo_cliente: espera ['cliente', 'inicio_janela', 'fim_janela']
JANELA_TEMPO_CLIENTE_COLUMN_MAP = {
    'cliente': ['cliente', 'client', 'id_cliente', 'client_id', 'nome_cliente'],
    'inicio_janela': ['inicio_janela', 'start_time', 'time_window_start', 'inicio', 'hora_inicial', 'start_window', 'janela_inicio'],
    'fim_janela': ['fim_janela', 'end_time', 'time_window_end', 'fim', 'hora_final', 'end_window', 'janela_fim']
}

def _standardize_and_map_columns(df, target_column_map):
    """
    Padroniza nomes de colunas de um DataFrame: converte nomes atuais para minúsculas,
    substitui espaços e hífens por underscores e mapeia para nomes de coluna alvo.

    Args:
        df (pd.DataFrame): O DataFrame de entrada.
        target_column_map (dict): Um dicionário onde as chaves são os nomes de coluna exatos
                                  esperados e os valores são listas de possíveis nomes de coluna
                                  (já normalizados) que podem corresponder a eles.

    Returns:
        pd.DataFrame: DataFrame com as colunas renomeadas, ou None se df for None.
    """
    if df is None:
        return None

    df_copy = df.copy()
    rename_dict = {}

    # Normalização dos possiveis nomes para as colunas objetivo
    normalized_to_target_map = {}
    for target_col, possible_normalized_names in target_column_map.items():
        for pnn in possible_normalized_names:
            normalized_to_target_map[pnn] = target_col

        normalized_to_target_map[target_col.lower().replace(" ", "_").replace("-", "_")] = target_col


    for current_col_name in df_copy.columns:
        normalized_current_col_name = current_col_name.lower().replace(" ", "_").replace("-", "_")

        if normalized_current_col_name in normalized_to_target_map:
            target_col_name = normalized_to_target_map[normalized_current_col_name]
            if current_col_name != target_col_name:
                rename_dict[current_col_name] = target_col_name

    df_copy.rename(columns=rename_dict, inplace=True)

    return df_copy


def _load_and_read_data(filename_base, file_type_input):
    """
    Carrega dados de um arquivo local, sendo robusto a tipos de arquivo e extensões.
    Prioriza o upload se o arquivo não estiver presente.
    """
    df = None
    loaded_filename = None
    possible_extensions = []

    # Prioriza a extensão baseada no input do usuário
    if file_type_input.lower() == "excel":
        possible_extensions = ["xlsx", "xls", "csv"] # Tenta extensões comuns de Excel, depois CSV
    elif file_type_input.lower() == "csv":
        possible_extensions = ["csv", "xlsx", "xls"] # Tenta CSV primeiro, depois Excel
    else:
        # Fallback para input não reconhecido (como "rxcel") - tenta extensões comuns
        print(f"Aviso: Tipo de arquivo '{file_type_input}' não reconhecido. Tentando extensões comuns (.xlsx, .csv).")
        possible_extensions = ["xlsx", "xls", "csv"]

    # 1. Tentativa de carregar arquivo existente no ambiente Colab (/content/)
    for ext in possible_extensions:
        current_filename = os.path.join('/content', f"{filename_base}.{ext}")
        if os.path.exists(current_filename):
            loaded_filename = current_filename
            print(f"Arquivo '{loaded_filename}' encontrado no ambiente Colab.")
            break

    # 2. Se o arquivo não foi encontrado localmente, solicita upload
    if loaded_filename is None:
        print(f"O arquivo base '{filename_base}' não foi encontrado com extensões esperadas. Por favor, faça o upload.")
        uploaded = files.upload()
        if uploaded:
            # Tenta encontrar o arquivo uploaded que corresponda ao filename_base
            for uploaded_name in uploaded:
                # Considera o nome do arquivo uploaded pelo usuário
                if filename_base in uploaded_name:
                    loaded_filename = uploaded_name
                    print(f"Arquivo '{loaded_filename}' enviado com sucesso.")
                    break
            if loaded_filename is None:
                print(f"Nenhum arquivo relevante para '{filename_base}' foi enviado. Por favor, verifique o nome do arquivo.")
                return None
        else:
            print("Nenhum arquivo foi enviado.")
            return None

    # 3. Tenta ler o arquivo com pandas
    try:
        if loaded_filename.lower().endswith(('.xlsx', '.xls')):
            df = pd.read_excel(loaded_filename)
        elif loaded_filename.lower().endswith('.csv'):
            df = pd.read_csv(loaded_filename, encoding='utf-8-sig')
        else:
            print(f"Erro: Extensão de arquivo desconhecida para '{loaded_filename}'. Não é .xlsx, .xls ou .csv.")
            return None

        # Retorna None se o DataFrame estiver vazio, para evitar erros de 'truth value ambiguous'
        if df.empty:
            print(f"Aviso: O arquivo '{loaded_filename}' foi carregado, mas está vazio. Retornando None.")
            return None

        print(f"DataFrame carregado com sucesso de '{loaded_filename}'.")
        return df
    except Exception as e:
        print(f"Erro ao carregar '{loaded_filename}': {e}")
        return None

# Função específica para carregar a planilha de dados dos clientes.
# Recebe o tipo de arquivo (Excel/CSV) e um nome base para o arquivo.
def load_clientes_data(file_type, filename_base="clientes"):
    print(f"Preparando para carregar a planilha de clientes (base: '{filename_base}', tipo de input: '{file_type}')")
    df = _load_and_read_data(filename_base, file_type)
    if df is not None:
        return _standardize_and_map_columns(df, CLIENTES_COLUMN_MAP)
    return None

# Função específica para carregar a planilha de capacidade dos veículos.
# Esta função será usada para a pergunta sobre a capacidade dos veículos.
def load_veiculos_capacidade_data(file_type, filename_base="veiculos_capacidade"):
    print(f"Preparando para carregar a planilha de capacidade dos veículos (base: '{filename_base}', tipo de input: '{file_type}')")
    df = _load_and_read_data(filename_base, file_type)
    if df is not None:
        return _standardize_and_map_columns(df, VEICULOS_CAPACIDADE_COLUMN_MAP)
    return None

# Função específica para carregar a planilha de custo por veículo.
# Esta função será usada APENAS se o objetivo for 'Minimizar custo' e o custo por KM for variável.
def load_veiculos_custo_km_data(file_type, filename_base="veiculos_custo_km"):
    print(f"Preparando para carregar a planilha de custo por KM dos veículos (base: '{filename_base}', tipo de input: '{file_type}')")
    df = _load_and_read_data(filename_base, file_type)
    if df is not None:
        return _standardize_and_map_columns(df, VEICULOS_CUSTO_KM_COLUMN_MAP)
    return None

# Função específica para carregar a planilha de janelas de tempo dos clientes.
# Esta função será usada se o usuário indicar que existem janelas de tempo e elas são variáveis por cliente.
def load_janela_tempo_cliente_data(file_type, filename_base="janela_tempo_cliente"):
    print(f"Preparando para carregar a planilha de janelas de tempo dos clientes (base: '{filename_base}', tipo de input: '{file_type}')")
    df = _load_and_read_data(filename_base, file_type)
    if df is not None:
        return _standardize_and_map_columns(df, JANELA_TEMPO_CLIENTE_COLUMN_MAP)
    return None

def validar_coerencia_manual(answers, df_clientes, df_veiculos_capacidade):
    erros = []
    try:
        n_veiculos = int(answers.get("Qual é o número de veiculos?", 0))
    except Exception:
        n_veiculos = 0
    try:
        n_clientes = int(answers.get("Quantos clientes existem?", 0))
    except Exception:
        n_clientes = 0

    if isinstance(df_veiculos_capacidade, pd.DataFrame) and "id_veiculo" in df_veiculos_capacidade.columns:
        n_planilha = df_veiculos_capacidade["id_veiculo"].nunique()
        if n_veiculos != n_planilha:
            erros.append(
                f"Erro de coerência: o usuário informou {n_veiculos} veículos, "
                f"mas a planilha contém {n_planilha} IDs únicos."
            )

    if isinstance(df_clientes, pd.DataFrame) and "Cliente" in df_clientes.columns:
        n_planilha = df_clientes.loc[df_clientes["Cliente"] != 0, "Cliente"].nunique()
        if n_clientes != n_planilha:
            erros.append(
                f"Erro de coerência: o usuário informou {n_clientes} clientes, "
                f"mas a planilha contém {n_planilha} clientes válidos (excluindo o depósito 0)."
            )
    return len(erros) == 0, erros




class CustomJsonEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, time):
            return str(obj)
        return super().default(obj)
````

## **AGENTES DO NÚCLEO INICIAL**

Esta é a equipe que prepara a primeira resposta. O orquestrador usa os agentes conforme o caminho de resolução escolhido. As descrições abaixo apresentam a responsabilidade de cada componente.


### **INDEPENDENTFEASIBILITYVERIFIER — VERIFICAÇÃO INDEPENDENTE DO PLANO**

Este componente verifica o plano de forma independente da busca que o produziu. Confere o atendimento dos clientes, as cargas e os horários, e recalcula distâncias e custos.





Tabela com células delimitadas: Verificação / Pergunta simples





**Aprovar a viabilidade não é provar o ótimo.** Significa que o plano respeita as regras verificadas. O código inclui o contexto da auditoria e a identificação do problema, para não misturar resultados de dados diferentes.


````python
class IndependentFeasibilityVerifier:
    """
    Verificador independente de factibilidade.

    Princípio central:
    - NÃO confia nos totais declarados pelo solver.
    - Reconstrói cobertura, carga, distância, tempos, custos e objetivo a partir
      de path + veículo físico + dados originais do problema.
    - É ferramenta interna do AuditorAgent. O Certificador recebe apenas o
      relatório produzido pelo Auditor; não instancia este verificador.
    """

    VERSION = "INDEPENDENT_FEASIBILITY_VERIFIER_10_1"
    TOL = 1e-7
    # Tolerância exclusiva para encadeamento temporal Multi-Trip.
    # 1e-5 minuto = 0,0006 segundo. Evita falsos positivos gerados
    # apenas por arredondamento/representação de ponto flutuante.
    SEQUENCE_TOL_MIN = 1e-5

    def __init__(self, shared_memory):
        self.shared_memory = shared_memory
        self.problem = None
        self.objective_context = None

    @staticmethod
    def _safe_float(value, default=None):
        try:
            if value is None:
                return default
            if isinstance(value, str):
                value = value.strip().replace(",", ".")
                if not value:
                    return default
            return float(value)
        except Exception:
            return default

    @staticmethod
    def _safe_int(value, default=None):
        try:
            return int(value)
        except Exception:
            try:
                return int(float(value))
            except Exception:
                return default

    @staticmethod
    def _parse_bool(value, default=False):
        if value is None:
            return default
        if isinstance(value, bool):
            return value
        if isinstance(value, (int, float)):
            return bool(value)
        txt = str(value).strip().lower()
        if txt in {"1", "true", "sim", "s", "yes", "y"}:
            return True
        if txt in {"0", "2", "false", "não", "nao", "n", "no"}:
            return False
        return default

    @staticmethod
    def _parse_time(value, default=0.0):
        if value is None:
            return float(default)
        if isinstance(value, (int, float)):
            return float(value)
        if hasattr(value, "hour") and hasattr(value, "minute"):
            return float(value.hour) * 60.0 + float(value.minute) + float(getattr(value, "second", 0)) / 60.0
        parts = str(value).strip().split(":")
        try:
            if len(parts) >= 2:
                return float(parts[0]) * 60.0 + float(parts[1]) + (float(parts[2]) / 60.0 if len(parts) > 2 else 0.0)
        except Exception:
            pass
        return float(default)

    @staticmethod
    def _find_column(df, candidates):
        if df is None:
            return None
        lower = {str(c).strip().lower(): c for c in df.columns}
        for candidate in candidates:
            key = str(candidate).strip().lower()
            if key in lower:
                return lower[key]
        return None

    @staticmethod
    def _haversine(lat1, lon1, lat2, lon2):
        R = 6371.0
        p1 = math.radians(float(lat1))
        p2 = math.radians(float(lat2))
        dphi = math.radians(float(lat2) - float(lat1))
        dlambda = math.radians(float(lon2) - float(lon1))
        a = math.sin(dphi / 2.0) ** 2 + math.cos(p1) * math.cos(p2) * math.sin(dlambda / 2.0) ** 2
        return 2.0 * R * math.asin(math.sqrt(max(0.0, min(1.0, a))))

    @staticmethod
    def _normalize_fixed_cost_mode(value):
        txt = str(value or "").strip().upper()
        if txt in {"ALWAYS", "INCONDICIONAL", "UNCONDITIONAL", "SEMPRE", "2"}:
            return "ALWAYS"
        return "WHEN_USED"

    @staticmethod
    def _vehicle_numeric_id(value, fallback):
        if isinstance(value, (int, float)):
            return int(value)
        digits = "".join(ch for ch in str(value or "") if ch.isdigit())
        return int(digits) if digits else int(fallback)

    def _translated_description(self):
        raw = self.shared_memory.data.get("translated_json")
        if isinstance(raw, str):
            try:
                raw = json.loads(raw)
            except Exception:
                raw = {}
        if not isinstance(raw, dict):
            raw = {}
        return raw.get("problem_description", {}) or {}

    def _load_problem(self):
        mem = self.shared_memory.data
        ans = mem.get("user_answers", {}) or {}
        desc = self._translated_description()

        ctx = mem.get("objective_context")
        if not isinstance(ctx, dict):
            raise ValueError("IndependentFeasibilityVerifier: objective_context ausente.")
        self.objective_context = ctx

        dfc = mem.get("df_clientes")
        dfv = mem.get("df_veiculos_capacidade")
        dfcost = mem.get("df_veiculos_custo_km")
        dfw = mem.get("df_janela_tempo_cliente")
        if dfc is None or getattr(dfc, "empty", True):
            raise ValueError("IndependentFeasibilityVerifier: df_clientes ausente.")
        if dfv is None or getattr(dfv, "empty", True):
            raise ValueError("IndependentFeasibilityVerifier: df_veiculos_capacidade ausente.")

        cidc = self._find_column(dfc, ["Cliente", "cliente", "id_cliente"])
        latc = self._find_column(dfc, ["latitude", "lat"])
        lonc = self._find_column(dfc, ["longitude", "lon", "lng"])
        demc = self._find_column(dfc, ["Demanda", "demanda", "demand"])
        if None in (cidc, latc, lonc, demc):
            raise ValueError("IndependentFeasibilityVerifier: colunas de clientes incompletas.")

        clients = {}
        for _, row in dfc.iterrows():
            cid = self._safe_int(row[cidc], None)
            if cid is None:
                continue
            clients[cid] = {
                "lat": float(row[latc]),
                "lon": float(row[lonc]),
                "demand": float(row[demc]),
            }
        if 0 not in clients:
            raise ValueError("IndependentFeasibilityVerifier: depósito ID 0 ausente.")
        customers = sorted(k for k in clients if k != 0)

        capc = self._find_column(dfv, ["capacidade", "CAPACIDADE"])
        vidc = self._find_column(dfv, ["id_veiculo", "ID_VEICULO", "veiculo"])
        if capc is None:
            raise ValueError("IndependentFeasibilityVerifier: coluna capacidade ausente.")
        vehicles = {}
        for idx, (_, row) in enumerate(dfv.reset_index(drop=True).iterrows()):
            vid = self._vehicle_numeric_id(row[vidc], idx + 1) if vidc is not None else idx + 1
            vehicles[vid] = {"vehicle_id": vid, "capacity": float(row[capc]), "cost_per_km": 0.0, "fixed_cost": 0.0}

        if dfcost is not None and not getattr(dfcost, "empty", True):
            cvid = self._find_column(dfcost, ["id_veiculo", "ID_VEICULO", "veiculo"])
            cpk = self._find_column(dfcost, ["custo_por_km", "CUSTO_POR_KM", "cost_per_km"])
            cfix = self._find_column(dfcost, ["custo_diario", "CUSTO_DIARIO", "custo_fixo"])
            for idx, (_, row) in enumerate(dfcost.reset_index(drop=True).iterrows()):
                vid = self._vehicle_numeric_id(row[cvid], idx + 1) if cvid is not None else idx + 1
                if vid in vehicles:
                    if cpk is not None:
                        vehicles[vid]["cost_per_km"] = float(row[cpk])
                    if cfix is not None:
                        vehicles[vid]["fixed_cost"] = float(row[cfix])

        windows = {}
        window_mode = str(desc.get("time_window_mode", "NONE") or "NONE").upper()
        has_time_windows = self._parse_bool(desc.get("has_time_windows"), False)
        if has_time_windows and window_mode == "UNIFORM":
            a = self._safe_float(desc.get("uniform_time_window_start_minutes"), None)
            b = self._safe_float(desc.get("uniform_time_window_end_minutes"), None)
            if a is None or b is None or b <= a:
                raise ValueError("IndependentFeasibilityVerifier: janela uniforme inválida no contrato.")
            windows = {cid: (float(a), float(b)) for cid in customers}
        elif has_time_windows and window_mode == "PER_CLIENT":
            if dfw is None or getattr(dfw, "empty", True):
                raise ValueError("IndependentFeasibilityVerifier: janelas por cliente ausentes.")
            wid = self._find_column(dfw, ["cliente", "Cliente", "id_cliente"])
            ws_col = self._find_column(dfw, ["inicio_janela", "início_janela", "inicio"])
            we_col = self._find_column(dfw, ["fim_janela", "fim"])
            if None in (wid, ws_col, we_col):
                raise ValueError("IndependentFeasibilityVerifier: colunas de janelas ausentes.")
            for _, row in dfw.iterrows():
                cid = self._safe_int(row[wid], None)
                if cid is not None and cid != 0:
                    a, b = self._parse_time(row[ws_col]), self._parse_time(row[we_col])
                    if b <= a:
                        raise ValueError(f"IndependentFeasibilityVerifier: janela inválida para cliente {cid}.")
                    windows[cid] = (a, b)

        speed = self._safe_float(desc.get("average_vehicle_speed_kmh"), None)
        if speed is None or speed <= 0:
            speed = self._safe_float(ans.get("Qual é a velocidade média dos veículos (em Km/h)?"), 40.0)
        service = self._safe_float(desc.get("service_time_per_delivery_minutes"), None)
        if service is None:
            service = self._safe_float(ans.get("Qual é o tempo médio que cada entrega consume (em minutos)?"), 0.0)
        max_time = self._safe_float(desc.get("max_time_per_vehicle_minutes"), None)
        if max_time is None or max_time <= 0:
            max_time = self._safe_float(ans.get("Tempo disponível por veículo por dia (minutos)?"), 480.0)
        min_usage = self._safe_float(desc.get("min_usage_time_minutes", desc.get("minimum_vehicle_usage_minutes")), None)
        if min_usage is None:
            min_usage = self._safe_float(ans.get("tempo_minimo_minutos"), 0.0)
        has_min = self._parse_bool(desc.get("has_min_usage_time"), min_usage > 0)
        if not has_min:
            min_usage = 0.0
        allow_multi = self._parse_bool(desc.get("allow_multi_trip"), self._parse_bool(ans.get("Um veiculo pode fazer mais de uma viagem por dia?"), True))
        must_use_all = self._parse_bool(desc.get("must_use_all_vehicles"), self._parse_bool(ans.get("must_use_all_vehicles"), False))
        fixed_mode = self._normalize_fixed_cost_mode(desc.get("fixed_cost_mode", ans.get("fixed_cost_mode", "WHEN_USED")))

        op_start = self._safe_float(desc.get("vehicle_start_time_minutes"), 0.0)
        op_end = self._safe_float(desc.get("vehicle_end_time_minutes"), op_start + float(max_time))
        if op_end <= op_start:
            raise ValueError("IndependentFeasibilityVerifier: horizonte operacional inválido.")
        full_windows = {cid: windows.get(cid, (op_start, op_end)) for cid in customers}

        p = {
            "customers": customers,
            "clients": clients,
            "vehicles": vehicles,
            "windows": full_windows,
            "speed": float(speed),
            "service_time": float(service),
            "max_time": float(max_time),
            "minimum_usage": float(min_usage),
            "allow_multi_trip": bool(allow_multi),
            "must_use_all_vehicles": bool(must_use_all),
            "fixed_cost_mode": fixed_mode,
            "cost_data_available": bool(dfcost is not None and not getattr(dfcost, "empty", True)),
            "operation_start": float(op_start),
            "operation_end": float(op_end),
        }
        self.problem = p
        return p

    def _canonical_score(self, components):
        ctx = self.objective_context
        active = ctx.get("active_objectives", []) or []
        coeff = ctx.get("coefficients", {}) or {}
        return float(sum(float(coeff.get(obj, 0.0)) * float(components.get(obj, 0.0)) for obj in active))

    def _simulate_trip_from_path(self, path, start_time, vehicle):
        p = self.problem
        if len(path) < 3 or int(path[0]) != 0 or int(path[-1]) != 0:
            return {"feasible": False, "reason": "TRIP_MUST_START_AND_END_AT_DEPOT"}

        sequence = [int(x) for x in path[1:-1]]
        if not sequence:
            return {"feasible": False, "reason": "EMPTY_TRIP"}
        if any(cid not in p["customers"] for cid in sequence):
            return {"feasible": False, "reason": "UNKNOWN_CUSTOMER"}
        if len(sequence) != len(set(sequence)):
            return {"feasible": False, "reason": "DUPLICATE_WITHIN_TRIP"}

        now = float(start_time)
        current = 0
        distance = 0.0
        load = 0.0
        waiting = 0.0
        details = []

        if now < p["operation_start"] - self.TOL or now > p["operation_end"] + self.TOL:
            return {"feasible": False, "reason": "START_OUTSIDE_OPERATION_HORIZON"}

        for cid in sequence:
            d = self._haversine(
                p["clients"][current]["lat"], p["clients"][current]["lon"],
                p["clients"][cid]["lat"], p["clients"][cid]["lon"],
            )
            travel = d / p["speed"] * 60.0
            arrival = now + travel
            ws, we = p["windows"][cid]
            service_start = max(arrival, float(ws))
            if service_start > float(we) + self.TOL:
                return {"feasible": False, "reason": "TIME_WINDOW", "customer": cid, "arrival": arrival, "window": (ws, we)}
            waiting += max(0.0, service_start - arrival)
            load += float(p["clients"][cid]["demand"])
            if load > float(vehicle["capacity"]) + self.TOL:
                return {"feasible": False, "reason": "CAPACITY", "load": load, "capacity": vehicle["capacity"]}
            departure = service_start + p["service_time"]
            details.append({
                "node_id": cid,
                "arrival_time_minutes": arrival,
                "service_start_time_minutes": service_start,
                "departure_time_minutes": departure,
                "cumulative_load": load,
            })
            distance += d
            now = departure
            current = cid

        dback = self._haversine(
            p["clients"][current]["lat"], p["clients"][current]["lon"],
            p["clients"][0]["lat"], p["clients"][0]["lon"],
        )
        distance += dback
        now += dback / p["speed"] * 60.0
        if now > p["operation_end"] + self.TOL:
            return {"feasible": False, "reason": "RETURN_AFTER_OPERATION_END", "end": now, "operation_end": p["operation_end"]}

        return {
            "feasible": True,
            "start": float(start_time),
            "end": float(now),
            "duration": float(now - start_time),
            "distance_km": float(distance),
            "load": float(load),
            "waiting_minutes": float(waiting),
            "details": details,
            "customers": sequence,
        }

    def verify(self, solution=None, store_key="independent_feasibility_report"):
        p = self._load_problem()
        mem = self.shared_memory.data
        if solution is None:
            solution = mem.get("best_incumbent") or mem.get("solver_result")

        report = {
            "version": self.VERSION,
            "feasible": False,
            "checks": {},
            "violations": [],
            "recomputed": {},
            "solver_declared": {},
            "differences": {},
            "normalized_solution": None,
        }

        if not isinstance(solution, dict) or solution.get("status") != "SUCCESS":
            report["violations"].append("Solução ausente ou status diferente de SUCCESS.")
            mem[store_key] = report
            return report

        routes = solution.get("routes", []) or []
        if not routes:
            report["violations"].append("Solução sem rotas.")
            mem[store_key] = report
            return report

        coverage = Counter()
        per_vehicle = defaultdict(list)
        normalized_routes = []
        depot_ok = True
        capacity_ok = True
        windows_ok = True
        route_time_ok = True
        unknown_vehicle_ok = True

        for idx, route in enumerate(routes, start=1):
            raw_path = route.get("path", []) or []
            try:
                path = [int(x) for x in raw_path]
            except Exception:
                report["violations"].append(f"Rota {idx}: path inválido.")
                depot_ok = False
                continue

            if len(path) < 3 or path[0] != 0 or path[-1] != 0:
                depot_ok = False
                report["violations"].append(f"Rota {idx}: deve iniciar e terminar no depósito 0.")
                continue

            vid = self._safe_int(route.get("physical_vehicle_id", route.get("vehicle_id")), None)
            if vid not in p["vehicles"]:
                unknown_vehicle_ok = False
                report["violations"].append(f"Rota {idx}: veículo físico inválido {vid}.")
                continue

            start = self._safe_float(route.get("start_time_minutes"), None)
            if start is None:
                report["violations"].append(f"Rota {idx}: start_time_minutes ausente; impossível reconstruir sequência temporal independentemente.")
                route_time_ok = False
                continue

            sim = self._simulate_trip_from_path(path, start, p["vehicles"][vid])
            if not sim.get("feasible"):
                reason = sim.get("reason")
                if reason == "CAPACITY": capacity_ok = False
                if reason == "TIME_WINDOW": windows_ok = False
                if reason in {"TRIP_MUST_START_AND_END_AT_DEPOT", "EMPTY_TRIP"}: depot_ok = False
                if reason in {"RETURN_AFTER_OPERATION_END", "START_OUTSIDE_OPERATION_HORIZON"}: route_time_ok = False
                report["violations"].append(f"Rota {idx}: {reason}.")
                continue

            for cid in sim["customers"]:
                coverage[cid] += 1

            nr = copy.deepcopy(route)
            nr.update({
                "vehicle_id": vid,
                "physical_vehicle_id": vid,
                "path": path,
                "start_time_minutes": sim["start"],
                "end_time_minutes": sim["end"],
                "total_route_time_minutes": sim["duration"],
                "distance_km": sim["distance_km"],
                "total_route_load": sim["load"],
                "waiting_time_minutes": sim["waiting_minutes"],
                "path_details": sim["details"],
            })
            normalized_routes.append(nr)
            per_vehicle[vid].append(nr)

        missing = [cid for cid in p["customers"] if coverage[cid] == 0]
        duplicated = [cid for cid in p["customers"] if coverage[cid] > 1]
        unknown_covered = [cid for cid in coverage if cid not in p["customers"]]
        coverage_ok = not missing and not duplicated and not unknown_covered
        if missing:
            report["violations"].append(f"Clientes não atendidos: {missing[:30]}")
        if duplicated:
            report["violations"].append(f"Clientes duplicados: {duplicated[:30]}")
        if unknown_covered:
            report["violations"].append(f"Clientes desconhecidos nas rotas: {unknown_covered[:30]}")

        sequence_ok = True
        daily_time_ok = True
        minimum_usage_ok = True
        multitrip_ok = True
        total_vehicle_time = {}

        for vid, vroutes in per_vehicle.items():
            vroutes.sort(key=lambda r: (float(r["start_time_minutes"]), int(r.get("trip_number", r.get("trip_id", 0)) or 0)))
            total_work = sum(float(r["total_route_time_minutes"]) for r in vroutes)
            total_vehicle_time[vid] = total_work
            if total_work > p["max_time"] + self.TOL:
                daily_time_ok = False
                report["violations"].append(f"V{vid}: jornada {total_work:.6f} > {p['max_time']:.6f} min.")
            if p["minimum_usage"] > 0 and total_work + self.TOL < p["minimum_usage"]:
                minimum_usage_ok = False
                report["violations"].append(f"V{vid}: utilização {total_work:.6f} < mínimo {p['minimum_usage']:.6f} min.")
            if not p["allow_multi_trip"] and len(vroutes) > 1:
                multitrip_ok = False
                report["violations"].append(f"V{vid}: múltiplas viagens não permitidas.")
            for a, b in zip(vroutes, vroutes[1:]):
                previous_end = float(a["end_time_minutes"])
                next_start = float(b["start_time_minutes"])

                # O sequenciamento Multi-Trip usa uma tolerância temporal própria.
                # Isso evita reprovar rotas por diferenças microscópicas causadas
                # por arredondamento (por exemplo, 626.181732 vs 626.181731).
                # A tolerância NÃO cria folga operacional relevante: 1e-5 min
                # corresponde a apenas 0,0006 segundo.
                if previous_end > next_start + self.SEQUENCE_TOL_MIN:
                    sequence_ok = False
                    report["violations"].append(
                        f"V{vid}: sobreposição entre viagens "
                        f"({previous_end:.6f} > {next_start:.6f}; "
                        f"tol={self.SEQUENCE_TOL_MIN:.6f} min)."
                    )

        must_use_all_ok = True
        if p["must_use_all_vehicles"]:
            missing_vehicles = sorted(set(p["vehicles"]) - set(per_vehicle))
            if missing_vehicles:
                must_use_all_ok = False
                report["violations"].append(f"Veículos obrigatórios não utilizados: {missing_vehicles}")

        total_distance = sum(float(r["distance_km"]) for r in normalized_routes)
        total_time = sum(float(r["total_route_time_minutes"]) for r in normalized_routes)
        variable_cost = sum(
            float(r["distance_km"]) * float(p["vehicles"][r["physical_vehicle_id"]]["cost_per_km"])
            for r in normalized_routes
        )
        used_ids = sorted(per_vehicle)
        if p["fixed_cost_mode"] == "WHEN_USED":
            fixed_cost = sum(float(p["vehicles"][vid]["fixed_cost"]) for vid in used_ids)
        else:
            fixed_cost = sum(float(v["fixed_cost"]) for v in p["vehicles"].values())
        total_cost = variable_cost + fixed_cost
        components = {
            "DISTANCE": total_distance,
            "TIME": total_time,
            "COST": total_cost,
            "VEHICLES": float(len(used_ids)),
        }
        objective = self._canonical_score(components)

        declared_summary = solution.get("summary", {}) or {}
        declared_obj = self._safe_float(solution.get("canonical_objective_score", solution.get("solver_objective", solution.get("fitness"))), None)
        report["solver_declared"] = {
            "distance_km": self._safe_float(declared_summary.get("total_distance_km"), None),
            "time_minutes": self._safe_float(declared_summary.get("total_time_minutes"), None),
            "variable_cost": self._safe_float(declared_summary.get("total_variable_cost"), None),
            "fixed_cost": self._safe_float(declared_summary.get("total_fixed_cost"), None),
            "total_cost": self._safe_float(declared_summary.get("total_cost"), None),
            "canonical_objective_score": declared_obj,
        }
        report["recomputed"] = {
            "total_distance_km": total_distance,
            "total_time_minutes": total_time,
            "total_variable_cost": variable_cost,
            "total_fixed_cost": fixed_cost,
            "total_cost": total_cost,
            "num_vehicles_used": len(used_ids),
            "num_routes": len(normalized_routes),
            "vehicle_total_time_minutes": total_vehicle_time,
            "objective_components": components,
            "canonical_objective_score": objective,
        }
        for key, declared_key in [
            ("total_distance_km", "distance_km"),
            ("total_time_minutes", "time_minutes"),
            ("total_variable_cost", "variable_cost"),
            ("total_fixed_cost", "fixed_cost"),
            ("total_cost", "total_cost"),
            ("canonical_objective_score", "canonical_objective_score"),
        ]:
            declared = report["solver_declared"].get(declared_key)
            actual = report["recomputed"].get(key)
            report["differences"][key] = None if declared is None else float(declared - actual)

        report["checks"] = {
            "all_customers_served_once": coverage_ok,
            "no_duplicates": not duplicated,
            "start_end_depot": depot_ok,
            "known_physical_vehicle": unknown_vehicle_ok,
            "capacity_per_trip": capacity_ok,
            "time_windows": windows_ok,
            "route_operation_horizon": route_time_ok,
            "multi_trip_sequence_nonoverlap": sequence_ok,
            "multi_trip_allowed_rule": multitrip_ok,
            "daily_work_time_per_physical_vehicle": daily_time_ok,
            "minimum_usage_per_used_vehicle": minimum_usage_ok,
            "must_use_all_vehicles": must_use_all_ok,
            "distance_recomputed": True,
            "variable_cost_recomputed": True,
            "fixed_cost_recomputed": True,
            "canonical_objective_recomputed": True,
        }

        feasible = all(report["checks"].values()) and len(normalized_routes) > 0
        report["feasible"] = bool(feasible)
        report["normalized_solution"] = {
            "status": "SUCCESS" if feasible else "INFEASIBLE",
            "method": solution.get("method", "INDEPENDENTLY_RECONSTRUCTED"),
            "routes": normalized_routes,
            "summary": {
                "total_distance_km": total_distance,
                "total_time_minutes": total_time,
                "total_variable_cost": variable_cost,
                "total_fixed_cost": fixed_cost,
                "total_cost": total_cost,
                "num_vehicles_used": len(used_ids),
                "num_routes": len(normalized_routes),
            },
            "objective_components": components,
            "canonical_objective_score": objective,
            "objective_context": copy.deepcopy(self.objective_context),
        }

        mem[store_key] = copy.deepcopy(report)
        return report

    def compute_fast_bounds(self):
        """Bounds simples e seguros da Camada 1. Não substituem o dual bound do MIP."""
        if self.problem is None:
            self._load_problem()
        p = self.problem
        total_demand = sum(float(p["clients"][cid]["demand"]) for cid in p["customers"])
        max_capacity = max(float(v["capacity"]) for v in p["vehicles"].values())
        min_trips_capacity = int(math.ceil(total_demand / max_capacity - 1e-12)) if max_capacity > 0 else None

        if not p["allow_multi_trip"]:
            min_vehicles_capacity = min_trips_capacity
        else:
            min_vehicles_capacity = 1 if p["customers"] else 0

        return {
            "total_demand": total_demand,
            "max_vehicle_capacity": max_capacity,
            "capacity_lower_bound_on_trips": min_trips_capacity,
            "safe_lower_bound_on_vehicles_from_capacity": min_vehicles_capacity,
            "distance_nonnegative_lower_bound": 0.0,
            "variable_cost_nonnegative_lower_bound": 0.0,
            "note": "Com Multi-Trip, capacidade por viagem não implica ceil(demanda/capacidade) veículos; por isso o bound de veículos é conservador.",
        }


@dataclass
class AuditContext:
    user_answers: Dict[str, Any]
    translated_json: Any
    problem_classification: str
    optimization_approach: str
    solver_result: Dict[str, Any]
    df_clientes: Any
    df_veiculos_capacidade: Any
    df_veiculos_custo_km: Optional[Any] = None
    df_janela_tempo_cliente: Optional[Any] = None
    tentativas: int = 0
    metodo_utilizado: Optional[str] = None
    teve_solucao: bool = False
    estabilidade_solucao: Optional[float] = None
    tempo_execucao_segundos: Optional[float] = None
    motivo_parada: Optional[str] = None
    erros_internos: List[str] = None


def _vrp_fingerprint(value):
    import hashlib, json
    def freeze(obj):
        if hasattr(obj, "to_dict") and hasattr(obj, "columns"):
            return {"columns": list(map(str, obj.columns)), "rows": freeze(obj.to_dict("records"))}
        if isinstance(obj, dict):
            return {str(k): freeze(v) for k, v in sorted(obj.items(), key=lambda kv: str(kv[0]))}
        if isinstance(obj, (list, tuple)):
            return [freeze(v) for v in obj]
        if isinstance(obj, set):
            return sorted((freeze(v) for v in obj), key=repr)
        if hasattr(obj, "item"):
            return freeze(obj.item())
        if obj is None or isinstance(obj, (str, int, float, bool)):
            return obj
        return str(obj)
    return hashlib.sha256(json.dumps(freeze(value), sort_keys=True, ensure_ascii=False).encode()).hexdigest()


def _vrp_ensure_proof_scope(shared_memory):
    """Descarta provas antigas quando os dados ou o objetivo canônico mudam."""
    mem = shared_memory.data
    ctx = mem.get("objective_context") or {}
    payload = {key: mem.get(key) for key in (
        "df_clientes", "df_veiculos_capacidade", "df_veiculos_custo_km",
        "df_janela_tempo_cliente", "translated_json", "user_answers")}
    payload["objective_coefficients"] = ctx.get("coefficients", {})
    signature = _vrp_fingerprint(payload)
    if mem.get("proof_scope_v5") != signature:
        for key in list(mem):
            if key.startswith(("k_exact_", "k_vehicle_feasibility_", "bound_", "ave_v2_", "scf_lp_")) or key in {
                "certified_vehicle_lower_bound", "continuity_best_vehicle_lower_bound",
                "certified_lower_bound", "lower_bound", "continuity_best_valid_lower_bound",
                "continuity_baseline_valid_lower_bound", "continuity_carried_valid_lower_bound",
                "certification_strengthened_bounds", "certification_bound_catalog",
            }:
                mem.pop(key, None)
        mem["proof_scope_v5"] = signature
    return signature


def _vrp_vehicle_id(value):
    try:
        x = float(value)
        if x.is_integer():
            return str(int(x))
    except (ValueError, TypeError, OverflowError):
        pass
    return str(value).strip()


def _vrp_interval_vehicle_lb(windows, service):
    """Janelas são de INÍCIO do serviço, não de término.

    Para cada intervalo, calcula a menor sobreposição obrigatória de cada
    serviço sobre todas as suas partidas permitidas. Ignora deslocamentos.
    """
    import math
    service = float(service)
    if not windows or service <= 0:
        return 1, None
    ends = sorted({x for a, b in windows for x in (a, b, a + service, b + service)})
    best, witness = 1, None
    for ia, a in enumerate(ends):
        for b in ends[ia + 1:]:
            if b <= a:
                continue
            def overlap(start):
                return max(0.0, min(b, start + service) - max(a, start))
            work = sum(min(overlap(ws), overlap(we)) for ws, we in windows)
            lb = int(math.ceil(work / (b - a) - 1e-9))
            if lb > best:
                best, witness = lb, {"start": a, "end": b, "mandatory_service": work}
    return best, witness


def _vrp_collect_jobs(ctx, specs, worker, deadline):
    """Uma identidade por processo; toda saída limpa todos os filhos.

    specs = [{job_id, payload, quota}]. O worker devolve o mesmo job_id.
    O deadline inclui construção, solução e comunicação, medido no pai.
    """
    import time, queue
    q = ctx.Queue()
    jobs, received = {}, {}
    try:
        for spec in specs:
            if time.monotonic() >= deadline:
                break
            proc = ctx.Process(target=worker, args=(spec, q))
            proc.start()
            jobs[spec["job_id"]] = (proc, spec)
        while len(received) < len(jobs):
            left = deadline - time.monotonic()
            if left <= 0:
                break
            try:
                item = q.get(timeout=min(0.25, left))
            except queue.Empty:
                if all(not p.is_alive() for p, _ in jobs.values()):
                    # O feeder pode entregar a mensagem logo após a saída.
                    try:
                        item = q.get(timeout=min(0.1, max(0.001, deadline-time.monotonic())))
                    except queue.Empty:
                        break
                else:
                    continue
            jid = item.get("job_id")
            if jid in jobs and jid not in received:
                received[jid] = item
                # Não encerra outro worker quando as respostas chegam fora de ordem.
                if item.get("has_model_witness"):
                    break
        for jid, (proc, spec) in jobs.items():
            if jid not in received:
                reason = "CANCELLED_AFTER_WITNESS" if any(r.get("has_model_witness") for r in received.values()) else (
                    "WALL_DEADLINE" if time.monotonic() >= deadline else "WORKER_EXIT_WITHOUT_RESULT")
                received[jid] = {"job_id": jid, "status": "unknown", "error": reason,
                                 "quota": spec["quota"], "payload": spec["payload"]}
        return list(received.values())
    finally:
        for proc, _ in jobs.values():
            proc.join(timeout=0.05)
            if proc.is_alive():
                proc.terminate()
                proc.join(timeout=1.0)
            if proc.is_alive() and hasattr(proc, "kill"):
                proc.kill()
                proc.join(timeout=1.0)
        q.close()
        q.cancel_join_thread()


def _vrp_scf_distance_bound(problem, trip_lb, budget_seconds, cache):
    """LP SCF de capacidade Qmax: relaxação global, sem restrição a rotas do GA.

    Só aceita ótimo do LP. Subtrai tolerância conservadora do valor numérico.
    Não usa o primal de um LP interrompido como limite inferior.
    """
    import time, math
    t0 = time.monotonic()
    signature = _vrp_fingerprint((problem, int(trip_lb)))
    if signature in cache:
        return dict(cache[signature], cache_hit=True)
    result = {"valid": False, "status": "NOT_RUN", "distance_lb": None, "cache_hit": False}
    if budget_seconds < 1:
        return dict(result, status="NO_BUDGET")
    try:
        import numpy as np
        from scipy.optimize import linprog
        from scipy.sparse import lil_matrix
        C = list(problem["customers"])
        if not C:
            return dict(result, valid=True, status="EMPTY", distance_lb=0.0)
        Q = max(float(v["capacity"]) for v in problem["vehicles"])
        if Q <= 0:
            return dict(result, status="INVALID_CAPACITY")
        nodes = [0] + C
        arcs = [(i, j) for i in nodes for j in nodes if i != j]
        index = {a: n for n, a in enumerate(arcs)}
        na = len(arcs)
        c = np.zeros(2*na)
        eq = lil_matrix((3*len(C)+1, 2*na)); rhs = np.zeros(eq.shape[0])
        ub = lil_matrix((na+1, 2*na)); urhs = np.zeros(na+1)
        for col, (i, j) in enumerate(arcs):
            c[col] = float(problem["distance"][i, j])
            ub[col, col] = -Q
            ub[col, na+col] = 1.0
        row = 0
        for j in C:
            for i in nodes:
                if i != j: eq[row, index[i, j]] = 1.0
            rhs[row] = 1.0; row += 1
            for i in nodes:
                if i != j: eq[row, index[j, i]] = 1.0
            rhs[row] = 1.0; row += 1
            for i in nodes:
                if i != j:
                    eq[row, na+index[i, j]] = 1.0
                    eq[row, na+index[j, i]] = -1.0
            rhs[row] = float(problem["client_rows"][j]["demand"]); row += 1
        for j in C:
            eq[row, index[0, j]] = 1.0
            eq[row, index[j, 0]] = -1.0
            ub[na, index[0, j]] = -1.0
        urhs[na] = -max(1, int(trip_lb))
        bounds = [(0.0, 1.0)]*na + [(0.0, 0.0 if j == 0 else Q) for i, j in arcs]
        left = float(budget_seconds)-(time.monotonic()-t0)
        if left <= 0:
            return dict(result, status="BUILD_BUDGET_EXHAUSTED")
        sol = linprog(c, A_ub=ub.tocsr(), b_ub=urhs, A_eq=eq.tocsr(), b_eq=rhs,
                      bounds=bounds, method="highs", options={"time_limit":left})
        result.update(status=str(sol.message), elapsed_seconds=time.monotonic()-t0)
        if sol.success and sol.status == 0 and math.isfinite(float(sol.fun)):
            value = max(0.0, float(sol.fun)-1e-7*max(1.0, abs(float(sol.fun))))
            result.update(valid=True, status="OPTIMAL_SCF_LP", distance_lb=value)
            cache[signature] = dict(result)
    except Exception as exc:
        result.update(status="LP_UNAVAILABLE_OR_ERROR", error=f"{type(exc).__name__}: {exc}")
    return result
````

Antes de organizar as viagens, ele confere se a ficha do problema está preenchida de maneira coerente. Essa conferência evita iniciar a busca com dados incompletos ou incompatíveis.





Tabela com células delimitadas: Responsabilidade / Explicação



````python
class ValidatorAgent:
    VERSION = "VALIDATOR_V10_CANONICAL_GATE"

    def __init__(self, shared_memory):
        self.shared_memory = shared_memory
        self.role = "Agente Validador de Dados e Gate de Integridade."
        self.goal = "Impedir que dados inválidos avancem para tradução e otimização."
        self.errors = []

    @staticmethod
    def _parse_time(value):
        if value is None:
            return None
        if hasattr(value, "hour") and hasattr(value, "minute"):
            return float(value.hour * 60 + value.minute + getattr(value, "second", 0) / 60.0)
        txt = str(value).strip()
        for fmt in ("%H:%M:%S", "%H:%M"):
            try:
                t = pd.to_datetime(txt, format=fmt).time()
                return float(t.hour * 60 + t.minute + t.second / 60.0)
            except Exception:
                pass
        return None

    def _validate_clientes_df(self, df):
        required = ["Cliente", "latitude", "longitude", "Demanda"]
        if not isinstance(df, pd.DataFrame):
            self.errors.append("DF Clientes: deve ser um DataFrame pandas.")
            return
        missing = [c for c in required if c not in df.columns]
        if missing:
            self.errors.append(f"DF Clientes: colunas obrigatórias ausentes: {', '.join(missing)}.")
            return
        for c in required:
            if df[c].isnull().any():
                self.errors.append(f"DF Clientes: coluna '{c}' contém valores nulos.")
        if not pd.api.types.is_numeric_dtype(df["Cliente"]):
            self.errors.append("DF Clientes: 'Cliente' deve ser numérico.")
        else:
            if df["Cliente"].duplicated().any():
                self.errors.append("DF Clientes: IDs de clientes duplicados.")
            depot_count = int((df["Cliente"] == 0).sum())
            if depot_count != 1:
                self.errors.append(f"DF Clientes: deve existir exatamente um depósito com Cliente=0; encontrados {depot_count}.")
            elif "Demanda" in df.columns:
                try:
                    depot_demand = float(df.loc[df["Cliente"] == 0, "Demanda"].iloc[0])
                    if abs(depot_demand) > 1e-9:
                        self.errors.append("DF Clientes: a demanda do depósito Cliente=0 deve ser zero.")
                except Exception:
                    pass
        if not pd.api.types.is_numeric_dtype(df["latitude"]):
            self.errors.append("DF Clientes: 'latitude' deve ser numérica.")
        elif not df["latitude"].between(-90, 90).all():
            self.errors.append("DF Clientes: latitude fora do intervalo [-90, 90].")
        if not pd.api.types.is_numeric_dtype(df["longitude"]):
            self.errors.append("DF Clientes: 'longitude' deve ser numérica.")
        elif not df["longitude"].between(-180, 180).all():
            self.errors.append("DF Clientes: longitude fora do intervalo [-180, 180].")
        if not pd.api.types.is_numeric_dtype(df["Demanda"]):
            self.errors.append("DF Clientes: 'Demanda' deve ser numérica.")
        elif (df["Demanda"] < 0).any():
            self.errors.append("DF Clientes: demanda negativa não é permitida.")

    def _validate_veiculos_capacidade_df(self, df):
        required = ["id_veiculo", "capacidade"]
        if not isinstance(df, pd.DataFrame):
            self.errors.append("DF Veículos Capacidade: deve ser um DataFrame pandas.")
            return
        missing = [c for c in required if c not in df.columns]
        if missing:
            self.errors.append(f"DF Veículos Capacidade: colunas ausentes: {', '.join(missing)}.")
            return
        for c in required:
            if df[c].isnull().any():
                self.errors.append(f"DF Veículos Capacidade: coluna '{c}' contém nulos.")
        if not pd.api.types.is_numeric_dtype(df["id_veiculo"]):
            self.errors.append("DF Veículos Capacidade: 'id_veiculo' deve ser numérico.")
        elif df["id_veiculo"].duplicated().any():
            self.errors.append("DF Veículos Capacidade: IDs de veículos duplicados.")
        if not pd.api.types.is_numeric_dtype(df["capacidade"]):
            self.errors.append("DF Veículos Capacidade: 'capacidade' deve ser numérica.")
        elif (df["capacidade"] <= 0).any():
            self.errors.append("DF Veículos Capacidade: capacidade deve ser maior que zero.")

    def _validate_veiculos_custo_df(self, df):
        required = ["id_veiculo", "custo_por_km", "custo_diario"]
        if not isinstance(df, pd.DataFrame):
            self.errors.append("DF Veículos Custo: deve ser um DataFrame pandas.")
            return
        missing = [c for c in required if c not in df.columns]
        if missing:
            self.errors.append(f"DF Veículos Custo: colunas ausentes: {', '.join(missing)}.")
            return
        for c in required:
            if df[c].isnull().any():
                self.errors.append(f"DF Veículos Custo: coluna '{c}' contém nulos.")
        if not pd.api.types.is_numeric_dtype(df["id_veiculo"]):
            self.errors.append("DF Veículos Custo: 'id_veiculo' deve ser numérico.")
        elif df["id_veiculo"].duplicated().any():
            self.errors.append("DF Veículos Custo: IDs de veículos duplicados.")
        for c in ("custo_por_km", "custo_diario"):
            if not pd.api.types.is_numeric_dtype(df[c]):
                self.errors.append(f"DF Veículos Custo: '{c}' deve ser numérico.")
            elif (df[c] < 0).any():
                self.errors.append(f"DF Veículos Custo: '{c}' não pode ser negativo.")

    def _validate_multiobjective_config(self, answers):
        if "multiobjetivo" not in str(answers.get("objective_otimizacao", "")).lower():
            return
        cfg = answers.get("multiobjective_config")
        if not isinstance(cfg, dict):
            self.errors.append("Multiobjetivo: configuração deve ser um dicionário.")
            return
        objectives = cfg.get("objectives", [])
        weights = cfg.get("weights", {})
        valid = {"DISTANCE", "TIME", "COST", "VEHICLES"}
        if not isinstance(objectives, list) or len(set(objectives)) < 2:
            self.errors.append("Multiobjetivo: selecione pelo menos dois objetivos distintos.")
            return
        if any(o not in valid for o in objectives):
            self.errors.append("Multiobjetivo: existe objetivo inválido.")
        if not isinstance(weights, dict):
            self.errors.append("Multiobjetivo: pesos devem ser um dicionário.")
            return
        try:
            vals = [float(weights[o]) for o in objectives]
            if any(v < 0 for v in vals):
                self.errors.append("Multiobjetivo: pesos não podem ser negativos.")
            if abs(sum(vals) - 1.0) > 1e-6:
                self.errors.append(f"Multiobjetivo: a soma dos pesos deve ser 1.0; encontrada {sum(vals):.6f}.")
        except Exception:
            self.errors.append("Multiobjetivo: pesos ausentes ou inválidos.")

    def _validate_windows(self, answers, df_windows):
        exists = str(answers.get("Existem janelas de tempo? (1 - sim, 2 - não)", "2")) == "1"
        if not exists:
            if df_windows is not None and not getattr(df_windows, "empty", True):
                self.errors.append("Janelas: planilha fornecida apesar de a UI indicar que não existem janelas.")
            return

        mode = str(answers.get("As janelas de tempo são iguais para todos os clientes? (1 - sim, 2 - não)", ""))
        if mode == "1":
            raw = answers.get("Qual é a janela de tempo? (Ex: 08:00-17:00)")
            if not raw or "-" not in str(raw):
                self.errors.append("Janelas: informe a janela única no formato HH:MM-HH:MM.")
                return
            a, b = [x.strip() for x in str(raw).split("-", 1)]
            start, end = self._parse_time(a), self._parse_time(b)
            if start is None or end is None or end <= start:
                self.errors.append("Janelas: janela única inválida; o fim deve ser posterior ao início.")
            if df_windows is not None and not getattr(df_windows, "empty", True):
                self.errors.append("Janelas: não forneça planilha quando a janela é uniforme para todos os clientes.")
            return

        if mode != "2":
            self.errors.append("Janelas: informe se as janelas são iguais para todos (1) ou diferentes (2).")
            return
        if not isinstance(df_windows, pd.DataFrame) or df_windows.empty:
            self.errors.append("Janelas: a planilha por cliente é obrigatória quando as janelas são diferentes.")
            return
        required = ["cliente", "inicio_janela", "fim_janela"]
        missing = [c for c in required if c not in df_windows.columns]
        if missing:
            self.errors.append(f"Janelas: colunas ausentes: {', '.join(missing)}.")
            return
        if df_windows["cliente"].duplicated().any():
            self.errors.append("Janelas: clientes duplicados na planilha de janelas.")
        for _, row in df_windows.iterrows():
            start = self._parse_time(row["inicio_janela"])
            end = self._parse_time(row["fim_janela"])
            if start is None or end is None or end <= start:
                self.errors.append(f"Janelas: janela inválida para cliente {row['cliente']}.")

    def validate(self, answers, df_clientes, df_veiculos_capacidade=None,
                 df_veiculos_custo_km=None, df_janela_tempo_cliente=None):
        self.errors = []
        self._validate_multiobjective_config(answers)
        self._validate_clientes_df(df_clientes)
        self._validate_veiculos_capacidade_df(df_veiculos_capacidade)

        use_all_raw = answers.get("must_use_all_vehicles", answers.get("Todos os veiculos devem ser utilizados?"))
        if isinstance(use_all_raw, str) and use_all_raw.strip() in ("1", "2"):
            answers["must_use_all_vehicles"] = use_all_raw.strip() == "1"
        elif isinstance(use_all_raw, bool):
            answers["must_use_all_vehicles"] = use_all_raw
        elif use_all_raw is None:
            answers["must_use_all_vehicles"] = False
        else:
            self.errors.append("UI: 'Todos os veículos devem ser utilizados?' deve ser Sim/Não ou 1/2.")

        has_min = str(answers.get("veiculos_tem_tempo_minimo", "2")) == "1"
        try:
            min_usage = float(answers.get("tempo_minimo_minutos", 0) or 0)
            if min_usage < 0 or (has_min and min_usage <= 0):
                self.errors.append("UI: tempo mínimo deve ser > 0 quando habilitado e nunca negativo.")
        except Exception:
            self.errors.append("UI: tempo mínimo deve ser numérico.")

        cfg = answers.get("multiobjective_config", {}) or {}
        active = [str(x).upper() for x in cfg.get("objectives", [])]
        requires_cost = "COST" in active or (not active and "custo" in str(answers.get("objective_otimizacao", "")).lower())
        if requires_cost:
            if df_veiculos_custo_km is None or getattr(df_veiculos_custo_km, "empty", True):
                self.errors.append("Custos: planilha de custos é obrigatória quando COST é objetivo ativo.")
            else:
                self._validate_veiculos_custo_df(df_veiculos_custo_km)
            if answers.get("tipo_custo_fixo") not in ("ativacao", "incondicional"):
                self.errors.append("Custos: informe se o custo fixo é por ativação ou incondicional.")

        self._validate_windows(answers, df_janela_tempo_cliente)
        ok, coherence_errors = validar_coerencia_manual(answers, df_clientes, df_veiculos_capacidade)
        if not ok:
            self.errors.extend(coherence_errors)

        result = {
            "valid": not self.errors,
            "errors": list(self.errors),
            "version": self.VERSION,
        }
        if self.shared_memory is not None:
            self.shared_memory.data["validation_result"] = result

        if self.errors:
            return "Erros encontrados durante a validação:\n" + "\n".join(f"  - {e}" for e in self.errors)
        return "Validação concluída com sucesso! Todos os dados estão corretos."
````

### **TRANSLATORAGENT — PREPARAÇÃO DO PROBLEMA**

O usuário escreve informações de uma forma; os cálculos precisam delas em outra. Esse agente transforma os dados conferidos em uma estrutura comum para os demais componentes.





Tabela com células delimitadas: Responsabilidade / Explicação



````python
class TranslatorAgent:
    """Traduz a entrada validada em um contrato operacional canônico."""

    VERSION = "TRANSLATOR_OPERATIONAL_CONTRACT_10"

    def __init__(self, shared_memory):
        self.shared_memory = shared_memory
        self.role = "Agente Tradutor e Guardião do Contrato Operacional."
        self.goal = "Produzir uma especificação única consumida por solvers, Auditor e Certificador."

    @staticmethod
    def _parse_bool(value, default=False):
        if value is None:
            return default
        if isinstance(value, bool):
            return value
        if isinstance(value, (int, float)):
            return bool(value)
        txt = str(value).strip().lower()
        if txt in {"1", "true", "sim", "s", "yes", "y"}:
            return True
        if txt in {"0", "2", "false", "não", "nao", "n", "no"}:
            return False
        return default

    @staticmethod
    def _time_to_minutes(value):
        if value is None:
            return None
        if isinstance(value, (int, float)):
            return float(value)
        if isinstance(value, time):
            return float(value.hour * 60 + value.minute + value.second / 60.0)
        txt = str(value).strip()
        for fmt in ("%H:%M:%S", "%H:%M"):
            try:
                t = datetime.strptime(txt, fmt).time()
                return float(t.hour * 60 + t.minute + t.second / 60.0)
            except ValueError:
                pass
        return None

    def _uniform_window(self, answers):
        exists = str(answers.get("Existem janelas de tempo? (1 - sim, 2 - não)", "2")) == "1"
        same = str(answers.get("As janelas de tempo são iguais para todos os clientes? (1 - sim, 2 - não)", "")) == "1"
        if not (exists and same):
            return None
        raw = answers.get("Qual é a janela de tempo? (Ex: 08:00-17:00)")
        if not raw or "-" not in str(raw):
            raise ValueError("Janela uniforme inválida: use HH:MM-HH:MM.")
        start_raw, end_raw = [x.strip() for x in str(raw).split("-", 1)]
        start = self._time_to_minutes(start_raw)
        end = self._time_to_minutes(end_raw)
        if start is None or end is None or end <= start:
            raise ValueError("Janela uniforme inválida: o fim deve ser posterior ao início.")
        return float(start), float(end)

    def _derive_operation_start(self, answers, df_windows, uniform_window):
        for key in (
            "vehicle_start_time_minutes", "vehicle_start_time", "start_time",
            "Horário inicial da jornada", "Horario inicial da jornada",
        ):
            if answers.get(key) is not None:
                val = self._time_to_minutes(answers[key])
                if val is not None:
                    return float(val)
        if uniform_window is not None:
            return float(uniform_window[0])
        if df_windows is not None and not getattr(df_windows, "empty", True):
            for col in ("inicio_janela", "início_janela", "inicio"):
                if col in df_windows.columns:
                    vals = [self._time_to_minutes(v) for v in df_windows[col].tolist()]
                    vals = [v for v in vals if v is not None]
                    if vals:
                        return float(min(vals))
        return 0.0

    def translate_to_json(self, answers, df_clientes, df_veiculos_capacidade=None,
                          df_veiculos_custo_km=None, df_janela_tempo_cliente=None):
        has_min_usage = str(answers.get("veiculos_tem_tempo_minimo", "2")) == "1"
        min_usage = float(answers.get("tempo_minimo_minutos", 0) or 0)
        must_use_all = self._parse_bool(
            answers.get("must_use_all_vehicles", answers.get("Todos os veiculos devem ser utilizados?")), False
        )
        allow_multi = self._parse_bool(
            answers.get("Um veiculo pode fazer mais de uma viagem por dia?", answers.get("allow_multi_trip")), False
        )

        multi_cfg = answers.get("multiobjective_config", {"enabled": False, "objectives": [], "weights": {}}) or {}
        active = [str(x).upper() for x in multi_cfg.get("objectives", [])]
        requires_cost = "COST" in active or (not active and "custo" in str(answers.get("objective_otimizacao", "")).lower())
        cost_available = df_veiculos_custo_km is not None and not getattr(df_veiculos_custo_km, "empty", True)
        if requires_cost and not cost_available:
            raise ValueError("O objetivo inclui CUSTO, mas a planilha de custos por veículo não foi fornecida.")

        fixed_type = str(answers.get("tipo_custo_fixo", "ativacao") or "ativacao").lower()
        if fixed_type in {"incondicional", "always", "unconditional"}:
            fixed_type, fixed_mode = "incondicional", "ALWAYS"
        else:
            fixed_type, fixed_mode = "ativacao", "WHEN_USED"

        windows_exist = str(answers.get("Existem janelas de tempo? (1 - sim, 2 - não)", "2")) == "1"
        same_windows = str(answers.get("As janelas de tempo são iguais para todos os clientes? (1 - sim, 2 - não)", "")) == "1"
        uniform = self._uniform_window(answers)
        if windows_exist and same_windows:
            window_mode = "UNIFORM"
        elif windows_exist:
            window_mode = "PER_CLIENT"
        else:
            window_mode = "NONE"

        service_time = float(answers.get("Qual é o tempo médio que cada entrega consume (em minutos)?", 0) or 0)
        speed = float(answers.get("Qual é a velocidade média dos veículos (em Km/h)?", 0) or 0)
        max_time = float(answers.get("Tempo disponível por veículo por dia (minutos)?", 0) or 0)
        if speed <= 0 or max_time <= 0 or service_time < 0:
            raise ValueError("Velocidade e jornada devem ser positivas; tempo de serviço não pode ser negativo.")

        operation_start = self._derive_operation_start(answers, df_janela_tempo_cliente, uniform)
        operation_end = operation_start + max_time

        description = {
            "objective": answers.get("objective_otimizacao", "Não especificado"),
            "multiobjective_config": multi_cfg,
            "requires_cost_data": requires_cost,
            "cost_data_available": cost_available,
            "has_vehicles": True,
            "has_vehicle_capacity": df_veiculos_capacidade is not None and not getattr(df_veiculos_capacidade, "empty", True),
            "has_vehicle_variable_costs": cost_available,
            "has_time_windows": windows_exist,
            "time_window_mode": window_mode,
            "uniform_time_window_start_minutes": uniform[0] if uniform else None,
            "uniform_time_window_end_minutes": uniform[1] if uniform else None,
            "has_clients": df_clientes is not None and not getattr(df_clientes, "empty", True),
            "has_client_demand": df_clientes is not None and "Demanda" in df_clientes.columns and bool((df_clientes["Demanda"] > 0).any()),
            "number_of_vehicles": int(answers.get("Qual é o número de veiculos?", 0)),
            "number_of_clients": int(answers.get("Quantos clientes existem?", 0)),
            "service_time_per_delivery_minutes": service_time,
            "average_vehicle_speed_kmh": speed,
            "max_time_per_vehicle_minutes": max_time,
            "allow_multi_trip": allow_multi,
            "has_min_usage_time": has_min_usage,
            "min_usage_time_minutes": min_usage if has_min_usage else 0.0,
            "minimum_vehicle_usage_minutes": min_usage if has_min_usage else 0.0,
            "minimum_utilization_time_minutes": min_usage if has_min_usage else 0.0,
            "must_use_all_vehicles": must_use_all,
            "fixed_cost_type": fixed_type,
            "fixed_cost_mode": fixed_mode,
            "vehicle_start_time_minutes": operation_start,
            "vehicle_end_time_minutes": operation_end,
            "operational_time_definition": "SUM_TRIP_ELAPSED_TIME_INCLUDING_TRAVEL_WAIT_SERVICE",
            "translator_contract_version": self.VERSION,
        }
        problem_data = {
            "problem_description": description,
            "raw_data_summary": {
                "df_clientes_cols": list(df_clientes.columns) if df_clientes is not None else [],
                "df_veiculos_capacidade_cols": list(df_veiculos_capacidade.columns) if df_veiculos_capacidade is not None else [],
                "df_veiculos_custo_km_cols": list(df_veiculos_custo_km.columns) if df_veiculos_custo_km is not None else [],
                "df_janela_tempo_cliente_cols": list(df_janela_tempo_cliente.columns) if df_janela_tempo_cliente is not None else [],
            },
        }
        self.shared_memory.data["operational_contract"] = dict(description)
        return json.dumps(problem_data, indent=4, cls=CustomJsonEncoder)
````

### **CLASSIFIERAGENT — SELEÇÃO DA ESTRATÉGIA INICIAL**

 Este agente observa as características do pedido e ajuda a escolher o caminho inicial de resolução.





Tabela com células delimitadas: Responsabilidade / Explicação



````python
class ClassifierAgent:
    def __init__(self, shared_memory):
        self.shared_memory = shared_memory
        # === Definição dos Meta-atributos do Agente Classificador ===

        # Este agente é responsável por analisar as características do problema de roteirização
        # para determinar seu tipo específico (e.g., CVRP, VRPTW).
        self.role = "Agente Classificador de Problemas: Analisa o JSON traduzido para determinar o tipo específico do problema de roteirização."

        # Objetivo (Goal): classificar o problema com precisão para que o MathAgent possa selecionar a abordagem de otimização mais adequada.
        self.goal = "Classificar o problema de roteirização com base nas suas características presentes no JSON de entrada, distinguindo entre tipos como CVRP e VRPTW."

        # Conhecimento Especializado (Backstory): Ele compreende os atributos que definem diferentes tipos de problemas de roteirização,
        # como a presença de janelas de tempo ou capacidades de veículos.
        self.backstory = "Especialista em taxonomia de problemas de roteirização de veículos. Conhece as regras e heurísticas para identificar se um problema é um CVRP, VRPTW ou outra variação, baseado em seus atributos chave."

        # Ferramentas (Tools): Ele utiliza a biblioteca 'json' para carregar e interpretar o JSON de entrada e
        # opera com lógica condicional para realizar a classificação.
        self.tools = [
            "json.loads para desserialização de JSON",
            "Lógica condicional (if/elif/else) para análise de atributos do problema"
        ]

        # Memória (Memory): Sua memória é de curto prazo, focando apenas no JSON do problema atual que está sendo classificado.
        self.memory = None # Este agente não mantém um estado de memória próprio além da execução da função.

        # Restrições (Constraints): Ele depende de um JSON de entrada bem formatado e com os atributos necessários para a classificação.
        # Não corrige erros nos dados de entrada; apenas os analisa.
        self.constraints = [
            "Depende da saída estruturada e traduzida do TranslatorAgent.",
            "Não realiza validação de dados; apenas interpreta os atributos presentes no JSON.",
            "A classificação é baseada em regras predefinidas e pode ser limitada para tipos de problemas não contemplados."
        ]

        # Planejamento (Planning): Ele segue um plano sequencial: carregar o JSON, extrair a descrição do problema
        # e aplicar uma série de verificações condicionais para determinar o tipo.
        self.planning = "Executa um plano de análise hierárquica: carregar o JSON de entrada, extrair os atributos da descrição do problema e aplicar regras condicionais para identificar o tipo de problema de roteirização."

        # Ações (Actions): Ele desserializa o JSON, acessa atributos específicos do dicionário
        # 'problem_description' e define a string do tipo de problema.
        self.actions = [
            "Desserializar a string JSON de entrada.",
            "Acessar os atributos dentro da chave 'problem_description'.",
            "Aplicar lógica condicional para classificar o problema.",
            "Retornar a string resultante da classificação."
        ]

        # Avaliação (Evaluation): Seu sucesso é medido pela atribuição de um tipo de problema válido e reconhecido
        # com base nos atributos fornecidos no JSON, sem resultar em 'não classificado' quando dados suficientes estão presentes.
        self.evaluation = "O sucesso é medido pela capacidade de atribuir um tipo de problema de roteirização válido e específico, com base nas características presentes no JSON, evitando classificações genéricas ou 'não classificado' para problemas com atributos claros."

        # === Fim da Definição dos Meta-atributos ===

    def classify_problem(self, translated_problem_json):
        # Carrega a string JSON em um dicionário Python para fácil acesso aos dados.
        problem_data = json.loads(translated_problem_json)
        # Extrai a seção 'problem_description' que contém os atributos para a classificação.
        description = problem_data.get("problem_description", {})

        # Extrai atributos relevantes para a classificação
        objective = description.get("objective", "").lower()
        has_time_windows = description.get("has_time_windows", False)
        has_vehicle_capacity = description.get("has_vehicle_capacity", False)
        has_client_demand = description.get("has_client_demand", False)
        has_clients = description.get("has_clients", False)
        has_vehicles = description.get("has_vehicles", False)

        # Inicializa a variável para armazenar o tipo de problema classificado.
        problem_type = "Tipo de problema de roteirização não classificado."

        # -- Lógica de classificação revisada ---
        # A ordem das verificações é crucial, priorizando as classificações mais específicas e a intenção do usuário.

        # 1. Prioriza o objetivo 'Multiobjetivo' se explicitamente selecionado pelo usuário
        if objective == "multiobjetivo":
            problem_type = "Multiobjetivo (Problema com Múltiplos Objetivos Definidos pelo Usuário)"
        # 2. CVRPTW (Capacitated Vehicle Routing Problem with Time Windows)
        # Se tem janelas de tempo, capacidade de veículos e demanda dos clientes.
        elif has_time_windows and has_vehicle_capacity and has_client_demand:
            problem_type = "CVRPTW (Capacitated Vehicle Routing Problem with Time Windows)"
        # 3. VRPTW (Vehicle Routing Problem with Time Windows)
        # Se tem janelas de tempo, mas não necessariamente capacidade ou demanda (ou já foi tratado como CVRPTW)
        elif has_time_windows:
            problem_type = "VRPTW (Vehicle Routing Problem with Time Windows)"
        # 4. CVRP (Capacitated Vehicle Routing Problem)
        # Se tem capacidade e demanda dos clientes, mas sem janelas de tempo
        elif has_vehicle_capacity and has_client_demand:
            problem_type = "CVRP (Capacitated Vehicle Routing Problem)"
        # 5. VRP Básico
        # Se tem clientes e veículos, mas sem as restrições mais complexas acima
        elif has_clients and has_vehicles:
            problem_type = "VRP Básico (Vehicle Routing Problem)"
        # Se nenhuma das condições acima for atendida, mantém o default "não classificado."

        # Imprime o resultado da classificação para o usuário, tornando o processo transparente.
        print(f"\nCLASSIFICADOR: O problema de roteirização foi classificado como: {problem_type}")

        # Retorna o tipo de problema classificado, que será usado por outros agentes (e.g., MathAgent).
        return problem_type
````

### **MATHAGENT — MODELAGEM DAS REGRAS**

Este agente representa as decisões permitidas, as restrições da operação e o objetivo que será minimizado.





Tabela com células delimitadas: Responsabilidade / Explicação


````python
class MathAgent:
    """
    Agente Matemático Decisor e Guardião da Função Objetivo Canônica.

    Responsabilidades:
    1. Selecionar a estratégia de otimização conforme o porte do problema.
    2. Interpretar os objetivos ativos e seus pesos.
    3. Construir uma função objetivo multiobjetivo fixa e certificável.
    4. Definir escalas determinísticas de normalização.
    5. Armazenar o objective_context na SharedMemory.
    6. Fornecer uma função única para avaliar soluções de qualquer solver.

    IMPORTANTE:
    - Este agente não resolve o VRP.
    - Este agente não altera restrições operacionais.
    - A função objetivo canônica é congelada antes da busca e não pode
      depender das soluções encontradas por GA, OR-Tools ou outro solver.
    """

    VERSION = "MATH_AGENT_CANONICAL_OBJECTIVE_2_OPERATIONAL_CONTRACT"

    VALID_OBJECTIVES = ("COST", "DISTANCE", "TIME", "VEHICLES")

    def __init__(self, shared_memory):
        self.shared_memory = shared_memory
        self.role = (
            "Agente Matemático Decisor: seleciona estratégias de otimização "
            "e define a função objetivo canônica compartilhada pelos solvers."
        )
        self.goal = (
            "Maximizar a qualidade da solução, manter consistência matemática "
            "entre os solvers e permitir posterior certificação de optimalidade."
        )
        self.backstory = (
            "Especialista em pesquisa operacional, seleção de solvers, "
            "otimização multiobjetivo e formulações matemáticas certificáveis."
        )

    @staticmethod
    def _safe_float(value, default=0.0):
        try:
            if value is None:
                return default
            if isinstance(value, str):
                value = value.strip().replace(",", ".")
                if not value:
                    return default
            return float(value)
        except (TypeError, ValueError):
            return default

    @staticmethod
    def _safe_int(value, default=0):
        try:
            return int(value)
        except (TypeError, ValueError):
            try:
                return int(float(value))
            except Exception:
                return default

    def _parse_problem_data(self, translated_problem_json):
        import json
        if isinstance(translated_problem_json, str):
            problem_data = json.loads(translated_problem_json)
        elif isinstance(translated_problem_json, dict):
            problem_data = translated_problem_json
        else:
            raise TypeError(
                "MATH_AGENT: translated_problem_json deve ser str JSON ou dict."
            )
        description = problem_data.get("problem_description", {}) or {}
        return problem_data, description

    def _infer_active_objectives(self, description):
        objective_text = str(description.get("objective", "") or "").lower()
        multi_cfg = description.get("multiobjective_config", {}) or {}
        active = [
            str(x).upper().strip()
            for x in (multi_cfg.get("objectives", []) or [])
            if str(x).strip()
        ]
        active = [x for x in active if x in self.VALID_OBJECTIVES]

        if active:
            return active
        if "custo" in objective_text:
            return ["COST"]
        if "tempo" in objective_text:
            return ["TIME"]
        if "veículo" in objective_text or "veiculo" in objective_text:
            return ["VEHICLES"]
        return ["DISTANCE"]

    def _normalize_weights(self, active_objectives, raw_weights, is_multi):
        if not active_objectives:
            raise ValueError("MATH_AGENT: nenhum objetivo ativo foi identificado.")

        if len(active_objectives) == 1 and not is_multi:
            return {active_objectives[0]: 1.0}

        raw_weights = raw_weights or {}
        normalized_input = {}

        for obj in active_objectives:
            value = self._safe_float(
                raw_weights.get(obj, raw_weights.get(obj.lower(), 0.0)),
                0.0
            )
            if value > 1.0:
                value /= 100.0
            normalized_input[obj] = value

        total = sum(normalized_input.values())

        if total <= 0:
            raise ValueError(
                "MATH_AGENT: configuração multiobjetivo sem pesos válidos."
            )

        if abs(total - 1.0) > 1e-6 and not (0.999 <= total <= 1.001):
            raise ValueError(
                "MATH_AGENT: os pesos multiobjetivo devem somar 100% "
                f"(ou 1.0). Soma recebida: {total:.6f}."
            )

        return {
            obj: value / total
            for obj, value in normalized_input.items()
        }

    def _extract_number_of_vehicles(self, description):
        mem = getattr(self.shared_memory, "data", {}) or {}
        ans = mem.get("user_answers", {}) or {}

        candidates = [
            description.get("number_of_vehicles"),
            description.get("numero_veiculos"),
            description.get("num_vehicles"),
            ans.get("Qual é o número de veiculos?"),
            ans.get("Qual é o número de veículos?"),
            ans.get("numero_veiculos"),
            ans.get("num_veiculos"),
        ]

        for value in candidates:
            n = self._safe_int(value, 0)
            if n > 0:
                return n
        return 1

    def _extract_available_time(self, description):
        mem = getattr(self.shared_memory, "data", {}) or {}
        ans = mem.get("user_answers", {}) or {}

        candidates = [
            description.get("available_time_minutes"),
            description.get("tempo_disponivel_por_veiculo"),
            description.get("vehicle_available_time_minutes"),
            ans.get("Tempo disponível por veículo por dia (minutos)?"),
            ans.get("tempo_disponivel_por_veiculo"),
            ans.get("available_time_minutes"),
        ]

        for value in candidates:
            t = self._safe_float(value, 0.0)
            if t > 0:
                return t
        return 1440.0

    def _compute_distance_scale(self, description):
        """
        Escala determinística para DISTANCE.

        Quando há coordenadas na SharedMemory, usa 2 * soma das distâncias
        depósito-cliente. Caso contrário, usa fallback baseado no número de
        clientes. Esta é uma escala de normalização, não um bound ótimo.
        """
        import math

        mem = getattr(self.shared_memory, "data", {}) or {}
        df_clientes = mem.get("df_clientes")

        if df_clientes is not None and not getattr(df_clientes, "empty", True):
            try:
                id_col = "Cliente" if "Cliente" in df_clientes.columns else "cliente"
                lat_col = "latitude"
                lon_col = "longitude"

                if (
                    id_col in df_clientes.columns
                    and lat_col in df_clientes.columns
                    and lon_col in df_clientes.columns
                ):
                    deposito_rows = df_clientes[
                        df_clientes[id_col].astype(int) == 0
                    ]

                    if not deposito_rows.empty:
                        depot_lat = float(deposito_rows.iloc[0][lat_col])
                        depot_lon = float(deposito_rows.iloc[0][lon_col])

                        def haversine(lat1, lon1, lat2, lon2):
                            R = 6371.0
                            p1 = math.radians(lat1)
                            p2 = math.radians(lat2)
                            dphi = math.radians(lat2 - lat1)
                            dlambda = math.radians(lon2 - lon1)
                            a = (
                                math.sin(dphi / 2.0) ** 2
                                + math.cos(p1)
                                * math.cos(p2)
                                * math.sin(dlambda / 2.0) ** 2
                            )
                            return 2.0 * R * math.asin(math.sqrt(max(0.0, a)))

                        sum_round_trips = 0.0

                        for _, row in df_clientes.iterrows():
                            cid = self._safe_int(row[id_col], -1)
                            if cid == 0:
                                continue

                            lat = self._safe_float(row[lat_col], None)
                            lon = self._safe_float(row[lon_col], None)
                            if lat is None or lon is None:
                                continue

                            sum_round_trips += 2.0 * haversine(
                                depot_lat, depot_lon, lat, lon
                            )

                        if sum_round_trips > 0:
                            return max(sum_round_trips, 1.0)
            except Exception:
                pass

        n_clients = self._safe_int(
            description.get("number_of_clients", 0), 0
        )
        return max(1.0, float(max(1, n_clients)) * 20.0)

    def _compute_cost_scale(self, description, distance_scale):
        """
        Escala determinística para COST:
            soma dos custos fixos disponíveis
            + distance_scale * maior custo/km conhecido.
        """
        mem = getattr(self.shared_memory, "data", {}) or {}
        candidate_frames = [
            mem.get("df_veiculos_custo_km"),
            mem.get("df_veiculos_custos"),
            mem.get("df_custos_veiculos"),
        ]

        fixed_sum = 0.0
        max_cost_km = 0.0

        for df in candidate_frames:
            if df is None or getattr(df, "empty", True):
                continue

            try:
                columns_lower = {str(c).lower(): c for c in df.columns}

                cost_km_col = None
                for key in (
                    "custo_por_km", "custo_km",
                    "cost_per_km", "custo por km"
                ):
                    if key in columns_lower:
                        cost_km_col = columns_lower[key]
                        break

                fixed_col = None
                for key in (
                    "custo_diario", "custo_fixo",
                    "fixed_cost", "custo diário"
                ):
                    if key in columns_lower:
                        fixed_col = columns_lower[key]
                        break

                if cost_km_col is not None:
                    values = [
                        self._safe_float(x, 0.0)
                        for x in df[cost_km_col]
                    ]
                    if values:
                        max_cost_km = max(max_cost_km, max(values))

                if fixed_col is not None:
                    fixed_sum = max(
                        fixed_sum,
                        sum(self._safe_float(x, 0.0) for x in df[fixed_col])
                    )

            except Exception:
                continue

        max_cost_km = max(
            max_cost_km,
            self._safe_float(description.get("max_cost_per_km", 0.0), 0.0)
        )

        fixed_sum = max(
            fixed_sum,
            self._safe_float(
                description.get("total_fixed_cost_reference", 0.0), 0.0
            )
        )

        if max_cost_km <= 0:
            max_cost_km = 1.0

        return max(
            fixed_sum + distance_scale * max_cost_km,
            1.0
        )

    def _compute_time_scale(self, description):
        return max(
            1.0,
            float(self._extract_number_of_vehicles(description))
            * float(self._extract_available_time(description))
        )

    def _compute_vehicle_scale(self, description):
        return max(
            1.0,
            float(self._extract_number_of_vehicles(description))
        )

    def build_objective_context(self, translated_problem_json):
        """
        Constrói e congela a função objetivo canônica.

        Fórmula:
            score = Σ coefficient[obj] * metric[obj]

        onde:
            coefficient[obj] = weight[obj] / scale[obj]

        As escalas são fixadas antes da busca e não dependem de soluções
        encontradas pelos solvers.
        """
        _, description = self._parse_problem_data(
            translated_problem_json
        )

        objective_text = str(
            description.get("objective", "") or ""
        ).lower()

        multi_cfg = (
            description.get("multiobjective_config", {})
            or {}
        )

        active_objectives = self._infer_active_objectives(
            description
        )

        is_multi = (
            bool(multi_cfg.get("enabled", False))
            or len(active_objectives) > 1
            or "multiobjetivo" in objective_text
        )

        weights = self._normalize_weights(
            active_objectives,
            multi_cfg.get("weights", {}) or {},
            is_multi
        )

        requires_cost_data = "COST" in active_objectives
        cost_data_available = bool(
            description.get("cost_data_available", False)
        )

        if requires_cost_data and not cost_data_available:
            mem = getattr(self.shared_memory, "data", {}) or {}
            cost_data_available = any(
                df is not None and not getattr(df, "empty", True)
                for df in (
                    mem.get("df_veiculos_custo_km"),
                    mem.get("df_veiculos_custos"),
                    mem.get("df_custos_veiculos"),
                )
            )

        if requires_cost_data and not cost_data_available:
            raise ValueError(
                "MATH_AGENT: o objetivo inclui COST, mas não há dados "
                "de custo por veículo disponíveis."
            )

        distance_scale = self._compute_distance_scale(description)

        scales = {
            "DISTANCE": distance_scale,
            "COST": self._compute_cost_scale(
                description, distance_scale
            ),
            "TIME": self._compute_time_scale(description),
            "VEHICLES": self._compute_vehicle_scale(description),
        }

        coefficients = {
            obj: weights[obj] / scales[obj]
            for obj in active_objectives
        }

        mem = getattr(self.shared_memory, "data", {}) or {}
        ans = mem.get("user_answers", {}) or {}

        fixed_cost_mode = (
            description.get("fixed_cost_mode")
            or description.get("custo_fixo_mode")
            or ans.get("fixed_cost_mode")
            or ans.get("custo_fixo_mode")
            or "UNKNOWN"
        )

        context = {
            "version": "CANONICAL_OBJECTIVE_2",
            "type": "WEIGHTED_NORMALIZED_SUM",
            "direction": "MINIMIZE",
            "active_objectives": list(active_objectives),
            "weights": dict(weights),
            "scales": dict(scales),
            "coefficients": dict(coefficients),
            "fixed_cost_mode": str(fixed_cost_mode),
            "number_of_vehicles": int(
                self._extract_number_of_vehicles(description)
            ),
            "normalization_policy": "DETERMINISTIC_PROBLEM_DATA_SCALES",
            "metric_contract": {
                "DISTANCE": "SUM_ROUTE_HAVERSINE_KM",
                "TIME": "SUM_TRIP_ELAPSED_TIME_INCLUDING_TRAVEL_WAIT_SERVICE",
                "COST": "VARIABLE_DISTANCE_COST_PLUS_APPLICABLE_FIXED_COST",
                "VEHICLES": "COUNT_DISTINCT_PHYSICAL_VEHICLES_USED",
            },
            "operational_contract_version": description.get("translator_contract_version", "LEGACY"),
            "frozen": True,
            "certification_ready": True,
        }

        self.validate_objective_context(context)

        try:
            self.shared_memory.data["objective_context"] = context
        except Exception:
            pass

        return context

    def validate_objective_context(self, context):
        if not isinstance(context, dict):
            raise TypeError(
                "MATH_AGENT: objective_context deve ser um dicionário."
            )

        if not context.get("frozen", False):
            raise ValueError(
                "MATH_AGENT: objective_context precisa estar congelado."
            )

        active = context.get("active_objectives", []) or []
        weights = context.get("weights", {}) or {}
        scales = context.get("scales", {}) or {}
        coefficients = context.get("coefficients", {}) or {}

        if not active:
            raise ValueError(
                "MATH_AGENT: objective_context sem objetivos ativos."
            )

        total_weight = 0.0

        for obj in active:
            if obj not in self.VALID_OBJECTIVES:
                raise ValueError(
                    f"MATH_AGENT: objetivo inválido no contexto: {obj}."
                )

            weight = self._safe_float(weights.get(obj), -1.0)
            scale = self._safe_float(scales.get(obj), 0.0)
            coefficient = self._safe_float(
                coefficients.get(obj), -1.0
            )

            if weight < 0:
                raise ValueError(
                    f"MATH_AGENT: peso inválido para {obj}."
                )
            if scale <= 0:
                raise ValueError(
                    f"MATH_AGENT: escala deve ser positiva para {obj}."
                )

            expected = weight / scale
            if abs(coefficient - expected) > 1e-12:
                raise ValueError(
                    f"MATH_AGENT: coeficiente inconsistente para {obj}."
                )

            total_weight += weight

        if abs(total_weight - 1.0) > 1e-6:
            raise ValueError(
                "MATH_AGENT: pesos do contexto canônico devem somar 1.0."
            )

        return True

    def evaluate_canonical_objective(
        self,
        metrics,
        objective_context=None
    ):
        """
        Avalia uma solução com a função objetivo canônica.

        metrics pode conter:
            COST, DISTANCE, TIME, VEHICLES
        """
        if objective_context is None:
            objective_context = (
                getattr(self.shared_memory, "data", {})
                .get("objective_context")
            )

        if not objective_context:
            raise ValueError(
                "MATH_AGENT: objective_context não foi construído."
            )

        self.validate_objective_context(objective_context)

        score = 0.0

        for obj in objective_context["active_objectives"]:
            metric_value = self._safe_float(
                metrics.get(obj, metrics.get(obj.lower(), 0.0)),
                0.0
            )

            score += (
                objective_context["coefficients"][obj]
                * metric_value
            )

        return float(score)

    def decide_optimization_approach(
        self,
        problem_classification,
        translated_problem_json,
        force_change=False
    ):
        """
        Seleciona a estratégia de otimização considerando o porte e a
        complexidade do problema.

        Perfis:
        - INTENSIVE: até 20 clientes -> portfólio amplo.
        - BALANCED: 21 a 50 clientes -> portfólio moderado.
        - LARGE_SCALE: mais de 50 clientes -> OR-Tools como construtor
          principal + GA curto de refinamento quando multiobjetivo.

        Antes da decisão, constrói e congela a função objetivo canônica.
        """
        _, description = self._parse_problem_data(
            translated_problem_json
        )

        objective = str(
            description.get("objective", "") or ""
        ).lower()

        number_of_clients = self._safe_int(
            description.get("number_of_clients", 0),
            0
        )

        multi_cfg = (
            description.get("multiobjective_config", {})
            or {}
        )

        active_objectives = self._infer_active_objectives(
            description
        )

        is_multi = (
            bool(multi_cfg.get("enabled", False))
            or len(active_objectives) > 1
            or "multiobjetivo" in objective
        )

        objective_context = self.build_objective_context(
            translated_problem_json
        )

        if number_of_clients <= 20:
            profile = "INTENSIVE"
            standard_approach = "PORTFOLIO_OR_TOOLS_GA_INTENSIVE"

        elif number_of_clients <= 50:
            profile = "BALANCED"
            standard_approach = "PORTFOLIO_OR_TOOLS_GA_BALANCED"

        else:
            profile = "LARGE_SCALE"

            if is_multi:
                standard_approach = "PORTFOLIO_OR_TOOLS_GA_LARGE_SCALE"
            else:
                standard_approach = "Solver (OR-Tools)"

        try:
            self.shared_memory.data["optimization_profile"] = {
                "profile": profile,
                "number_of_clients": number_of_clients,
                "is_multiobjective": is_multi,
                "strategy": standard_approach,
                "active_objectives": active_objectives,
                "objective_context_version": objective_context["version"],
            }
        except Exception:
            pass

        if force_change:
            if "PORTFOLIO" in standard_approach:
                if number_of_clients <= 50:
                    return "Algoritmos Genéticos"
                return "Solver (OR-Tools)"

            if standard_approach == "Algoritmos Genéticos":
                return "Solver (OR-Tools)"

            return "Algoritmos Genéticos"

        return standard_approach
````

### **ORTOOLSSOLVERAGENT — BUSCA DE PLANOS COM OR-TOOLS**

Este é um dos especialistas que o coordenador pode usar para montar viagens. Ele utiliza ferramentas de roteamento do OR-Tools para procurar uma boa organização.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class ORToolsSolverAgent:

    VERSION = "ORTOOLS_SOLVER_V11_1_DEADLINE_CONTRACT_TIME_PRECISION_FIX"

    def __init__(self, shared_memory):

        self.shared_memory = shared_memory


        # META-ATRIBUTOS


        self.role = (
            "Arquiteto de Otimização Combinatória (OR-Tools): "
            "Especialista sênior em pesquisa operacional e modelagem "
            "de problemas de roteirização."
        )

        self.goal = (
            "Resolver variantes de VRP com frota heterogênea, "
            "capacidade, janelas de tempo, jornada máxima, "
            "utilização mínima, custos fixos, custos variáveis "
            "e suporte a múltiplas viagens."
        )

        self.backstory = (
            "Especialista em traduzir requisitos logísticos "
            "em modelos matemáticos rigorosos usando OR-Tools."
        )

        self.tools = [
            "pywrapcp.RoutingIndexManager",
            "pywrapcp.RoutingModel",
            "routing_enums_pb2",
            "Fórmula de Haversine",
            "Guided Local Search",
            "Busca Local OR-Tools"
        ]

        self.memory = []

        self.constraints = [
            "Capacidade por veículo",
            "Frota heterogênea",
            "Janelas de tempo",
            "Jornada máxima",
            "Tempo mínimo de utilização",
            "Uso obrigatório ou opcional dos veículos",
            "Custos fixos",
            "Custos variáveis",
            "Multi-Trip",
            "Retorno ao depósito",
            "Escalonamento inteiro"
        ]

        self.planning = (
            "1. Validar entrada -> "
            "2. Criar Data Model -> "
            "3. Criar matriz de distância -> "
            "4. Criar callbacks -> "
            "5. Criar dimensões -> "
            "6. Aplicar restrições -> "
            "7. Configurar custos -> "
            "8. Resolver -> "
            "9. Decodificar -> "
            "10. Auditar solução"
        )

        self.actions = [
            "solve_problem",
            "_create_data_model",
            "_validate_data_model",
            "_build_routing_model",
            "_decode_solution",
            "_audit_solution"
        ]

        self.evaluation = (
            "A solução somente é considerada válida quando "
            "o solver encontra uma solução e o auditor confirma "
            "todas as restrições."
        )

        self.manager = None
        self.routing = None
        self.solution = None
        self.data = None

        self.errors = []
        self.warnings = []

        self.SCALING_FACTOR = 1000

        # Escala exclusiva para grandezas temporais.
        # Carga/capacidade continuam usando SCALING_FACTOR = 1000.
        self.TIME_SCALING_FACTOR = 1_000_000

        # Número máximo de viagens por veículo físico.
        # Não significa que todas serão utilizadas.
        self.DEFAULT_MAX_TRIPS = None  # sem limite arbitrário; derivado do número de clientes


    # ESCALAMENTO


    def _scale_value(self, value):
        return int(round(float(value) * self.SCALING_FACTOR))

    def _unscale_value(self, value):
        return float(value) / self.SCALING_FACTOR

    def _scale_time_duration(self, value):
        # Duração conservadora: nunca menor que o valor físico contínuo.
        scaled = float(value) * self.TIME_SCALING_FACTOR
        return int(math.ceil(scaled - 1e-12))

    def _scale_time_lower_bound(self, value):
        # Limite inferior conservador.
        scaled = float(value) * self.TIME_SCALING_FACTOR
        return int(math.ceil(scaled - 1e-12))

    def _scale_time_upper_bound(self, value):
        # Limite superior conservador.
        scaled = float(value) * self.TIME_SCALING_FACTOR
        return int(math.floor(scaled + 1e-12))

    def _unscale_time(self, value):
        return float(value) / self.TIME_SCALING_FACTOR


    # PARSING DE TEMPO


    def _parse_to_min(self, value):

        if value is None:
            return 0.0

        if isinstance(value, (int, float)):
            return float(value)

        if isinstance(value, time):
            return (
                value.hour * 60
                + value.minute
                + value.second / 60.0
            )

        if isinstance(value, datetime):
            return (
                value.hour * 60
                + value.minute
                + value.second / 60.0
            )

        if isinstance(value, str):

            value = value.strip()

            for fmt in (
                "%H:%M:%S",
                "%H:%M"
            ):
                try:
                    dt = datetime.strptime(value, fmt)

                    return (
                        dt.hour * 60
                        + dt.minute
                        + dt.second / 60.0
                    )

                except ValueError:
                    continue

        raise ValueError(
            f"Formato de tempo inválido: {value}"
        )


    # HAVERSINE


    def _haversine_distance(
        self,
        lat1,
        lon1,
        lat2,
        lon2
    ):

        R = 6371.0

        lat1_rad = math.radians(float(lat1))
        lon1_rad = math.radians(float(lon1))

        lat2_rad = math.radians(float(lat2))
        lon2_rad = math.radians(float(lon2))

        dlon = lon2_rad - lon1_rad
        dlat = lat2_rad - lat1_rad

        a = (
            math.sin(dlat / 2.0) ** 2
            +
            math.cos(lat1_rad)
            *
            math.cos(lat2_rad)
            *
            math.sin(dlon / 2.0) ** 2
        )

        a = min(1.0, max(0.0, a))

        c = 2.0 * math.atan2(
            math.sqrt(a),
            math.sqrt(1.0 - a)
        )

        return R * c


    # BOOLEAN ROBUSTO


    def _parse_bool(self, value, default=False):

        if value is None:
            return default

        if isinstance(value, bool):
            return value

        if isinstance(value, (int, float)):
            return bool(value)

        if isinstance(value, str):

            value = value.strip().lower()

            if value in (
                "true",
                "1",
                "yes",
                "sim",
                "s"
            ):
                return True

            if value in (
                "false",
                "0",
                "no",
                "não",
                "nao",
                "n"
            ):
                return False

        return default


    # NORMALIZAÇÃO DO MODO DE CUSTO FIXO


    def _normalize_fixed_cost_mode(self, value):

        if value is None:
            return "WHEN_USED"

        value = str(value).strip().upper()

        aliases = {
            "WHEN_USED": "WHEN_USED",
            "USED": "WHEN_USED",
            "UTILIZADO": "WHEN_USED",
            "QUANDO_USADO": "WHEN_USED",

            "ALWAYS": "ALWAYS",
            "EVERY_VEHICLE": "ALWAYS",
            "ALL": "ALWAYS",
            "SEMPRE": "ALWAYS"
        }

        return aliases.get(value, "WHEN_USED")


    # CRIAÇÃO DO DATA MODEL


    def _create_data_model(self, translated_json):

        try:

            problem_data = json.loads(translated_json)

            desc = problem_data.get(
                "problem_description",
                {}
            )


            # DATAFRAMES


            df_clientes = self.shared_memory.data.get(
                "df_clientes"
            )

            df_cap = self.shared_memory.data.get(
                "df_veiculos_capacidade"
            )

            df_cost = self.shared_memory.data.get(
                "df_veiculos_custo_km"
            )

            df_janela = self.shared_memory.data.get(
                "df_janela_tempo_cliente"
            )

            if df_clientes is None:
                raise ValueError(
                    "df_clientes não encontrado no shared_memory."
                )

            if df_cap is None:
                raise ValueError(
                    "df_veiculos_capacidade não encontrado."
                )


            # CLIENTES


            df_clientes = df_clientes.copy()
            df_cap = df_cap.copy()

            if df_cost is not None:
                df_cost = df_cost.copy()

            multiobjective_config = desc.get(
                "multiobjective_config",
                {"enabled": False, "objectives": [], "weights": {}}
            ) or {}
            active_objectives = [str(x).upper() for x in multiobjective_config.get("objectives", [])]
            objective_text = str(desc.get("objective", "")).lower()
            requires_cost_data = "COST" in active_objectives or (
                not active_objectives and "custo" in objective_text
            )

            if requires_cost_data and (df_cost is None or df_cost.empty):
                raise ValueError(
                    "O objetivo inclui CUSTO, mas df_veiculos_custo_km não foi fornecido."
                )

            if requires_cost_data:
                if "custo_por_km" not in df_cost.columns or "custo_diario" not in df_cost.columns:
                    raise ValueError(
                        "A planilha de custos é obrigatória e deve conter custo_por_km e custo_diario."
                    )

            if df_janela is not None:
                df_janela = df_janela.copy()

            required_client_columns = [
                "Cliente",
                "latitude",
                "longitude",
                "Demanda"
            ]

            for col in required_client_columns:

                if col not in df_clientes.columns:
                    raise ValueError(
                        f"Coluna obrigatória ausente em "
                        f"df_clientes: {col}"
                    )

            if "capacidade" not in df_cap.columns:
                raise ValueError(
                    "Coluna 'capacidade' ausente "
                    "em df_veiculos_capacidade."
                )

            df_clientes["Cliente"] = (
                df_clientes["Cliente"].astype(int)
            )

            df_clientes["latitude"] = (
                df_clientes["latitude"].astype(float)
            )

            df_clientes["longitude"] = (
                df_clientes["longitude"].astype(float)
            )

            df_clientes["Demanda"] = (
                df_clientes["Demanda"].astype(float)
            )

            df_cap["capacidade"] = (
                df_cap["capacidade"].astype(float)
            )


            # DEPÓSITO


            depot_rows = df_clientes[
                df_clientes["Cliente"] == 0
            ]

            if depot_rows.empty:
                raise ValueError(
                    "Cliente 0 (depósito) não encontrado."
                )

            depot_row = depot_rows.iloc[0]

            customers = (
                df_clientes[
                    df_clientes["Cliente"] != 0
                ]
                .sort_values("Cliente")
                .reset_index(drop=True)
            )


            # MAPEAMENTO CORRETO DOS CLIENTES


            node_to_client_id = {
                0: 0
            }

            client_id_to_node = {
                0: 0
            }

            locations = [
                {
                    "client_id": 0,
                    "lat": float(depot_row["latitude"]),
                    "lon": float(depot_row["longitude"])
                }
            ]

            demands = [0.0]

            for node, (_, row) in enumerate(
                customers.iterrows(),
                start=1
            ):

                client_id = int(row["Cliente"])

                node_to_client_id[node] = client_id
                client_id_to_node[client_id] = node

                locations.append(
                    {
                        "client_id": client_id,
                        "lat": float(row["latitude"]),
                        "lon": float(row["longitude"])
                    }
                )

                demands.append(
                    float(row["Demanda"])
                )


            # MATRIZ DE DISTÂNCIA


            dist_matrix = {}

            for i in range(len(locations)):

                for j in range(len(locations)):

                    dist_matrix[(i, j)] = (
                        self._haversine_distance(
                            locations[i]["lat"],
                            locations[i]["lon"],
                            locations[j]["lat"],
                            locations[j]["lon"]
                        )
                    )


            # JANELAS DE TEMPO — contrato canônico do Translator

            time_windows = {}
            window_mode = str(desc.get("time_window_mode", "NONE") or "NONE").upper()
            has_time_windows = self._parse_bool(desc.get("has_time_windows"), False)

            if has_time_windows and window_mode == "UNIFORM":
                start_tw = float(desc.get("uniform_time_window_start_minutes"))
                end_tw = float(desc.get("uniform_time_window_end_minutes"))
                if end_tw <= start_tw:
                    raise ValueError("Janela uniforme inválida no contrato operacional.")
                for client_id in client_id_to_node:
                    if client_id != 0:
                        time_windows[client_id] = {"start": start_tw, "end": end_tw}

            elif has_time_windows and window_mode == "PER_CLIENT":
                if df_janela is None or df_janela.empty:
                    raise ValueError("Contrato indica janelas por cliente, mas df_janela_tempo_cliente está ausente.")
                required_window_columns = ["cliente", "inicio_janela", "fim_janela"]
                for col in required_window_columns:
                    if col not in df_janela.columns:
                        raise ValueError(f"Coluna '{col}' ausente em df_janela_tempo_cliente.")
                for _, row in df_janela.iterrows():
                    client_id = int(row["cliente"])
                    if client_id not in client_id_to_node or client_id == 0:
                        continue
                    start_tw = self._parse_to_min(row["inicio_janela"])
                    end_tw = self._parse_to_min(row["fim_janela"])
                    if end_tw <= start_tw:
                        raise ValueError(f"Janela inválida para cliente {client_id}: {start_tw} >= {end_tw}")
                    time_windows[client_id] = {"start": start_tw, "end": end_tw}

            # Horizonte da jornada vem exclusivamente do contrato operacional.
            depot_start = float(desc.get("vehicle_start_time_minutes", 0.0))
            depot_end = float(desc.get("vehicle_end_time_minutes", depot_start + float(desc.get("max_time_per_vehicle_minutes", 480.0))))
            if depot_end <= depot_start:
                raise ValueError("Horizonte operacional inválido: vehicle_end_time_minutes deve ser maior que vehicle_start_time_minutes.")
            time_windows[0] = {"start": depot_start, "end": depot_end}


            # VEÍCULOS


            capacities = [
                float(x)
                for x in df_cap["capacidade"].tolist()
            ]

            num_physical_vehicles = len(capacities)

            if num_physical_vehicles == 0:
                raise ValueError(
                    "Nenhum veículo foi encontrado."
                )


            # CUSTOS


            if df_cost is not None:

                if "custo_por_km" in df_cost.columns:

                    costs_km = [
                        float(x)
                        for x in df_cost[
                            "custo_por_km"
                        ].tolist()
                    ]

                else:

                    costs_km = [
                        0.0
                    ] * num_physical_vehicles

                if "custo_diario" in df_cost.columns:

                    fixed_costs = [
                        float(x)
                        for x in df_cost[
                            "custo_diario"
                        ].tolist()
                    ]

                else:

                    fixed_costs = [
                        0.0
                    ] * num_physical_vehicles

            else:

                costs_km = [
                    0.0
                ] * num_physical_vehicles

                fixed_costs = [
                    0.0
                ] * num_physical_vehicles


            # ALINHAMENTO ENTRE VEÍCULOS E DADOS DE CUSTO


            if len(costs_km) < num_physical_vehicles:

                costs_km.extend(
                    [0.0] *
                    (
                        num_physical_vehicles
                        - len(costs_km)
                    )
                )

            if len(fixed_costs) < num_physical_vehicles:

                fixed_costs.extend(
                    [0.0] *
                    (
                        num_physical_vehicles
                        - len(fixed_costs)
                    )
                )

            costs_km = costs_km[
                :num_physical_vehicles
            ]

            fixed_costs = fixed_costs[
                :num_physical_vehicles
            ]


            # PARÂMETROS OPERACIONAIS


            avg_speed = float(
                desc.get(
                    "average_vehicle_speed_kmh",
                    40.0
                )
            )

            if avg_speed <= 0:
                raise ValueError(
                    "average_vehicle_speed_kmh "
                    "deve ser > 0."
                )

            service_time = float(
                desc.get(
                    "service_time_per_delivery_minutes",
                    0
                )
            )

            max_time = float(
                desc.get(
                    "max_time_per_vehicle_minutes",
                    480
                )
            )

            minimum_usage = float(
                desc.get(
                    "min_usage_time_minutes",
                    desc.get(
                        "minimum_vehicle_usage_minutes",
                        desc.get(
                            "minimum_utilization_time_minutes",
                            0
                        )
                    )
                )
            )

            allow_multi_trip = self._parse_bool(
                desc.get(
                    "allow_multi_trip",
                    False
                )
            )

            must_use_all_vehicles = self._parse_bool(
                desc.get(
                    "must_use_all_vehicles",
                    desc.get(
                        "must_use_all",
                        False
                    )
                )
            )

            fixed_cost_mode = (
                self._normalize_fixed_cost_mode(
                    desc.get(
                        "fixed_cost_mode",
                        desc.get("fixed_cost_type", "WHEN_USED")
                    )
                )
            )

            return_to_depot = self._parse_bool(
                desc.get(
                    "return_to_depot",
                    True
                ),
                default=True
            )

            # Limite matematicamente seguro: uma viagem útil atende pelo menos
            # um cliente. Portanto nenhum veículo precisa de mais slots de viagem
            # do que o número total de clientes.
            safe_trip_bound = max(1, len(locations) - 1) if allow_multi_trip else 1
            max_trips_per_vehicle = int(
                desc.get(
                    "max_trips_per_vehicle",
                    safe_trip_bound
                )
            )
            if allow_multi_trip:
                max_trips_per_vehicle = min(max_trips_per_vehicle, safe_trip_bound)
            else:
                max_trips_per_vehicle = 1

            if max_trips_per_vehicle < 1:
                max_trips_per_vehicle = 1


            # MODELO FINAL


            data = {

                "locations": locations,

                "demands": demands,

                "dist_matrix": dist_matrix,

                "node_to_client_id": node_to_client_id,

                "client_id_to_node": client_id_to_node,

                "capacities": capacities,

                "costs_km": costs_km,

                "fixed_costs": fixed_costs,

                "num_physical_vehicles":
                    num_physical_vehicles,

                "avg_speed":
                    avg_speed,

                "service_time":
                    service_time,

                "max_time":
                    max_time,

                "minimum_usage":
                    minimum_usage,

                "allow_multi_trip":
                    allow_multi_trip,

                "must_use_all_vehicles":
                    must_use_all_vehicles,

                "fixed_cost_mode":
                    fixed_cost_mode,
                "objective": desc.get('objective','Minimizar distância'),
                "multiobjective_config": multiobjective_config,
                "requires_cost_data": requires_cost_data,
                "cost_data_available": (df_cost is not None and not df_cost.empty),

                "return_to_depot":
                    return_to_depot,

                "max_trips_per_vehicle":
                    max_trips_per_vehicle,

                "time_windows":
                    time_windows,

                "depot_start":
                    depot_start,

                "depot_end":
                    depot_end,

                "horizon_minutes":
                    max(
                        depot_end,
                        depot_start + max_time
                    )
            }

            return data

        except Exception as e:

            self.errors.append(
                f"Erro DataModel: {str(e)}"
            )

            return None


    # VALIDAÇÃO DO DATA MODEL


    def _validate_data_model(self):

        data = self.data

        if data is None:
            return False

        demands = data["demands"]
        capacities = data["capacities"]


        # DEMANDAS NEGATIVAS


        if any(
            demand < 0
            for demand in demands
        ):

            self.errors.append(
                "Existem demandas negativas."
            )

            return False


        # CAPACIDADES


        if any(
            capacity <= 0
            for capacity in capacities
        ):

            self.errors.append(
                "Todas as capacidades dos veículos "
                "devem ser maiores que zero."
            )

            return False


        # DEMANDA TOTAL


        total_demand = sum(demands)

        total_capacity = sum(capacities)

        if (
            not data["allow_multi_trip"]
            and total_demand > total_capacity
        ):

            self.errors.append(
                "Demanda total superior à capacidade "
                "total da frota."
            )

            return False


        # CLIENTE INDIVIDUAL


        max_capacity = max(capacities)

        impossible_clients = []

        for node, demand in enumerate(demands):

            if node == 0:
                continue

            if demand > max_capacity:

                impossible_clients.append(
                    data["node_to_client_id"][node]
                )

        if impossible_clients:

            self.errors.append(
                "Clientes com demanda superior à "
                "capacidade máxima de qualquer veículo: "
                f"{impossible_clients}"
            )

            return False


        # TEMPO


        if data["max_time"] <= 0:

            self.errors.append(
                "max_time_per_vehicle_minutes "
                "deve ser maior que zero."
            )

            return False

        if data["minimum_usage"] < 0:

            self.errors.append(
                "minimum_vehicle_usage_minutes "
                "não pode ser negativo."
            )

            return False

        if (
            data["minimum_usage"]
            > data["max_time"]
        ):

            self.errors.append(
                "Tempo mínimo de utilização "
                "superior ao tempo máximo "
                "de jornada."
            )

            return False


        # DEPOT WINDOW


        if (
            data["depot_end"]
            <= data["depot_start"]
        ):

            self.errors.append(
                "Janela do depósito inválida."
            )

            return False

        return True


    # CRIAÇÃO DO ROUTING MODEL


    def _get_objective_context(self):
        """
        Obtém a função objetivo canônica criada pelo MathAgent.

        O Solver OR-Tools não calcula pesos ou escalas próprios. Isso garante
        que OR-Tools, GA e o futuro certificador Branch-and-Price resolvam
        exatamente a mesma função matemática.
        """
        context = self.shared_memory.data.get("objective_context")

        if not isinstance(context, dict):
            raise ValueError(
                "OR_TOOLS: objective_context ausente. "
                "Execute o MathAgent antes dos solvers."
            )

        if not context.get("frozen", False):
            raise ValueError(
                "OR_TOOLS: objective_context deve estar congelado."
            )

        active = context.get("active_objectives", []) or []
        scales = context.get("scales", {}) or {}
        coefficients = context.get("coefficients", {}) or {}

        if not active:
            raise ValueError(
                "OR_TOOLS: objective_context sem objetivos ativos."
            )

        for obj in active:
            scale = float(scales.get(obj, 0.0) or 0.0)
            coefficient = float(coefficients.get(obj, -1.0) or 0.0)

            if scale <= 0:
                raise ValueError(
                    f"OR_TOOLS: escala inválida para {obj}."
                )

            if coefficient < 0:
                raise ValueError(
                    f"OR_TOOLS: coeficiente inválido para {obj}."
                )

        return context

    def _canonical_objective_details(self, components):
        """
        Calcula score e componentes normalizados usando exclusivamente
        o objective_context do MathAgent.
        """
        context = self._get_objective_context()
        active = context["active_objectives"]
        scales = context["scales"]
        coefficients = context["coefficients"]

        normalized = {
            key: float(value) / max(1e-12, float(scales.get(key, 1.0)))
            for key, value in components.items()
        }

        score = sum(
            float(coefficients[obj]) * float(components.get(obj, 0.0))
            for obj in active
        )

        return float(score), normalized, context

    def _build_routing_model(self):

        data = self.data

        num_nodes = len(data["demands"])

        num_physical = (
            data["num_physical_vehicles"]
        )

        if data["allow_multi_trip"]:

            trips_per_vehicle = (
                data["max_trips_per_vehicle"]
            )

        else:

            trips_per_vehicle = 1

        total_routing_vehicles = (
            num_physical
            * trips_per_vehicle
        )


        # MAPEAMENTO DOS VEÍCULOS VIRTUAIS


        virtual_vehicle_info = []

        for physical_vehicle in range(
            num_physical
        ):

            for trip_number in range(
                trips_per_vehicle
            ):

                virtual_index = len(
                    virtual_vehicle_info
                )

                virtual_vehicle_info.append(
                    {
                        "virtual_index":
                            virtual_index,

                        "physical_vehicle":
                            physical_vehicle,

                        "trip_number":
                            trip_number + 1
                    }
                )

        self.data[
            "virtual_vehicle_info"
        ] = virtual_vehicle_info


        # ROUTING INDEX MANAGER


        self.manager = (
            pywrapcp.RoutingIndexManager(
                num_nodes,
                total_routing_vehicles,
                0
            )
        )

        self.routing = (
            pywrapcp.RoutingModel(
                self.manager
            )
        )


        # TIME CALLBACK


        def time_callback(
            from_index,
            to_index
        ):

            from_node = (
                self.manager.IndexToNode(
                    from_index
                )
            )

            to_node = (
                self.manager.IndexToNode(
                    to_index
                )
            )

            distance = data[
                "dist_matrix"
            ][
                (from_node, to_node)
            ]

            travel_time = (
                distance
                / data["avg_speed"]
                * 60.0
            )

            # O serviço ocorre no cliente de origem.
            if from_node != 0:

                travel_time += (
                    data["service_time"]
                )

            return self._scale_time_duration(
                travel_time
            )

        time_callback_index = (
            self.routing.RegisterTransitCallback(
                time_callback
            )
        )


        # DEMAND CALLBACK


        def demand_callback(
            from_index
        ):

            from_node = (
                self.manager.IndexToNode(
                    from_index
                )
            )

            return self._scale_value(
                data["demands"][from_node]
            )

        demand_callback_index = (
            self.routing.RegisterUnaryTransitCallback(
                demand_callback
            )
        )


        # CAPACIDADE


        # Escalonamos capacidade também para
        # manter tudo em inteiros.
        capacities_scaled = [
            self._scale_value(x)
            for x in data["capacities"]
        ]

        vehicle_capacities = []

        for info in virtual_vehicle_info:

            physical = info[
                "physical_vehicle"
            ]

            vehicle_capacities.append(
                capacities_scaled[physical]
            )

        self.routing.AddDimensionWithVehicleCapacity(
            demand_callback_index,
            0,
            vehicle_capacities,
            True,
            "Capacity"
        )


        # TIME DIMENSION


        horizon = max(
            data["horizon_minutes"],
            data["depot_end"]
        )

        self.routing.AddDimension(
            time_callback_index,

            # Slack = espera
            self._scale_time_upper_bound(horizon),

            # Capacidade máxima da dimensão
            self._scale_time_upper_bound(horizon),

            # NÃO fixar todos os Starts em zero.
            False,

            "Time"
        )

        time_dimension = (
            self.routing.GetDimensionOrDie(
                "Time"
            )
        )


        # JANELA DO DEPÓSITO


        depot_window = data[
            "time_windows"
        ][0]

        depot_start_scaled = (
            self._scale_time_lower_bound(
                depot_window["start"]
            )
        )

        depot_end_scaled = (
            self._scale_time_upper_bound(
                depot_window["end"]
            )
        )

        for v in range(
            total_routing_vehicles
        ):

            start = self.routing.Start(v)
            end = self.routing.End(v)

            time_dimension.CumulVar(
                start
            ).SetRange(
                depot_start_scaled,
                depot_end_scaled
            )

            time_dimension.CumulVar(
                end
            ).SetRange(
                depot_start_scaled,
                depot_end_scaled
            )


        # JANELAS DOS CLIENTES


        for client_id, window in (
            data["time_windows"].items()
        ):

            if client_id == 0:
                continue

            node = data[
                "client_id_to_node"
            ].get(client_id)

            if node is None:
                continue

            routing_index = (
                self.manager.NodeToIndex(node)
            )

            if routing_index < 0:
                continue

            start_scaled = (
                self._scale_time_lower_bound(
                    window["start"]
                )
            )

            end_scaled = (
                self._scale_time_upper_bound(
                    window["end"]
                )
            )

            time_dimension.CumulVar(
                routing_index
            ).SetRange(
                start_scaled,
                end_scaled
            )


        # TEMPO OPERACIONAL POR VEÍCULO FÍSICO

        # Definição canônica: soma da duração das viagens ativas.
        # Intervalos ociosos entre viagens não contam como utilização.
        solver = self.routing.solver()
        horizon_scaled = self._scale_time_upper_bound(horizon)
        max_time_scaled = self._scale_time_upper_bound(data["max_time"])
        physical_usage_exprs = {}

        for physical in range(num_physical):
            base = physical * trips_per_vehicle
            duration_exprs = []

            for trip in range(trips_per_vehicle):
                v = base + trip
                start_var = time_dimension.CumulVar(self.routing.Start(v))
                end_var = time_dimension.CumulVar(self.routing.End(v))
                active = self.routing.ActiveVehicleVar(v)
                duration = end_var - start_var

                # Uma viagem virtual inativa deve ter duração operacional zero.
                solver.Add(duration <= horizon_scaled * active)
                duration_exprs.append(duration)

            usage = solver.Sum(duration_exprs)
            physical_usage_exprs[physical] = usage
            solver.Add(usage <= max_time_scaled)

        self.data["physical_usage_exprs"] = physical_usage_exprs


        # ORDENAMENTO DAS VIAGENS


        if data["allow_multi_trip"]:

            for physical in range(
                num_physical
            ):

                base = (
                    physical
                    * trips_per_vehicle
                )

                for trip in range(
                    trips_per_vehicle - 1
                ):

                    current_v = (
                        base + trip
                    )

                    next_v = (
                        base + trip + 1
                    )

                    current_start = (
                        self.routing.Start(
                            current_v
                        )
                    )

                    current_end = (
                        self.routing.End(
                            current_v
                        )
                    )

                    next_start = (
                        self.routing.Start(
                            next_v
                        )
                    )

                    # A próxima viagem só pode
                    # começar depois da anterior.
                    self.routing.solver().Add(
                        time_dimension.CumulVar(
                            next_start
                        )
                        >=
                        time_dimension.CumulVar(
                            current_end
                        )
                    )


        # USO OBRIGATÓRIO DOS VEÍCULOS


        if data[
            "must_use_all_vehicles"
        ]:

            for physical in range(
                num_physical
            ):

                first_virtual = (
                    physical
                    * trips_per_vehicle
                )

                start = (
                    self.routing.Start(
                        first_virtual
                    )
                )

                end = (
                    self.routing.End(
                        first_virtual
                    )
                )

                # O primeiro trip precisa ser usado.
                self.routing.solver().Add(
                    self.routing.NextVar(start)
                    != end
                )


        # CONTIGUIDADE DAS VIAGENS


        if data["allow_multi_trip"]:

            for physical in range(
                num_physical
            ):

                base = (
                    physical
                    * trips_per_vehicle
                )

                for trip in range(
                    trips_per_vehicle - 1
                ):

                    current_v = (
                        base + trip
                    )

                    next_v = (
                        base + trip + 1
                    )

                    current_start = (
                        self.routing.Start(
                            current_v
                        )
                    )

                    current_end = (
                        self.routing.End(
                            current_v
                        )
                    )

                    next_start = (
                        self.routing.Start(
                            next_v
                        )
                    )

                    # Se a viagem seguinte for utilizada,
                    # a anterior precisa ser utilizada.

                    # Isso impede:

                    # Trip 1 = ociosa
                    # Trip 2 = utilizada

                    self.routing.solver().Add(
                        self.routing.ActiveVehicleVar(
                            next_v
                        )
                        <=
                        self.routing.ActiveVehicleVar(
                            current_v
                        )
                    )

                    # Ordem temporal
                    self.routing.solver().Add(
                        time_dimension.CumulVar(
                            next_start
                        )
                        >=
                        time_dimension.CumulVar(
                            current_end
                        )
                    )


        # TEMPO MÍNIMO DE UTILIZAÇÃO

        if data["minimum_usage"] > 0:
            minimum_usage_scaled = self._scale_time_lower_bound(data["minimum_usage"])

            for physical in range(num_physical):
                first_virtual = physical * trips_per_vehicle
                active_physical = self.routing.ActiveVehicleVar(first_virtual)
                usage = physical_usage_exprs[physical]

                # Veículo ocioso: uso = 0 e nenhum mínimo é exigido.
                # Veículo ativo: soma das durações das viagens >= mínimo.
                solver.Add(usage >= minimum_usage_scaled * active_physical)


        # OBJETIVO CANÔNICO / CUSTOS

        objective_context = self._get_objective_context()
        coefficients = objective_context["coefficients"]
        data["objective_context"] = objective_context

        # Escala inteira usada apenas pela API RoutingModel.
        # Não altera a função matemática, apenas converte coeficientes
        # contínuos para custos inteiros.
        ROUTING_OBJECTIVE_SCALE = 1_000_000_000

        for v, info in enumerate(virtual_vehicle_info):
            physical = info["physical_vehicle"]
            cost_per_km = float(data["costs_km"][physical])

            def canonical_arc_cost_callback(
                from_index,
                to_index,
                cost_per_km=cost_per_km
            ):
                i = self.manager.IndexToNode(from_index)
                j = self.manager.IndexToNode(to_index)

                distance = float(data["dist_matrix"][(i, j)])
                variable_cost = distance * cost_per_km

                value = (
                    float(coefficients.get("DISTANCE", 0.0)) * distance
                    + float(coefficients.get("COST", 0.0)) * variable_cost
                )

                return int(round(value * ROUTING_OBJECTIVE_SCALE))

            callback_idx = self.routing.RegisterTransitCallback(
                canonical_arc_cost_callback
            )
            self.routing.SetArcCostEvaluatorOfVehicle(
                callback_idx,
                v
            )

            # Custos ligados à ativação do veículo físico entram apenas
            # na primeira viagem virtual. A restrição de contiguidade garante
            # que viagens posteriores não possam ser usadas sem a primeira.
            trip_number = info["trip_number"]
            activation_term = 0.0

            if trip_number == 1:
                if (
                    data["fixed_cost_mode"] == "WHEN_USED"
                    and float(coefficients.get("COST", 0.0)) > 0.0
                ):
                    activation_term += (
                        float(coefficients["COST"])
                        * float(data["fixed_costs"][physical])
                    )

                if float(coefficients.get("VEHICLES", 0.0)) > 0.0:
                    activation_term += float(
                        coefficients["VEHICLES"]
                    )

            self.routing.SetFixedCostOfVehicle(
                int(round(
                    activation_term
                    * ROUTING_OBJECTIVE_SCALE
                )),
                v
            )

            # TIME é representado pelo span real da dimensão Time e, portanto,
            # inclui deslocamento, serviço e espera por janelas.
            if float(coefficients.get("TIME", 0.0)) > 0.0:
                time_coeff = int(round(
                    float(coefficients["TIME"])
                    * ROUTING_OBJECTIVE_SCALE
                    / self.TIME_SCALING_FACTOR
                ))

                if time_coeff > 0:
                    time_dimension.SetSpanCostCoefficientForVehicle(
                        time_coeff,
                        v
                    )


        # FINALIZADORES


        for v in range(
            total_routing_vehicles
        ):

            self.routing.AddVariableMinimizedByFinalizer(
                time_dimension.CumulVar(
                    self.routing.Start(v)
                )
            )

            self.routing.AddVariableMinimizedByFinalizer(
                time_dimension.CumulVar(
                    self.routing.End(v)
                )
            )

        return time_dimension


    # CONFIGURAÇÃO DO SOLVER


    def _configure_search_parameters(
        self
    ):

        search_parameters = (
            pywrapcp.DefaultRoutingSearchParameters()
        )


        # SOLUÇÃO INICIAL


        search_parameters.first_solution_strategy = (
            routing_enums_pb2.FirstSolutionStrategy
            .PATH_CHEAPEST_ARC
        )


        # META-HEURÍSTICA


        search_parameters.local_search_metaheuristic = (
            routing_enums_pb2.LocalSearchMetaheuristic
            .GUIDED_LOCAL_SEARCH
        )


        # TEMPO


        # Limite adaptativo por tamanho da instância.
        # Mantém controle de tempo sem enfraquecer a busca.
        num_clients = max(0, len(self.data.get("demands", [])) - 1)
        explicit_limit = self.data.get("solver_time_limit_seconds")

        if explicit_limit is not None:
            time_limit_seconds = int(explicit_limit)
        elif num_clients <= 20:
            time_limit_seconds = 120
        elif num_clients <= 50:
            time_limit_seconds = 180
        else:
            time_limit_seconds = 240

        search_parameters.time_limit.seconds = max(1, time_limit_seconds)


        # LOG


        search_parameters.log_search = False

        return search_parameters


    # DECODIFICAÇÃO DA SOLUÇÃO


    def _decode_solution(
        self,
        time_dimension
    ):

        routes = []

        data = self.data

        virtual_info = (
            data[
                "virtual_vehicle_info"
            ]
        )

        for v, info in enumerate(
            virtual_info
        ):

            start = (
                self.routing.Start(v)
            )

            first_next = (
                self.solution.Value(
                    self.routing.NextVar(
                        start
                    )
                )
            )


            # VIAGEM OCIOSA


            if self.routing.IsEnd(
                first_next
            ):

                continue

            physical = info[
                "physical_vehicle"
            ]

            trip_number = info[
                "trip_number"
            ]

            idx = start

            path_nodes = []
            path_details = []

            route_distance = 0.0
            route_load = 0.0

            previous_time = None

            while not self.routing.IsEnd(idx):

                node = (
                    self.manager.IndexToNode(
                        idx
                    )
                )

                client_id = (
                    data[
                        "node_to_client_id"
                    ][node]
                )

                current_time = (
                    self._unscale_time(
                        self.solution.Value(
                            time_dimension.CumulVar(
                                idx
                            )
                        )
                    )
                )

                path_nodes.append(
                    client_id
                )

                demand = (
                    data["demands"][node]
                )

                route_load += demand

                path_details.append(
                    {
                        "node_id": client_id,

                        "arrival_time_minutes":
                            current_time,

                        "demand":
                            demand
                    }
                )

                next_idx = (
                    self.solution.Value(
                        self.routing.NextVar(
                            idx
                        )
                    )
                )

                next_node = (
                    self.manager.IndexToNode(
                        next_idx
                    )
                )

                route_distance += (
                    data["dist_matrix"][
                        (node, next_node)
                    ]
                )

                previous_time = current_time

                idx = next_idx


            # DEPÓSITO FINAL


            final_node = (
                self.manager.IndexToNode(
                    idx
                )
            )

            final_client_id = (
                data[
                    "node_to_client_id"
                ][final_node]
            )

            final_time = (
                self._unscale_time(
                    self.solution.Value(
                        time_dimension.CumulVar(
                            idx
                        )
                    )
                )
            )

            path_nodes.append(
                final_client_id
            )

            path_details.append(
                {
                    "node_id":
                        final_client_id,

                    "arrival_time_minutes":
                        final_time,

                    "demand":
                        0
                }
            )


            # TEMPO DE INÍCIO


            start_time = (
                self._unscale_time(
                    self.solution.Value(
                        time_dimension.CumulVar(
                            start
                        )
                    )
                )
            )

            end_time = final_time

            route_duration = (
                end_time
                - start_time
            )


            # CUSTO VARIÁVEL


            variable_cost = (
                route_distance
                * data["costs_km"][physical]
            )


            # CUSTO FIXO


            # O custo fixo é associado ao veículo
            # físico e não a cada trip.

            # Portanto somente a primeira viagem
            # recebe o valor no relatório.

            fixed_cost = 0.0

            if (
                trip_number == 1
                and data[
                    "fixed_cost_mode"
                ] == "WHEN_USED"
            ):

                fixed_cost = (
                    data[
                        "fixed_costs"
                    ][physical]
                )

            route_cost = (
                variable_cost
                + fixed_cost
            )

            routes.append(
                {
                    "vehicle_id":
                        physical + 1,

                    "trip_number":
                        trip_number,

                    "physical_vehicle_id":
                        physical + 1,

                    "path":
                        path_nodes,

                    "path_details":
                        path_details,

                    "start_time_minutes":
                        start_time,

                    "end_time_minutes":
                        end_time,

                    "total_route_time_minutes":
                        route_duration,

                    "distance_km":
                        route_distance,

                    "total_route_load":
                        route_load,

                    "variable_cost":
                        variable_cost,

                    "fixed_cost":
                        fixed_cost,

                    "total_route_cost":
                        route_cost
                }
            )

        return routes


    # AUDITORIA DA SOLUÇÃO


    def _audit_solution(
        self,
        routes
    ):

        data = self.data

        audit = {
            "valid": True,
            "errors": [],
            "warnings": []
        }


        # CLIENTES ATENDIDOS


        visited_clients = []

        for route in routes:

            for client_id in route["path"]:

                if client_id != 0:
                    visited_clients.append(
                        client_id
                    )

        expected_clients = [
            data["node_to_client_id"][node]
            for node in range(
                1,
                len(data["demands"])
            )
        ]


        # DUPLICADOS


        duplicates = {
            client_id
            for client_id in visited_clients
            if visited_clients.count(client_id) > 1
        }

        if duplicates:

            audit["valid"] = False

            audit["errors"].append(
                "Clientes atendidos mais de uma vez: "
                f"{sorted(duplicates)}"
            )


        # NÃO ATENDIDOS


        missing = sorted(
            set(expected_clients)
            - set(visited_clients)
        )

        if missing:

            audit["valid"] = False

            audit["errors"].append(
                "Clientes não atendidos: "
                f"{missing}"
            )


        # CAPACIDADE


        for route in routes:

            vehicle_id = (
                route["physical_vehicle_id"]
                - 1
            )

            capacity = (
                data["capacities"][
                    vehicle_id
                ]
            )

            load = (
                route["total_route_load"]
            )

            if load > capacity + 1e-9:

                audit["valid"] = False

                audit["errors"].append(
                    f"Veículo {vehicle_id + 1}, "
                    f"Trip {route['trip_number']}: "
                    f"carga {load:.3f} > "
                    f"capacidade {capacity:.3f}."
                )


        # JORNADA POR VEÍCULO


        for physical in range(
            data["num_physical_vehicles"]
        ):

            vehicle_routes = [
                r
                for r in routes
                if r[
                    "physical_vehicle_id"
                ] == physical + 1
            ]

            if not vehicle_routes:

                if data[
                    "must_use_all_vehicles"
                ]:

                    audit["valid"] = False

                    audit["errors"].append(
                        f"Veículo {physical + 1} "
                        "deveria ser utilizado, "
                        "mas está ocioso."
                    )

                continue

            vehicle_routes.sort(
                key=lambda r:
                    r["trip_number"]
            )

            total_usage = sum(
                float(r.get("total_route_time_minutes", 0.0) or 0.0)
                for r in vehicle_routes
            )

            if total_usage > data["max_time"] + 1e-9:
                audit["valid"] = False
                audit["errors"].append(
                    f"Veículo {physical + 1}: utilização {total_usage:.2f} min "
                    f"> máximo {data['max_time']:.2f} min."
                )

            if (
                data["minimum_usage"] > 0
                and total_usage + 1e-9 < data["minimum_usage"]
            ):
                audit["valid"] = False
                audit["errors"].append(
                    f"Veículo {physical + 1}: utilização {total_usage:.2f} min "
                    f"< mínimo {data['minimum_usage']:.2f} min."
                )


        # SEQUÊNCIA DAS VIAGENS


        for physical in range(
            data["num_physical_vehicles"]
        ):

            vehicle_routes = [
                r
                for r in routes
                if r[
                    "physical_vehicle_id"
                ] == physical + 1
            ]

            vehicle_routes.sort(
                key=lambda r:
                    r["trip_number"]
            )

            for i in range(
                len(vehicle_routes) - 1
            ):

                current_route = (
                    vehicle_routes[i]
                )

                next_route = (
                    vehicle_routes[i + 1]
                )

                if (
                    next_route[
                        "start_time_minutes"
                    ]
                    <
                    current_route[
                        "end_time_minutes"
                    ]
                    - 1e-9
                ):

                    audit["valid"] = False

                    audit["errors"].append(
                        f"Sobreposição de viagens "
                        f"no veículo {physical + 1}."
                    )


        # JANELAS DE TEMPO


        for route in routes:

            for detail in route[
                "path_details"
            ]:

                client_id = detail[
                    "node_id"
                ]

                if client_id == 0:
                    continue

                window = data[
                    "time_windows"
                ].get(client_id)

                if window is None:
                    continue

                arrival = detail[
                    "arrival_time_minutes"
                ]

                if (
                    arrival
                    < window["start"]
                    - 1e-9
                    or
                    arrival
                    > window["end"]
                    + 1e-9
                ):

                    audit["valid"] = False

                    audit["errors"].append(
                        f"Cliente {client_id}: "
                        f"tempo {arrival:.2f} "
                        f"fora da janela "
                        f"[{window['start']:.2f}, "
                        f"{window['end']:.2f}]."
                    )


        # RETORNO AO DEPÓSITO


        if data["return_to_depot"]:

            for route in routes:

                if (
                    not route["path"]
                    or route["path"][-1] != 0
                ):

                    audit["valid"] = False

                    audit["errors"].append(
                        f"Veículo "
                        f"{route['vehicle_id']} "
                        f"Trip "
                        f"{route['trip_number']} "
                        "não retornou ao depósito."
                    )


        # CUSTO FIXO


        if (
            data["fixed_cost_mode"]
            == "ALWAYS"
        ):

            expected_fixed = sum(
                data["fixed_costs"]
            )

        else:

            used_physical = {
                r[
                    "physical_vehicle_id"
                ] - 1
                for r in routes
            }

            expected_fixed = sum(
                data["fixed_costs"][v]
                for v in used_physical
            )

        variable_cost = sum(
            r["variable_cost"]
            for r in routes
        )

        reported_fixed = sum(
            r["fixed_cost"]
            for r in routes
        )

        # Quando custo fixo = ALWAYS,
        # ele não é colocado individualmente
        # em cada rota para não duplicar.
        if (
            data["fixed_cost_mode"]
            == "ALWAYS"
        ):

            reported_fixed = (
                expected_fixed
            )

        expected_total = (
            variable_cost
            + expected_fixed
        )


        # RESULTADO DA AUDITORIA


        audit["total_distance_km"] = sum(
            r["distance_km"]
            for r in routes
        )

        audit["total_time_minutes"] = sum(
            float(r.get("total_route_time_minutes", 0.0) or 0.0)
            for r in routes
        )

        audit["total_variable_cost"] = (
            variable_cost
        )

        audit["total_fixed_cost"] = (
            expected_fixed
        )

        audit["total_cost"] = (
            expected_total
        )

        audit["num_routes"] = len(
            routes
        )

        audit["num_vehicles_used"] = len(
            {
                r[
                    "physical_vehicle_id"
                ]
                for r in routes
            }
        )

        return audit


    # CONSTRUÇÃO DO RESULTADO FINAL


    def _build_result(
        self,
        routes,
        audit
    ):

        data = self.data

        total_demand = sum(
            data["demands"]
        )

        total_capacity = sum(
            data["capacities"]
        )

        vehicles_used = (
            audit[
                "num_vehicles_used"
            ]
        )

        vehicles_total = (
            data[
                "num_physical_vehicles"
            ]
        )

        vehicle_utilization = (
            vehicles_used
            / vehicles_total
            * 100.0
            if vehicles_total > 0
            else 0.0
        )

        vehicle_idle_percentage = (
            100.0
            - vehicle_utilization
        )

        result = {

            "status": (
                "SUCCESS"
                if audit["valid"]
                else "INVALID_SOLUTION"
            ),

            "method":
                "OR-Tools",

            "solver":
                "OR-Tools RoutingModel",

            "routes":
                routes,

            "summary": {

                "total_distance_km":
                    audit[
                        "total_distance_km"
                    ],

                "total_time_minutes":
                    audit.get(
                        "total_time_minutes",
                        0.0
                    ),

                "total_variable_cost":
                    audit[
                        "total_variable_cost"
                    ],

                "total_fixed_cost":
                    audit[
                        "total_fixed_cost"
                    ],

                "total_cost":
                    audit[
                        "total_cost"
                    ],

                "total_demand":
                    total_demand,

                "total_fleet_capacity":
                    total_capacity,

                "num_vehicles_total":
                    vehicles_total,

                "num_vehicles_used":
                    vehicles_used,

                "vehicle_utilization_percentage":
                    vehicle_utilization,

                "vehicle_idle_percentage":
                    vehicle_idle_percentage,

                "num_routes":
                    audit[
                        "num_routes"
                    ],

                "allow_multi_trip":
                    data[
                        "allow_multi_trip"
                    ],

                "must_use_all_vehicles":
                    data[
                        "must_use_all_vehicles"
                    ],

                "fixed_cost_mode":
                    data[
                        "fixed_cost_mode"
                    ]
            },

            "audit":
                audit,

            "errors":
                self.errors,

            "warnings":
                self.warnings
        }

        return result


    # SOLVE PROBLEM


    def solve_problem(
        self,
        classification,
        translated_json
    ):

        import time as _time_runtime
        _solve_started = _time_runtime.monotonic()
        print("🔵 OR-Tools: iniciando construção e busca da solução...")


        # RESET DE ESTADO


        self.errors = []
        self.warnings = []

        self.manager = None
        self.routing = None
        self.solution = None
        self.data = None


        # CLASSIFICATION


        if classification is not None:

            problem_type = (
                classification.get(
                    "problem_type"
                )
                if isinstance(
                    classification,
                    dict
                )
                else None
            )

            if (
                problem_type
                and str(problem_type).upper()
                not in (
                    "VRP",
                    "CVRP",
                    "VRPTW",
                    "MTVRP",
                    "MULTI_TRIP_VRP"
                )
            ):

                self.warnings.append(
                    f"Tipo de problema "
                    f"'{problem_type}' "
                    "não é uma variante "
                    "explicitamente suportada "
                    "por este solver."
                )


        # DATA MODEL


        self.data = (
            self._create_data_model(
                translated_json
            )
        )

        if self.data is None:

            return {
                "status": "ERROR",
                "error_log": self.errors
            }

        # Contrato de orçamento: o Orchestrator publica o limite corrente em SharedMemory.
        # O tempo de construção já consumido é descontado do tempo entregue à busca.
        requested_limit = self.shared_memory.data.get("current_solver_time_limit_seconds")
        if requested_limit is not None:
            requested_limit = max(1.0, float(requested_limit))
            elapsed_build = _time_runtime.monotonic() - _solve_started
            remaining_search = requested_limit - elapsed_build
            if remaining_search <= 0.05:
                return {
                    "status": "TIME_LIMIT",
                    "method": "OR-Tools",
                    "error_log": ["Orçamento OR-Tools esgotado durante a construção do modelo."],
                    "warnings": self.warnings,
                    "elapsed_seconds": elapsed_build,
                }
            self.data["solver_time_limit_seconds"] = max(1, int(remaining_search))


        # VALIDATION


        if not self._validate_data_model():

            return {
                "status": "INFEASIBLE",
                "error_log":
                    self.errors,
                "warnings":
                    self.warnings
            }


        # BUILD ROUTING MODEL


        try:

            time_dimension = (
                self._build_routing_model()
            )

        except Exception as e:

            self.errors.append(
                "Erro ao construir modelo "
                f"OR-Tools: {str(e)}"
            )

            return {
                "status": "ERROR",
                "error_log":
                    self.errors
            }


        # SEARCH PARAMETERS


        search_parameters = (
            self._configure_search_parameters()
        )


        # SOLVE


        try:

            self.solution = (
                self.routing.SolveWithParameters(
                    search_parameters
                )
            )

        except Exception as e:

            self.errors.append(
                "Erro durante execução "
                f"do OR-Tools: {str(e)}"
            )

            return {
                "status": "ERROR",
                "error_log":
                    self.errors
            }


        # NO SOLUTION


        if self.solution is None:
            elapsed_total = _time_runtime.monotonic() - _solve_started
            requested_limit = self.shared_memory.data.get("current_solver_time_limit_seconds")
            timed_out = (
                requested_limit is not None
                and elapsed_total >= max(0.0, float(requested_limit) - 1.0)
            )

            self.errors.append(
                "OR-Tools não encontrou uma solução viável "
                + ("dentro do orçamento de busca." if timed_out else "nas condições fornecidas.")
            )

            return {
                "status": "TIME_LIMIT" if timed_out else "INFEASIBLE",
                "method": "OR-Tools",
                "error_log": self.errors,
                "warnings": self.warnings,
                "elapsed_seconds": elapsed_total,
            }


        # DECODE


        try:

            routes = self._decode_solution(
                time_dimension
            )

        except Exception as e:

            self.errors.append(
                "Erro ao decodificar solução: "
                f"{str(e)}"
            )

            return {
                "status": "ERROR",
                "error_log":
                    self.errors
            }


        # AUDIT


        audit = self._audit_solution(
            routes
        )


        # FINAL RESULT


        result = self._build_result(
            routes,
            audit
        )


        # FUNÇÃO OBJETIVO CANÔNICA


        # Valor inteiro interno do RoutingModel, útil apenas para diagnóstico.
        result["routing_objective_scaled"] = (
            self.solution.ObjectiveValue()
        )

        summary = result.get("summary", {}) or {}

        components = {
            "DISTANCE": float(
                summary.get(
                    "total_distance_km",
                    summary.get("total_distance", 0.0)
                ) or 0.0
            ),
            "TIME": float(
                summary.get(
                    "total_time_minutes",
                    summary.get("total_time", 0.0)
                ) or 0.0
            ),
            "COST": float(
                summary.get("total_cost", 0.0) or 0.0
            ),
            "VEHICLES": float(
                summary.get("num_vehicles_used", 0) or 0
            ),
        }

        canonical_score, normalized, objective_context = (
            self._canonical_objective_details(components)
        )

        result["solver_objective"] = canonical_score
        result["canonical_objective_score"] = canonical_score
        result["objective_components"] = components
        result["objective_normalized"] = normalized
        result["objective_context"] = objective_context
        result["objective_context_version"] = (
            objective_context.get("version")
        )
        result["optimization_objective"] = (
            "MULTI"
            if len(objective_context["active_objectives"]) > 1
            else objective_context["active_objectives"][0]
        )

        # Mantidos apenas como metadados de entrada para compatibilidade
        # com relatórios existentes; o solver não usa estes campos para
        # recalcular pesos ou escalas.
        result["multiobjective_config"] = self.data.get(
            "multiobjective_config",
            {}
        )
        result["objective_weights"] = objective_context.get(
            "weights",
            {}
        )
        result["objective_scales"] = objective_context.get(
            "scales",
            {}
        )


        # AVISO SOBRE CUSTO FIXO ALWAYS


        if (
            self.data[
                "fixed_cost_mode"
            ] == "ALWAYS"
        ):

            result["summary"][
                "fixed_cost_note"
            ] = (
                "O custo fixo da frota é "
                "adicionado como constante "
                "no relatório final, pois "
                "deve ser pago mesmo para "
                "veículos não utilizados."
            )


        # QUALIDADE MATEMÁTICA / OPTIMALIDADE

        # RoutingModel + Guided Local Search é uma busca heurística/meta-heurística.
        # SUCCESS significa solução viável e auditada, não prova de ótimo global.
        if result.get("status") == "SUCCESS":
            result["optimality"] = {
                "status": "BEST_FOUND_NOT_PROVEN",
                "proven_optimal": False,
                "termination_reason": "SEARCH_FINISHED_OR_TIME_LIMIT",
                "objective_value": result.get("solver_objective"),
                "best_bound": None,
                "optimality_gap": None,
                "explanation": (
                    "A solução é viável e foi a melhor encontrada pelo OR-Tools "
                    "na configuração de busca utilizada. Como o RoutingModel está "
                    "usando Guided Local Search, não há prova de ótimo global."
                )
            }
        else:
            result["optimality"] = {
                "status": "NO_VALID_SOLUTION",
                "proven_optimal": False,
                "termination_reason": result.get("status"),
                "objective_value": result.get("solver_objective"),
                "best_bound": None,
                "optimality_gap": None
            }

        result["solution_quality"] = {
            "feasible": result.get("status") == "SUCCESS",
            "audited": bool(result.get("audit", {}).get("valid", False)),
            "optimality_proven": False,
            "quality_label": (
                "MELHOR_SOLUCAO_ENCONTRADA"
                if result.get("status") == "SUCCESS"
                else "SEM_SOLUCAO_VALIDA"
            )
        }

        result["runtime_seconds"] = round(_time_runtime.monotonic() - _solve_started, 3)
        print(
            f"✅ OR-Tools: terminou em {result['runtime_seconds']:.1f}s | "
            f"status={result.get('status')} | "
            f"custo={float((result.get('summary') or {}).get('total_cost', result.get('total_cost', 0.0)) or 0.0):.2f} | "
            f"distância={float((result.get('summary') or {}).get('total_distance_km', 0.0) or 0.0):.2f} km"
        )

        return result
````

### **GENETICALGORITHMSOLVERAGENT — BUSCA POR COMBINAÇÃO DE ALTERNATIVAS**

Esse agente cria e compara propostas, aproveita partes promissoras e testa novas combinações. O nome “genético” vem dessa maneira de combinar e modificar propostas.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class GeneticAlgorithmSolverAgent:
    VERSION = "GA_SOLVER_V15_4_PHYSICAL_HARD_K_FLEET"

    """
    GeneticAlgorithmSolverAgent
    ===========================

    Solver híbrido para VRP / VRPTW / Multi-Trip.

    Principais características:

    - Permutação de clientes como cromossomo.
    - Decoder responsável pela distribuição dos clientes entre:
        * veículos
        * viagens
    - Multi-trip real.
    - Capacidade por viagem.
    - Janela de tempo com espera.
    - Tempo disponível acumulado por veículo.
    - Tempo mínimo de utilização.
    - Tempo mínimo aplicado apenas aos veículos efetivamente utilizados.
    - Uso obrigatório da frota controlado exclusivamente por must_use_all_vehicles.
    - Custo variável por km por veículo.
    - Custo fixo por ativação.
    - Custo fixo incondicional.
    - OX crossover.
    - Swap.
    - Relocate.
    - 2-opt.
    - Busca local.
    - Elitismo.
    - Mutação adaptativa.
    - Diversificação.
    - Penalização hierárquica das restrições.
    - Integração com SharedMemory.
    """


    # CONSTRUTOR


    def __init__(self, shared_memory):

        self.shared_memory = shared_memory

        self.role = (
            "Solver de Algoritmos Genéticos para VRP/VRPTW "
            "Multi-Trip."
        )

        self.goal = (
            "Encontrar soluções de alta qualidade respeitando "
            "capacidade, janelas de tempo, tempo disponível, "
            "tempo mínimo, utilização obrigatória da frota "
            "e custos operacionais."
        )

        self.backstory = (
            "Especialista em meta-heurísticas para problemas "
            "de roteirização combinatória."
        )

        self.tools = [
            "Order Crossover (OX)",
            "Swap",
            "Relocate",
            "2-opt",
            "Elitismo",
            "Mutação adaptativa",
            "Busca local",
            "Diversificação",
            "Decoder Multi-Trip",
        ]

        self.memory = []
        # Cache de decodificações especiais: permite preservar sementes de frota mínima sem alterar o cromossomo.
        self._special_decoded_cache = {}
        # Arquivo interno de soluções não dominadas para o modo multiobjetivo.
        self._pareto_archive = {}
        # Congela a primeira decodificação de cada cromossomo durante a execução.
        # Preserva o comportamento estocástico do decoder sem permitir que o mesmo
        # cromossomo mude de fitness ou de factibilidade em avaliações posteriores.
        self._fitness_cache = {}

        self.constraints = [
            "Cada cliente deve ser atendido exatamente uma vez.",
            "Capacidade deve ser respeitada em cada viagem.",
            "Janelas de tempo devem ser respeitadas.",
            "Chegada antecipada gera espera.",
            "Chegada após a janela gera violação.",
            "Tempo total do veículo não pode ultrapassar sua disponibilidade.",
            "Viagens do mesmo veículo são sequenciais.",
            "Veículos utilizados devem respeitar o tempo mínimo quando configurado.",
            "Veículos ociosos são permitidos quando must_use_all_vehicles=False.",
            "Custo fixo deve respeitar sua regra de cobrança.",
        ]

    #FUNÇÃO AUXILIAR PARA CONTROLAR O TEMPO DE EXECUÇÃO
    def _time_limit_reached(self,start_time,time_limit_seconds):
        """ Verifica se o tempo máximo paremitido para o AG foi atingido.
        Retorna:
        True -> limite atingido
        False -> ainda há tempo disponível
        """
        elapsed_time = _time_module.monotonic() - start_time
        deadline = getattr(self, "_active_deadline", None)
        if deadline is not None:
            return _time_module.monotonic() >= deadline
        return elapsed_time >= time_limit_seconds

    def _deadline_reached(self):
        deadline = getattr(self, "_active_deadline", None)
        return deadline is not None and _time_module.monotonic() >= deadline


    # NORMALIZAÇÃO DO OBJETIVO


    def _normalize_objective(
        self,
        objective
    ):
        """Normaliza o objetivo do usuário para o formato interno do GA."""

        if objective is None:
            return "DISTANCE"

        text = str(
            objective
        ).strip().lower()

        if "multiobjetivo" in text:
            return "MULTI"

        if "tempo" in text:
            return "TIME"

        if "custo" in text:
            return "COST"

        if (
            "veículo" in text
            or
            "veiculos" in text
            or
            "veículos" in text
        ):
            return "VEHICLES"

        if (
            "distância" in text
            or
            "distancia" in text
        ):
            return "DISTANCE"

        return "DISTANCE"


    # OBJETIVO CANÔNICO (compatível com MathAgent / Certificador)


    def _get_objective_context(self):
        context = self.shared_memory.data.get("objective_context")
        if not isinstance(context, dict):
            return None
        active = context.get("active_objectives", []) or []
        coefficients = context.get("coefficients", {}) or {}
        scales = context.get("scales", {}) or {}
        if not active or not coefficients:
            return None
        return {
            "active_objectives": [str(x).upper() for x in active],
            "coefficients": {str(k).upper(): float(v) for k, v in coefficients.items()},
            "scales": {str(k).upper(): float(v) for k, v in scales.items()},
            "weights": {
                str(k).upper(): float(v)
                for k, v in (context.get("weights", {}) or {}).items()
            },
            "version": context.get("version"),
            "frozen": bool(context.get("frozen", False)),
        }

    def _canonical_objective_details(self, components):
        context = self._get_objective_context()
        if context is None:
            return None

        normalized = {
            key: float(value) / max(1e-12, float(context["scales"].get(key, 1.0)))
            for key, value in components.items()
        }
        score = sum(
            float(context["coefficients"].get(obj, 0.0))
            * float(components.get(obj, 0.0))
            for obj in context["active_objectives"]
        )
        return float(score), normalized, context


    # UTILIDADES


    def _parse_to_min(self, value):

        if value is None:
            return 0.0

        if isinstance(value, (int, float, np.integer, np.floating)):
            return float(value)

        if isinstance(value, time):
            return (
                value.hour * 60
                + value.minute
                + value.second / 60.0
            )

        if isinstance(value, datetime):
            return (
                value.hour * 60
                + value.minute
                + value.second / 60.0
            )

        if isinstance(value, str):

            value = value.strip()

            formats = [
                "%H:%M:%S",
                "%H:%M",
                "%H.%M",
            ]

            for fmt in formats:

                try:

                    dt = datetime.strptime(
                        value,
                        fmt
                    )

                    return (
                        dt.hour * 60
                        + dt.minute
                        + dt.second / 60.0
                    )

                except ValueError:
                    pass

            try:
                return float(value)

            except ValueError:
                return 0.0

        return 0.0


    # DISTÂNCIA HAVERSINE


    def _haversine_distance(
        self,
        lat1,
        lon1,
        lat2,
        lon2
    ):

        R = 6371.0

        lat1 = math.radians(float(lat1))
        lon1 = math.radians(float(lon1))

        lat2 = math.radians(float(lat2))
        lon2 = math.radians(float(lon2))

        dlat = lat2 - lat1
        dlon = lon2 - lon1

        a = (
            math.sin(dlat / 2.0) ** 2
            +
            math.cos(lat1)
            * math.cos(lat2)
            * math.sin(dlon / 2.0) ** 2
        )

        c = (
            2.0
            * math.atan2(
                math.sqrt(a),
                math.sqrt(
                    max(
                        0.0,
                        1.0 - a
                    )
                )
            )
        )

        return R * c


    # LEITURA DOS DADOS


    def _create_data_lookup(
        self,
        translated_problem_json
    ):

        if isinstance(
            translated_problem_json,
            str
        ):

            prob_json = json.loads(
                translated_problem_json
            )

        else:

            prob_json = translated_problem_json

        desc = prob_json.get(
            "problem_description",
            {}
        )


        # SHARED MEMORY


        mem = self.shared_memory.data

        df_clientes = mem.get(
            "df_clientes"
        )

        df_cap = mem.get(
            "df_veiculos_capacidade"
        )

        df_custos = mem.get(
            "df_veiculos_custo_km"
        )

        df_janela = mem.get(
            "df_janela_tempo_cliente"
        )

        if df_clientes is None:

            raise ValueError(
                "df_clientes não foi encontrado "
                "no shared_memory."
            )

        if df_cap is None:

            raise ValueError(
                "df_veiculos_capacidade não foi encontrado."
            )


        # CLIENTES


        client_lookup = (
            df_clientes
            .set_index("Cliente")
            .to_dict("index")
        )

        depot_id = int(
            df_clientes["Cliente"].min()
        )


        # MATRIZ DE DISTÂNCIAS PRÉ-CALCULADA

        # As coordenadas são constantes durante toda a execução do GA.
        # Portanto, calculamos Haversine uma única vez para cada par
        # de nós e reutilizamos exatamente os mesmos valores depois.
        # Isso reduz fortemente o custo computacional sem alterar a
        # matemática, o decoder, o fitness ou a sequência genética.
        node_ids = [int(cid) for cid in client_lookup.keys()]
        distance_matrix = {i: {} for i in node_ids}
        for i in node_ids:
            pi = client_lookup[i]
            for j in node_ids:
                if i == j:
                    distance_matrix[i][j] = 0.0
                else:
                    pj = client_lookup[j]
                    distance_matrix[i][j] = self._haversine_distance(
                        pi["latitude"], pi["longitude"],
                        pj["latitude"], pj["longitude"]
                    )


        # JANELAS DE TEMPO


        janela_inicio = {}
        janela_fim = {}

        has_time_windows = bool(desc.get("has_time_windows", False))
        time_window_mode = str(desc.get("time_window_mode", "NONE") or "NONE").upper()

        if has_time_windows and time_window_mode == "UNIFORM":
            start_tw = float(desc.get("uniform_time_window_start_minutes"))
            end_tw = float(desc.get("uniform_time_window_end_minutes"))
            if end_tw <= start_tw:
                raise ValueError("Janela uniforme inválida no contrato operacional.")
            for cid in client_lookup:
                if int(cid) != depot_id:
                    janela_inicio[int(cid)] = start_tw
                    janela_fim[int(cid)] = end_tw

        elif has_time_windows and time_window_mode == "PER_CLIENT":
            if df_janela is None or df_janela.empty:
                raise ValueError(
                    "Contrato indica janelas por cliente, mas "
                    "df_janela_tempo_cliente está ausente."
                )
            for _, row in df_janela.iterrows():
                cid = int(row["cliente"])
                if cid == depot_id:
                    continue
                janela_inicio[cid] = self._parse_to_min(row["inicio_janela"])
                janela_fim[cid] = self._parse_to_min(row["fim_janela"])

        elif df_janela is not None and not df_janela.empty:
            # Compatibilidade com notebooks anteriores ao contrato.
            for _, row in df_janela.iterrows():
                cid = int(row["cliente"])
                if cid == depot_id:
                    continue
                janela_inicio[cid] = self._parse_to_min(row["inicio_janela"])
                janela_fim[cid] = self._parse_to_min(row["fim_janela"])


        # CAPACIDADES


        capacities = [
            float(x)
            for x in df_cap[
                "capacidade"
            ].tolist()
        ]


        # CUSTOS


        if df_custos is not None:

            if (
                "custo_por_km"
                in df_custos.columns
            ):

                costs_km = [
                    float(x)
                    for x in df_custos[
                        "custo_por_km"
                    ].tolist()
                ]

            else:

                costs_km = [
                    0.0
                    for _ in capacities
                ]

            if (
                "custo_diario"
                in df_custos.columns
            ):

                fixed_costs = [
                    float(x)
                    for x in df_custos[
                        "custo_diario"
                    ].tolist()
                ]

            else:

                fixed_costs = [
                    0.0
                    for _ in capacities
                ]

        else:

            costs_km = [
                0.0
                for _ in capacities
            ]

            fixed_costs = [
                0.0
                for _ in capacities
            ]


        # REGRAS OPERACIONAIS


        allow_multi = bool(
            desc.get(
                "allow_multi_trip",
                False
            )
        )

        max_time = float(
            desc.get(
                "max_time_per_vehicle_minutes",
                480
            )
        )

        service_time = float(
            desc.get(
                "service_time_per_delivery_minutes",
                0
            )
        )

        avg_speed = float(
            desc.get(
                "average_vehicle_speed_kmh",
                60
            )
        )

        if avg_speed <= 0:

            avg_speed = 60.0


        # TEMPO MÍNIMO


        has_min_usage = bool(
            desc.get(
                "has_min_usage_time",
                False
            )
        )

        min_usage = float(
            desc.get(
                "min_usage_time_minutes",
                0
            )
        )


        # REGRA DE UTILIZAÇÃO DA FROTA
        # must_use_all_vehicles e has_min_usage são independentes.

        must_use_raw = desc.get("must_use_all_vehicles", desc.get("must_use_all", False))
        if isinstance(must_use_raw, str):
            must_use_all_vehicles = must_use_raw.strip().lower() in ("1", "true", "sim", "s", "yes", "y")
        else:
            must_use_all_vehicles = bool(must_use_raw)


        # CUSTO FIXO


        fixed_cost_type = str(
            desc.get(
                "fixed_cost_type",
                "ativacao"
            )
        ).lower()

        if fixed_cost_type in [
            "activation",
            "activated",
            "uso",
            "utilizacao",
            "utilização",
        ]:

            fixed_cost_type = "ativacao"

        elif fixed_cost_type in [
            "always",
            "unconditional",
            "incondicional",
        ]:

            fixed_cost_type = (
                "incondicional"
            )

        else:

            fixed_cost_type = "ativacao"


        # INÍCIO DA OPERAÇÃO


        if "vehicle_start_time_minutes" in desc:
            vehicle_start = float(desc["vehicle_start_time_minutes"])
        elif "vehicle_start_time" in desc:
            vehicle_start = self._parse_to_min(desc["vehicle_start_time"])
        elif "start_time" in desc:
            vehicle_start = self._parse_to_min(desc["start_time"])
        elif janela_inicio:
            vehicle_start = min(janela_inicio.values())
        else:
            vehicle_start = 0.0

        vehicle_end = float(
            desc.get(
                "vehicle_end_time_minutes",
                vehicle_start + max_time
            )
        )
        if vehicle_end <= vehicle_start:
            raise ValueError(
                "vehicle_end_time_minutes deve ser maior que vehicle_start_time_minutes."
            )


        # CONFIGURAÇÃO MULTIOBJETIVO E NORMALIZAÇÃO ROBUSTA

        # As escalas representam limites superiores plausíveis do problema,
        # e não apenas a soma arbitrária das distâncias da matriz. Isso evita
        # que um objetivo fique artificialmente dominante ou irrelevante.
        multiobjective_config = desc.get('multiobjective_config', {'enabled': False, 'objectives': [], 'weights': {}}) or {}
        coords = {int(cid): (float(row['latitude']), float(row['longitude'])) for cid, row in client_lookup.items()}
        ids = list(coords.keys())
        max_pair_distance = 0.0
        for i in ids:
            for j in ids:
                if i != j:
                    max_pair_distance = max(max_pair_distance, self._haversine_distance(coords[i][0], coords[i][1], coords[j][0], coords[j][1]))
        client_count = max(1, len([c for c in client_lookup if c != depot_id]))
        distance_scale = max(1.0, 2.0 * client_count * max_pair_distance)
        time_scale = max(1.0, max_time * max(1, len(capacities)))
        cost_scale = max(1.0, distance_scale * max(costs_km or [1.0]) + (sum(fixed_costs) if fixed_cost_type == 'incondicional' else max(fixed_costs or [0.0]) * max(1, len(capacities))))
        vehicle_scale = max(1.0, float(len(capacities)))



        # RETORNO


        return {

            "client_lookup":
                client_lookup,

            "distance_matrix":
                distance_matrix,

            "depot_id":
                depot_id,

            "service_time":
                service_time,

            "avg_speed":
                avg_speed,

            "vehicle_capacities":
                capacities,

            "vehicle_costs_km":
                costs_km,

            "vehicle_fixed_costs":
                fixed_costs,

            "cost_data_available":
                bool(df_custos is not None and not df_custos.empty),

            "max_time":
                max_time,

            "vehicle_start_time":
                vehicle_start,

            "vehicle_end_time":
                vehicle_end,

            "has_time_windows":
                has_time_windows,

            "time_window_mode":
                time_window_mode,

            "janela_map_inicio":
                janela_inicio,

            "janela_map_fim":
                janela_fim,

            "allow_multi":
                allow_multi,

            "has_min_usage":
                has_min_usage,

            "min_usage_val":
                min_usage,

            "fixed_cost_type":
                fixed_cost_type,

            "must_use_all_vehicles":
                must_use_all_vehicles,
            "multiobjective_config": multiobjective_config,
            "objective": desc.get("objective","DISTANCE"),
            "objective_scales": {
                "DISTANCE": distance_scale, "TIME": time_scale,
                "COST": cost_scale, "VEHICLES": vehicle_scale
            },
        }


    # DISTÂNCIA ENTRE CLIENTES


    def _distance(
        self,
        a,
        b,
        data
    ):

        # Caminho rápido: matriz Haversine pré-calculada.
        matrix = data.get("distance_matrix")
        if matrix is not None:
            try:
                return float(matrix[int(a)][int(b)])
            except (KeyError, TypeError, ValueError):
                pass

        # Alternativa para chamadas sem o contrato completo.
        pa = data[
            "client_lookup"
        ][a]

        pb = data[
            "client_lookup"
        ][b]

        return self._haversine_distance(
            pa["latitude"],
            pa["longitude"],
            pb["latitude"],
            pb["longitude"]
        )


    # AVALIA UMA VIAGEM


    def _evaluate_trip(
        self,
        path,
        data,
        start_clock
    ):

        depot = data[
            "depot_id"
        ]

        if len(path) < 2:

            return {
                "feasible": True,
                "distance": 0.0,
                "work_time": 0.0,
                "arrival_times": [],
                "service_start_times": [],
                "loads": [],
                "waiting_time": 0.0,
                "window_violation": 0.0,
                "capacity_violation": 0.0,
            }

        current_time = float(
            start_clock
        )

        distance = 0.0
        waiting = 0.0
        window_violation = 0.0

        arrival_times = []
        service_start_times = []
        loads = []

        current_load = 0.0


        # PERCORRE A VIAGEM


        for i in range(
            1,
            len(path)
        ):

            previous = path[
                i - 1
            ]

            current = path[i]

            dist = self._distance(
                previous,
                current,
                data
            )

            distance += dist

            travel_time = (
                dist
                / data["avg_speed"]
            ) * 60.0

            arrival = (
                current_time
                + travel_time
            )


            # DEPÓSITO


            if current == depot:

                current_time = arrival

                arrival_times.append(
                    current_time
                )

                service_start_times.append(
                    current_time
                )

                loads.append(
                    current_load
                )

                continue


            # JANELA


            window_start = (
                data[
                    "janela_map_inicio"
                ].get(
                    current,
                    0.0
                )
            )

            window_end = (
                data[
                    "janela_map_fim"
                ].get(
                    current,
                    1440.0
                )
            )


            # ESPERA


            if arrival < window_start:

                waiting_here = (
                    window_start
                    - arrival
                )

                waiting += (
                    waiting_here
                )

                service_start = (
                    window_start
                )

            else:

                service_start = arrival


            # ATRASO


            if service_start > window_end:

                window_violation += (
                    service_start
                    - window_end
                )


            # SERVIÇO


            current_time = (
                service_start
                + data[
                    "service_time"
                ]
            )

            demand = float(
                data[
                    "client_lookup"
                ][current]["Demanda"]
            )

            current_load += demand

            arrival_times.append(
                arrival
            )

            service_start_times.append(
                service_start
            )

            loads.append(
                current_load
            )


        # TEMPO TOTAL DA VIAGEM


        work_time = (
            current_time
            - start_clock
        )

        return {

            "feasible":
                window_violation
                <= 1e-9,

            "distance":
                distance,

            "work_time":
                work_time,

            "arrival_times":
                arrival_times,

            "service_start_times":
                service_start_times,

            "loads":
                loads,

            "waiting_time":
                waiting,

            "window_violation":
                window_violation,

            "capacity_violation":
                0.0,
        }


    # CAPACIDADE DA VIAGEM


    def _trip_demand(
        self,
        trip,
        data
    ):

        depot = data[
            "depot_id"
        ]

        total = 0.0

        for client in trip:

            if client == depot:
                continue

            total += float(
                data[
                    "client_lookup"
                ][client]["Demanda"]
            )

        return total


    # DISTÂNCIA DA VIAGEM


    def _trip_distance(
        self,
        trip,
        data
    ):

        if len(trip) < 2:
            return 0.0

        total = 0.0

        for i in range(
            len(trip) - 1
        ):

            total += self._distance(
                trip[i],
                trip[i + 1],
                data
            )

        return total


    # REAGENDAMENTO COMPLETO DO VEÍCULO



    # Sempre que uma viagem muda, todas as viagens seguintes
    # são recalculadas.


    def _best_trip_start(self, path, data, earliest_start):
        """Escolhe um início factível que minimize espera improdutiva."""
        earliest_start = float(earliest_start)
        candidates = {earliest_start}
        depot = data["depot_id"]
        raw_elapsed = 0.0

        for i in range(1, len(path)):
            prev_node = path[i - 1]
            node = path[i]
            dist = self._distance(prev_node, node, data)
            raw_elapsed += (dist / data["avg_speed"]) * 60.0

            if node != depot:
                ws = float(data["janela_map_inicio"].get(node, 0.0))
                we = float(data["janela_map_fim"].get(node, 1440.0))
                candidates.add(max(earliest_start, ws - raw_elapsed))
                candidates.add(max(earliest_start, we - raw_elapsed))
                raw_elapsed += float(data["service_time"])

        best = None
        for start in sorted(candidates):
            ev = self._evaluate_trip(path, data, start)
            if ev["window_violation"] > 1e-9:
                continue
            key = (float(ev["work_time"]), float(ev["waiting_time"]), float(start))
            if best is None or key < best[0]:
                best = (key, float(start), ev)

        if best is None:
            ev = self._evaluate_trip(path, data, earliest_start)
            return earliest_start, ev
        return best[1], best[2]

    def _reschedule_vehicle(self, vehicle, data):
        current_clock = float(data["vehicle_start_time"])
        total_work = 0.0

        for trip in vehicle["trips"]:
            start_clock, evaluation = self._best_trip_start(
                trip["path"], data, current_clock
            )
            trip["evaluation"] = evaluation
            trip["start_clock"] = start_clock
            total_work += float(evaluation["work_time"])
            current_clock = start_clock + float(evaluation["work_time"])

        vehicle["total_work"] = total_work
        vehicle["used"] = len(vehicle["trips"]) > 0
        return vehicle


    # VERIFICAÇÃO DE VEÍCULO


    def _vehicle_is_feasible(
        self,
        vehicle,
        data
    ):

        if not vehicle["used"]:
            return True


        # TEMPO TOTAL


        if (
            vehicle["total_work"]
            > data["max_time"] + 1e-9
        ):

            return False


        # HORIZONTE OPERACIONAL

        if vehicle["trips"]:
            last_trip = vehicle["trips"][-1]
            last_end = (
                float(last_trip["start_clock"])
                + float(last_trip["evaluation"]["work_time"])
            )
            if last_end > float(data.get("vehicle_end_time", float("inf"))) + 1e-9:
                return False


        # CADA VIAGEM


        for trip in vehicle[
            "trips"
        ]:

            evaluation = trip[
                "evaluation"
            ]

            if (
                evaluation[
                    "window_violation"
                ]
                > 1e-9
            ):

                return False

            demand = (
                self._trip_demand(
                    trip["path"],
                    data
                )
            )

            capacity = (
                data[
                    "vehicle_capacities"
                ][
                    vehicle[
                        "vehicle_idx"
                    ]
                ]
            )

            if (
                demand
                > capacity + 1e-9
            ):

                return False

        return True


    # MELHOR INSERÇÃO EM VIAGEM


    def _best_insertion_in_trip(
        self,
        trip,
        client,
        vehicle_idx,
        data,
        start_clock
    ):

        best = None

        capacity = (
            data[
                "vehicle_capacities"
            ][vehicle_idx]
        )

        old_distance = (
            self._trip_distance(
                trip,
                data
            )
        )

        # Inserir antes do depósito final.
        for pos in range(
            1,
            len(trip)
        ):
            if self._deadline_reached():
                break

            candidate = (
                trip[:pos]
                + [client]
                + trip[pos:]
            )


            # CAPACIDADE


            demand = (
                self._trip_demand(
                    candidate,
                    data
                )
            )

            if (
                demand
                > capacity + 1e-9
            ):

                continue


            # AVALIAÇÃO


            evaluation = (
                self._evaluate_trip(
                    candidate,
                    data,
                    start_clock
                )
            )


            # TEMPO


            if (
                evaluation[
                    "work_time"
                ]
                > data["max_time"]
                + 1e-9
            ):

                continue


            # JANELA


            if (
                evaluation[
                    "window_violation"
                ]
                > 1e-9
            ):

                continue


            # DISTÂNCIA


            new_distance = (
                evaluation[
                    "distance"
                ]
            )

            delta_distance = (
                new_distance
                - old_distance
            )

            candidate_info = {

                "trip":
                    candidate,

                "evaluation":
                    evaluation,

                "delta_distance":
                    delta_distance,
            }

            if best is None:

                best = candidate_info

            elif (
                delta_distance
                < best[
                    "delta_distance"
                ]
            ):

                best = candidate_info

        return best


    # CRIA NOVA VIAGEM


    def _create_new_trip(
        self,
        client,
        vehicle_idx,
        data,
        start_clock
    ):

        depot = data[
            "depot_id"
        ]

        trip = [
            depot,
            client,
            depot
        ]

        demand = float(
            data[
                "client_lookup"
            ][client]["Demanda"]
        )

        capacity = (
            data[
                "vehicle_capacities"
            ][vehicle_idx]
        )

        if (
            demand
            > capacity + 1e-9
        ):

            return None

        evaluation = (
            self._evaluate_trip(
                trip,
                data,
                start_clock
            )
        )

        if (
            evaluation[
                "work_time"
            ]
            > data["max_time"]
            + 1e-9
        ):

            return None

        if (
            evaluation[
                "window_violation"
            ]
            > 1e-9
        ):

            return None

        return {

            "trip":
                trip,

            "evaluation":
                evaluation,

            "delta_distance":
                evaluation[
                    "distance"
                ],
        }


    # CANDIDATO DE INSERÇÃO EM VEÍCULO

    # Testa a alteração em uma cópia do veículo.
    # Assim conseguimos verificar se a alteração de uma viagem
    # prejudica as viagens posteriores.


    def _evaluate_vehicle_insertion(self, vehicle, client, data):
        # Testa inserção em viagens existentes e também a criação de uma nova viagem.
        vehicle_idx=vehicle["vehicle_idx"]; candidates=[]
        for trip_idx,trip_data in enumerate(vehicle["trips"]):
            if self._deadline_reached(): break
            result=self._best_insertion_in_trip(trip_data["path"],client,vehicle_idx,data,trip_data["start_clock"])
            if result is None: continue
            test_vehicle=copy.deepcopy(vehicle); test_vehicle["trips"][trip_idx]={"path":result["trip"],"evaluation":result["evaluation"],"start_clock":trip_data["start_clock"]}; self._reschedule_vehicle(test_vehicle,data)
            if not self._vehicle_is_feasible(test_vehicle,data): continue
            old_distance=sum(float(t["evaluation"]["distance"]) if isinstance(t.get("evaluation"), dict) and "distance" in t["evaluation"] else self._trip_distance(t["path"], data) for t in vehicle["trips"]); new_distance=sum(float(t["evaluation"]["distance"]) if isinstance(t.get("evaluation"), dict) and "distance" in t["evaluation"] else self._trip_distance(t["path"], data) for t in test_vehicle["trips"]); candidates.append({"type":"existing_trip","vehicle":test_vehicle,"delta_distance":new_distance-old_distance,"delta_work":test_vehicle["total_work"]-vehicle["total_work"]})
        if data.get("allow_multi",False):
            test_vehicle=copy.deepcopy(vehicle); start_clock=data["vehicle_start_time"]+vehicle["total_work"]; result=self._create_new_trip(client,vehicle_idx,data,start_clock)
            if result is not None:
                test_vehicle["trips"].append({"path":result["trip"],"evaluation":result["evaluation"],"start_clock":start_clock}); self._reschedule_vehicle(test_vehicle,data)
                if self._vehicle_is_feasible(test_vehicle,data): candidates.append({"type":"new_trip","vehicle":test_vehicle,"delta_distance":result["delta_distance"],"delta_work":test_vehicle["total_work"]-vehicle["total_work"]})
        if not candidates: return None
        return min(candidates,key=lambda c:self._score_decoder_candidate(c,vehicle,data))

    def _score_decoder_candidate(self, candidate, vehicle, data):
        """Pontua uma inserção sem alterar a lógica construtiva histórica."""
        v_idx = vehicle["vehicle_idx"]
        delta_distance = float(candidate.get("delta_distance", 0.0))
        delta_time = float(candidate.get("delta_work", 0.0))
        activates = 0.0 if vehicle.get("used", False) else 1.0
        fixed_delta = (
            float(data["vehicle_fixed_costs"][v_idx])
            if activates and data.get("fixed_cost_type") == "ativacao"
            else 0.0
        )
        delta_cost = (
            delta_distance * float(data["vehicle_costs_km"][v_idx])
            + fixed_delta
        )
        components = {
            "DISTANCE": delta_distance,
            "TIME": delta_time,
            "COST": delta_cost,
            "VEHICLES": activates,
        }

        canonical = self._canonical_objective_details(components)
        if canonical is not None:
            score, _, context = canonical
            active = context["active_objectives"]
            fleet_tiebreak = (
                1
                if activates and ("VEHICLES" in active or "COST" in active)
                else 0
            )
            return (
                score,
                fleet_tiebreak,
                delta_cost,
                delta_time,
                delta_distance,
            )

        # Alternativa para dados sem objective_context.
        config = data.get("multiobjective_config", {}) or {}
        objectives = [
            str(x).upper()
            for x in (config.get("objectives", []) or [])
        ]
        weights = {
            str(k).upper(): float(v)
            for k, v in (config.get("weights", {}) or {}).items()
        }
        objective = self._normalize_objective(data.get("objective", "DISTANCE"))
        scales = data.get("objective_scales", {}) or {}
        multi = bool(config.get("enabled", False) or len(objectives) > 1)

        if not objectives:
            objectives = [objective if objective != "MULTI" else "DISTANCE"]

        if multi:
            positive = {
                k: max(0.0, float(v))
                for k, v in weights.items()
                if max(0.0, float(v)) > 0.0
            }
            total = sum(positive.values())
            weights = (
                {k: v / total for k, v in positive.items()}
                if total > 0
                else {objectives[0]: 1.0}
            )
        else:
            weights = {objectives[0]: 1.0}

        score = sum(
            float(weight)
            * float(components.get(obj, 0.0))
            / max(1e-9, float(scales.get(obj, 1.0)))
            for obj, weight in weights.items()
        )
        fleet_tiebreak = (
            1
            if activates and ("VEHICLES" in weights or "COST" in weights)
            else 0
        )
        return (
            score,
            fleet_tiebreak,
            delta_cost,
            delta_time,
            delta_distance,
        )


    # DECODER MULTI-TRIP



    # DIRETIVAS DO AGENTE DE CONTINUIDADE

    def _get_continuity_directive(self, data=None):
        """
         lê a diretiva de Continuidade com isolamento por execução.

        Prioridade:
        1. diretiva imutável/local da chamada (data["_continuity_directive"]);
        2. SharedMemory apenas como fallback de compatibilidade.

        Isso evita colisão entre Tracks A/B ou execuções concorrentes, nas quais
        a SharedMemory pode já conter a diretiva de outra hipótese.
        """
        if isinstance(data, dict):
            local_directive = data.get("_continuity_directive")
            if isinstance(local_directive, dict):
                return copy.deepcopy(local_directive)

        mem = getattr(self.shared_memory, "data", {}) or {}
        directive = mem.get("continuity_search_directive")
        return copy.deepcopy(directive) if isinstance(directive, dict) else {}

    def _continuity_target_max_vehicles(self, data):
        directive = self._get_continuity_directive(data)
        raw = directive.get("target_max_vehicles")
        try:
            value = int(raw)
        except Exception:
            return None
        total = len(data.get("vehicle_capacities", []) or [])
        if total <= 0:
            return None
        return max(1, min(total, value))

    def _continuity_strategy_type(self, data):
        return str(self._get_continuity_directive(data).get("type", "") or "").upper()

    def _hard_k_allowed_vehicle_indices(self, chromosome, data):
        """
         escolhe FISICAMENTE o subconjunto de veículos disponível ao
        decoder durante FLEET_REDUCTION(target_max_vehicles=K).

        Fora de FLEET_REDUCTION retorna todos os veículos.

        A escolha é determinística por cromossomo para manter o fitness estável:
        cromossomos diferentes exploram subconjuntos físicos diferentes, enquanto
        o mesmo cromossomo sempre enxerga o mesmo subconjunto.

        Não existe K=2 hard-coded: K vem da diretiva de Continuidade.
        """
        total = len(data.get("vehicle_capacities", []) or [])
        all_indices = list(range(total))

        strategy = self._continuity_strategy_type(data)
        target_k = self._continuity_target_max_vehicles(data)

        if (
            strategy != "FLEET_REDUCTION"
            or target_k is None
            or target_k >= total
        ):
            return all_indices

        if data.get("must_use_all_vehicles", False):
            # Restrição original do problema prevalece sobre a diretiva de busca.
            return all_indices

        target_k = max(1, min(int(target_k), total))

        capacities = list(data.get("vehicle_capacities", []) or [])
        costs_km = list(data.get("vehicle_costs_km", []) or [])
        fixed_costs = list(data.get("vehicle_fixed_costs", []) or [])

        def _safe(seq, idx, default):
            try:
                return float(seq[idx])
            except Exception:
                return float(default)

        # Ranking estrutural: favorece veículos capazes, baratos por km e com
        # custo fixo baixo. Serve apenas como ordem-base; não elimina os demais.
        max_cap = max([_safe(capacities, i, 0.0) for i in all_indices] or [1.0])
        max_ckm = max([_safe(costs_km, i, 0.0) for i in all_indices] or [1.0])
        max_fix = max([_safe(fixed_costs, i, 0.0) for i in all_indices] or [1.0])
        max_cap = max(1e-9, max_cap)
        max_ckm = max(1e-9, max_ckm)
        max_fix = max(1e-9, max_fix)

        ranked = sorted(
            all_indices,
            key=lambda i: (
                -(_safe(capacities, i, 0.0) / max_cap),
                (_safe(costs_km, i, 0.0) / max_ckm),
                (_safe(fixed_costs, i, 0.0) / max_fix),
                i,
            )
        )

        # Assinatura estável do cromossomo (não usa hash() do Python).
        # Permutações diferentes espalham a população por subconjuntos distintos.
        signature = 2166136261
        for pos, client in enumerate(chromosome):
            try:
                value = int(client)
            except Exception:
                value = pos + 1
            signature ^= (value + 131 * (pos + 1)) & 0xFFFFFFFF
            signature = (signature * 16777619) & 0xFFFFFFFF

        # Parte da população testa deliberadamente o conjunto estruturalmente
        # mais promissor; o restante explora outras combinações físicas.
        mode = signature % 5

        if mode == 0:
            allowed = ranked[:target_k]
        elif mode == 1:
            # Maior capacidade.
            allowed = sorted(
                all_indices,
                key=lambda i: (-_safe(capacities, i, 0.0), i)
            )[:target_k]
        elif mode == 2:
            # Menor custo fixo + variável.
            allowed = sorted(
                all_indices,
                key=lambda i: (
                    _safe(fixed_costs, i, 0.0),
                    _safe(costs_km, i, 0.0),
                    -_safe(capacities, i, 0.0),
                    i,
                )
            )[:target_k]
        else:
            # Exploração determinística: rotação do ranking para que cromossomos
            # diferentes possam testar pares/subconjuntos diferentes.
            offset = signature % total
            rotated = ranked[offset:] + ranked[:offset]
            step = 1 + ((signature // max(1, total)) % max(1, total - 1))
            picked = []
            cursor = 0
            visited = set()
            while len(picked) < target_k and len(visited) < total:
                idx = rotated[cursor % total]
                visited.add(idx)
                if idx not in picked:
                    picked.append(idx)
                cursor = (cursor + step) % total
                if cursor == 0 and len(picked) < target_k:
                    # fallback para evitar ciclos quando step e total não são coprimos
                    for idx2 in rotated:
                        if idx2 not in picked:
                            picked.append(idx2)
                            if len(picked) >= target_k:
                                break
                    break
            allowed = picked[:target_k]

        # Segurança contratual.
        allowed = list(dict.fromkeys(int(i) for i in allowed))
        if len(allowed) > target_k:
            allowed = allowed[:target_k]
        if not allowed:
            allowed = ranked[:target_k]

        return allowed

    def _decode_chromosome(self, chromosome, data):
        """
        Decoder Multi-Trip  — PHYSICAL HARD-K.

        Em FLEET_REDUCTION(K), o decoder NÃO cria primeiro uma solução com toda
        a frota para depois verificar used<=K. Ele seleciona um subconjunto
        físico de no máximo K veículos e SOMENTE esses veículos participam da
        construção. Os demais permanecem materializados no resultado, porém
        indisponíveis e obrigatoriamente ociosos.

        Consequência:
            used_vehicles <= K por construção.

        Se os K veículos permitidos não conseguirem atender todos os clientes,
        esses clientes ficam em `unassigned`; a solução recebe penalidade e não
        pode virar SUCCESS. Isso é NOT_FOUND heurístico, nunca INFEASIBLE.
        """
        num_vehicles = len(data["vehicle_capacities"])
        vehicles = [
            {
                "vehicle_idx": v,
                "trips": [],
                "total_work": 0.0,
                "used": False,
                "hard_k_allowed": True,
            }
            for v in range(num_vehicles)
        ]

        unassigned = []
        remaining_clients = list(chromosome)

        directive = self._get_continuity_directive(data)
        strategy_type = str(directive.get("type", "") or "").upper()
        target_max = self._continuity_target_max_vehicles(data)

        hard_k_active = (
            strategy_type == "FLEET_REDUCTION"
            and target_max is not None
            and not data.get("must_use_all_vehicles", False)
        )

        if hard_k_active:
            allowed_indices = self._hard_k_allowed_vehicle_indices(
                chromosome, data
            )
            allowed_set = set(allowed_indices)

            if len(allowed_set) > int(target_max):
                raise RuntimeError(
                    "CONTRACT_VIOLATION: decoder HARD-K liberou "
                    f"{len(allowed_set)} veículos para K={target_max}"
                )

            for vehicle in vehicles:
                vehicle["hard_k_allowed"] = (
                    vehicle["vehicle_idx"] in allowed_set
                )

        else:
            allowed_indices = list(range(num_vehicles))
            allowed_set = set(allowed_indices)

        # must_use_all é restrição ORIGINAL do problema e prevalece.
        if data["must_use_all_vehicles"]:
            target_max = None
            allowed_indices = list(range(num_vehicles))
            allowed_set = set(allowed_indices)

            for vehicle in vehicles:
                vehicle["hard_k_allowed"] = True

            for vehicle_idx in allowed_indices:
                vehicle = vehicles[vehicle_idx]

                if not remaining_clients:
                    break

                seeds = []

                for client in remaining_clients:
                    result = self._create_new_trip(
                        client,
                        vehicle["vehicle_idx"],
                        data,
                        data["vehicle_start_time"],
                    )

                    if result is not None:
                        seeds.append(
                            (result["delta_distance"], client, result)
                        )

                if seeds:
                    seeds.sort(key=lambda x: x[0])
                    _, client, result = random.choice(
                        seeds[: min(3, len(seeds))]
                    )

                    vehicle["trips"].append(
                        {
                            "path": result["trip"],
                            "evaluation": result["evaluation"],
                            "start_clock": data["vehicle_start_time"],
                        }
                    )

                    vehicle["used"] = True
                    self._reschedule_vehicle(vehicle, data)
                    remaining_clients.remove(client)


        # CONSTRUÇÃO

        for client in remaining_clients:
            candidates = []
            active_candidates = []


            for vehicle_idx in allowed_indices:
                vehicle = vehicles[vehicle_idx]

                if hard_k_active and not vehicle.get(
                    "hard_k_allowed", False
                ):
                    raise RuntimeError(
                        "CONTRACT_VIOLATION: veículo fora do subconjunto "
                        "HARD-K alcançou o decoder."
                    )

                result = self._evaluate_vehicle_insertion(
                    vehicle, client, data
                )

                if result is None:
                    continue

                item = {
                    "vehicle_idx": vehicle["vehicle_idx"],
                    "result": result,
                    "score": self._score_decoder_candidate(
                        result, vehicle, data
                    ),
                }

                candidates.append(item)

                if vehicle.get("used", False):
                    active_candidates.append(item)

            if not candidates:
                unassigned.append(client)
                continue

            config = data.get("multiobjective_config", {}) or {}
            objectives = [
                str(x).upper()
                for x in (config.get("objectives", []) or [])
            ]
            weights = {
                str(k).upper(): float(v)
                for k, v in (config.get("weights", {}) or {}).items()
            }
            fleet_weight = weights.get("VEHICLES", 0.0)

            # Em HARD-K reutilizar veículo já aberto continua sendo prioridade,
            # pois isso favorece multi-trip antes de consumir outro veículo do K.
            force_active_first = (
                strategy_type == "FLEET_REDUCTION"
                and bool(active_candidates)
            )

            canonical_fleet_first = (
                config.get("enabled", False)
                and fleet_weight > 0
                and bool(active_candidates)
            )

            if force_active_first or canonical_fleet_first:
                active_candidates.sort(key=lambda x: x["score"])
                selected_pool = active_candidates[
                    : min(3, len(active_candidates))
                ]
            else:
                candidates.sort(key=lambda x: x["score"])
                selected_pool = candidates[: min(3, len(candidates))]

            selected = (
                selected_pool[0]
                if data.get("deterministic_decoder", False)
                else (
                    random.choice(selected_pool)
                    if random.random() < 0.10
                    else selected_pool[0]
                )
            )

            selected_idx = int(selected["vehicle_idx"])

            if hard_k_active and selected_idx not in allowed_set:
                raise RuntimeError(
                    "CONTRACT_VIOLATION: tentativa de ativar veículo "
                    f"{selected_idx} fora de HARD-K={sorted(allowed_set)}"
                )

            vehicle = vehicles[selected_idx]
            vehicle.clear()
            vehicle.update(selected["result"]["vehicle"])
            vehicle["hard_k_allowed"] = True

        # Reagenda apenas os veículos que participaram da construção.
        for vehicle_idx in allowed_indices:
            self._reschedule_vehicle(vehicles[vehicle_idx], data)

        # Veículos bloqueados permanecem obrigatoriamente vazios.
        if hard_k_active:
            for vehicle in vehicles:
                if vehicle["vehicle_idx"] not in allowed_set:
                    if vehicle.get("trips") or vehicle.get("used", False):
                        raise RuntimeError(
                            "CONTRACT_VIOLATION: veículo bloqueado por HARD-K "
                            "foi ativado."
                        )
                    vehicle["trips"] = []
                    vehicle["total_work"] = 0.0
                    vehicle["used"] = False
                    vehicle["hard_k_allowed"] = False

            actual_used = sum(
                1 for v in vehicles if v.get("used", False)
            )

            if actual_used > int(target_max):
                raise RuntimeError(
                    "CONTRACT_VIOLATION: decoder terminou com "
                    f"{actual_used} veículos para K={target_max}"
                )

        return {
            "vehicles": vehicles,
            "unassigned": unassigned,
            "continuity_strategy": strategy_type or None,
            "continuity_target_max_vehicles": target_max,
            "hard_k_active": bool(hard_k_active),
            "hard_k_allowed_vehicle_indices": list(allowed_indices),
            "hard_k_allowed_vehicle_ids": [
                int(i) + 1 for i in allowed_indices
            ],
        }


    # FITNESS


    def _calculate_solution_metrics(
        self,
        decoded,
        data
    ):

        total_distance_cost = 0.0
        total_fixed_cost = 0.0

        total_window_violation = 0.0
        total_capacity_violation = 0.0
        total_time_violation = 0.0
        total_min_usage_violation = 0.0

        used_vehicles = 0


        # VEÍCULOS


        for vehicle in decoded[
            "vehicles"
        ]:

            v_idx = vehicle[
                "vehicle_idx"
            ]


            # VEÍCULO NÃO UTILIZADO


            if not vehicle[
                "used"
            ]:

                continue

            used_vehicles += 1


            # CUSTO FIXO POR ATIVAÇÃO


            if (
                data[
                    "fixed_cost_type"
                ]
                == "ativacao"
            ):

                total_fixed_cost += (
                    data[
                        "vehicle_fixed_costs"
                    ][v_idx]
                )


            # VIAGENS


            for trip in vehicle[
                "trips"
            ]:

                evaluation = trip[
                    "evaluation"
                ]


                # CUSTO KM


                total_distance_cost += (
                    evaluation[
                        "distance"
                    ]
                    *
                    data[
                        "vehicle_costs_km"
                    ][v_idx]
                )


                # JANELA


                total_window_violation += (
                    evaluation[
                        "window_violation"
                    ]
                )


                # CAPACIDADE


                demand = (
                    self._trip_demand(
                        trip["path"],
                        data
                    )
                )

                capacity = (
                    data[
                        "vehicle_capacities"
                    ][v_idx]
                )

                if (
                    demand
                    > capacity + 1e-9
                ):

                    total_capacity_violation += (
                        demand - capacity
                    )


            # TEMPO TOTAL DO VEÍCULO


            work_time = (
                vehicle[
                    "total_work"
                ]
            )

            if (
                work_time
                > data["max_time"]
                + 1e-9
            ):

                total_time_violation += (
                    work_time
                    - data["max_time"]
                )


            # TEMPO MÍNIMO

            # IMPORTANTE:
            # 0 minutos também é violação quando o mínimo
            # está ativo.


            if data["has_min_usage"] and vehicle["used"]:
                # O mínimo só é exigido de veículos efetivamente utilizados.
                minimum = float(data["min_usage_val"])
                if work_time < minimum - 1e-9:
                    total_min_usage_violation += minimum - work_time


        # CUSTO FIXO INCONDICIONAL


        if (
            data[
                "fixed_cost_type"
            ]
            == "incondicional"
        ):

            total_fixed_cost = sum(
                data[
                    "vehicle_fixed_costs"
                ]
            )


        # VEÍCULOS OBRIGATÓRIOS


        vehicle_usage_violation = 0

        if data[
            "must_use_all_vehicles"
        ]:

            total_vehicles = len(
                data[
                    "vehicle_capacities"
                ]
            )

            vehicle_usage_violation = max(
                0,
                total_vehicles
                - used_vehicles
            )


        # CLIENTES NÃO ATENDIDOS


        expected_clients = {int(c) for c in data["client_lookup"].keys() if int(c) != int(data["depot_id"])}
        served_sequence = [
            int(c)
            for vehicle in decoded.get("vehicles", [])
            for trip in vehicle.get("trips", [])
            for c in trip.get("path", [])
            if int(c) != int(data["depot_id"])
        ]
        served_set = set(served_sequence)
        missing_client_ids = sorted(expected_clients - served_set)
        duplicate_client_ids = sorted({c for c in served_sequence if served_sequence.count(c) > 1})
        unexpected_client_ids = sorted(served_set - expected_clients)
        # Não confiamos apenas em decoded["unassigned"]: a cobertura é reconstruída
        # diretamente das rotas, evitando SUCCESS falso em caches/sementes especiais.
        missing_clients = len(missing_client_ids)
        duplicate_clients = len(duplicate_client_ids)
        unexpected_clients = len(unexpected_client_ids)


        # MÉTRICAS AGREGADAS PARA OBJETIVOS

        total_distance=sum(float(trip.get('evaluation',{}).get('distance',0.0)) for v in decoded['vehicles'] if v.get('used',False) for trip in v.get('trips',[]))
        total_time=sum(float(v.get('total_work',0.0)) for v in decoded['vehicles'] if v.get('used',False))


        # CUSTO REAL


        real_cost = (
            total_distance_cost
            +
            total_fixed_cost
        )


        # PENALIDADES

        # Hierarquia:

        # 1. clientes não atendidos
        # 2. veículos obrigatórios
        # 3. janela
        # 4. capacidade
        # 5. tempo disponível
        # 6. tempo mínimo
        # 7. custo


        penalty = 0.0


        # 1. CLIENTES NÃO ATENDIDOS


        penalty += (
            missing_clients
            * 1_000_000_000
        )


        # 2. VEÍCULOS OBRIGATÓRIOS


        penalty += (
            vehicle_usage_violation
            * 100_000_000
        )


        # 3. JANELAS


        penalty += (
            total_window_violation
            * 10_000_000
        )


        # 4. CAPACIDADE


        penalty += (
            total_capacity_violation
            * 10_000_000
        )


        # 5. TEMPO DISPONÍVEL


        penalty += (
            total_time_violation
            * 10_000_000
        )


        # 6. TEMPO MÍNIMO


        if data[
            "has_min_usage"
        ]:

            penalty += (
                total_min_usage_violation
                * 10_000_000
            )


        # FITNESS


        objective_input={'total_distance':total_distance,'total_time':total_time,'real_cost':real_cost,'used_vehicles':used_vehicles}
        multiobjective_value=self._calculate_multiobjective_value(objective_input,data)
        objective_components=objective_input.get('objective_components',{})
        objective_normalized=objective_input.get('objective_normalized',{})
        objective_weights=objective_input.get('objective_weights',{})
        active_objectives=objective_input.get('active_objectives',[])
        objective_value=real_cost if multiobjective_value is None else multiobjective_value
        fitness = objective_value + penalty


        # FACTIBILIDADE


        feasible = (

            missing_clients == 0

            and duplicate_clients == 0

            and unexpected_clients == 0

            and
            vehicle_usage_violation
            == 0

            and
            total_window_violation
            <= 1e-9

            and
            total_capacity_violation
            <= 1e-9

            and
            total_time_violation
            <= 1e-9

            and
            total_min_usage_violation
            <= 1e-9
        )

        return {

            "fitness":
                fitness,

            "real_cost":
                real_cost,
            "total_distance": total_distance,
            "total_time": total_time,
            "objective_value": objective_value,

            "feasible":
                feasible,

            "distance_cost":
                total_distance_cost,

            "fixed_cost":
                total_fixed_cost,

            "window_violation":
                total_window_violation,

            "capacity_violation":
                total_capacity_violation,

            "time_violation":
                total_time_violation,

            "min_usage_violation":
                total_min_usage_violation,

            "vehicle_usage_violation":
                vehicle_usage_violation,

            "missing_clients":
                missing_clients,
            "missing_client_ids": missing_client_ids,
            "duplicate_clients": duplicate_clients,
            "duplicate_client_ids": duplicate_client_ids,
            "unexpected_client_ids": unexpected_client_ids,

            "used_vehicles":
                used_vehicles,

            # Dados explícitos dos objetivos usados pelo restante do sistema.
            "objective_components": objective_components if "objective_components" in locals() else {},
            "objective_normalized": objective_normalized if "objective_normalized" in locals() else {},
            "objective_weights": objective_weights if "objective_weights" in locals() else {},
            "active_objectives": active_objectives if "active_objectives" in locals() else [],
        }


    # OBJETIVO MULTIOBJETIVO

    def _calculate_multiobjective_value(self, metrics, data):
        components = {
            "DISTANCE": float(metrics.get("total_distance", 0.0)),
            "TIME": float(metrics.get("total_time", 0.0)),
            "COST": float(metrics.get("real_cost", 0.0)),
            "VEHICLES": float(metrics.get("used_vehicles", 0)),
        }

        canonical = self._canonical_objective_details(components)
        if canonical is not None:
            score, normalized, context = canonical
            metrics["objective_components"] = components
            metrics["objective_normalized"] = normalized
            metrics["objective_weights"] = dict(context.get("weights", {}))
            metrics["active_objectives"] = list(context["active_objectives"])
            metrics["objective_context_version"] = context.get("version")
            return score

        # Alternativa de compatibilidade para chamadas sem configuração explícita.
        config = data.get("multiobjective_config", {}) or {}
        objectives = [
            str(x).upper()
            for x in (config.get("objectives", []) or [])
        ]
        raw_weights = {
            str(k).upper(): float(v)
            for k, v in (config.get("weights", {}) or {}).items()
        }
        scales = data.get("objective_scales", {}) or {}
        objective = self._normalize_objective(data.get("objective", "DISTANCE"))
        multi = bool(config.get("enabled", False) or len(objectives) > 1)

        if not objectives:
            objectives = [objective if objective != "MULTI" else "DISTANCE"]

        if multi:
            weights = {
                k: max(0.0, v)
                for k, v in raw_weights.items()
                if max(0.0, v) > 0.0
            }
            total_weight = sum(weights.values())
            weights = (
                {k: v / total_weight for k, v in weights.items()}
                if total_weight > 0
                else {objectives[0]: 1.0}
            )
        else:
            weights = {objectives[0]: 1.0}

        normalized = {}
        for key, value in components.items():
            scale = max(1e-9, float(scales.get(key, 1.0)))
            if key == "VEHICLES" and scale > 1.0:
                normalized[key] = max(0.0, (value - 1.0) / (scale - 1.0))
            else:
                normalized[key] = max(0.0, value / scale)

        metrics["objective_components"] = components
        metrics["objective_normalized"] = normalized
        metrics["objective_weights"] = weights
        metrics["active_objectives"] = objectives

        if not multi:
            return components.get(objectives[0], 0.0)

        return sum(
            float(weights.get(k, 0.0)) * normalized.get(k, 0.0)
            for k in objectives
        )


    # FITNESS PÚBLICO


    def _calculate_fitness(self, chromosome, data, return_decoded=False):
        key = tuple(chromosome)

        cached_fitness = self._fitness_cache.get(key)
        if cached_fitness is not None:
            fitness, metrics, decoded = cached_fitness
            if return_decoded:
                return (
                    float(fitness),
                    copy.deepcopy(metrics),
                    copy.deepcopy(decoded),
                )
            return float(fitness)

        special = self._special_decoded_cache.get(key)
        decoded = (
            copy.deepcopy(special)
            if special is not None
            else self._decode_chromosome(chromosome, data)
        )
        metrics = self._calculate_solution_metrics(decoded, data)
        fitness = float(metrics["fitness"])

        self._fitness_cache[key] = (
            fitness,
            copy.deepcopy(metrics),
            copy.deepcopy(decoded),
        )

        if return_decoded:
            return fitness, copy.deepcopy(metrics), copy.deepcopy(decoded)
        return fitness


    # SOLUÇÃO INICIAL


    def _generate_initial_solution(self, data, strategy='random'):
        # Mantém as estratégias existentes e acrescenta a estratégia de frota mínima.
        clients = [c for c in data['client_lookup'].keys() if c != data['depot_id']]
        if strategy == 'random': random.shuffle(clients)
        elif strategy == 'window': clients.sort(key=lambda c: (data['janela_map_inicio'].get(c, 0.0), data['janela_map_fim'].get(c, 1440.0)))
        elif strategy == 'demand': clients.sort(key=lambda c: float(data['client_lookup'][c]['Demanda']), reverse=True)
        elif strategy == 'distance': clients.sort(key=lambda c: self._distance(data['depot_id'], c, data))
        elif strategy == 'hybrid': clients.sort(key=lambda c: (data['janela_map_fim'].get(c, 1440.0), -float(data['client_lookup'][c]['Demanda'])))
        elif strategy == 'minimum_fleet':
            seed = self._build_minimum_fleet_seed(data, clients)
            if seed is not None: return seed['chromosome']
        return clients

    def _construct_single_vehicle_solution(self, data, clients, vehicle_idx, attempts=12):
        if not data.get("allow_multi", False):
            return None

        best = None
        base_orders = [
            list(clients),
            sorted(clients, key=lambda c: (
                data["janela_map_inicio"].get(c, 0.0),
                data["janela_map_fim"].get(c, 1440.0)
            )),
            sorted(clients, key=lambda c: data["janela_map_fim"].get(c, 1440.0)),
            sorted(clients, key=lambda c: (
                data["janela_map_fim"].get(c, 1440.0),
                -float(data["client_lookup"][c]["Demanda"])
            )),
        ]

        for attempt in range(max(1, attempts)):
            if self._deadline_reached():
                break
            order = base_orders[attempt % len(base_orders)][:]
            if attempt >= len(base_orders):
                random.shuffle(order)

            vehicle = {"vehicle_idx": vehicle_idx, "trips": [], "total_work": 0.0, "used": False}
            remaining = order[:]

            while remaining:
                if self._deadline_reached():
                    break
                candidates = []
                for client in remaining:
                    result = self._evaluate_vehicle_insertion(vehicle, client, data)
                    if result is not None:
                        candidates.append((self._score_decoder_candidate(result, vehicle, data), client, result))
                if not candidates:
                    break
                candidates.sort(key=lambda x: x[0])
                top = candidates[:min(4, len(candidates))]
                _, client, result = top[0] if attempt == 0 else random.choice(top)
                vehicle = copy.deepcopy(result["vehicle"])
                remaining.remove(client)

            if remaining or not vehicle.get("used", False):
                continue

            self._reschedule_vehicle(vehicle, data)
            if not self._vehicle_is_feasible(vehicle, data):
                continue

            decoded = {
                "vehicles": [
                    {"vehicle_idx": v, "trips": [], "total_work": 0.0, "used": False}
                    for v in range(len(data["vehicle_capacities"]))
                ],
                "unassigned": []
            }
            decoded["vehicles"][vehicle_idx] = vehicle
            metrics = self._calculate_solution_metrics(decoded, data)
            candidate = {
                "vehicle_idx": vehicle_idx,
                "decoded": decoded,
                "metrics": metrics,
                "chromosome": [
                    c for trip in vehicle["trips"] for c in trip["path"]
                    if c != data["depot_id"]
                ]
            }


            if best is None or float(metrics["fitness"]) < float(best["metrics"]["fitness"]) - 1e-12:
                best = candidate

        return best

    def _build_minimum_fleet_seed(self, data, clients=None):
        # Procura primeiro uma solução factível com um único veículo; só depois considera mais veículos.
        clients = list(clients) if clients is not None else [c for c in data['client_lookup'].keys() if c != data['depot_id']]
        if not clients or not data.get('allow_multi', False) or data.get('must_use_all_vehicles', False): return None
        candidates = []
        for vehicle_idx in range(len(data['vehicle_capacities'])):
            if self._deadline_reached():
                break
            result = self._construct_single_vehicle_solution(data, clients, vehicle_idx, attempts=10)
            if result is not None and result['metrics']['feasible']: candidates.append(result)
        if candidates:
            candidates.sort(key=lambda x: (x['metrics']['used_vehicles'], x['metrics']['real_cost'], x['metrics']['total_time']))
            chosen = candidates[0]; key = tuple(chosen['chromosome']); self._special_decoded_cache[key] = copy.deepcopy(chosen['decoded']); self._fitness_cache.pop(key, None); return chosen
        # Se nenhum veículo sozinho comportar todos os clientes, constrói uma solução de frota mínima progressivamente.
        order = sorted(clients, key=lambda c: (data['janela_map_inicio'].get(c, 0.0), data['janela_map_fim'].get(c, 1440.0)))
        return {'chromosome': order, 'decoded': None, 'metrics': {'feasible': False, 'used_vehicles': len(data['vehicle_capacities'])}}

    def _build_minimum_fleet_chromosome(self, data, clients=None):
        # Mantém compatibilidade com chamadas existentes.
        seed = self._build_minimum_fleet_seed(data, clients)
        return seed['chromosome'] if seed is not None else list(clients or [])


    # ORDER CROSSOVER


    def _order_crossover(
        self,
        parent1,
        parent2
    ):

        n = len(
            parent1
        )

        if n < 2:

            return copy.deepcopy(
                parent1
            )

        start, end = sorted(
            random.sample(
                range(n),
                2
            )
        )

        child = [
            None
        ] * n

        child[
            start:end + 1
        ] = parent1[
            start:end + 1
        ]

        remaining = [

            x

            for x in parent2

            if x not in child
        ]

        pointer = 0

        for i in range(n):

            if child[i] is None:

                child[i] = (
                    remaining[
                        pointer
                    ]
                )

                pointer += 1

        return child


    # SWAP


    def _swap_mutation(
        self,
        chromosome
    ):

        child = chromosome[:]

        if len(child) < 2:

            return child

        i, j = random.sample(
            range(len(child)),
            2
        )

        child[i], child[j] = (
            child[j],
            child[i]
        )

        return child


    # RELOCATE


    def _relocate_mutation(
        self,
        chromosome
    ):

        child = chromosome[:]

        if len(child) < 2:

            return child

        i = random.randrange(
            len(child)
        )

        client = child.pop(i)

        j = random.randrange(
            len(child) + 1
        )

        child.insert(
            j,
            client
        )

        return child


    # 2-OPT


    def _two_opt_mutation(
        self,
        chromosome
    ):

        child = chromosome[:]

        if len(child) < 4:

            return child

        i, j = sorted(
            random.sample(
                range(len(child)),
                2
            )
        )

        child[
            i:j + 1
        ] = reversed(
            child[
                i:j + 1
            ]
        )

        return child


    # MUTAÇÃO


    def _mutate(
        self,
        chromosome,
        mutation_rate=0.20
    ):

        child = chromosome[:]

        if (
            random.random()
            > mutation_rate
        ):

            return child

        operator = random.random()

        if operator < 0.33:

            return (
                self._swap_mutation(
                    child
                )
            )

        elif operator < 0.66:

            return (
                self._relocate_mutation(
                    child
                )
            )

        else:

            return (
                self._two_opt_mutation(
                    child
                )
            )


    # BUSCA LOCAL


    def _local_search(self, chromosome, data, max_moves=10):
        # Aplica operadores clássicos e, quando possível, consolida a frota.
        current = chromosome[:]; current_fitness = self._calculate_fitness(current, data); moves = 0
        while moves < max_moves:
            improved = False; indices = list(range(len(current))); random.shuffle(indices); limit = min(len(indices), 12)
            for a in range(limit):
                for b in range(a + 1, limit):
                    candidate = current[:]; i, j = indices[a], indices[b]; candidate[i], candidate[j] = candidate[j], candidate[i]; f = self._calculate_fitness(candidate, data)
                    if f + 1e-9 < current_fitness: current, current_fitness, moves, improved = candidate, f, moves + 1, True; break
                if improved: break
            if improved: continue
            if len(current) >= 2:
                for _ in range(10):
                    i, j = random.sample(range(len(current)), 2); candidate = current[:]; client = candidate.pop(i); candidate.insert(j, client); f = self._calculate_fitness(candidate, data)
                    if f + 1e-9 < current_fitness: current, current_fitness, moves, improved = candidate, f, moves + 1, True; break
            if improved: continue
            if len(current) >= 4:
                for _ in range(10):
                    i, j = sorted(random.sample(range(len(current)), 2)); candidate = current[:]; candidate[i:j + 1] = reversed(candidate[i:j + 1]); f = self._calculate_fitness(candidate, data)
                    if f + 1e-9 < current_fitness: current, current_fitness, moves, improved = candidate, f, moves + 1, True; break
            if not improved: break
        if data.get('allow_multi', False): current = self._fleet_consolidation(current, data, max_moves=max(2, max_moves // 2))
        return current

    def _fleet_consolidation(self, chromosome, data, max_moves=6):
        # Tenta remover veículos de forma real: reconstrói as viagens usando apenas um subconjunto menor da frota.
        current = chromosome[:]; current_fitness = self._calculate_fitness(current, data); decoded = self._decode_chromosome(current, {**data, 'deterministic_decoder': True}); used_indices = [v['vehicle_idx'] for v in decoded['vehicles'] if v.get('used', False)]
        if len(used_indices) <= 1 or not data.get('allow_multi', False): return current
        all_clients = [c for v in decoded['vehicles'] for trip in v.get('trips', []) for c in trip.get('path', []) if c != data['depot_id']]
        if len(all_clients) != len(chromosome) or len(set(all_clients)) != len(all_clients): return current
        attempts = 0; target_count = len(used_indices) - 1
        vehicle_orders = []
        # Primeiro tenta os veículos já utilizados; depois testa os veículos disponíveis quando a frota é heterogênea.
        vehicle_orders.append(used_indices[:])
        for idx in range(len(data['vehicle_capacities'])):
            if idx not in used_indices: vehicle_orders.append([idx])
        while target_count >= 1 and attempts < max_moves * 4:
            subsets = []
            if target_count == 1: subsets = [[v] for v in range(len(data['vehicle_capacities']))]
            else:
                from itertools import combinations
                subsets = [list(x) for x in combinations(range(len(data['vehicle_capacities'])), target_count)]
            random.shuffle(subsets)
            for subset in subsets[:max_moves * 2]:
                for order_mode in range(4):
                    order = all_clients[:]
                    if order_mode == 1: order.sort(key=lambda c: (data['janela_map_inicio'].get(c, 0.0), data['janela_map_fim'].get(c, 1440.0)))
                    elif order_mode == 2: order.sort(key=lambda c: data['janela_map_fim'].get(c, 1440.0))
                    elif order_mode == 3: order.sort(key=lambda c: (data['janela_map_fim'].get(c, 1440.0), -float(data['client_lookup'][c]['Demanda'])))
                    else: random.shuffle(order)
                    built = self._construct_solution_for_vehicle_subset(data, order, subset, attempts=2)
                    attempts += 1
                    if built is None or not built['metrics']['feasible']: continue
                    key = tuple(built['chromosome']); self._special_decoded_cache[key] = copy.deepcopy(built['decoded']); self._fitness_cache.pop(key, None); f = built['metrics']['fitness']
                    # A consolidação é uma tentativa; só é aceita se melhorar o objetivo real do problema.
                    if f + 1e-9 < current_fitness: return built['chromosome']
            target_count -= 1
        return current

    def _construct_solution_for_vehicle_subset(self, data, order, vehicle_indices, attempts=2):
        # Constrói uma solução usando exclusivamente os veículos do subconjunto e permitindo várias viagens.
        best = None
        for attempt in range(max(1, attempts)):
            vehicles = [{'vehicle_idx': v, 'trips': [], 'total_work': 0.0, 'used': False} for v in range(len(data['vehicle_capacities']))]; remaining = list(order)
            while remaining:
                candidates = []
                for client in remaining:
                    for v_idx in vehicle_indices:
                        result = self._evaluate_vehicle_insertion(vehicles[v_idx], client, data)
                        if result is not None: candidates.append((self._score_decoder_candidate(result, vehicles[v_idx], data), v_idx, client, result))
                if not candidates: break
                candidates.sort(key=lambda x: x[0]); top = candidates[:min(5, len(candidates))]; _, v_idx, client, result = top[0] if attempt == 0 else random.choice(top); vehicles[v_idx].clear(); vehicles[v_idx].update(result['vehicle']); remaining.remove(client)
            if not remaining:
                for vehicle in vehicles: self._reschedule_vehicle(vehicle, data)
                decoded = {'vehicles': vehicles, 'unassigned': []}; metrics = self._calculate_solution_metrics(decoded, data); chromosome = [c for v in vehicles for trip in v.get('trips', []) for c in trip.get('path', []) if c != data['depot_id']]
                candidate = {'decoded': decoded, 'metrics': metrics, 'chromosome': chromosome}
                if best is None or (metrics['feasible'], -metrics['used_vehicles'], -metrics['real_cost']) > (best['metrics']['feasible'], -best['metrics']['used_vehicles'], -best['metrics']['real_cost']): best = candidate
        return best


    # DIVERSIFICAÇÃO


    def _diversify(
        self,
        chromosome,
        intensity=0.20
    ):

        child = chromosome[:]

        n = len(child)

        if n < 2:

            return child

        number = max(
            1,
            int(
                n * intensity
            )
        )

        for _ in range(
            number
        ):

            operation = random.choice(
                [
                    "swap",
                    "relocate",
                    "reverse",
                ]
            )

            if operation == "swap":

                child = (
                    self._swap_mutation(
                        child
                    )
                )

            elif operation == "relocate":

                child = (
                    self._relocate_mutation(
                        child
                    )
                )

            else:

                if len(child) >= 3:

                    i, j = sorted(
                        random.sample(
                            range(
                                len(child)
                            ),
                            2
                        )
                    )

                    child[
                        i:j + 1
                    ] = reversed(
                        child[
                            i:j + 1
                        ]
                    )

        return child


    # CONTINUIDADE: VIZINHANÇAS LNS LEVES ()


    def _continuity_block_relocate(self, chromosome, max_block=6):
        """Move um bloco contíguo para outra posição sem perder a permutação.

        É barato (opera apenas no cromossomo) e cria uma vizinhança maior que um
        relocate unitário, útil quando o incumbent já está em um ótimo local forte.
        """
        child = list(chromosome)
        n = len(child)
        if n < 4:
            return self._relocate_mutation(child)
        block_len = random.randint(2, min(max_block, max(2, n // 8)))
        start = random.randint(0, n - block_len)
        block = child[start:start + block_len]
        del child[start:start + block_len]
        pos = random.randint(0, len(child))
        child[pos:pos] = block
        return child

    def _continuity_ruin_recreate_order(self, chromosome, data, fraction=0.12):
        """Ruin-and-recreate barato no espaço de permutações.

        Remove alguns clientes e os reinsere pela menor inserção Haversine na
        sequência. Não declara factibilidade: o decoder e, externamente, o Auditor
        continuam sendo as autoridades.
        """
        base = list(chromosome)
        n = len(base)
        if n < 6:
            return self._diversify(base, intensity=0.20)
        q = max(2, min(max(3, int(round(n * float(fraction)))), max(2, n // 3)))
        remove_idx = sorted(random.sample(range(n), q), reverse=True)
        removed = []
        for idx in remove_idx:
            removed.append(base.pop(idx))
        random.shuffle(removed)
        if not base:
            return list(chromosome)
        depot = int(data.get("depot_id", 0))
        for client in removed:
            if self._deadline_reached():
                # Completa de forma válida se o deadline chegar durante a reconstrução.
                base.append(client)
                continue
            best_pos = 0
            best_delta = float("inf")
            for pos in range(len(base) + 1):
                prev_node = depot if pos == 0 else base[pos - 1]
                next_node = depot if pos == len(base) else base[pos]
                delta = (
                    self._distance(prev_node, client, data)
                    + self._distance(client, next_node, data)
                    - self._distance(prev_node, next_node, data)
                )
                # Pequeno ruído só desempata e evita reconstruções idênticas.
                delta *= (1.0 + random.uniform(-0.015, 0.015))
                if delta < best_delta:
                    best_delta = delta
                    best_pos = pos
            base.insert(best_pos, client)
        return base

    def _continuity_shake(self, chromosome, data, strength=None):
        """Vizinhança composta usada SOMENTE em continuity_mode."""
        child = list(chromosome)
        n = len(child)
        if n < 2:
            return child
        if strength is None:
            strength = random.choice(("near", "medium", "strong"))
        if strength == "near":
            operations = random.randint(2, 3)
            ruin_prob = 0.10
        elif strength == "medium":
            operations = random.randint(3, 5)
            ruin_prob = 0.25
        else:
            operations = random.randint(5, 8)
            ruin_prob = 0.45
        for _ in range(operations):
            if self._deadline_reached():
                break
            r = random.random()
            if r < 0.28:
                child = self._swap_mutation(child)
            elif r < 0.56:
                child = self._relocate_mutation(child)
            elif r < 0.78:
                child = self._two_opt_mutation(child)
            else:
                child = self._continuity_block_relocate(child)
        if random.random() < ruin_prob and not self._deadline_reached():
            fraction = 0.08 if strength == "near" else (0.12 if strength == "medium" else 0.18)
            child = self._continuity_ruin_recreate_order(child, data, fraction=fraction)
        return child


    # ARQUIVO DE SOLUÇÕES NÃO DOMINADAS

    def _is_multiobjective(self, data):
        # Identifica se o problema possui mais de um objetivo ativo.
        config = data.get('multiobjective_config', {}) or {}; objectives = config.get('objectives', []) or []; return bool(config.get('enabled', False) or len(objectives) > 1)

    def _dominates_metrics(self, a, b, tol=1e-12):
        # Retorna True quando A é pelo menos tão boa quanto B em todos os objetivos e estritamente melhor em pelo menos um.
        if not a.get('feasible', False): return False
        if not b.get('feasible', False): return True
        objectives = list(a.get('active_objectives', [])) or ['DISTANCE']
        na, nb = a.get('objective_normalized', {}), b.get('objective_normalized', {}); better = False
        for obj in objectives:
            va, vb = float(na.get(obj, 0.0)), float(nb.get(obj, 0.0))
            if va > vb + tol: return False
            if va < vb - tol: better = True
        return better

    def _update_pareto_archive(self, population, data):
        # Preserva soluções não dominadas para impedir que uma boa solução seja perdida pela evolução.
        if not self._is_multiobjective(data): return
        candidates = list(self._pareto_archive.values())
        for chromosome in population:
            fitness, metrics, decoded = self._calculate_fitness(chromosome, data, return_decoded=True); candidates.append({'chromosome': copy.deepcopy(chromosome), 'metrics': copy.deepcopy(metrics), 'decoded': copy.deepcopy(decoded), 'fitness': fitness})
        nondominated = []
        for item in candidates:
            dominated = False
            for other in candidates:
                if item is other: continue
                if self._dominates_metrics(other['metrics'], item['metrics']): dominated = True; break
            if not dominated: nondominated.append(item)
        unique = {}
        for item in nondominated: unique[tuple(item['chromosome'])] = item
        self._pareto_archive = unique

    def _inject_pareto_archive(self, population, fitnesses, data, elite_size):
        # Substitui indivíduos piores por soluções não dominadas preservadas no arquivo.
        if not self._is_multiobjective(data) or not self._pareto_archive: return population, fitnesses
        ranked = sorted(range(len(population)), key=lambda i: fitnesses[i], reverse=True); archive_items = sorted(self._pareto_archive.values(), key=lambda x: (x['metrics'].get('used_vehicles', 999999), x['metrics'].get('real_cost', float('inf')))); replace_count = min(len(archive_items), max(1, elite_size))
        for pos in range(replace_count):
            idx = ranked[pos]; population[idx] = copy.deepcopy(archive_items[pos]['chromosome']); fitnesses[idx] = self._calculate_fitness(population[idx], data)
        return population, fitnesses


    # SELEÇÃO POR TORNEIO


    def _tournament_selection(
        self,
        population,
        fitnesses,
        tournament_size=3
    ):

        indices = random.sample(
            range(
                len(population)
            ),
            min(
                tournament_size,
                len(population)
            )
        )

        best_idx = min(
            indices,
            key=lambda i:
            fitnesses[i]
        )

        return population[
            best_idx
        ]


    # SEMENTE DE FACTIBILIDADE PARA INSTÂNCIAS GRANDES


    def _build_large_scale_feasibility_seed(self, data):
        """Constrói rapidamente um incumbent priorizando factibilidade, não custo.

        Clientes com janelas mais críticas entram primeiro. A escolha de veículo
        privilegia menor trabalho acumulado e folga temporal, permitindo abrir
        veículos adicionais quando isso evita clientes não atendidos.
        """
        if self._deadline_reached():
            return None
        depot = int(data["depot_id"])
        clients = [int(c) for c in data["client_lookup"] if int(c) != depot]
        clients.sort(key=lambda c: (
            float(data["janela_map_fim"].get(c, 1440.0)),
            float(data["janela_map_inicio"].get(c, 0.0)),
            -float(data["client_lookup"][c]["Demanda"]),
        ))
        vehicles = [
            {"vehicle_idx": v, "trips": [], "total_work": 0.0, "used": False}
            for v in range(len(data["vehicle_capacities"]))
        ]
        unassigned = []
        for client in clients:
            if self._deadline_reached():
                unassigned.extend([c for c in clients if c != client and c not in unassigned])
                unassigned.append(client)
                break
            choices = []
            # Para factibilidade testamos primeiro veículos menos carregados.
            order = sorted(range(len(vehicles)), key=lambda v: (vehicles[v]["total_work"], len(vehicles[v]["trips"])))
            for v_idx in order:
                if self._deadline_reached(): break
                vehicle = vehicles[v_idx]
                result = self._evaluate_vehicle_insertion(vehicle, client, data)
                if result is None:
                    continue
                candidate_vehicle = result["vehicle"]
                end_clock = float(data["vehicle_start_time"])
                if candidate_vehicle.get("trips"):
                    last = candidate_vehicle["trips"][-1]
                    end_clock = float(last["start_clock"]) + float(last["evaluation"]["work_time"])
                # Factibilidade primeiro: menor término, menor trabalho e maior capacidade.
                key = (
                    end_clock,
                    float(candidate_vehicle.get("total_work", 0.0)),
                    -float(data["vehicle_capacities"][v_idx]),
                    len(candidate_vehicle.get("trips", [])),
                )
                choices.append((key, v_idx, result))
            if not choices:
                unassigned.append(client)
                continue
            choices.sort(key=lambda x: x[0])
            _, v_idx, result = choices[0]
            vehicles[v_idx] = copy.deepcopy(result["vehicle"])

        for vehicle in vehicles:
            if self._deadline_reached(): break
            self._reschedule_vehicle(vehicle, data)
        decoded = {"vehicles": vehicles, "unassigned": list(unassigned)}
        metrics = self._calculate_solution_metrics(decoded, data)
        chromosome = [
            c for vehicle in vehicles for trip in vehicle.get("trips", [])
            for c in trip.get("path", []) if c != depot
        ]
        # Completa o cromossomo com clientes ausentes para preservar a permutação.
        chromosome += [c for c in clients if c not in set(chromosome)]
        if len(chromosome) != len(clients) or len(set(chromosome)) != len(clients):
            return None
        key = tuple(chromosome)
        self._special_decoded_cache[key] = copy.deepcopy(decoded)
        self._fitness_cache.pop(key, None)
        return {"chromosome": chromosome, "decoded": decoded, "metrics": metrics}


    # POPULAÇÃO INICIAL


    def _build_continuity_warm_start(self, data):
        """Converte a solução preservada do Relatório 1 em uma semente real do GA.

        A rota auditada é reconstruída com a mesma estrutura veículo/viagem e
        armazenada em _special_decoded_cache. Assim, o indivíduo baseline entra
        na população com sua solução decodificada preservada, em vez de ser
        reinterpretado pelo decoder estocástico.
        """
        mem = getattr(self.shared_memory, "data", {}) or {}
        baseline = mem.get("continuity_seed_solution") or mem.get("continuity_baseline_solution")
        if not isinstance(baseline, dict):
            ctx = mem.get("continuity_context", {}) or {}
            baseline = ctx.get("baseline_solution") if isinstance(ctx, dict) else None
        if not isinstance(baseline, dict):
            return None
        routes = baseline.get("routes", []) or []
        if isinstance(routes, dict):
            # Compatibilidade defensiva: algumas camadas podem armazenar rotas
            # agrupadas por veículo/viagem. Nunca iteramos chaves string como rotas.
            flattened = []
            for value in routes.values():
                if isinstance(value, dict):
                    flattened.append(value)
                elif isinstance(value, (list, tuple)):
                    flattened.extend(x for x in value if isinstance(x, dict))
            routes = flattened
        elif isinstance(routes, (list, tuple)):
            routes = [r for r in routes if isinstance(r, dict)]
        else:
            routes = []

        if not routes:
            return None

        depot = int(data["depot_id"])
        expected = {int(c) for c in data["client_lookup"].keys() if int(c) != depot}
        vehicles = [
            {"vehicle_idx": v, "trips": [], "total_work": 0.0, "used": False}
            for v in range(len(data["vehicle_capacities"]))
        ]
        seen = []
        try:
            ordered_routes = sorted(
                [r for r in routes if len(r.get("path", []) or []) > 2],
                key=lambda r: (
                    int(r.get("physical_vehicle_id", r.get("vehicle_id", 1)) or 1),
                    float(r.get("start_time_minutes", 0.0) or 0.0),
                    int(r.get("trip_id", r.get("trip_number", 0)) or 0),
                ),
            )
            for r in ordered_routes:
                vehicle_id = int(r.get("physical_vehicle_id", r.get("vehicle_id", 1)) or 1)
                v_idx = vehicle_id - 1
                if v_idx < 0 or v_idx >= len(vehicles):
                    return None
                path = [int(x) for x in (r.get("path", []) or [])]
                if not path or path[0] != depot:
                    path = [depot] + path
                if path[-1] != depot:
                    path = path + [depot]
                clients = [x for x in path if x != depot]
                if any(c not in expected for c in clients):
                    return None
                seen.extend(clients)
                start_clock = float(r.get("start_time_minutes", data["vehicle_start_time"]) or data["vehicle_start_time"])
                evaluation = self._evaluate_trip(path, data, start_clock)
                vehicles[v_idx]["trips"].append({
                    "path": path,
                    "evaluation": evaluation,
                    "start_clock": start_clock,
                })
                vehicles[v_idx]["used"] = True

            if len(seen) != len(expected) or set(seen) != expected or len(seen) != len(set(seen)):
                return None

            for vehicle in vehicles:
                vehicle["trips"].sort(key=lambda t: float(t["start_clock"]))
                vehicle["total_work"] = sum(float(t["evaluation"].get("work_time", 0.0)) for t in vehicle["trips"])
                vehicle["used"] = bool(vehicle["trips"])
                if vehicle["used"] and not self._vehicle_is_feasible(vehicle, data):
                    return None

            decoded = {"vehicles": vehicles, "unassigned": []}
            metrics = self._calculate_solution_metrics(decoded, data)
            if not metrics.get("feasible", False):
                return None

            chromosome = [c for r in ordered_routes for c in r.get("path", []) if int(c) != depot]
            chromosome = [int(c) for c in chromosome]
            if len(chromosome) != len(expected) or set(chromosome) != expected:
                return None
            key = tuple(chromosome)
            self._special_decoded_cache[key] = copy.deepcopy(decoded)
            self._fitness_cache.pop(key, None)
            return {"chromosome": chromosome, "decoded": decoded, "metrics": metrics}
        except Exception:
            return None

    def _initialize_population(self, data, population_size, continuity_mode=False):
        # Constrói a semente minimum_fleet uma vez e reutiliza o resultado.

        population = []
        strategies = ['random', 'window', 'demand', 'distance', 'hybrid']
        n_clients = max(0, len(data.get('client_lookup', {})) - 1)

        # Continuidade: preserva o baseline auditado, mas evita concentrar
        # toda a busca em uma vizinhança excessivamente estreita. A população
        # recebe perturbações em três escalas; o restante continua independente.
        # warm start e modo de continuidade são conceitos independentes.
        # A população diversificada só é criada quando o Orchestrator autoriza
        # explicitamente continuity_mode=True após o Relatório 1.
        self._continuity_warm_start_loaded = False
        continuity_seed = self._build_continuity_warm_start(data) if bool(continuity_mode) else None
        if continuity_seed is not None:
            self._continuity_warm_start_loaded = True
            base = list(continuity_seed['chromosome'])
            population.append(base)  # baseline exato; decoded preservado no cache especial

            # Faixas deliberadamente separadas. Em 60 clientes, por exemplo,
            # 0.03-0.08 altera poucos genes; 0.12-0.25 cruza bacias próximas;
            # 0.35-0.55 força exploração estrutural sem eliminar a parte aleatória.
            bands = [
                (0.15, (0.03, 0.08)),
                (0.15, (0.12, 0.25)),
                (0.10, (0.35, 0.55)),
            ]
            for fraction, (lo, hi) in bands:
                target = max(2, int(round(population_size * fraction)))
                created = 0
                attempts = 0
                max_attempts = max(8, target * 4)
                while created < target and attempts < max_attempts:
                    if self._deadline_reached() or len(population) >= population_size:
                        break
                    attempts += 1
                    intensity = random.uniform(lo, hi)
                    variant = self._diversify(copy.deepcopy(base), intensity=intensity)
                    if variant == base or variant in population:
                        continue
                    population.append(variant)
                    created += 1

        directive = self._get_continuity_directive(data)
        continuity_strategy = str(directive.get("type", "") or "").upper()
        target_max = self._continuity_target_max_vehicles(data)

        # elimina o bypass do warm start. Uma incumbente de 3 veículos
        # não pode ficar no cache especial durante uma busca HARD K=2.
        if continuity_strategy == "FLEET_REDUCTION" and target_max is not None:
            population = []
            self._special_decoded_cache = {}
            self._fitness_cache = {}
            self._continuity_warm_start_loaded = False
            continuity_seed = None

        # DIVERSIFICATION não pode ser uma cópia do incumbent com pequenas
        # mutações. Se a diretiva é de escape, removemos o baseline exato da população
        # e substituímos a maior parte das sementes por perturbações fortes/frescas.
        if continuity_strategy == "DIVERSIFICATION" and continuity_seed is not None:
            base = list(continuity_seed['chromosome'])
            population = []
            target_strong = max(6, int(round(population_size * 0.65)))
            attempts = 0
            while len(population) < target_strong and attempts < target_strong * 8:
                attempts += 1
                variant = self._continuity_shake(copy.deepcopy(base), data, strength="strong")
                variant = self._diversify(variant, intensity=random.uniform(0.35, 0.65))
                if variant != base and variant not in population:
                    population.append(variant)
            # O restante nasce sem warm-start, para abrir bacias realmente independentes.
            while len(population) < min(population_size, target_strong + max(4, int(population_size*0.20))):
                cand = self._generate_initial_solution(data, random.choice(strategies))
                cand = self._diversify(cand, intensity=random.uniform(0.20, 0.45))
                if cand not in population:
                    population.append(cand)
            self._continuity_warm_start_loaded = False

        # Em FLEET_REDUCTION aumentamos a presença de seeds de frota mínima.
        if continuity_strategy == "FLEET_REDUCTION" and data.get('allow_multi', False):
            min_fleet_count = max(2, int(population_size * 0.40))
        else:
            min_fleet_count = max(1, int(population_size * 0.20)) if data.get('allow_multi', False) and n_clients <= 50 else 0
        feasibility_seed = self._build_large_scale_feasibility_seed(data) if n_clients > 50 else None
        if feasibility_seed is not None and len(population) < population_size:
            candidate = list(feasibility_seed['chromosome'])
            if candidate not in population:
                population.append(candidate)

        minimum_seed = None
        if min_fleet_count > 0 and not self._deadline_reached():
            seed_obj = self._build_minimum_fleet_seed(data)
            if seed_obj is not None:
                minimum_seed = list(seed_obj.get('chromosome', []))

        for i in range(len(population), population_size):
            if self._deadline_reached():
                break
            if i < min_fleet_count and minimum_seed:
                chromosome = copy.deepcopy(minimum_seed)
                # Mantém a primeira cópia exatamente ligada ao decoded especial;
                # as demais ganham pequena diversidade e serão decodificadas normalmente.
                if i > 0:
                    chromosome = self._diversify(chromosome, intensity=0.06)
            else:
                strategy = strategies[(i - min_fleet_count) % len(strategies)]
                chromosome = self._generate_initial_solution(data, strategy)
                if i >= min_fleet_count + len(strategies):
                    chromosome = self._diversify(chromosome, intensity=0.10)
            population.append(chromosome)

        # Garante população mínima operacional mesmo se o deadline estiver muito curto.
        if not population:
            population.append(self._generate_initial_solution(data, 'distance'))
        return population


    # SOLVE PROBLEM


    def solve_problem(
        self,
        classification,
        translated_json,
        population_size=None,
        num_generations=None,
        time_limit_seconds=600,
        seed=None,
        feasibility_first=False,
        continuity_mode=False
    ):

        # INICIO DO CONTROLE DE TEMPO
        start_time = _time_module.monotonic()
        self._active_deadline = start_time + max(0.1, float(time_limit_seconds))

        stop_reason = "UNKNOWN"
        status = "UNKNOWN"
        last_generation = -1

        try:


            # SEMENTE

            # Cada execução começa com um arquivo limpo de soluções não dominadas.
            self._pareto_archive = {}
            self._special_decoded_cache = {}
            self._fitness_cache = {}


            if seed is not None:
                random.seed(seed)

                np.random.seed(
                    seed
                )


            # DADOS



            call_local_directive = None
            _directive_source = translated_json
            if isinstance(translated_json, str):
                try:
                    _directive_source = json.loads(translated_json)
                except Exception:
                    _directive_source = None
            if isinstance(_directive_source, dict):
                embedded = _directive_source.get("_continuity_directive")
                if isinstance(embedded, dict):
                    call_local_directive = copy.deepcopy(embedded)

            data = (
                self._create_data_lookup(
                    translated_json
                )
            )
            data["feasibility_first"] = bool(feasibility_first)

            if isinstance(call_local_directive, dict):
                directive = call_local_directive
            else:
                directive = self._get_continuity_directive(None)

            data["_continuity_directive"] = copy.deepcopy(directive)
            data["_continuity_strategy"] = str(directive.get("type", "") or "").upper()
            data["_continuity_target_max_vehicles"] = self._continuity_target_max_vehicles(data)

            if str(data.get("_continuity_strategy", "")).upper() == "FLEET_REDUCTION":
                if data.get("_continuity_target_max_vehicles") is None:
                    raise RuntimeError("CONTRACT_ERROR: FLEET_REDUCTION chegou ao GA sem K")
                print(
                    "   🔒 GA HARD-K confirmado | "
                    f"strategy=FLEET_REDUCTION | K={data['_continuity_target_max_vehicles']} | "
                    f"input_type={type(translated_json).__name__}"
                )


            # CLIENTES


            clients = [

                c

                for c in data[
                    "client_lookup"
                ].keys()

                if c != data[
                    "depot_id"
                ]
            ]

            num_clients = len(
                clients
            )

            num_vehicles = len(
                data[
                    "vehicle_capacities"
                ]
            )

            if num_clients == 0:

                return {

                    "status":
                        "SUCCESS",

                    "method":
                        "GA-Hybrid-MultiTrip-VRPTW",

                    "total_cost":
                        0.0,

                    "fitness":
                        0.0,

                    "feasible":
                        True,

                    "metrics": {

                        "distance_cost":
                            0.0,

                        "fixed_cost":
                            0.0,

                        "used_vehicles":
                            0,

                        "missing_clients":
                            0,

                        "window_violation":
                            0.0,

                        "capacity_violation":
                            0.0,

                        "time_violation":
                            0.0,

                        "min_usage_violation":
                            0.0,
                    },

                    "routes":
                        [],
                }


            # VALIDAÇÃO BÁSICA


            if num_vehicles <= 0:

                return {

                    "status":
                        "ERROR",

                    "method":
                        "GA-Hybrid-MultiTrip-VRPTW",

                    "error_log":
                        [
                            "Nenhum veículo disponível."
                        ],

                    "routes":
                        [],
                }

            if len(
                data["vehicle_costs_km"]
            ) < num_vehicles:

                data[
                    "vehicle_costs_km"
                ] = (
                    data[
                        "vehicle_costs_km"
                    ]
                    + [1.0]
                    * (
                        num_vehicles
                        - len(
                            data[
                                "vehicle_costs_km"
                            ]
                        )
                    )
                )

            if len(
                data["vehicle_fixed_costs"]
            ) < num_vehicles:

                data[
                    "vehicle_fixed_costs"
                ] = (
                    data[
                        "vehicle_fixed_costs"
                    ]
                    + [0.0]
                    * (
                        num_vehicles
                        - len(
                            data[
                                "vehicle_fixed_costs"
                            ]
                        )
                    )
                )


            # PARÂMETROS


            # Parâmetros adaptativos: instâncias grandes precisam de busca seletiva,
            # não simplesmente de populações e gerações maiores.
            if population_size is None:
                if num_clients <= 20:
                    population_size = 180
                elif num_clients <= 50:
                    population_size = 140
                else:
                    population_size = 100

            if num_generations is None:
                if num_clients <= 20:
                    num_generations = 240
                elif num_clients <= 50:
                    num_generations = 180
                else:
                    num_generations = 120

            population_size = max(
                20,
                int(
                    population_size
                )
            )
            if num_clients > 50:
                # Decodificação VRPTW Multi-Trip é a operação dominante; populações
                # menores dão mais iterações úteis dentro do deadline real.
                population_size = min(population_size, 48 if feasibility_first else 60)

            num_generations = max(
                20,
                int(
                    num_generations
                )
            )


            # POPULAÇÃO


            population = (
                self._initialize_population(
                    data,
                    population_size,
                    continuity_mode=bool(continuity_mode)
                )
            )
            # Não inferir continuidade a partir do cache especial: o fluxo principal
            # também usa seeds/cache. Continuidade só existe por autorização explícita.
            continuity_warm_start_active = bool(
                continuity_mode and getattr(self, "_continuity_warm_start_loaded", False)
            )


            # AVALIAÇÃO INICIAL


            fitnesses = []
            evaluated_population = []
            for chromosome in population:
                if self._deadline_reached() and evaluated_population:
                    break
                fitnesses.append(self._calculate_fitness(chromosome, data))
                evaluated_population.append(chromosome)
            population = evaluated_population
            if not population:
                return {
                    "status": "TIME_LIMIT",
                    "method": "GA-Hybrid-MultiTrip-VRPTW",
                    "routes": [],
                    "error_log": ["Deadline atingido antes da primeira avaliação completa."],
                }
            # Registra imediatamente as soluções não dominadas da população inicial.
            if not self._deadline_reached() and num_clients <= 50:
                self._update_pareto_archive(population, data)
            best_idx = min(
                range(
                    len(population)
                ),
                key=lambda i:
                fitnesses[i]
            )

            best_global = (
                copy.deepcopy(
                    population[
                        best_idx
                    ]
                )
            )

            best_fitness = (
                fitnesses[
                    best_idx
                ]
            )

            (
                _,
                best_metrics,
                _
            ) = self._calculate_fitness(
                best_global,
                data,
                return_decoded=True
            )


            # PARÂMETROS EVOLUTIVOS


            elite_size = max(
                2,
                int(
                    population_size
                    * 0.05
                )
            )

            elite_size = min(
                elite_size,
                population_size
            )

            base_mutation_rate = 0.20

            stagnation = 0

            if num_clients <= 20:
                patience = max(35, int(num_generations * 0.18))
            elif num_clients <= 50:
                patience = max(30, int(num_generations * 0.15))
            else:
                patience = max(24, int(num_generations * 0.12))

            # somente quando existe um baseline de continuidade válido,
            # damos mais tempo para a população atravessar platôs. O solver
            # principal usa a política de parada definida acima.
            if continuity_warm_start_active:
                if num_clients > 50:
                    patience = max(patience, 60, int(num_generations * 0.30))
                elif num_clients > 20:
                    patience = max(patience, 45, int(num_generations * 0.25))

            # Busca local seletiva. Para >50 clientes continua DESLIGADA no fluxo
            # principal; na continuidade usamos taxa baixa e apenas 1 movimento,
            # suficiente para intensificar alguns descendentes sem dominar o deadline.
            local_search_rate = 0.18 if num_clients <= 20 else (0.10 if num_clients <= 50 else 0.0)
            local_search_moves = 5 if num_clients <= 20 else (3 if num_clients <= 50 else 2)
            if continuity_warm_start_active and num_clients > 50:
                # Em LARGE_SCALE a busca local por descendente mostrou custo alto e
                # derrubou o número de gerações úteis. A continuidade mantém a
                # diversificação multiescala da população e a paciência ampliada,
                # mas deixa a busca local desligada para recuperar throughput.
                local_search_rate = 0.0
                local_search_moves = 1

            # configuração LNS coerente com a hipótese de Continuidade.
            # A mesma configuração é usada tanto no log quanto na reprodução.
            directive_type = str(data.get("_continuity_strategy", "") or "").upper()
            lns_probability = 0.45
            lns_weights = [0.45, 0.40, 0.15]

            if directive_type == "FLEET_REDUCTION":
                lns_probability = 0.60
                lns_weights = [0.20, 0.50, 0.30]
            elif directive_type in ("ROUTE_DISTANCE_REDUCTION", "ROUTE_TIME_REDUCTION"):
                lns_probability = 0.50
                lns_weights = [0.60, 0.35, 0.05]
            elif directive_type == "VEHICLE_COST_REALLOCATION":
                lns_probability = 0.50
                lns_weights = [0.35, 0.45, 0.20]
            elif directive_type == "DIVERSIFICATION":
                # o GA não é mais o executor normal do Track A.
                # Aqui sua função é escapar deliberadamente da bacia local do AVE.
                lns_probability = 0.72
                lns_weights = [0.20, 0.35, 0.45]
                local_search_rate = 0.0
                local_search_moves = 1
                patience = max(patience, 70, int(num_generations * 0.35))

            progress_every = 20 if num_clients <= 50 else 10
            print(
                f"🧬 GA: início | clientes={num_clients} | população={population_size} | "
                f"gerações={num_generations} | limite={time_limit_seconds}s | seed={seed} | continuity_mode={'SIM' if continuity_mode else 'NÃO'} | warm_start_continuidade={'SIM' if continuity_warm_start_active else 'NÃO'}"
            )
            if continuity_warm_start_active:
                print(
                    f"   🔁 Continuidade V14 Objective-Driven LNS | estratégia={directive_type or 'GENERIC'} | "
                    f"população diversificada=SIM | patience={patience} | "
                    f"busca_local_clássica={local_search_rate:.2%} | "
                    f"LNS_composto={lns_probability:.0%} | movimentos_clássicos={local_search_moves}"
                )


            # EVOLUÇÃO


            for generation in range(num_generations):

                last_generation = generation

            #verifica limite de tempo
                if self._time_limit_reached(start_time, time_limit_seconds):
                    stop_reason = "TIME_LIMIT"
                    break

                ranked = sorted(
                    zip(
                        population,
                        fitnesses
                    ),
                    key=lambda x:
                    x[1]
                )

                generation_best = (
                    ranked[0]
                )


                # MELHOR GLOBAL


                if (
                    generation_best[1]
                    + 1e-9
                    < best_fitness
                ):

                    best_fitness = (
                        generation_best[1]
                    )

                    best_global = (
                        copy.deepcopy(
                            generation_best[0]
                        )
                    )

                    (
                        _,
                        best_metrics,
                        _
                    ) = (
                        self._calculate_fitness(
                            best_global,
                            data,
                            return_decoded=True
                        )
                    )

                    stagnation = 0

                else:

                    stagnation += 1


                # MUTAÇÃO ADAPTATIVA


                if stagnation < 10:

                    mutation_rate = (
                        base_mutation_rate
                    )

                elif stagnation < 30:

                    mutation_rate = 0.30

                elif stagnation < 60:

                    mutation_rate = 0.45

                else:

                    mutation_rate = 0.60


                # NOVA POPULAÇÃO


                new_population = []


                # ELITISMO


                for i in range(
                    elite_size
                ):

                    new_population.append(
                        copy.deepcopy(
                            ranked[i][0]
                        )
                    )


                # REPRODUÇÃO


                while len(
                    new_population
                ) < population_size:
                    if self._deadline_reached():
                        stop_reason = "TIME_LIMIT"
                        break

                    parent1 = (
                        self._tournament_selection(
                            population,
                            fitnesses,
                            3
                        )
                    )

                    parent2 = (
                        self._tournament_selection(
                            population,
                            fitnesses,
                            3
                        )
                    )


                    # OX


                    if random.random() < 0.85:

                        child = (
                            self._order_crossover(
                                parent1,
                                parent2
                            )
                        )

                    else:

                        child = (
                            copy.deepcopy(
                                parent1
                            )
                        )


                    # MUTAÇÃO


                    # Continuidade LARGE_SCALE: complementa a mutação unitária;
                    # parte dos descendentes recebe uma vizinhança composta barata.
                    # A regra adicional depende de uma continuidade com plano inicial válido.
                    if continuity_warm_start_active and num_clients > 50 and random.random() < lns_probability:
                        strength = random.choices(
                            ["near", "medium", "strong"],
                            weights=lns_weights,
                            k=1
                        )[0]
                        child = self._continuity_shake(child, data, strength=strength)
                        # Mutação adaptativa adicional apenas com metade da taxa para
                        # não destruir sistematicamente a estrutura criada pelo LNS.
                        child = self._mutate(child, mutation_rate * 0.50)
                    else:
                        child = self._mutate(
                            child,
                            mutation_rate
                        )


                    # BUSCA LOCAL


                    if random.random() < local_search_rate:
                        child = self._local_search(
                            child,
                            data,
                            max_moves=local_search_moves
                        )

                    new_population.append(
                        child
                    )

                if stop_reason == "TIME_LIMIT":
                    break

                population = (
                    new_population
                )


                # FITNESS


                new_fitnesses = []
                evaluated_population = []
                for chromosome in population:
                    if self._deadline_reached() and evaluated_population:
                        stop_reason = "TIME_LIMIT"
                        break
                    new_fitnesses.append(self._calculate_fitness(chromosome, data))
                    evaluated_population.append(chromosome)
                population = evaluated_population
                fitnesses = new_fitnesses
                if stop_reason == "TIME_LIMIT":
                    break
                # Atualiza e reinjeta o arquivo Pareto a cada geração.
                if num_clients <= 50 and not self._deadline_reached():
                    self._update_pareto_archive(population, data)
                    population, fitnesses = self._inject_pareto_archive(population, fitnesses, data, elite_size)


                # DIVERSIFICAÇÃO


                # continuidade LARGE_SCALE: quando o incumbent fica
                # estagnado, substitui apenas a cauda da população por vizinhanças
                # multiescala dos melhores indivíduos e por algumas soluções frescas.
                # O elite (incluindo o baseline quando ele ainda é o melhor) é preservado.
                if (
                    continuity_warm_start_active
                    and num_clients > 50
                    and stagnation >= 10
                    and generation > 0
                    and generation % 8 == 0
                    and not self._deadline_reached()
                ):
                    ranked_now = sorted(zip(population, fitnesses), key=lambda x: x[1])
                    replace_count = max(4, int(round(population_size * (0.50 if directive_type == "DIVERSIFICATION" else 0.25))))
                    keep_count = max(elite_size, population_size - replace_count)
                    refreshed = [copy.deepcopy(ch) for ch, _ in ranked_now[:keep_count]]
                    top_pool = [ch for ch, _ in ranked_now[:max(elite_size, min(6, len(ranked_now)))]]
                    while len(refreshed) < population_size and not self._deadline_reached():
                        slot = len(refreshed) - keep_count
                        if slot < max(2, replace_count // 5):
                            candidate = self._generate_initial_solution(
                                data, random.choice(["random", "window", "demand", "distance", "hybrid"])
                            )
                            candidate = self._diversify(candidate, intensity=random.uniform(0.10, 0.25))
                        else:
                            parent = copy.deepcopy(random.choice(top_pool))
                            strength = random.choices(
                                ["near", "medium", "strong"],
                                weights=[0.25, 0.50, 0.25],
                                k=1
                            )[0]
                            candidate = self._continuity_shake(parent, data, strength=strength)
                        refreshed.append(candidate)
                    population = refreshed
                    fitnesses = [self._calculate_fitness(ch, data) for ch in population]

                if (
                    stagnation > max(30, patience // 2)
                    and num_clients <= 50
                    and
                    generation % 10 == 0
                ):

                    number_to_replace = max(
                        1,
                        int(
                            population_size
                            * 0.15
                        )
                    )

                    ranked = sorted(
                        zip(
                            population,
                            fitnesses
                        ),
                        key=lambda x:
                        x[1]
                    )

                    new_population = [

                        copy.deepcopy(
                            ranked[i][0]
                        )

                        for i in range(
                            elite_size
                        )
                    ]


                    # MANTER PARTE DA POPULAÇÃO


                    keep_until = (
                        population_size
                        - number_to_replace
                    )

                    while len(
                        new_population
                    ) < keep_until:

                        parent = (
                            self._tournament_selection(
                                population,
                                fitnesses,
                                3
                            )
                        )

                        candidate = (
                            self._diversify(
                                parent,
                                intensity=0.20
                            )
                        )

                        new_population.append(
                            candidate
                        )


                    # NOVOS INDIVÍDUOS


                    while len(
                        new_population
                    ) < population_size:

                        candidate = (
                            self._generate_initial_solution(
                                data,
                                random.choice(
                                    [
                                        "random",
                                        "window",
                                        "demand",
                                        "distance",
                                        "hybrid",
                                    ]
                                )
                            )
                        )

                        candidate = (
                            self._diversify(
                                candidate,
                                intensity=0.20
                            )
                        )

                        new_population.append(
                            candidate
                        )

                    population = (
                        new_population
                    )

                    fitnesses = [

                        self._calculate_fitness(
                            chromosome,
                            data
                        )

                        for chromosome
                        in population
                    ]

                if generation == 0 or (generation + 1) % progress_every == 0:
                    elapsed = _time_module.monotonic() - start_time
                    print(
                        f"   GA geração {generation + 1}/{num_generations} | "
                        f"melhor fitness={best_fitness:.6f} | estagnação={stagnation} | "
                        f"tempo={elapsed:.1f}s"
                    )


                # EARLY STOPPING


                if (
                    stagnation
                    >= patience
                    and
                    best_metrics[
                        "feasible"
                    ]
                ):

                    stop_reason = "EARLY_STOPPING"
                    break


            # MELHOR SOLUÇÃO


            # O melhor indivíduo ponderado continua sendo a regra principal; o arquivo Pareto impede perda de alternativas não dominadas.
            if self._is_multiobjective(data) and self._pareto_archive:
                archive_best = min(self._pareto_archive.values(), key=lambda x: x['fitness'])
                if float(archive_best['fitness']) + 1e-9 < float(best_fitness): best_global = copy.deepcopy(archive_best['chromosome']); best_fitness = float(archive_best['fitness'])

            # Última tentativa de consolidação antes de devolver a solução ao restante do sistema.
            if data.get("allow_multi", False) and num_clients <= 50 and not self._deadline_reached():
                consolidated = self._fleet_consolidation(copy.deepcopy(best_global), data, max_moves=24)
                c_fit, c_metrics, _ = self._calculate_fitness(consolidated, data, return_decoded=True)
                if c_metrics.get("feasible", False) and c_fit <= best_fitness + 1e-12:
                    best_global = consolidated
                    best_fitness = c_fit

            (
                final_fitness,
                final_metrics,
                decoded
            ) = self._calculate_fitness(
                best_global,
                data,
                return_decoded=True
            )


            # GERAR ROTAS


            final_routes = []

            for vehicle in decoded[
                "vehicles"
            ]:

                v_idx = vehicle[
                    "vehicle_idx"
                ]

                if not vehicle[
                    "used"
                ]:

                    continue

                for trip_idx, trip in enumerate(
                    vehicle[
                        "trips"
                    ]
                ):

                    path = trip[
                        "path"
                    ]

                    evaluation = (
                        trip[
                            "evaluation"
                        ]
                    )

                    distance = (
                        evaluation[
                            "distance"
                        ]
                    )

                    arrival_times = (
                        evaluation[
                            "arrival_times"
                        ]
                    )

                    service_times = (
                        evaluation[
                            "service_start_times"
                        ]
                    )

                    loads = (
                        evaluation[
                            "loads"
                        ]
                    )


                    # DETALHAMENTO


                    path_details = []

                    path_details.append({

                        "node_id":
                            int(
                                data[
                                    "depot_id"
                                ]
                            ),

                        "arrival_time_minutes":
                            round(
                                trip[
                                    "start_clock"
                                ],
                                2
                            ),

                        "service_start_time_minutes":
                            round(
                                trip[
                                    "start_clock"
                                ],
                                2
                            ),

                        "load":
                            0,
                    })

                    current_load = 0.0

                    for i in range(
                        1,
                        len(path)
                    ):

                        node = path[i]

                        if (
                            node
                            != data[
                                "depot_id"
                            ]
                        ):

                            current_load += float(
                                data[
                                    "client_lookup"
                                ][node][
                                    "Demanda"
                                ]
                            )

                        arrival = (

                            arrival_times[
                                i - 1
                            ]

                            if (
                                i - 1
                                <
                                len(
                                    arrival_times
                                )
                            )

                            else None
                        )

                        service_start = (

                            service_times[
                                i - 1
                            ]

                            if (
                                i - 1
                                <
                                len(
                                    service_times
                                )
                            )

                            else arrival
                        )

                        path_details.append({

                            "node_id":
                                int(node),

                            "arrival_time_minutes":
                                (
                                    None
                                    if arrival is None
                                    else round(
                                        arrival,
                                        2
                                    )
                                ),

                            "service_start_time_minutes":
                                (
                                    None
                                    if service_start is None
                                    else round(
                                        service_start,
                                        2
                                    )
                                ),

                            "load":
                                round(
                                    current_load,
                                    2
                                ),
                        })


                    # CUSTO DA VIAGEM


                    trip_cost = (
                        distance
                        *
                        data[
                            "vehicle_costs_km"
                        ][v_idx]
                    )


                    # CUSTO FIXO POR ATIVAÇÃO

                    # O custo fixo aparece somente na primeira
                    # viagem do veículo.

                    # Não é somado novamente ao total final,
                    # pois já está incluído em final_metrics.


                    if (
                        trip_idx == 0
                        and
                        data[
                            "fixed_cost_type"
                        ]
                        == "ativacao"
                    ):

                        trip_cost += (
                            data[
                                "vehicle_fixed_costs"
                            ][v_idx]
                        )


                    # CARGA


                    total_load = (

                        max(
                            loads
                        )
                        if loads
                        else 0.0
                    )


                    # ROTA FINAL


                    final_routes.append({

                        "vehicle_id":
                            v_idx + 1,

                        "trip_id":
                            trip_idx + 1,

                        # Contrato explícito com o Auditor independente.
                        # O horário já foi calculado pelo decoder; apenas o expomos
                        # na rota final sem alterar a lógica matemática do GA.
                        "start_time_minutes":
                            float(trip["start_clock"]),

                        "path":
                            path,

                        "distance_km":
                            round(
                                distance,
                                3
                            ),

                        "total_route_cost":
                            round(
                                trip_cost,
                                2
                            ),

                        "total_route_load":
                            round(
                                total_load,
                                2
                            ),

                        "total_route_time_minutes":
                            round(
                                evaluation[
                                    "work_time"
                                ],
                                2
                            ),

                        "waiting_time_minutes":
                            round(
                                evaluation[
                                    "waiting_time"
                                ],
                                2
                            ),

                        "path_details":
                            path_details,
                    })


            # CUSTO FINAL


            final_total_cost = (
                final_metrics[
                    "real_cost"
                ]
            )


            # STATUS


            status = (

                "SUCCESS"

                if final_metrics[
                    "feasible"
                ]

                else "INFEASIBLE"
            )


            # RESULTADO


            result = {

                "status":
                    status,

                "method":
                    "GA-Hybrid-MultiTrip-VRPTW",

                "classification":
                    classification,

                "total_cost":
                    round(
                        final_total_cost,
                        2
                    ),

                "fitness":
                    round(
                        final_fitness,
                        2
                    ),

                "feasible":
                    final_metrics[
                        "feasible"
                    ],

                "metrics": {

                    "distance_cost":
                        round(
                            final_metrics[
                                "distance_cost"
                            ],
                            2
                        ),

                    "fixed_cost":
                        round(
                            final_metrics[
                                "fixed_cost"
                            ],
                            2
                        ),

                    "used_vehicles":
                        final_metrics[
                            "used_vehicles"
                        ],

                    # Resultados explícitos dos objetivos para o Diagnostic/Explainer/Auditor.
                    "objective_value": round(float(final_metrics.get("objective_value", final_fitness)), 8),
                    "objective_components": final_metrics.get("objective_components", {}),
                    "objective_normalized": final_metrics.get("objective_normalized", {}),
                    "objective_weights": final_metrics.get("objective_weights", {}),
                    "active_objectives": final_metrics.get("active_objectives", []),
                    "objective_context_version": final_metrics.get("objective_context_version"),
                    "total_distance": round(float(final_metrics.get("total_distance", 0.0)), 3),
                    "total_time": round(float(final_metrics.get("total_time", 0.0)), 2),

                    "required_vehicles":
                        num_vehicles
                        if data[
                            "must_use_all_vehicles"
                        ]
                        else None,

                    "missing_clients":
                        final_metrics[
                            "missing_clients"
                        ],
                    "missing_client_ids": final_metrics.get("missing_client_ids", []),
                    "duplicate_clients": final_metrics.get("duplicate_clients", 0),
                    "duplicate_client_ids": final_metrics.get("duplicate_client_ids", []),

                    "window_violation":
                        round(
                            final_metrics[
                                "window_violation"
                            ],
                            4
                        ),

                    "capacity_violation":
                        round(
                            final_metrics[
                                "capacity_violation"
                            ],
                            4
                        ),

                    "time_violation":
                        round(
                            final_metrics[
                                "time_violation"
                            ],
                            4
                        ),

                    "min_usage_violation":
                        round(
                            final_metrics[
                                "min_usage_violation"
                            ],
                            4
                        ),

                    "vehicle_usage_violation":
                        final_metrics[
                            "vehicle_usage_violation"
                        ],
                },

                "routes":
                    final_routes,
                "continuity_warm_start_used": bool(continuity_warm_start_active),
                "continuity_mode": bool(continuity_mode),
                "continuity_strategy": str(data.get("_continuity_strategy", "") or "") or None,
                "continuity_target_max_vehicles": data.get("_continuity_target_max_vehicles"),
                "continuity_directive_applied": copy.deepcopy(data.get("_continuity_directive", {}) or {}),
                "continuity_directive_source": (
                    "CALL_LOCAL"
                    if isinstance(call_local_directive, dict)
                    else ("SHARED_MEMORY_FALLBACK" if data.get("_continuity_directive") else "NONE")
                ),
                "continuity_lns_probability": float(lns_probability),
                "continuity_lns_weights": list(lns_weights),
                "ga_version": self.VERSION,
                "ga_role": ("DIVERSIFICATION_ESCAPE_ENGINE" if str(data.get("_continuity_strategy", "")).upper() == "DIVERSIFICATION" else "PRIMARY_OR_PORTFOLIO_SOLVER"),
            }


            # QUALIDADE MATEMÁTICA / CONVERGÊNCIA

            if stop_reason == "UNKNOWN":
                stop_reason = "MAX_GENERATIONS"

            elapsed_seconds = max(0.0, _time_module.monotonic() - start_time)
            result["optimality"] = {
                "status": "BEST_FOUND_NOT_PROVEN" if status == "SUCCESS" else "NO_VALID_SOLUTION",
                "proven_optimal": False,
                "termination_reason": stop_reason,
                "objective_value": result.get("fitness"),
                "best_bound": None,
                "optimality_gap": None,
                "generations_executed": max(0, last_generation + 1),
                "execution_time_seconds": round(elapsed_seconds, 3),
                "explanation": (
                    "Algoritmos Genéticos são meta-heurísticos: a solução reportada "
                    "é a melhor solução factível encontrada, sem prova de ótimo global."
                )
            }
            # CONTRATO DURO DE REDUÇÃO DE FROTA.
            # Uma execução FLEET_REDUCTION(target=K) jamais pode retornar SUCCESS
            # com mais de K veículos. Isso transforma bugs de integração em
            # estados explícitos, em vez de falsos sucessos silenciosos.
            _strategy_contract = str(data.get("_continuity_strategy", "") or "").upper()
            _target_contract = data.get("_continuity_target_max_vehicles")
            if _strategy_contract == "FLEET_REDUCTION":
                if _target_contract is None:
                    raise RuntimeError("CONTRACT_ERROR: GA recebeu FLEET_REDUCTION sem target_max_vehicles")
                _used_contract = int(final_metrics.get("used_vehicles", 0) or 0)
                _decoded_allowed = (
                    decoded.get("hard_k_allowed_vehicle_indices", [])
                    if isinstance(decoded, dict) else []
                )
                result["fleet_reduction_contract"] = {
                    "target_max_vehicles": int(_target_contract),
                    "used_vehicles": _used_contract,
                    "satisfied": bool(
                        _used_contract <= int(_target_contract)
                        and bool(final_metrics.get("feasible", False))
                    ),
                    "physical_hard_k": True,
                    "allowed_vehicle_indices": list(_decoded_allowed),
                    "allowed_vehicle_ids": [int(i) + 1 for i in _decoded_allowed],
                    "unassigned_clients": int(
                        final_metrics.get("missing_clients", 0) or 0
                    ),
                }
                _hard_k_feasible = bool(final_metrics.get("feasible", False))
                _hard_k_satisfied = (
                    _hard_k_feasible
                    and _used_contract <= int(_target_contract)
                )
                if status == "SUCCESS" and not _hard_k_satisfied:
                    result["status"] = "NOT_FOUND"
                    status = "NOT_FOUND"
                    result["contract_status"] = "TARGET_NOT_SATISFIED"
                    result["contract_message"] = (
                        f"Busca PHYSICAL HARD-K <= {_target_contract} não encontrou "
                        "candidato completamente factível dentro do subconjunto físico "
                        "permitido. Isto é NOT_FOUND heurístico e NÃO prova inviabilidade."
                    )

            result["solution_quality"] = {
                "feasible": bool(final_metrics.get("feasible", False)) and status == "SUCCESS",
                "audited": False,
                "optimality_proven": False,
                "quality_label": "MELHOR_SOLUCAO_ENCONTRADA" if status == "SUCCESS" else "SEM_SOLUCAO_VALIDA"
            }


            # SALVAR NA SHARED MEMORY

            # Mantemos a integração com o restante do sistema.


            try:

                self.shared_memory.data[
                    "ga_solver_result"
                ] = result

            except Exception:

                pass

            result["runtime_seconds"] = round(_time_module.monotonic() - start_time, 3)
            result["generations_executed"] = max(0, last_generation + 1)
            result["stop_reason"] = stop_reason if stop_reason != "UNKNOWN" else "GENERATIONS_COMPLETED"

            # Diagnóstico temporário: imprime exatamente as rotas e os
            # tempos por veículo que serão entregues ao Orchestrator/Auditor.

            print(
                f"✅ GA: terminou em {result['runtime_seconds']:.1f}s | "
                f"gerações={result['generations_executed']} | motivo={result['stop_reason']} | "
                f"status={result.get('status')} | custo={float(result.get('total_cost', 0.0) or 0.0):.2f}"
            )
            return result


        # ERRO CONTROLADO


        except Exception as e:

            error_message = (
                f"Erro no GeneticAlgorithmSolverAgent: "
                f"{type(e).__name__}: {str(e)}"
            )

            try:

                self.shared_memory.data[
                    "ga_solver_error"
                ] = error_message

            except Exception:

                pass

            return {

                "status":
                    "ERROR",

                "method":
                    "GA-Hybrid-MultiTrip-VRPTW",

                "error":
                    error_message,

                "error_type":
                    type(e).__name__,

                "routes":
                    [],
            }
````

### **OPTIMALITYCERTIFICATIONAGENT — CERTIFICAÇÃO INICIAL**

Depois de encontrar um plano, surge outra pergunta: “quanto sabemos sobre a qualidade dele?”. Esse agente reúne evidências para responder, em vez de confiar só na aparência da solução.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class OptimalityCertificationAgent:
    """Avalia a qualidade do plano por meio de limites matemáticos globais.

    Calcula limites analíticos e oferece a formulação exata quando aplicável.
    A certificação depende da validade global da evidência, da consistência do
    objetivo e da auditoria independente, respeitando as tolerâncias numéricas.
    Um resultado de busca local, por si só, não demonstra o ótimo global."""

    VERSION = "SCIP_CASCADE_CERTIFIER_16_K_FEASIBILITY_SCF"

    STATUS_RUNNING = "CERTIFICATION_RUNNING"
    STATUS_OPTIMAL = "PROVEN_OPTIMAL"
    STATUS_INFEASIBLE = "PROVEN_INFEASIBLE"
    STATUS_TIME_LIMIT = "TIME_LIMIT"
    STATUS_RESOURCE_LIMIT = "RESOURCE_LIMIT"
    STATUS_ERROR = "CERTIFICATION_ERROR"
    STATUS_NOT_PROVEN = "NOT_PROVEN"

    def __init__(
        self,
        shared_memory,
        time_limit_seconds=300,
        scip_threads=0,
        scip_memory_mb=0,
        relative_gap_target=0.0,
        use_incumbent_cutoff=True,
        verbose_solver=True,
        relative_gap_tolerance=0.0001,
        business_gap_tolerance=0.01,
        intensive_certification=False,
        intensive_time_limit_seconds=1800,
        adaptive_stop_enabled=True,
        adaptive_min_runtime_seconds=30,
        adaptive_patience_seconds=20,
        adaptive_check_interval_seconds=5,
        adaptive_min_relative_lb_improvement=0.0001,
    ):
        self.shared_memory = shared_memory
        self.time_limit_seconds = max(1, int(time_limit_seconds))
        self.scip_threads = max(0, int(scip_threads))
        self.scip_memory_mb = max(0, int(scip_memory_mb))
        self.relative_gap_target = max(0.0, float(relative_gap_target))
        self.use_incumbent_cutoff = bool(use_incumbent_cutoff)
        self.verbose_solver = bool(verbose_solver)
        self.relative_gap_tolerance = max(0.0, float(relative_gap_tolerance))
        self.business_gap_tolerance = max(self.relative_gap_tolerance, float(business_gap_tolerance))
        self.intensive_certification = bool(intensive_certification)
        self.intensive_time_limit_seconds = max(1, int(intensive_time_limit_seconds))

        # Parada adaptativa do SCIP: preserva a validade matemática do dual bound,
        # mas interrompe a busca quando o lower bound deixa de avançar por tempo
        # suficiente. O limite rígido time_limit_seconds continua sendo o teto.
        self.adaptive_stop_enabled = bool(adaptive_stop_enabled)
        self.adaptive_min_runtime_seconds = max(1.0, float(adaptive_min_runtime_seconds))
        self.adaptive_patience_seconds = max(1.0, float(adaptive_patience_seconds))
        self.adaptive_check_interval_seconds = max(0.5, float(adaptive_check_interval_seconds))
        self.adaptive_min_relative_lb_improvement = max(
            0.0, float(adaptive_min_relative_lb_improvement)
        )

        # política de escalabilidade. Bounds exponenciais são excelentes em
        # instâncias pequenas, mas nunca devem bloquear a certificação de instâncias
        # médias/grandes. Os limites abaixo são conservadores e independentes do objetivo.
        self.exact_tsp_dp_max_customers = 14
        self.capacitated_partition_dp_max_customers = 14
        self.medium_instance_threshold = 30
        self.large_instance_threshold = 60
        self.very_large_instance_threshold = 100
        self.bound_preprocessing_time_budget_seconds = max(2.0, min(30.0, 0.15 * self.time_limit_seconds))
        self._bound_preprocessing_start = None
        self._scalability_profile = None

        # guardas de complexidade antes da construção do MILP.
        # Estes limites NÃO alteram a região factível: apenas decidem se vale a pena
        # construir o SCIP completo ou emitir um certificado analítico válido.
        self.max_exact_arc_binaries = 120_000
        self.max_exact_total_binary_variables = 180_000
        self.max_exact_sequence_constraints = 100_000
        self.enable_model_complexity_guard = True

        self.objective_context = None
        self.problem = None
        self.global_start_time = None
        self._model_stats = {}

        self.role = (
            "Agente Certificador Exato: especialista em MILP, SCIP, bounds duais, "
            "VRPTW Multi-Trip e prova matemática de ótimo global."
        )
        self.goal = (
            "Provar o ótimo global com um modelo MILP integrado e liberar resultado "
            "definitivo somente quando o solver retorna status OPTIMAL."
        )


    # Utilitários / contrato de dados

    @staticmethod
    def _safe_float(value, default=0.0):
        try:
            if value is None:
                return default
            if isinstance(value, str):
                value = value.strip().replace(",", ".")
                if not value:
                    return default
            return float(value)
        except Exception:
            return default

    @staticmethod
    def _safe_int(value, default=0):
        try:
            return int(value)
        except Exception:
            try:
                return int(float(value))
            except Exception:
                return default

    @staticmethod
    def _parse_bool(value, default=False):
        if value is None:
            return default
        if isinstance(value, bool):
            return value
        if isinstance(value, (int, float)):
            return bool(value)
        txt = str(value).strip().lower()
        if txt in {"1", "true", "sim", "s", "yes", "y"}:
            return True
        if txt in {"0", "2", "false", "não", "nao", "n", "no"}:
            return False
        return default

    @staticmethod
    def _parse_time_to_minutes(value, default=0.0):
        if value is None:
            return float(default)
        if isinstance(value, (int, float)):
            return float(value)
        if hasattr(value, "hour") and hasattr(value, "minute"):
            return float(value.hour) * 60.0 + float(value.minute) + float(getattr(value, "second", 0)) / 60.0
        txt = str(value).strip()
        parts = txt.split(":")
        try:
            if len(parts) >= 2:
                return float(parts[0]) * 60.0 + float(parts[1]) + (float(parts[2]) / 60.0 if len(parts) > 2 else 0.0)
        except Exception:
            pass
        return float(default)

    @staticmethod
    def _normalize_fixed_cost_mode(value):
        txt = str(value or "").strip().upper()
        if txt in {"ALWAYS", "INCONDICIONAL", "UNCONDITIONAL", "SEMPRE", "2"}:
            return "ALWAYS"
        return "WHEN_USED"

    @staticmethod
    def _vehicle_numeric_id(value, fallback):
        if isinstance(value, (int, float)):
            return int(value)
        digits = "".join(ch for ch in str(value or "") if ch.isdigit())
        return int(digits) if digits else int(fallback)

    @staticmethod
    def _find_column(df, candidates):
        if df is None:
            return None
        lower = {str(c).strip().lower(): c for c in df.columns}
        for candidate in candidates:
            key = str(candidate).strip().lower()
            if key in lower:
                return lower[key]
        return None

    @staticmethod
    def _haversine(lat1, lon1, lat2, lon2):
        R = 6371.0
        p1 = math.radians(float(lat1)); p2 = math.radians(float(lat2))
        dphi = math.radians(float(lat2) - float(lat1))
        dlambda = math.radians(float(lon2) - float(lon1))
        a = math.sin(dphi / 2.0) ** 2 + math.cos(p1) * math.cos(p2) * math.sin(dlambda / 2.0) ** 2
        return 2.0 * R * math.asin(math.sqrt(max(0.0, min(1.0, a))))

    def _load_translated_description(self):
        raw = self.shared_memory.data.get("translated_json")
        if isinstance(raw, str):
            try:
                raw = json.loads(raw)
            except Exception:
                raw = {}
        if not isinstance(raw, dict):
            raw = {}
        return raw.get("problem_description", {}) or {}

    def _load_objective_context(self):
        ctx = self.shared_memory.data.get("objective_context")
        if not isinstance(ctx, dict):
            raise ValueError("CERTIFICADOR V14.2: objective_context ausente.")
        if not ctx.get("frozen", False) or not ctx.get("certification_ready", False):
            raise ValueError("CERTIFICADOR V14.2: objective_context precisa estar frozen e certification_ready.")
        active = list(ctx.get("active_objectives", []) or [])
        coeff = ctx.get("coefficients", {}) or {}
        if not active:
            raise ValueError("CERTIFICADOR V14.2: nenhum objetivo ativo.")
        for obj in active:
            if float(coeff.get(obj, -1.0)) < 0:
                raise ValueError(f"CERTIFICADOR V14.2: coeficiente inválido para {obj}.")
        self.objective_context = ctx
        return ctx


    def _external_global_lower_bound(self):
        """Retorna o melhor LB GLOBAL já certificado fora desta camada SCIP.

        Fontes aceitas:
        - continuidade (LB previamente preservado para o mesmo problema/contexto);
        - BoundStrengtheningAgent, somente quando explicitamente GLOBAL_VALID.

        Nunca promove evidência local/heurística a prova global.
        """
        mem = self.shared_memory.data
        candidates = []

        for key in ("continuity_best_valid_lower_bound", "continuity_baseline_valid_lower_bound"):
            val = self._safe_float(mem.get(key), None)
            if val is not None and math.isfinite(float(val)):
                candidates.append((key, float(val)))

        bs = mem.get("bound_strengthening_last_result", {}) or {}
        if isinstance(bs, dict) and str(bs.get("validity", "")).upper().startswith("GLOBAL_VALID"):
            val = self._safe_float(bs.get("best_lower_bound"), None)
            if val is not None and math.isfinite(float(val)):
                candidates.append(("BOUND_STRENGTHENING_GLOBAL", float(val)))

        if not candidates:
            return None, None
        source, value = max(candidates, key=lambda kv: kv[1])
        return float(value), source

    def _merge_external_global_lower_bound(self, lower_bound, lower_source=None):
        """Combina LB da camada com LB global externo via máximo (minimização)."""
        ext_lb, ext_source = self._external_global_lower_bound()
        base = self._safe_float(lower_bound, None)
        if ext_lb is None:
            return base, lower_source, None
        if base is None or ext_lb > base + 1e-15:
            return float(ext_lb), ext_source, float(ext_lb)
        return float(base), lower_source, float(ext_lb)

    def _build_problem(self):
        """Lê os parâmetros do translated_json, o contrato utilizado na modelagem.

        Serviço, utilização mínima, janelas e demais restrições usam as mesmas
        informações fornecidas ao processo de otimização."""
        _vrp_ensure_proof_scope(self.shared_memory)
        mem = self.shared_memory.data
        ans = mem.get("user_answers", {}) or {}
        desc = self._load_translated_description()

        df_clients = mem.get("df_clientes")
        df_caps = mem.get("df_veiculos_capacidade")
        df_costs = mem.get("df_veiculos_custo_km")
        df_windows = mem.get("df_janela_tempo_cliente")
        if df_clients is None or getattr(df_clients, "empty", True):
            raise ValueError("CERTIFICADOR V14.2: df_clientes ausente.")
        if df_caps is None or getattr(df_caps, "empty", True):
            raise ValueError("CERTIFICADOR V14.2: df_veiculos_capacidade ausente.")

        cid_col = self._find_column(df_clients, ["Cliente", "cliente", "id_cliente"])
        lat_col = self._find_column(df_clients, ["latitude", "lat"])
        lon_col = self._find_column(df_clients, ["longitude", "lon", "lng"])
        dem_col = self._find_column(df_clients, ["Demanda", "demanda", "demand"])
        if None in (cid_col, lat_col, lon_col, dem_col):
            raise ValueError("CERTIFICADOR V14.2: colunas obrigatórias de clientes ausentes.")

        client_rows = {}
        for _, row in df_clients.iterrows():
            cid = self._safe_int(row[cid_col], None)
            if cid is None:
                continue
            client_rows[cid] = {
                "lat": float(row[lat_col]), "lon": float(row[lon_col]), "demand": float(row[dem_col])
            }
        if 0 not in client_rows:
            raise ValueError("CERTIFICADOR V14.2: depósito ID 0 não encontrado.")
        customers = sorted(cid for cid in client_rows if cid != 0)

        cap_col = self._find_column(df_caps, ["capacidade", "CAPACIDADE"])
        cap_id_col = self._find_column(df_caps, ["id_veiculo", "ID_VEICULO", "veiculo"])
        if cap_col is None:
            raise ValueError("CERTIFICADOR V14.2: capacidade dos veículos ausente.")
        vehicles = []
        for idx, (_, row) in enumerate(df_caps.reset_index(drop=True).iterrows()):
            vid = self._vehicle_numeric_id(row[cap_id_col], idx + 1) if cap_id_col is not None else idx + 1
            vehicles.append({"idx": idx, "vehicle_id": vid, "capacity": float(row[cap_col]), "cost_per_km": 0.0, "fixed_cost": 0.0})

        if df_costs is not None and not getattr(df_costs, "empty", True):
            idc = self._find_column(df_costs, ["id_veiculo", "ID_VEICULO", "veiculo"])
            cpk = self._find_column(df_costs, ["custo_por_km", "CUSTO_POR_KM", "cost_per_km"])
            fix = self._find_column(df_costs, ["custo_diario", "CUSTO_DIARIO", "custo_fixo"])
            cost_map = {}
            for idx, (_, row) in enumerate(df_costs.reset_index(drop=True).iterrows()):
                vid = self._vehicle_numeric_id(row[idc], idx + 1) if idc is not None else idx + 1
                cost_map[vid] = {
                    "cost_per_km": float(row[cpk]) if cpk is not None else 0.0,
                    "fixed_cost": float(row[fix]) if fix is not None else 0.0,
                }
            for v in vehicles:
                if v["vehicle_id"] in cost_map:
                    v.update(cost_map[v["vehicle_id"]])

        windows = {}
        window_mode = str(desc.get("time_window_mode", "NONE") or "NONE").upper()
        has_time_windows = self._parse_bool(desc.get("has_time_windows"), False)
        if has_time_windows and window_mode == "UNIFORM":
            a = self._safe_float(desc.get("uniform_time_window_start_minutes"), None)
            b = self._safe_float(desc.get("uniform_time_window_end_minutes"), None)
            if a is None or b is None or b <= a:
                raise ValueError("CERTIFICADOR V14.2: janela uniforme inválida no contrato.")
            windows = {cid: (float(a), float(b)) for cid in customers}
        elif has_time_windows and window_mode == "PER_CLIENT":
            if df_windows is None or getattr(df_windows, "empty", True):
                raise ValueError("CERTIFICADOR V14.2: contrato exige janelas por cliente, mas a planilha está ausente.")
            wid = self._find_column(df_windows, ["cliente", "Cliente", "id_cliente"])
            ws_col = self._find_column(df_windows, ["inicio_janela", "início_janela", "inicio"])
            we_col = self._find_column(df_windows, ["fim_janela", "fim"])
            if None in (wid, ws_col, we_col):
                raise ValueError("CERTIFICADOR V14.2: colunas de janelas ausentes.")
            for _, row in df_windows.iterrows():
                cid = self._safe_int(row[wid], None)
                if cid is not None and cid != 0:
                    a = self._parse_time_to_minutes(row[ws_col])
                    b = self._parse_time_to_minutes(row[we_col])
                    if b <= a:
                        raise ValueError(f"CERTIFICADOR V14.2: janela inválida para cliente {cid}.")
                    windows[cid] = (a, b)

        speed = self._safe_float(desc.get("average_vehicle_speed_kmh"), None)
        if speed is None or speed <= 0:
            speed = self._safe_float(ans.get("Qual é a velocidade média dos veículos (em Km/h)?"), 40.0)
        service = self._safe_float(desc.get("service_time_per_delivery_minutes"), None)
        if service is None:
            service = self._safe_float(ans.get("Qual é o tempo médio que cada entrega consume (em minutos)?"), 0.0)
        max_time = self._safe_float(desc.get("max_time_per_vehicle_minutes"), None)
        if max_time is None or max_time <= 0:
            max_time = self._safe_float(ans.get("Tempo disponível por veículo por dia (minutos)?"), 480.0)
        min_usage = self._safe_float(
            desc.get("min_usage_time_minutes", desc.get("minimum_vehicle_usage_minutes")), None
        )
        if min_usage is None:
            min_usage = self._safe_float(ans.get("tempo_minimo_minutos"), 0.0)
        has_min = self._parse_bool(desc.get("has_min_usage_time"), min_usage > 0)
        if not has_min:
            min_usage = 0.0
        allow_multi = self._parse_bool(desc.get("allow_multi_trip"), self._parse_bool(ans.get("Um veiculo pode fazer mais de uma viagem por dia?"), True))
        must_use_all = self._parse_bool(desc.get("must_use_all_vehicles"), self._parse_bool(ans.get("must_use_all_vehicles"), False))
        fixed_mode = self._normalize_fixed_cost_mode(desc.get("fixed_cost_mode", ans.get("fixed_cost_mode", "WHEN_USED")))

        operation_start = self._safe_float(desc.get("vehicle_start_time_minutes"), 0.0)
        operation_end = self._safe_float(desc.get("vehicle_end_time_minutes"), operation_start + max_time)
        if operation_end <= operation_start:
            raise ValueError("CERTIFICADOR V14.2: horizonte operacional inválido.")

        ids = [0] + customers
        dist = {}
        travel = {}
        for i in ids:
            for j in ids:
                d = self._haversine(client_rows[i]["lat"], client_rows[i]["lon"], client_rows[j]["lat"], client_rows[j]["lon"])
                dist[(i, j)] = d
                travel[(i, j)] = d / speed * 60.0

        # Clientes sem janela recebem o horizonte operacional.
        full_windows = {cid: windows.get(cid, (operation_start, operation_end)) for cid in customers}

        p = {
            "customers": customers,
            "client_rows": client_rows,
            "vehicles": vehicles,
            "windows": full_windows,
            "distance": dist,
            "travel_time": travel,
            "speed": speed,
            "service_time": float(service),
            "max_time": float(max_time),
            "minimum_usage": float(min_usage),
            "allow_multi_trip": bool(allow_multi),
            "must_use_all_vehicles": bool(must_use_all),
            "fixed_cost_mode": fixed_mode,
            "cost_data_available": bool(df_costs is not None and not getattr(df_costs, "empty", True)),
            "operation_start": float(operation_start),
            "operation_end": float(operation_end),
        }

        max_cap = max(v["capacity"] for v in vehicles)
        infeasible_demand = [cid for cid in customers if client_rows[cid]["demand"] > max_cap + 1e-9]
        if infeasible_demand:
            raise ValueError(f"CERTIFICADOR V14.2: demanda maior que toda capacidade: {infeasible_demand[:20]}")

        self.problem = p
        mem["certification_operational_contract"] = {
            "service_time_per_delivery_minutes": p["service_time"],
            "minimum_vehicle_usage_minutes": p["minimum_usage"],
            "max_time_per_vehicle_minutes": p["max_time"],
            "average_vehicle_speed_kmh": p["speed"],
            "allow_multi_trip": p["allow_multi_trip"],
            "must_use_all_vehicles": p["must_use_all_vehicles"],
            "fixed_cost_mode": p["fixed_cost_mode"],
            "time_window_mode": window_mode,
            "operation_start_minutes": p["operation_start"],
            "operation_end_minutes": p["operation_end"],
            "time_metric_definition": "SUM_TRIP_ELAPSED_TIME_INCLUDING_TRAVEL_WAIT_SERVICE",
        }
        return p

    def _canonical_score(self, components):
        coeff = self.objective_context["coefficients"]
        return float(sum(
            float(coeff.get(obj, 0.0)) * float(components.get(obj, 0.0))
            for obj in self.objective_context["active_objectives"]
        ))


    def _select_scalability_profile(self):
        """Seleciona uma política de prova sem alterar a matemática do problema.

        A política controla apenas QUAIS bounds opcionais são calculados e o tamanho
        do modelo exato construído. Nunca transforma um lower bound em restrição de
        igualdade nem reduz o espaço de busca sem uma prova de segurança.
        """
        n = len(self.problem.get("customers", [])) if self.problem else 0
        if n <= self.capacitated_partition_dp_max_customers:
            profile = {
                "name": "SMALL_EXACT_BOUNDS",
                "use_tsp_dp": True,
                "use_capacitated_partition_dp": True,
                "build_full_scip": True,
            }
        elif n <= self.medium_instance_threshold:
            profile = {
                "name": "MEDIUM_POLYNOMIAL_BOUNDS",
                "use_tsp_dp": False,
                "use_capacitated_partition_dp": False,
                "build_full_scip": True,
            }
        elif n <= self.large_instance_threshold:
            profile = {
                "name": "LARGE_LIGHT_BOUNDS",
                "use_tsp_dp": False,
                "use_capacitated_partition_dp": False,
                "build_full_scip": True,
            }
        else:
            profile = {
                "name": "VERY_LARGE_LIGHT_BOUNDS",
                "use_tsp_dp": False,
                "use_capacitated_partition_dp": False,
                "build_full_scip": True,
            }
        profile["n_customers"] = n
        self._scalability_profile = profile
        self.shared_memory.data["certification_scalability_profile"] = copy.deepcopy(profile)
        return profile

    def _bound_budget_exhausted(self):
        if self._bound_preprocessing_start is None:
            return False
        return (_cert_wallclock_time.time() - self._bound_preprocessing_start) >= self.bound_preprocessing_time_budget_seconds

    def _held_karp_tsp_lower_bound(self, metric):
        """Resolve exatamente o TSP relaxado depósito+clientes para n<=14.

        Pela desigualdade triangular (Haversine/tempo proporcional à distância),
        o custo do melhor ciclo TSP é <= soma das rotas de qualquer solução VRP:
        concatenar as rotas no depósito e aplicar shortcuts nunca aumenta o custo.
        Logo este valor é um lower bound global seguro para DISTANCE/TRAVEL_TIME.
        """
        p = self.problem
        customers = list(p["customers"])
        n = len(customers)
        if n == 0:
            return 0.0
        profile = self._scalability_profile or self._select_scalability_profile()
        if n > self.exact_tsp_dp_max_customers or not profile.get("use_tsp_dp", False):
            return None
        if self._bound_budget_exhausted():
            return None

        # DP Held-Karp: estado (mask,last) = menor caminho 0 -> subset -> last.
        dp = {}
        for j, cid in enumerate(customers):
            dp[(1 << j, j)] = float(metric[(0, cid)])

        full = (1 << n) - 1
        for mask in range(1, full + 1):
            if (mask & 255) == 0 and self._bound_budget_exhausted():
                return None
            if mask & (mask - 1) == 0:
                continue
            for j in range(n):
                if not (mask & (1 << j)):
                    continue
                prev = mask ^ (1 << j)
                best = float("inf")
                cj = customers[j]
                for i in range(n):
                    if prev & (1 << i):
                        val = dp.get((prev, i))
                        if val is not None:
                            best = min(best, val + float(metric[(customers[i], cj)]))
                if math.isfinite(best):
                    dp[(mask, j)] = best

        best_cycle = min(
            dp[(full, j)] + float(metric[(customers[j], 0)])
            for j in range(n)
            if (full, j) in dp
        )
        return float(best_cycle) if math.isfinite(best_cycle) else None

    def _capacitated_partition_dp_lower_bound(self):
        """Lower bound CVRP por DP exata para n<=14 usando capacidade máxima.

        Relaxações deliberadas: ignora janelas, frota física, utilização mínima,
        horizonte diário e heterogeneidade de capacidade (usa max(capacidade)).
        Cada subconjunto capacitado recebe o menor ciclo depósito-subconjunto-depósito.
        A melhor partição desses subconjuntos é, portanto, um lower bound global
        seguro para a DISTANCE do problema original Multi-Trip/VRPTW.
        """
        p = self.problem
        C = list(p["customers"])
        n = len(C)
        if n == 0:
            return {"distance_lb": 0.0, "trips": 0, "applied": True}
        profile = self._scalability_profile or self._select_scalability_profile()
        if n > self.capacitated_partition_dp_max_customers or not profile.get("use_capacitated_partition_dp", False):
            return {"distance_lb": None, "trips": None, "applied": False, "skip_reason": "SCALABILITY_PROFILE"}
        if self._bound_budget_exhausted():
            return {"distance_lb": None, "trips": None, "applied": False, "skip_reason": "BOUND_PREPROCESSING_BUDGET"}

        max_cap = max(float(v["capacity"]) for v in p["vehicles"])
        demands = [float(p["client_rows"][cid]["demand"]) for cid in C]
        full = (1 << n) - 1

        # Demanda de cada subconjunto em O(2^n).
        subset_demand = [0.0] * (full + 1)
        for mask in range(1, full + 1):
            if (mask & 255) == 0 and self._bound_budget_exhausted():
                return {"distance_lb": None, "trips": None, "applied": False, "skip_reason": "BOUND_PREPROCESSING_BUDGET"}
            bit = mask & -mask
            j = bit.bit_length() - 1
            subset_demand[mask] = subset_demand[mask ^ bit] + demands[j]

        # Held-Karp para todos os subconjuntos: hk[mask,last].
        hk = {}
        for j, cid in enumerate(C):
            hk[(1 << j, j)] = float(p["distance"][(0, cid)])
        for mask in range(1, full + 1):
            if (mask & 255) == 0 and self._bound_budget_exhausted():
                return {"distance_lb": None, "trips": None, "applied": False, "skip_reason": "BOUND_PREPROCESSING_BUDGET"}
            if mask & (mask - 1) == 0:
                continue
            for j in range(n):
                if not (mask & (1 << j)):
                    continue
                prev = mask ^ (1 << j)
                cj = C[j]
                best = float("inf")
                for i in range(n):
                    if prev & (1 << i):
                        val = hk.get((prev, i))
                        if val is not None:
                            best = min(best, val + float(p["distance"][(C[i], cj)]))
                if math.isfinite(best):
                    hk[(mask, j)] = best

        route_cost = [float("inf")] * (full + 1)
        route_cost[0] = 0.0
        for mask in range(1, full + 1):
            if subset_demand[mask] > max_cap + 1e-12:
                continue
            best = float("inf")
            for j in range(n):
                if mask & (1 << j):
                    val = hk.get((mask, j))
                    if val is not None:
                        best = min(best, val + float(p["distance"][(C[j], 0)]))
            route_cost[mask] = best

        # Partição capacitada. Fixar o primeiro bit do mask no submask elimina
        # permutações da mesma partição sem alterar o ótimo da relaxação.
        dp = [float("inf")] * (full + 1)
        trip_count = [10**9] * (full + 1)
        dp[0] = 0.0
        trip_count[0] = 0
        for mask in range(1, full + 1):
            if (mask & 127) == 0 and self._bound_budget_exhausted():
                return {"distance_lb": None, "trips": None, "applied": False, "skip_reason": "BOUND_PREPROCESSING_BUDGET"}
            anchor = mask & -mask
            sub = mask
            while sub:
                if (sub & anchor) and math.isfinite(route_cost[sub]):
                    rem = mask ^ sub
                    cand = dp[rem] + route_cost[sub]
                    cand_trips = trip_count[rem] + 1
                    if cand < dp[mask] - 1e-12 or (
                        abs(cand - dp[mask]) <= 1e-12 and cand_trips < trip_count[mask]
                    ):
                        dp[mask] = cand
                        trip_count[mask] = cand_trips
                sub = (sub - 1) & mask

        if not math.isfinite(dp[full]):
            return {"distance_lb": None, "trips": None, "applied": True}
        return {
            "distance_lb": float(dp[full]),
            "trips": int(trip_count[full]),
            "applied": True,
            "relaxations": [
                "TIME_WINDOWS", "PHYSICAL_VEHICLE_COUNT", "MINIMUM_USAGE",
                "DAILY_HORIZON", "HETEROGENEOUS_CAPACITY_TO_MAX_CAPACITY",
            ],
        }

    def _mst_distance_lower_bound(self):
        """MST em depósito+clientes: lower bound métrico O(n²) para distância total."""
        p = self.problem
        nodes = [0] + list(p["customers"])
        if len(nodes) <= 1:
            return 0.0
        in_tree = {nodes[0]}
        best = {j: float(p["distance"][(nodes[0], j)]) for j in nodes[1:]}
        total = 0.0
        while len(in_tree) < len(nodes):
            j = min((u for u in nodes if u not in in_tree), key=lambda u: best.get(u, float("inf")))
            w = best.get(j, float("inf"))
            if not math.isfinite(w):
                return None
            total += w
            in_tree.add(j)
            for u in nodes:
                if u not in in_tree:
                    d = float(p["distance"][(j, u)])
                    if d < best.get(u, float("inf")):
                        best[u] = d
        return float(total)

    def _radial_capacity_distance_lower_bound(self):
        """LB radial CVRP: (2/Qmax) * sum(q_i*d(0,i)), válido com capacidade máxima."""
        p = self.problem
        C = list(p["customers"])
        if not C:
            return 0.0
        qmax = max(float(v["capacity"]) for v in p["vehicles"])
        if qmax <= 1e-12:
            return None
        weighted_radius = sum(
            float(p["client_rows"][i]["demand"]) * float(p["distance"][(0, i)])
            for i in C
        )
        return float(2.0 * weighted_radius / qmax)

    def _build_certified_bound_catalog(self, values):
        """Registra proveniência e aplicabilidade dos bounds usados no certificado."""
        active = list(self.objective_context.get("active_objectives", []) or [])
        coeff = self.objective_context.get("coefficients", {}) or {}
        catalog = []

        def add(name, component, value, method, assumptions=None, used=True):
            valid = value is not None and math.isfinite(float(value))
            catalog.append({
                "name": name,
                "component": component,
                "value": (float(value) if valid else None),
                "method": method,
                "valid_global_lower_bound": bool(valid),
                "active_in_objective": component in active,
                "canonical_coefficient": float(coeff.get(component, 0.0)),
                "used_in_canonical_bound": bool(valid and component in active and used),
                "assumptions": list(assumptions or []),
            })

        add("DISTANCE_CERTIFIED_LB", "DISTANCE", values.get("distance_lb"),
            values.get("distance_lb_source", "SAFE_RELAXATION"))
        add("TIME_CERTIFIED_LB", "TIME", values.get("time_lb"),
            values.get("time_lb_source", "SERVICE_PLUS_TRAVEL_SAFE_RELAXATION"),
            ["fixed nonnegative service time", "uniform positive vehicle speed"])
        add("COST_CERTIFIED_LB", "COST", values.get("cost_lb"),
            "NONNEGATIVE_COST_RELAXATION",
            ["nonnegative cost_per_km", "nonnegative fixed costs"])
        add("VEHICLES_CERTIFIED_LB", "VEHICLES", values.get("vehicles_lb"),
            "WORKLOAD_CAPACITY_HORIZON_RELAXATION")
        return catalog

    def _physical_certificate_interval(self, upper_bound, lower_bound):
        """Intervalo legível quando há exatamente um objetivo ativo."""
        active = list(self.objective_context.get("active_objectives", []) or [])
        if len(active) != 1 or upper_bound is None or lower_bound is None:
            return None
        obj = active[0]
        coeff = float((self.objective_context.get("coefficients", {}) or {}).get(obj, 0.0))
        if coeff <= 0:
            return None
        lb_phys = float(lower_bound) / coeff
        ub_phys = float(upper_bound) / coeff
        units = {"TIME": "min", "DISTANCE": "km", "COST": "R$", "VEHICLES": "veículos"}
        return {
            "objective": obj,
            "lower_bound_physical": lb_phys,
            "upper_bound_physical": ub_phys,
            "remaining_absolute_uncertainty_physical": max(0.0, ub_phys - lb_phys),
            "unit": units.get(obj, "unidade_objetivo"),
        }

    def _safe_trip_slot_upper_bound_from_incumbent(self, incumbent_ub, base_upper):
        """Aperta K somente quando a prova é segura.

        Por prudência, a redução por incumbent é aplicada apenas ao objetivo puro TIME.
        Se há k viagens, cada viagem ativa contém >=1 cliente e exige ao menos o menor
        ciclo depósito-cliente-depósito. Assim:
            TIME >= n*service + k*min_roundtrip_travel.
        Se uma solução deve melhorar o UB atual, essa desigualdade fornece um upper
        bound seguro para k. Nunca usa simplesmente o número de viagens do incumbent.
        """
        p = self.problem
        active = list(self.objective_context.get("active_objectives", []) or [])
        if active != ["TIME"] or incumbent_ub is None or not math.isfinite(float(incumbent_ub)):
            return int(base_upper), "STRUCTURAL_SAFE_UPPER_BOUND"

        coeff_time = float((self.objective_context.get("coefficients", {}) or {}).get("TIME", 0.0))
        if coeff_time <= 0:
            return int(base_upper), "STRUCTURAL_SAFE_UPPER_BOUND"

        customers = list(p["customers"])
        if not customers:
            return 0, "NO_CUSTOMERS"

        # UB canônico / coeficiente TIME = UB físico em minutos no objetivo puro TIME.
        ub_time = float(incumbent_ub) / coeff_time
        service_total = len(customers) * float(p["service_time"])
        min_roundtrip = min(
            float(p["travel_time"][(0, i)]) + float(p["travel_time"][(i, 0)])
            for i in customers
        )
        if min_roundtrip <= 1e-12:
            return int(base_upper), "STRUCTURAL_SAFE_UPPER_BOUND"

        # Para uma solução estritamente melhor que o incumbent:
        # service_total + k*min_roundtrip <= ub_time.
        k_ub = int(math.floor((ub_time - service_total) / min_roundtrip + 1e-12))
        k_ub = max(1, min(int(base_upper), k_ub))
        return k_ub, "INCUMBENT_TIME_SAFE_UPPER_BOUND"

    def _analytical_lower_bounds(self, incumbent_ub=None):
        """Calcula bounds globais seguros, independentes do incumbent.

        Os bounds são válidos para mono e multiobjetivo porque cada componente
        recebe um limite inferior próprio e a função objetivo canônica usa
        coeficientes não negativos.
        """
        p = self.problem
        C = list(p["customers"])
        V = list(p["vehicles"])
        self._bound_preprocessing_start = _cert_wallclock_time.time()
        scalability_profile = self._select_scalability_profile()
        if not C:
            bounds = {
                "distance_lb": 0.0,
                "travel_time_lb": 0.0,
                "service_time_lb": 0.0,
                "time_lb": 0.0,
                "cost_lb": 0.0,
                "vehicles_lb": 0,
                "trips_lb": 0,
                "canonical_objective_lb": 0.0,
                "trip_slots_safe_upper_bound": 0,
            }
            self.shared_memory.data["certification_strengthened_bounds"] = copy.deepcopy(bounds)
            return bounds

        nodes = [0] + C
        min_in_dist = 0.0
        min_out_dist = 0.0
        min_in_time = 0.0
        min_out_time = 0.0

        for i in C:
            min_in_dist += min(float(p["distance"][(j, i)]) for j in nodes if j != i)
            min_out_dist += min(float(p["distance"][(i, j)]) for j in nodes if j != i)
            min_in_time += min(float(p["travel_time"][(j, i)]) for j in nodes if j != i)
            min_out_time += min(float(p["travel_time"][(i, j)]) for j in nodes if j != i)

        degree_distance_lb = max(min_in_dist, min_out_dist)
        degree_travel_time_lb = max(min_in_time, min_out_time)
        mst_distance_lb = self._mst_distance_lower_bound()
        radial_capacity_distance_lb = self._radial_capacity_distance_lower_bound()

        tsp_distance_lb = self._held_karp_tsp_lower_bound(p["distance"])
        tsp_travel_time_lb = (
            float(tsp_distance_lb) / float(p["speed"]) * 60.0
            if tsp_distance_lb is not None and float(p["speed"]) > 0
            else None
        )
        capacitated_dp = self._capacitated_partition_dp_lower_bound()
        capacitated_distance_lb = capacitated_dp.get("distance_lb")
        capacitated_travel_time_lb = (
            float(capacitated_distance_lb) / float(p["speed"]) * 60.0
            if capacitated_distance_lb is not None and float(p["speed"]) > 0
            else None
        )

        distance_candidates = {
            "DEGREE": float(degree_distance_lb),
            "MST": (float(mst_distance_lb) if mst_distance_lb is not None else None),
            "RADIAL_CAPACITY": (float(radial_capacity_distance_lb) if radial_capacity_distance_lb is not None else None),
            "TSP_HELD_KARP": float(tsp_distance_lb) if tsp_distance_lb is not None else None,
            "CAPACITATED_PARTITION_DP": (
                float(capacitated_distance_lb) if capacitated_distance_lb is not None else None
            ),
        }
        valid_distance_candidates = {k: v for k, v in distance_candidates.items() if v is not None}
        distance_lb_source, distance_lb = max(valid_distance_candidates.items(), key=lambda kv: kv[1])

        travel_candidates = {
            name: (float(value) / float(p["speed"]) * 60.0 if value is not None and float(p["speed"]) > 0 else None)
            for name, value in distance_candidates.items()
        }
        valid_travel_candidates = {k: v for k, v in travel_candidates.items() if v is not None}
        travel_time_lb_source, travel_time_lb = max(valid_travel_candidates.items(), key=lambda kv: kv[1])
        service_time_lb = float(len(C)) * float(p["service_time"])
        time_lb = service_time_lb + travel_time_lb

        total_demand = sum(float(p["client_rows"][i]["demand"]) for i in C)
        max_capacity = max(float(v["capacity"]) for v in V)
        trips_lb = int(math.ceil(total_demand / max_capacity - 1e-12)) if max_capacity > 0 else len(C)
        trips_lb = max(1, min(len(C), trips_lb))

        H = max(1e-12, float(p["max_time"]))
        if p["must_use_all_vehicles"]:
            vehicles_lb = len(V)
        else:
            vehicles_lb = max(1, int(math.ceil(time_lb / H - 1e-12)))
            if not p["allow_multi_trip"]:
                vehicles_lb = max(vehicles_lb, trips_lb)
        # incorporar diretamente no MILP o melhor vehicle-LB global
        # já provado pelo K-Certifier/Bound. Isso fortalece a relaxação e evita
        # que a certificação global esqueça uma prova estrutural válida.
        mem = self.shared_memory.data
        persisted_vehicle_lb = self._safe_int(mem.get("certified_vehicle_lower_bound"), 1)
        try:
            for _kp in (mem.get("k_vehicle_feasibility_history", []) or []):
                if (
                    isinstance(_kp, dict)
                    and str(_kp.get("status", "")).upper() == "PROVEN_INFEASIBLE"
                    and _kp.get("global_valid") is True
                ):
                    persisted_vehicle_lb = max(
                        persisted_vehicle_lb,
                        self._safe_int(_kp.get("proven_vehicle_lower_bound"), 1),
                    )
        except Exception:
            pass
        vehicles_lb = max(vehicles_lb, int(persisted_vehicle_lb or 1))
        vehicles_lb = min(len(V), vehicles_lb)

        if p["minimum_usage"] > 0:
            if p["must_use_all_vehicles"]:
                time_lb = max(time_lb, float(p["minimum_usage"]) * len(V))
            else:
                time_lb = max(time_lb, float(p["minimum_usage"]) * vehicles_lb)

        min_single_trip_duration = min(
            float(p["service_time"])
            + float(p["travel_time"][(0, i)])
            + float(p["travel_time"][(i, 0)])
            for i in C
        )
        if not p["allow_multi_trip"]:
            trip_slots_upper = min(len(C), len(V))
        elif min_single_trip_duration > 1e-12:
            total_vehicle_work_capacity = float(len(V)) * H
            trip_slots_upper = min(
                len(C),
                max(1, int(math.floor(total_vehicle_work_capacity / min_single_trip_duration + 1e-12)))
            )
        else:
            trip_slots_upper = len(C)

        structural_trip_slots_upper = int(trip_slots_upper)
        trip_slots_upper, trip_slots_source = self._safe_trip_slot_upper_bound_from_incumbent(
            incumbent_ub, structural_trip_slots_upper
        )

        all_nonnegative_costs = all(
            float(v["cost_per_km"]) >= -1e-12 and float(v["fixed_cost"]) >= -1e-12
            for v in V
        )
        cost_lb = None
        if all_nonnegative_costs:
            min_cost_per_km = min(float(v["cost_per_km"]) for v in V)
            variable_cost_lb = min_cost_per_km * distance_lb
            if p["fixed_cost_mode"] == "ALWAYS":
                fixed_cost_lb = sum(float(v["fixed_cost"]) for v in V)
            else:
                cheapest_fixed = sorted(float(v["fixed_cost"]) for v in V)
                fixed_cost_lb = sum(cheapest_fixed[:vehicles_lb])
            cost_lb = variable_cost_lb + fixed_cost_lb

        component_lbs = {
            "DISTANCE": distance_lb,
            "TIME": time_lb,
            "COST": cost_lb,
            "VEHICLES": float(vehicles_lb),
        }
        canonical_lb = 0.0
        canonical_valid = True
        coeff = self.objective_context.get("coefficients", {}) or {}
        for obj in self.objective_context.get("active_objectives", []) or []:
            lb = component_lbs.get(obj)
            if lb is None:
                canonical_valid = False
                break
            canonical_lb += float(coeff.get(obj, 0.0)) * float(lb)

        bounds = {
            "distance_lb": float(distance_lb),
            "travel_time_lb": float(travel_time_lb),
            "service_time_lb": float(service_time_lb),
            "time_lb": float(time_lb),
            "cost_lb": (None if cost_lb is None else float(cost_lb)),
            "vehicles_lb": int(vehicles_lb),
            "trips_lb": int(trips_lb),
            "canonical_objective_lb": float(canonical_lb) if canonical_valid else None,
            "canonical_objective_lb_valid": bool(canonical_valid),
            "trip_slots_safe_upper_bound": int(trip_slots_upper),
            "trip_slots_source": trip_slots_source,
            "trip_slots_structural_upper_bound": int(structural_trip_slots_upper),
            "tsp_dp_applied": bool(tsp_distance_lb is not None),
            "scalability_profile": copy.deepcopy(scalability_profile),
            "bound_preprocessing_elapsed_seconds": float(_cert_wallclock_time.time() - self._bound_preprocessing_start),
            "bound_preprocessing_budget_seconds": float(self.bound_preprocessing_time_budget_seconds),
            "tsp_distance_lb": (None if tsp_distance_lb is None else float(tsp_distance_lb)),
            "tsp_travel_time_lb": (None if tsp_travel_time_lb is None else float(tsp_travel_time_lb)),
            "degree_distance_lb": float(degree_distance_lb),
            "degree_travel_time_lb": float(degree_travel_time_lb),
            "mst_distance_lb": (None if mst_distance_lb is None else float(mst_distance_lb)),
            "radial_capacity_distance_lb": (None if radial_capacity_distance_lb is None else float(radial_capacity_distance_lb)),
            "capacitated_partition_dp_applied": bool(capacitated_dp.get("applied", False)),
            "capacitated_partition_distance_lb": (
                None if capacitated_distance_lb is None else float(capacitated_distance_lb)
            ),
            "capacitated_partition_travel_time_lb": (
                None if capacitated_travel_time_lb is None else float(capacitated_travel_time_lb)
            ),
            "capacitated_partition_relaxed_trip_count": capacitated_dp.get("trips"),
            "capacitated_partition_relaxations": copy.deepcopy(capacitated_dp.get("relaxations", [])),
            "distance_lb_source": distance_lb_source,
            "travel_time_lb_source": travel_time_lb_source,
            "time_lb_source": f"SERVICE_PLUS_{travel_time_lb_source}",
            "note": (
                "Bounds analíticos seguros: graus mínimos, MST, radial-capacity, serviço inevitável, "
                "capacidade por viagem e jornada máxima por veículo."
            ),
        }
        bounds["certified_bound_catalog"] = self._build_certified_bound_catalog(bounds)
        bounds["objective_aware"] = True
        bounds["active_objectives"] = list(self.objective_context.get("active_objectives", []) or [])
        self.shared_memory.data["certification_bound_catalog"] = copy.deepcopy(bounds["certified_bound_catalog"])
        self.shared_memory.data["certification_strengthened_bounds"] = copy.deepcopy(bounds)
        return bounds


    # Pré-processamento de arcos

    def _build_feasible_arcs(self):
        p = self.problem
        C = p["customers"]; maxcap = max(v["capacity"] for v in p["vehicles"])
        start = p["operation_start"]; end = p["operation_end"]; service = p["service_time"]
        arcs = []
        pruned_window = 0; pruned_capacity_pair = 0; possible = 0
        nodes = [0] + C
        for i in nodes:
            for j in nodes:
                if i == j:
                    continue
                if i == 0 and j == 0:
                    continue
                possible += 1
                feasible = True
                if i == 0 and j != 0:
                    wsj, wej = p["windows"][j]
                    earliest_j = max(start + p["travel_time"][(0, j)], wsj)
                    if earliest_j > wej + 1e-9 or earliest_j + service + p["travel_time"][(j, 0)] > end + 1e-9:
                        feasible = False; pruned_window += 1
                elif i != 0 and j == 0:
                    wsi, _ = p["windows"][i]
                    earliest_i = max(wsi, start + p["travel_time"][(0, i)])
                    if earliest_i + service + p["travel_time"][(i, 0)] > end + 1e-9:
                        feasible = False; pruned_window += 1
                else:
                    if p["client_rows"][i]["demand"] + p["client_rows"][j]["demand"] > maxcap + 1e-9:
                        feasible = False; pruned_capacity_pair += 1
                    if feasible:
                        wsi, _ = p["windows"][i]; _, wej = p["windows"][j]
                        earliest_i = max(wsi, start + p["travel_time"][(0, i)])
                        if earliest_i + service + p["travel_time"][(i, j)] > wej + 1e-9:
                            feasible = False; pruned_window += 1
                if feasible:
                    arcs.append((i, j))
        stats = {
            "possible_arcs": possible,
            "feasible_arcs": len(arcs),
            "pruned_arcs": possible - len(arcs),
            "pruned_due_window": pruned_window,
            "pruned_due_capacity_pair": pruned_capacity_pair,
        }
        self.shared_memory.data["certification_preprocessing"] = stats
        return arcs, stats

    def _estimate_model_complexity(self, arcs, strengthened_bounds):
        """Estima o tamanho do MILP antes de alocar variáveis SCIP."""
        p = self.problem
        n = len(p["customers"]); nv = len(p["vehicles"])
        K = int(strengthened_bounds.get("trip_slots_safe_upper_bound", n))
        K = max(0, K)
        arc_bin = K * len(arcs)
        visit_bin = K * n
        assign_bin = K * nv
        vehicle_bin = nv
        total_bin = arc_bin + visit_bin + assign_bin + vehicle_bin
        seq = nv * K * max(0, K - 1) // 2
        reasons = []
        if arc_bin > self.max_exact_arc_binaries:
            reasons.append("ARC_BINARIES")
        if total_bin > self.max_exact_total_binary_variables:
            reasons.append("TOTAL_BINARIES")
        if seq > self.max_exact_sequence_constraints:
            reasons.append("SEQUENCE_CONSTRAINTS")
        stats = {
            "n_customers": n, "n_vehicles": nv, "trip_slots": K,
            "feasible_arcs_per_slot": len(arcs),
            "estimated_binary_arc_variables": arc_bin,
            "estimated_binary_visit_variables": visit_bin,
            "estimated_binary_assignment_variables": assign_bin,
            "estimated_binary_vehicle_variables": vehicle_bin,
            "estimated_total_binary_variables": total_bin,
            "estimated_sequence_constraints": seq,
            "guard_triggered": bool(self.enable_model_complexity_guard and reasons),
            "guard_reasons": reasons,
            "thresholds": {
                "max_exact_arc_binaries": self.max_exact_arc_binaries,
                "max_exact_total_binary_variables": self.max_exact_total_binary_variables,
                "max_exact_sequence_constraints": self.max_exact_sequence_constraints,
            },
        }
        self.shared_memory.data["certification_model_complexity_estimate"] = copy.deepcopy(stats)
        return stats


    # Construção do MILP SCIP

    def _import_scip(self):
        try:
            from pyscipopt import Model, quicksum
            return Model, quicksum
        except Exception as exc:
            raise ImportError(
                "PySCIPOpt não está instalado. No Colab execute antes: !pip install -q pyscipopt"
            ) from exc

    def _build_scip_model(self, incumbent_ub=None, feasibility_only=False, max_active_vehicles=None):
        Model, quicksum = self._import_scip()
        p = self.problem; C = p["customers"]; V = p["vehicles"]
        arcs, prep = self._build_feasible_arcs()
        outgoing = defaultdict(list); incoming = defaultdict(list)
        for i, j in arcs:
            outgoing[i].append(j); incoming[j].append(i)

        strengthened_bounds = self._analytical_lower_bounds(incumbent_ub=incumbent_ub)
        # Recompute with precisely the same distance matrix as this MILP.
        # The helper only returns a bound after an optimal SCF relaxation.
        scf_cache = self.shared_memory.data.setdefault("scf_model_cache_v51", {})
        scf = _vrp_scf_distance_bound(p, strengthened_bounds["trips_lb"], 30.0, scf_cache)
        if scf.get("valid") and scf.get("distance_lb") is not None:
            safe_d = max(0.0, float(scf["distance_lb"])-1e-7)
            strengthened_bounds["distance_lb"] = max(strengthened_bounds["distance_lb"],safe_d)
            rates=[float(v["cost_per_km"]) for v in V]
            fixed=sorted(float(v["fixed_cost"]) for v in V)
            if rates and min(rates)>=0 and min(fixed)>=0:
                fixed_lb=(sum(fixed) if p["fixed_cost_mode"] != "WHEN_USED" or p["must_use_all_vehicles"]
                          else sum(fixed[:int(strengthened_bounds["vehicles_lb"])]))
                strengthened_bounds["cost_lb"]=max(strengthened_bounds["cost_lb"],fixed_lb+min(rates)*safe_d)
        strengthened_bounds["scf_model_result"] = scf
        strengthened_bounds["canonical_objective_lb"] = sum(
            float(self.objective_context["coefficients"].get(key,0))*float(strengthened_bounds[field])
            for key,field in [("DISTANCE","distance_lb"),("COST","cost_lb"),("TIME","time_lb"),("VEHICLES","vehicles_lb")])

        # Limite superior seguro de slots. Continua exato: nunca usa o número de
        # viagens do incumbent como limite, apenas dados estruturais do problema.
        K = int(strengthened_bounds["trip_slots_safe_upper_bound"])
        if C and K <= 0:
            K = 1
        # Limite por trabalho: n*serviço + viagens*(menor saída + menor retorno).
        # É seguro também sem desigualdade triangular; cada viagem contém esses arcos.
        active_cap = len(V) if max_active_vehicles is None else min(len(V),int(max_active_vehicles))
        if C and active_cap > 0:
            edge_time = min(p["travel_time"][0,i] for i in C) + min(p["travel_time"][i,0] for i in C)
            if edge_time > 1e-12:
                residual = active_cap*float(p["max_time"])-len(C)*float(p["service_time"])
                K = min(K,max(1,int(math.floor(residual/edge_time+1e-9))))
            if not p["allow_multi_trip"]:
                K = min(K,active_cap)
        T = list(range(K))
        H = float(p["max_time"])
        op0 = float(p["operation_start"]); op1 = float(p["operation_end"])
        horizon_span = max(1.0, op1 - op0)
        maxdist = max(p["distance"].values()) if p["distance"] else 0.0
        Mdist = max(1.0, (len(C) + 1) * maxdist)

        model = Model("VRPTW_MULTITRIP_EXACT_V14_1_SCALABLE")
        if not self.verbose_solver:
            model.hideOutput()
        model.setParam("limits/time", float(self.time_limit_seconds))
        if self.relative_gap_target > 0:
            model.setParam("limits/gap", float(self.relative_gap_target))
        else:
            model.setParam("limits/gap", 0.0)
        if self.scip_memory_mb > 0:
            try: model.setParam("limits/memory", float(self.scip_memory_mb))
            except Exception: pass
        if self.scip_threads > 0:
            try: model.setParam("parallel/maxnthreads", int(self.scip_threads))
            except Exception: pass

        z = {t: model.addVar(vtype="B", name=f"trip_active_{t}") for t in T}
        y = {(t, i): model.addVar(vtype="B", name=f"visit_{t}_{i}") for t in T for i in C}
        x = {(t, i, j): model.addVar(vtype="B", name=f"arc_{t}_{i}_{j}") for t in T for (i, j) in arcs}
        assign = {(t, v["idx"]): model.addVar(vtype="B", name=f"assign_{t}_v{v['idx']}") for t in T for v in V}
        used = {v["idx"]: model.addVar(vtype="B", name=f"vehicle_used_{v['idx']}") for v in V}

        s = {t: model.addVar(lb=op0, ub=op1, vtype="C", name=f"trip_start_{t}") for t in T}
        e = {t: model.addVar(lb=op0, ub=op1, vtype="C", name=f"trip_end_{t}") for t in T}
        dur = {t: model.addVar(lb=0.0, ub=H, vtype="C", name=f"trip_duration_{t}") for t in T}
        # tau fica diretamente limitado pela janela do cliente, mesmo quando o
        # cliente não pertence ao slot. Isso é seguro e permite Big-M muito menor.
        tau = {
            (t, i): model.addVar(
                lb=float(p["windows"][i][0]),
                ub=float(p["windows"][i][1]),
                vtype="C",
                name=f"service_{t}_{i}",
            )
            for t in T for i in C
        }
        # Ordem MTZ elimina subtours mesmo em casos degenerados com tempo de serviço
        # e distância iguais a zero. Isso torna a conectividade independente da
        # propagação temporal.
        pos = {} if float(p["service_time"]) > 0 else {(t, i): model.addVar(lb=0.0, ub=float(len(C)), vtype="C", name=f"pos_{t}_{i}") for t in T for i in C}
        tripdist = {t: model.addVar(lb=0.0, ub=Mdist, vtype="C", name=f"trip_dist_{t}") for t in T}
        qdist = {} if feasibility_only else {(t, v["idx"]): model.addVar(lb=0.0, ub=Mdist, vtype="C", name=f"dist_t{t}_v{v['idx']}") for t in T for v in V}
        qdur = {(t, v["idx"]): model.addVar(lb=0.0, ub=H, vtype="C", name=f"dur_t{t}_v{v['idx']}") for t in T for v in V}

        # Cobertura exata dos clientes.
        for i in C:
            model.addCons(quicksum(y[t, i] for t in T) == 1, name=f"cover_{i}")

        # Estrutura de cada viagem.
        for t in T:
            model.addCons(quicksum(x[t, 0, j] for j in outgoing[0] if (t, 0, j) in x) == z[t], name=f"depart_{t}")
            model.addCons(quicksum(x[t, i, 0] for i in incoming[0] if (t, i, 0) in x) == z[t], name=f"return_{t}")
            for i in C:
                model.addCons(quicksum(x[t, i, j] for j in outgoing[i] if (t, i, j) in x) == y[t, i], name=f"flowout_{t}_{i}")
                model.addCons(quicksum(x[t, j, i] for j in incoming[i] if (t, j, i) in x) == y[t, i], name=f"flowin_{t}_{i}")
                model.addCons(y[t, i] <= z[t])
            model.addCons(dur[t] == e[t] - s[t])
            model.addCons(dur[t] <= H * z[t])
            model.addCons(tripdist[t] == quicksum(float(p["distance"][(i, j)]) * x[t, i, j] for (i, j) in arcs))

            # Uma viagem ativa deve conter ao menos um cliente.
            model.addCons(quicksum(y[t, i] for i in C) >= z[t], name=f"nonempty_{t}")

            for i in C:
                ws, we = p["windows"][i]
                if pos:
                    model.addCons(pos[t, i] >= y[t, i])
                    model.addCons(pos[t, i] <= float(len(C)) * y[t, i])

                # Cliente só pode ser visitado por veículo capaz de transportar
                # individualmente sua demanda. Fortalece a relaxação LP.
                eligible = [
                    v["idx"] for v in V
                    if float(v["capacity"]) + 1e-12 >= float(p["client_rows"][i]["demand"])
                ]
                model.addCons(
                    y[t, i] <= quicksum(assign[t, vi] for vi in eligible),
                    name=f"eligible_vehicle_{t}_{i}",
                )

                if (t, 0, i) in x:
                    m_0i = max(
                        0.0,
                        op1 + float(p["travel_time"][(0, i)]) - float(ws)
                    )
                    model.addCons(
                        tau[t, i] >= s[t] + float(p["travel_time"][(0, i)])
                        - m_0i * (1 - x[t, 0, i])
                    )

            for i, j in arcs:
                if i != 0 and j != 0:
                    _, we_i = p["windows"][i]
                    ws_j, _ = p["windows"][j]
                    m_ij = max(
                        0.0,
                        float(we_i) + float(p["service_time"])
                        + float(p["travel_time"][(i, j)]) - float(ws_j)
                    )
                    model.addCons(
                        tau[t, j] >= tau[t, i]
                        + float(p["service_time"] + p["travel_time"][(i, j)])
                        - m_ij * (1 - x[t, i, j])
                    )
                    if pos:
                        model.addCons(
                            pos[t, j] >= pos[t, i] + 1.0
                            - float(len(C)) * (1 - x[t, i, j])
                        )
                elif i != 0 and j == 0:
                    ws_i, we_i = p["windows"][i]
                    return_time = float(p["service_time"] + p["travel_time"][(i, 0)])
                    rhs = tau[t, i] + return_time
                    m_ret_lb = max(0.0, float(we_i) + return_time - op0)
                    m_ret_ub = max(0.0, op1 - (float(ws_i) + return_time))
                    model.addCons(e[t] >= rhs - m_ret_lb * (1 - x[t, i, 0]))
                    model.addCons(e[t] <= rhs + m_ret_ub * (1 - x[t, i, 0]))

            # Uma viagem ativa pertence a exatamente um veículo físico.
            model.addCons(quicksum(assign[t, v["idx"]] for v in V) == z[t])

            # Capacidade por veículo: formulação mais forte que a capacidade média
            # ponderada pela atribuição na relaxação LP.
            load_expr = quicksum(
                float(p["client_rows"][i]["demand"]) * y[t, i] for i in C
            )
            total_demand = sum(float(p["client_rows"][i]["demand"]) for i in C)

            for v in V:
                vi = v["idx"]
                model.addCons(assign[t, vi] <= used[vi])
                model.addCons(
                    load_expr <= float(v["capacity"]) + total_demand * (1 - assign[t, vi]),
                    name=f"capacity_t{t}_v{vi}",
                )
                # No modo de factibilidade, custo não participa da prova.
                if not feasibility_only:
                    model.addCons(qdist[t, vi] <= tripdist[t])
                    model.addCons(qdist[t, vi] <= Mdist * assign[t, vi])
                    model.addCons(qdist[t, vi] >= tripdist[t] - Mdist * (1 - assign[t, vi]))
                # linearização duração * atribuição
                model.addCons(qdur[t, vi] <= dur[t])
                model.addCons(qdur[t, vi] <= H * assign[t, vi])
                model.addCons(qdur[t, vi] >= dur[t] - H * (1 - assign[t, vi]))

        # Fortalecimento da relaxação, sem excluir soluções inteiras válidas.
        for t in T:
            model.addCons(dur[t] >= float(p["service_time"])*quicksum(y[t,i] for i in C)
                          + quicksum(float(p["travel_time"][i,j])*x[t,i,j] for i,j in arcs),
                          name=f"trip_workload_{t}")
            model.addCons(quicksum(float(p["client_rows"][i]["demand"])*y[t,i] for i in C)
                          <= quicksum(float(v["capacity"])*assign[t,v["idx"]] for v in V),
                          name=f"aggregate_capacity_{t}")
            model.addCons(quicksum(qdur[t,v["idx"]] for v in V) == dur[t])
            if not feasibility_only:
                model.addCons(quicksum(qdist[t,v["idx"]] for v in V) == tripdist[t])
        if max_active_vehicles is not None:
            model.addCons(quicksum(used.values()) <= int(max_active_vehicles), name="hard_k_cardinality")

        # Redução de simetria dos slots globais:
        # 1) slots ativos formam um prefixo;
        # 2) slots ativos são ordenados cronologicamente.
        # Como os slots são rótulos artificiais, essa ordenação não remove
        # nenhuma solução física e elimina grande parte das permutações equivalentes.
        for t in range(K - 1):
            model.addCons(z[t] >= z[t + 1], name=f"active_prefix_{t}")
            model.addCons(
                s[t] <= s[t + 1] + horizon_span * (1 - z[t + 1]),
                name=f"chronological_slots_{t}",
            )

        # Uso de veículos, Multi-Trip e utilização mínima.
        for v in V:
            vi = v["idx"]
            model.addCons(used[vi] <= quicksum(assign[t, vi] for t in T))
            if p["must_use_all_vehicles"]:
                model.addCons(used[vi] == 1)
            if not p["allow_multi_trip"]:
                model.addCons(quicksum(assign[t, vi] for t in T) <= 1)
            model.addCons(quicksum(qdur[t, vi] for t in T) <= H * used[vi])
            if p["minimum_usage"] > 0:
                model.addCons(quicksum(qdur[t, vi] for t in T) >= float(p["minimum_usage"]) * used[vi])

        # Não sobreposição sem variáveis binárias pairwise de ordem.
        # A ordem cronológica global dos slots permite impor diretamente que,
        # se dois slots a<b pertencem ao mesmo veículo, b só inicia após o fim de a.
        # Prefix end dominates the end of every earlier trip of each vehicle.
        # Together with chronological slots this is equivalent to all pairwise
        # nonoverlap constraints, with O(K*V) instead of O(K*K*V) rows.
        prefix_end = {(t,v["idx"]): model.addVar(lb=op0,ub=op1,vtype="C",
                        name=f"prefix_end_{t}_{v['idx']}") for t in T for v in V}
        sequence_constraints = 0
        for t in T:
            for v in V:
                vi=v["idx"]
                model.addCons(prefix_end[t,vi] >= e[t]-horizon_span*(1-assign[t,vi]))
                sequence_constraints += 1
                if t:
                    model.addCons(prefix_end[t,vi] >= prefix_end[t-1,vi])
                    model.addCons(s[t] >= prefix_end[t-1,vi]-horizon_span*(1-assign[t,vi]))
                    sequence_constraints += 2

        # Componentes da função objetivo canônica.
        total_distance = quicksum(tripdist[t] for t in T)
        total_time = quicksum(dur[t] for t in T)
        total_variable_cost = 0.0 if feasibility_only else quicksum(float(v["cost_per_km"]) * qdist[t, v["idx"]] for t in T for v in V)
        if p["fixed_cost_mode"] == "WHEN_USED":
            total_fixed_cost = quicksum(float(v["fixed_cost"]) * used[v["idx"]] for v in V)
        else:
            total_fixed_cost = float(sum(v["fixed_cost"] for v in V))
        total_cost = total_variable_cost + total_fixed_cost
        total_vehicles = quicksum(used[v["idx"]] for v in V)

        # Bounds estruturais seguros fortalecem a relaxação LP sem alterar o
        # conjunto de soluções inteiras factíveis.
        model.addCons(
            quicksum(z[t] for t in T) >= int(strengthened_bounds["trips_lb"]),
            name="minimum_number_of_trips",
        )
        model.addCons(
            total_distance >= float(strengthened_bounds["distance_lb"]),
            name="analytical_distance_lower_bound",
        )
        model.addCons(
            total_time >= float(strengthened_bounds["time_lb"]),
            name="analytical_time_lower_bound",
        )
        model.addCons(
            total_vehicles >= int(strengthened_bounds["vehicles_lb"]),
            name="analytical_vehicle_lower_bound",
        )
        if not feasibility_only and strengthened_bounds.get("cost_lb") is not None:
            model.addCons(
                total_cost >= float(strengthened_bounds["cost_lb"]),
                name="analytical_cost_lower_bound",
            )

        coeff = self.objective_context["coefficients"]
        objective = (
            float(coeff.get("DISTANCE", 0.0)) * total_distance
            + float(coeff.get("TIME", 0.0)) * total_time
            + float(coeff.get("COST", 0.0)) * total_cost
            + float(coeff.get("VEHICLES", 0.0)) * total_vehicles
        )

        # UB externo validado: adicioná-lo como cutoff é matematicamente seguro e
        # evita explorar soluções sabidamente piores. Uma pequena tolerância impede
        # excluir o incumbent por ruído numérico.
        if not feasibility_only and incumbent_ub is not None and self.use_incumbent_cutoff:
            cutoff_eps = max(1e-8, abs(float(incumbent_ub)) * 1e-9)
            model.addCons(objective <= float(incumbent_ub) + cutoff_eps, name="validated_incumbent_cutoff")

        model.setObjective(0.0 if feasibility_only else objective, "minimize")

        self._model_stats = {
            "feasibility_only": bool(feasibility_only),
            "max_active_vehicles": max_active_vehicles,
            "trip_slots_safe_upper_bound": K,
            "trip_slots_source": strengthened_bounds.get("trip_slots_source"),
            "binary_arc_variables": len(x),
            "binary_visit_variables": len(y),
            "binary_assignment_variables": len(assign),
            "binary_vehicle_variables": len(used),
            "binary_order_variables": 0,
            "sequence_constraints_without_order_binaries": sequence_constraints,
            "preprocessing": prep,
            "strengthened_analytical_bounds": copy.deepcopy(strengthened_bounds),
            "incumbent_cutoff_used": incumbent_ub is not None and self.use_incumbent_cutoff,
            "scalability_profile": copy.deepcopy(self._scalability_profile),
        }
        handles = {
            "T": T, "C": C, "V": V, "z": z, "y": y, "x": x, "assign": assign,
            "used": used, "s": s, "e": e, "dur": dur, "tau": tau, "tripdist": tripdist,
            "qdist": qdist, "qdur": qdur, "pos": pos, "arcs": arcs, "prefix_end": prefix_end,
            "total_distance": total_distance, "total_time": total_time,
            "total_variable_cost": total_variable_cost, "total_fixed_cost": total_fixed_cost,
            "total_cost": total_cost, "total_vehicles": total_vehicles,
        }
        return model, handles

    def _extract_solution(self, model, h):
        sol = model.getBestSol()
        if sol is None:
            return None
        p = self.problem
        routes = []
        for t in h["T"]:
            if model.getSolVal(sol, h["z"][t]) < 0.5:
                continue
            vehicle = None
            for v in h["V"]:
                if model.getSolVal(sol, h["assign"][t, v["idx"]]) > 0.5:
                    vehicle = v; break
            if vehicle is None:
                continue
            succ = {}
            for i, j in h["arcs"]:
                var = h["x"].get((t, i, j))
                if var is not None and model.getSolVal(sol, var) > 0.5:
                    succ[i] = j
            path = [0]; cur = 0; guard = 0
            while cur in succ and guard <= len(h["C"]) + 2:
                cur = succ[cur]; path.append(cur); guard += 1
                if cur == 0:
                    break
            if path[-1] != 0 or len(path) <= 2:
                continue
            start = float(model.getSolVal(sol, h["s"][t])); end = float(model.getSolVal(sol, h["e"][t]))
            details = []; load = 0.0
            prev = 0
            for cid in path[1:-1]:
                service_start = float(model.getSolVal(sol, h["tau"][t, cid]))
                arrival = start + p["travel_time"][(0, cid)] if prev == 0 else details[-1]["departure_time_minutes"] + p["travel_time"][(prev, cid)]
                load += p["client_rows"][cid]["demand"]
                details.append({
                    "node_id": cid, "arrival_time_minutes": float(arrival),
                    "service_start_time_minutes": service_start,
                    "departure_time_minutes": service_start + p["service_time"], "load": load,
                })
                prev = cid
            routes.append({
                "vehicle_id": vehicle["vehicle_id"], "physical_vehicle_id": vehicle["vehicle_id"],
                "trip_slot": t + 1, "path": path,
                "start_time_minutes": start, "end_time_minutes": end,
                "total_route_time_minutes": end - start,
                "distance_km": float(model.getSolVal(sol, h["tripdist"][t])),
                "total_route_load": float(sum(p["client_rows"][cid]["demand"] for cid in path if cid != 0)),
                "path_details": details,
            })
        routes.sort(key=lambda r: (r["physical_vehicle_id"], r["start_time_minutes"]))
        counter = defaultdict(int)
        for r in routes:
            counter[r["physical_vehicle_id"]] += 1
            r["trip_number"] = counter[r["physical_vehicle_id"]]

        used_ids = sorted(set(r["physical_vehicle_id"] for r in routes))
        vmap = {v["vehicle_id"]: v for v in p["vehicles"]}
        total_distance = sum(r["distance_km"] for r in routes)
        total_time = sum(r["total_route_time_minutes"] for r in routes)
        variable = sum(r["distance_km"] * vmap[r["physical_vehicle_id"]]["cost_per_km"] for r in routes)
        fixed = sum(vmap[vid]["fixed_cost"] for vid in used_ids) if p["fixed_cost_mode"] == "WHEN_USED" else sum(v["fixed_cost"] for v in p["vehicles"])
        comps = {"DISTANCE": total_distance, "TIME": total_time, "COST": variable + fixed, "VEHICLES": float(len(used_ids))}
        score = self._canonical_score(comps)
        return {
            "status": "SUCCESS", "method": "SCIP_EXACT_GLOBAL_MILP", "routes": routes,
            "summary": {
                "total_distance_km": total_distance, "total_time_minutes": total_time,
                "total_variable_cost": variable, "total_fixed_cost": fixed,
                "total_cost": variable + fixed, "num_vehicles_used": len(used_ids), "num_routes": len(routes),
            },
            "objective_components": comps,
            "canonical_objective_score": score,
            "objective_context": copy.deepcopy(self.objective_context),
            "proven_optimal": False,
        }

    def _publish(self, status, certificate):
        mem = self.shared_memory.data
        mem["certification_status"] = status
        mem["optimality_certificate"] = copy.deepcopy(certificate)
        mem["certification_progress"] = {
            "elapsed_seconds": certificate.get("elapsed_seconds"),
            "upper_bound": certificate.get("upper_bound"),
            "lower_bound": certificate.get("lower_bound"),
            "gap_pct": certificate.get("gap_pct"),
            "status": status,
            "method": "SCIP_EXACT_GLOBAL_MILP",
        }


    # níveis A/B/C/D e certificação em cascata

    def _classify_certification_level(self, feasible, scip_status, upper_bound, lower_bound):
        """Application scale; labels do not alter the mathematical bounds."""
        import math
        valid=False
        try:
            ub=float(upper_bound);lb=float(lower_bound)
            valid=math.isfinite(ub) and math.isfinite(lb) and lb<=ub+max(1e-7,abs(ub)*1e-6)
        except (ValueError,TypeError):pass
        if not feasible or not valid:
            return dict(level='D',gap_relative=None,gap_pct=None,declaration='Sem classificação: auditoria ou LB válido ausente',report_allowed=False,valid_lower_bound=valid,optimal_within_tolerance=False)
        gap=max(0.,(ub-lb)/max(abs(ub),1e-12))
        close=abs(ub-lb)<=max(1e-7,abs(ub)*1e-6)
        level='A' if close else ('B+' if gap<=.01 else ('B' if gap<=.35 else 'C'))
        labels={'A':'Ótimo certificado dentro da tolerância numérica','B+':'Garantia forte: GAP até 1%','B':'Garantia intermediária: GAP acima de 1% e até 35%','C':'Garantia ampla: GAP acima de 35%'}
        return dict(level=level,gap_absolute=max(0.,ub-lb),gap_relative=gap,gap_pct=gap*100,declaration=labels[level]+' — escala do sistema V5.9.1',report_allowed=True,valid_lower_bound=True,optimal_within_tolerance=close)

    def _try_add_incumbent_mip_start(self, model, handles, normalized_solution):
        """Complete start, validated against every original model constraint."""
        report={"submitted":False,"validated":False,"accepted":False,"reason":None}
        self._mip_start_report=report
        sol=None
        try:
            routes=sorted(normalized_solution.get("routes",[]),key=lambda r:float(r["start_time_minutes"]))
            if not routes or len(routes)>len(handles["T"]):
                report["reason"]="NO_ROUTES_OR_TOO_MANY_TRIPS"; return False
            p=self.problem
            sol=model.createSol()
            # Every variable receives an explicit value, including unused arcs.
            for var in model.getVars(): model.setSolVal(sol,var,0.0)
            for t in handles["T"]:
                model.setSolVal(sol,handles["s"][t],float(p["operation_start"]))
                model.setSolVal(sol,handles["e"][t],float(p["operation_start"]))
                for cid in handles["C"]:
                    model.setSolVal(sol,handles["tau"][t,cid],float(p["windows"][cid][0]))
            ends={v["idx"]:float(p["operation_start"]) for v in handles["V"]}
            for t in handles["T"]:
                if t<len(routes):
                    r=routes[t]; path=[int(c) for c in r["path"]]
                    vid=int(r.get("physical_vehicle_id",r.get("vehicle_id")))
                    vi=next(v["idx"] for v in handles["V"] if int(v["vehicle_id"])==vid)
                    start=float(r["start_time_minutes"]); end=float(r["end_time_minutes"])
                    dist=sum(float(p["distance"][i,j]) for i,j in zip(path,path[1:]))
                    for group,key,val in [("z",t,1),("used",vi,1),("assign",(t,vi),1),
                        ("s",t,start),("e",t,end),("dur",t,end-start),("tripdist",t,dist),
                        ("qdur",(t,vi),end-start)]:
                        model.setSolVal(sol,handles[group][key],val)
                    if handles["qdist"]:model.setSolVal(sol,handles["qdist"][t,vi],dist)
                    for rank,cid in enumerate(path[1:-1],1):
                        model.setSolVal(sol,handles["y"][t,cid],1)
                        if handles["pos"]:model.setSolVal(sol,handles["pos"][t,cid],rank)
                    for i,j in zip(path,path[1:]):model.setSolVal(sol,handles["x"][t,i,j],1)
                    for d in r["path_details"]:
                        model.setSolVal(sol,handles["tau"][t,int(d["node_id"])],float(d["service_start_time_minutes"]))
                    ends[vi]=max(ends[vi],end)
                for vi,end in ends.items():model.setSolVal(sol,handles["prefix_end"][t,vi],end)
            report["validated"]=bool(model.checkSol(sol,printreason=True,completely=True))
            if not report["validated"]:
                report["reason"]="MODEL_REJECTED_COMPLETE_START"; return False
            report["submitted"]=True
            report["accepted"]=bool(model.addSol(sol,free=False))
            report["reason"]="ACCEPTED" if report["accepted"] else "NOT_STORED"
            return report["accepted"]
        except Exception as exc:
            report["reason"]="START_EXCEPTION: "+str(exc)
            return False
        finally:
            if sol is not None:model.freeSol(sol)

    def _install_adaptive_lb_monitor(self, model, analytical_lb):
        """
        Instala um EventHandler SCIP que interrompe a solução quando o lower bound
        global deixa de apresentar progresso material por um período configurável.

        A interrupção NÃO invalida a certificação: o dual bound corrente do SCIP
        continua sendo um lower bound matematicamente válido. O certificado final
        usa max(SCIP_DUAL, ANALYTICAL_LB).
        """
        if not self.adaptive_stop_enabled:
            return None

        try:
            from pyscipopt import Eventhdlr, SCIP_EVENTTYPE
        except Exception:
            return None

        class _AdaptiveLowerBoundMonitor(Eventhdlr):
            def eventinit(self):
                self._event_types = [SCIP_EVENTTYPE.NODEFOCUSED]
                # LPSOLVED permite verificar estagnação também durante trabalho pesado
                # no nó raiz, quando NODEFOCUSED pode demorar muito para voltar a disparar.
                lp_event = getattr(SCIP_EVENTTYPE, "LPSOLVED", None)
                if lp_event is not None:
                    self._event_types.append(lp_event)
                for evtype in self._event_types:
                    try:
                        self.model.catchEvent(evtype, self)
                    except Exception:
                        pass

            def eventexit(self):
                for evtype in getattr(self, "_event_types", []):
                    try:
                        self.model.dropEvent(evtype, self)
                    except Exception:
                        pass

            def eventexec(self, event):
                now = _time_module.monotonic()
                if now - self.last_check < self.check_interval:
                    return
                self.last_check = now

                try:
                    dual = float(self.model.getDualbound())
                except Exception:
                    return

                if not math.isfinite(dual):
                    return

                combined_lb = dual
                if self.analytical_lb is not None and math.isfinite(self.analytical_lb):
                    combined_lb = max(combined_lb, self.analytical_lb)

                threshold = max(
                    1e-12,
                    abs(self.best_lb) * self.min_relative_improvement
                    if math.isfinite(self.best_lb)
                    else 1e-12,
                )

                if (not math.isfinite(self.best_lb)) or combined_lb > self.best_lb + threshold:
                    self.best_lb = combined_lb
                    self.last_progress = now
                    self.progress_events += 1
                    return

                elapsed = now - self.started
                stagnant = now - self.last_progress
                if elapsed >= self.min_runtime and stagnant >= self.patience:
                    self.stop_reason = "LOWER_BOUND_STAGNATION"
                    self.stagnation_seconds = stagnant
                    try:
                        self.model.interruptSolve()
                    except Exception:
                        pass

        monitor = _AdaptiveLowerBoundMonitor()
        now = _time_module.monotonic()
        monitor.analytical_lb = (
            float(analytical_lb)
            if analytical_lb is not None and math.isfinite(float(analytical_lb))
            else None
        )
        monitor.best_lb = (
            monitor.analytical_lb
            if monitor.analytical_lb is not None
            else float("-inf")
        )
        monitor.started = now
        monitor.last_progress = now
        monitor.last_check = now
        monitor.min_runtime = float(self.adaptive_min_runtime_seconds)
        monitor.patience = float(self.adaptive_patience_seconds)
        monitor.check_interval = float(self.adaptive_check_interval_seconds)
        monitor.min_relative_improvement = float(self.adaptive_min_relative_lb_improvement)
        monitor.stop_reason = None
        monitor.stagnation_seconds = None
        monitor.progress_events = 0

        try:
            model.includeEventhdlr(
                monitor,
                "adaptive_lb_monitor",
                "Interrompe SCIP quando o lower bound global fica estagnado.",
            )
            return monitor
        except Exception:
            return None

    def _run_scip_layer(self, layer_name, time_limit_seconds, incumbent_report):
        """Executa uma camada SCIP e retorna bounds, solução e metadados."""
        incumbent = incumbent_report.get("normalized_solution") if isinstance(incumbent_report, dict) else None
        incumbent_ub = None
        if incumbent_report and incumbent_report.get("feasible"):
            incumbent_ub = float(incumbent_report["recomputed"]["canonical_objective_score"])

        old_time = self.time_limit_seconds
        old_gap = self.relative_gap_target
        self.time_limit_seconds = max(1, int(time_limit_seconds))
        # Camada rápida pode encerrar quando chegar ao nível A numérico.
        self.relative_gap_target = self.relative_gap_tolerance
        try:
            layer_wall_start = _time_module.time()

            # preflight barato e certificado ANTES de construir o MILP completo.
            # Em instâncias grandes, evita alocar centenas de milhares de binárias
            # quando o modelo ultrapassa os limites operacionais configurados.
            strengthened = self._analytical_lower_bounds(incumbent_ub=incumbent_ub)
            preflight_arcs, preflight_prep = self._build_feasible_arcs()
            complexity = self._estimate_model_complexity(preflight_arcs, strengthened)
            analytical_lb_pre = strengthened.get("canonical_objective_lb")
            if complexity.get("guard_triggered", False):
                upper = incumbent_ub
                lower = (float(analytical_lb_pre) if analytical_lb_pre is not None and math.isfinite(float(analytical_lb_pre)) else None)
                lower, lower_source, external_lb = self._merge_external_global_lower_bound(
                    lower, "ANALYTICAL_COMPLEXITY_GUARD"
                )
                status_guard = "skipped_model_complexity_guard"
                level = self._classify_certification_level(incumbent is not None, status_guard, upper, lower)
                self._model_stats = {
                    "model_build_skipped": True,
                    "skip_reason": "MODEL_COMPLEXITY_GUARD",
                    "complexity_estimate": copy.deepcopy(complexity),
                    "preprocessing": copy.deepcopy(preflight_prep),
                    "strengthened_analytical_bounds": copy.deepcopy(strengthened),
                    "scalability_profile": copy.deepcopy(self._scalability_profile),
                }
                elapsed = _time_module.time() - layer_wall_start
                print(
                    f"🛡️ {layer_name}: SCIP completo não construído pelo complexity guard | "
                    f"arc_bin≈{complexity['estimated_binary_arc_variables']:,} | "
                    f"bin_total≈{complexity['estimated_total_binary_variables']:,} | "
                    f"seq≈{complexity['estimated_sequence_constraints']:,}"
                )
                print(f"   Certificação analítica preservada | UB={upper} | LB={lower}")
                return {
                    "layer": layer_name, "time_limit_seconds": self.time_limit_seconds,
                    "elapsed_seconds": elapsed, "solve_elapsed_seconds": 0.0,
                    "model_build_elapsed_seconds": 0.0, "scip_status": status_guard,
                    "upper_bound": upper, "lower_bound": lower,
                    "scip_dual_bound_raw": None,
                    "analytical_lower_bound": (float(analytical_lb_pre) if analytical_lb_pre is not None else None),
                    "external_global_lower_bound": external_lb,
                    "lower_bound_source": lower_source,
                    "branch_nodes_processed": 0, "solution": copy.deepcopy(incumbent),
                    "mip_start_added": False, "adaptive_stop_enabled": False,
                    "adaptive_stop_triggered": False, "adaptive_stop_reason": None,
                    "adaptive_stagnation_seconds": None, "adaptive_lb_progress_events": 0,
                    "classification": level, "model_statistics": copy.deepcopy(self._model_stats),
                    "model_complexity_guard_triggered": True,
                }

            model, handles = self._build_scip_model(incumbent_ub=incumbent_ub)
            build_elapsed = _time_module.time() - layer_wall_start
            remaining_solver_time = max(0.1, float(self.time_limit_seconds) - build_elapsed)
            try:
                model.setParam("limits/time", remaining_solver_time)
            except Exception:
                pass
            analytical = copy.deepcopy(
                self.shared_memory.data.get("certification_strengthened_bounds", {}) or {}
            )
            analytical_lb = analytical.get("canonical_objective_lb")
            adaptive_monitor = self._install_adaptive_lb_monitor(model, analytical_lb)

            mip_start_added = self._try_add_incumbent_mip_start(model, handles, incumbent)
            t0 = _time_module.time()
            print(
                f"🔍 {layer_name}: SCIP iniciado | orçamento máximo={self.time_limit_seconds}s | "
                f"build={build_elapsed:.2f}s | solve máximo={remaining_solver_time:.2f}s | "
                f"gap alvo={self.relative_gap_tolerance*100:.4f}%"
            )
            if adaptive_monitor is not None:
                print(
                    "   ⚡ Parada adaptativa ativa | "
                    f"mínimo={self.adaptive_min_runtime_seconds:.0f}s | "
                    f"paciência={self.adaptive_patience_seconds:.0f}s | "
                    f"checagem={self.adaptive_check_interval_seconds:.0f}s"
                )
            if mip_start_added:
                print("   ✅ MIP-start independente aceito pelo SCIP.")
            elif incumbent_ub is not None:
                print("   ℹ️ Incumbent usado como cutoff seguro; MIP-start parcial não disponível/aceito.")
            model.optimize()
            solve_elapsed = _time_module.time() - t0
            elapsed = _time_module.time() - layer_wall_start

            scip_status = str(model.getStatus()).lower()
            try:
                primal = float(model.getPrimalbound())
            except Exception:
                primal = float("inf")
            try:
                dual = float(model.getDualbound())
            except Exception:
                dual = float("-inf")
            try:
                nodes = int(model.getNNodes())
            except Exception:
                nodes = None

            scip_solution = self._extract_solution(model, handles)
            best_solution = scip_solution
            if incumbent is not None:
                if best_solution is None or float(incumbent["canonical_objective_score"]) < float(best_solution["canonical_objective_score"]) - 1e-9:
                    best_solution = copy.deepcopy(incumbent)

            upper = (
                float(best_solution["canonical_objective_score"])
                if best_solution is not None
                else (primal if math.isfinite(primal) else None)
            )

            # Combina o dual bound do SCIP com o lower bound analítico independente.
            # O máximo dos dois continua sendo lower bound válido em minimização.
            lower_candidates = []
            if math.isfinite(dual):
                lower_candidates.append(("SCIP_DUAL", float(dual)))
            if analytical_lb is not None and math.isfinite(float(analytical_lb)):
                lower_candidates.append(("ANALYTICAL", float(analytical_lb)))

            if lower_candidates:
                lower_source, lower = max(lower_candidates, key=lambda item: item[1])
                if len(lower_candidates) == 2:
                    scip_lb = next(v for k, v in lower_candidates if k == "SCIP_DUAL")
                    analytical_v = next(v for k, v in lower_candidates if k == "ANALYTICAL")
                    eq_tol = max(1e-10, 1e-8 * max(1.0, abs(scip_lb), abs(analytical_v)))
                    if abs(scip_lb - analytical_v) <= eq_tol:
                        lower_source = "SCIP_DUAL_EQUALS_ANALYTICAL"
                    elif scip_lb > analytical_v:
                        lower_source = "SCIP_DUAL"
                    else:
                        lower_source = "ANALYTICAL"
            else:
                lower_source, lower = None, None

            # o LB externo globalmente válido participa já da camada provisória,
            # inclusive antes da auditoria final. Assim o gap exibido nunca regride.
            lower, lower_source, external_lb = self._merge_external_global_lower_bound(
                lower, lower_source
            )

            feasible = best_solution is not None
            level = self._classify_certification_level(feasible, scip_status, upper, lower)

            return {
                "layer": layer_name,
                "time_limit_seconds": self.time_limit_seconds,
                "elapsed_seconds": elapsed,
                "solve_elapsed_seconds": solve_elapsed,
                "model_build_elapsed_seconds": build_elapsed,
                "scip_status": scip_status,
                "upper_bound": upper,
                "lower_bound": lower,
                "scip_dual_bound_raw": (float(dual) if math.isfinite(dual) else None),
                "analytical_lower_bound": (
                    float(analytical_lb)
                    if analytical_lb is not None and math.isfinite(float(analytical_lb))
                    else None
                ),
                "lower_bound_source": lower_source,
                "external_global_lower_bound": external_lb,
                "branch_nodes_processed": nodes,
                "solution": best_solution,
                "mip_start_added": mip_start_added, "mip_start_report": copy.deepcopy(getattr(self,"_mip_start_report",{})),
                "adaptive_stop_enabled": bool(adaptive_monitor is not None),
                "adaptive_stop_triggered": bool(
                    adaptive_monitor is not None
                    and adaptive_monitor.stop_reason == "LOWER_BOUND_STAGNATION"
                ),
                "adaptive_stop_reason": (
                    adaptive_monitor.stop_reason if adaptive_monitor is not None else None
                ),
                "adaptive_stagnation_seconds": (
                    adaptive_monitor.stagnation_seconds if adaptive_monitor is not None else None
                ),
                "adaptive_lb_progress_events": (
                    adaptive_monitor.progress_events if adaptive_monitor is not None else 0
                ),
                "classification": level,
                "model_statistics": copy.deepcopy(self._model_stats),
            }
        finally:
            self.time_limit_seconds = old_time
            self.relative_gap_target = old_gap

    def _get_pre_audit_report(self):
        """Obtém a verificação independente já executada pelo AuditorAgent.

        O Certificador NÃO instancia nem chama IndependentFeasibilityVerifier.
        A arquitetura é deliberadamente: Orchestrator -> Auditor -> Certificador.
        """
        mem = self.shared_memory.data
        report = mem.get("independent_feasibility_report_pre_certification")
        if not isinstance(report, dict):
            raise RuntimeError(
                "CERTIFICADOR V14.2: não existe relatório independente pré-certificação. "
                "O Orchestrator deve executar o AuditorAgent antes do Certificador."
            )
        if not bool(report.get("feasible", False)):
            raise RuntimeError(
                "CERTIFICADOR V14.2: o Auditor não aprovou a factibilidade independente do incumbent."
            )
        normalized = report.get("normalized_solution")
        if not isinstance(normalized, dict) or normalized.get("status") != "SUCCESS":
            raise RuntimeError(
                "CERTIFICADOR V14.2: relatório do Auditor não contém normalized_solution válida."
            )
        return copy.deepcopy(report)

    def _provisional_result(self, chosen, layer2, layer3, pre_report, fast_bounds, layer1_elapsed):
        """Monta resultado matemático PROVISÓRIO.

        O nível só se torna final após o Orchestrator mandar a solução produzida
        pelo SCIP novamente ao Auditor e chamar finalize_after_independent_audit().
        """
        cls = chosen["classification"]
        scip_status = chosen["scip_status"]
        candidate = copy.deepcopy(chosen.get("solution")) if isinstance(chosen.get("solution"), dict) else None
        certificate = {
            "version": self.VERSION,
            "method": "CASCADE_AUDITOR_GATE_THEN_SCIP",
            "formulation": "GLOBAL_TRIP_SLOTS_WITH_PHYSICAL_VEHICLE_SCHEDULING",
            "status": "PENDING_FINAL_INDEPENDENT_AUDIT",
            "scip_status": scip_status,
            "proven_optimal": False,
            "optimal_within_tolerance": False,
            "certification_finalized": False,
            "provisional_certification_level": cls["level"],
            "certification_level": None,
            "certification_declaration": "Aguardando auditoria independente final",
            "report_allowed": False,
            "upper_bound": chosen.get("upper_bound"),
            "lower_bound": chosen.get("lower_bound"),
            "gap_absolute": cls.get("gap_absolute"),
            "gap_relative": cls.get("gap_relative"),
            "gap_pct": cls.get("gap_pct"),
            "relative_gap_tolerance": self.relative_gap_tolerance,
            "business_gap_tolerance": self.business_gap_tolerance,
            "lower_bound_certified": bool(cls.get("valid_lower_bound")),
            "pre_certification_independent_audit_passed": True,
            "pre_certification_independent_report": copy.deepcopy(pre_report),
            "fast_bounds": copy.deepcopy(fast_bounds),
            "layer1_elapsed_seconds": layer1_elapsed,
            "layer2": {k: copy.deepcopy(v) for k, v in layer2.items() if k != "solution"},
            "layer3": ({k: copy.deepcopy(v) for k, v in layer3.items() if k != "solution"} if layer3 else None),
            "selected_layer": chosen["layer"],
            "elapsed_seconds": _time_module.time() - self.global_start_time,
            "objective_context_version": self.objective_context.get("version"),
            "operational_contract": copy.deepcopy(self.shared_memory.data.get("certification_operational_contract")),
            "strengthened_analytical_bounds": copy.deepcopy(
                self.shared_memory.data.get("certification_strengthened_bounds", {}) or {}
            ),
            "certified_bound_catalog": copy.deepcopy(
                self.shared_memory.data.get("certification_bound_catalog", []) or []
            ),
            "physical_optimality_interval": self._physical_certificate_interval(
                chosen.get("upper_bound"), chosen.get("lower_bound")
            ),
        }
        return {
            "status": certificate["status"],
            "proven_optimal": False,
            "optimal_within_tolerance": False,
            "certification_finalized": False,
            "provisional_certification_level": cls["level"],
            "certification_level": None,
            "report_allowed": False,
            "certificate": certificate,
            "certified_solution": candidate,
        }

    def finalize_after_independent_audit(self, provisional_result, final_audit_report):
        """Finaliza A/B/C/D SOMENTE depois da auditoria independente pós-SCIP.

        Regras:
        - falha de factibilidade => D;
        - sem LB válido => D;
        - gap <= tolerância numérica => A;
        - gap <= tolerância empresarial => B;
        - gap maior, mas conhecido => C;
        - proven_optimal=True somente se SCIP=optimal, a auditoria passou e o
          objetivo reconstruído é consistente com o UB do modelo.
        """
        mem = self.shared_memory.data
        result = copy.deepcopy(provisional_result or {})
        cert = copy.deepcopy(result.get("certificate", {}) or {})
        candidate = result.get("certified_solution")

        feasible = bool(isinstance(final_audit_report, dict) and final_audit_report.get("feasible", False))
        if not feasible or not isinstance(candidate, dict):
            cert.update({
                "status": "CERTIFICATION_INCONCLUSIVE",
                "proven_optimal": False,
                "optimal_within_tolerance": False,
                "certification_finalized": True,
                "certification_level": "D",
                "certification_declaration": "Certificação inconclusiva: solução pós-SCIP falhou na auditoria independente",
                "report_allowed": False,
                "post_certification_independent_audit_passed": False,
                "post_certification_independent_report": copy.deepcopy(final_audit_report),
            })
            result.update({
                "status": cert["status"], "proven_optimal": False,
                "optimal_within_tolerance": False, "certification_finalized": True,
                "certification_level": "D", "report_allowed": False,
                "certificate": cert, "certified_solution": None,
            })
            mem["certification_result"] = copy.deepcopy(result)
            mem["optimality_certificate"] = copy.deepcopy(cert)
            return result

        normalized = final_audit_report.get("normalized_solution")
        recomputed = final_audit_report.get("recomputed", {}) or {}
        audited_ub = self._safe_float(recomputed.get("canonical_objective_score"), None)
        lb = self._safe_float(cert.get("lower_bound"), None)

        # Continuidade: um lower bound certificado no Relatório 1 continua válido
        # para o MESMO problema e a MESMA objective_context congelada. Portanto,
        # a segunda rodada nunca enfraquece a prova já obtida.
        baseline_lb = self._safe_float(
            mem.get("continuity_baseline_valid_lower_bound"), None
        )
        if baseline_lb is not None:
            lb = baseline_lb if lb is None else max(float(lb), float(baseline_lb))

        # consome o melhor LB GLOBAL produzido pelo BoundStrengtheningAgent.
        # Somente resultados explicitamente marcados como globais entram na certificação.
        bs = mem.get("bound_strengthening_last_result", {}) or {}
        bs_lb = None
        if isinstance(bs, dict) and str(bs.get("validity", "")).upper().startswith("GLOBAL_VALID"):
            bs_lb = self._safe_float(bs.get("best_lower_bound"), None)
        if bs_lb is not None:
            lb = bs_lb if lb is None else max(float(lb), float(bs_lb))

        scip_status = str(cert.get("scip_status", "")).lower()
        model_ub = self._safe_float(cert.get("upper_bound"), None)

        # A prova só é transferível para a solução auditada se o objetivo do
        # modelo e o objetivo reconstruído coincidirem dentro de tolerância
        # estrita de integração (muito menor que a tolerância empresarial).
        objective_consistent = True
        objective_difference = None
        if audited_ub is None:
            objective_consistent = False
        elif model_ub is not None:
            objective_difference = abs(audited_ub - model_ub)
            integration_tol = max(1e-7, 1e-6 * max(1.0, abs(model_ub), abs(audited_ub)))
            objective_consistent = objective_difference <= integration_tol

        if not objective_consistent:
            cls = {
                "level": "D", "gap_absolute": None, "gap_relative": None,
                "gap_pct": None, "declaration": "Certificação inconclusiva: objetivo SCIP divergente da reconstrução independente",
                "report_allowed": False, "valid_lower_bound": lb is not None,
                "optimal_within_tolerance": False,
            }
            proven = False
        else:
            # Pequenas inversões UB<LB por tolerância numérica são absorvidas;
            # inversão material invalida a classificação.
            if lb is not None and audited_ub is not None and lb > audited_ub:
                inversion = lb - audited_ub
                integration_tol = max(1e-7, 1e-6 * max(1.0, abs(lb), abs(audited_ub)))
                if inversion <= integration_tol:
                    lb = audited_ub
                else:
                    lb = None
            cls = self._classify_certification_level(True, scip_status, audited_ub, lb)
            proven = bool(scip_status == "optimal" and cls["level"] == "A")

        level_to_status = {
            "A": "PROVEN_OPTIMAL" if proven else "CERTIFIED_OPTIMAL_WITHIN_TOLERANCE",
            "B+": "CERTIFIED_NEAR_OPTIMAL",
            "B": "FEASIBLE_WITH_INTERMEDIATE_GUARANTEE",
            "C": "FEASIBLE_WITH_CERTIFIED_GAP",
            "D": "CERTIFICATION_INCONCLUSIVE",
        }
        status = level_to_status[cls["level"]]

        if isinstance(normalized, dict):
            candidate = copy.deepcopy(normalized)
        candidate["status"] = "SUCCESS"
        candidate["proven_optimal"] = proven
        candidate["optimal_within_tolerance"] = bool(cls.get("optimal_within_tolerance"))
        candidate["certification_level"] = cls["level"]
        candidate["certification_declaration"] = cls["declaration"]
        candidate["optimality_gap_pct"] = cls.get("gap_pct")

        cert.update({
            "status": status,
            "proven_optimal": proven,
            "optimal_within_tolerance": bool(cls.get("optimal_within_tolerance")),
            "certification_finalized": True,
            "certification_level": cls["level"],
            "certification_declaration": cls["declaration"],
            "report_allowed": bool(cls["report_allowed"]),
            "upper_bound": audited_ub,
            "lower_bound": lb,
            "gap_absolute": cls.get("gap_absolute"),
            "gap_relative": cls.get("gap_relative"),
            "gap_pct": cls.get("gap_pct"),
            "lower_bound_certified": bool(cls.get("valid_lower_bound")),
            "bound_strengthening_consumed": bool(bs_lb is not None),
            "bound_strengthening_lower_bound": bs_lb,
            "bound_strengthening_result": copy.deepcopy(bs) if isinstance(bs, dict) else None,
            "post_certification_independent_audit_passed": True,
            "post_certification_independent_report": copy.deepcopy(final_audit_report),
            "objective_consistent_with_independent_reconstruction": objective_consistent,
            "objective_difference_model_vs_audit": objective_difference,
            "physical_optimality_interval": self._physical_certificate_interval(audited_ub, lb),
            "certified_bound_catalog": copy.deepcopy(
                self.shared_memory.data.get("certification_bound_catalog", []) or []
            ),
        })
        result.update({
            "status": status,
            "proven_optimal": proven,
            "optimal_within_tolerance": bool(cls.get("optimal_within_tolerance")),
            "certification_finalized": True,
            "certification_level": cls["level"],
            "report_allowed": bool(cls["report_allowed"]),
            "certificate": cert,
            "certified_solution": candidate,
        })
        mem["certification_result"] = copy.deepcopy(result)
        mem["optimality_certificate"] = copy.deepcopy(cert)
        mem["certified_solution"] = copy.deepcopy(candidate)
        mem["certification_level"] = cls["level"]
        mem["certification_report_allowed"] = bool(cls["report_allowed"])
        mem["certification_checkpoint"] = {
            "version": self.VERSION,
            "best_solution": copy.deepcopy(candidate),
            "upper_bound": audited_ub,
            "lower_bound": lb,
            "gap_pct": cls.get("gap_pct"),
            "certification_level": cls["level"],
            "selected_layer": cert.get("selected_layer"),
            "note": "Checkpoint lógico de incumbent/bounds; não contém a árvore B&B interna do SCIP.",
        }
        return result

    def certify_optimality(self):
        """Executa a certificação rápida em cascata sem verificar factibilidade internamente.

        Camada 1 já deve ter sido executada pelo AuditorAgent sob comando do
        Orchestrator. Camada 2 executa SCIP rápido. Camada 3 é opcional.
        O resultado retornado é PROVISÓRIO até a auditoria independente final.
        """
        self.global_start_time = _time_module.time()
        mem = self.shared_memory.data
        try:
            self._load_objective_context()
            self._build_problem()

            print("\n" + "=" * 70)
            print(f"🏆 CERTIFICAÇÃO {self.VERSION} — AUDITOR → BOUNDS/K-PROOF → SCIP/GUARD → AUDITOR")
            print("=" * 70)

            layer1_start = _time_module.time()
            pre_report = self._get_pre_audit_report()
            fast_bounds = copy.deepcopy(mem.get("certification_fast_bounds", {}) or {})
            layer1_elapsed = _time_module.time() - layer1_start
            incumbent = copy.deepcopy(pre_report["normalized_solution"])
            mem["best_incumbent"] = copy.deepcopy(incumbent)
            print(f"⚡ Gate do Auditor confirmado em {layer1_elapsed:.3f}s | factível=True")
            print(f"   UB reconstruído pelo Auditor = {incumbent['canonical_objective_score']:.12f}")

            layer2 = self._run_scip_layer("CAMADA_2_FAST_EXACT", self.time_limit_seconds, pre_report)
            chosen = layer2

            layer3 = None
            if self.intensive_certification and layer2["classification"]["level"] != "A":
                # Sem verificador interno: se o SCIP encontrou solução melhor, ela
                # será auditada somente após a cascata. Para warm-start da Camada 3,
                # usamos a solução do próprio SCIP como candidato técnico.
                if isinstance(layer2.get("solution"), dict):
                    pseudo_report = {
                        "feasible": True,
                        "normalized_solution": copy.deepcopy(layer2["solution"]),
                        "recomputed": {"canonical_objective_score": layer2.get("upper_bound")},
                    }
                else:
                    pseudo_report = pre_report
                print("🧠 Camada 3 intensiva solicitada pelo usuário.")
                layer3 = self._run_scip_layer("CAMADA_3_INTENSIVE", self.intensive_time_limit_seconds, pseudo_report)
                rank = {"A": 0, "B+": 1, "B": 2, "C": 3, "D": 4}
                c2, c3 = layer2["classification"], layer3["classification"]
                if rank[c3["level"]] < rank[c2["level"]] or (
                    rank[c3["level"]] == rank[c2["level"]]
                    and c3.get("gap_relative") is not None
                    and (c2.get("gap_relative") is None or c3["gap_relative"] < c2["gap_relative"])
                ):
                    chosen = layer3

            result = self._provisional_result(chosen, layer2, layer3, pre_report, fast_bounds, layer1_elapsed)

            # rastreabilidade da prova estrutural Exact-K.
            # O LB canônico continua vindo do Bound/global LB; aqui não somamos
            # bounds novamente. Apenas anexamos a evidência matemática.
            kproof = copy.deepcopy(mem.get("k_vehicle_feasibility_last_result") or {})
            if isinstance(kproof, dict) and kproof:
                result["k_vehicle_feasibility"] = kproof
                cert_obj = result.get("certificate")
                if isinstance(cert_obj, dict):
                    cert_obj["k_vehicle_feasibility"] = copy.deepcopy(kproof)
                    cert_obj["certified_vehicle_lower_bound"] = mem.get(
                        "certified_vehicle_lower_bound"
                    )

            mem["certification_result_provisional"] = copy.deepcopy(result)
            mem["certification_result"] = copy.deepcopy(result)
            print("\n⏳ Resultado matemático provisório pronto; aguardando AuditorAgent pós-SCIP.")
            print(f"   Nível provisório: {result.get('provisional_certification_level')}")
            print(f"   UB={result['certificate'].get('upper_bound')} | LB={result['certificate'].get('lower_bound')} | gap={result['certificate'].get('gap_pct')}%")
            return result

        except Exception as exc:
            elapsed = _time_module.time() - self.global_start_time if self.global_start_time else 0.0
            certificate = {
                "version": self.VERSION,
                "method": "CASCADE_AUDITOR_GATE_THEN_SCIP",
                "status": self.STATUS_ERROR,
                "proven_optimal": False,
                "certification_finalized": True,
                "certification_level": "D",
                "certification_declaration": "Certificação inconclusiva por erro",
                "report_allowed": False,
                "upper_bound": None,
                "lower_bound": None,
                "gap_pct": None,
                "elapsed_seconds": elapsed,
                "error_type": type(exc).__name__,
                "message": str(exc),
            }
            result = {
                "status": self.STATUS_ERROR, "proven_optimal": False,
                "certification_finalized": True, "certification_level": "D",
                "report_allowed": False, "certificate": certificate,
                "certified_solution": None,
            }
            mem["certification_result"] = copy.deepcopy(result)
            mem["optimality_certificate"] = copy.deepcopy(certificate)
            print(f"❌ CERTIFICADOR V15.1: {type(exc).__name__}: {exc}")
            return result
````

### **EXPLAINERAGENT — EXPLICAÇÃO DOS RESULTADOS**

Ele ajuda a transformar resultados técnicos em uma explicação que possa ser lida. Pense nele como quem apresenta o trabalho da equipe ao cliente.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class ExplainerAgent:
    """
    Agente Explicador consolidado para comunicação dos resultados finais.

    Responsabilidades:
    - comunicar o status da auditoria;
    - explicar o nível de certificação A/B/C/D sem confundir factibilidade
      com prova de otimalidade;
    - explicar o gap certificado de forma simples e tecnicamente correta;
    - evitar repetir métricas operacionais já exibidas no infográfico;
    - preservar neutralidade: nunca chamar uma solução de ótima sem evidência
      matemática compatível com o certificado.
    """

    VERSION = "EXPLAINER_V6_4_CONTINUITY_BOUND_AWARE"

    def __init__(self, shared_memory):
        self.shared_memory = shared_memory
        self.role = (
            "Agente Explicador: Especialista em comunicação de dados, "
            "otimização e logística."
        )
        self.goal = (
            "Comunicar performance, auditoria e nível de garantia matemática "
            "de forma clara, distinguindo melhor solução encontrada de ótimo provado."
        )
        self.backstory = (
            "Traduz resultados de otimização combinatória e certificação matemática "
            "em linguagem operacional compreensível, sem exagerar o grau de certeza."
        )
        self.tools = [
            "Formatador Markdown",
            "Interpretador de Auditoria",
            "Interpretador de Certificação A/B/C/D",
            "Integrador de BI",
        ]
        self.memory = None
        self.constraints = [
            "Depende da SharedMemory",
            "Deve manter neutralidade",
            "Não declarar ótimo global sem certificado compatível",
        ]
        self.planning = (
            "1. Resultado operacional -> 2. Objetivo -> 3. Auditoria -> "
            "4. Certificação -> 5. Recomendações."
        )
        self.actions = ["explain_solution"]

    @staticmethod
    def _safe_float(value, default=None):
        try:
            if value is None:
                return default
            return float(value)
        except Exception:
            return default

    def _extract_certificate(self, final_solution):
        """Obtém o certificado final sem depender de um único formato legado."""
        cert = final_solution.get("certificate", {}) or {}
        optimality = final_solution.get("optimality", {}) or {}

        level = (
            final_solution.get("certification_level")
            or cert.get("certification_level")
            or optimality.get("certification_level")
        )
        proven = bool(
            final_solution.get(
                "proven_optimal",
                cert.get("proven_optimal", optimality.get("proven_optimal", False)),
            )
        )
        within = bool(
            final_solution.get(
                "optimal_within_tolerance",
                cert.get(
                    "optimal_within_tolerance",
                    optimality.get("optimal_within_tolerance", False),
                ),
            )
        )
        gap_pct = self._safe_float(
            cert.get(
                "gap_pct",
                final_solution.get(
                    "optimality_gap_pct",
                    optimality.get("gap_pct"),
                ),
            ),
            None,
        )
        lower_bound = self._safe_float(cert.get("lower_bound"), None)
        upper_bound = self._safe_float(cert.get("upper_bound"), None)
        declaration = (
            cert.get("certification_declaration")
            or final_solution.get("certification_declaration")
            or optimality.get("explanation")
        )
        scip_status = cert.get("scip_status")
        lower_bound_certified = bool(cert.get("lower_bound_certified", False))

        return {
            "level": str(level).upper() if level else None,
            "proven_optimal": proven,
            "optimal_within_tolerance": within,
            "gap_pct": gap_pct,
            "lower_bound": lower_bound,
            "upper_bound": upper_bound,
            "declaration": declaration,
            "scip_status": scip_status,
            "lower_bound_certified": lower_bound_certified,
        }

    def _append_certification_explanation(self, lines, final_solution):
        cert = self._extract_certificate(final_solution)
        level = cert["level"]
        gap_pct = cert["gap_pct"]
        lb = cert["lower_bound"]
        ub = cert["upper_bound"]
        proven = cert["proven_optimal"]
        within = cert["optimal_within_tolerance"]

        lines += ["", "#### 🏆 Qualidade matemática e certificação"]

        if proven:
            lines.append(
                "✅ **ÓTIMO GLOBAL COMPROVADO (Nível A).** "
                "O solver exato provou que não existe solução factível melhor "
                "para o modelo matemático considerado."
            )
        elif level == "A" and within:
            lines.append(
                "✅ **Ótimo certificado dentro da tolerância numérica (Nível A).** "
                "A diferença restante está abaixo da tolerância numérica definida "
                "para a certificação."
            )
        elif level == "B+":
            lines.append(
                "🟢 **Solução quase ótima com garantia matemática (Nível B+).** "
                "A solução foi auditada e o gap certificado está dentro da "
                "limite de 1% do GAP."
            )
        elif level == "B":
            lines.append(
                "🟢 **Garantia intermediária na escala do sistema (Nível B).** "
                "A solução foi auditada e o gap certificado está dentro da "
                "faixa acima de 1% e até 35% da escala do sistema."
            )
        elif level == "C":
            lines.append(
                "🟡 **Melhor solução encontrada, com gap matemático certificado (Nível C).** "
                "A solução é factível e foi aprovada pelo Auditor, e existe um "
                "limite inferior matemático válido. A otimalidade global, porém, "
                "não foi provada dentro do tempo de certificação."
            )
        elif level == "D":
            lines.append(
                "🟠 **Certificação matemática inconclusiva (Nível D).** "
                "A certificação não produziu evidência suficiente para quantificar "
                "de forma confiável a distância até o ótimo global."
            )
        elif str(final_solution.get("status", "")).upper() == "SUCCESS":
            lines.append(
                "🟢 **Melhor solução factível encontrada.** A solução foi aceita "
                "operacionalmente, mas não há um nível de certificação matemática "
                "final disponível neste resultado."
            )
        else:
            lines.append(
                "⚠️ Não há solução válida suficiente para afirmar qualidade matemática."
            )
            return

        if gap_pct is not None and level in {"B+", "B", "C"}:
            lines.append(f"- **Gap certificado:** `{gap_pct:.4f}%`.")

            if lb is not None and ub is not None:
                lines.append(
                    f"- **Limite inferior (LB):** `{lb:.12g}` | "
                    f"**melhor valor factível conhecido (UB):** `{ub:.12g}`."
                )

            lines.append(
                "- **Como interpretar o gap:** em um problema de minimização, o "
                "UB representa o valor da melhor solução factível conhecida e o LB "
                "é um limite matemático abaixo do qual o ótimo global não pode estar. "
                "O gap mede a distância relativa ainda não resolvida entre esses dois "
                "limites. Portanto, ele expressa o quanto ainda falta provar sobre a "
                "otimalidade — **não significa que a solução encontrada seja esse "
                "percentual pior que o ótimo real**."
            )

            if cert["lower_bound_certified"]:
                lines.append(
                    "- O limite inferior utilizado no cálculo foi marcado pelo "
                    "Certificador como matematicamente válido."
                )

        if cert["scip_status"]:
            lines.append(
                f"- **Status do solver exato:** `{cert['scip_status']}`."
            )

        if cert["declaration"]:
            lines.append(
                f"- **Declaração do certificado:** {cert['declaration']}"
            )

    def _objective_display(self, user_answers):
        mem = self.shared_memory.data
        ctx = mem.get("objective_context", {}) or {}
        moc = (user_answers or {}).get("multiobjective_config", {}) or {}
        labels = {"DISTANCE": "Distância", "TIME": "Tempo operacional total", "COST": "Custo", "VEHICLES": "Veículos"}
        objectives = [str(x).upper() for x in (moc.get("objectives", []) or [])]
        if not objectives:
            objectives = [str(x).upper() for x in (ctx.get("active_objectives", []) or ctx.get("active_components", []) or ctx.get("objectives", []) or [])]
        if not objectives:
            txt = str((user_answers or {}).get("objective_otimizacao", "")).lower()
            objectives = ["COST"] if "custo" in txt else ["TIME"] if "tempo" in txt else ["VEHICLES"] if ("veíc" in txt or "veic" in txt) else ["DISTANCE"]
        weights = moc.get("weights", {}) or ctx.get("weights", {}) or {}
        multi = bool(moc.get("enabled", False)) or len(objectives) > 1
        parts = []
        for obj in objectives:
            w = self._safe_float(weights.get(obj, weights.get(obj.lower())), None)
            if w is not None and w > 1:
                w /= 100.0
            parts.append(f"{labels.get(obj,obj)} {w*100:.0f}%" if multi and w is not None and w > 0 else labels.get(obj,obj))
        return objectives, multi, " + ".join(parts)

    def _operational_summary(self, final_solution, user_answers):
        mem = self.shared_memory.data
        routes = final_solution.get("routes", []) or []
        valid = [r for r in routes if len(r.get("path", []) or []) > 2]
        used = sorted({int(r.get("physical_vehicle_id", r.get("vehicle_id", 0)) or 0) for r in valid if int(r.get("physical_vehicle_id", r.get("vehicle_id", 0)) or 0) > 0})
        clients = sorted({int(x) for r in valid for x in (r.get("path", []) or []) if int(x) != 0})
        distance = sum(float(r.get("distance_km", 0.0) or 0.0) for r in valid)
        total_time = sum(float(r.get("total_route_time_minutes", 0.0) or 0.0) for r in valid)
        load = sum(float(r.get("total_route_load", 0.0) or 0.0) for r in valid)
        summary = final_solution.get("summary", {}) or {}
        if not distance:
            distance = self._safe_float(final_solution.get("total_distance_km", summary.get("total_distance_km")), 0.0) or 0.0
        if not total_time:
            total_time = self._safe_float(final_solution.get("total_time_minutes", summary.get("total_time_minutes")), 0.0) or 0.0
        fin = mem.get("diagnostic_financial_breakdown", {}) or {}
        total_cost = self._safe_float(fin.get("total_cost"), None)
        if total_cost is None:
            total_cost = self._safe_float(final_solution.get("total_cost", summary.get("total_cost")), None)
        df_cost = mem.get("df_veiculos_custo_km")
        objectives, multi, objective_text = self._objective_display(user_answers)
        has_cost = bool("COST" in objectives or (df_cost is not None and not getattr(df_cost, "empty", True)) or (total_cost is not None and abs(total_cost) > 1e-12))

        # Total esperado de clientes: preferir a tabela original do problema.
        # Se ela não estiver disponível, usar o total de clientes distintos presentes
        # na solução sem inventar um denominador externo.
        expected_clients = len(clients)
        df_clients = mem.get("df_clientes")
        if df_clients is not None and not getattr(df_clients, "empty", True):
            try:
                id_col = next((c for c in ("cliente", "id_cliente", "id", "node_id") if c in df_clients.columns), None)
                if id_col is not None:
                    ids = {int(v) for v in df_clients[id_col].tolist() if int(v) != 0}
                    if ids:
                        expected_clients = len(ids)
                else:
                    expected_clients = max(0, len(df_clients) - 1)
            except Exception:
                expected_clients = len(clients)

        return {"routes": valid, "vehicles": used, "trips": len(valid), "clients": clients, "expected_clients": expected_clients,
                "distance": distance, "time": total_time, "load": load, "cost": total_cost,
                "has_cost": has_cost, "objectives": objectives, "multiobjective": multi,
                "objective_text": objective_text,
                "unit_load": (user_answers or {}).get("unidade_medida_carga", "un"),
                "currency": (user_answers or {}).get("unidade_monetaria", "R$")}

    def _append_executive_certification(self, lines, final_solution, multiobjective=False):
        cert = self._extract_certificate(final_solution)
        level, gap = cert["level"], cert["gap_pct"]
        lines += ["", "#### 📐 Qualidade matemática da solução"]
        if cert["proven_optimal"]:
            lines.append("✅ **Nível A — ótimo global comprovado.** Não existe solução factível melhor para o modelo matemático certificado.")
        elif level == "A" and cert["optimal_within_tolerance"]:
            lines.append("✅ **Nível A — ótimo certificado dentro da tolerância numérica.**")
        elif level == "B+":
            lines.append("🟢 **Nível B+ — solução quase ótima com garantia matemática.** O gap certificado está dentro da limite de 1% do GAP.")
        elif level == "B":
            lines.append("🟢 **Nível B — garantia intermediária na escala do sistema.** O gap certificado está dentro da faixa acima de 1% e até 35% da escala do sistema.")
        elif level == "C":
            lines.append("🟡 **Nível C — solução factível com limite matemático conhecido.** A solução operacional foi validada, mas a otimalidade global não foi demonstrada.")
        elif level == "D":
            lines.append("🟠 **Nível D — certificação matemática inconclusiva.** A solução factível pode ser utilizada, mas não há limite suficiente para quantificar a incerteza de otimalidade.")
        else:
            lines.append("ℹ️ A solução operacional está disponível, mas não há certificação matemática conclusiva associada a este resultado.")

        if gap is not None and level in {"B+", "B", "C"}:
            lines.append(f"- **Gap de certificação:** {gap:.2f}%.")
            lines.append("- Esse gap mede a **incerteza matemática que ainda resta para provar o ótimo global**; não significa que a solução encontrada seja esse percentual pior que o ótimo real.")

        # Valores normalizados LB/UB não são protagonistas em multiobjetivo.
        if not multiobjective and cert["lower_bound"] is not None and cert["upper_bound"] is not None:
            lines.append("- Os limites matemáticos detalhados permanecem registrados no certificado técnico do sistema.")

    def explain_solution(
        self,
        final_solution,
        problem_classification,
        user_answers,
        audit_recommendation=None,
        tentativa_atual=1,
        recomendacoes_diag=None,
    ):
        """Relatório executivo: solução operacional primeiro; prova matemática depois."""
        mem = self.shared_memory.data
        diagnostic = mem.get("diagnostic_result", {}) or mem.get("diagnostic_analysis", {}) or {}
        op = self._operational_summary(final_solution, user_answers)
        approved = bool((audit_recommendation or {}).get("solucao_aprovada", False))

        lines = [
            "=" * 72,
            "### 📑 RELATÓRIO EXECUTIVO DA OTIMIZAÇÃO",
            "=" * 72,
            "",
            "#### 🚚 Plano logístico encontrado",
            f"O sistema encontrou um plano para atender **{len(op['clients'])} clientes**, "
            f"utilizando **{len(op['vehicles'])} {'veículo' if len(op['vehicles']) == 1 else 'veículos'}** e realizando **{op['trips']} {'viagem' if op['trips'] == 1 else 'viagens'}**.",
            f"**Objetivo da otimização:** {op['objective_text']}.",
            "",
            "**Principais resultados:**",
        ]

        # Objetivos ativos aparecem primeiro; métricas complementares depois, sem score normalizado.
        shown = set()
        for obj in op["objectives"]:
            if obj == "COST" and op["has_cost"] and op["cost"] is not None:
                lines.append(f"- **Custo total:** {op['currency']} {op['cost']:.2f}"); shown.add("COST")
            elif obj == "DISTANCE":
                lines.append(f"- **Distância total:** {op['distance']:.2f} km"); shown.add("DISTANCE")
            elif obj == "TIME":
                lines.append(f"- **Tempo somado das viagens:** {op['time']:.2f} min (soma de deslocamento, atendimento e espera durante as viagens; não é a duração do expediente)"); shown.add("TIME")
            elif obj == "VEHICLES":
                lines.append(f"- **Veículos utilizados:** {len(op['vehicles'])}"); shown.add("VEHICLES")

        if "VEHICLES" not in shown:
            lines.append(f"- **Veículos utilizados:** {len(op['vehicles'])}")
        lines.append(f"- **Viagens realizadas:** {op['trips']}")
        lines.append(f"- **Clientes atendidos:** {len(op['clients'])}/{op['expected_clients']}")
        if op["load"] > 0:
            lines.append(f"- **Carga transportada:** {op['load']:.2f} {op['unit_load']}")
        if "DISTANCE" not in shown:
            lines.append(f"- **Distância total:** {op['distance']:.2f} km")
        if "TIME" not in shown:
            lines.append(f"- **Tempo somado das viagens:** {op['time']:.2f} min (soma de deslocamento, atendimento e espera durante as viagens; não é a duração do expediente)")
        if op["has_cost"] and op["cost"] is not None and "COST" not in shown:
            lines.append(f"- **Custo total:** {op['currency']} {op['cost']:.2f}")

        lines += ["", "#### ✅ Validação operacional"]
        if approved:
            lines.append("**Auditoria operacional: APROVADA.** As restrições verificadas pelo Auditor foram atendidas para a solução final.")
        else:
            lines.append("**Auditoria operacional: NÃO APROVADA ou não confirmada.** Consulte os pontos de atenção antes de utilizar o plano.")

        self._append_executive_certification(lines, final_solution, multiobjective=op["multiobjective"])

        recs = diagnostic.get("recomendacoes", []) or recomendacoes_diag or []
        if recs:
            lines += ["", "#### 💡 Recomendações operacionais prioritárias"]
            for i, rec in enumerate(recs[:5], 1):
                lines.append(f"{i}. {rec}")

        alerts = [str(a) for a in (diagnostic.get("alertas", []) or []) if str(a).strip()]
        if alerts:
            lines += ["", "#### ⚠️ Pontos de atenção"]
            for alert in alerts:
                lines.append(f"- {alert}")

        lines += [
            "",
            "#### 📊 Onde consultar os detalhes",
            "O **Plano Logístico Otimizado** acima concentra a leitura visual de frota, viagens, clientes, carga, distância, tempo e custos quando aplicáveis. O mapa/diagrama apresenta o detalhamento operacional das rotas.",
        ]
        return "\n".join(lines)
    def explain_continuity_result(self, comparison, continuity_diagnostic=None):
        """Relatório 2 comparativo; não reescreve nem apaga o Relatório 1."""
        c = comparison or {}
        before, after = c.get("before", {}) or {}, c.get("after", {}) or {}
        lines = [
            "=" * 72,
            "### 🔄 RELATÓRIO 2 — CONTINUIDADE DA OTIMIZAÇÃO",
            "=" * 72,
            "",
            "A solução do Relatório 1 foi preservada enquanto o sistema utilizou "
            "tempo computacional adicional para tentar melhorar a solução e/ou sua certificação.",
            "",
            "#### 📊 Comparação",
        ]
        def fmt(v, nd=4):
            try: return f"{float(v):.{nd}f}"
            except Exception: return "—"
        lines += [
            f"- **Objetivo canônico:** {fmt(before.get('objective'), 6)} → {fmt(after.get('objective'), 6)}",
            f"- **Distância:** {fmt(before.get('distance'), 2)} km → {fmt(after.get('distance'), 2)} km",
            f"- **Tempo:** {fmt(before.get('time'), 2)} min → {fmt(after.get('time'), 2)} min",
            f"- **Custo:** {fmt(before.get('cost'), 2)} → {fmt(after.get('cost'), 2)}",
            f"- **Veículos:** {fmt(before.get('vehicles'), 0)} → {fmt(after.get('vehicles'), 0)}",
            f"- **Certificação:** Nível {before.get('level','—')} → Nível {after.get('level','—')}",
            f"- **Gap certificado:** {fmt(before.get('gap_pct'), 2)}% → {fmt(after.get('gap_pct'), 2)}%",
            "",
            "#### ✅ Resultado da continuidade",
        ]
        def inferred_lb(side):
            lb = self._safe_float(side.get("lower_bound"), None)
            if lb is not None:
                return lb
            ub = self._safe_float(side.get("objective"), None)
            gap = self._safe_float(side.get("gap_pct"), None)
            if ub is None or gap is None:
                return None
            return ub * (1.0 - gap / 100.0)

        before_lb, after_lb = inferred_lb(before), inferred_lb(after)
        lb_strengthened = (
            before_lb is not None and after_lb is not None
            and after_lb > before_lb + max(1e-12, abs(before_lb) * 1e-9)
        )
        lb_unchanged = (
            before_lb is not None and after_lb is not None
            and abs(after_lb - before_lb) <= max(1e-12, abs(before_lb) * 1e-9)
        )

        if c.get("solution_improved") and c.get("certification_improved"):
            if lb_strengthened:
                lines.append("Foi encontrada uma solução auditada melhor e o limite inferior matemático também foi fortalecido; por isso, o intervalo de certificação diminuiu pelos dois lados.")
            elif lb_unchanged:
                lines.append("Foi encontrada uma solução auditada melhor. O limite inferior matemático permaneceu inalterado; o gap diminuiu porque o novo upper bound (melhor solução factível conhecida) ficou mais próximo desse limite.")
            else:
                lines.append("Foi encontrada uma solução auditada melhor e o gap de certificação diminuiu. O relatório não dispõe de informação suficiente para atribuir essa redução especificamente a um fortalecimento do lower bound.")
        elif c.get("solution_improved"):
            lines.append("Foi encontrada uma solução operacional auditada melhor; a força da certificação não aumentou.")
        elif c.get("certification_improved"):
            if lb_strengthened:
                lines.append("A solução operacional foi preservada e o limite inferior matemático foi fortalecido, reduzindo a incerteza da prova.")
            else:
                lines.append("A solução operacional foi preservada, mas o gap/nível de certificação foi fortalecido.")
        else:
            lines.append("O processamento adicional não encontrou solução auditada melhor nem certificação mais forte; o Relatório 1 permanece como referência.")
        if after.get("level") in {"B", "C"} and after.get("gap_pct") is not None:
            lines.append("O gap continua representando incerteza da prova matemática, e não o percentual real de piora da solução.")
        report = "\n".join(lines)
        self.shared_memory.data["continuity_report"] = report
        return report
````

### **AUDITORAGENT — REVISÃO DO RESULTADO**

Este agente revisa a coerência do resultado e de sua apresentação dentro do fluxo do sistema.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class AuditorAgent:
    """Auditor  com verificação independente incorporada.

    O Auditor não usa carga, distância, tempo, custo ou objetivo declarados pelo
    solver para decidir factibilidade. Ele reconstrói tudo através do verificador
    independente e apenas depois avalia metadados de qualidade/certificação.
    """

    VERSION = "AUDITOR_10_2_CONTINUITY_COMPATIBLE"

    def __init__(self, shared_memory=None):
        self.shared_memory = shared_memory
        self.role = "Agente Auditor Independente de Factibilidade e Certificação."
        self.goal = "Aprovar somente soluções reconstruídas independentemente a partir das rotas e dados originais."
        self.constraints = [
            "Não confiar em totais declarados pelo solver.",
            "Não declarar ótimo sem certificado matemático.",
            "Não alterar silenciosamente a solução; a versão normalizada é armazenada separadamente.",
        ]

    @staticmethod
    def _load_json(value):
        if isinstance(value, dict):
            return value
        if isinstance(value, str):
            try:
                return json.loads(value)
            except Exception:
                return {}
        return {}

    def _multiobjective_audit(self, context: AuditContext, independent_report):
        raw = self._load_json(context.translated_json)
        desc = raw.get("problem_description", {}) or {}
        cfg = desc.get("multiobjective_config", {}) or {}
        objectives = [str(x).upper() for x in cfg.get("objectives", []) or []]
        weights = cfg.get("weights", {}) or {}
        details = []
        if cfg.get("enabled"):
            if len(objectives) < 2:
                details.append("Configuração multiobjetivo sem pelo menos dois objetivos.")
            total = 0.0
            for obj in objectives:
                try:
                    w = float(weights.get(obj, 0.0))
                    if w > 1.0:
                        w /= 100.0
                    total += w
                    if w < 0:
                        details.append(f"Peso negativo para {obj}.")
                except Exception:
                    details.append(f"Peso inválido para {obj}.")
            if objectives and abs(total - 1.0) > 1e-6:
                details.append(f"Soma dos pesos = {total:.8f}; esperado 1.0.")

        recomputed = independent_report.get("recomputed", {}) or {}
        if "COST" in objectives and recomputed.get("total_cost") is None:
            details.append("CUSTO ativo mas custo total não pôde ser recalculado independentemente.")
        return {
            "status": not details,
            "enabled": bool(cfg.get("enabled", False)),
            "objectives": objectives,
            "weights": weights,
            "detalhes": details,
        }

    @staticmethod
    def _certification_assessment(result):
        cert = result.get("certificate", {}) or {}
        level = result.get("certification_level", cert.get("certification_level"))
        gap = cert.get("gap_pct", result.get("optimality_gap_pct"))
        proven = bool(result.get("proven_optimal", cert.get("proven_optimal", False)))
        within = bool(result.get("optimal_within_tolerance", cert.get("optimal_within_tolerance", False)))
        return {
            "certification_level": level,
            "proven_optimal": proven,
            "optimal_within_tolerance": within,
            "gap_pct": gap,
            "lower_bound": cert.get("lower_bound"),
            "upper_bound": cert.get("upper_bound"),
            "declaration": cert.get("certification_declaration", result.get("certification_declaration")),
        }

    def gerar_diagnostico_completo(self, context: AuditContext):
        if self.shared_memory is None:
            raise ValueError("AuditorAgent V10 exige shared_memory para a verificação independente.")
        verifier = IndependentFeasibilityVerifier(self.shared_memory)
        store_key = "independent_feasibility_report_auditor"
        report = verifier.verify(context.solver_result, store_key=store_key)
        # Camada 1 da certificação em cascata: limites rápidos e seguros são
        # calculados pelo componente interno do Auditor e armazenados na memória.
        fast_bounds = verifier.compute_fast_bounds()
        self.shared_memory.data["certification_fast_bounds"] = copy.deepcopy(fast_bounds)
        self.shared_memory.data["last_independent_feasibility_report"] = copy.deepcopy(report)
        feasible = bool(report.get("feasible", False))
        multi = self._multiobjective_audit(context, report)
        certification = self._certification_assessment(context.solver_result or {})

        dominated = bool((context.solver_result or {}).get("dominated_by_incumbent", False))
        approved = feasible and multi.get("status", True) and not dominated

        if not feasible:
            action = "TROCAR_ESTRATEGIA_MATEMATICA"
            justification = "A solução falhou na reconstrução independente de factibilidade."
        elif not multi.get("status", True):
            action = "TROCAR_ESTRATEGIA_MATEMATICA"
            justification = "A solução é fisicamente factível, mas a configuração multiobjetivo é inconsistente."
        elif dominated:
            action = "REEXECUTAR_MESMO_SOLVER"
            justification = "A solução é factível, porém dominada por um incumbent melhor."
        else:
            action = "ACEITAR_SOLUCAO"
            justification = "Factibilidade reconstruída independentemente e aprovada."

        rec = {
            "solucao_aprovada": approved,
            "factibilidade_aprovada": feasible,
            "qualidade_aprovada": approved,
            "proxima_acao_orquestrador": action,
            "justificativa_tecnica_da_recomendacao": justification,
            "independent_verification": {
                "version": report.get("version"),
                "feasible": feasible,
                "checks": copy.deepcopy(report.get("checks", {})),
                "violations": copy.deepcopy(report.get("violations", [])),
            },
            "certification_assessment": certification,
            "multiobjective_audit": multi,
        }

        diag = {
            "status_final": "APROVADO" if approved else "REPROVADO",
            "factibilidade_independente": copy.deepcopy(report),
            "fast_bounds": copy.deepcopy(fast_bounds),
            "multiobjective": multi,
            "certification_assessment": certification,
            "dominated_by_incumbent": dominated,
            "recomendacao": rec,
        }

        self.shared_memory.data["audit_diagnostic"] = copy.deepcopy(diag)
        self.shared_memory.data["audit_recommendation"] = copy.deepcopy(rec)
        return diag, rec
    def auditar_continuidade(self, context: AuditContext):
        """Audita candidato da continuidade e preserva uma cópia em namespace próprio.

        O método usa exatamente o mesmo IndependentFeasibilityVerifier do fluxo
        principal; não cria uma regra de factibilidade paralela.
        """
        diag, rec = self.gerar_diagnostico_completo(context)
        if self.shared_memory is not None:
            self.shared_memory.data["continuity_candidate_audit"] = copy.deepcopy({
                "diagnostic": diag, "recommendation": rec,
                "independent_report": self.shared_memory.data.get("independent_feasibility_report_auditor"),
            })
        return diag, rec
````

### **DIAGNOSTICAGENT — DIAGNÓSTICO E RECOMENDAÇÕES**

 Seu papel é procurar explicações, para facilitar o entendimento. Quando algo não funciona bem, ele ajuda a entender o que aconteceu.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class DiagnosticAgent:
    """
    Agente de Diagnóstico e Business Intelligence.

    Responsabilidades:
    ---------------------------------------------------------
    1. Analisar a solução produzida pelo Solver.
    2. Calcular KPIs operacionais e financeiros.
    3. Consolidar múltiplas viagens por veículo físico.
    4. Analisar ociosidade da frota.
    5. Analisar capacidade e carga.
    6. Analisar SLA / janelas de tempo.
    7. Analisar objetivos de otimização.
    8. Detectar possíveis anomalias na solução.
    9. Gerar recomendações inteligentes em linguagem natural.
    10. Gerar painel BI.
    11. Gerar mapa interativo das rotas usando PyDeck.
    12. Disponibilizar todos os resultados na SharedMemory.

    IMPORTANTE:
    ---------------------------------------------------------
    Este agente NÃO altera a solução produzida pelo Solver.
    Ele apenas analisa, diagnostica, explica e visualiza.
    """

    def __init__(self, shared_memory):

        self.shared_memory = shared_memory


        # META-ATRIBUTOS


        self.role = (
            "Agente de Diagnóstico e Business Intelligence: "
            "Especialista em análise de performance, qualidade da "
            "solução e visualização de dados logísticos."
        )

        self.goal = (
            "Transformar resultados brutos do Solver em indicadores "
            "de performance, diagnóstico operacional, alertas, "
            "recomendações estratégicas e visualizações compreensíveis "
            "para tomada de decisão."
        )

        self.backstory = (
            "Um analista de dados veterano no setor de logística. "
            "Sua expertise permite identificar gargalos operacionais, "
            "ociosidade de frota, baixa utilização de capacidade, "
            "riscos de SLA, custos excessivos, problemas de configuração "
            "e possíveis anomalias na solução matemática. "
            "Também consolida múltiplas viagens por veículo físico."
        )

        self.tools = [
            "Analisador de KPIs de Frota Consolidado",
            "Gerador de Painéis Visuais",
            "Gerador de Mapas Interativos PyDeck",
            "Calculador de Riscos de SLA e Pontualidade",
            "Analisador de Capacidade",
            "Analisador de Custos",
            "Analisador de Multiobjetivo",
            "Detector de Anomalias",
            "Formatador de Tabelas Estilizadas"
        ]

        self.memory = None

        self.constraints = [
            "Depende da existência de uma solução na SharedMemory.",
            "Não altera as rotas produzidas pelo Solver.",
            "As recomendações são baseadas nos dados disponíveis.",
            "Não deve declarar uma solução como ótima apenas porque foi "
            "marcada como SUCCESS.",
            "Deve distinguir solução ótima, factível e reprovada.",
            "Deve preservar o contexto Multi-Trip.",
            "Deve considerar must_use_all_vehicles quando disponível."
        ]

        self.planning = (
            "1. Extrair resultados do Solver. "
            "2. Validar o status da solução. "
            "3. Consolidar viagens por veículo físico. "
            "4. Calcular distância, tempo, custo, carga e utilização. "
            "5. Analisar SLA. "
            "6. Analisar objetivos e pesos. "
            "7. Detectar anomalias. "
            "8. Gerar recomendações. "
            "9. Gerar painel BI. "
            "10. Gerar mapa das rotas. "
            "11. Armazenar os resultados estruturados na SharedMemory."
        )

        self.actions = [
            "analisar_performance_e_gerar_recomendacoes",
            "analisar_resultado_otimizacao",
            "exibir_painel_visual_completo",
            "exibir_mapa_maplibre_deck",
            "Análise de ociosidade temporal por veículo real",
            "Análise de folga de tempo por cliente",
            "Análise de capacidade da frota",
            "Detecção de anomalias na solução"
        ]


    # UTILITÁRIOS


    def _safe_float(self, value, default=0.0):

        try:

            if value is None:
                return default

            if isinstance(value, str):
                value = value.replace(",", ".").strip()

                if value == "":
                    return default

            return float(value)

        except (TypeError, ValueError):
            return default



    def _safe_int(self, value, default=0):

        try:
            return int(value)

        except (TypeError, ValueError):
            try:
                return int(float(value))
            except Exception:
                return default






    def _parse_to_min(self, t):

        if isinstance(t, (int, float)):
            return float(t)

        if isinstance(t, time):

            return (
                t.hour * 60
                + t.minute
                + t.second / 60
            )

        if isinstance(t, str):

            for fmt in (
                "%H:%M:%S",
                "%H:%M"
            ):

                try:

                    dt = datetime.strptime(t, fmt)

                    return (
                        dt.hour * 60
                        + dt.minute
                        + dt.second / 60
                    )

                except ValueError:
                    continue

        return 1440.0



    def _min_to_hhmm(self, minutes):

        minutes = self._safe_float(minutes)

        h = int(minutes // 60) % 24
        m = int(minutes % 60)

        return f"{h:02d}:{m:02d}"



    def _obter_numero_veiculos(self, ans):

        keys = [
            "Qual é o número de veiculos?",
            "Qual é o número de veículos?",
            "numero_veiculos",
            "num_veiculos",
            "number_of_vehicles",
            "vehicles"
        ]

        for key in keys:

            if key in ans:

                valor = self._safe_int(
                    ans.get(key),
                    0
                )

                if valor > 0:
                    return valor

        return 0



    def _obter_tempo_maximo(self, ans):

        keys = [
            "Tempo disponível por veículo por dia (minutos)?",
            "tempo_disponivel_por_veiculo",
            "available_time_minutes",
            "tempo_maximo"
        ]

        for key in keys:

            if key in ans:

                valor = self._safe_float(
                    ans.get(key),
                    0
                )

                if valor > 0:
                    return valor

        return 1440.0


    # 1. ANÁLISE DE PERFORMANCE ORIGINAL




    # 2. NOVA FUNÇÃO:
    # ANÁLISE INTELIGENTE DA SOLUÇÃO




    # 3. PAINEL BI ORIGINAL + MELHORADO




    # 4. NOVA FUNÇÃO:
    # MAPA INTERATIVO PYDECK




    # 5. MÉTODO PRINCIPAL DO DIAGNÓSTICO




    # DIAGNÓSTICO ANALÍTICO

    def _get_vehicle_tables(self):
        mem = self.shared_memory.data
        return mem.get("df_veiculos_capacidade"), mem.get("df_veiculos_custo_km")

    def _build_trip_metrics(self):
        mem = self.shared_memory.data
        result = mem.get("solver_result", {}) or {}
        routes = result.get("routes", []) or []
        ans = mem.get("user_answers", {}) or {}
        df_cap, _ = self._get_vehicle_tables()
        cap_map = {}
        if df_cap is not None and not df_cap.empty:
            id_col = "id_veiculo" if "id_veiculo" in df_cap.columns else None
            if id_col:
                cap_map = {int(r[id_col]): float(r["capacidade"]) for _, r in df_cap.iterrows()}
            else:
                cap_map = {i + 1: float(v) for i, v in enumerate(df_cap["capacidade"].tolist())}

        out = []
        for idx, r in enumerate(routes, 1):
            path = r.get("path", []) or []
            if len(path) <= 2:
                continue
            v = int(r.get("physical_vehicle_id", r.get("vehicle_id", 1)) or 1)
            trip = int(r.get("trip_number", r.get("trip_id", idx)) or idx)
            load = float(r.get("total_route_load", 0.0) or 0.0)
            capacity = float(cap_map.get(v, 0.0) or 0.0)
            util = (100.0 * load / capacity) if capacity > 0 else 0.0
            out.append({
                "vehicle_id": v,
                "trip_number": trip,
                "path": path,
                "clients": [int(x) for x in path if int(x) != 0],
                "num_clients": len([x for x in path if int(x) != 0]),
                "load": load,
                "capacity": capacity,
                "capacity_utilization_pct": util,
                "distance_km": float(r.get("distance_km", 0.0) or 0.0),
                "time_min": float(r.get("total_route_time_minutes", 0.0) or 0.0),
                "cost": float(r.get("total_route_cost", 0.0) or 0.0),
            })
        return sorted(out, key=lambda x: (x["vehicle_id"], x["trip_number"]))

    def _vehicle_choice_assessment(self, trip_metrics):
        mem = self.shared_memory.data
        result = mem.get("solver_result", {}) or {}
        df_cap, df_cost = self._get_vehicle_tables()
        used = sorted({t["vehicle_id"] for t in trip_metrics})
        if not used:
            return {}
        v = used[0]
        cap = next((t["capacity"] for t in trip_metrics if t["vehicle_id"] == v), 0.0)
        ntrips = sum(1 for t in trip_metrics if t["vehicle_id"] == v)
        dist = sum(t["distance_km"] for t in trip_metrics if t["vehicle_id"] == v)
        cost = sum(t["cost"] for t in trip_metrics if t["vehicle_id"] == v)
        cost_km = fixed = None
        if df_cost is not None and not df_cost.empty:
            try:
                row = df_cost.iloc[v-1]
                cost_km = float(row.get("custo_por_km", 0.0))
                fixed = float(row.get("custo_diario", 0.0))
            except Exception:
                pass
        expl = (
            f"O veículo V{v} executou {ntrips} viagens, percorreu {dist:.2f} km e acumulou "
            f"custo reportado de {cost:.2f}. Sua capacidade é {cap:.2f}. "
        )
        if cost_km is not None:
            expl += f"Seu custo variável é {cost_km:.2f}/km e o custo fixo é {fixed:.2f}. "
        expl += (
            "A escolha de um veículo menor pode exigir mais retornos ao depósito; portanto, "
            "ela é economicamente interessante apenas quando a economia de custo fixo/quilômetro "
            "compensa as viagens adicionais. O Solver encontrou essa configuração como a melhor "
            "dentro da busca executada, mas isso não deve ser chamado de ótimo global sem certificado de optimalidade."
        )
        return {"vehicle_id": v, "trips": ntrips, "distance_km": dist, "cost": cost, "explanation": expl}

    def _capacity_trip_lower_bound(self):
        """Retorna um LB seguro para o número de viagens por capacidade.

        Prioriza o bound já produzido pelo Certificador. Se ele ainda não estiver
        disponível, usa ceil(demanda total / maior capacidade), que é um lower
        bound válido para capacidade por viagem.
        """
        mem = self.shared_memory.data
        result = mem.get("solver_result", {}) or {}

        def walk(obj):
            if isinstance(obj, dict):
                fb = obj.get("fast_bounds")
                if isinstance(fb, dict):
                    val = fb.get("capacity_lower_bound_on_trips")
                    if val is not None:
                        try:
                            return max(0, int(math.ceil(float(val))))
                        except Exception:
                            pass
                for key in ("certificate", "certification", "optimality", "layer1", "layer2"):
                    if key in obj:
                        found = walk(obj.get(key))
                        if found is not None:
                            return found
            return None

        certified = walk(result)
        if certified is not None:
            return certified, "CERTIFIER"

        df_clients = mem.get("df_clientes")
        df_cap = mem.get("df_veiculos_capacidade")
        try:
            if df_clients is None or df_cap is None or df_clients.empty or df_cap.empty:
                return None, None
            demand_col = next((c for c in ("demanda", "Demanda", "demand") if c in df_clients.columns), None)
            if demand_col is None or "capacidade" not in df_cap.columns:
                return None, None
            total_demand = sum(max(0.0, self._safe_float(v, 0.0)) for v in df_clients[demand_col].tolist())
            max_capacity = max(self._safe_float(v, 0.0) for v in df_cap["capacidade"].tolist())
            if max_capacity <= 0:
                return None, None
            return int(math.ceil(max(0.0, total_demand) / max_capacity - 1e-12)), "CAPACITY_FALLBACK"
        except Exception:
            return None, None

    def analisar_resultado_otimizacao(self):
        mem = self.shared_memory.data
        result = mem.get("solver_result", {}) or {}
        ans = mem.get("user_answers", {}) or {}
        summary = result.get("summary", {}) or {}
        trip_metrics = self._build_trip_metrics()
        used = sorted({t["vehicle_id"] for t in trip_metrics})
        try:
            total_vehicles = int(ans.get("Qual é o número de veiculos?", summary.get("num_vehicles_total", 0)) or 0)
        except Exception:
            total_vehicles = int(summary.get("num_vehicles_total", 0) or 0)

        must_use = bool(ans.get("must_use_all_vehicles", False))
        active_obj = []
        moc = ans.get("multiobjective_config", {}) or {}
        if moc.get("enabled"):
            active_obj = [str(x).upper() for x in moc.get("objectives", [])]
        else:
            txt = str(ans.get("objective_otimizacao", "")).lower()
            active_obj = ["COST"] if "custo" in txt else ["TIME"] if "tempo" in txt else ["VEHICLES"] if "veíc" in txt or "veic" in txt else ["DISTANCE"]

        alerts, recs, positives = [], [], []
        if total_vehicles and len(used) < total_vehicles:
            reduction = 100.0 * (total_vehicles - len(used)) / total_vehicles
            if must_use:
                alerts.append(f"Apenas {len(used)} de {total_vehicles} veículos foram utilizados, embora todos sejam obrigatórios.")
            else:
                positives.append(f"Consolidação de frota: {len(used)} de {total_vehicles} veículos utilizados ({reduction:.1f}% dos veículos permaneceram disponíveis).")
                if "COST" in active_obj or "VEHICLES" in active_obj:
                    recs.append("Em uma próxima otimização, vale testar a manutenção desta consolidação de frota; qualquer nova configuração deve ser novamente otimizada e auditada antes do uso operacional.")

        if trip_metrics:
            avg_util = sum(t["capacity_utilization_pct"] for t in trip_metrics) / len(trip_metrics)
            low = [t for t in trip_metrics if t["capacity_utilization_pct"] < 50]
            current_trips = len(trip_metrics)
            trip_lb, trip_lb_source = self._capacity_trip_lower_bound()
            if trip_lb is not None and current_trips <= trip_lb:
                recs.append(
                    f"Utilização média da capacidade por viagem: {avg_util:.1f}%. "
                    f"O plano utiliza {current_trips} {'viagem' if current_trips == 1 else 'viagens'}, "
                    f"valor que coincide com o limite inferior seguro de {trip_lb} imposto pela capacidade. "
                    "Portanto, reduzir o número de viagens não é possível sem alterar dados ou restrições do cenário."
                )
            else:
                recs.append(
                    f"Utilização média da capacidade por viagem: {avg_util:.1f}%. Esse indicador, isoladamente, "
                    "não demonstra que seja possível reduzir o número de viagens. Em uma próxima otimização, "
                    "o Solver pode testar alternativas de consolidação, sempre respeitando capacidade, jornada "
                    "e janelas de tempo e submetendo qualquer nova solução ao Auditor."
                )
            if low:
                n_low = len(low)
                viagem_txt = "viagem" if n_low == 1 else "viagens"
                recs.append(f"Foi identificada {n_low} {viagem_txt} com utilização inferior a 50% da capacidade." if n_low == 1 else f"Foram identificadas {n_low} {viagem_txt} com utilização inferior a 50% da capacidade.")
                if trip_lb is not None and current_trips <= trip_lb:
                    recs[-1] += " A baixa ocupação não indica redundância: o número total de viagens já está no limite inferior seguro de capacidade."
                else:
                    recs[-1] += " Isso não implica que a viagem possa ser eliminada, pois sua necessidade pode decorrer de capacidade, janelas de tempo, jornada ou da própria estrutura das rotas. Qualquer redução só deve ser considerada se uma nova solução factível for encontrada e novamente aprovada pelo Auditor."

        optimality = result.get("optimality", {}) or {}
        if result.get("status") == "SUCCESS" and not optimality.get("proven_optimal", False):
            positives.append("A solução é factível e aprovada, mas a optimalidade global não foi comprovada pelo método utilizado.")

        fleet_message = (positives[0] if positives else "A utilização da frota é compatível com as restrições informadas.")
        diagnostic = {
            "resumo": "Solução factível analisada sob custo, frota, viagens, capacidade e qualidade matemática.",
            "alertas": alerts,
            "recomendacoes": recs,
            "pontos_positivos": positives,
            "trip_metrics": trip_metrics,
            "optimality_assessment": optimality,
            "fleet_assessment": {"used": len(used), "total": total_vehicles, "message": fleet_message},
            "vehicle_choice_assessment": self._vehicle_choice_assessment(trip_metrics),
        }
        mem["diagnostic_analysis"] = diagnostic
        return diagnostic

    def analisar_performance_e_gerar_recomendacoes(self):
        d = self.analisar_resultado_otimizacao()
        return d.get("recomendacoes", [])



    # MÉTRICAS FINANCEIRAS E OPERACIONAIS


    def _financial_breakdown(self):
        """
        Reconstrói a decomposição financeira sem alterar solver_result.

        Prioridade:
        1) campos explicitamente reportados nas rotas;
        2) summary/resultado do solver;
        3) reconstrução por distância x custo/km + custo fixo por veículo.

        Retorna:
            dict(total_cost, variable_cost, fixed_cost, source)
        """
        mem = self.shared_memory.data
        result = mem.get("solver_result", {}) or {}
        summary = result.get("summary", {}) or {}
        routes = result.get("routes", []) or []
        ans = mem.get("user_answers", {}) or {}
        df_cost = mem.get("df_veiculos_custo_km")

        valid_routes = [
            r for r in routes
            if len(r.get("path", []) or []) > 2
        ]


        # 1) Valores explícitos

        route_variable_present = any(
            "variable_cost" in r and r.get("variable_cost") is not None
            for r in valid_routes
        )
        route_fixed_present = any(
            "fixed_cost" in r and r.get("fixed_cost") is not None
            for r in valid_routes
        )

        route_variable = sum(
            self._safe_float(r.get("variable_cost", 0.0))
            for r in valid_routes
        )
        route_fixed = sum(
            self._safe_float(r.get("fixed_cost", 0.0))
            for r in valid_routes
        )

        summary_variable = self._safe_float(
            summary.get(
                "total_variable_cost",
                result.get("total_variable_cost", 0.0)
            ),
            0.0
        )
        summary_fixed = self._safe_float(
            summary.get(
                "total_fixed_cost",
                result.get("total_fixed_cost", 0.0)
            ),
            0.0
        )


        # 2) Tabela de custos

        cost_km_map = {}
        fixed_map = {}

        if df_cost is not None and not getattr(df_cost, "empty", True):
            try:
                if "id_veiculo" in df_cost.columns:
                    for _, row in df_cost.iterrows():
                        vid = self._safe_int(row.get("id_veiculo"), 0)
                        if vid <= 0:
                            continue
                        cost_km_map[vid] = self._safe_float(
                            row.get("custo_por_km", 0.0)
                        )
                        fixed_map[vid] = self._safe_float(
                            row.get("custo_diario", 0.0)
                        )
                else:
                    for i, (_, row) in enumerate(df_cost.iterrows(), 1):
                        cost_km_map[i] = self._safe_float(
                            row.get("custo_por_km", 0.0)
                        )
                        fixed_map[i] = self._safe_float(
                            row.get("custo_diario", 0.0)
                        )
            except Exception:
                cost_km_map = {}
                fixed_map = {}

        reconstructed_variable = 0.0
        used_vehicles = set()

        num_real = self._obter_numero_veiculos(ans)

        for r in valid_routes:
            v = self._safe_int(
                r.get(
                    "physical_vehicle_id",
                    r.get("vehicle_id", 1)
                ),
                1
            )

            if (
                "physical_vehicle_id" not in r
                and num_real > 0
            ):
                v = ((v - 1) % num_real) + 1

            used_vehicles.add(v)

            distance = self._safe_float(
                r.get("distance_km", 0.0)
            )
            reconstructed_variable += (
                distance * cost_km_map.get(v, 0.0)
            )

        # Modo de custo fixo.
        fixed_mode = str(
            ans.get(
                "fixed_cost_mode",
                ans.get(
                    "fixed_cost_type",
                    ans.get(
                        "tipo_custo_fixo",
                        "WHEN_USED"
                    )
                )
            )
        ).strip().upper()

        aliases_always = {
            "ALWAYS",
            "ALL",
            "EVERY_VEHICLE",
            "SEMPRE",
            "INCONDICIONAL",
            "2"
        }

        if fixed_mode in aliases_always:
            reconstructed_fixed = sum(
                fixed_map.values()
            )
        else:
            reconstructed_fixed = sum(
                fixed_map.get(v, 0.0)
                for v in used_vehicles
            )


        # 3) Escolha robusta

        if route_variable_present and route_variable > 0:
            variable_cost = route_variable
            variable_source = "routes"
        elif summary_variable > 0:
            variable_cost = summary_variable
            variable_source = "summary"
        else:
            variable_cost = reconstructed_variable
            variable_source = "reconstructed"

        if route_fixed_present and route_fixed > 0:
            fixed_cost = route_fixed
            fixed_source = "routes"
        elif summary_fixed > 0:
            fixed_cost = summary_fixed
            fixed_source = "summary"
        else:
            fixed_cost = reconstructed_fixed
            fixed_source = "reconstructed"

        reported_total = self._safe_float(
            result.get(
                "total_cost",
                summary.get("total_cost", 0.0)
            ),
            0.0
        )

        components_total = variable_cost + fixed_cost

        # O total oficial do solver continua tendo prioridade,
        # mas os componentes são preservados e reconciliados.
        total_cost = (
            reported_total
            if reported_total > 0
            else components_total
        )

        reconciliation_gap = total_cost - components_total

        return {
            "total_cost": round(total_cost, 6),
            "variable_cost": round(variable_cost, 6),
            "fixed_cost": round(fixed_cost, 6),
            "components_total": round(components_total, 6),
            "reconciliation_gap": round(reconciliation_gap, 6),
            "source": {
                "variable": variable_source,
                "fixed": fixed_source,
                "total": (
                    "solver"
                    if reported_total > 0
                    else "components"
                )
            }
        }



    # COMPONENTES GRÁFICOS DO INFOGRÁFICO


    def _draw_truck(self, ax, x, y, scale=1.0, color="#1565C0"):
        """Desenha um caminhão vetorial simples, sem depender de arquivo externo."""
        from matplotlib.patches import Rectangle, Circle, FancyBboxPatch, Polygon

        # Baú
        ax.add_patch(FancyBboxPatch(
            (x, y), 0.12 * scale, 0.045 * scale,
            boxstyle="round,pad=0.002,rounding_size=0.004",
            facecolor="#E9F2FF", edgecolor="#0B3A6E", linewidth=1.2
        ))
        # Cabine
        ax.add_patch(FancyBboxPatch(
            (x + 0.12 * scale, y), 0.055 * scale, 0.045 * scale,
            boxstyle="round,pad=0.002,rounding_size=0.006",
            facecolor=color, edgecolor="#0B3A6E", linewidth=1.2
        ))
        # Para-brisa
        ax.add_patch(Polygon([
            (x + 0.132 * scale, y + 0.036 * scale),
            (x + 0.163 * scale, y + 0.036 * scale),
            (x + 0.169 * scale, y + 0.022 * scale),
            (x + 0.132 * scale, y + 0.022 * scale),
        ], closed=True, facecolor="#CDEBFF", edgecolor="#0B3A6E", linewidth=0.7))
        # Rodas
        for wx in (x + 0.035 * scale, x + 0.135 * scale):
            ax.add_patch(Circle((wx, y - 0.002 * scale), 0.010 * scale,
                                facecolor="#263238", edgecolor="white", linewidth=0.8))
            ax.add_patch(Circle((wx, y - 0.002 * scale), 0.004 * scale,
                                facecolor="#90A4AE", edgecolor="none"))


    def _objective_context_for_display(self):
        """Retorna objetivos/pesos em formato estável exclusivamente para apresentação."""
        mem = self.shared_memory.data
        ans = mem.get("user_answers", {}) or {}
        ctx = mem.get("objective_context", {}) or {}
        moc = ans.get("multiobjective_config", {}) or {}
        labels = {"TIME": "Tempo operacional total", "COST": "Custo", "DISTANCE": "Distância", "VEHICLES": "Veículos"}

        objectives = [str(x).upper() for x in (moc.get("objectives", []) or [])]
        if not objectives:
            objectives = [str(x).upper() for x in (ctx.get("active_objectives", []) or ctx.get("active_components", []) or ctx.get("objectives", []) or [])]
        if not objectives:
            txt = str(ans.get("objective_otimizacao", ans.get("objective", ""))).lower()
            if "custo" in txt or "cost" in txt:
                objectives = ["COST"]
            elif "tempo" in txt or "time" in txt:
                objectives = ["TIME"]
            elif "veíc" in txt or "veic" in txt or "vehicle" in txt:
                objectives = ["VEHICLES"]
            else:
                objectives = ["DISTANCE"]

        weights = moc.get("weights", {}) or ctx.get("weights", {}) or {}
        normalized_weights = {}
        for obj in objectives:
            w = self._safe_float(weights.get(obj, weights.get(obj.lower(), 0.0)), 0.0)
            if w > 1.0:
                w /= 100.0
            normalized_weights[obj] = w

        multi = bool(moc.get("enabled", False)) or len(objectives) > 1
        if multi:
            parts = []
            for obj in objectives:
                w = normalized_weights.get(obj, 0.0)
                parts.append(f"{labels.get(obj, obj)} {w*100:.0f}%" if w > 0 else labels.get(obj, obj))
            text = " + ".join(parts)
        else:
            text = labels.get(objectives[0], objectives[0])

        return {"objectives": objectives, "weights": normalized_weights, "multiobjective": multi, "text": text}

    def _extract_certification_display(self):
        """Lê o certificado sem recalcular bounds ou alterar a solução."""
        mem = self.shared_memory.data
        result = mem.get("solver_result", {}) or {}
        candidates = [
            result,
            mem.get("final_solution", {}) or {},
            mem.get("certification_result", {}) or {},
            mem.get("optimality_certificate", {}) or {},
        ]
        cert = {}
        optimality = {}
        source = result
        for c in candidates:
            if not isinstance(c, dict):
                continue
            cc = c.get("certificate", {}) or {}
            oo = c.get("optimality", {}) or {}
            if cc or oo or c.get("certification_level"):
                source, cert, optimality = c, cc, oo
                break

        level = source.get("certification_level") or cert.get("certification_level") or optimality.get("certification_level")
        proven = bool(source.get("proven_optimal", cert.get("proven_optimal", optimality.get("proven_optimal", False))))
        within = bool(source.get("optimal_within_tolerance", cert.get("optimal_within_tolerance", optimality.get("optimal_within_tolerance", False))))
        gap = cert.get("gap_pct", source.get("optimality_gap_pct", optimality.get("gap_pct")))
        gap = self._safe_float(gap, None)
        lb = self._safe_float(cert.get("lower_bound", optimality.get("lower_bound")), None)
        ub = self._safe_float(cert.get("upper_bound", optimality.get("upper_bound")), None)
        level = str(level).upper() if level else None

        if proven:
            short = "Nível A · ótimo global comprovado"
        elif level == "A" and within:
            short = "Nível A · ótimo dentro da tolerância"
        elif level == "B+":
            short = "Nível B+ · garantia matemática forte"
        elif level == "B":
            short = "Nível B · garantia intermediária"
        elif level == "C":
            short = "Nível C · limite matemático conhecido"
        elif level == "D":
            short = "Nível D · certificação inconclusiva"
        else:
            short = "Sem certificação matemática conclusiva"
        return {"level": level, "proven_optimal": proven, "within_tolerance": within,
                "gap_pct": gap, "lower_bound": lb, "upper_bound": ub, "short": short}

    def _has_cost_context(self, objectives, financial):
        mem = self.shared_memory.data
        df_cost = mem.get("df_veiculos_custo_km")
        table_present = df_cost is not None and not getattr(df_cost, "empty", True)
        return bool("COST" in objectives or table_present or abs(self._safe_float(financial.get("total_cost"), 0.0)) > 1e-12)

    def _financial_breakdown_v2(self, trips=None):
        """Compatibilidade: usa a decomposição financeira oficial já consolidada."""
        return self._financial_breakdown()

    def _build_infographic_payload(self):
        """Consolida dados executivos do infográfico sem alterar solver_result."""
        mem = self.shared_memory.data
        result = mem.get("solver_result", {}) or {}
        ans = mem.get("user_answers", {}) or {}
        trips = self._build_trip_metrics()
        financial = self._financial_breakdown()
        diagnostic = mem.get("diagnostic_analysis", {}) or {}
        obj = self._objective_context_for_display()
        cert = self._extract_certification_display()

        total_vehicles = self._obter_numero_veiculos(ans)
        used = sorted({int(t.get("vehicle_id", 1)) for t in trips})
        clients = sorted({int(c) for t in trips for c in (t.get("clients", []) or []) if int(c) != 0})
        total_load = sum(self._safe_float(t.get("load", 0)) for t in trips)
        total_distance = sum(self._safe_float(t.get("distance_km", 0)) for t in trips)
        total_time = sum(self._safe_float(t.get("time_min", 0)) for t in trips)
        avg_util = (sum(self._safe_float(t.get("capacity_utilization_pct", 0)) for t in trips) / len(trips)) if trips else 0.0
        has_cost = self._has_cost_context(obj["objectives"], financial)

        # SLA: somente leitura dos detalhes já produzidos pelo solver.
        df_j = mem.get("df_janela_tempo_cliente")
        late, total_client_visits, window_map = 0, 0, {}
        if df_j is not None and not getattr(df_j, "empty", True):
            try:
                for _, row in df_j.iterrows():
                    window_map[int(row["cliente"])] = self._parse_to_min(row["fim_janela"])
            except Exception:
                window_map = {}
        for r in result.get("routes", []) or []:
            for d in r.get("path_details", []) or []:
                cid = self._safe_int(d.get("node_id", 0), 0)
                if cid == 0:
                    continue
                total_client_visits += 1
                service_or_arrival = d.get("service_start_time_minutes", d.get("arrival_time_minutes", 0))
                if cid in window_map and self._safe_float(service_or_arrival, 0) > window_map[cid] + 1e-7:
                    late += 1
        sla_pct = 100.0 * (total_client_visits - late) / total_client_visits if total_client_visits else 100.0

        audit = mem.get("audit_recommendation", {}) or mem.get("audit_result", {}) or {}
        audit_approved = bool(audit.get("solucao_aprovada", audit.get("approved", result.get("status") in {"SUCCESS", "OPTIMAL"})))
        if cert["proven_optimal"]:
            quality = "ÓTIMO GLOBAL COMPROVADO"
        elif cert["level"] == "A" and cert["within_tolerance"]:
            quality = "ÓTIMO DENTRO DA TOLERÂNCIA"
        else:
            quality = "SOLUÇÃO FACTÍVEL AUDITADA" if audit_approved else "MELHOR SOLUÇÃO ENCONTRADA"

        recs = diagnostic.get("recomendacoes", []) or []
        recommendation = recs[0] if recs else "Utilizar o plano auditado como referência operacional; qualquer alteração de rotas deve ser submetida a nova otimização e auditoria."

        return {
            "trips": trips, "financial": financial, "has_cost": has_cost,
            "total_vehicles": total_vehicles, "used": used, "clients": clients,
            "clients_served": len(clients), "total_load": total_load,
            "total_distance": total_distance, "total_time": total_time, "avg_util": avg_util,
            "objective_text": obj["text"], "objectives": obj["objectives"], "weights": obj["weights"],
            "multiobjective": obj["multiobjective"], "sla_pct": sla_pct, "late": late,
            "quality": quality, "audit_approved": audit_approved, "certification": cert,
            "recommendation": recommendation,
            "unit_load": ans.get("unidade_medida_carga", "un"),
            "unit_currency": ans.get("unidade_monetaria", "R$"),
        }


    # TABELAS INTERNAS DE BI (SEM EXIBIÇÃO AO CLIENTE)

    def _atualizar_tabelas_bi_sem_exibir(self):
        """
        Mantém as estruturas tabulares na SharedMemory para uso interno,
        sem exibi-las ao cliente no relatório final.
        """
        mem = self.shared_memory.data
        ans = mem.get("user_answers", {}) or {}
        trips = self._build_trip_metrics()
        financial = self._financial_breakdown_v2(trips)

        unit_load = ans.get("unidade_medida_carga", "un")
        unit_currency = ans.get("unidade_monetaria", "R$")
        total_vehicles = self._obter_numero_veiculos(ans)
        used = sorted({int(t.get("vehicle_id", 0)) for t in trips if int(t.get("vehicle_id", 0)) > 0})

        # Objetivos
        moc = ans.get("multiobjective_config", {}) or {}
        objectives = [str(x).upper() for x in moc.get("objectives", [])]
        weights = moc.get("weights", {}) or {}
        if not objectives:
            txt = str(ans.get("objective_otimizacao", "")).lower()
            if "custo" in txt:
                objectives = ["COST"]
            elif "tempo" in txt:
                objectives = ["TIME"]
            elif "veíc" in txt or "veic" in txt:
                objectives = ["VEHICLES"]
            else:
                objectives = ["DISTANCE"]

        total_distance = sum(self._safe_float(t.get("distance_km", 0)) for t in trips)
        total_time = sum(self._safe_float(t.get("time_min", 0)) for t in trips)

        labels = {
            "DISTANCE": "Distância",
            "TIME": "Tempo operacional total",
            "COST": "Custo",
            "VEHICLES": "Número de veículos",
        }
        objective_rows = []
        for obj in objectives:
            if obj == "DISTANCE":
                value = f"{total_distance:.2f} km"
            elif obj == "TIME":
                value = f"{total_time:.2f} min"
            elif obj == "COST":
                value = f"{unit_currency} {financial['total_cost']:.2f}"
            elif obj == "VEHICLES":
                value = str(len(used))
            else:
                value = "N/A"
            w = self._safe_float(weights.get(obj, 0.0))
            if w > 1:
                w /= 100.0
            objective_rows.append({
                "Objetivo": labels.get(obj, obj),
                "Resultado otimizado": value,
                "Peso": f"{w*100:.1f}%" if moc.get("enabled") else "-",
            })

        # Frota
        stats = {}
        for t in trips:
            v = int(t.get("vehicle_id", 0))
            st = stats.setdefault(v, {"viagens": 0, "carga": 0.0, "distancia": 0.0, "tempo": 0.0, "custo": 0.0})
            st["viagens"] += 1
            st["carga"] += self._safe_float(t.get("load", 0))
            st["distancia"] += self._safe_float(t.get("distance_km", 0))
            st["tempo"] += self._safe_float(t.get("time_min", 0))
            st["custo"] += self._safe_float(t.get("cost", 0))

        tempo_max = self._obter_tempo_maximo(ans)
        fleet_rows = []
        for v in range(1, max(total_vehicles, max(used or [0])) + 1):
            st = stats.get(v, {"viagens": 0, "carga": 0.0, "distancia": 0.0, "tempo": 0.0, "custo": 0.0})
            idle_pct = max(0.0, 100.0 * (1.0 - st["tempo"] / tempo_max)) if tempo_max > 0 else 0.0
            fleet_rows.append({
                "Veículo": f"V{v}",
                "Viagens": st["viagens"],
                f"Custo ({unit_currency})": round(st["custo"], 2),
                "Km Total": round(st["distancia"], 2),
                "Tempo em Rota (min)": round(st["tempo"], 1),
                f"Carga Total ({unit_load})": round(st["carga"], 2),
                "Ociosidade (%)": f"{idle_pct:.1f}%",
            })

        trip_rows = []
        for t in trips:
            trip_rows.append({
                "Veículo": f"V{t['vehicle_id']}",
                "Viagem": t["trip_number"],
                "Clientes": " → ".join(f"C{x}" for x in t.get("clients", [])),
                f"Carga ({unit_load})": round(self._safe_float(t.get("load", 0)), 2),
                "Capacidade (%)": f"{self._safe_float(t.get('capacity_utilization_pct', 0)):.1f}%",
                "Distância (km)": round(self._safe_float(t.get("distance_km", 0)), 2),
                "Tempo (min)": round(self._safe_float(t.get("time_min", 0)), 1),
                f"Custo ({unit_currency})": round(self._safe_float(t.get("cost", 0)), 2),
                "Horário": f"{t.get('start_hhmm','--:--')} → {t.get('end_hhmm','--:--')}",
            })

        mem["diagnostic_bi_tables"] = {
            "objectives": objective_rows,
            "fleet_summary": fleet_rows,
            "financial_summary": [{
                f"Custo Total ({unit_currency})": round(financial["total_cost"], 2),
                f"Custo Variável ({unit_currency})": round(financial["variable_cost"], 2),
                f"Custo Fixo ({unit_currency})": round(financial["fixed_cost"], 2),
                "Veículos Utilizados": len(used),
                "Veículos Disponíveis": total_vehicles,
                "Distância Total (Km)": round(total_distance, 2),
                "Tempo Total (min)": round(total_time, 2),
            }],
            "trip_summary": trip_rows,
        }
        mem["diagnostic_financial_breakdown"] = financial
        return mem["diagnostic_bi_tables"]

    def exibir_painel_visual_completo(self, objective=None, exibir=False):
        """
        Interface pública mantida para integração com o Orchestrator.

        Por padrão, não exibe tabelas no relatório do cliente.
        Os dados continuam sendo calculados e armazenados na SharedMemory.
        """
        tables = self._atualizar_tabelas_bi_sem_exibir()
        if exibir:
            print("ℹ️ As tabelas BI estão armazenadas em memory.data['diagnostic_bi_tables'].")
        return self.analisar_resultado_otimizacao()


    VERSION = "DIAGNOSTIC_EXECUTIVE_DASHBOARD_V2_3_CONTINUITY"


    # PLANO DE VIAGENS



    VERSION = "DIAGNOSTIC_EXECUTIVE_DASHBOARD_V2_3_CONTINUITY"



    # DASHBOARD LOGÍSTICO ADAPTATIVO

    def exibir_diagrama_fluxo_logistico(
        self,
        figsize_width=None,
        salvar_png=False,
        filename="plano_logistico_otimizado_v13.png",
        modo_tamanho="auto"
    ):
        """
        Infográfico logístico em formato de dashboard tabular.

        Organização:
        - cabeçalho executivo;
        - 6 cartões KPI;
        - faixa de contexto (objetivo, janelas, frota ativa e custos);
        - plano de viagens em formato tabular por colunas;
        - rota visual D -> clientes -> D;
        - resumo operacional;
        - utilização da capacidade;
        - destaques operacionais;
        - composição do custo;
        - recomendação principal e legenda.

        O método é estritamente visual e NÃO altera a solução do Solver.
        """
        import numpy as np
        import matplotlib.pyplot as plt
        from matplotlib.patches import FancyBboxPatch, Rectangle, Circle, Wedge, Polygon

        p = self._build_infographic_payload()
        trips = list(p.get("trips", []) or [])
        if not trips:
            print("⚠️ Não há viagens para gerar o infográfico logístico.")
            return None

        trips = sorted(
            trips,
            key=lambda t: (
                int(t.get("vehicle_id", 0)),
                int(t.get("trip_number", 0))
            )
        )

        used = sorted({
            int(t.get("vehicle_id", 0))
            for t in trips
            if int(t.get("vehicle_id", 0)) > 0
        })
        n = len(trips)


        # Tamanho adaptativo — quadrado para casos típicos.
        # Para muitos registros aumenta discretamente a altura.

        if figsize_width is None:
            if modo_tamanho == "compacto":
                figsize_width = 15
            elif modo_tamanho == "detalhado":
                figsize_width = 20
            else:
                figsize_width = 17 if n <= 20 else 19

        if n <= 20:
            figsize_height = float(figsize_width)
        else:
            figsize_height = float(figsize_width) * min(1.45, 1.0 + (n - 20) * 0.018)


        # Paleta

        NAVY = "#0B356A"
        BLUE = "#1565C0"
        GREEN = "#2FA84F"
        ORANGE = "#FB8C00"
        PURPLE = "#7E57C2"
        TEAL = "#12A7B5"
        GOLD = "#F6B400"
        RED = "#F0624D"
        BG = "#F3F7FB"
        CARD = "#FFFFFF"
        TEXT = "#18324A"
        MUTED = "#6F8496"
        GRID = "#D9E5F0"

        palette = [
            BLUE, GREEN, ORANGE, PURPLE, TEAL, GOLD,
            "#EC407A", "#26A69A", "#5C6BC0", "#8D6E63"
        ]
        vehicle_color = {
            v: palette[i % len(palette)]
            for i, v in enumerate(used)
        }

        fig = plt.figure(
            figsize=(figsize_width, figsize_height),
            dpi=150,
            facecolor=BG
        )
        ax = fig.add_axes([0, 0, 1, 1])
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        ax.axis("off")

        # Escala tipográfica pensada para visualização no Colab.
        # Não depende diretamente do tamanho do canvas para evitar encolhimento visual.
        font_boost = 1.0
        if modo_tamanho == "detalhado":
            font_boost = 1.12
        elif modo_tamanho == "compacto":
            font_boost = 0.94

        def fsz(v):
            return float(v) * font_boost

        def card(x, y, w, h, fc=CARD, ec=GRID, radius=0.009, lw=1.0):
            patch = FancyBboxPatch(
                (x, y), w, h,
                boxstyle=f"round,pad=0.003,rounding_size={radius}",
                facecolor=fc,
                edgecolor=ec,
                linewidth=lw
            )
            ax.add_patch(patch)
            return patch

        def txt(x, y, s, size=9, weight="normal", color=TEXT,
                ha="left", va="center"):
            ax.text(
                x, y, str(s),
                fontsize=fsz(size),
                fontweight=weight,
                color=color,
                ha=ha,
                va=va,
                family="DejaVu Sans"
            )

        def depot(cx, cy, color=NAVY, scale=1.0):
            w = 0.017 * scale
            h = 0.018 * scale
            ax.add_patch(Rectangle(
                (cx-w/2, cy-h/2), w, h*0.70,
                facecolor="#EDF4FB", edgecolor=color, linewidth=0.9
            ))
            ax.add_patch(Polygon([
                (cx-w*0.62, cy+h*0.12),
                (cx, cy+h*0.62),
                (cx+w*0.62, cy+h*0.12)
            ], closed=True, facecolor=color, edgecolor=color))
            txt(cx, cy-h*0.03, "D", 7.2, "bold", color, ha="center")


        # CABEÇALHO

        card(0.015, 0.888, 0.97, 0.098, fc=NAVY, ec=NAVY, radius=0.018)
        self._draw_truck(ax, 0.030, 0.914, scale=1.20, color="#1976D2")
        txt(0.255, 0.954, "PLANO LOGÍSTICO OTIMIZADO", 19.5, "bold", "#A9D6FF")
        txt(0.255, 0.925, "Solução completa gerada pelo sistema multiagente", 10.8, "normal", "#DCEBFA")
        txt(0.255, 0.902, f"Objetivo: {p['objective_text']}", 10.4, "bold", "white")

        qfc = "#2E7D32"
        card(0.742, 0.905, 0.228, 0.060, fc=qfc, ec="#62B865", radius=0.012)
        ax.add_patch(Circle((0.777, 0.935), 0.018, facecolor="white", edgecolor="none"))
        txt(0.777, 0.935, "✓", 14, "bold", qfc, ha="center")
        txt(0.815, 0.948, "QUALIDADE DA SOLUÇÃO", 8.3, "bold", "white")
        txt(0.815, 0.929, p.get("quality", "MELHOR SOLUÇÃO ENCONTRADA"), 8.2, "bold", "white")
        txt(0.815, 0.912, p.get("certification", {}).get("short", "Auditoria operacional aprovada"), 6.8, "normal", "#DDF5DE")


        # KPIs

        kpi_y, kpi_h = 0.817, 0.056
        # KPIs adaptativos: objetivos ativos primeiro; custo só aparece quando existe contexto de custo.
        metric_map = {
            "DISTANCE": ("DISTÂNCIA", f"{p['total_distance']:.2f} km", "percorridos", BLUE),
            "TIME": ("TEMPO", f"{p['total_time']:.2f} min", "soma das viagens", PURPLE),
            "COST": ("CUSTO TOTAL", f"{p['unit_currency']} {p['financial']['total_cost']:.2f}", "fixo + variável", GREEN),
            "VEHICLES": ("VEÍCULOS", f"{len(used)}", f"de {p['total_vehicles']} disponíveis", BLUE),
        }
        kpis = []
        for obj in p.get("objectives", []):
            if obj == "COST" and not p.get("has_cost"):
                continue
            item = metric_map.get(obj)
            if item and item not in kpis:
                kpis.append(item)
        complements = [
            ("CLIENTES", f"{p['clients_served']}", "atendidos", TEAL),
            ("VEÍCULOS", f"{len(used)}", f"de {p['total_vehicles']} disponíveis", BLUE),
            ("VIAGENS", f"{len(trips)}", "realizadas", GREEN),
            ("DEMANDA", f"{p['total_load']:.0f}", f"{p['unit_load']} entregues", ORANGE),
            ("DISTÂNCIA", f"{p['total_distance']:.2f} km", "percorridos", BLUE),
            ("TEMPO", f"{p['total_time']:.2f} min", "soma das viagens", PURPLE),
        ]
        if p.get("has_cost"):
            complements.append(("CUSTO TOTAL", f"{p['unit_currency']} {p['financial']['total_cost']:.2f}", "fixo + variável", GREEN))
        titles = {x[0] for x in kpis}
        for item in complements:
            if item[0] not in titles and len(kpis) < 6:
                kpis.append(item); titles.add(item[0])
        kpis = kpis[:6]
        gap = 0.008
        kpi_w = (0.97 - 5*gap) / 6
        for i, (title, value, sub, c) in enumerate(kpis):
            x = 0.015 + i*(kpi_w+gap)
            card(x, kpi_y, kpi_w, kpi_h, fc="white", ec=c, radius=0.010, lw=1.0)
            txt(x+kpi_w/2, kpi_y+0.041, title, 7.8, "bold", c, ha="center")
            txt(x+kpi_w/2, kpi_y+0.023, value, 12.6, "bold", c, ha="center")
            txt(x+kpi_w/2, kpi_y+0.007, sub, 6.6, "normal", MUTED, ha="center")


        # FAIXA DE CONTEXTO

        card(0.015, 0.758, 0.97, 0.039, fc="#FFFDF9", ec="#EBDCC5", radius=0.009)
        used_txt = " · ".join(f"V{v}" for v in used) or "—"
        fin = p["financial"]
        fourth_context = (
            ("CUSTOS", f"Fixo: {p['unit_currency']} {fin['fixed_cost']:.0f} | Variável: {p['unit_currency']} {fin['variable_cost']:.0f}", GREEN)
            if p.get("has_cost") else
            ("GARANTIA MATEMÁTICA", p.get("certification", {}).get("short", "Sem certificação conclusiva"), PURPLE)
        )
        context = [
            ("OBJETIVO", p["objective_text"], ORANGE),
            ("JANELAS DE TEMPO", f"{p['sla_pct']:.0f}% atendidas", BLUE),
            ("FROTA ATIVA", used_txt, ORANGE),
            fourth_context,
        ]
        cx = [0.035, 0.285, 0.535, 0.785]
        for (title, value, c), x in zip(context, cx):
            ax.add_patch(Circle((x, 0.778), 0.0068, facecolor=c, edgecolor="none"))
            txt(x+0.014, 0.784, title, 7.4, "bold", NAVY)
            txt(x+0.014, 0.769, value, 7.1, "normal", TEXT)


        # LAYOUT PRINCIPAL

        left_x, left_w = 0.015, 0.735
        right_x, right_w = 0.765, 0.220
        body_top, body_bottom = 0.744, 0.090

        # Título do plano
        card(left_x, body_top-0.035, left_w, 0.032, fc=NAVY, ec=NAVY, radius=0.007)
        txt(left_x+0.014, body_top-0.019,
            f"PLANO DE VIAGENS ({len(used)} VEÍCULOS · {len(trips)} VIAGENS)",
            8.6, "bold", "white")

        # Legenda de cores por veículo no cabeçalho
        lx = left_x + left_w - 0.205
        for v in used[:5]:
            c = vehicle_color[v]
            ax.add_patch(Circle((lx, body_top-0.019), 0.006, facecolor=c, edgecolor="none"))
            txt(lx+0.010, body_top-0.019, f"V{v}", 7.0, "bold", "white")
            lx += 0.050


        # Cabeçalho de colunas

        table_top = body_top - 0.048
        header_h = 0.024
        card(left_x, table_top-header_h, left_w, header_h, fc="#F7FAFD", ec=GRID, radius=0.004)

        # colunas: Veículo, Viagem, Carga, Distância, Custo, Rota
        cols = {
            "vehicle": (left_x+0.012, 0.065),
            "trip":    (left_x+0.078, 0.050),
            "load":    (left_x+0.132, 0.128),
            "dist":    (left_x+0.267, 0.075),
            "cost":    (left_x+0.347, 0.090),
            "route":   (left_x+0.445, left_w-0.455),
        }
        txt(cols["vehicle"][0], table_top-header_h/2, "Veículo", 6.8, "bold", NAVY)
        txt(cols["trip"][0], table_top-header_h/2, "Viagem", 6.8, "bold", NAVY)
        txt(cols["load"][0], table_top-header_h/2, f"Carga ({p['unit_load']})", 6.8, "bold", NAVY)
        txt(cols["dist"][0], table_top-header_h/2, "Distância", 6.8, "bold", NAVY)
        txt(cols["cost"][0], table_top-header_h/2, "Custo", 6.8, "bold", NAVY)
        txt(cols["route"][0], table_top-header_h/2, "Rota (Depósito → Clientes → Depósito)", 6.8, "bold", NAVY)


        # Linhas da tabela — altura adaptativa

        table_bottom = body_bottom + 0.005
        usable = (table_top-header_h) - table_bottom
        group_gap = 0.0035
        group_breaks = max(0, len(used)-1)
        row_h = (usable - group_breaks*group_gap) / n
        row_h = min(row_h, 0.0305)

        current_y = table_top - header_h
        prev_v = None

        for t in trips:
            v = int(t.get("vehicle_id", 0))
            trip_no = int(t.get("trip_number", 0))
            c = vehicle_color.get(v, BLUE)

            if prev_v is not None and v != prev_v:
                current_y -= group_gap

            y = current_y - row_h
            card(left_x, y, left_w, row_h, fc="white", ec=GRID, radius=0.003, lw=0.7)

            # veículo — faixa colorida
            veh_w = 0.060
            card(left_x+0.004, y+0.003, veh_w, row_h-0.006, fc=c, ec=c, radius=0.004)
            txt(left_x+0.004+veh_w/2, y+row_h*0.62, f"V{v}", 7.5, "bold", "white", ha="center")
            txt(left_x+0.004+veh_w/2, y+row_h*0.28, f"{self._safe_float(t.get('time_min',0)):.1f} min", 5.6, "bold", "white", ha="center")

            # viagem
            txt(left_x+0.094, y+row_h/2, f"T{trip_no}", 7.3, "bold", NAVY, ha="center")

            # carga + barra
            load = self._safe_float(t.get("load", 0))
            cap = max(0.0, self._safe_float(t.get("capacity", 0)))
            util = self._safe_float(t.get("capacity_utilization_pct", 0))
            txt(left_x+0.132, y+row_h*0.64,
                f"{load:.0f}/{cap:.0f} ({util:.0f}%)", 6.5, "bold", TEXT)
            bx, by, bw, bh = left_x+0.132, y+row_h*0.22, 0.085, 0.0055
            ax.add_patch(Rectangle((bx, by), bw, bh, facecolor="#E5EBF1", edgecolor="none"))
            ax.add_patch(Rectangle((bx, by), bw*min(1,util/100), bh, facecolor=c, edgecolor="none"))

            # distância / custo
            txt(left_x+0.300, y+row_h/2, f"{self._safe_float(t.get('distance_km',0)):.2f} km", 6.5, "bold", TEXT, ha="center")
            txt(left_x+0.392, y+row_h/2, f"{p['unit_currency']} {self._safe_float(t.get('cost',0)):.2f}", 6.5, "bold", TEXT, ha="center")

            # rota
            path = list(t.get("path", []) or [])
            if not path or int(path[0]) != 0:
                path = [0] + path
            if int(path[-1]) != 0:
                path = path + [0]

            flow_x0 = left_x + 0.455
            flow_x1 = left_x + left_w - 0.015
            positions = np.linspace(flow_x0, flow_x1, max(2, len(path)))
            fy = y + row_h/2

            for j in range(len(positions)-1):
                ax.annotate(
                    "",
                    xy=(positions[j+1]-0.007, fy),
                    xytext=(positions[j]+0.007, fy),
                    arrowprops=dict(arrowstyle="->", color="#54708C", lw=0.8)
                )

            for px, node in zip(positions, path):
                node = int(node)
                if node == 0:
                    depot(px, fy, NAVY, scale=0.76)
                else:
                    ax.add_patch(Circle(
                        (px, fy), 0.0080,
                        facecolor=c, edgecolor="white", linewidth=0.7
                    ))
                    txt(px, fy, str(node), 5.2, "bold", "white", ha="center")

            current_y = y
            prev_v = v


        # RESUMO OPERACIONAL

        card(right_x, 0.570, right_w, 0.126, fc="white", ec=GRID, radius=0.008)
        card(right_x, 0.671, right_w, 0.025, fc=NAVY, ec=NAVY, radius=0.007)
        txt(right_x+0.014, 0.6835, "RESUMO OPERACIONAL", 8.1, "bold", "white")

        preserved = max(0, p['total_vehicles'] - len(used))
        bullets = [
            (f"{len(used)} " + ("veículo utilizado" if len(used) == 1 else "veículos utilizados"), f"{preserved} " + ("veículo ocioso" if preserved == 1 else "veículos ociosos")),
            (f"{len(trips)} viagens realizadas", ""),
            (f"{p['total_load']:.0f} {p['unit_load']} entregues", ""),
            (f"Janelas de tempo: {p['sla_pct']:.0f}% atendidas", ""),
            (f"Utilização média por viagem: {p['avg_util']:.0f}%", ""),
        ]
        yy = 0.654
        for main, sub in bullets:
            ax.add_patch(Circle((right_x+0.016, yy), 0.0052, facecolor=GREEN, edgecolor="none"))
            txt(right_x+0.030, yy, main, 6.2, "normal", TEXT)
            if sub:
                txt(right_x+0.030, yy-0.011, sub, 5.3, "normal", MUTED)
                yy -= 0.027
            else:
                yy -= 0.020

        # Caminhão no canto, fora da área de texto
        self._draw_truck(ax, right_x+right_w-0.048, 0.581, scale=0.22, color=BLUE)


        # UTILIZAÇÃO DA CAPACIDADE

        card(right_x, 0.390, right_w, 0.166, fc="white", ec=GRID, radius=0.008)
        card(right_x, 0.531, right_w, 0.025, fc=NAVY, ec=NAVY, radius=0.007)
        txt(right_x+0.014, 0.5435, "UTILIZAÇÃO DA CAPACIDADE", 7.8, "bold", "white")

        # Mostra até 6 viagens. O plano principal já contém todas as viagens;
        # aqui o objetivo é oferecer uma leitura-resumo sem poluir o painel.
        shown = trips[:6]
        y0 = 0.510
        for i, t in enumerate(shown):
            v = int(t.get("vehicle_id", 0))
            c = vehicle_color.get(v, BLUE)
            util = self._safe_float(t.get("capacity_utilization_pct", 0))
            yy = y0 - i*0.0185
            txt(right_x+0.014, yy, f"V{v}-T{t['trip_number']}", 5.7, "normal", TEXT)
            bx = right_x+0.075
            ax.add_patch(Rectangle((bx, yy-0.004), 0.078, 0.007, facecolor="#E5EAF0", edgecolor="none"))
            ax.add_patch(Rectangle((bx, yy-0.004), 0.078*min(1,util/100), 0.007, facecolor=c, edgecolor="none"))
            txt(right_x+0.164, yy, f"{util:.0f}%", 5.7, "bold", TEXT)

        if len(trips) > len(shown):
            txt(right_x+0.014, 0.405, f"+ {len(trips)-len(shown)} " + ("viagem detalhada no plano" if len(trips)-len(shown) == 1 else "viagens detalhadas no plano"), 5.4, "normal", MUTED)

        # Média separada e claramente delimitada — sem qualquer sobreposição.
        avg_y = 0.397
        txt(right_x+0.014, avg_y-0.016, "MÉDIA POR VIAGEM", 5.6, "bold", MUTED)
        txt(right_x+0.014, avg_y-0.032, f"{p['avg_util']:.0f}%", 9.4, "bold", NAVY)
        ax.add_patch(Rectangle((right_x+0.070, avg_y-0.036), 0.104, 0.008, facecolor="#E5EAF0", edgecolor="none"))
        ax.add_patch(Rectangle((right_x+0.070, avg_y-0.036), 0.104*min(1,p['avg_util']/100), 0.008, facecolor=GREEN, edgecolor="none"))


        # DESTAQUES

        card(right_x, 0.258, right_w, 0.112, fc="white", ec=GRID, radius=0.008)
        card(right_x, 0.345, right_w, 0.025, fc=NAVY, ec=NAVY, radius=0.007)
        txt(right_x+0.014, 0.3575, "DESTAQUES OPERACIONAIS", 7.7, "bold", "white")

        best = max(trips, key=lambda z: self._safe_float(z.get("capacity_utilization_pct",0)))
        worst = min(trips, key=lambda z: self._safe_float(z.get("capacity_utilization_pct",0)))
        longest = max(trips, key=lambda z: self._safe_float(z.get("distance_km",0)))
        highlights = [
            ("MAIOR OCUPAÇÃO", f"V{best['vehicle_id']}-T{best['trip_number']} · {best['capacity_utilization_pct']:.0f}%", GOLD),
            ("MENOR OCUPAÇÃO", f"V{worst['vehicle_id']}-T{worst['trip_number']} · {worst['capacity_utilization_pct']:.0f}%", RED),
            ("MAIOR PERCURSO", f"V{longest['vehicle_id']}-T{longest['trip_number']} · {longest['distance_km']:.2f} km", TEAL),
        ]
        yy = 0.326
        for title, value, c in highlights:
            ax.add_patch(Circle((right_x+0.017, yy), 0.006, facecolor=c, edgecolor="none"))
            txt(right_x+0.032, yy+0.005, title, 5.4, "bold", MUTED)
            txt(right_x+0.032, yy-0.008, value, 6.1, "bold", TEXT)
            yy -= 0.028


        # COMPOSIÇÃO DO CUSTO

        card(right_x, 0.105, right_w, 0.136, fc="white", ec=GRID, radius=0.008)
        card(right_x, 0.216, right_w, 0.025, fc=NAVY, ec=NAVY, radius=0.007)
        section_title = "COMPOSIÇÃO DO CUSTO" if p.get("has_cost") else "GARANTIA MATEMÁTICA"
        txt(right_x+0.014, 0.2285, section_title, 7.7, "bold", "white")

        if p.get("has_cost"):
            fcost = max(0.0, fin['fixed_cost'])
            vcost = max(0.0, fin['variable_cost'])
            total = max(1e-9, fcost + vcost)
            center = (right_x+0.055, 0.160)
            rr = 0.030
            fixed_pct = fcost/total
            ax.add_patch(Wedge(center, rr, 90, 90-360*fixed_pct, facecolor=BLUE, edgecolor="white", linewidth=0.6))
            ax.add_patch(Wedge(center, rr, 90-360*fixed_pct, 90-360, facecolor=GREEN, edgecolor="white", linewidth=0.6))
            ax.add_patch(Circle(center, rr*0.58, facecolor="white", edgecolor="none"))
            tx = right_x+0.105
            txt(tx, 0.178, "Custo fixo", 5.8, "bold", BLUE)
            txt(right_x+right_w-0.015, 0.178, f"{p['unit_currency']} {fcost:.2f}", 5.8, "bold", TEXT, ha="right")
            txt(tx, 0.157, "Custo variável", 5.8, "bold", GREEN)
            txt(right_x+right_w-0.015, 0.157, f"{p['unit_currency']} {vcost:.2f}", 5.8, "bold", TEXT, ha="right")
            txt(tx, 0.131, "TOTAL", 6.3, "bold", NAVY)
            txt(right_x+right_w-0.015, 0.131, f"{p['unit_currency']} {fin['total_cost']:.2f}", 6.7, "bold", NAVY, ha="right")
        else:
            cert = p.get("certification", {}) or {}
            txt(right_x+0.014, 0.188, cert.get("short", "Sem certificação matemática conclusiva"), 6.3, "bold", NAVY)
            gap_pct = cert.get("gap_pct")
            if gap_pct is not None and cert.get("level") in {"B", "C"}:
                txt(right_x+0.014, 0.160, f"Gap de certificação: {gap_pct:.2f}%", 6.2, "bold", PURPLE)
                txt(right_x+0.014, 0.137, "Mede incerteza de prova, não perda real.", 5.5, "normal", MUTED)
            else:
                txt(right_x+0.014, 0.154, "Consulte o relatório executivo para detalhes.", 5.6, "normal", MUTED)


        # RODAPÉ

        card(0.015, 0.020, 0.685, 0.054, fc="#F6FBEF", ec="#CCE5B5", radius=0.010)
        ax.add_patch(Circle((0.045, 0.049), 0.011, facecolor="#FFD54F", edgecolor="#E0A800", linewidth=1.0))
        ax.add_patch(Rectangle((0.041, 0.029), 0.008, 0.008, facecolor="#FFB300", edgecolor="none"))
        txt(0.070, 0.059, "RECOMENDAÇÃO PRINCIPAL", 7.7, "bold", "#33691E")
        rec = str(p['recommendation']).replace("💡", "").strip()
        if len(rec) > 170:
            rec = rec[:167] + "..."
        txt(0.070, 0.038, rec, 6.4, "normal", TEXT)

        card(0.715, 0.020, 0.270, 0.054, fc="white", ec="#DCE7D5", radius=0.010)
        txt(0.730, 0.059, "LEGENDA DA ROTA", 7.3, "bold", NAVY)
        depot(0.750, 0.038, NAVY, scale=0.64)
        txt(0.767, 0.038, "Depósito", 6.0, "normal", TEXT)
        ax.add_patch(Circle((0.835, 0.038), 0.0062, facecolor=BLUE, edgecolor="none"))
        txt(0.847, 0.038, "Cliente", 6.0, "normal", TEXT)
        ax.annotate("", xy=(0.945,0.038), xytext=(0.895,0.038), arrowprops=dict(arrowstyle="->", color=NAVY, lw=0.9))
        txt(0.950, 0.038, "Rota", 6.0, "normal", TEXT)

        if salvar_png:
            fig.savefig(
                filename,
                dpi=190,
                bbox_inches="tight",
                facecolor=BG
            )
            self.shared_memory.data["logistics_infographic_file"] = filename

        plt.show()

        self.shared_memory.data["logistics_flow_diagram_generated"] = True
        self.shared_memory.data["logistics_infographic_version"] = "DIAGNOSTIC_EXECUTIVE_DASHBOARD_V2_3_CONTINUITY"
        self.shared_memory.data["logistics_infographic_figsize"] = (figsize_width, figsize_height)
        return fig

    def exibir_mapa_maplibre_deck(self, *args, **kwargs):
        """Interface de compatibilidade com o Orchestrator para exibir o infográfico logístico."""
        return self.exibir_diagrama_fluxo_logistico()

    def executar_diagnostico(self, gerar_painel=True, gerar_mapa=True, verbose=False):
        print("[DiagnosticAgent DIAGNOSTIC_EXECUTIVE_DASHBOARD_V2_3_CONTINUITY]")
        diagnostic = self.analisar_resultado_otimizacao()

        try:
            self._atualizar_tabelas_bi_sem_exibir()
        except Exception as e:
            diagnostic.setdefault("visual_warnings", []).append(f"BI interno: {e}")

        if gerar_mapa:
            try:
                self.exibir_diagrama_fluxo_logistico(modo_tamanho="auto")
            except Exception as e:
                diagnostic.setdefault("visual_warnings", []).append(f"Infográfico logístico: {e}")

        self.shared_memory.data["diagnostic_result"] = diagnostic

        if verbose:
            print(diagnostic.get("resumo", ""))
            for a in diagnostic.get("alertas", []):
                print(f"⚠️ {a}")
            for r in diagnostic.get("recomendacoes", []):
                print(f"💡 {r}")

        return diagnostic


    def analisar_continuidade(self, baseline_solution, continuation_solution, comparison=None):
        """Gera diagnóstico comparativo do Relatório 2 sem alterar nenhuma rota."""
        comparison = copy.deepcopy(comparison or self.shared_memory.data.get("continuity_comparison") or {})
        result = {
            "resumo": "Comparação pós-solução entre o Relatório 1 e a rodada opcional de continuidade.",
            "baseline_preservado": True,
            "solution_improved": bool(comparison.get("solution_improved", False)),
            "certification_improved": bool(comparison.get("certification_improved", False)),
            "before": copy.deepcopy(comparison.get("before", {})),
            "after": copy.deepcopy(comparison.get("after", {})),
            "recomendacoes": [],
            "alertas": [],
        }
        if result["solution_improved"]:
            result["recomendacoes"].append(
                "A solução da continuidade somente deve substituir operacionalmente a anterior porque foi novamente auditada."
            )
        if result["certification_improved"]:
            result["recomendacoes"].append(
                "A certificação foi fortalecida; o ganho representa redução da incerteza matemática da prova."
            )
        if not result["solution_improved"] and not result["certification_improved"]:
            result["recomendacoes"].append(
                "O processamento adicional não superou o baseline; manter o plano do Relatório 1."
            )
        self.shared_memory.data["continuity_diagnostic"] = copy.deepcopy(result)
        return result

    VERSION = "DIAGNOSTIC_EXECUTIVE_DASHBOARD_V2_3_CONTINUITY"
````

### **INITIALSOLUTIONREFINERAGENT — REFINAMENTO DA SOLUÇÃO INICIAL**

Antes de entregar o plano inicial, este agente tenta pequenos ajustes com um tempo curto. O objetivo é aproveitar melhorias rápidas antes de apresentar a primeira resposta.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class InitialSolutionRefinerAgent:
    """Acabamento do núcleo inicial; estado próprio, sem iniciar continuidade."""
    def __init__(self, memory):
        self.memory = memory

    def refine(self, solution, budget_seconds=60., runner=None):
        import tempfile
        budget_seconds=float(budget_seconds)
        if not _pd_math.isfinite(budget_seconds) or budget_seconds<0:
            raise ValueError('Orçamento inicial deve ser finito e não negativo.')
        started=_pd_time.monotonic();deadline=started+budget_seconds
        inputs=_pd_memory_inputs(self.memory);scope=_ct_scope(inputs)
        audit=_pd_audit(inputs,solution)
        if not audit.get('feasible'):raise ValueError('Refinamento exige incumbent auditado.')
        best=audit['normalized_solution'];before=float(best['canonical_objective_score'])
        search={};records=[];run=runner or _ct_run_process
        # State is deliberately separate from continuity checkpoints.
        with tempfile.TemporaryDirectory(prefix='initial_refinement_') as td:
            pending=1.
            for index,(stage,share) in enumerate([('MELHORIA',.35),('EXATO_LOCAL',.45),('MELHORIA',.20)]):
                seconds=max(0.,deadline-_pd_time.monotonic())*share/pending;pending-=share
                if seconds<3:continue
                folder=_PDPath(td)/str(index);inputs['search_state_v58']=_pd_copy.deepcopy(search)
                cfg=dict(WALL_STAGE_SECONDS=seconds,TEMPO_CONSTRUCAO_S=seconds,TEMPO_LP_S=seconds,
                    TEMPO_SOLVER_OBJETIVO_S=seconds,TEMPO_SOLVER_K_S=seconds,MARGEM_FINALIZACAO_S=1,
                    THREADS_SCIP=1,INTERVALO_IMPRESSAO_S=15)
                try:r=run(stage,inputs,_pd_copy.deepcopy(best),cfg,folder)
                except Exception as exc:r=dict(error=str(exc),termination_reason='STAGE_LAUNCH_ERROR')
                if r.get('fingerprint')==scope:
                    recovered=_pd_read(folder/'search_state.json')
                    if isinstance(recovered,dict):search=recovered
                    candidate=_pd_read(folder/'candidate.json')
                    check=_pd_audit(inputs,candidate) if candidate else {}
                    if check.get('feasible') and check['recomputed']['canonical_objective_score']<best['canonical_objective_score']:
                        best=check['normalized_solution']
                records.append(dict(stage=stage,result=r))
                if r.get('parent_stop_reason')=='USER_INTERRUPTED':break
        audit=_pd_audit(inputs,best)
        if not audit.get('feasible'):raise RuntimeError('Refinamento rejeitado pela auditoria final.')
        report=dict(agent=type(self).__name__,fingerprint=scope,budget_seconds=budget_seconds,
            elapsed_seconds=_pd_time.monotonic()-started,ub_before=before,
            ub_after=best['canonical_objective_score'],ub_gain=before-best['canonical_objective_score'],stages=records,
            final_audit_passed=True, final_audit=_pd_copy.deepcopy(audit))
        self.memory.data['initial_refinement_state']=dict(report=report,search_state=search)
        print(f"Refinamento inicial: UB {before:.8f} → {best['canonical_objective_score']:.8f}; {report['elapsed_seconds']:.1f}s adicionais.")
        return best,audit,report
````

## **AGENTES DA CONTINUIDADE**

Esta equipe só entra na rodada extra quando o usuário autoriza. Ela começa com um plano completo aprovado. Há dois trabalhos diferentes: **melhorar a solução, diminuindo UB**, e **fortalecer a garantia, aumentando LB**. A investigação de menos veículos é uma possibilidade de busca, sem bloquear as outras melhorias.


### **CONTINUITYOPTIMIZATIONAGENT — ORGANIZAÇÃO DA CONTINUIDADE**

O usuário pode dizer: “temos mais alguns minutos; tentem melhorar”. Este agente participa da organização dessa rodada, sempre partindo do que já foi aprovado.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class ContinuityOptimizationAgent:
    """Coordena a continuidade a partir do melhor plano auditado.

    A busca procura reduzir o objetivo; a certificação procura fortalecer o limite
    global. As tarefas respeitam seus prazos e compartilham os resultados válidos.
    O Auditor verifica a factibilidade, o Certificador classifica a garantia e o
    Orquestrador administra recursos e decisões. Evidência local não é prova global."""

    VERSION = "CONTINUITY_OPTIMIZATION_AGENT_V10_PROCESS_PARALLEL_ABC"

    def __init__(self, shared_memory):
        self.shared_memory = shared_memory

    @staticmethod
    def _safe_float(v, default=None):
        try:
            return default if v is None else float(v)
        except Exception:
            return default

    @staticmethod
    def _safe_int(v, default=0):
        try:
            return default if v is None else int(v)
        except Exception:
            return default

    def _translated_dict(self):
        """Retorna translated_json como dict, aceitando dict ou string JSON."""
        import copy
        import json
        raw = (getattr(self.shared_memory, "data", {}) or {}).get("translated_json", {}) or {}
        if isinstance(raw, dict):
            return copy.deepcopy(raw)
        if isinstance(raw, str):
            try:
                obj = json.loads(raw)
                return obj if isinstance(obj, dict) else {}
            except Exception:
                return {}
        return {}

    @staticmethod
    def _recursive_find(obj, keys):
        """Busca tolerante por chaves em estruturas dict/list aninhadas."""
        wanted = {str(k).strip().lower() for k in keys}
        queue = [obj]
        seen = set()
        while queue:
            cur = queue.pop(0)
            oid = id(cur)
            if oid in seen:
                continue
            seen.add(oid)
            if isinstance(cur, dict):
                for k, v in cur.items():
                    if str(k).strip().lower() in wanted and v is not None:
                        return v
                queue.extend(cur.values())
            elif isinstance(cur, (list, tuple)):
                queue.extend(cur)
        return None

    @staticmethod
    def _as_bool(v, default=None):
        if isinstance(v, bool):
            return v
        if isinstance(v, (int, float)):
            return bool(v)
        if isinstance(v, str):
            t = v.strip().upper()
            if t in {"TRUE", "T", "YES", "Y", "SIM", "S", "1"}:
                return True
            if t in {"FALSE", "F", "NO", "N", "NAO", "NÃO", "2", "0"}:
                return False
        return default

    def _objective_user_weights(self, ctx, active):
        """
        Recupera os pesos semânticos escolhidos pelo usuário (ex. 0.4/0.6).
        Nunca confunde coefficients normalizados com weights.
        """
        raw = None
        for key in ("weights", "objective_weights", "user_weights", "multiobjective_weights"):
            candidate = ctx.get(key) if isinstance(ctx, dict) else None
            if isinstance(candidate, dict) and candidate:
                raw = candidate
                break
        if raw is None:
            translated = self._translated_dict()
            candidate = self._recursive_find(translated, ("weights", "objective_weights", "user_weights", "multiobjective_weights"))
            if isinstance(candidate, dict):
                raw = candidate
        weights = {}
        if isinstance(raw, dict):
            for obj in active:
                val = None
                for k, v in raw.items():
                    if str(k).upper() == obj:
                        val = self._safe_float(v, None)
                        break
                if val is not None:
                    weights[obj] = float(val / 100.0 if val > 1.0 + 1e-12 else val)
        # Fallback seguro: reconstruir peso = coefficient * scale, quando scales existe.
        if len(weights) < len(active):
            coeff = (ctx or {}).get("coefficients", {}) or {}
            scales = (ctx or {}).get("scales", {}) or {}
            for obj in active:
                if obj in weights:
                    continue
                c = self._safe_float(coeff.get(obj), None)
                sc = self._safe_float(scales.get(obj), None)
                if c is not None and sc is not None and sc > 0:
                    weights[obj] = max(0.0, float(c) * float(sc))
        total = sum(max(0.0, float(weights.get(o, 0.0))) for o in active)
        if total > 0:
            weights = {o: max(0.0, float(weights.get(o, 0.0))) / total for o in active}
        return weights

    def _operational_flags(self):
        """Lê flags estruturais relevantes antes dos Tracks A/B."""
        data = getattr(self.shared_memory, "data", {}) or {}
        translated = self._translated_dict()
        def first(keys):
            v = self._recursive_find(translated, keys)
            if v is None:
                v = self._recursive_find(data, keys)
            return v
        must = self._as_bool(first(("must_use_all_vehicles", "must_use_all", "use_all_vehicles")), False)
        multi = self._as_bool(first(("multi_trip", "allow_multi_trip", "multiple_trips_per_vehicle", "multi_trip_allowed")), None)
        return {"must_use_all_vehicles": bool(must), "multi_trip": multi}

    @staticmethod
    def _successful(solution):
        return isinstance(solution, dict) and str(solution.get("status", "")).upper() == "SUCCESS"

    def _metrics(self, solution):
        s = solution or {}
        sm = s.get("summary", {}) or {}
        m = s.get("metrics", {}) or {}
        raw_routes = s.get("routes", []) or []

        if isinstance(raw_routes, dict):
            routes = []
            for value in raw_routes.values():
                if isinstance(value, dict):
                    routes.append(value)
                elif isinstance(value, (list, tuple)):
                    routes.extend(x for x in value if isinstance(x, dict))
        elif isinstance(raw_routes, (list, tuple)):
            routes = [r for r in raw_routes if isinstance(r, dict)]
        else:
            routes = []

        routes = [r for r in routes if len(r.get("path", []) or []) > 2]
        vehicles = {
            self._safe_int(r.get("physical_vehicle_id", r.get("vehicle_id", 0)), 0)
            for r in routes
        }

        distance_fallback = sum(self._safe_float(r.get("distance_km"), 0.0) or 0.0 for r in routes)
        time_fallback = sum(
            self._safe_float(
                r.get("total_route_time_minutes", r.get("route_time_minutes")),
                0.0,
            ) or 0.0
            for r in routes
        )

        return {
            "DISTANCE": self._safe_float(
                sm.get("total_distance_km", m.get("total_distance")),
                distance_fallback,
            ) or 0.0,
            "TIME": self._safe_float(
                sm.get("total_time_minutes", m.get("total_time")),
                time_fallback,
            ) or 0.0,
            "COST": self._safe_float(
                sm.get("total_cost", s.get("total_cost", m.get("real_cost"))),
                0.0,
            ) or 0.0,
            "VEHICLES": self._safe_float(
                sm.get("num_vehicles_used", m.get("used_vehicles")),
                len([v for v in vehicles if v > 0]),
            ) or 0.0,
        }

    def _score(self, solution):
        ctx = self.shared_memory.data.get("objective_context", {}) or {}
        coeff = ctx.get("coefficients", {}) or {}
        metrics = self._metrics(solution)
        active = ctx.get("active_objectives", []) or []
        return float(
            sum(
                float(coeff.get(o, 0.0) or 0.0) * float(metrics.get(o, 0.0) or 0.0)
                for o in active
            )
        )

    def _client_count(self):
        data = self.shared_memory.data

        df = data.get("df_clientes")
        try:
            if df is not None and len(df) > 0:
                cols = {str(c).strip().lower(): c for c in getattr(df, "columns", [])}
                id_col = None
                for key in ("id", "id_cliente", "cliente", "node_id"):
                    if key in cols:
                        id_col = cols[key]
                        break
                if id_col is not None:
                    vals = []
                    for v in df[id_col].tolist():
                        try:
                            vals.append(int(v))
                        except Exception:
                            pass
                    non_depot = {v for v in vals if v != 0}
                    if non_depot:
                        return len(non_depot)

                # Fallback conservador: se houver linha do depósito, normalmente é +1.
                return max(0, len(df) - 1)
        except Exception:
            pass

        translated = self._translated_dict()

        for key in ("clients", "clientes"):
            obj = translated.get(key)
            if isinstance(obj, dict):
                ids = []
                for k in obj.keys():
                    try:
                        ids.append(int(k))
                    except Exception:
                        pass
                non_depot = {x for x in ids if x != 0}
                if non_depot:
                    return len(non_depot)
            elif isinstance(obj, (list, tuple)):
                count = 0
                for item in obj:
                    if not isinstance(item, dict):
                        continue
                    ident = item.get("id", item.get("ID", item.get("node_id")))
                    try:
                        if int(ident) != 0:
                            count += 1
                    except Exception:
                        count += 1
                if count:
                    return count

        for key in ("num_clients", "number_of_clients", "n_clients"):
            n = self._safe_int(data.get(key), 0)
            if n > 0:
                return n
        return 0


    # OBJECTIVE-DRIVEN IMPROVEMENT PLANNING

    def _active_vehicle_ids(self, solution):
        """Retorna os veículos físicos efetivamente usados pela solução incumbente."""
        raw_routes = (solution or {}).get("routes", []) or []
        if isinstance(raw_routes, dict):
            tmp = []
            for value in raw_routes.values():
                if isinstance(value, dict):
                    tmp.append(value)
                elif isinstance(value, (list, tuple)):
                    tmp.extend(x for x in value if isinstance(x, dict))
            raw_routes = tmp

        ids = set()
        for r in raw_routes if isinstance(raw_routes, (list, tuple)) else []:
            if not isinstance(r, dict) or len(r.get("path", []) or []) <= 2:
                continue
            vid = self._safe_int(
                r.get("physical_vehicle_id", r.get("vehicle_id", 0)), 0
            )
            if vid > 0:
                ids.add(vid)
        return sorted(ids)

    def _flatten_solution_routes(self, solution):
        """Normaliza as viagens da incumbente em uma lista de dicts não vazios."""
        raw = (solution or {}).get("routes", []) or []
        if isinstance(raw, dict):
            out = []
            for value in raw.values():
                if isinstance(value, dict):
                    out.append(value)
                elif isinstance(value, (list, tuple)):
                    out.extend(x for x in value if isinstance(x, dict))
            raw = out
        if not isinstance(raw, (list, tuple)):
            return []
        return [r for r in raw if isinstance(r, dict) and len(r.get("path", []) or []) > 2]

    def _node_coordinate_map(self):
        """Mapa id -> (lat, lon), usado somente para diagnóstico heurístico pré-Track."""
        data = getattr(self.shared_memory, "data", {}) or {}
        df = data.get("df_clientes")
        coords = {}
        if df is not None:
            try:
                cols = {str(c).strip().lower(): c for c in df.columns}
                idc = next((cols[k] for k in ("id", "id_cliente", "cliente", "node_id") if k in cols), None)
                latc = next((cols[k] for k in ("latitude", "lat") if k in cols), None)
                lonc = next((cols[k] for k in ("longitude", "lon", "lng") if k in cols), None)
                if idc is not None and latc is not None and lonc is not None:
                    for _, row in df.iterrows():
                        try:
                            coords[int(row[idc])] = (float(row[latc]), float(row[lonc]))
                        except Exception:
                            pass
            except Exception:
                pass
        return coords

    def _best_2opt_hint(self, path, coords):
        """Estima o melhor ganho geométrico 2-opt; NÃO afirma factibilidade temporal."""
        if not isinstance(path, (list, tuple)) or len(path) < 5 or not coords:
            return None
        try:
            nodes = [int(x) for x in path]
        except Exception:
            return None
        if any(x not in coords for x in nodes):
            return None
        best = None
        # Troca arestas (a,b),(c,d) por (a,c),(b,d). O GA/Auditor ainda deve
        # verificar TW, jornada, capacidade e todas as demais restrições.
        for i in range(0, len(nodes) - 3):
            for k in range(i + 2, len(nodes) - 1):
                a,b,c,d = nodes[i],nodes[i+1],nodes[k],nodes[k+1]
                ab = self._haversine_km(*coords[a], *coords[b])
                cd = self._haversine_km(*coords[c], *coords[d])
                ac = self._haversine_km(*coords[a], *coords[c])
                bd = self._haversine_km(*coords[b], *coords[d])
                gain = (ab + cd) - (ac + bd)
                if gain > 1e-8 and (best is None or gain > best["estimated_distance_gain_km"]):
                    best = {
                        "i": int(i + 1), "k": int(k),
                        "estimated_distance_gain_km": float(gain),
                        "feasibility_status": "UNVERIFIED_REQUIRES_SOLVER_AND_AUDITOR",
                    }
        return best

    def _structural_incumbent_analysis(self, baseline_solution):
        """
         diagnóstico estrutural ANTES dos Tracks A/B.

        Não altera a solução e não declara movimentos factíveis. Identifica onde
        vale concentrar a busca: redução de frota, sequência intra-viagem,
        relocate/swap entre viagens, consolidação de viagens e realocação econômica.
        """
        import copy
        routes = self._flatten_solution_routes(baseline_solution)
        coords = self._node_coordinate_map()
        by_vehicle = {}
        trip_rows = []
        two_opt = []
        for idx, r in enumerate(routes):
            vid = self._safe_int(r.get("physical_vehicle_id", r.get("vehicle_id", 0)), 0)
            tid = r.get("trip_id", r.get("trip_number", idx + 1))
            path = list(r.get("path", []) or [])
            dist = self._safe_float(r.get("distance_km"), 0.0) or 0.0
            tm = self._safe_float(r.get("total_route_time_minutes", r.get("route_time_minutes")), 0.0) or 0.0
            load = self._safe_float(r.get("load", r.get("total_load", r.get("transported_load"))), None)
            clients = [x for x in path if self._safe_int(x, 0) != 0]
            row = {
                "vehicle_id": vid, "trip_id": tid, "route_index": idx,
                "clients": copy.deepcopy(clients), "client_count": len(clients),
                "distance_km": float(dist), "time_minutes": float(tm), "load": load,
            }
            trip_rows.append(row)
            by_vehicle.setdefault(vid, {"vehicle_id": vid, "trip_count": 0, "distance_km": 0.0, "time_minutes": 0.0, "client_count": 0})
            by_vehicle[vid]["trip_count"] += 1
            by_vehicle[vid]["distance_km"] += float(dist)
            by_vehicle[vid]["time_minutes"] += float(tm)
            by_vehicle[vid]["client_count"] += len(clients)
            hint = self._best_2opt_hint(path, coords)
            if hint:
                hint.update({"vehicle_id": vid, "trip_id": tid, "route_index": idx})
                two_opt.append(hint)

        two_opt.sort(key=lambda x: -x.get("estimated_distance_gain_km", 0.0))
        vehicles = sorted(by_vehicle.values(), key=lambda x: x["vehicle_id"])
        trips_by_distance = sorted(trip_rows, key=lambda x: -x["distance_km"])
        trips_by_clients = sorted(trip_rows, key=lambda x: x["client_count"])

        opportunities = {
            "fleet_reduction": {
                "active_vehicle_ids": [v["vehicle_id"] for v in vehicles if v["vehicle_id"] > 0],
                "candidate_elimination_order": [v["vehicle_id"] for v in sorted(vehicles, key=lambda x: (x["client_count"], x["time_minutes"], x["distance_km"])) if v["vehicle_id"] > 0],
                "status": "HYPOTHESIS_REQUIRES_FEASIBILITY_TEST",
            },
            "intra_trip_sequence": {
                "operator": "2opt",
                "candidate_trips": copy.deepcopy(two_opt[:12]),
                "status": "GEOMETRIC_HINT_REQUIRES_TW_AND_AUDIT_VALIDATION",
            },
            "inter_trip_reallocation": {
                "operators": ["relocate", "swap"],
                "priority_source_trips": copy.deepcopy(trips_by_distance[:12]),
                "status": "HYPOTHESIS_REQUIRES_FEASIBILITY_TEST",
            },
            "trip_consolidation": {
                "operator": "trip_consolidation",
                "small_trip_candidates": copy.deepcopy(trips_by_clients[:12]),
                "vehicles_with_multiple_trips": [copy.deepcopy(v) for v in vehicles if v["trip_count"] > 1],
                "status": "HYPOTHESIS_REQUIRES_FEASIBILITY_TEST",
            },
            "vehicle_cost_reallocation": {
                "active_vehicle_ids": [v["vehicle_id"] for v in vehicles if v["vehicle_id"] > 0],
                "status": "HYPOTHESIS_REQUIRES_COST_AND_FEASIBILITY_TEST",
            },
        }
        return {
            "version": "STRUCTURAL_ANALYSIS_V1",
            "num_routes": len(routes),
            "num_active_vehicles": len([v for v in vehicles if v["vehicle_id"] > 0]),
            "vehicle_summary": vehicles,
            "trip_summary": trip_rows,
            "opportunities": opportunities,
            "safety_note": "Hints heurísticos não são prova de melhora nem de factibilidade; Solver + Auditor continuam soberanos.",
        }

    def _objective_improvement_plan(self, baseline_solution):
        """
        Analisa PRIMEIRO a solução incumbente e o objetivo congelado.

        A função não substitui Solver/Auditor. Ela responde:
          1) o que precisa melhorar;
          2) por quais mecanismos isso pode melhorar;
          3) quais hipóteses devem orientar a busca primal.

        O plano é publicado em shared_memory para que o GA/LNS possa consumi-lo.
        Se uma versão antiga do GA ignorar os hints, a continuidade continua
        compatível e preserva a incumbente.
        """
        import copy

        ctx = self.shared_memory.data.get("objective_context", {}) or {}
        active = [str(x).upper() for x in (ctx.get("active_objectives", []) or [])]
        coeff = ctx.get("coefficients", {}) or {}  # somente para score canônico
        user_weights = self._objective_user_weights(ctx, active)
        operational = self._operational_flags()
        structural = self._structural_incumbent_analysis(baseline_solution)
        metrics = self._metrics(baseline_solution)
        active_vehicles = self._active_vehicle_ids(baseline_solution)
        n_used = int(round(metrics.get("VEHICLES", len(active_vehicles)) or 0))
        if active_vehicles:
            n_used = len(active_vehicles)

        min_fixed, fixed_mode, _ = self._extract_fixed_cost_lower_bound_inputs()

        # Ordena pelo PESO SEMÂNTICO DO USUÁRIO (40/60 etc.), não pelo coefficient
        # normalizado usado no score. Isso evita imprimir 0.1%/0.0% e evita
        # prioridades analíticas erradas.
        weighted = []
        for obj in active:
            w = self._safe_float(user_weights.get(obj), 0.0) or 0.0
            if w > 0:
                weighted.append((float(w), obj))
        weighted.sort(key=lambda x: (-x[0], x[1]))

        strategies = []

        def add_strategy(kind, priority, reason, **kwargs):
            item = {
                "type": str(kind),
                "priority": int(priority),
                "reason": str(reason),
                "status": "TO_TEST",
            }
            item.update(kwargs)
            # Evita duplicação quando vários objetivos pedem o mesmo mecanismo.
            for old in strategies:
                if old.get("type") == item["type"]:
                    old["priority"] = min(old["priority"], item["priority"])
                    reasons = old.setdefault("reasons", [old.get("reason")])
                    if item["reason"] not in reasons:
                        reasons.append(item["reason"])
                    return
            strategies.append(item)

        # COST: decomposição explícita = custo fixo de ativação + custo variável/km.
        if "COST" in active:
            if fixed_mode == "WHEN_USED" and n_used > 1 and not operational.get("must_use_all_vehicles", False):
                target = max(1, n_used - 1)
                add_strategy(
                    "FLEET_REDUCTION",
                    1,
                    "Custo fixo é pago somente quando o veículo é utilizado e must_use_all=False; testar explicitamente n-1 veículos ativos.",
                    current_used_vehicles=n_used,
                    target_max_vehicles=target,
                    analytical_question=f"É factível atender todos os clientes com {target} veículos?",
                    elimination_candidates=(structural.get("opportunities", {}).get("fleet_reduction", {}).get("candidate_elimination_order") or list(active_vehicles)),
                    structural_test="ELIMINATE_ONE_ACTIVE_VEHICLE_AND_REINSERT_ITS_CLIENTS",
                    allow_new_trips_on_remaining_vehicles=bool(operational.get("multi_trip") is not False),
                    failure_semantics="NOT_FOUND_NOT_INFEASIBLE",
                    preserve_incumbent=True,
                )
            add_strategy(
                "ROUTE_DISTANCE_REDUCTION",
                2,
                "Reduzir quilômetros reduz a parcela variável do custo.",
                operators=["2opt", "relocate", "swap", "trip_consolidation"],
                structural_hints=copy.deepcopy({
                    "intra_trip_sequence": structural.get("opportunities", {}).get("intra_trip_sequence", {}),
                    "inter_trip_reallocation": structural.get("opportunities", {}).get("inter_trip_reallocation", {}),
                    "trip_consolidation": structural.get("opportunities", {}).get("trip_consolidation", {}),
                }),
                preserve_vehicle_count=True,
            )
            add_strategy(
                "VEHICLE_COST_REALLOCATION",
                3,
                "Redistribuir viagens para veículos economicamente mais eficientes quando factível.",
                structural_hints=copy.deepcopy(structural.get("opportunities", {}).get("vehicle_cost_reallocation", {})),
                preserve_incumbent=True,
            )

        if "VEHICLES" in active and n_used > 1:
            add_strategy(
                "FLEET_REDUCTION",
                1,
                "O objetivo inclui minimizar veículos; testar consolidação da incumbente em n-1 veículos.",
                current_used_vehicles=n_used,
                target_max_vehicles=max(1, n_used - 1),
                preserve_incumbent=True,
            )

        if "DISTANCE" in active:
            add_strategy(
                "ROUTE_DISTANCE_REDUCTION",
                1,
                "O objetivo inclui distância; intensificar o desenho das viagens a partir das rotas atuais.",
                operators=["2opt", "relocate", "swap", "trip_consolidation"],
                structural_hints=copy.deepcopy({
                    "intra_trip_sequence": structural.get("opportunities", {}).get("intra_trip_sequence", {}),
                    "inter_trip_reallocation": structural.get("opportunities", {}).get("inter_trip_reallocation", {}),
                    "trip_consolidation": structural.get("opportunities", {}).get("trip_consolidation", {}),
                }),
                preserve_vehicle_count=True,
            )

        if "TIME" in active:
            add_strategy(
                "ROUTE_TIME_REDUCTION",
                1,
                "O objetivo inclui tempo; reduzir deslocamento, espera e duração das viagens.",
                operators=["2opt", "relocate", "swap", "trip_consolidation"],
                preserve_vehicle_count=True,
            )
            add_strategy(
                "WORKLOAD_TIME_REALLOCATION",
                2,
                "Redistribuir clientes/viagens quando isso reduzir o tempo canônico sem violar restrições.",
                preserve_incumbent=True,
            )

        if not strategies:
            add_strategy(
                "GENERIC_INCUMBENT_INTENSIFICATION",
                1,
                "Nenhum mecanismo específico pôde ser inferido; intensificar a incumbente sem destruí-la.",
                preserve_incumbent=True,
            )

        strategies.sort(key=lambda x: (x.get("priority", 999), x.get("type", "")))

        plan = {
            "version": self.VERSION,
            "stage": "PRE_DUAL_TRACK_OBJECTIVE_DRIVEN_PLANNING",
            "baseline_score": float(self._score(baseline_solution)),
            "baseline_metrics": copy.deepcopy(metrics),
            "active_objectives": active,
            "objective_weights": copy.deepcopy(user_weights),
            "canonical_coefficients": {o: self._safe_float(coeff.get(o), 0.0) or 0.0 for o in active},
            "operational_flags": copy.deepcopy(operational),
            "structural_incumbent_analysis": copy.deepcopy(structural),
            "master_question": "Como posso tornar esta solução incumbente ainda melhor considerando os objetivos ativos?",
            "weighted_objective_order": [o for _, o in weighted],
            "active_vehicle_ids": active_vehicles,
            "current_used_vehicles": n_used,
            "fixed_cost_mode": fixed_mode,
            "fleet_reduction_question": (
                f"É factível reduzir a frota ativa de {n_used} para {max(1, n_used-1)} veículos?"
                if n_used > 1 and fixed_mode == "WHEN_USED" and not operational.get("must_use_all_vehicles", False)
                else None
            ),
            "minimum_nonnegative_fixed_cost": min_fixed,
            "question_1": "O que exatamente estou tentando melhorar?",
            "question_2": "Quais componentes da solução atual podem produzir essa melhoria?",
            "question_3": "Quais modificações concretas da incumbente devo testar?",
            "strategies": strategies,
            "incumbent_protection": (
                "A incumbente só pode ser substituída por solução SUCCESS, auditada e "
                "estritamente melhor no objetivo canônico."
            ),
        }
        return plan

    def _publish_improvement_plan(self, baseline_solution):
        import copy
        plan = self._objective_improvement_plan(baseline_solution)
        self.shared_memory.data["continuity_improvement_plan"] = copy.deepcopy(plan)
        self.shared_memory.data["continuity_search_directive"] = None
        return plan

    def _strategy_directive(self, strategy, baseline_solution):
        """Converte uma hipótese do plano em hint de busca, sem alterar restrições reais."""
        import copy
        directive = copy.deepcopy(strategy or {})
        directive["seed_from_incumbent"] = True
        directive["incumbent_score"] = float(self._score(baseline_solution))
        directive["incumbent_vehicle_ids"] = self._active_vehicle_ids(baseline_solution)
        directive["do_not_relax_constraints"] = True
        directive["require_audit_before_acceptance"] = True
        plan = self.shared_memory.data.get("continuity_improvement_plan", {}) or {}
        directive["structural_incumbent_analysis"] = copy.deepcopy(plan.get("structural_incumbent_analysis", {}))
        return directive

    def build_context(self, baseline_solution, baseline_certificate, budget_seconds):
        import copy
        import time

        cert = copy.deepcopy(baseline_certificate or {})
        baseline_lb = self._safe_float(cert.get("lower_bound"), None)
        lb_certified = bool(cert.get("lower_bound_certified", False))

        baseline = {
            "solution": copy.deepcopy(baseline_solution),
            "certificate": cert,
            "objective": self._score(baseline_solution),
            "lower_bound": baseline_lb,
            "lower_bound_certified": lb_certified,
            "gap_pct": self._safe_float(cert.get("gap_pct"), None),
            "level": str(cert.get("certification_level", "D") or "D").upper(),
            "metrics": self._metrics(baseline_solution),
        }

        improvement_plan = self._publish_improvement_plan(baseline_solution)

        ctx = {
            "version": self.VERSION,
            "improvement_plan": copy.deepcopy(improvement_plan),
            "baseline": baseline,
            "objective_context": copy.deepcopy(
                self.shared_memory.data.get("objective_context", {}) or {}
            ),
            "budget_seconds": max(60, int(budget_seconds)),
            "started_at_monotonic": time.monotonic(),
            "dual_track": True,
            "parallel": True,
            "history": [],
        }

        self.shared_memory.data["continuity_baseline"] = copy.deepcopy(baseline)
        self.shared_memory.data["continuity_context"] = copy.deepcopy(ctx)
        self.shared_memory.data["continuity_seed_solution"] = copy.deepcopy(baseline_solution)
        self.shared_memory.data["continuity_baseline_solution"] = copy.deepcopy(baseline_solution)

        self.shared_memory.data["continuity_baseline_valid_lower_bound"] = (
            baseline_lb if lb_certified and baseline_lb is not None else None
        )
        self.shared_memory.data["continuity_best_valid_lower_bound"] = (
            baseline_lb if lb_certified and baseline_lb is not None else None
        )
        self.shared_memory.data["continuity_bound_candidates"] = []

        return ctx

    def _audit_candidate(self, audit_callback, candidate, label):
        try:
            result = audit_callback(candidate, label)
            if isinstance(result, tuple) and len(result) >= 2:
                diag, rec = result[0], result[1]
            elif isinstance(result, dict):
                diag, rec = None, result
            else:
                diag, rec = None, {"solucao_aprovada": False}
            if not isinstance(rec, dict):
                rec = {"solucao_aprovada": False}
            return diag, rec
        except Exception as exc:
            return None, {
                "solucao_aprovada": False,
                "continuity_audit_error": f"{type(exc).__name__}: {exc}",
            }

    def _translated_with_continuity_payload(self, translated_json, directive, seed_solution):
        """Inclui a diretiva de continuidade no formato recebido pelo algoritmo genético.

        Aceita translated_json como dicionário ou como texto JSON."""
        import copy
        import json

        original_is_string = isinstance(translated_json, str)

        if original_is_string:
            try:
                payload = json.loads(translated_json)
            except Exception:
                payload = {}
        elif isinstance(translated_json, dict):
            payload = copy.deepcopy(translated_json)
        else:
            payload = {}

        if not isinstance(payload, dict):
            payload = {}

        payload["_continuity_directive"] = copy.deepcopy(directive)
        payload["_continuity_seed_solution"] = copy.deepcopy(seed_solution)
        payload["_continuity_bridge_version"] = self.VERSION

        # Também publica campos planos para GAs que leem aliases simples.
        payload["continuity_strategy"] = str(directive.get("type", "") or "")
        if directive.get("target_max_vehicles") is not None:
            payload["continuity_target_max_vehicles"] = int(directive["target_max_vehicles"])
        payload["continuity_mode"] = True

        if original_is_string:
            return json.dumps(payload, ensure_ascii=False)
        return payload

    def _print_improvement_plan(self, plan):
        """Rastreabilidade obrigatória antes de consumir orçamento do GA."""
        metrics = (plan or {}).get("baseline_metrics", {}) or {}
        strategies = (plan or {}).get("strategies", []) or []
        weights = (plan or {}).get("objective_weights", {}) or {}

        print("\n   🧠 PLANO DE MELHORIA OBJETIVO-DIRIGIDO")
        print(
            "      incumbente: "
            f"veículos={int(round(float(metrics.get('VEHICLES', 0) or 0)))} | "
            f"distância={float(metrics.get('DISTANCE', 0) or 0):.3f} km | "
            f"tempo={float(metrics.get('TIME', 0) or 0):.3f} min | "
            f"custo={float(metrics.get('COST', 0) or 0):.2f}"
        )
        if weights:
            print(
                "      objetivos (pesos do usuário): "
                + " | ".join(f"{k}={100.0*float(v):.1f}%" for k, v in weights.items())
            )
        flags = (plan or {}).get("operational_flags", {}) or {}
        print(
            f"      contexto estrutural: fixed_cost_mode={(plan or {}).get('fixed_cost_mode')} | "
            f"must_use_all={flags.get('must_use_all_vehicles')} | multi_trip={flags.get('multi_trip')}"
        )
        print(f"      ❓ pergunta mestre: {(plan or {}).get('master_question', 'Como posso melhorar a incumbente?')}")
        if (plan or {}).get("fleet_reduction_question"):
            print(f"      ❓ pergunta de frota: {(plan or {}).get('fleet_reduction_question')}")
        structural = (plan or {}).get("structural_incumbent_analysis", {}) or {}
        opp = structural.get("opportunities", {}) or {}
        seq = (opp.get("intra_trip_sequence", {}) or {}).get("candidate_trips", []) or []
        consol = (opp.get("trip_consolidation", {}) or {}).get("vehicles_with_multiple_trips", []) or []
        print(f"      análise estrutural: viagens={structural.get('num_routes', 0)} | candidatos_2opt={len(seq)} | veículos_multiviagem={len(consol)}")
        if seq:
            top = seq[0]
            print(f"      ↳ melhor hint geométrico 2-opt: V{top.get('vehicle_id')} / T{top.get('trip_id')} | ganho_estimado={float(top.get('estimated_distance_gain_km',0.0)):.3f} km | factibilidade=não verificada")

        for i, st in enumerate(strategies, 1):
            extra = ""
            if st.get("type") == "FLEET_REDUCTION":
                extra = (
                    f" | testar {st.get('current_used_vehicles')}→"
                    f"{st.get('target_max_vehicles')} veículos"
                )
            print(
                f"      hipótese {i}: {st.get('type')} "
                f"(prioridade={st.get('priority')}){extra}"
            )


    # TRACK A — PRIMAL / INCUMBENT IMPROVEMENT

    def _run_primal_track(
        self,
        ga_solver,
        audit_callback,
        classification,
        translated_json,
        baseline_solution,
        budget_seconds,
        seed,
        improvement_plan=None,
    ):
        """
         Track A:
        - parte SEMPRE da incumbente;
        - executa hipóteses orientadas ao objetivo;
        - tenta primeiro mudanças estruturais de maior impacto (ex.: n-1 veículos);
        - se não melhorarem, mantém a incumbente e concentra o restante da busca
          no redesenho das viagens;
        - nenhuma candidata substitui a incumbente sem auditoria e melhora estrita.
        """
        import copy
        import time

        started = time.monotonic()
        deadline = started + max(20.0, float(budget_seconds))
        best = copy.deepcopy(baseline_solution)
        best_score = self._score(best)
        n_clients = self._client_count()

        plan = copy.deepcopy(
            improvement_plan
            or self.shared_memory.data.get("continuity_improvement_plan")
            or self._objective_improvement_plan(best)
        )
        strategies = list(plan.get("strategies", []) or [])
        if not strategies:
            strategies = [{
                "type": "GENERIC_INCUMBENT_INTENSIFICATION",
                "priority": 1,
                "reason": "Fallback de intensificação da incumbente.",
            }]

        self.shared_memory.data["continuity_mode"] = True
        self.shared_memory.data["continuity_seed_solution"] = copy.deepcopy(best)
        self.shared_memory.data["continuity_current_best_score"] = float(best_score)
        self.shared_memory.data["continuity_improvement_plan"] = copy.deepcopy(plan)

        pop = 48 if n_clients > 50 else (70 if n_clients > 20 else 90)
        base_gens = 180 if n_clients > 50 else (220 if n_clients > 20 else 240)

        attempts = []
        accepted_count = 0
        last_error = None

        # No máximo 3 mecanismos por rodada para não pulverizar o orçamento.
        selected = strategies[:3]
        for idx, strategy in enumerate(selected):
            remaining = deadline - time.monotonic()
            if remaining < 8.0:
                break

            slots_left = max(1, len(selected) - idx)
            phase_budget = max(8, int(remaining / slots_left))
            directive = self._strategy_directive(strategy, best)

            # Publicação dupla: shared_memory + cópia do translated_json.
            # O segundo canal permite que o algoritmo genético receba a orientação
            # sem quebrar chamadas que ignoram a chave privada.
            self.shared_memory.data["continuity_search_directive"] = copy.deepcopy(directive)
            self.shared_memory.data["continuity_seed_solution"] = copy.deepcopy(best)

            # ponte robusta: translated_json pode ser dict OU string JSON.

            tj = self._translated_with_continuity_payload(
                translated_json,
                directive,
                best,
            )


            cand = None
            error = None
            phase_seed = int(seed) + idx * 1009
            phase_gens = max(60, int(base_gens / max(1, len(selected))))

            print(
                f"      ↳ GA dirigido | estratégia={directive.get('type')} | "
                f"seed={phase_seed} | orçamento={phase_budget}s | "
                f"target_max_vehicles={directive.get('target_max_vehicles', 'N/A')}"
            )

            try:
                cand = ga_solver.solve_problem(
                    classification,
                    tj,
                    population_size=pop,
                    num_generations=phase_gens,
                    time_limit_seconds=phase_budget,
                    seed=phase_seed,
                    feasibility_first=(n_clients > 50),
                    continuity_mode=True,
                )
            except TypeError:
                try:
                    cand = ga_solver.solve_problem(
                        classification,
                        tj,
                        population_size=pop,
                        num_generations=phase_gens,
                        time_limit_seconds=phase_budget,
                        seed=phase_seed,
                        feasibility_first=(n_clients > 50),
                    )
                except Exception as exc:
                    error = f"{type(exc).__name__}: {exc}"
            except Exception as exc:
                error = f"{type(exc).__name__}: {exc}"

            score = self._score(cand) if self._successful(cand) else None
            better = score is not None and float(score) < float(best_score) - 1e-10

            returned_strategy = None
            returned_target = None
            if isinstance(cand, dict):
                returned_strategy = (
                    cand.get("continuity_strategy")
                    or cand.get("strategy")
                    or (cand.get("continuity", {}) or {}).get("strategy")
                )
                returned_target = (
                    cand.get("continuity_target_max_vehicles")
                    or (cand.get("continuity", {}) or {}).get("target_max_vehicles")
                )

            print(
                f"      ↳ retorno GA | status="
                f"{cand.get('status') if isinstance(cand, dict) else 'INVALID'} | "
                f"score={score} | melhor={better} | "
                f"estratégia_reportada={returned_strategy or 'N/D'} | "
                f"target_reportado={returned_target if returned_target is not None else 'N/D'}"
            )

            audited = False
            accepted = False
            audit_error = None

            if better:
                _, rec = self._audit_candidate(
                    audit_callback,
                    cand,
                    f"CONTINUITY_V6_3_{strategy.get('type', 'PRIMAL')}",
                )
                audited = bool(rec.get("solucao_aprovada", False))
                audit_error = rec.get("continuity_audit_error")
                accepted = audited

            if accepted:
                best = copy.deepcopy(cand)
                best_score = float(score)
                accepted_count += 1
                self.shared_memory.data["continuity_seed_solution"] = copy.deepcopy(best)
                self.shared_memory.data["continuity_current_best_score"] = float(best_score)

            if error or audit_error:
                last_error = error or audit_error

            attempts.append({
                "strategy": copy.deepcopy(strategy),
                "directive": copy.deepcopy(directive),
                "candidate_score": score,
                "better_than_current_incumbent": bool(better),
                "audited": bool(audited),
                "accepted": bool(accepted),
                "phase_seed": int(phase_seed),
                "phase_budget_seconds": int(phase_budget),
                "generations_requested": int(phase_gens),
                "error": error or audit_error,
            })

            # Se a hipótese estrutural de redução de frota falhou, ela é encerrada
            # nesta rodada. A próxima estratégia (normalmente redesenho de rotas)
            # recebe a incumbente intacta.
            if strategy.get("type") == "FLEET_REDUCTION" and not accepted:
                strategy["status"] = "TESTED_NO_IMPROVEMENT"
            elif accepted:
                strategy["status"] = "ACCEPTED_IMPROVEMENT"
            else:
                strategy["status"] = "TESTED_NO_IMPROVEMENT"

        self.shared_memory.data["continuity_search_directive"] = None
        self.shared_memory.data["continuity_improvement_plan"]["strategies"] = copy.deepcopy(strategies)

        return {
            "track": "A_PRIMAL",
            "status": "IMPROVED" if accepted_count > 0 else ("ERROR" if last_error and not attempts else "NO_IMPROVEMENT"),
            "candidate": best,
            "candidate_score": float(best_score),
            "best_ub": float(best_score),
            "better_than_baseline": float(best_score) < float(self._score(baseline_solution)) - 1e-10,
            "audited": bool(accepted_count > 0),
            "accepted": bool(accepted_count > 0),
            "accepted_improvements": int(accepted_count),
            "seed": int(seed),
            "population": int(pop),
            "improvement_plan": plan,
            "strategy_attempts": attempts,
            "elapsed_seconds": float(time.monotonic() - started),
            "error": last_error,
        }


    # TRACK B — DUAL / SAFE LOWER-BOUND STRENGTHENING (RADIAL + DEGREE + MST)

    @staticmethod
    def _haversine_km(lat1, lon1, lat2, lon2):
        import math
        r = 6371.0088
        p1 = math.radians(float(lat1))
        p2 = math.radians(float(lat2))
        dp = math.radians(float(lat2) - float(lat1))
        dl = math.radians(float(lon2) - float(lon1))
        a = (
            math.sin(dp / 2.0) ** 2
            + math.cos(p1) * math.cos(p2) * math.sin(dl / 2.0) ** 2
        )
        a = min(1.0, max(0.0, a))
        return 2.0 * r * math.asin(math.sqrt(a))

    def _extract_nodes_for_bounds(self):
        """
        Retorna lista [{id, lat, lon, demand}] e depósito.
        Se não for possível extrair com segurança, retorna ([], None).
        """
        data = self.shared_memory.data
        df = data.get("df_clientes")
        if df is None:
            return [], None

        try:
            columns = {str(c).strip().lower(): c for c in df.columns}

            def pick(*names):
                for name in names:
                    if name in columns:
                        return columns[name]
                return None

            id_col = pick("id", "id_cliente", "cliente", "node_id")
            lat_col = pick("latitude", "lat")
            lon_col = pick("longitude", "lon", "lng")
            dem_col = pick("demanda", "demand")

            if id_col is None or lat_col is None or lon_col is None:
                return [], None

            nodes = []
            for _, row in df.iterrows():
                try:
                    ident = int(row[id_col])
                    lat = float(row[lat_col])
                    lon = float(row[lon_col])
                    demand = float(row[dem_col]) if dem_col is not None else 0.0
                    nodes.append(
                        {"id": ident, "lat": lat, "lon": lon, "demand": max(0.0, demand)}
                    )
                except Exception:
                    continue

            depot = next((x for x in nodes if int(x["id"]) == 0), None)
            clients = [x for x in nodes if int(x["id"]) != 0]
            if depot is None or not clients:
                return [], None
            return clients, depot
        except Exception:
            return [], None

    def _extract_vehicle_capacities(self):
        data = self.shared_memory.data
        df = data.get("df_veiculos_capacidade")
        if df is None:
            # nomes alternativos reconhecidos na entrada
            for key in (
                "df_capacidade_veiculos",
                "df_vehicle_capacity",
                "df_veiculos",
            ):
                if data.get(key) is not None:
                    df = data.get(key)
                    break

        caps = []
        if df is not None:
            try:
                columns = {str(c).strip().lower(): c for c in df.columns}
                cap_col = None
                for key in ("capacidade", "capacity", "vehicle_capacity"):
                    if key in columns:
                        cap_col = columns[key]
                        break
                if cap_col is not None:
                    for v in df[cap_col].tolist():
                        x = self._safe_float(v, None)
                        if x is not None and x > 0:
                            caps.append(float(x))
            except Exception:
                pass

        if caps:
            return caps

        translated = data.get("translated_json", {}) or {}
        if isinstance(translated, dict):
            for key in ("vehicle_capacities", "capacities"):
                obj = translated.get(key)
                if isinstance(obj, (list, tuple)):
                    for v in obj:
                        x = self._safe_float(v, None)
                        if x is not None and x > 0:
                            caps.append(float(x))
                    if caps:
                        return caps
        return caps

    def _extract_min_cost_per_km(self):
        data = self.shared_memory.data
        candidates = []

        for key in (
            "df_veiculos_custo",
            "df_custo_veiculos",
            "df_vehicle_cost",
            "df_veiculos_custos",
        ):
            df = data.get(key)
            if df is None:
                continue
            try:
                columns = {str(c).strip().lower(): c for c in df.columns}
                ccol = None
                for name in (
                    "custo_por_km",
                    "cost_per_km",
                    "custo_km",
                    "variable_cost_per_km",
                ):
                    if name in columns:
                        ccol = columns[name]
                        break
                if ccol is not None:
                    for v in df[ccol].tolist():
                        x = self._safe_float(v, None)
                        if x is not None and x >= 0:
                            candidates.append(float(x))
            except Exception:
                pass

        if candidates:
            return min(candidates)

        translated = data.get("translated_json", {}) or {}
        if isinstance(translated, dict):
            for key in ("cost_per_km", "vehicle_cost_per_km", "variable_cost_per_km"):
                obj = translated.get(key)
                if isinstance(obj, (list, tuple)):
                    vals = [
                        self._safe_float(x, None)
                        for x in obj
                    ]
                    vals = [x for x in vals if x is not None and x >= 0]
                    if vals:
                        return min(vals)
                else:
                    x = self._safe_float(obj, None)
                    if x is not None and x >= 0:
                        return float(x)
        return None

    def _extract_fixed_cost_lower_bound_inputs(self):
        """Extrai custo fixo mínimo não-negativo, modo de cobrança e jornada máxima."""
        data = self.shared_memory.data
        translated = self._translated_dict()

        fixed_values = []
        for key in (
            "df_veiculos_custo", "df_custo_veiculos", "df_vehicle_cost",
            "df_veiculos_custos",
        ):
            df = data.get(key)
            if df is None:
                continue
            try:
                columns = {str(c).strip().lower(): c for c in df.columns}
                fcol = None
                for name in (
                    "custo_diario", "custo_fixo", "fixed_cost", "daily_fixed_cost",
                    "fixed_daily_cost",
                ):
                    if name in columns:
                        fcol = columns[name]
                        break
                if fcol is not None:
                    for v in df[fcol].tolist():
                        x = self._safe_float(v, None)
                        if x is not None and x >= 0:
                            fixed_values.append(float(x))
            except Exception:
                pass

        min_fixed = min(fixed_values) if fixed_values else None

        mode = None
        mode_keys = (
            "fixed_cost_mode", "fixed_cost_policy", "fixed_cost_payment_mode",
            "fixed_cost_rule", "cost_fixed_mode", "fixed_cost_type",
            "vehicle_fixed_cost_mode", "custo_fixo_modo", "modo_custo_fixo",
        )
        raw_mode = self._recursive_find(translated, mode_keys)
        if raw_mode is None:
            raw_mode = self._recursive_find(data, mode_keys)
        if raw_mode is not None:
            mode = str(raw_mode).strip().upper()
        # Canonicalizações tolerantes apenas para reconhecer WHEN_USED/ALWAYS.
        if mode:
            if any(t in mode for t in ("WHEN_USED", "ATIV", "UTILIZ", "USED", "ONLY_IF", "CONDIT", "POR_USO", "SE_USADO")):
                mode = "WHEN_USED"
            elif any(t in mode for t in ("ALWAYS", "INCOND", "SEMPRE")):
                mode = "ALWAYS"

        max_time = None
        time_keys = (
            "max_time_per_vehicle_minutes", "available_time_per_vehicle_minutes",
            "vehicle_available_time_minutes", "max_vehicle_time_minutes",
        )
        raw_time = self._recursive_find(translated, time_keys)
        if raw_time is None:
            raw_time = self._recursive_find(data, time_keys)
        x = self._safe_float(raw_time, None)
        if x is not None and x > 0:
            max_time = float(x)

        return min_fixed, mode, max_time

    def _extract_speed_service(self):
        data = self.shared_memory.data
        translated = self._translated_dict()

        speed = None
        service = None
        for source in (translated, data):
            if speed is None:
                for key in (
                    "average_vehicle_speed_kmh",
                    "average_speed_kmh",
                    "speed_kmh",
                ):
                    x = self._safe_float(source.get(key), None)
                    if x is not None and x > 0:
                        speed = x
                        break
            if service is None:
                for key in (
                    "service_time_per_delivery_minutes",
                    "service_time_minutes",
                    "average_service_time_minutes",
                ):
                    x = self._safe_float(source.get(key), None)
                    if x is not None and x >= 0:
                        service = x
                        break
        return speed, service

    def _distance_lower_bounds(self, clients, depot, qmax):
        """
        Bounds válidos para distância total em métrica Haversine.

        radial_capacity:
            (2/Qmax) * sum_i q_i d(0,i)
        degree:
            1/2 * soma, por cliente, das duas menores arestas incidentes
            considerando depósito + demais clientes.

        O resultado usado é max(radial, degree), nunca soma.
        """
        n = len(clients)
        if n <= 0:
            return {}

        radial = None
        if qmax is not None and qmax > 0:
            radial = (
                2.0
                / float(qmax)
                * sum(
                    float(c.get("demand", 0.0))
                    * self._haversine_km(
                        depot["lat"], depot["lon"], c["lat"], c["lon"]
                    )
                    for c in clients
                )
            )

        all_nodes = [depot] + clients
        degree_sum = 0.0
        degree_valid = True

        for c in clients:
            distances = []
            for other in all_nodes:
                if int(other["id"]) == int(c["id"]):
                    continue
                distances.append(
                    self._haversine_km(
                        c["lat"], c["lon"], other["lat"], other["lon"]
                    )
                )
            distances.sort()
            if len(distances) < 2:
                degree_valid = False
                break
            degree_sum += distances[0] + distances[1]

        degree = 0.5 * degree_sum if degree_valid else None

        # MST sobre depósito + clientes: bound métrico global seguro para a
        # distância total de qualquer conjunto de rotas que conecta todos os clientes
        # ao depósito. Prim O(n^2), adequado para instâncias grandes.
        mst = None
        try:
            nodes = [depot] + clients
            m = len(nodes)
            if m >= 2:
                in_tree = [False] * m
                key = [float("inf")] * m
                key[0] = 0.0
                total = 0.0
                for _ in range(m):
                    u = -1
                    best = float("inf")
                    for i in range(m):
                        if not in_tree[i] and key[i] < best:
                            best = key[i]
                            u = i
                    if u < 0:
                        break
                    in_tree[u] = True
                    total += key[u]
                    for v in range(m):
                        if in_tree[v] or v == u:
                            continue
                        d = self._haversine_km(
                            nodes[u]["lat"], nodes[u]["lon"],
                            nodes[v]["lat"], nodes[v]["lon"],
                        )
                        if d < key[v]:
                            key[v] = d
                if all(in_tree):
                    mst = float(total)
        except Exception:
            mst = None

        vals = [x for x in (radial, degree, mst) if x is not None and x >= 0]
        return {
            "radial_capacity_distance_lb": radial,
            "degree_distance_lb": degree,
            "mst_distance_lb": mst,
            "best_distance_lb": max(vals) if vals else None,
        }

    def _canonical_lp_lower_bound(self, clients, depot, qmax, budget_seconds, incumbent_ub=None):
        """
        Relaxação LP globalmente válida baseada em fluxo de uma mercadoria (SCF).

        Variáveis contínuas:
          x_ij in [0,1]: uso fracionário do arco dirigido i->j
          f_ij >= 0: fluxo de carga entregue através do arco

        Restrições relaxadas:
          - cada cliente tem exatamente uma entrada e uma saída;
          - balanço de fluxo de demanda em cada cliente;
          - f_ij <= Qmax*x_ij;
          - número de saídas do depósito = número de retornos ao depósito.

        Qmax é a MAIOR capacidade física da frota. Portanto, trocar cada veículo
        heterogêneo por um veículo relaxado de capacidade Qmax só amplia a região
        factível: o ótimo deste LP é um lower bound do problema original.

        A função objetivo é montada DIRETAMENTE no objetivo canônico congelado:
          DISTANCE: alpha_D * distancia
          TIME:     alpha_T * (travel + n*service)
          COST:     alpha_C * (min_cost_km * distancia)
          VEHICLES: alpha_V * 1, se há demanda

        Custos fixos, espera, janelas, jornada, tempo mínimo e heterogeneidade são
        omitidos no LP. Isso torna o bound mais fraco, mas continua seguro.

        Só aceitamos o LP como certificado quando scipy.optimize.linprog/HiGHS
        retorna status ótimo. Timeout/inconclusivo NUNCA é promovido a lower bound.
        """
        import math
        import time

        started = time.monotonic()
        result = {
            "name": "continuity_v5_scf_lp_bound",
            "valid": False,
            "value": None,
            "scope": "CANONICAL_OBJECTIVE",
            "source": "CONTINUITY_TRACK_B_SCF_LP_HIGHS",
            "method": "SCF_LP_RELAXATION_QMAX",
            "solver": "scipy.optimize.linprog(method='highs')",
            "status": "NOT_RUN",
            "elapsed_seconds": 0.0,
        }

        if not clients or depot is None or qmax is None or qmax <= 0:
            result["status"] = "SKIPPED_MISSING_DATA"
            return result

        try:
            import numpy as np
            from scipy.optimize import linprog
            from scipy.sparse import lil_matrix
        except Exception as exc:
            result["status"] = "SCIPY_UNAVAILABLE"
            result["error"] = f"{type(exc).__name__}: {exc}"
            return result

        ctx = self.shared_memory.data.get("objective_context", {}) or {}
        coeff = ctx.get("coefficients", {}) or {}
        active = [str(x).upper() for x in (ctx.get("active_objectives", []) or [])]

        # Coeficientes canônicos devem ser não negativos em um problema de minimização.
        alphas = {}
        for obj in ("DISTANCE", "TIME", "COST", "VEHICLES"):
            a = self._safe_float(coeff.get(obj), 0.0) or 0.0
            if a < -1e-15:
                result["status"] = "SKIPPED_NEGATIVE_CANONICAL_COEFFICIENT"
                return result
            alphas[obj] = max(0.0, float(a))

        speed, service = self._extract_speed_service()
        min_cost_km = self._extract_min_cost_per_km()

        # Se uma componente ativa não puder ser representada com um bound seguro,
        # usamos zero para essa componente. Zero é válido porque as métricas/custos
        # do modelo são não negativos. Para TIME, serviço só é somado se conhecido.
        distance_multiplier = alphas["DISTANCE"]
        if "TIME" in active and speed is not None and speed > 0:
            distance_multiplier += alphas["TIME"] * 60.0 / float(speed)
        if "COST" in active and min_cost_km is not None and min_cost_km >= 0:
            distance_multiplier += alphas["COST"] * float(min_cost_km)

        # Lower bound seguro do número de veículos por jornada agregada.
        # Soma do trabalho de todos os veículos >= serviço total + viagem mínima.
        min_fixed, fixed_mode, max_vehicle_time = self._extract_fixed_cost_lower_bound_inputs()
        analytical_dist = self._distance_lower_bounds(clients, depot, qmax)
        analytical_distance_lb = self._safe_float(analytical_dist.get("best_distance_lb"), 0.0) or 0.0
        vehicle_lb = 1 if len(clients) > 0 else 0
        if (
            len(clients) > 0 and max_vehicle_time is not None and max_vehicle_time > 0
            and speed is not None and speed > 0 and service is not None and service >= 0
        ):
            total_work_lb = (
                len(clients) * float(service)
                + float(analytical_distance_lb) * 60.0 / float(speed)
            )
            vehicle_lb = max(1, int(math.ceil(total_work_lb / float(max_vehicle_time) - 1e-12)))
        else:
            total_work_lb = None

        constant_lb = 0.0
        if "TIME" in active and service is not None and service >= 0:
            constant_lb += alphas["TIME"] * len(clients) * float(service)
        if "VEHICLES" in active and len(clients) > 0:
            constant_lb += alphas["VEHICLES"] * float(vehicle_lb)

        # COST: se o custo fixo é por ativação, qualquer solução precisa pagar
        # ao menos vehicle_lb vezes o MENOR custo fixo não-negativo da frota.
        # Se o modo é desconhecido, omitimos esta parcela (zero = bound seguro).
        fixed_cost_lb = 0.0
        if (
            "COST" in active and min_fixed is not None and min_fixed >= 0
            and fixed_mode == "WHEN_USED" and vehicle_lb > 0
        ):
            fixed_cost_lb = float(vehicle_lb) * float(min_fixed)
            constant_lb += alphas["COST"] * fixed_cost_lb

        nodes = [depot] + list(clients)
        m = len(nodes)
        # Todos os arcos dirigidos i!=j.
        arcs = [(i, j) for i in range(m) for j in range(m) if i != j]
        A = len(arcs)
        arc_index = {a: k for k, a in enumerate(arcs)}
        nvar = 2 * A  # x[0:A], f[A:2A]

        c = np.zeros(nvar, dtype=float)
        distances = np.zeros(A, dtype=float)
        for k, (i, j) in enumerate(arcs):
            d = self._haversine_km(
                nodes[i]["lat"], nodes[i]["lon"],
                nodes[j]["lat"], nodes[j]["lon"],
            )
            distances[k] = d
            c[k] = float(distance_multiplier) * float(d)

        # Igualdades: entrada/saída dos clientes + balanço de carga +
        # equilíbrio de viagens no depósito.
        neq = 3 * (m - 1) + 1
        Aeq = lil_matrix((neq, nvar), dtype=float)
        beq = np.zeros(neq, dtype=float)
        row = 0

        # Cada cliente exatamente uma entrada.
        for j in range(1, m):
            for i in range(m):
                if i != j:
                    Aeq[row, arc_index[(i, j)]] = 1.0
            beq[row] = 1.0
            row += 1

        # Cada cliente exatamente uma saída.
        for i in range(1, m):
            for j in range(m):
                if i != j:
                    Aeq[row, arc_index[(i, j)]] = 1.0
            beq[row] = 1.0
            row += 1

        # Fluxo entregue: inflow - outflow = demanda do cliente.
        for j in range(1, m):
            for i in range(m):
                if i != j:
                    Aeq[row, A + arc_index[(i, j)]] += 1.0
            for k2 in range(m):
                if k2 != j:
                    Aeq[row, A + arc_index[(j, k2)]] -= 1.0
            beq[row] = max(0.0, float(nodes[j].get("demand", 0.0)))
            row += 1

        # Cada viagem que sai do depósito retorna ao depósito.
        for j in range(1, m):
            Aeq[row, arc_index[(0, j)]] += 1.0
            Aeq[row, arc_index[(j, 0)]] -= 1.0
        beq[row] = 0.0

        # Desigualdades f_ij - Qmax*x_ij <= 0.
        Aub = lil_matrix((A, nvar), dtype=float)
        bub = np.zeros(A, dtype=float)
        for k in range(A):
            Aub[k, k] = -float(qmax)
            Aub[k, A + k] = 1.0

        bounds = [(0.0, 1.0)] * A + [(0.0, None)] * A

        # Reserva pequena para serialização/coordenação do Track B.
        lp_limit = max(5.0, min(float(budget_seconds) * 0.90, float(budget_seconds) - 1.0))
        result.update({
            "num_nodes_including_depot": int(m),
            "num_arcs": int(A),
            "num_variables": int(nvar),
            "num_equalities": int(neq),
            "num_inequalities": int(A),
            "qmax_relaxation": float(qmax),
            "distance_multiplier_canonical": float(distance_multiplier),
            "constant_component_lb": float(constant_lb),
            "analytical_distance_lb_for_vehicle_count": float(analytical_distance_lb),
            "vehicle_lower_bound": int(vehicle_lb),
            "aggregate_work_lower_bound_minutes": total_work_lb,
            "min_fixed_cost": min_fixed,
            "fixed_cost_mode": fixed_mode,
            "fixed_cost_lower_bound": float(fixed_cost_lb),
            "max_vehicle_time_minutes": max_vehicle_time,
            "time_limit_seconds": float(lp_limit),
        })

        try:
            res = linprog(
                c,
                A_ub=Aub.tocsr(), b_ub=bub,
                A_eq=Aeq.tocsr(), b_eq=beq,
                bounds=bounds,
                method="highs",
                options={"time_limit": float(lp_limit), "presolve": True},
            )
        except Exception as exc:
            result["status"] = "LP_EXCEPTION"
            result["error"] = f"{type(exc).__name__}: {exc}"
            result["elapsed_seconds"] = float(time.monotonic() - started)
            return result

        result["solver_status_code"] = int(getattr(res, "status", -999))
        result["solver_message"] = str(getattr(res, "message", ""))
        result["elapsed_seconds"] = float(time.monotonic() - started)

        # HiGHS status 0 = ótimo. Só neste caso usamos fun como LB certificado.
        if int(getattr(res, "status", -1)) != 0 or not bool(getattr(res, "success", False)):
            result["status"] = "LP_NOT_OPTIMAL_NO_BOUND_ACCEPTED"
            return result

        lp_variable_part = self._safe_float(getattr(res, "fun", None), None)
        if lp_variable_part is None or not math.isfinite(lp_variable_part):
            result["status"] = "LP_INVALID_OBJECTIVE"
            return result

        value = max(0.0, float(lp_variable_part) + float(constant_lb))

        # Sanity gate: um lower bound válido não pode exceder um incumbent factível
        # materialmente. Pequena inversão numérica é truncada; inversão grande rejeita.
        if incumbent_ub is not None and math.isfinite(float(incumbent_ub)):
            tol = max(1e-8, 1e-7 * max(1.0, abs(float(incumbent_ub))))
            if value > float(incumbent_ub) + tol:
                result["status"] = "REJECTED_BOUND_EXCEEDS_FEASIBLE_UB"
                result["raw_value"] = float(value)
                return result
            if value > float(incumbent_ub):
                value = float(incumbent_ub)

        result.update({
            "status": "OPTIMAL_VALID_GLOBAL_LB",
            "valid": True,
            "value": float(value),
            "lp_variable_part": float(lp_variable_part),
            "proof_note": (
                "Ótimo de uma relaxação LP SCF que contém toda solução factível "
                "do problema original após relaxar heterogeneidade para Qmax e "
                "omitir restrições/custos não negativos."
            ),
        })
        return result

    def _run_dual_track(self, baseline_solution, budget_seconds):
        import copy
        import math
        import time

        started = time.monotonic()
        objective_ctx = self.shared_memory.data.get("objective_context", {}) or {}
        coeff = objective_ctx.get("coefficients", {}) or {}
        active = objective_ctx.get("active_objectives", []) or []

        baseline_valid_lb = self._safe_float(
            self.shared_memory.data.get("continuity_baseline_valid_lower_bound"),
            None,
        )

        clients, depot = self._extract_nodes_for_bounds()
        caps = self._extract_vehicle_capacities()
        qmax = max(caps) if caps else None
        total_capacity = sum(caps) if caps else None
        total_demand = sum(float(c.get("demand", 0.0)) for c in clients)

        candidates = []
        details = {
            "clients_used": len(clients),
            "qmax": qmax,
            "total_demand": total_demand,
        }

        if baseline_valid_lb is not None:
            candidates.append(
                {
                    "name": "baseline_certified_lower_bound",
                    "value": float(baseline_valid_lb),
                    "valid": True,
                    "scope": "CANONICAL_OBJECTIVE",
                    "source": "BASELINE_CERTIFIER",
                }
            )

        if clients and depot is not None:
            dist = self._distance_lower_bounds(clients, depot, qmax)
            details.update(dist)

            distance_lb = self._safe_float(dist.get("best_distance_lb"), None)
            speed, service = self._extract_speed_service()
            min_cost_km = self._extract_min_cost_per_km()

            component_lb = {
                "DISTANCE": 0.0,
                "TIME": 0.0,
                "COST": 0.0,
                "VEHICLES": 0.0,
            }

            if distance_lb is not None:
                component_lb["DISTANCE"] = max(0.0, float(distance_lb))

            if (
                distance_lb is not None
                and speed is not None
                and speed > 0
            ):
                travel_lb = float(distance_lb) * 60.0 / float(speed)
                service_lb = (
                    len(clients) * float(service)
                    if service is not None and service >= 0
                    else 0.0
                )
                component_lb["TIME"] = max(0.0, travel_lb + service_lb)
                details["travel_time_lb_minutes"] = travel_lb
                details["service_time_lb_minutes"] = service_lb

            if distance_lb is not None and min_cost_km is not None:
                component_lb["COST"] = max(
                    0.0,
                    float(min_cost_km) * float(distance_lb),
                )
                details["min_nonnegative_cost_per_km"] = min_cost_km

            if caps and qmax and qmax > 0:
                # Lower bound seguro do número de veículos por capacidade total
                # é delicado com multi-trip. Com multi-trip, capacidade por viagem
                # não implica veículos. Portanto usamos apenas 1 se há demanda.
                component_lb["VEHICLES"] = 1.0 if total_demand > 0 else 0.0
                details["capacity_lower_bound_on_trips"] = int(
                    math.ceil(total_demand / qmax - 1e-12)
                )

            canonical_lb = 0.0
            valid_for_active = True

            for obj in active:
                alpha = self._safe_float(coeff.get(obj), 0.0) or 0.0
                if alpha < -1e-15:
                    valid_for_active = False
                    break
                canonical_lb += alpha * float(component_lb.get(obj, 0.0))

            if valid_for_active:
                candidates.append(
                    {
                        "name": "continuity_v5_safe_component_bound",
                        "value": float(canonical_lb),
                        "valid": True,
                        "scope": "CANONICAL_OBJECTIVE",
                        "source": "CONTINUITY_TRACK_B_ANALYTICAL",
                        "components": component_lb,
                        "details": details,
                    }
                )


        # LP Bound Engine (SCF relaxation)

        incumbent_ub = self._score(baseline_solution)
        lp_candidate = self._canonical_lp_lower_bound(
            clients, depot, qmax, budget_seconds, incumbent_ub=incumbent_ub
        )
        details["lp_engine_status"] = lp_candidate.get("status")
        details["lp_engine_elapsed_seconds"] = lp_candidate.get("elapsed_seconds")
        if isinstance(lp_candidate, dict) and lp_candidate.get("valid"):
            candidates.append(lp_candidate)

        valid_values = [
            self._safe_float(x.get("value"), None)
            for x in candidates
            if isinstance(x, dict) and x.get("valid")
        ]
        valid_values = [x for x in valid_values if x is not None]

        best_lb = max(valid_values) if valid_values else baseline_valid_lb
        strengthened = (
            best_lb is not None
            and baseline_valid_lb is not None
            and float(best_lb) > float(baseline_valid_lb) + 1e-12
        )

        # Publica em namespace próprio. O Certifier continua sendo a autoridade
        # final e pode decidir consumir ou ignorar estes bounds.
        self.shared_memory.data["continuity_bound_candidates"] = candidates
        self.shared_memory.data["continuity_best_valid_lower_bound"] = best_lb
        self.shared_memory.data["continuity_bound_strengthened"] = bool(strengthened)
        self.shared_memory.data["continuity_bound_certificate"] = {
            "version": self.VERSION,
            "value": best_lb,
            "valid_global_lower_bound": bool(best_lb is not None),
            "scope": "CANONICAL_OBJECTIVE",
            "objective_context_version": str(objective_ctx.get("version", "") or ""),
            "method": "MAX_OF_CERTIFIED_BASELINE_ANALYTICAL_AND_SCF_LP_BOUNDS",
            "candidates": copy.deepcopy(candidates),
            "note": (
                "Bound candidate produzido pela continuidade. O Certifier permanece "
                "a única autoridade para classificação A/B/C/D."
            ),
        }

        return {
            "track": "B_DUAL",
            "status": "STRENGTHENED" if strengthened else "NO_STRONGER_BOUND",
            "baseline_valid_lb": baseline_valid_lb,
            "best_valid_lb": best_lb,
            "strengthened": bool(strengthened),
            "bound_candidates": candidates,
            "elapsed_seconds": float(time.monotonic() - started),
            "error": None,
        }


    # DUAL-TRACK COORDINATION

    def search_better_incumbent(
        self,
        ga_solver,
        audit_callback,
        classification,
        translated_json,
        baseline_solution,
        budget_seconds,
        seed_base=20260917,
        ave_agent=None,
        bound_agent=None,
        k_vehicle_certifier=None,
        optimality_certifier=None,
    ):
        """CONTINUITY: alterna explicitamente UB_IMPROVEMENT_MODE e K_PROOF_MODE persistente."""
        import copy, json, time
        from concurrent.futures import ThreadPoolExecutor

        started = time.monotonic()
        budget_seconds = max(20, int(budget_seconds))
        baseline = copy.deepcopy(baseline_solution)
        baseline_score = self._score(baseline)
        improvement_plan = self._publish_improvement_plan(baseline)
        self._print_improvement_plan(improvement_plan)
        seed = self._safe_int(self.shared_memory.data.get("continuity_restart_seed"), seed_base)
        if seed <= 0: seed = int(seed_base)

        ave_agent = ave_agent or self.shared_memory.data.get("_ave_agent")
        bound_agent = bound_agent or self.shared_memory.data.get("_bound_strengthening_agent")
        if ave_agent is None or bound_agent is None:
            raise RuntimeError("CONTINUITY V9 requer AVE e BoundStrengthening instanciados pelo Orchestrator.")


        # MÁQUINA DE ESTADOS: após duas tentativas UNKNOWN para o mesmo K,
        # interrompe AVE/FLEET_DROP/GA-diversification/Bound repetitivo e dedica
        # a rodada à pergunta estrutural K. O progresso Exact-K é persistente.

        k_vehicle_certifier = (
            k_vehicle_certifier
            or self.shared_memory.data.get("_k_vehicle_feasibility_certifier")
        )
        optimality_certifier = (
            optimality_certifier
            or self.shared_memory.data.get("_optimality_certifier_agent")
        )
        baseline_used = int(round(self._metrics(baseline).get("VEHICLES", 0.0) or 0.0))
        proof_target = baseline_used - 1 if baseline_used > 1 else None
        khist = self.shared_memory.data.get("k_vehicle_feasibility_history", []) or []
        prior_unknown = sum(
            1 for r in khist
            if isinstance(r, dict)
            and self._safe_int(r.get("target_k"), -1) == int(proof_target or -1)
            and str(r.get("status", "")).upper() == "UNKNOWN"
        )
        # não existe mais modo exclusivo que suspenda A/B.
        # Track C roda concorrentemente com Primal e Bound em processo isolado.
        proof_mode = False

        if proof_mode:
            print(
                f"   🔐 K_PROOF_MODE V9 ATIVO | UB veículos={baseline_used} | "
                f"testar K={proof_target} | UNKNOWN prévios={prior_unknown}"
            )
            print("      ⏸️ AVE/FLEET_DROP/GA-diversification suspensos nesta rodada.")
            proof_budget = max(20, int(0.92 * budget_seconds))
            kres = k_vehicle_certifier.certify(
                target_k=int(proof_target),
                translated_problem=translated_json,
                classification=classification,
                incumbent=baseline,
                ga_solver=ga_solver,
                auditor=audit_callback,
                time_limit_seconds=int(proof_budget),
                seed=int(seed + 32452843),
                optimality_certifier=optimality_certifier,
                exact_k_enabled=True,
            )
            kstatus = str((kres or {}).get("status", "UNKNOWN")).upper()
            best = copy.deepcopy(baseline)
            best_score = float(baseline_score)

            if kstatus == "FEASIBLE" and isinstance(kres.get("candidate"), dict):
                kcand = copy.deepcopy(kres["candidate"])
                kscore = self._score(kcand)
                if kscore < best_score - 1e-10:
                    best, best_score = kcand, float(kscore)

            # Só recalcular Bound quando surgiu prova global nova.
            base_lb = self._safe_float(
                self.shared_memory.data.get("continuity_best_valid_lower_bound"),
                self._safe_float(self.shared_memory.data.get("continuity_baseline_valid_lower_bound"), None),
            )
            best_lb = base_lb
            bound_details = {"status": "SKIPPED_NO_NEW_GLOBAL_EVIDENCE"}
            if kstatus == "PROVEN_INFEASIBLE" and kres.get("global_valid") is True:
                try:
                    cert0 = copy.deepcopy(
                        self.shared_memory.data.get("certification_result")
                        or self.shared_memory.data.get("optimality_certificate")
                        or {}
                    )
                    bres = bound_agent.strengthen(
                        best, cert0, None,
                        max(5, min(30, int(0.08 * budget_seconds)))
                    )
                    best_lb = self._safe_float(bres.get("best_lower_bound"), base_lb)
                    bound_details = copy.deepcopy(bres)
                except Exception as exc:
                    bound_details = {"status": "ERROR", "error": f"{type(exc).__name__}: {exc}"}

            ub_improved = best_score < baseline_score - 1e-10
            lb_improved = (
                best_lb is not None and base_lb is not None
                and best_lb > base_lb + 1e-12
            )
            if kstatus == "PROVEN_INFEASIBLE":
                status = "K_PROVEN_INFEASIBLE"
            elif kstatus == "FEASIBLE":
                status = "K_FEASIBLE"
            else:
                status = "K_PROOF_PENDING"

            more_time = bool((kres or {}).get("more_time_required", False))
            exact = (kres or {}).get("exact_proof") or {}
            if kstatus == "UNKNOWN":
                more_time = True
                self.shared_memory.data["continuity_more_time_required"] = True
                self.shared_memory.data["continuity_more_time_reason"] = {
                    "target_k": int(proof_target),
                    "status": "K_PROOF_PENDING",
                    "proven_infeasible_subsets": exact.get("proven_infeasible_subsets", kres.get("proven_infeasible_subsets")),
                    "unresolved_subsets": exact.get("unresolved_subsets", kres.get("unresolved_subsets")),
                    "total_subsets": exact.get("total_subsets", kres.get("total_subsets")),
                    "message": (
                        f"A factibilidade com K={proof_target} ainda não foi decidida. "
                        "O progresso Exact-K foi preservado e a próxima continuidade retomará dos subconjuntos pendentes."
                    ),
                }
                print(
                    f"   ⏳ K={proof_target} ainda NÃO DECIDIDO | "
                    f"fechados={exact.get('proven_infeasible_subsets', kres.get('proven_infeasible_subsets'))}/"
                    f"{exact.get('total_subsets', kres.get('total_subsets'))}. "
                    "Mais tempo é necessário; o progresso não será perdido."
                )
            else:
                self.shared_memory.data["continuity_more_time_required"] = False
                self.shared_memory.data["continuity_more_time_reason"] = None

            history_entry = {
                "seed": int(seed),
                "budget_seconds": int(budget_seconds),
                "elapsed_seconds": float(time.monotonic() - started),
                "status": status,
                "mode": "K_PROOF_MODE",
                "target_k": int(proof_target),
                "baseline_ub": float(baseline_score),
                "best_ub": float(best_score),
                "baseline_lb": base_lb,
                "best_lb": best_lb,
                "ub_improved": bool(ub_improved),
                "lb_improved": bool(lb_improved),
                "more_time_required": bool(more_time),
            }
            self.shared_memory.data["continuity_candidate"] = copy.deepcopy(best)
            self.shared_memory.data["continuity_best_valid_lower_bound"] = best_lb
            hist0 = self.shared_memory.data.get("continuity_history", [])
            self.shared_memory.data["continuity_history"] = (
                hist0 if isinstance(hist0, list) else []
            ) + [copy.deepcopy(history_entry)]

            return {
                "status": status,
                "best_candidate": best,
                "best_ub": float(best_score),
                "best_valid_lb": best_lb,
                "history": [history_entry],
                "elapsed_seconds": float(time.monotonic() - started),
                "audited_improvements": 1 if ub_improved else 0,
                "bound_improvements": 1 if lb_improved else 0,
                "track_a": {"track": "A_PRIMAL", "status": "SUSPENDED_DURING_K_PROOF_MODE"},
                "track_b": {"track": "B_DUAL", "status": bound_details.get("status"), "details": bound_details},
                "track_c": {"track": "C_K_EXACT_PROOF", "status": kstatus, "details": copy.deepcopy(kres)},
                "dual_track_status": status,
                "improvement_plan": copy.deepcopy(improvement_plan),
                "more_time_required": bool(more_time),
            }

        # Orçamento normal: usado somente enquanto ainda estamos em UB_IMPROVEMENT_MODE.
        bound_budget = max(5, min(45, int(0.15 * budget_seconds)))
        ave_budget = max(10, int(0.60 * budget_seconds))
        print(f"   ⚖️ Continuidade V10 | A=PRIMAL | B=BOUND | C=K-PROOF | "
              f"AVE≈{ave_budget}s | Bound≈{bound_budget}s | GA=escape-only")

        def eval_cb(sol):
            return self._score(sol)

        cert = copy.deepcopy(self.shared_memory.data.get("certification_result") or
                             self.shared_memory.data.get("optimality_certificate") or {})
        # nunca deixe o Track B esquecer o melhor LB global já conhecido.
        carried_lb = self._safe_float(
            self.shared_memory.data.get("continuity_best_valid_lower_bound"),
            self._safe_float(self.shared_memory.data.get("continuity_baseline_valid_lower_bound"), None),
        )
        if carried_lb is not None:
            cert_lb = self._safe_float(cert.get("lower_bound"), None) if isinstance(cert, dict) else None
            if not isinstance(cert, dict):
                cert = {}
            cert["lower_bound"] = max(float(carried_lb), float(cert_lb or 0.0))
            cert["certified_lower_bound"] = cert["lower_bound"]


        # TRACKS A/B/C CONCORRENTES EM PROCESSOS INDEPENDENTES
        # Nenhum worker escreve na SharedMemory oficial do processo pai.
        # O Orchestrator/Continuity faz merge somente após receber resultados.

        import multiprocessing as mp
        k_parallel_res = None
        k_parallel_state = None

        baseline_used_parallel = int(round(
            self._metrics(baseline).get("VEHICLES", 0.0) or 0.0
        ))
        target_k_parallel = (
            baseline_used_parallel - 1 if baseline_used_parallel > 1 else None
        )
        k_budget_parallel = max(30, int(0.88 * budget_seconds))

        print(
            f"   ⚡ Continuidade V10 PROCESS-PARALLEL | "
            f"A(PRIMAL)≈{ave_budget}s | B(BOUND)≈{bound_budget}s | "
            f"C(K-PROOF)≈{k_budget_parallel}s"
        )

        def _worker_a(q):
            try:
                r = ave_agent.optimize(
                    copy.deepcopy(baseline),
                    copy.deepcopy(improvement_plan),
                    eval_cb,
                    audit_callback,
                    int(ave_budget),
                    int(seed),
                )
                q.put(("A", r, None, copy.deepcopy(
                    ave_agent.shared_memory.data.get("ave_v2_persistent_state", {}))))
            except Exception as exc:
                q.put(("A", None, f"{type(exc).__name__}: {exc}"))

        def _worker_b(q):
            try:
                r = bound_agent.strengthen(
                    copy.deepcopy(baseline),
                    copy.deepcopy(cert),
                    None,
                    int(bound_budget),
                )
                bmem = bound_agent.shared_memory.data
                bstate = {key:copy.deepcopy(bmem[key]) for key in (
                    "bound_v2_persistent_state", "bound_strengthening_last_fingerprint",
                    "bound_strengthening_last_result", "bound_strengthening_best_lb",
                    "bound_cut_pool", "scf_lp_cache_v5", "scf_lp_last_result_v5",
                    "certified_lower_bound") if key in bmem}
                q.put(("B", r, None, bstate))
            except Exception as exc:
                q.put(("B", None, f"{type(exc).__name__}: {exc}"))

        def _worker_c(q):
            try:
                if (
                    k_vehicle_certifier is None
                    or optimality_certifier is None
                    or target_k_parallel is None
                ):
                    q.put(("C", None, "K_PROOF_NOT_APPLICABLE", None))
                    return
                r = k_vehicle_certifier.certify(
                    target_k=int(target_k_parallel),
                    translated_problem=translated_json,
                    classification=classification,
                    incumbent=copy.deepcopy(baseline),
                    ga_solver=ga_solver,
                    auditor=audit_callback,
                    time_limit_seconds=int(k_budget_parallel),
                    seed=int(seed + 32452843),
                    optimality_certifier=optimality_certifier,
                    exact_k_enabled=True,
                )
                # Persistência do filho precisa voltar explicitamente ao pai.
                child_mem = getattr(k_vehicle_certifier.shared_memory, "data", {}) or {}
                proof_state = {
                    "k_exact_persistent_state":
                        copy.deepcopy(child_mem.get("k_exact_persistent_state", {})),
                    "k_vehicle_feasibility_history":
                        copy.deepcopy(child_mem.get("k_vehicle_feasibility_history", [])),
                    "k_vehicle_feasibility_last_result":
                        copy.deepcopy(child_mem.get("k_vehicle_feasibility_last_result")),
                    "certified_vehicle_lower_bound":
                        child_mem.get("certified_vehicle_lower_bound"),
                    "continuity_best_vehicle_lower_bound":
                        child_mem.get("continuity_best_vehicle_lower_bound"),
                }
                q.put(("C", r, None, proof_state))
            except Exception as exc:
                q.put(("C", None, f"{type(exc).__name__}: {exc}", None))

        try:
            ctx = mp.get_context("fork")
            q = ctx.Queue()
            procs = [
                ctx.Process(target=_worker_a, args=(q,)),
                ctx.Process(target=_worker_b, args=(q,)),
            ]
            if (
                k_vehicle_certifier is not None
                and optimality_certifier is not None
                and target_k_parallel is not None
            ):
                procs.append(ctx.Process(target=_worker_c, args=(q,)))

            for p in procs:
                p.start()

            received = {}
            deadline_parallel = time.monotonic() + max(
                float(ave_budget), float(bound_budget),
                float(k_budget_parallel if len(procs) >= 3 else 0)
            ) + 5.0

            while len(received) < len(procs) and time.monotonic() < deadline_parallel:
                try:
                    item = q.get(timeout=5.0)
                except Exception:
                    continue
                tag = item[0]
                received[tag] = item

            for p in procs:
                try:
                    p.join(timeout=2.0)
                    if p.is_alive():
                        p.terminate()
                        p.join(timeout=2.0)
                except Exception:
                    pass

            aitem = received.get("A")
            if aitem and isinstance(aitem[1], dict):
                ave_res = aitem[1]
                if len(aitem)>3 and isinstance(aitem[3],dict):
                    self.shared_memory.data["ave_v2_persistent_state"] = copy.deepcopy(aitem[3])
                    ave_agent._load_persistent_state()
            else:
                err = aitem[2] if aitem else "PROCESS_A_NO_RESULT"
                ave_res = {
                    "status":"ERROR", "solution":copy.deepcopy(baseline),
                    "best_score":baseline_score, "improved":False,
                    "need_diversification":True, "error":err,
                }

            bitem = received.get("B")
            if bitem and isinstance(bitem[1], dict):
                bound_res = bitem[1]
                if len(bitem)>3 and isinstance(bitem[3],dict):
                    self.shared_memory.data.update(copy.deepcopy(bitem[3]))
                    bound_agent._load_persistent()
            else:
                err = bitem[2] if bitem else "PROCESS_B_NO_RESULT"
                base_lb = self._safe_float(
                    self.shared_memory.data.get("continuity_baseline_valid_lower_bound"),
                    None,
                )
                bound_res = {
                    "status":"ERROR", "base_lower_bound":base_lb,
                    "best_lower_bound":base_lb, "improved":False, "error":err,
                }

            citem = received.get("C")
            if citem:
                k_parallel_res = citem[1] if isinstance(citem[1], dict) else None
                k_parallel_state = citem[3] if len(citem) > 3 else None

                # Merge controlado: só o pai altera SharedMemory oficial.
                if isinstance(k_parallel_state, dict):
                    child_exact = k_parallel_state.get("k_exact_persistent_state")
                    if isinstance(child_exact, dict):
                        self.shared_memory.data["k_exact_persistent_state"] = child_exact
                    child_hist = k_parallel_state.get("k_vehicle_feasibility_history")
                    if isinstance(child_hist, list):
                        self.shared_memory.data["k_vehicle_feasibility_history"] = child_hist[-100:]
                    if k_parallel_state.get("k_vehicle_feasibility_last_result") is not None:
                        self.shared_memory.data["k_vehicle_feasibility_last_result"] = \
                            k_parallel_state["k_vehicle_feasibility_last_result"]
                    for lbkey in (
                        "certified_vehicle_lower_bound",
                        "continuity_best_vehicle_lower_bound",
                    ):
                        val = self._safe_int(k_parallel_state.get(lbkey), None)
                        if val is not None:
                            old = self._safe_int(self.shared_memory.data.get(lbkey), 1) or 1
                            self.shared_memory.data[lbkey] = max(old, val)

        except Exception as exc:
            # Fallback seguro e explícito: se multiprocessing falhar, A/B rodam
            # sequencialmente; K continuará no bloco normal abaixo.
            print(f"   ⚠️ PROCESS-PARALLEL indisponível: {type(exc).__name__}: {exc}")
            try:
                ave_res = ave_agent.optimize(
                    baseline, improvement_plan, eval_cb,
                    audit_callback, ave_budget, seed
                )
            except Exception as e:
                ave_res = {
                    "status":"ERROR", "solution":copy.deepcopy(baseline),
                    "best_score":baseline_score, "improved":False,
                    "need_diversification":True,
                    "error":f"{type(e).__name__}: {e}",
                }
            try:
                bound_res = bound_agent.strengthen(
                    baseline, cert, None, bound_budget
                )
            except Exception as e:
                base_lb = self._safe_float(
                    self.shared_memory.data.get("continuity_baseline_valid_lower_bound"), None
                )
                bound_res = {
                    "status":"ERROR","base_lower_bound":base_lb,
                    "best_lower_bound":base_lb,"improved":False,
                    "error":f"{type(e).__name__}: {e}",
                }

        best = copy.deepcopy(ave_res.get("solution")) if isinstance(ave_res.get("solution"), dict) else copy.deepcopy(baseline)
        best_score = self._score(best)
        ga_info = {"used": False, "reason": None}

        # investigação estrutural K centralizada no novo certificador.
        # O AVE continua livre para reduzir frota heurísticamente, mas a pergunta
        # formal "existe solução <=K?" pertence ao KVehicleFeasibilityCertifier.
        remaining = budget_seconds - (time.monotonic() - started)
        k_info = {"used": False, "status": None}

        used_now = int(round(self._metrics(best).get("VEHICLES", 0.0) or 0.0))
        k_vehicle_certifier = (
            k_vehicle_certifier
            or self.shared_memory.data.get("_k_vehicle_feasibility_certifier")
        )
        optimality_certifier = (
            optimality_certifier
            or self.shared_memory.data.get("_optimality_certifier_agent")
        )

        # O melhor LB estrutural de veículos é persistente. Se ainda houver gap
        # de cardinalidade, investigamos K=UB-1. Nada é hard-coded em K=2.
        known_vehicle_lb = 1
        for _key in (
            "certified_vehicle_lower_bound",
            "continuity_best_vehicle_lower_bound",
            "vehicle_lower_bound",
        ):
            try:
                known_vehicle_lb = max(
                    known_vehicle_lb,
                    int(self.shared_memory.data.get(_key, 1) or 1),
                )
            except Exception:
                pass
        try:
            _br = self.shared_memory.data.get("bound_strengthening_last_result") or {}
            _gv = (_br.get("global_component_bounds") or {}).get("VEHICLES")
            if _gv is not None:
                known_vehicle_lb = max(known_vehicle_lb, int(float(_gv)))
        except Exception:
            pass

        target_k = used_now - 1 if used_now > 1 else None
        already = self.shared_memory.data.get("k_vehicle_feasibility_last_result") or {}
        already_target = self._safe_int(already.get("target_k"), -1)
        already_status = str(already.get("status", "") or "").upper()

        should_investigate_k = (
            k_vehicle_certifier is not None
            and target_k is not None
            and target_k >= known_vehicle_lb
            and remaining >= 45
            and not (
                already_target == target_k
                and already_status in {"PROVEN_INFEASIBLE", "FEASIBLE"}
            )
        )

        if should_investigate_k:
            # transferência adaptativa de orçamento para prova K.
            # Após UNKNOWN repetido no mesmo K, evita gastar a rodada repetindo
            # heurísticas equivalentes e entrega a maior parte do saldo ao Exact-K.
            _hist = self.shared_memory.data.get("k_vehicle_feasibility_history", []) or []
            _prior_unknown_k = sum(
                1 for _r in _hist
                if isinstance(_r, dict)
                and self._safe_int(_r.get("target_k"), -1) == int(target_k)
                and str(_r.get("status", "")).upper() == "UNKNOWN"
            )
            _k_fraction = 0.55 if _prior_unknown_k == 0 else (0.75 if _prior_unknown_k == 1 else 0.88)
            _reserve = 15.0 if _prior_unknown_k == 0 else 8.0
            k_budget = max(30, int(min(remaining * _k_fraction, max(30.0, remaining - _reserve))))
            k_info["prior_unknown_same_k"] = int(_prior_unknown_k)
            k_info["budget_transfer_fraction"] = float(_k_fraction)
            k_info["used"] = True
            k_info["target_k"] = int(target_k)
            k_info["vehicle_lb_before"] = int(known_vehicle_lb)
            k_info["budget_seconds"] = int(k_budget)

            print(
                f"   🧮 K-CERTIFIER | vehicleLB={known_vehicle_lb} | "
                f"vehicleUB={used_now} | testar K={target_k} | orçamento≈{k_budget}s"
            )

            try:
                if (
                    isinstance(k_parallel_res, dict)
                    and self._safe_int(k_parallel_res.get("target_k"), -1) == int(target_k)
                ):
                    kres = copy.deepcopy(k_parallel_res)
                    k_info["source"] = "PARALLEL_TRACK_C"
                    print(
                        f"   🧮 K-CERTIFIER V4 | resultado recebido do Track C paralelo | "
                        f"K={target_k} | status={kres.get('status')}"
                    )
                else:
                    kres = k_vehicle_certifier.certify(
                        target_k=int(target_k),
                        translated_problem=translated_json,
                        classification=classification,
                        incumbent=best,
                        ga_solver=ga_solver,
                        auditor=audit_callback,
                        time_limit_seconds=int(k_budget),
                        seed=int(seed + 32452843),
                        optimality_certifier=optimality_certifier,
                        exact_k_enabled=True,
                    )
                    k_info["source"] = "SEQUENTIAL_FALLBACK"
                k_info["result"] = copy.deepcopy(kres)
                k_info["status"] = (
                    kres.get("status") if isinstance(kres, dict) else "INVALID"
                )

                if isinstance(kres, dict):
                    if (
                        kres.get("status") == "PROVEN_INFEASIBLE"
                        and kres.get("global_valid") is True
                    ):
                        new_vlb = self._safe_int(
                            kres.get("proven_vehicle_lower_bound"),
                            int(target_k) + 1,
                        )
                        self.shared_memory.data["certified_vehicle_lower_bound"] = max(
                            self._safe_int(
                                self.shared_memory.data.get(
                                    "certified_vehicle_lower_bound"
                                ),
                                1,
                            ),
                            int(new_vlb),
                        )
                        self.shared_memory.data[
                            "continuity_best_vehicle_lower_bound"
                        ] = max(
                            self._safe_int(
                                self.shared_memory.data.get(
                                    "continuity_best_vehicle_lower_bound"
                                ),
                                1,
                            ),
                            int(new_vlb),
                        )
                        print(
                            f"   🔒 EXACT-K PROOF | K={target_k} impossível → "
                            f"vehicleLB≥{new_vlb}"
                        )

                        # Reexecuta Bound imediatamente para transformar a prova
                        # estrutural em COST/canonical LB fortalecido.
                        remaining_after_k = budget_seconds - (
                            time.monotonic() - started
                        )
                        if remaining_after_k >= 3:
                            try:
                                bound_res = bound_agent.strengthen(
                                    best,
                                    cert,
                                    None,
                                    min(15, max(3, int(remaining_after_k))),
                                )
                                k_info["bound_recomputed_after_proof"] = True
                            except Exception as exc:
                                k_info["bound_recompute_error"] = (
                                    f"{type(exc).__name__}: {exc}"
                                )

                    elif kres.get("status") == "FEASIBLE":
                        kcand = kres.get("candidate")
                        if isinstance(kcand, dict):
                            # O próprio K-Certifier já exigiu Auditor.
                            kscore = self._score(kcand)
                            k_info["candidate_score"] = kscore
                            if kscore < best_score - 1e-10:
                                best = copy.deepcopy(kcand)
                                best_score = float(kscore)
                                k_info["promoted"] = True
                                print(
                                    f"   ✅ K-CERTIFIER encontrou solução auditada "
                                    f"≤{target_k} | score={best_score:.12f}"
                                )
            except Exception as exc:
                k_info["status"] = "ERROR"
                k_info["error"] = f"{type(exc).__name__}: {exc}"

            remaining = budget_seconds - (time.monotonic() - started)
        elif target_k is not None:
            k_info.update({
                "target_k": int(target_k),
                "vehicle_lb_before": int(known_vehicle_lb),
                "reason": (
                    "ALREADY_RESOLVED"
                    if already_target == target_k
                    and already_status in {"PROVEN_INFEASIBLE", "FEASIBLE"}
                    else "INSUFFICIENT_BUDGET_OR_NO_CARDINALITY_GAP"
                ),
            })

        # GA genérico entra apenas como mecanismo de escape de bacia.
        if (not ave_res.get("improved", False)
                and ave_res.get("need_diversification", False)
                and remaining >= 20):
            ga_info["used"] = True
            ga_info["reason"] = "AVE_LOCAL_NEIGHBORHOODS_EXHAUSTED"
            ga_budget = max(10, int(min(remaining * 0.55, remaining - 5)))
            directive = {
                "type": "DIVERSIFICATION",
                "purpose": "ESCAPE_FROM_AVE_LOCAL_BASIN",
                "avoid_repeating_local_neighborhoods": True,
                "seed": int(seed + 104729),
            }
            self.shared_memory.data["continuity_search_directive"] = copy.deepcopy(directive)
            self.shared_memory.data["continuity_seed_solution"] = copy.deepcopy(best)
            payload = copy.deepcopy(translated_json)
            original_string = isinstance(payload, str)
            if original_string:
                try: payload = json.loads(payload)
                except Exception: payload = {}
            if not isinstance(payload, dict): payload = {}
            payload["_continuity_directive"] = copy.deepcopy(directive)
            payload["_continuity_seed_solution"] = copy.deepcopy(best)
            payload["continuity_mode"] = True
            if original_string: payload = json.dumps(payload, ensure_ascii=False)

            print(f"   🌐 AVE esgotou vizinhanças locais → GA DIVERSIFICATION | orçamento≈{ga_budget}s")
            try:
                diversified = ga_solver.solve_problem(
                    classification, payload, time_limit_seconds=ga_budget,
                    seed=int(seed + 104729), continuity_mode=True
                )
                ga_info["status"] = diversified.get("status") if isinstance(diversified, dict) else "INVALID"
                if isinstance(diversified, dict) and diversified.get("status") == "SUCCESS":
                    diag, rec = self._audit_candidate(audit_callback, diversified, "GA_DIVERSIFICATION_GATE")
                    if isinstance(rec, dict) and rec.get("solucao_aprovada", False):
                        report = (self.shared_memory.data.get("independent_feasibility_report_auditor")
                                  or self.shared_memory.data.get("last_independent_feasibility_report") or {})
                        norm = report.get("normalized_solution") if isinstance(report, dict) else None
                        diversified_seed = copy.deepcopy(norm) if isinstance(norm, dict) else copy.deepcopy(diversified)

                        # Structural Diversity Gate. Score diferente não basta:
                        # o GA só abre uma nova bacia para o AVE se a estrutura realmente mudou.
                        gate = ave_res.get("ga_diversity_gate", {}) if isinstance(ave_res, dict) else {}
                        min_dist = self._safe_float(gate.get("minimum_structural_distance"), 0.08) or 0.08
                        try:
                            structural_distance = float(ave_agent.structural_distance(best, diversified_seed))
                        except Exception:
                            structural_distance = 0.0
                        ga_info["structural_distance"] = structural_distance
                        ga_info["minimum_structural_distance"] = float(min_dist)
                        ga_info["structurally_different"] = bool(structural_distance >= min_dist)

                        if structural_distance < min_dist:
                            ga_info["status"] = "GA_ESCAPE_FAILED_SAME_BASIN"
                            ga_info["reason"] = "STRUCTURAL_DIVERSITY_BELOW_THRESHOLD"
                            print(
                                f"   🧱 GA permaneceu na mesma bacia | distância estrutural="
                                f"{structural_distance:.4f} < {min_dist:.4f} → AVE não será repetido."
                            )
                        else:
                            remaining2 = budget_seconds - (time.monotonic() - started)
                            if remaining2 >= 5:
                                print(
                                    f"   🔬 GA encontrou região estruturalmente nova "
                                    f"(distância={structural_distance:.4f}) → AVE volta a intensificar."
                                )
                                ave2 = ave_agent.optimize(diversified_seed, improvement_plan, eval_cb,
                                                          audit_callback, remaining2, seed + 104729)
                                ga_info["ave_after_diversification"] = {
                                    k:v for k,v in ave2.items() if k not in ("solution","search_memory")
                                }
                                candidate2 = ave2.get("solution") if isinstance(ave2, dict) else None
                                if isinstance(candidate2, dict) and self._score(candidate2) < best_score - 1e-10:
                                    best = copy.deepcopy(candidate2); best_score = self._score(best)

                            # A solução diversificada pode ser promovida diretamente apenas se
                            # for auditada, melhor e estruturalmente distinta.
                            dscore = self._score(diversified_seed)
                            if dscore < best_score - 1e-10:
                                best = copy.deepcopy(diversified_seed); best_score = dscore
            except Exception as exc:
                ga_info["status"] = "ERROR"
                ga_info["error"] = f"{type(exc).__name__}: {exc}"

        # Evidências exatas/globais produzidas pelo AVE podem alimentar Track B.
        evidence = []
        # AVE  publica evidências estruturadas diretamente;  permanece compatível.
        # AVE  publica em bound_evidence; "evidence" é mantido por compatibilidade.
        for key in ("bound_evidence", "evidence"):
            for e in (ave_res.get(key, []) or []):
                if isinstance(e, dict):
                    evidence.append(copy.deepcopy(e))
        for item in (ave_res.get("search_memory", []) or []):
            if isinstance(item, dict):
                move = item.get("move", {}) or {}
                meta = move.get("metadata", {}) if isinstance(move, dict) else {}
                if isinstance(meta, dict) and str(meta.get("proof_status", "")).upper() in {"PROVEN","EXACT","CERTIFIED"}:
                    evidence.append(meta)
        if evidence:
            try:
                bound_res2 = bound_agent.strengthen(best, cert, evidence, min(10, max(5, budget_seconds//10)))
                if self._safe_float(bound_res2.get("best_lower_bound"), -1) > self._safe_float(bound_res.get("best_lower_bound"), -1):
                    bound_res = bound_res2
            except Exception:
                pass

        base_lb = self._safe_float(bound_res.get("base_lower_bound"), self._safe_float(self.shared_memory.data.get("continuity_baseline_valid_lower_bound"), None))
        best_lb = self._safe_float(bound_res.get("best_lower_bound"), base_lb)
        ub_improved = best_score < baseline_score - 1e-10
        lb_improved = best_lb is not None and base_lb is not None and best_lb > base_lb + 1e-12
        status = "UB_AND_LB_IMPROVED" if ub_improved and lb_improved else "UB_IMPROVED" if ub_improved else "LB_IMPROVED" if lb_improved else "NO_IMPROVEMENT"

        track_a = {"track":"A_PRIMAL_AVE", "status":ave_res.get("status"),
                   "accepted":bool(ub_improved), "best_ub":float(best_score),
                   "ave":{k:v for k,v in ave_res.items() if k not in ("solution","search_memory")},
                   "ga_diversification":ga_info}
        track_b = {"track":"B_DUAL_BOUND_STRENGTHENING", "status":bound_res.get("status"),
                   "baseline_valid_lb":base_lb, "best_valid_lb":best_lb,
                   "strengthened":bool(lb_improved), "details":copy.deepcopy(bound_res)}
        history_entry = {"seed":int(seed), "budget_seconds":int(budget_seconds),
                         "elapsed_seconds":float(time.monotonic()-started), "status":status,
                         "baseline_ub":float(baseline_score), "best_ub":float(best_score),
                         "baseline_lb":base_lb, "best_lb":best_lb,
                         "ub_improved":bool(ub_improved), "lb_improved":bool(lb_improved),
                         "ga_used_only_for_diversification":bool(ga_info.get("used"))}
        self.shared_memory.data["continuity_candidate"] = copy.deepcopy(best)
        self.shared_memory.data["continuity_track_a_result"] = copy.deepcopy(track_a)
        self.shared_memory.data["continuity_track_b_result"] = copy.deepcopy(track_b)
        self.shared_memory.data["continuity_best_valid_lower_bound"] = best_lb
        self.shared_memory.data["continuity_dual_track_summary"] = copy.deepcopy(history_entry)
        hist = self.shared_memory.data.get("continuity_history", [])
        self.shared_memory.data["continuity_history"] = (hist if isinstance(hist,list) else []) + [copy.deepcopy(history_entry)]
        return {"status":status, "best_candidate":best, "best_ub":float(best_score),
                "best_valid_lb":best_lb, "history":[history_entry],
                "elapsed_seconds":float(time.monotonic()-started),
                "audited_improvements":1 if ub_improved else 0,
                "bound_improvements":1 if lb_improved else 0,
                "track_a":track_a, "track_b":track_b,
                "dual_track_status":status, "improvement_plan":copy.deepcopy(improvement_plan)}
````

### **NEIGHBORHOODOPTIMIZATIONAGENT — REORGANIZAÇÃO DE VIAGENS**

Este agente seleciona uma parte do plano para reorganizar, mantendo as demais viagens sob controle. Uma “região” aqui é um conjunto de viagens escolhido para esse trabalho.





Tabela com células delimitadas: Responsabilidade / Explicação





**Um exemplo:** Pode liberar três viagens para trocar clientes entre elas. Mesmo que encontre o melhor arranjo dessa parte, isso não prova que o plano inteiro seja ótimo.

**Por que há duas classes dentro desta célula?** A classe interna reúne métodos de base; a classe pública os utiliza e acrescenta métodos. Aqui elas formam a implementação de um mesmo papel na equipe, não dois integrantes independentes.



````python
class _NeighborhoodOptimizationAgentV1:
    """
    AVE  — Agente de Otimização por Vizinhança Exata / Matheuristic.

    Papel:
      - protagonista do Track A durante a Continuidade;
      - parte de um incumbent auditado;
      - executa primeiro movimentos dirigidos e determinísticos;
      - depois tenta redução de frota por destroy + repair;
      - usa callbacks do sistema para avaliação/auditoria, evitando duplicar
        a função objetivo canônica e a lógica independente do Auditor;
      - solicita GA apenas como DIVERSIFICAÇÃO quando a vizinhança se esgota.

    Contrato de integração:
      optimize(
          incumbent,
          improvement_plan=None,
          evaluator_callback=None,
          audit_callback=None,
          budget_seconds=120,
          seed=0,
      )

    evaluator_callback(candidate) pode retornar:
      - float score; ou
      - dict com canonical_score/score/objective_value.

    audit_callback(candidate, stage) pode retornar:
      - bool; ou
      - (diagnostic, recommendation); ou
      - dict contendo solucao_aprovada / feasible / approved.

    Segurança:
      - nunca substitui o incumbent sem score estritamente menor;
      - NOT_FOUND nunca significa INFEASIBLE;
      - mantém memória das vizinhanças testadas;
      - não chama GA internamente: quando esgotado, retorna
        NEED_DIVERSIFICATION=True para o Continuity decidir.
    """

    VERSION = "AVE_V1_1_AUDIT_NORMALIZED_NEIGHBORHOOD"

    def __init__(self, shared_memory):
        import copy
        self.shared_memory = shared_memory
        self.memory = []
        self._tested = set()
        self._copy = copy.deepcopy

    # --------------------------- utilidades ----------------------------

    @staticmethod
    def _f(x, default=0.0):
        try:
            return float(x)
        except Exception:
            return default

    @staticmethod
    def _successful(sol):
        return isinstance(sol, dict) and str(sol.get("status", "")).upper() in {
            "SUCCESS", "FEASIBLE", "OPTIMAL", "FEASIBLE_WITH_CERTIFIED_GAP"
        }

    def _score_from_solution(self, sol):
        if not isinstance(sol, dict):
            return float("inf")
        for k in ("canonical_score", "objective_value", "fitness", "score"):
            if sol.get(k) is not None:
                try:
                    return float(sol[k])
                except Exception:
                    pass
        summary = sol.get("summary", {}) or {}
        for k in ("canonical_score", "objective_value", "score"):
            if summary.get(k) is not None:
                try:
                    return float(summary[k])
                except Exception:
                    pass
        return float("inf")

    def _evaluate(self, candidate, evaluator_callback):
        if evaluator_callback is None:
            return self._score_from_solution(candidate)
        out = evaluator_callback(candidate)
        if isinstance(out, (int, float)):
            return float(out)
        if isinstance(out, dict):
            return self._score_from_solution(out)
        return float("inf")

    def _audit(self, candidate, audit_callback, stage):
        if audit_callback is None:
            return True, {}
        try:
            out = audit_callback(candidate, stage)
        except TypeError:
            out = audit_callback(candidate)
        if isinstance(out, bool):
            return out, {}
        if isinstance(out, tuple) and len(out) >= 2:
            rec = out[1] if isinstance(out[1], dict) else {}
            ok = bool(rec.get("solucao_aprovada",
                              rec.get("feasible",
                              rec.get("approved", False))))
            return ok, rec
        if isinstance(out, dict):
            ok = bool(out.get("solucao_aprovada",
                              out.get("feasible",
                              out.get("approved", False))))
            return ok, out
        return False, {}

    @staticmethod
    def _routes(sol):
        if not isinstance(sol, dict):
            return []
        routes = sol.get("routes")
        if isinstance(routes, list):
            return routes
        ns = sol.get("normalized_solution", {}) or {}
        routes = ns.get("routes")
        return routes if isinstance(routes, list) else []

    @staticmethod
    def _path(route):
        p = route.get("path", route.get("route", []))
        return list(p) if isinstance(p, (list, tuple)) else []

    @staticmethod
    def _vehicle(route):
        return route.get("physical_vehicle_id", route.get("vehicle_id"))

    def _signature(self, kind, payload):
        return (str(kind), repr(payload))

    def _publish_memory(self):
        try:
            self.shared_memory.data["ave_search_memory"] = self._copy(self.memory)
        except Exception:
            pass

    def _candidate_with_routes(self, incumbent, routes, tag, meta=None):
        cand = self._copy(incumbent)
        cand["routes"] = self._copy(routes)
        cand["status"] = "SUCCESS"
        cand["ave_candidate"] = True
        cand["ave_version"] = self.VERSION
        cand["ave_move"] = tag
        cand["ave_metadata"] = self._copy(meta or {})
        # Os totais precisam ser recalculados após mudança estrutural.
        # O callback canônico/Auditor deve reconstruí-los.
        for k in ("canonical_score", "objective_value", "fitness", "score"):
            cand.pop(k, None)
        return cand

    def _consider(self, candidate, best, best_score, evaluator_callback,
                  audit_callback, stage, move_info):
        # primeiro reconstrói/audita operacionalmente a candidata.
        # Isto evita comparar métricas desatualizadas depois de alterar apenas o path.
        approved, audit_info = self._audit(candidate, audit_callback, stage)
        normalized = None
        try:
            report = (self.shared_memory.data.get("independent_feasibility_report_auditor")
                      or self.shared_memory.data.get("last_independent_feasibility_report")
                      or {})
            normalized = report.get("normalized_solution") if isinstance(report, dict) else None
        except Exception:
            normalized = None
        evaluated = normalized if isinstance(normalized, dict) else candidate
        score = self._evaluate(evaluated, evaluator_callback) if approved else float("inf")
        better = approved and score < best_score - 1e-10
        record = {
            "stage": stage, "move": self._copy(move_info),
            "score": None if score == float("inf") else float(score),
            "better_than_incumbent": bool(better),
            "audited": True, "approved": bool(approved),
        }
        self.memory.append(record)
        if better:
            evaluated = self._copy(evaluated)
            evaluated["status"] = "SUCCESS"
            evaluated["canonical_score"] = float(score)
            return evaluated, float(score), True
        return best, best_score, False

    # -------------------- vizinhança determinística --------------------

    def _two_opt_candidates(self, incumbent, structural, limit=40):
        routes = self._routes(incumbent)
        hints = []
        if isinstance(structural, dict):
            hints = structural.get("intra_trip_2opt_candidates", []) or []
        hinted = []
        for h in hints:
            if not isinstance(h, dict):
                continue
            hinted.append((h.get("vehicle_id"), h.get("trip_id")))

        order = []
        for idx, r in enumerate(routes):
            key = (self._vehicle(r), r.get("trip_id"))
            priority = 0 if key in hinted else 1
            order.append((priority, idx, r))
        order.sort(key=lambda x: x[0])

        produced = 0
        for _, ridx, route in order:
            p = self._path(route)
            if len(p) < 5:
                continue
            # depósito nas extremidades: inverte somente clientes internos.
            for i in range(1, len(p) - 2):
                for j in range(i + 1, len(p) - 1):
                    sig = self._signature("2OPT", (ridx, i, j, tuple(p)))
                    if sig in self._tested:
                        continue
                    self._tested.add(sig)
                    nroutes = self._copy(routes)
                    np = p[:i] + list(reversed(p[i:j + 1])) + p[j + 1:]
                    nroutes[ridx]["path"] = np
                    yield self._candidate_with_routes(
                        incumbent, nroutes, "2OPT",
                        {"route_index": ridx, "vehicle_id": self._vehicle(route),
                         "trip_id": route.get("trip_id"), "i": i, "j": j}
                    )
                    produced += 1
                    if produced >= limit:
                        return

    def _relocate_candidates(self, incumbent, limit=80):
        routes = self._routes(incumbent)
        produced = 0
        for a, ra in enumerate(routes):
            pa = self._path(ra)
            if len(pa) <= 3:
                continue
            for pos in range(1, len(pa) - 1):
                node = pa[pos]
                for b, rb in enumerate(routes):
                    if a == b:
                        continue
                    pb = self._path(rb)
                    for ins in range(1, max(2, len(pb))):
                        sig = self._signature("RELOCATE", (a, pos, b, ins, node))
                        if sig in self._tested:
                            continue
                        self._tested.add(sig)
                        nr = self._copy(routes)
                        npa = self._path(nr[a]); npb = self._path(nr[b])
                        moved = npa.pop(pos)
                        if ins >= len(npb):
                            ins = len(npb) - 1
                        npb.insert(ins, moved)
                        nr[a]["path"] = npa
                        nr[b]["path"] = npb
                        yield self._candidate_with_routes(
                            incumbent, nr, "RELOCATE",
                            {"node": node, "from_route": a, "to_route": b,
                             "insert_position": ins}
                        )
                        produced += 1
                        if produced >= limit:
                            return

    def _swap_candidates(self, incumbent, limit=60):
        routes = self._routes(incumbent)
        produced = 0
        for a in range(len(routes)):
            pa = self._path(routes[a])
            for b in range(a + 1, len(routes)):
                pb = self._path(routes[b])
                for ia in range(1, max(1, len(pa) - 1)):
                    if ia >= len(pa) - 1: break
                    for ib in range(1, max(1, len(pb) - 1)):
                        if ib >= len(pb) - 1: break
                        sig = self._signature("SWAP", (a, ia, b, ib, pa[ia], pb[ib]))
                        if sig in self._tested:
                            continue
                        self._tested.add(sig)
                        nr = self._copy(routes)
                        npa = self._path(nr[a]); npb = self._path(nr[b])
                        npa[ia], npb[ib] = npb[ib], npa[ia]
                        nr[a]["path"] = npa; nr[b]["path"] = npb
                        yield self._candidate_with_routes(
                            incumbent, nr, "SWAP",
                            {"route_a": a, "route_b": b,
                             "node_a": pa[ia], "node_b": pb[ib]}
                        )
                        produced += 1
                        if produced >= limit:
                            return

    # ------------------------- redução de frota -------------------------

    def _fleet_elimination_candidates(self, incumbent, structural, limit=120):
        """
        Destroy + repair dirigido.

        Remove todas as viagens de um veículo e tenta reinserir seus clientes
        nas viagens dos veículos restantes. A enumeração é deliberadamente
        limitada por budget/limit; portanto falha = NOT_FOUND, nunca INFEASIBLE.

        A factibilidade real é decidida pelo evaluator/Auditor do sistema.
        """
        routes = self._routes(incumbent)
        vehicles = []
        for r in routes:
            v = self._vehicle(r)
            if v is not None and v not in vehicles:
                vehicles.append(v)

        order = []
        fr = (structural or {}).get("fleet_reduction_opportunity", {}) if isinstance(structural, dict) else {}
        for x in fr.get("elimination_order", []) or []:
            v = x.get("vehicle_id") if isinstance(x, dict) else x
            if v in vehicles and v not in order:
                order.append(v)
        order += [v for v in vehicles if v not in order]

        produced = 0
        for victim in order:
            kept = [self._copy(r) for r in routes if self._vehicle(r) != victim]
            removed = [r for r in routes if self._vehicle(r) == victim]
            pool = []
            for r in removed:
                p = self._path(r)
                pool.extend([n for n in p if n not in (0, "0", None)])
            if not pool or not kept:
                continue

            # Repair determinístico por inserção: cada cliente é inserido na
            # menor viagem corrente; gera variantes rotacionando a ordem do pool.
            variants = min(max(1, len(pool)), 12)
            for shift in range(variants):
                ordered = pool[shift:] + pool[:shift]
                nr = self._copy(kept)
                for node in ordered:
                    target = min(
                        range(len(nr)),
                        key=lambda z: max(0, len(self._path(nr[z])) - 2)
                    )
                    p = self._path(nr[target])
                    if not p:
                        p = [0, 0]
                    p.insert(max(1, len(p) - 1), node)
                    nr[target]["path"] = p

                sig = self._signature("FLEET_ELIMINATION", (victim, shift, tuple(ordered)))
                if sig in self._tested:
                    continue
                self._tested.add(sig)
                yield self._candidate_with_routes(
                    incumbent, nr, "FLEET_ELIMINATION_EXACT_REPAIR_SEED",
                    {"eliminated_vehicle": victim,
                     "target_vehicle_count": max(1, len(vehicles) - 1),
                     "destroyed_clients": list(pool),
                     "repair_variant": shift,
                     "proof_status": "HEURISTIC_REPAIR_REQUIRES_AUDIT"}
                )
                produced += 1
                if produced >= limit:
                    return

    # ----------------------------- API --------------------------------

    def optimize(self, incumbent, improvement_plan=None,
                 evaluator_callback=None, audit_callback=None,
                 budget_seconds=120, seed=0):
        import time
        started = time.monotonic()
        deadline = started + max(5.0, float(budget_seconds))
        self.memory = []
        self._tested = set()

        best = self._copy(incumbent)
        best_score = self._evaluate(best, evaluator_callback)
        baseline_score = best_score

        plan = self._copy(improvement_plan or {})
        structural = plan.get("structural_incumbent_analysis", {}) or {}
        strategies = [str(s.get("type", "")).upper()
                      for s in (plan.get("strategies", []) or [])
                      if isinstance(s, dict)]

        print(f"🧩 AVE {self.VERSION} | incumbent={best_score:.12f} | orçamento={budget_seconds}s")
        print(f"   estratégias={strategies or ['GENERIC']}")

        improvements = 0

        def run_stream(name, stream):
            nonlocal best, best_score, improvements
            tested = 0
            for cand in stream:
                if time.monotonic() >= deadline:
                    break
                tested += 1
                best2, score2, accepted = self._consider(
                    cand, best, best_score, evaluator_callback, audit_callback,
                    f"AVE_{name}", {"type": name, "metadata": cand.get("ave_metadata", {})}
                )
                if accepted:
                    best, best_score = best2, score2
                    improvements += 1
                    print(f"   ✅ {name}: UB {baseline_score:.12f} → {best_score:.12f}")
            print(f"   ↳ {name}: {tested} vizinhanças avaliadas")

        # Ordem: movimentos baratos -> redução estrutural -> inter-trip.
        if time.monotonic() < deadline:
            run_stream("DIRECT_2OPT", self._two_opt_candidates(best, structural))
        if ("FLEET_REDUCTION" in strategies or not strategies) and time.monotonic() < deadline:
            run_stream("FLEET_ELIMINATION", self._fleet_elimination_candidates(best, structural))
        if time.monotonic() < deadline:
            run_stream("RELOCATE", self._relocate_candidates(best))
        if time.monotonic() < deadline:
            run_stream("SWAP", self._swap_candidates(best))

        self._publish_memory()
        improved = best_score < baseline_score - 1e-10
        exhausted = time.monotonic() < deadline and not improved

        result = {
            "status": "IMPROVED" if improved else "NOT_FOUND",
            "version": self.VERSION,
            "solution": self._copy(best),
            "baseline_score": float(baseline_score),
            "best_score": float(best_score),
            "improved": bool(improved),
            "improvements_accepted": int(improvements),
            "tested_neighborhoods": len(self._tested),
            "search_memory": self._copy(self.memory),
            "need_diversification": bool(exhausted),
            "diversification_reason": (
                "LOCAL_NEIGHBORHOODS_EXHAUSTED_WITHOUT_IMPROVEMENT"
                if exhausted else None
            ),
            "proof": {
                "global_infeasibility_proven": False,
                "note": "NOT_FOUND não implica INFEASIBLE."
            },
            "elapsed_seconds": time.monotonic() - started,
        }
        try:
            self.shared_memory.data["ave_last_result"] = self._copy(result)
        except Exception:
            pass
        return result


class NeighborhoodOptimizationAgent(_NeighborhoodOptimizationAgentV1):
    """
    AVE  — Adaptive Memory + Ejection Chains + Fleet Reduction Restricted Exact Repair.
    Responsabilidades:
      1) memória persistente por assinatura estrutural do incumbent;
      2) não repete vizinhanças declaradas esgotadas para o MESMO incumbent;
      3) adiciona EJECTION_CHAIN de dois movimentos coordenados;
      4) redução de frota 3→2 (ou n→n-1) por níveis progressivos;
      5) restricted exact repair para pools pequenos, com enumeração sistemática;
      6) diagnóstico explícito por veículo eliminado e nível de reparo;
      7) evidências locais são publicadas, mas nunca promovidas a prova global;
      8) fornece assinatura estrutural para o Orquestrador aplicar o
         Structural Diversity Gate ao retorno do GA.

    IMPORTANTE:
      - "restricted exact" significa exato SOMENTE dentro da vizinhança enumerada.
      - falha de reparo continua sendo NOT_FOUND, nunca INFEASIBLE global.
      - o Auditor continua soberano sobre factibilidade.
    """
    VERSION = "AVE_V2_2_DEDICATED_K_FLEET_ARM"

    def __init__(self, shared_memory):
        super().__init__(shared_memory)
        self._load_persistent_state()

    # ---------------------- memória adaptativa -----------------------

    def _load_persistent_state(self):
        data = getattr(self.shared_memory, "data", {}) or {}
        state = data.get("ave_v2_persistent_state", {}) or {}
        self._persistent = self._copy(state) if isinstance(state, dict) else {}
        self._persistent.setdefault("solutions", {})
        self._persistent.setdefault("evidence", [])
        self._persistent.setdefault("version", self.VERSION)

    def _save_persistent_state(self):
        try:
            self.shared_memory.data["ave_v2_persistent_state"] = self._copy(self._persistent)
        except Exception:
            pass

    def structural_signature(self, sol):
        """Assinatura independente de métricas; representa a bacia/estrutura de rotas."""
        normalized = []
        for r in self._routes(sol):
            p = tuple(str(x) for x in self._path(r))
            normalized.append((
                str(self._vehicle(r)),
                str(r.get("trip_id", "")),
                p,
            ))
        normalized.sort(key=lambda x: (x[0], x[1], x[2]))
        return repr(tuple(normalized))

    def structural_distance(self, a, b):
        """
        Distância estrutural [0,1] baseada em:
          - arestas dirigidas das rotas;
          - atribuição cliente→veículo.
        Útil para o gate do GA.
        """
        def features(sol):
            edges, assign = set(), set()
            for r in self._routes(sol):
                v = str(self._vehicle(r))
                p = [str(x) for x in self._path(r)]
                for i in range(len(p) - 1):
                    edges.add((p[i], p[i + 1], v))
                for n in p:
                    if n not in ("0", "0.0", "None"):
                        assign.add((n, v))
            return edges, assign
        ea, aa = features(a); eb, ab = features(b)
        union = len(ea | eb) + len(aa | ab)
        if union == 0:
            return 0.0
        inter = len(ea & eb) + len(aa & ab)
        return max(0.0, min(1.0, 1.0 - inter / union))

    def _state_for(self, sig):
        s = self._persistent["solutions"].setdefault(sig, {})
        s.setdefault("exhausted", {})
        s.setdefault("fleet", {})
        s.setdefault("visits", 0)
        s["visits"] += 1
        return s

    def _is_exhausted(self, sig, neighborhood):
        return bool(self._state_for(sig)["exhausted"].get(neighborhood, False))

    def _mark_exhausted(self, sig, neighborhood, details=None):
        st = self._state_for(sig)
        st["exhausted"][neighborhood] = True
        if details is not None:
            st.setdefault("exhaustion_details", {})[neighborhood] = self._copy(details)

    # ---------------------- ejection chains --------------------------

    def _ejection_chain_candidates(self, incumbent, limit=80):
        """
        Cadeia de dois movimentos coordenados:
          A: rota a -> rota b
          B: rota b -> rota c (ou a)
        Permite atravessar barreiras que relocate/swap isolados não atravessam.
        A factibilidade é sempre decidida pelo Auditor.
        """
        routes = self._routes(incumbent)
        produced = 0
        n = len(routes)
        for a in range(n):
            pa = self._path(routes[a])
            if len(pa) <= 3:
                continue
            for ia in range(1, len(pa) - 1):
                node_a = pa[ia]
                for b in range(n):
                    if b == a:
                        continue
                    pb = self._path(routes[b])
                    if len(pb) <= 2:
                        continue
                    # poucas posições representativas para controlar explosão
                    insert_positions = sorted(set([1, max(1, len(pb)//2), max(1, len(pb)-1)]))
                    for ins_a in insert_positions:
                        for ib in range(1, len(pb) - 1):
                            node_b = pb[ib]
                            for c in range(n):
                                if c == b:
                                    continue
                                pc = self._path(routes[c])
                                ins_b = max(1, len(pc) - 1)
                                sig = self._signature(
                                    "EJECTION_CHAIN",
                                    (a, ia, b, ins_a, ib, c, node_a, node_b)
                                )
                                if sig in self._tested:
                                    continue
                                self._tested.add(sig)
                                nr = self._copy(routes)
                                npa = self._path(nr[a])
                                npb = self._path(nr[b])
                                npc = self._path(nr[c])

                                moved_a = npa.pop(ia)
                                # localizar node_b novamente, pois b pode receber A antes da ejeção
                                try:
                                    idx_b = npb.index(node_b)
                                except ValueError:
                                    continue
                                moved_b = npb.pop(idx_b)
                                pos_a = min(max(1, ins_a), max(1, len(npb)-1))
                                npb.insert(pos_a, moved_a)
                                pos_b = min(max(1, ins_b), max(1, len(npc)-1))
                                npc.insert(pos_b, moved_b)

                                nr[a]["path"] = npa
                                nr[b]["path"] = npb
                                nr[c]["path"] = npc
                                yield self._candidate_with_routes(
                                    incumbent, nr, "EJECTION_CHAIN",
                                    {
                                        "chain": [
                                            {"node": node_a, "from": a, "to": b},
                                            {"node": node_b, "from": b, "to": c},
                                        ],
                                        "proof_status": "HEURISTIC_REQUIRES_AUDIT",
                                    }
                                )
                                produced += 1
                                if produced >= limit:
                                    return

    # ------------------ fleet reduction progressiva -----------------

    def _fleet_order(self, incumbent, structural):
        vehicles = []
        for r in self._routes(incumbent):
            v = self._vehicle(r)
            if v is not None and v not in vehicles:
                vehicles.append(v)
        order = []
        fr = (structural or {}).get("fleet_reduction_opportunity", {}) if isinstance(structural, dict) else {}
        for x in fr.get("elimination_order", []) or []:
            v = x.get("vehicle_id") if isinstance(x, dict) else x
            if v in vehicles and v not in order:
                order.append(v)
        return order + [v for v in vehicles if v not in order]

    def _victim_pool(self, incumbent, victim):
        routes = self._routes(incumbent)
        kept = [self._copy(r) for r in routes if self._vehicle(r) != victim]
        removed = [self._copy(r) for r in routes if self._vehicle(r) == victim]
        pool = []
        for r in removed:
            pool.extend([n for n in self._path(r) if n not in (0, "0", None)])
        return kept, removed, pool

    @staticmethod
    def _insert_node(routes, route_idx, pos, node):
        nr = [dict(r) for r in routes]
        # paths precisam de cópia própria
        for i, r in enumerate(routes):
            nr[i] = dict(r)
            p = r.get("path", r.get("route", []))
            nr[i]["path"] = list(p) if isinstance(p, (list, tuple)) else []
        p = nr[route_idx]["path"]
        if not p:
            p[:] = [0, 0]
        pos = min(max(1, pos), max(1, len(p)-1))
        p.insert(pos, node)
        return nr

    def _fleet_repair_level1(self, incumbent, victim, limit=18):
        """Seeds baratos: balanceamento por tamanho de viagem + rotações do pool."""
        kept, _, pool = self._victim_pool(incumbent, victim)
        if not kept or not pool:
            return
        variants = min(max(1, len(pool)), limit)
        for shift in range(variants):
            ordered = pool[shift:] + pool[:shift]
            nr = self._copy(kept)
            for node in ordered:
                target = min(range(len(nr)), key=lambda z: max(0, len(self._path(nr[z])) - 2))
                p = self._path(nr[target]) or [0, 0]
                p.insert(max(1, len(p)-1), node)
                nr[target]["path"] = p
            yield self._candidate_with_routes(
                incumbent, nr, "FLEET_REDUCTION_L1_BALANCED",
                {"eliminated_vehicle": victim, "level": 1,
                 "destroyed_clients": list(pool),
                 "target_vehicle_count": len({self._vehicle(r) for r in kept}),
                 "proof_status": "HEURISTIC_REPAIR_REQUIRES_AUDIT"}
            )

    def _fleet_repair_level2(self, incumbent, victim, limit=36):
        """
        Inserção incremental com diferentes ordens e posições representativas.
        Mais profunda que L1, ainda heurística.
        """
        kept, _, pool = self._victim_pool(incumbent, victim)
        if not kept or not pool:
            return
        orders = [
            list(pool),
            list(reversed(pool)),
            pool[::2] + pool[1::2],
            pool[1::2] + pool[::2],
        ]
        produced = 0
        for ordered in orders:
            # três políticas de rota: menor, round-robin, maior
            for policy in ("SMALLEST", "ROUND_ROBIN", "LARGEST"):
                nr = self._copy(kept)
                for k, node in enumerate(ordered):
                    sizes = [max(0, len(self._path(r))-2) for r in nr]
                    if policy == "SMALLEST":
                        target = min(range(len(nr)), key=lambda z: sizes[z])
                    elif policy == "LARGEST":
                        target = max(range(len(nr)), key=lambda z: sizes[z])
                    else:
                        target = k % len(nr)
                    p = self._path(nr[target]) or [0, 0]
                    # alterna início/fim/meio
                    poss = [1, max(1, len(p)-1), max(1, len(p)//2)]
                    pos = poss[k % len(poss)]
                    p.insert(min(pos, max(1, len(p)-1)), node)
                    nr[target]["path"] = p
                yield self._candidate_with_routes(
                    incumbent, nr, "FLEET_REDUCTION_L2_REPAIR",
                    {"eliminated_vehicle": victim, "level": 2,
                     "policy": policy, "destroyed_clients": list(pool),
                     "target_vehicle_count": len({self._vehicle(r) for r in kept}),
                     "proof_status": "HEURISTIC_REPAIR_REQUIRES_AUDIT"}
                )
                produced += 1
                if produced >= limit:
                    return

    def _fleet_repair_level3_restricted_exact(self, incumbent, victim, limit=80, exact_pool_max=7):
        """
        Enumeração sistemática de uma vizinhança restrita.

        Para até exact_pool_max clientes destruídos, enumera combinações de
        rota/posição de inserção (limitadas por 'limit'). Para pools maiores,
        seleciona uma janela de até exact_pool_max clientes e mantém os demais
        por um seed L2. Portanto a palavra EXACT refere-se somente ao subproblema
        restrito enumerado, não ao VRP global.
        """
        import itertools
        kept, _, pool = self._victim_pool(incumbent, victim)
        if not kept or not pool:
            return

        # base: para pool grande, posiciona a cauda heuristicamente e enumera a cabeça.
        exact_nodes = list(pool[:exact_pool_max])
        tail = list(pool[exact_pool_max:])
        base = self._copy(kept)
        for k, node in enumerate(tail):
            target = k % len(base)
            p = self._path(base[target]) or [0, 0]
            p.insert(max(1, len(p)-1), node)
            base[target]["path"] = p

        # opções compactas rota/posição para cada cliente.
        options = []
        for node in exact_nodes:
            op = []
            for ridx, r in enumerate(base):
                p = self._path(r) or [0, 0]
                for pos in sorted(set([1, max(1, len(p)//2), max(1, len(p)-1)])):
                    op.append((ridx, pos))
            options.append(op)

        # Cursor por estrutura/nível: retoma uma enumeração interrompida.
        state = self._state_for(self.structural_signature(incumbent))
        cursor_key = f"{victim}|{limit}|{exact_pool_max}"
        cursors = state.setdefault("fleet_enumeration_cursors",{})
        progress = cursors.setdefault(cursor_key,{"offset":0,"produced":0})
        produced = int(progress["produced"])
        if produced >= limit:
            return
        offset = int(progress["offset"])
        combinations = itertools.islice(itertools.product(*options),offset,None)
        for combo_index,combo in enumerate(combinations,start=offset):
            nr = self._copy(base)
            # inserir sequencialmente, reajustando posição ao tamanho corrente
            for node, (ridx, pos) in zip(exact_nodes, combo):
                p = self._path(nr[ridx]) or [0, 0]
                p.insert(min(max(1, pos), max(1, len(p)-1)), node)
                nr[ridx]["path"] = p
            sig = self._signature("FLEET_RESTRICTED_EXACT", (victim, tuple(combo), tuple(exact_nodes)))
            if sig in self._tested:
                progress["offset"] = combo_index+1
                continue
            self._tested.add(sig)
            yield self._candidate_with_routes(
                incumbent, nr, "FLEET_REDUCTION_L3_RESTRICTED_EXACT",
                {"eliminated_vehicle": victim, "level": 3,
                 "exact_nodes": list(exact_nodes), "heuristic_tail": list(tail),
                 "target_vehicle_count": len({self._vehicle(r) for r in kept}),
                 "scope": "LOCAL_RESTRICTED_NEIGHBORHOOD",
                 "proof_status": "EXACT_ENUMERATION_WITHIN_RESTRICTED_NEIGHBORHOOD"}
            )
            # Só avança após o consumidor processar o candidato e pedir o próximo.
            produced += 1
            progress.update(offset=combo_index+1,produced=produced)
            if produced >= limit:
                return

    # ----------------------------- API  ----------------------------

    def optimize(self, incumbent, improvement_plan=None,
                 evaluator_callback=None, audit_callback=None,
                 budget_seconds=120, seed=0):
        import time
        _vrp_ensure_proof_scope(self.shared_memory)
        self._load_persistent_state()
        started = time.monotonic()
        deadline = started + max(0.0, float(budget_seconds))
        self.memory = []
        self._tested = set()

        best = self._copy(incumbent)
        best_score = self._evaluate(best, evaluator_callback)
        baseline_score = best_score
        plan = self._copy(improvement_plan or {})
        structural = plan.get("structural_incumbent_analysis", {}) or {}
        strategies = [str(s.get("type", "")).upper()
                      for s in (plan.get("strategies", []) or [])
                      if isinstance(s, dict)]

        start_sig = self.structural_signature(best)
        self._state_for(start_sig)
        print(f"🧩 AVE {self.VERSION} | incumbent={best_score:.12f} | orçamento={budget_seconds}s")
        print(f"   estratégias={strategies or ['GENERIC']}")

        improvements = 0
        evidence = []
        diagnostics = []

        def run_stream(name, stream, signature_at_start):
            nonlocal best, best_score, improvements
            if self._is_exhausted(signature_at_start, name):
                print(f"   ⏭️ {name}: SKIP — vizinhança já esgotada para este incumbent")
                return 0, False
            tested = 0
            accepted_any = False
            for cand in stream:
                if time.monotonic() >= deadline:
                    break
                tested += 1
                old = best_score
                best2, score2, accepted = self._consider(
                    cand, best, best_score, evaluator_callback, audit_callback,
                    f"AVE_V2_{name}", {"type": name, "metadata": cand.get("ave_metadata", {})}
                )
                if accepted:
                    best, best_score = best2, score2
                    improvements += 1
                    accepted_any = True
                    print(f"   ✅ {name}: UB {old:.12f} → {best_score:.12f}")
            # Só declara esgotada se terminou antes do deadline e não melhorou.
            if time.monotonic() < deadline and not accepted_any:
                self._mark_exhausted(signature_at_start, name, {"tested": tested})
            print(f"   ↳ {name}: {tested} vizinhanças avaliadas")
            return tested, accepted_any

        # 1) movimentos baratos, com memória por incumbent.
        sig = self.structural_signature(best)
        if time.monotonic() < deadline:
            run_stream("DIRECT_2OPT", self._two_opt_candidates(best, structural), sig)

        # 2) FleetReductionExactRepair progressivo, por vítima.
        if ("FLEET_REDUCTION" in strategies or not strategies) and time.monotonic() < deadline:
            for victim in self._fleet_order(best, structural):
                if time.monotonic() >= deadline:
                    break
                sig = self.structural_signature(best)
                victim_key = f"FLEET_DROP_{victim}"
                vst = self._state_for(sig)["fleet"].setdefault(str(victim), {"max_level_attempted": 0})
                victim_improved = False

                # o nível 4 é deliberadamente muito mais profundo. Ele NÃO é
                # prova global de inviabilidade; serve para não concluir a tentativa 3→2
                # com a mesma profundidade do nível 3.
                levels = [
                    (1, self._fleet_repair_level1(best, victim)),
                    (2, self._fleet_repair_level2(best, victim)),
                    (3, self._fleet_repair_level3_restricted_exact(best, victim)),
                    (4, self._fleet_repair_level3_restricted_exact(
                        best, victim, limit=500, exact_pool_max=12
                    )),
                    # braço exclusivo para redução K→K-1.
                    # É muito mais profundo que L4, mas continua sendo uma
                    # vizinhança restrita: NOT_FOUND nunca vira INFEASIBLE.
                    (5, self._fleet_repair_level3_restricted_exact(
                        best, victim, limit=3000, exact_pool_max=16
                    )),
                ]
                for level, stream in levels:
                    if time.monotonic() >= deadline or victim_improved:
                        break
                    level_key = (
                        f"FLEET_K_REDUCTION_ARM_{victim}_L5"
                        if level == 5 else f"{victim_key}_L{level}"
                    )
                    if level == 5:
                        print(
                            f"   🎯 AVE BRAÇO K-FLEET | eliminar V{victim} | "
                            "L5 até 3000 reconstruções | prova_global=NÃO"
                        )
                    if self._is_exhausted(sig, level_key):
                        continue
                    tested, accepted = run_stream(level_key, stream, sig)
                    vst["max_level_attempted"] = max(vst.get("max_level_attempted", 0), level)
                    vst[f"level_{level}_tested"] = tested
                    if accepted:
                        victim_improved = True
                        vst["success_level"] = level
                        vst["status"] = "IMPROVED"
                        break
                    else:
                        vst["status"] = "NOT_FOUND"
                diagnostics.append({
                    "type": "FLEET_REDUCTION",
                    "eliminated_vehicle": victim,
                    "status": "IMPROVED" if victim_improved else "NOT_FOUND",
                    "max_level_attempted": vst.get("max_level_attempted", 0),
                    "global_infeasibility_proven": False,
                    "deep_repair_level_4": bool(vst.get("max_level_attempted", 0) >= 4),
                })
                if not victim_improved:
                    evidence.append({
                        "type": "FLEET_ELIMINATION_LOCAL_FAILURE",
                        "vehicle": victim,
                        "scope": "LOCAL_RESTRICTED_NEIGHBORHOOD",
                        "proof_status": "NOT_FOUND",
                        "global_valid": False,
                        "note": "Falha local/restrita; não prova inviabilidade de reduzir frota."
                    })

        # 3) inter-trip simples.
        sig = self.structural_signature(best)
        if time.monotonic() < deadline:
            run_stream("RELOCATE", self._relocate_candidates(best), sig)
        sig = self.structural_signature(best)
        if time.monotonic() < deadline:
            run_stream("SWAP", self._swap_candidates(best), sig)

        # 4) movimento coordenado somente após vizinhanças simples.
        sig = self.structural_signature(best)
        if time.monotonic() < deadline:
            run_stream("EJECTION_CHAIN", self._ejection_chain_candidates(best), sig)

        final_sig = self.structural_signature(best)
        self._persistent["evidence"].extend(self._copy(evidence))
        self._persistent["evidence"] = self._persistent["evidence"][-500:]
        self._save_persistent_state()
        self._publish_memory()

        improved = best_score < baseline_score - 1e-10

        # Diversificação só faz sentido se não houve melhoria e as vizinhanças
        # principais do incumbent final já foram esgotadas.
        fst = self._state_for(final_sig)
        exhausted_names = {k for k, v in fst.get("exhausted", {}).items() if v}
        core = {"DIRECT_2OPT", "RELOCATE", "SWAP", "EJECTION_CHAIN"}
        exhausted_core = core.issubset(exhausted_names)
        need_div = (not improved) and exhausted_core and time.monotonic() < deadline

        result = {
            "status": "IMPROVED" if improved else "NOT_FOUND",
            "version": self.VERSION,
            "solution": self._copy(best),
            "baseline_score": float(baseline_score),
            "best_score": float(best_score),
            "improved": bool(improved),
            "improvements_accepted": int(improvements),
            "tested_neighborhoods": len(self._tested),
            "search_memory": self._copy(self.memory),
            "persistent_memory": self._copy(fst),
            "fleet_reduction_diagnostics": diagnostics,
            "bound_evidence": evidence,
            "need_diversification": bool(need_div),
            "diversification_reason": (
                "ADAPTIVE_NEIGHBORHOODS_EXHAUSTED_FOR_SAME_INCUMBENT"
                if need_div else None
            ),
            "structural_signature_before": start_sig,
            "structural_signature_after": final_sig,
            "ga_diversity_gate": {
                "required": bool(need_div),
                "minimum_structural_distance": 0.08,
                "instruction": (
                    "Aceitar retorno do GA como 'nova região' somente se "
                    "structural_distance(incumbent, ga_solution) >= 0.08."
                ),
            },
            "proof": {
                "global_infeasibility_proven": False,
                "note": (
                    "NOT_FOUND não implica INFEASIBLE. Deep/restricted repair "
                    "é exato apenas dentro da vizinhança enumerada."
                ),
            },
            "elapsed_seconds": time.monotonic() - started,
        }
        try:
            self.shared_memory.data["ave_last_result"] = self._copy(result)
            self.shared_memory.data["ave_bound_evidence"] = self._copy(evidence)
        except Exception:
            pass
        return result

    def improve_incumbent(self, inputs, baseline, seconds, folder, state):
        return _v512_local_portfolio(inputs, baseline, seconds, folder, state)

    def optimize_region(self, inputs, baseline, seconds, folder, state):
        return _v512_structural_finish(_v59_exact_search, inputs, baseline, seconds, folder, state)
````

### **BOUNDSTRENGTHENINGAGENT — FORTALECIMENTO DO LIMITE INFERIOR**

Enquanto outros procuram um plano melhor, ele investiga o quanto qualquer plano poderia melhorar. Seu trabalho é tornar mais forte o piso matemático, sem excluir soluções realmente possíveis.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class _BoundStrengtheningAgentV1:
    """
    Bound Strengthening Agent.

    Papel:
      - especialista do Track B;
      - recebe o LB válido já existente e tenta fortalecê-lo;
      - NÃO repete uma relaxação se não houver informação nova;
      - mantém Cut Pool / Evidence Pool;
      - aceita evidências do AVE (conflitos/provas locais) somente quando
        marcadas como matematicamente válidas;
      - nunca converte NOT_FOUND heurístico em prova de inviabilidade.

    O agente implementa bounds globais seguros e uma infraestrutura incremental.
    Bounds condicionais/restritos são armazenados separadamente e NÃO são
    promovidos automaticamente a global LB.

    API principal:
      strengthen(incumbent, current_certification=None,
                 ave_evidence=None, budget_seconds=30)
    """

    VERSION = "BOUND_STRENGTHENING_AGENT_V1_1_INTEGRATED_SAFE"

    def __init__(self, shared_memory):
        import copy
        self.shared_memory = shared_memory
        self._copy = copy.deepcopy
        self.cut_pool = []
        self.evidence_pool = []
        self._fingerprints = set()

    @staticmethod
    def _f(x, default=None):
        try:
            return float(x)
        except Exception:
            return default

    def _translated(self):
        import json
        raw = (getattr(self.shared_memory, "data", {}) or {}).get("translated_json", {}) or {}
        if isinstance(raw, dict):
            return raw
        if isinstance(raw, str):
            try:
                obj = json.loads(raw)
                return obj if isinstance(obj, dict) else {}
            except Exception:
                return {}
        return {}

    def _recursive(self, obj, keys):
        wanted = {str(k).lower() for k in keys}
        if isinstance(obj, dict):
            for k, v in obj.items():
                if str(k).lower() in wanted:
                    return v
            for v in obj.values():
                hit = self._recursive(v, keys)
                if hit is not None:
                    return hit
        elif isinstance(obj, (list, tuple)):
            for v in obj:
                hit = self._recursive(v, keys)
                if hit is not None:
                    return hit
        return None

    def _objective_context(self):
        return (getattr(self.shared_memory, "data", {}) or {}).get("objective_context", {}) or {}

    def _canonical_coefficients(self):
        ctx = self._objective_context()
        coeff = ctx.get("coefficients", {}) or {}
        out = {}
        for k, v in coeff.items():
            x = self._f(v, None)
            if x is not None and x >= 0:
                out[str(k).upper()] = x
        return out

    def _current_lb(self, cert):
        cert = cert or {}
        for k in ("lower_bound", "LB", "best_lower_bound", "certified_lower_bound"):
            x = self._f(cert.get(k), None)
            if x is not None:
                return x
        mem = getattr(self.shared_memory, "data", {}) or {}
        for k in ("certified_lower_bound", "lower_bound"):
            x = self._f(mem.get(k), None)
            if x is not None:
                return x
        return 0.0

    def _client_rows(self):
        """Reúne clientes, demanda, geometria e janelas de atendimento para a modelagem."""
        data = getattr(self.shared_memory, "data", {}) or {}
        df = data.get("df_clientes")
        rows=[]
        if df is not None:
            try:
                cols={str(c).strip().lower():c for c in df.columns}
                idc=next((cols[k] for k in ("id","id_cliente","cliente","node_id") if k in cols),None)
                dc=next((cols[k] for k in ("demanda","demand") if k in cols),None)
                latc=next((cols[k] for k in ("latitude","lat") if k in cols),None)
                lonc=next((cols[k] for k in ("longitude","lon","lng") if k in cols),None)
                for idx,r in df.iterrows():
                    rows.append({
                        "id": r[idc] if idc is not None else idx,
                        "demand": self._f(r[dc],0.0) if dc is not None else 0.0,
                        "lat": self._f(r[latc],None) if latc is not None else None,
                        "lon": self._f(r[lonc],None) if lonc is not None else None,
                    })
            except Exception:
                pass
        try:
            wmap=self._window_map()
        except Exception:
            wmap={}
        for r in rows:
            candidates=[str(r.get("id"))]
            try:
                fv=float(r.get("id"))
                candidates += [str(int(fv)),str(fv)]
            except Exception:
                pass
            w=next((wmap.get(k) for k in candidates if wmap.get(k) is not None),None)
            if isinstance(w,(list,tuple)) and len(w)>=2:
                r["tw_start"]=self._f(w[0],None)
                r["tw_end"]=self._f(w[1],None)
        return rows

    def _vehicle_data(self):
        data = getattr(self.shared_memory, "data", {}) or {}
        caps, cpkm, fixed = [], [], []
        for key in ("df_veiculos_capacidade", "df_capacidade_veiculos", "df_vehicle_capacity"):
            df = data.get(key)
            if df is None: continue
            try:
                cols = {str(c).strip().lower(): c for c in df.columns}
                cc = next((cols[k] for k in ("capacidade", "capacity") if k in cols), None)
                if cc is not None:
                    caps += [self._f(v, None) for v in df[cc].tolist()]
            except Exception: pass
        for key in ("df_veiculos_custo_km", "df_veiculos_custo", "df_custo_veiculos", "df_vehicle_cost", "df_veiculos_custos"):
            df = data.get(key)
            if df is None: continue
            try:
                cols = {str(c).strip().lower(): c for c in df.columns}
                vc = next((cols[k] for k in ("custo_por_km", "cost_per_km", "custo_km") if k in cols), None)
                fc = next((cols[k] for k in ("custo_diario", "custo_fixo", "fixed_cost", "daily_fixed_cost") if k in cols), None)
                if vc is not None: cpkm += [self._f(v, None) for v in df[vc].tolist()]
                if fc is not None: fixed += [self._f(v, None) for v in df[fc].tolist()]
            except Exception: pass
        caps = [x for x in caps if x is not None and x > 0]
        cpkm = [x for x in cpkm if x is not None and x >= 0]
        fixed = [x for x in fixed if x is not None and x >= 0]
        return caps, cpkm, fixed

    @staticmethod
    def _haversine(a, b):
        import math
        if None in (a.get("lat"), a.get("lon"), b.get("lat"), b.get("lon")):
            return None
        R = 6371.0088
        p1, p2 = math.radians(a["lat"]), math.radians(b["lat"])
        dp = math.radians(b["lat"] - a["lat"])
        dl = math.radians(b["lon"] - a["lon"])
        h = math.sin(dp/2)**2 + math.cos(p1)*math.cos(p2)*math.sin(dl/2)**2
        return 2*R*math.asin(min(1.0, math.sqrt(h)))

    def _safe_global_bounds(self):
        """
        Produz apenas bounds que permanecem válidos globalmente.
        """
        import math
        rows = self._client_rows()
        caps, cpkm, fixed = self._vehicle_data()
        tr = self._translated()

        clients = [r for r in rows if str(r["id"]) not in ("0", "0.0") and r["demand"] >= 0]
        depot = next((r for r in rows if str(r["id"]) in ("0", "0.0")), None)
        total_demand = sum(r["demand"] for r in clients)
        qmax = max(caps) if caps else None

        trip_lb = math.ceil(total_demand / qmax) if qmax and qmax > 0 else None

        service = self._f(self._recursive(tr, (
            "service_time_per_delivery_minutes", "service_time_minutes",
            "average_service_time_minutes")), 0.0)
        max_day = self._f(self._recursive(tr, (
            "max_time_per_vehicle_minutes", "available_time_per_vehicle_minutes",
            "vehicle_available_time_minutes", "tempo_disponivel")), None)

        # LB de veículos por tempo de serviço puro: viagens/deslocamentos só aumentam tempo.
        service_work_lb = len(clients) * max(0.0, service or 0.0)
        vehicle_lb_time = math.ceil(service_work_lb / max_day) if max_day and max_day > 0 else 1
        vehicle_lb = max(1, vehicle_lb_time)

        radial = None
        degree = None
        if depot and qmax and qmax > 0:
            vals = []
            valid = True
            for c in clients:
                d = self._haversine(depot, c)
                if d is None:
                    valid = False; break
                vals.append(c["demand"] * d)
            if valid:
                radial = 2.0 / qmax * sum(vals)

        if depot and clients:
            nodes = [depot] + clients
            mins = []
            valid = True
            for c in clients:
                ds = []
                for o in nodes:
                    if o is c: continue
                    d = self._haversine(c, o)
                    if d is None:
                        valid = False; break
                    ds.append(d)
                if not valid or len(ds) < 2:
                    valid = False; break
                ds.sort()
                mins.append(ds[0] + ds[1])
            if valid:
                degree = 0.5 * sum(mins)

        distance_lb = max([x for x in (radial, degree, 0.0) if x is not None])

        min_cpkm = min(cpkm) if cpkm else 0.0
        min_fixed = min(fixed) if fixed else 0.0
        mode = str(self._recursive(tr, (
            "fixed_cost_mode", "fixed_cost_policy", "fixed_cost_payment_mode",
            "custo_fixo_modo")) or "").upper()
        when_used = any(t in mode for t in ("WHEN_USED", "ATIV", "UTILIZ", "USED", "CONDIT"))
        always = any(t in mode for t in ("ALWAYS", "INCOND", "SEMPRE"))

        if always and fixed:
            fixed_lb = sum(sorted(fixed))
        elif when_used and fixed:
            fixed_lb = vehicle_lb * min_fixed
        else:
            fixed_lb = 0.0

        cost_lb = distance_lb * min_cpkm + fixed_lb

        # Tempo total global: serviço + mínimo de deslocamento convertido por velocidade.
        speed = self._f(self._recursive(tr, ("average_vehicle_speed_kmh", "average_speed_kmh", "speed_kmh")), None)
        time_lb = service_work_lb
        if speed and speed > 0:
            time_lb += distance_lb / speed * 60.0

        return {
            "DISTANCE": float(distance_lb),
            "COST": float(cost_lb),
            "TIME": float(time_lb),
            "VEHICLES": float(vehicle_lb),
            "TRIPS": None if trip_lb is None else int(trip_lb),
            "components": {
                "distance_radial_capacity": radial,
                "distance_degree": degree,
                "service_work_minutes": service_work_lb,
                "vehicle_lb_time": vehicle_lb_time,
                "min_cost_per_km": min_cpkm,
                "fixed_cost_lb": fixed_lb,
            }
        }

    def _canonical_lb(self, bounds):
        coeff = self._canonical_coefficients()
        if not coeff:
            return None
        total = 0.0
        used = False
        for obj in ("DISTANCE", "COST", "TIME", "VEHICLES"):
            if obj in coeff and bounds.get(obj) is not None:
                total += coeff[obj] * float(bounds[obj])
                used = True
        return total if used else None

    def _ingest_ave_evidence(self, evidence):
        """
        Só evidência explicitamente certificada/exata entra no cut pool global.
        Heurística/NOT_FOUND é mantida apenas como informação local.
        """
        accepted = 0
        for e in evidence or []:
            if not isinstance(e, dict):
                continue
            fp = repr(sorted((str(k), repr(v)) for k, v in e.items()))
            if fp in self._fingerprints:
                continue
            self._fingerprints.add(fp)
            self.evidence_pool.append(self._copy(e))
            proof = str(e.get("proof_status", "")).upper()
            scope = str(e.get("scope", "")).upper()
            if proof in {"PROVEN", "EXACT", "CERTIFIED"} and scope == "GLOBAL":
                self.cut_pool.append(self._copy(e))
                accepted += 1
        return accepted

    def strengthen(self, incumbent, current_certification=None,
                   ave_evidence=None, budget_seconds=30):
        import time
        started = time.monotonic()
        base_lb = self._current_lb(current_certification)
        new_evidence = self._ingest_ave_evidence(ave_evidence or [])

        fingerprint = (
            round(base_lb, 12),
            len(self.cut_pool),
            len(self.evidence_pool),
            repr(self._objective_context()),
        )
        mem = getattr(self.shared_memory, "data", {}) or {}
        previous_fp = mem.get("bound_strengthening_last_fingerprint")

        bounds = self._safe_global_bounds()
        analytical = self._canonical_lb(bounds)
        candidates = [base_lb]
        if analytical is not None:
            candidates.append(float(analytical))

        # somente cuts globais que tragam explicitamente um LB canônico válido.
        cut_lbs = []
        for cut in self.cut_pool:
            x = self._f(cut.get("canonical_lower_bound"), None)
            if x is not None:
                cut_lbs.append(x)
        candidates += cut_lbs
        best_lb = max(candidates)

        improved = best_lb > base_lb + 1e-12
        no_new_information = (previous_fp == fingerprint and new_evidence == 0)

        status = (
            "IMPROVED"
            if improved else
            "NO_NEW_BOUND_INFORMATION"
            if no_new_information else
            "VALID_LB_UNCHANGED"
        )

        result = {
            "status": status,
            "version": self.VERSION,
            "base_lower_bound": float(base_lb),
            "best_lower_bound": float(best_lb),
            "improved": bool(improved),
            "analytical_canonical_lb": analytical,
            "global_component_bounds": bounds,
            "new_global_evidence": int(new_evidence),
            "cut_pool_size": len(self.cut_pool),
            "evidence_pool_size": len(self.evidence_pool),
            "repeat_same_relaxation": False,
            "should_rerun_without_new_information": False,
            "validity": "GLOBAL_VALID_LOWER_BOUND_COMPONENTS_ONLY",
            "warnings": [
                "NOT_FOUND heurístico nunca é promovido a prova.",
                "Bounds condicionais/restritos não são usados como LB global.",
            ],
            "elapsed_seconds": time.monotonic() - started,
        }
        try:
            self.shared_memory.data["bound_strengthening_last_fingerprint"] = fingerprint
            self.shared_memory.data["bound_strengthening_last_result"] = self._copy(result)
            self.shared_memory.data["bound_cut_pool"] = self._copy(self.cut_pool)
        except Exception:
            pass

        print(
            f"📐 BoundStrengthening V1 | LB {base_lb:.12f} → {best_lb:.12f} | "
            f"status={status} | cuts={len(self.cut_pool)}"
        )
        return result


class BoundStrengtheningAgent(_BoundStrengtheningAgentV1):
    """
    Bound Strengthening  — Persistent LB + geometric/workload strengthening
    + temporal incompatibility cuts.

    Princípios:
      - começa SEMPRE do maior LB global já certificado/conhecido;
      - nunca reduz LB;
      - pools persistem em SharedMemory entre restarts;
      - cortes temporais são globais somente quando derivados de condição
        necessária matematicamente segura;
      - evidência local do AVE nunca vira prova global automaticamente;
      - não repete separação sem informação/estado novo.

    Esta versão não afirma que um corte local prova inviabilidade global.
    """
    VERSION = "BOUND_STRENGTHENING_AGENT_V6_SCF_LP"

    def __init__(self, shared_memory):
        super().__init__(shared_memory)
        self._load_persistent()

    def _load_persistent(self):
        self.cut_pool = []
        self.evidence_pool = []
        self._fingerprints = set()
        mem = getattr(self.shared_memory, "data", {}) or {}
        state = mem.get("bound_v2_persistent_state", {}) or {}
        if isinstance(state, dict):
            self.best_persistent_lb = self._f(state.get("best_lower_bound"), 0.0) or 0.0
            cp = state.get("cut_pool")
            ep = state.get("evidence_pool")
            fps = state.get("fingerprints")
            if isinstance(cp, list):
                self.cut_pool = self._copy(cp)
            if isinstance(ep, list):
                self.evidence_pool = self._copy(ep)
            if isinstance(fps, (list, set, tuple)):
                self._fingerprints = set(fps)
        else:
            self.best_persistent_lb = 0.0

    def _save_persistent(self, best_lb):
        self.best_persistent_lb = max(self.best_persistent_lb, float(best_lb))
        state = {
            "version": self.VERSION,
            "best_lower_bound": self.best_persistent_lb,
            "cut_pool": self._copy(self.cut_pool),
            "evidence_pool": self._copy(self.evidence_pool[-1000:]),
            "fingerprints": list(self._fingerprints)[-5000:],
        }
        try:
            self.shared_memory.data["bound_v2_persistent_state"] = state
            self.shared_memory.data["certified_lower_bound"] = max(
                self._f(self.shared_memory.data.get("certified_lower_bound"), 0.0) or 0.0,
                self.best_persistent_lb
            )
        except Exception:
            pass

    # --------------------- recuperação robusta do LB ----------------------

    def _all_numeric_lbs(self, obj):
        vals = []
        wanted = {
            "lower_bound", "lb", "best_lower_bound", "certified_lower_bound",
            "global_lower_bound", "canonical_lower_bound"
        }
        def walk(x):
            if isinstance(x, dict):
                for k, v in x.items():
                    if str(k).lower() in wanted:
                        z = self._f(v, None)
                        if z is not None and z >= 0:
                            vals.append(z)
                    walk(v)
            elif isinstance(x, (list, tuple)):
                for v in x:
                    walk(v)
        walk(obj)
        return vals

    def _current_lb(self, cert):
        mem = getattr(self.shared_memory, "data", {}) or {}
        vals = [self.best_persistent_lb]
        vals += self._all_numeric_lbs(cert or {})
        # Fontes conhecidas do sistema/certificador.
        for key in (
            "certified_lower_bound", "lower_bound", "best_lower_bound",
            "optimality_certification", "certification_result",
            "last_certification", "global_certification"
        ):
            if key in mem:
                vals += self._all_numeric_lbs({key: mem.get(key)})
                z = self._f(mem.get(key), None)
                if z is not None and z >= 0:
                    vals.append(z)
        return max([x for x in vals if x is not None] or [0.0])

    # --------------------- dados temporais ----------------------

    def _window_map(self):
        # Usa exatamente a interpretação de janelas do modelo exato.
        provider = OptimalityCertificationAgent(self.shared_memory,verbose_solver=False)
        problem = provider._build_problem()
        return {_vrp_vehicle_id(cid):tuple(win) for cid,win in problem["windows"].items()}

    def _client_map(self):
        return {str(r["id"]): r for r in self._client_rows()}

    def _service_minutes(self):
        tr = self._translated()
        return max(0.0, self._f(self._recursive(tr, (
            "service_time_per_delivery_minutes", "service_time_minutes",
            "average_service_time_minutes", "tempo_medio_entrega",
            "tempo_servico"
        )), 0.0) or 0.0)

    def _speed_kmh(self):
        tr = self._translated()
        return self._f(self._recursive(tr, (
            "average_vehicle_speed_kmh", "average_speed_kmh", "speed_kmh",
            "velocidade_media"
        )), None)

    # --------------------- bounds geométricos seguros ----------------------

    def _mst_lower_bound(self, rows):
        """
        Toda união de rotas depósito→...→depósito é um grafo conectado que
        contém depósito e clientes. Logo seu comprimento é >= MST desses nós.
        """
        nodes = [r for r in rows if r.get("lat") is not None and r.get("lon") is not None]
        if len(nodes) != len(rows) or len(nodes) <= 1:
            return None
        used = {0}
        best = [float("inf")] * len(nodes)
        for j in range(1, len(nodes)):
            d = self._haversine(nodes[0], nodes[j])
            if d is None:
                return None
            best[j] = d
        total = 0.0
        for _ in range(len(nodes) - 1):
            j = min((k for k in range(len(nodes)) if k not in used), key=lambda k: best[k], default=None)
            if j is None or best[j] == float("inf"):
                return None
            total += best[j]
            used.add(j)
            for k in range(len(nodes)):
                if k in used:
                    continue
                d = self._haversine(nodes[j], nodes[k])
                if d is not None and d < best[k]:
                    best[k] = d
        return total

    # --------------------- cortes temporais globais ----------------------

    def _temporal_incompatibility_cuts(self):
        """
        Gera incompatibilidade GLOBAL de dois clientes NA MESMA VIAGEM quando
        nem i→j nem j→i podem respeitar janelas, mesmo usando tempo de viagem
        Haversine (lower bound geométrico) e serviço.

        Isso é um corte global de mesma-viagem. Em ambiente multi-trip ele
        fortalece TRIPS, mas NÃO é automaticamente um corte de veículos.
        """
        windows = self._window_map()
        cmap = self._client_map()
        speed = self._speed_kmh()
        service = self._service_minutes()
        if not windows or not speed or speed <= 0:
            return []

        ids = [i for i in windows if i in cmap and i not in ("0", "0.0")]
        cuts = []
        for a in range(len(ids)):
            i = ids[a]
            wi = windows[i]
            for b in range(a + 1, len(ids)):
                j = ids[b]
                wj = windows[j]
                dij = self._haversine(cmap[i], cmap[j])
                if dij is None:
                    continue
                tij = dij / speed * 60.0
                # condição necessária para existir i antes de j:
                # começar i no mais cedo possível + serviço + viagem <= fim de j
                i_to_j_possible = wi[0] + service + tij <= wj[1] + 1e-9
                j_to_i_possible = wj[0] + service + tij <= wi[1] + 1e-9
                if not i_to_j_possible and not j_to_i_possible:
                    cuts.append({
                        "type": "TEMPORAL_PAIR_SAME_TRIP_INCOMPATIBILITY",
                        "clients": [i, j],
                        "scope": "GLOBAL",
                        "proof_status": "CERTIFIED",
                        "global_valid": True,
                        "meaning": "Os dois clientes não podem pertencer à mesma viagem.",
                    })
        return cuts

    def _max_clique_size(self, pair_cuts, node_limit=80):
        """Clique exata por Bron–Kerbosch para grafo pequeno; retorna LB de viagens."""
        adj = {}
        for c in pair_cuts:
            p = c.get("clients", [])
            if len(p) != 2:
                continue
            a, b = map(str, p)
            adj.setdefault(a, set()).add(b)
            adj.setdefault(b, set()).add(a)
        if not adj or len(adj) > node_limit:
            # fallback seguro: qualquer aresta implica clique 2
            return 2 if adj else 1

        best = 1
        def bk(R, P, X):
            nonlocal best
            if not P and not X:
                best = max(best, len(R)); return
            if len(R) + len(P) <= best:
                return
            # pivot
            U = P | X
            u = max(U, key=lambda z: len(adj.get(z, set()) & P), default=None)
            candidates = P - (adj.get(u, set()) if u is not None else set())
            for v in list(candidates):
                bk(R | {v}, P & adj.get(v, set()), X & adj.get(v, set()))
                P.remove(v); X.add(v)
        bk(set(), set(adj), set())
        return best


    def _interval_energy_vehicle_lb(self, clients, service_minutes):
        windows = self._window_map()
        intervals = []
        for c in clients:
            cid = str(c.get("id"))
            win = windows.get(cid)
            if win is None:
                a, b = self._f(c.get("tw_start"),None), self._f(c.get("tw_end"),None)
                if a is not None and b is not None: win = (a,b)
            if win is not None:
                intervals.append(tuple(map(float,win)))
        return _vrp_interval_vehicle_lb(intervals,service_minutes)

    def _held_karp_one_tree_lb(self, rows, iterations=80):
        """LB Held-Karp por 1-tree e potenciais; válido para métrica Haversine."""
        nodes=[r for r in rows if r.get("lat") is not None and r.get("lon") is not None]
        n=len(nodes)
        if n<3: return None
        root=next((i for i,r in enumerate(nodes) if str(r.get("id")) in ("0","0.0")),0)
        C=[[0.0]*n for _ in range(n)]
        for i in range(n):
            for j in range(i+1,n):
                d=self._haversine(nodes[i],nodes[j])
                if d is None: return None
                C[i][j]=C[j][i]=float(d)
        pi=[0.0]*n; best=-float("inf")
        def one_tree():
            others=[i for i in range(n) if i!=root]
            inside={others[0]}; edges=[]
            while len(inside)<len(others):
                ch=None
                for i in inside:
                    for j in others:
                        if j in inside: continue
                        w=C[i][j]+pi[i]+pi[j]
                        if ch is None or w<ch[0]: ch=(w,i,j)
                if ch is None: return None
                edges.append((ch[1],ch[2])); inside.add(ch[2])
            re=sorted((C[root][j]+pi[root]+pi[j],root,j) for j in others)[:2]
            edges += [(a,b) for _,a,b in re]
            deg=[0]*n
            for a,b in edges: deg[a]+=1; deg[b]+=1
            mod=sum(C[a][b]+pi[a]+pi[b] for a,b in edges)
            return mod-2.0*sum(pi),deg
        for it in range(max(1,int(iterations))):
            z=one_tree()
            if z is None: break
            lb,deg=z; best=max(best,lb)
            g=[d-2 for d in deg]; norm=sum(x*x for x in g)
            if norm<=1e-12: break
            step=2.0/(1.0+0.08*it)
            for i in range(n): pi[i]+=step*g[i]
        return None if best==-float("inf") else float(best)

    def _radial_capacity_distance_lb(self, rows):
        """LB radial CVRP: (2/Qmax)*sum(demand_i*dist(depot,i))."""
        depot=next((r for r in rows if str(r.get("id")) in ("0","0.0")),None)
        caps,_,_=self._vehicle_data()
        if depot is None or not caps or max(caps)<=0: return None
        qmax=max(caps); total=0.0
        for r in rows:
            if str(r.get("id")) in ("0","0.0"): continue
            d=self._haversine(depot,r)
            if d is None: return None
            total += max(0.0,float(self._f(r.get("demand"),0.0) or 0.0))*float(d)
        return float(2.0*total/qmax)

    def _safe_global_bounds_v2(self):
        import math
        b = super()._safe_global_bounds()
        rows = self._client_rows()
        clients = [r for r in rows if str(r["id"]) not in ("0", "0.0")]
        depot = next((r for r in rows if str(r["id"]) in ("0", "0.0")), None)

        # MST fortalece LB de distância quando maior.
        all_rows = ([depot] if depot else []) + clients
        mst = self._mst_lower_bound(all_rows) if depot else None
        one_tree = self._one_tree_lower_bound(all_rows) if depot else None
        old_distance = self._f(b.get("DISTANCE"), 0.0) or 0.0
        lp_result = {"valid":False,"status":"NO_BUDGET"}
        left = max(0.0,getattr(self,"_lp_deadline_v5",0.0)-__import__("time").monotonic())
        if left >= 1.0:
            try:
                provider = OptimalityCertificationAgent(self.shared_memory,verbose_solver=False)
                provider._load_objective_context()
                problem = provider._build_problem()
                cache = self.shared_memory.data.setdefault("scf_lp_cache_v5",{})
                lp_result = _vrp_scf_distance_bound(problem,int(b.get("TRIPS") or 1),left,cache)
                if lp_result.get("valid"):
                    old_distance = max(old_distance,float(lp_result["distance_lb"]))
            except Exception as exc:
                lp_result = {"valid":False,"status":"LP_SETUP_ERROR","error":str(exc)}
        self.shared_memory.data["scf_lp_last_result_v5"] = lp_result
        hk_one_tree = self._held_karp_one_tree_lb(all_rows, iterations=80) if depot else None
        radial_capacity = self._radial_capacity_distance_lb(all_rows) if depot else None
        distance_lb = max(
            old_distance, mst or 0.0, one_tree or 0.0,
            hk_one_tree or 0.0, radial_capacity or 0.0
        )

        service_work = self._f((b.get("components") or {}).get("service_work_minutes"), 0.0) or 0.0
        speed = self._speed_kmh()
        time_lb = service_work + (distance_lb / speed * 60.0 if speed and speed > 0 else 0.0)

        tr = self._translated()
        max_day = self._f(self._recursive(tr, (
            "max_time_per_vehicle_minutes", "available_time_per_vehicle_minutes",
            "vehicle_available_time_minutes", "tempo_disponivel"
        )), None)
        vehicle_lb_workload = math.ceil(time_lb / max_day) if max_day and max_day > 0 else 1
        vehicle_lb_interval, interval_witness = self._interval_energy_vehicle_lb(clients, service_work / max(1, len(clients)))
        vehicle_lb = max(int(self._f(b.get("VEHICLES"), 1) or 1), vehicle_lb_workload, vehicle_lb_interval)

        # Somente PROVEN_INFEASIBLE + global_valid pode elevar vehicleLB.
        mem = getattr(self.shared_memory, "data", {}) or {}
        kproof = mem.get("k_vehicle_feasibility_last_result") or {}
        kproof_vehicle_lb = None
        if (
            isinstance(kproof, dict)
            and str(kproof.get("status", "")).upper() == "PROVEN_INFEASIBLE"
            and kproof.get("global_valid") is True
        ):
            try:
                kproof_vehicle_lb = int(kproof.get("proven_vehicle_lower_bound"))
                vehicle_lb = max(vehicle_lb, kproof_vehicle_lb)
            except Exception:
                kproof_vehicle_lb = None

        # nunca perder uma prova anterior se last_result for posteriormente UNKNOWN.
        history_k_lb = 1
        try:
            for _kp in (mem.get("k_vehicle_feasibility_history", []) or []):
                if (
                    isinstance(_kp, dict)
                    and str(_kp.get("status", "")).upper() == "PROVEN_INFEASIBLE"
                    and _kp.get("global_valid") is True
                ):
                    history_k_lb = max(history_k_lb, int(_kp.get("proven_vehicle_lower_bound", 1) or 1))
            vehicle_lb = max(vehicle_lb, history_k_lb)
        except Exception:
            history_k_lb = 1
        try:
            persisted_k_lb = int(mem.get("certified_vehicle_lower_bound", 1) or 1)
            vehicle_lb = max(vehicle_lb, persisted_k_lb)
        except Exception:
            persisted_k_lb = 1

        # Cortes temporais → clique = LB de viagens, não de veículos em multi-trip.
        temporal = self._temporal_incompatibility_cuts()
        clique_trip_lb = self._max_clique_size(temporal)
        trip_lb = max(int(b.get("TRIPS") or 1), int(clique_trip_lb))
        depot_trip_edges = self._depot_trip_edge_lower_bound(all_rows, trip_lb)
        if depot_trip_edges is not None:
            distance_lb = max(distance_lb, float(depot_trip_edges))
            time_lb = service_work + (distance_lb / speed * 60.0 if speed and speed > 0 else 0.0)
            vehicle_lb_workload = math.ceil(time_lb / max_day) if max_day and max_day > 0 else 1
            vehicle_lb = max(vehicle_lb, vehicle_lb_workload)

        # Recalcular custo com bounds fortalecidos.
        _, cpkm, fixed = self._vehicle_data()
        min_cpkm = min(cpkm) if cpkm else 0.0
        mode = str(self._recursive(tr, (
            "fixed_cost_mode", "fixed_cost_policy", "fixed_cost_payment_mode",
            "custo_fixo_modo"
        )) or "").upper()
        when_used = any(t in mode for t in ("WHEN_USED", "ATIV", "UTILIZ", "USED", "CONDIT"))
        always = any(t in mode for t in ("ALWAYS", "INCOND", "SEMPRE"))
        if always and fixed:
            fixed_lb = sum(sorted(fixed))
        elif when_used and fixed:
            # LB seguro: k veículos ativos custam pelo menos a soma dos k menores custos fixos.
            k = min(vehicle_lb, len(fixed))
            fixed_lb = sum(sorted(fixed)[:k])
        else:
            fixed_lb = 0.0
        cost_lb = distance_lb * min_cpkm + fixed_lb

        # bound de custo acoplado por veículo. Evita combinar
        # o menor custo/km de um veículo com os menores custos fixos de outros,
        # uma relaxação excessivamente otimista. Agora minimizamos a expressão
        # sobre subconjuntos ativos de forma segura.
        pairs = self._paired_vehicle_costs()
        coupled_cost_lb = None
        if when_used and pairs:
            import itertools
            vals=[]
            n=len(pairs)
            # Para 12 veículos a enumeração completa (4095 subconjuntos) é barata.
            if n <= 18:
                for k in range(max(1,int(vehicle_lb)), n+1):
                    for subset in itertools.combinations(pairs,k):
                        vals.append(sum(fc for _,fc in subset) + distance_lb*min(cp for cp,_ in subset))
                if vals: coupled_cost_lb=min(vals)
            if coupled_cost_lb is not None:
                cost_lb=max(cost_lb,float(coupled_cost_lb))

        b["DISTANCE"] = float(distance_lb)
        b["TIME"] = float(max(self._f(b.get("TIME"), 0.0) or 0.0, time_lb))
        b["VEHICLES"] = float(vehicle_lb)
        b["TRIPS"] = int(trip_lb)
        b["COST"] = float(max(self._f(b.get("COST"), 0.0) or 0.0, cost_lb))
        b.setdefault("components", {})
        b["components"].update({
            "scf_lp": lp_result,
            "distance_mst": mst,
            "distance_one_tree": one_tree,
            "distance_held_karp_one_tree": hk_one_tree,
            "distance_radial_capacity": radial_capacity,
            "distance_depot_trip_edges": depot_trip_edges,
            "vehicle_lb_workload": vehicle_lb_workload,
            "vehicle_lb_interval_energy": vehicle_lb_interval,
            "vehicle_lb_exact_k_proof": kproof_vehicle_lb,
            "vehicle_lb_persisted_certified": persisted_k_lb,
            "interval_energy_witness": interval_witness,
            "temporal_pair_cuts": len(temporal),
            "temporal_clique_trip_lb": clique_trip_lb,
            "fixed_cost_lb_v2": fixed_lb,
            "coupled_heterogeneous_cost_lb": coupled_cost_lb,
        })
        return b, temporal

    def _one_tree_lower_bound(self, rows):
        """LB métrico 1-tree: MST sem o depósito + 2 menores arestas do depósito.
        Seguro para VRP métrico porque concatenar/atalhar rotas produz um tour TSP
        não mais longo que a soma das rotas VRP.
        """
        depot = next((r for r in rows if str(r.get("id")) in ("0", "0.0")), None)
        clients = [r for r in rows if str(r.get("id")) not in ("0", "0.0")]
        if depot is None or len(clients) < 2:
            return None
        mst_clients = self._mst_lower_bound(clients)
        if mst_clients is None:
            return None
        ds = sorted(self._haversine(depot, c) for c in clients)
        return float(mst_clients + ds[0] + ds[1])

    def _depot_trip_edge_lower_bound(self, rows, trips_lb):
        """LB das arestas incidentes ao depósito para pelo menos m viagens.
        Cada viagem exige saída+retorno. Duplicar cada distância de cliente duas
        vezes é uma relaxação otimista e portanto um lower bound seguro.
        """
        depot = next((r for r in rows if str(r.get("id")) in ("0", "0.0")), None)
        clients = [r for r in rows if str(r.get("id")) not in ("0", "0.0")]
        m = int(trips_lb or 0)
        if depot is None or not clients or m <= 0:
            return None
        edges=[]
        for c in clients:
            d=float(self._haversine(depot,c)); edges.extend([d,d])
        edges.sort()
        return float(sum(edges[:min(2*m,len(edges))]))

    def _paired_vehicle_costs(self):
        """Preserva o pareamento (custo/km, custo fixo) do mesmo veículo."""
        data=getattr(self.shared_memory,"data",{}) or {}; pairs=[]
        for key in ("df_veiculos_custo_km","df_veiculos_custo","df_custo_veiculos","df_vehicle_cost","df_veiculos_custos"):
            df=data.get(key)
            if df is None: continue
            try:
                cols={str(c).strip().lower():c for c in df.columns}
                vc=next((cols[k] for k in ("custo_por_km","cost_per_km","custo_km") if k in cols),None)
                fc=next((cols[k] for k in ("custo_diario","custo_fixo","fixed_cost","daily_fixed_cost") if k in cols),None)
                if vc is not None and fc is not None:
                    for _,r in df.iterrows():
                        a=self._f(r[vc],None); b=self._f(r[fc],None)
                        if a is not None and b is not None and a>=0 and b>=0: pairs.append((float(a),float(b)))
                    if pairs: break
            except Exception: pass
        return pairs

    # ----------------------------- API  ----------------------------

    def strengthen(self, incumbent, current_certification=None,
                   ave_evidence=None, budget_seconds=30):
        import time
        _vrp_ensure_proof_scope(self.shared_memory)
        self._load_persistent()
        started = time.monotonic()
        self._lp_deadline_v5 = started + max(0.0,float(budget_seconds))

        # 1. Nunca começar do zero se já existe LB conhecido.
        base_lb = self._current_lb(current_certification)

        # 2. Evidência AVE: só global/exata entra no cut pool.
        new_ave = self._ingest_ave_evidence(ave_evidence or [])

        # 3. Separação temporal + bounds fortalecidos.
        bounds, temporal_cuts = self._safe_global_bounds_v2()
        new_temporal = 0
        for cut in temporal_cuts:
            fp = repr(sorted((str(k), repr(v)) for k, v in cut.items()))
            if fp in self._fingerprints:
                continue
            self._fingerprints.add(fp)
            self.cut_pool.append(self._copy(cut))
            self.evidence_pool.append(self._copy(cut))
            new_temporal += 1

        analytical = self._canonical_lb(bounds)
        candidates = [base_lb, self.best_persistent_lb]
        if analytical is not None:
            candidates.append(float(analytical))

        # Cortes que tragam LB canônico explícito.
        explicit_cut_lbs = []
        for cut in self.cut_pool:
            x = self._f(cut.get("canonical_lower_bound"), None)
            if x is not None and x >= 0:
                explicit_cut_lbs.append(x)
        candidates.extend(explicit_cut_lbs)

        best_lb = max(candidates)
        # Não publica um limite que contradiga uma solução auditada conhecida.
        ub = self._f((incumbent or {}).get("canonical_objective_score"),None)
        if ub is not None and best_lb > ub + max(1e-7,abs(ub)*1e-7):
            raise ValueError(f"Limite inferior {best_lb} excede incumbent {ub}; certificação rejeitada")
        improved = best_lb > base_lb + 1e-12

        # Fingerprint após incorporação de informação.
        fingerprint = (
            round(best_lb, 12),
            len(self.cut_pool),
            len(self.evidence_pool),
            repr(self._objective_context()),
            int((bounds.get("components") or {}).get("temporal_pair_cuts", 0)),
        )
        mem = getattr(self.shared_memory, "data", {}) or {}
        previous_fp = mem.get("bound_strengthening_last_fingerprint")
        no_new_information = (
            previous_fp == fingerprint and new_ave == 0 and new_temporal == 0
        )

        if improved:
            status = "IMPROVED"
        elif no_new_information:
            status = "NO_NEW_BOUND_INFORMATION"
        else:
            status = "VALID_LB_UNCHANGED"

        self._save_persistent(best_lb)

        result = {
            "status": status,
            "version": self.VERSION,
            "base_lower_bound": float(base_lb),
            "best_lower_bound": float(best_lb),
            "persistent_best_lower_bound": float(self.best_persistent_lb),
            "improved": bool(improved),
            "analytical_canonical_lb": analytical,
            "global_component_bounds": bounds,
            "new_global_ave_evidence": int(new_ave),
            "new_temporal_cuts": int(new_temporal),
            "temporal_cut_count_total": sum(
                1 for c in self.cut_pool
                if str(c.get("type", "")).startswith("TEMPORAL_")
            ),
            "cut_pool_size": len(self.cut_pool),
            "evidence_pool_size": len(self.evidence_pool),
            "repeat_same_relaxation": False,
            "should_rerun_without_new_information": False,
            "validity": "GLOBAL_VALID_PERSISTENT_LOWER_BOUND_COMPONENTS_AND_SAFE_CUTS",
            "warnings": [
                "NOT_FOUND heurístico nunca é promovido a prova.",
                "Restricted exact do AVE é prova apenas da vizinhança restrita.",
                "Clique temporal em modo multi-trip fortalece TRIPS; não é automaticamente LB de veículos.",
                "O LB global nunca é reduzido entre restarts.",
            ],
            "elapsed_seconds": time.monotonic() - started,
        }
        try:
            self.shared_memory.data["bound_strengthening_last_fingerprint"] = fingerprint
            self.shared_memory.data["bound_strengthening_last_result"] = self._copy(result)
            self.shared_memory.data["bound_cut_pool"] = self._copy(self.cut_pool)
            self.shared_memory.data["bound_strengthening_best_lb"] = float(best_lb)
        except Exception:
            pass

        print(
            f"📐 BoundStrengthening V5.3 | LB {base_lb:.12f} → {best_lb:.12f} | "
            f"status={status} | cuts={len(self.cut_pool)} | "
            f"temporal_novos={new_temporal}"
        )
        try:
            _c=(bounds.get("components") or {})
            print(
                "   📏 LB components | "
                f"MST={_c.get('distance_mst')} | 1tree={_c.get('distance_one_tree')} | "
                f"HK={_c.get('distance_held_karp_one_tree')} | "
                f"radial={_c.get('distance_radial_capacity')} | "
                f"TW_vehicle={_c.get('vehicle_lb_interval_energy')} | "
                f"vehicleLB={bounds.get('VEHICLES')} | tripLB={bounds.get('TRIPS')} | "
                f"costLB={bounds.get('COST')}"
            )
        except Exception:
            pass
        return result
````

### **KVEHICLEFEASIBILITYCERTIFIER — INVESTIGAÇÃO DE MENOS VEÍCULOS**

Este agente pergunta se é possível atender todos usando menos veículos.



Tabela com células delimitadas: Responsabilidade / Explicação

````python
class KVehicleFeasibilityCertifier:
    VERSION = "K_VEHICLE_FEASIBILITY_CERTIFIER_V5_FAIR_DEADLINES_SCOPE"

    """
    Certificador generalizado de factibilidade de cardinalidade de frota com braço Exact-K SCIP integrado.

    Pergunta:
        existe uma solução factível do problema original usando <= K veículos?

    Estados:
        FEASIBLE          -> encontrou candidato <= K; precisa Auditor independente.
        PROVEN_INFEASIBLE -> somente quando uma prova global segura fecha K.
        UNKNOWN           -> não houve prova no orçamento.
        CONFIGURATION_ERROR

    Regra fundamental:
        NOT_FOUND heurístico NUNCA é convertido em PROVEN_INFEASIBLE.

    O agente é de suporte à Continuidade/Certificação. Ele não substitui
    OR-Tools, GA, AVE, BoundStrengtheningAgent, Auditor ou Certifier global.
    """

    def __init__(self, shared_memory):
        self.shared_memory = shared_memory
        self.role = "Certificador estrutural de cardinalidade K da frota."
        self.goal = (
            "Fechar a lacuna entre vehicle lower bound e vehicle upper bound "
            "por filtros matemáticos globais seguros e, quando disponível, "
            "busca HARD-K para encontrar testemunhos factíveis."
        )
        self.memory = []

    # ------------------------ utilidades -------------------------
    def _mem(self):
        return self.shared_memory.data

    def _f(self, x, default=0.0):
        try:
            if x is None:
                return default
            return float(x)
        except Exception:
            return default

    def _i(self, x, default=None):
        try:
            return int(x)
        except Exception:
            return default

    def _used_vehicles(self, solution):
        if not isinstance(solution, dict):
            return None
        for key in ("used_vehicles", "num_used_vehicles", "vehicles_used"):
            if solution.get(key) is not None:
                return self._i(solution.get(key))
        metrics = solution.get("metrics") or solution.get("solution_metrics") or {}
        for key in ("used_vehicles", "num_used_vehicles", "vehicles_used"):
            if metrics.get(key) is not None:
                return self._i(metrics.get(key))
        vehicles = solution.get("vehicles")
        if isinstance(vehicles, list):
            return sum(
                1 for v in vehicles
                if isinstance(v, dict) and (
                    v.get("used", False)
                    or bool(v.get("trips"))
                    or bool(v.get("routes"))
                )
            )
        return None

    def _candidate_feasible_flag(self, solution):
        if not isinstance(solution, dict):
            return False
        status = str(solution.get("status", "") or "").upper()
        metrics = solution.get("metrics") or solution.get("solution_metrics") or {}
        if metrics.get("feasible") is True:
            return True
        if solution.get("feasible") is True:
            return True
        return status in ("SUCCESS", "FEASIBLE", "OPTIMAL")

    def _vehicle_table(self):
        mem=self._mem()
        df=mem.get("df_veiculos_capacidade")
        if df is None:
            return []
        rows=[]
        try:
            for pos,(_,r) in enumerate(df.iterrows()):
                cols={str(c).strip().lower():c for c in df.columns}
                capc=next((cols[k] for k in ("capacidade","capacity") if k in cols),None)
                idc=next((cols[k] for k in ("veiculo","vehicle","id","id_veiculo") if k in cols),None)
                rows.append({
                    "idx":pos,
                    "id": r[idc] if idc is not None else pos+1,
                    "capacity":self._f(r[capc],0.0) if capc is not None else 0.0,
                })
        except Exception:
            return []
        return rows

    def _client_rows(self):
        mem=self._mem()
        df=mem.get("df_clientes")
        if df is None:
            return []
        rows=[]
        try:
            cols={str(c).strip().lower():c for c in df.columns}
            idc=next((cols[k] for k in ("cliente","id_cliente","id","node_id") if k in cols),None)
            dc=next((cols[k] for k in ("demanda","demand") if k in cols),None)
            for pos,(_,r) in enumerate(df.iterrows()):
                rows.append({
                    "id":r[idc] if idc is not None else pos,
                    "demand":self._f(r[dc],0.0) if dc is not None else 0.0,
                })
        except Exception:
            return []
        return rows

    def _problem_desc(self, translated_problem=None):
        if isinstance(translated_problem, str):
            try:
                import json
                translated_problem=json.loads(translated_problem)
            except Exception:
                translated_problem=None
        if isinstance(translated_problem, dict):
            d=translated_problem.get("problem_description")
            if isinstance(d, dict):
                return d
        mem=self._mem()
        for key in ("translated_problem","translated_problem_json","problem_json"):
            obj=mem.get(key)
            if isinstance(obj, dict):
                d=obj.get("problem_description",obj)
                if isinstance(d,dict):
                    return d
        return {}

    # --------------------- lower bounds seguros ----------------------
    def _known_vehicle_lb(self):
        mem=self._mem()
        vals=[]
        for key in (
            "continuity_best_vehicle_lower_bound",
            "certified_vehicle_lower_bound",
            "vehicle_lower_bound",
        ):
            v=self._i(mem.get(key))
            if v is not None and v>=0:
                vals.append(v)
        br=mem.get("bound_strengthening_last_result")
        if isinstance(br,dict):
            for path in (
                ("bounds","VEHICLES"),
                ("components","vehicle_lb"),
            ):
                obj=br
                try:
                    for p in path:
                        obj=obj[p]
                    v=self._i(obj)
                    if v is not None: vals.append(v)
                except Exception:
                    pass
        return max(vals) if vals else 1

    def _demand_capacity_vehicle_lb(self, desc):
        """
        Bound seguro para MULTI-TRIP:
        demanda/capacidade sozinha normalmente limita VIAGENS, não veículos.
        Portanto só eleva vehicle-LB quando multi-trip está DESATIVADO.
        """
        allow_multi=bool(desc.get("allow_multi_trip",False))
        if allow_multi:
            return 1, {"applicable":False,"reason":"MULTI_TRIP_ACTIVE"}
        clients=self._client_rows()
        vehicles=self._vehicle_table()
        if not clients or not vehicles:
            return 1, {"applicable":False,"reason":"DATA_UNAVAILABLE"}
        total_demand=sum(max(0.0,r["demand"]) for r in clients)
        caps=sorted((max(0.0,v["capacity"]) for v in vehicles),reverse=True)
        acc=0.0
        for k,c in enumerate(caps,1):
            acc+=c
            if acc+1e-9>=total_demand:
                return k, {
                    "applicable":True,
                    "total_demand":total_demand,
                    "top_k_capacity":acc,
                }
        return len(caps)+1, {
            "applicable":True,
            "total_demand":total_demand,
            "total_capacity":sum(caps),
        }

    def _service_work_vehicle_lb(self, desc):
        """
        Relaxação global segura: cada cliente consome service_time de algum
        veículo. Ignora viagem, espera e deslocamento, logo é otimista.
        """
        clients=self._client_rows()
        if not clients:
            return 1, {"applicable":False}
        depot_candidates={str(min((r["id"] for r in clients), default=0))}
        n=sum(1 for r in clients if str(r["id"]) not in depot_candidates)
        service=self._f(desc.get("service_time_per_delivery_minutes"),0.0)
        max_time=self._f(desc.get("max_time_per_vehicle_minutes"),0.0)
        if max_time<=0:
            return 1, {"applicable":False}
        import math
        work=max(0.0,n*service)
        return max(1,int(math.ceil(work/max_time-1e-12))), {
            "applicable":True,
            "mandatory_service_minutes":work,
            "max_time_per_vehicle":max_time,
        }

    def _global_safe_vehicle_lb(self, translated_problem=None):
        desc=self._problem_desc(translated_problem)
        known=self._known_vehicle_lb()
        cap_lb,cap_ev=self._demand_capacity_vehicle_lb(desc)
        svc_lb,svc_ev=self._service_work_vehicle_lb(desc)
        lb=max(1,known,cap_lb,svc_lb)
        return lb,{
            "known_vehicle_lb":known,
            "capacity_vehicle_lb":cap_lb,
            "capacity_evidence":cap_ev,
            "service_work_vehicle_lb":svc_lb,
            "service_work_evidence":svc_ev,
        }

    # --------------------- busca HARD-K ----------------------
    def _make_directive(self,k,seed=0):
        return {
            "type":"FLEET_REDUCTION",
            "target_max_vehicles":int(k),
            "hard_constraint":True,
            "purpose":"K_VEHICLE_FEASIBILITY_CERTIFICATION_SUPPORT",
            "source":"KVehicleFeasibilityCertifier",
            "seed":int(seed),
        }

    def _call_hard_k_solver(
        self, ga_solver, classification, translated_problem,
        k, time_limit_seconds, seed
    ):
        """
        O GA é usado apenas como gerador de TESTEMUNHO FEASIBLE.
        Falha do GA => UNKNOWN, nunca PROVEN_INFEASIBLE.
        """
        if ga_solver is None:
            return None
        import copy, json
        if isinstance(translated_problem,str):
            try:
                payload=json.loads(translated_problem)
            except Exception:
                return None
        else:
            payload=copy.deepcopy(translated_problem)
        if not isinstance(payload,dict):
            return None
        payload["_continuity_directive"]=self._make_directive(k,seed)
        payload["continuity_mode"]=True
        try:
            return ga_solver.solve_problem(
                classification,
                payload,
                time_limit_seconds=max(1,int(time_limit_seconds)),
                seed=int(seed),
                continuity_mode=True,
            )
        except TypeError:
            try:
                return ga_solver.solve_problem(
                    classification,
                    payload,
                    time_limit_seconds=max(1,int(time_limit_seconds)),
                    seed=int(seed),
                )
            except Exception:
                return None
        except Exception:
            return None

    # --------------------- auditor independente ----------------------
    def _audit_candidate(self,auditor,candidate):
        """
        Compatibilidade defensiva. O agente NÃO se auto-certifica:
        se não souber chamar o Auditor, retorna None e não promove FEASIBLE
        para evidência certificada.
        """
        if auditor is None or not isinstance(candidate,dict):
            return None

        # Continuidade normalmente fornece um callback(candidate, label).
        if callable(auditor):
            try:
                r = auditor(candidate, "K_VEHICLE_FEASIBILITY_CERTIFIER")
            except TypeError:
                try:
                    r = auditor(candidate)
                except Exception as e:
                    return False, {"approved": False, "error": str(e)}
            except Exception as e:
                return False, {"approved": False, "error": str(e)}

            # Callback do Orchestrator costuma retornar (diagnóstico, recommendation).
            if isinstance(r, tuple) and len(r) >= 2:
                diag, rec = r[0], r[1]
                if isinstance(rec, dict):
                    ok = rec.get("solucao_aprovada")
                    if ok is None:
                        ok = rec.get("approved", rec.get("feasible", rec.get("valid")))
                    return bool(ok), {"diagnostic": diag, "recommendation": rec}
            if isinstance(r, dict):
                ok = r.get("solucao_aprovada")
                if ok is None:
                    ok = r.get("approved", r.get("feasible", r.get("valid")))
                return bool(ok), r
            if isinstance(r, bool):
                return r, {"approved": r}

        for name in ("audit_solution","audit","validate_solution","run"):
            fn=getattr(auditor,name,None)
            if not callable(fn):
                continue
            for args in ((candidate,),):
                try:
                    r=fn(*args)
                    if isinstance(r,dict):
                        ok = r.get("approved")
                        if ok is None: ok=r.get("feasible")
                        if ok is None: ok=r.get("valid")
                        return bool(ok),r
                    if isinstance(r,bool):
                        return r,{"approved":r}
                except TypeError:
                    continue
                except Exception as e:
                    return False,{"approved":False,"error":str(e)}
        return None



    # --------------------- K-PROOF  INDEPENDENTE/HÍBRIDO ----------------------
    def _incumbent_vehicle_ids(self, incumbent):
        ids = set()
        def walk(obj):
            if isinstance(obj, dict):
                path = obj.get("path", obj.get("route", []))
                active = bool(obj.get("used") or obj.get("trips") or obj.get("routes"))
                active = active or (isinstance(path, (list, tuple)) and any(str(x) not in ("0", "0.0") for x in path))
                vid = obj.get("physical_vehicle_id", obj.get("vehicle_id", obj.get("id_veiculo")))
                if vid is not None and active:
                    ids.add(_vrp_vehicle_id(vid))
                for value in obj.values():
                    if isinstance(value, (dict, list, tuple)): walk(value)
            elif isinstance(obj, (list, tuple)):
                for value in obj: walk(value)
        walk(incumbent)
        return sorted(ids)

    def _fast_subset_proof(self, subset_vehicles, full_problem):
        import math
        required = ("customers", "client_rows", "service_time", "max_time", "allow_multi_trip")
        missing = [key for key in required if key not in full_problem]
        if missing:
            raise ValueError(f"Contrato Exact-K incompleto: {missing}")
        C = list(full_problem["customers"])
        if not C:
            return {"proven_infeasible":False, "reasons":[]}
        caps = [float(v["capacity"]) for v in subset_vehicles]
        demands = [float(full_problem["client_rows"][i]["demand"]) for i in C]
        service = float(full_problem["service_time"])
        H = float(full_problem["max_time"])
        if any(not math.isfinite(x) or x < 0 for x in caps+demands+[service]) or not math.isfinite(H) or H <= 0:
            raise ValueError("Capacidades, demandas e tempos inválidos no contrato Exact-K")
        reasons = []
        qmax = max(caps, default=0.0)
        if not caps or max(demands) > qmax + 1e-9:
            reasons.append("CLIENT_DEMAND_EXCEEDS_EVERY_SELECTED_VEHICLE_CAPACITY")
        if not full_problem["allow_multi_trip"] and sum(demands) > sum(caps)+1e-9:
            reasons.append("TOTAL_DEMAND_EXCEEDS_SELECTED_CAPACITY_NO_MULTITRIP")
        work = len(C)*service
        travel = full_problem.get("travel_time")
        if travel:
            nodes = [0]+C
            degree = max(sum(min(travel[j,i] for j in nodes if j != i) for i in C),
                         sum(min(travel[i,j] for j in nodes if j != i) for i in C))
            trip_lb = max(1, math.ceil(sum(demands)/qmax-1e-12)) if qmax > 0 else 1
            depot_lb = trip_lb*(min(travel[0,i] for i in C)+min(travel[i,0] for i in C))
            work += max(degree, depot_lb)
        if work > len(caps)*H + 1e-7:
            reasons.append("MANDATORY_WORK_EXCEEDS_SELECTED_DAILY_TIME")
        wins = full_problem.get("windows", {})
        if wins and not reasons:
            intervals = [tuple(map(float, wins[i])) for i in C]
            temporal_lb = full_problem.get("_k_temporal_lb_v5")
            if temporal_lb is None:
                temporal_lb, _ = _vrp_interval_vehicle_lb(intervals, service)
            if temporal_lb > len(caps):
                reasons.append("MANDATORY_SERVICE_INTERVAL_ENERGY_EXCEEDS_K")
        return {"proven_infeasible":bool(reasons), "reasons":reasons,
                "mandatory_work_lower_bound":work}

    def _exact_k_with_global_certifier(self, k, optimality_certifier, time_limit_seconds, incumbent=None):
        import time, copy, itertools, math, os
        t0 = time.monotonic()
        deadline = t0 + max(0.0, float(time_limit_seconds))
        mem = self.shared_memory.data
        def unknown(reason, **extra):
            return dict(status="UNKNOWN", global_valid=False, target_k=int(k), reason=reason,
                        elapsed_seconds=time.monotonic()-t0, **extra)
        if optimality_certifier is None:
            return unknown("EXACT_MODEL_PROVIDER_NOT_AVAILABLE")
        try:
            optimality_certifier._load_objective_context()
            full = optimality_certifier._build_problem()
            vehicles = list(full["vehicles"])
            n = len(vehicles)
            if not 1 <= k <= n: return unknown("INVALID_K")
            if full["must_use_all_vehicles"] and k < n:
                return dict(status="PROVEN_INFEASIBLE", global_valid=True, target_k=k,
                            proven_vehicle_lower_bound=n, reason="MUST_USE_ALL_VEHICLES_EXCEEDS_K")
            import multiprocessing as mp
            ctx = mp.get_context("fork")  # Colab/Linux; falha explícita em plataformas sem fork.
        except Exception as exc:
            return unknown("EXACT_SETUP_ERROR", error=f"{type(exc).__name__}: {exc}")
        scope = _vrp_fingerprint(full)
        root = mem.setdefault("k_exact_persistent_state", {})
        state = root.setdefault(f"V5|{scope}|K={k}", {"records":{}, "calls":0,
                              "global_fallback_attempts":0, "total_solver_seconds":0.0})
        state["calls"] += 1
        total = math.comb(n,k)
        # Evita materializar milhões de combinações: GLOBAL-K continua exato.
        enumerate_subsets = total <= 10000
        subsets = list(itertools.combinations(range(n),k)) if enumerate_subsets else []
        vids = [_vrp_vehicle_id(v["vehicle_id"]) for v in vehicles]
        incids = set(self._incumbent_vehicle_ids(incumbent))
        def key(sub): return ",".join(map(str,sub))
        # As janelas não mudam entre subsets: separar a mesma energia 66 vezes
        # desperdiçaria o orçamento antes da construção do primeiro modelo.
        fast_problem = dict(full)
        if full.get("windows"):
            fast_problem["_k_temporal_lb_v5"], _ = _vrp_interval_vehicle_lb(
                [tuple(map(float,full["windows"][i])) for i in full["customers"]],
                float(full["service_time"]))
        for sub in subsets:
            if time.monotonic() >= deadline: break
            previous = state["records"].get(key(sub), {})
            if previous.get("status") == "infeasible": continue
            fast = self._fast_subset_proof([vehicles[i] for i in sub],fast_problem)
            if fast["proven_infeasible"]:
                state["records"][key(sub)] = dict(previous, status="infeasible",
                    vehicle_ids=[vids[i] for i in sub], proof_type="FAST_ANALYTICAL_SAFE", evidence=fast)
        def closed_count(): return sum(r.get("status")=="infeasible" for r in state["records"].values())
        def proven(reason):
            return dict(status="PROVEN_INFEASIBLE", global_valid=True, target_k=k,
                        proven_vehicle_lower_bound=k+1, reason=reason,
                        proven_infeasible_subsets=total, total_subsets=total,
                        elapsed_seconds=time.monotonic()-t0)
        if closed_count() == total: return proven("ALL_K_SUBSETS_CLOSED")
        def rank(sub):
            rec = state["records"].get(key(sub), {})
            # Justiça entre pares: nenhum par recebe a 2a tentativa antes dos não tentados.
            return (int(rec.get("attempts",0)),
                    -len(set(vids[i] for i in sub)&incids),
                    -sum(float(vehicles[i]["capacity"]) for i in sub), sub)
        pending = sorted((s for s in subsets if state["records"].get(key(s),{}).get("status") != "infeasible"), key=rank)
        workers = max(1,min(2,(os.cpu_count() or 2)-1))
        print(f"      🧠 K-CERTIFIER V5 | K={k} | fechados={closed_count()}/{total} | workers={workers}")
        print(f"         prioridade incumbent={sorted(incids)} | ledger persistente; árvore SCIP não persistida")

        def worker(spec,q):
            started = time.monotonic()
            out = {"job_id":spec["job_id"], "payload":spec["payload"], "quota":spec["quota"]}
            model = None
            try:
                provider = copy.copy(optimality_certifier)
                provider.problem = copy.deepcopy(full)
                sub = spec["payload"]
                if sub is not None:
                    provider.problem["vehicles"] = [copy.deepcopy(vehicles[i]) for i in sub]
                provider.verbose_solver = False
                provider.scip_threads = 1
                provider.relative_gap_target = 0.0
                provider.time_limit_seconds = max(1,int(spec["quota"]))
                # Prova de factibilidade SEM cutoff no objetivo econômico.
                model,h = provider._build_scip_model(incumbent_ub=None, feasibility_only=True, max_active_vehicles=k)
                build_elapsed = time.monotonic()-started
                remaining = min(spec["quota"]-build_elapsed, deadline-time.monotonic()-0.2)
                out.update(build_seconds=build_elapsed, model_stats=copy.deepcopy(provider._model_stats))
                if remaining <= 0:
                    out.update(status="unknown", error="BUILD_BUDGET_EXHAUSTED")
                else:
                    model.setParam("limits/time",float(remaining))
                    model.setParam("limits/gap",0.0)
                    model.setParam("limits/solutions",1)
                    model.optimize()
                    sol = model.getBestSol()
                    # Custos incondicionais referem-se à frota original, mesmo num subset.
                    provider.problem = full
                    candidate = provider._extract_solution(model,h) if sol is not None else None
                    out.update(status=str(model.getStatus()).lower(), candidate=candidate,
                               has_model_witness=sol is not None, nodes=int(model.getNNodes()),
                               solve_seconds=float(model.getSolvingTime()))
            except Exception as exc:
                out.update(status="error", error=f"{type(exc).__name__}: {exc}")
            out["elapsed"] = time.monotonic()-started
            q.put(out)

        records = []
        def consume(results):
            for rec in results:
                sub = rec.get("payload")
                if rec.get("error") == "CANCELLED_AFTER_WITNESS": continue
                if sub is None:
                    state["global_fallback_attempts"] += 1
                    state["last_global_fallback"] = {a:b for a,b in rec.items() if a != "candidate"}
                    print(f"         GLOBAL-K: {rec.get('status')} | construção={rec.get('build_seconds',0):.1f}s | nós={rec.get('nodes',0)}")
                    if rec.get("status") == "infeasible": return proven("GLOBAL_K_SCIP_PROVEN_INFEASIBLE")
                else:
                    old = state["records"].get(key(sub),{})
                    stored = {a:b for a,b in rec.items() if a != "candidate"}
                    stored.update(attempts=int(old.get("attempts",0))+1, vehicle_ids=[vids[i] for i in sub],
                                  cumulative_elapsed_seconds=float(old.get("cumulative_elapsed_seconds",0))+float(rec.get("elapsed",0)))
                    state["records"][key(sub)] = stored
                    records.append(stored)
                    print(f"         Exact-K {stored['vehicle_ids']}: {stored.get('status')} | tentativa={stored['attempts']} | nós={stored.get('nodes',0)}")
                state["total_solver_seconds"] += float(rec.get("solve_seconds",0))
                if rec.get("has_model_witness"):
                    return dict(status="FEASIBLE" if isinstance(rec.get("candidate"),dict) else "FEASIBLE_MODEL_WITNESS",
                                global_valid=True, target_k=k, candidate=rec.get("candidate"),
                                reason="EXACT_K_WITNESS_REQUIRES_INDEPENDENT_AUDIT",
                                elapsed_seconds=time.monotonic()-t0)
            return None

        # Rodadas 1,4,7... dedicam orçamento substancial ao GLOBAL-K.
        # Outras rodadas cobrem os pares pendentes, sem fallback de poucos segundos.
        remaining = deadline-time.monotonic()
        global_round = not enumerate_subsets or state["calls"]%3 == 1
        if global_round and remaining >= 3:
            quota = remaining*(0.70 if pending else 0.98)
            spec = {"job_id":"GLOBAL", "payload":None, "quota":quota}
            answer = consume(_vrp_collect_jobs(ctx,[spec],worker,min(deadline,time.monotonic()+quota)))
            if answer: return answer
        while pending and deadline-time.monotonic() >= 3:
            wave = pending[:workers]; pending = pending[workers:]
            remaining = deadline-time.monotonic()
            specs = []
            for sub in wave:
                attempts = int(state["records"].get(key(sub),{}).get("attempts",0))
                quota = min(remaining-0.2,120.0*(2**min(attempts,2)))
                specs.append({"job_id":key(sub), "payload":sub, "quota":quota})
            answer = consume(_vrp_collect_jobs(ctx,specs,worker,min(deadline,time.monotonic()+max(s["quota"] for s in specs))))
            if answer: return answer
            if closed_count() == total: return proven("ALL_K_SUBSETS_CLOSED")
        return unknown("EXACT_K_PROOF_INCOMPLETE", more_time_required=True,
                       proven_infeasible_subsets=closed_count(), total_subsets=total,
                       unresolved_subsets=total-closed_count(), tested_subsets_this_call=len(records),
                       solver_seconds_accumulated=state["total_solver_seconds"],
                       global_fallback=state.get("last_global_fallback"),
                       persistence="LEDGER_ONLY_NOT_BRANCH_AND_BOUND_TREE")

    # --------------------- persistência da prova ----------------------
    def _persist(self,result):
        mem=self._mem()
        mem["k_vehicle_feasibility_last_result"]=result
        hist=mem.setdefault("k_vehicle_feasibility_history",[])
        if isinstance(hist,list):
            hist.append(result)
            if len(hist)>100:
                del hist[:-100]

        if result.get("status")=="PROVEN_INFEASIBLE":
            new_lb=self._i(result.get("proven_vehicle_lower_bound"))
            if new_lb is not None:
                old=self._i(mem.get("certified_vehicle_lower_bound"),1) or 1
                mem["certified_vehicle_lower_bound"]=max(old,new_lb)
                mem["continuity_best_vehicle_lower_bound"]=max(
                    self._i(mem.get("continuity_best_vehicle_lower_bound"),1) or 1,
                    new_lb
                )
        return result

    # --------------------- API principal ----------------------
    def certify(
        self,
        target_k,
        translated_problem=None,
        classification=None,
        incumbent=None,
        ga_solver=None,
        auditor=None,
        time_limit_seconds=120,
        seed=101,
        exact_prover=None,
        optimality_certifier=None,
        exact_k_enabled=True,
    ):
        """
        exact_prover é opcional e deve obedecer contrato estrito:
          prove_k_infeasible(k, translated_problem, classification, time_limit_seconds)
        retornando dict com:
          status = PROVEN_INFEASIBLE | FEASIBLE | UNKNOWN
          global_valid = bool

        Somente global_valid=True + PROVEN_INFEASIBLE eleva o LB.
        """
        import time, copy
        t0=time.monotonic()
        _vrp_ensure_proof_scope(self.shared_memory)
        k=self._i(target_k)
        if k is None or k<1:
            return self._persist({
                "agent_version":self.VERSION,
                "status":"CONFIGURATION_ERROR",
                "message":"target_k deve ser inteiro >=1.",
            })

        fleet=self._vehicle_table()
        if fleet and k>=len(fleet):
            return self._persist({
                "agent_version": self.VERSION,
                "target_k": k,
                "status": "UNKNOWN",
                "global_valid": False,
                "message": "K não restringe a cardinalidade, mas isso não prova factibilidade operacional.",
                "reason": "CARDINALITY_NON_BINDING_NO_FEASIBILITY_PROOF",
            })

        safe_lb,lb_ev=self._global_safe_vehicle_lb(translated_problem)

        result={
            "agent_version":self.VERSION,
            "target_k":k,
            "status":"UNKNOWN",
            "global_valid":False,
            "safe_vehicle_lower_bound_before":safe_lb,
            "lower_bound_evidence":lb_ev,
            "candidate":None,
            "audited":False,
            "proven_vehicle_lower_bound":None,
            "proof_source":None,
            "elapsed_seconds":0.0,
        }

        # 1) Prova imediata por lower bound global já conhecido.
        if safe_lb>k:
            result.update({
                "status":"PROVEN_INFEASIBLE",
                "global_valid":True,
                "proven_vehicle_lower_bound":safe_lb,
                "proof_source":"GLOBAL_SAFE_VEHICLE_LOWER_BOUND",
                "message":f"LB global de veículos={safe_lb} exclui K={k}.",
                "elapsed_seconds":time.monotonic()-t0,
            })
            return self._persist(result)

        # 2) Incumbente já pode ser testemunho <=K.
        inc_used=self._used_vehicles(incumbent)
        if (
            isinstance(incumbent,dict)
            and inc_used is not None and inc_used<=k
            and self._candidate_feasible_flag(incumbent)
        ):
            aud=self._audit_candidate(auditor,incumbent)
            if aud is not None and aud[0]:
                result.update({
                    "status":"FEASIBLE",
                    "candidate":copy.deepcopy(incumbent),
                    "audited":True,
                    "audit":aud[1],
                    "proof_source":"AUDITED_INCUMBENT_WITNESS",
                    "message":f"Incumbente auditada usa {inc_used} <= K={k}.",
                    "elapsed_seconds":time.monotonic()-t0,
                })
                return self._persist(result)

        # 3) Busca heurística HARD-K: serve SOMENTE para encontrar testemunho.
        remaining=max(1.0,float(time_limit_seconds)-(time.monotonic()-t0))
        hist=self._mem().get("k_vehicle_feasibility_history",[]) or []
        prior_unknown=sum(1 for r in hist if isinstance(r,dict) and self._i(r.get("target_k"),-1)==k and str(r.get("status","")).upper()=="UNKNOWN")
        # Primeira tentativa ainda dá espaço ao GA; após UNKNOWN repetido, transfere orçamento ao Exact-K.
        ga_fraction = 0.10 if prior_unknown == 0 else 0.0
        ga_budget = int(min(remaining * ga_fraction, remaining)) if ga_fraction > 0 else 0
        result["prior_unknown_same_k"] = int(prior_unknown)
        result["ga_budget_seconds"] = int(ga_budget)
        result["ga_hard_k_suspended"] = bool(prior_unknown >= 1)
        if ga_budget >= 2:
            cand = self._call_hard_k_solver(
                ga_solver, classification, translated_problem, k, ga_budget, seed
            )
        else:
            cand = None
            if prior_unknown >= 1:
                print(f"      ⏭️ GA HARD-K suspenso para K={k}: tentativa anterior UNKNOWN; orçamento transferido ao Exact-K.")
        cand_used = self._used_vehicles(cand)

        if (
            isinstance(cand,dict)
            and cand_used is not None and cand_used<=k
            and self._candidate_feasible_flag(cand)
        ):
            aud=self._audit_candidate(auditor,cand)
            if aud is not None and aud[0]:
                result.update({
                    "status":"FEASIBLE",
                    "candidate":copy.deepcopy(cand),
                    "audited":True,
                    "audit":aud[1],
                    "proof_source":"AUDITED_HARD_K_WITNESS",
                    "message":f"GA HARD-K encontrou testemunho auditado com {cand_used} <= K={k}.",
                    "elapsed_seconds":time.monotonic()-t0,
                })
                return self._persist(result)

        # 4) EXACT-K integrado.
        # Falha heurística continua sem valor de prova.
        remaining=max(0.0,float(time_limit_seconds)-(time.monotonic()-t0))

        if (
            bool(exact_k_enabled)
            and optimality_certifier is not None
            and remaining>=1.0
        ):
            proof=self._exact_k_with_global_certifier(
                k=k,
                optimality_certifier=optimality_certifier,
                time_limit_seconds=max(1,int(remaining)),
                incumbent=incumbent,
            )

            if isinstance(proof,dict):
                ps=str(proof.get("status","UNKNOWN") or "UNKNOWN").upper()
                gv=bool(proof.get("global_valid",False))

                if ps=="PROVEN_INFEASIBLE" and gv:
                    result.update({
                        "status":"PROVEN_INFEASIBLE",
                        "global_valid":True,
                        "proven_vehicle_lower_bound":max(
                            k+1,
                            self._i(
                                proof.get("proven_vehicle_lower_bound"),
                                k+1
                            ) or (k+1)
                        ),
                        "proof_source":"INTEGRATED_EXACT_K_SCIP",
                        "exact_proof":proof,
                        "message":(
                            f"SCIP Exact-K provou globalmente que não existe "
                            f"solução com <= K={k} veículos."
                        ),
                        "elapsed_seconds":time.monotonic()-t0,
                    })
                    return self._persist(result)

                if ps in ("FEASIBLE","FEASIBLE_MODEL_WITNESS"):
                    pc=proof.get("candidate")
                    pu=self._used_vehicles(pc)

                    # Para promoção a FEASIBLE externo, ainda exigimos o Auditor.
                    if (
                        isinstance(pc,dict)
                        and pu is not None
                        and pu<=k
                    ):
                        aud=self._audit_candidate(auditor,pc)
                        if aud is not None and aud[0]:
                            result.update({
                                "status":"FEASIBLE",
                                "candidate":copy.deepcopy(pc),
                                "audited":True,
                                "audit":aud[1],
                                "proof_source":"INTEGRATED_EXACT_K_SCIP_WITNESS",
                                "exact_proof":proof,
                                "message":(
                                    f"SCIP Exact-K encontrou testemunho auditado "
                                    f"com {pu} <= K={k}."
                                ),
                                "elapsed_seconds":time.monotonic()-t0,
                            })
                            return self._persist(result)

                    # O modelo encontrou existência, mas não conseguimos promover
                    # uma solução operacional sem extração + Auditor.
                    result["exact_proof"]=proof
                    result["exact_model_found_feasible"]=True

        # 5) Prover externo opcional mantido por compatibilidade/extensibilidade.
        remaining=max(0.0,float(time_limit_seconds)-(time.monotonic()-t0))
        if exact_prover is not None and remaining>=1.0:
            fn=getattr(exact_prover,"prove_k_infeasible",None)
            if callable(fn):
                try:
                    proof=fn(
                        k,
                        translated_problem,
                        classification,
                        max(1,int(remaining)),
                    )
                except Exception as e:
                    proof={"status":"UNKNOWN","global_valid":False,"error":str(e)}
                if isinstance(proof,dict):
                    ps=str(proof.get("status","UNKNOWN") or "UNKNOWN").upper()
                    gv=bool(proof.get("global_valid",False))
                    if ps=="PROVEN_INFEASIBLE" and gv:
                        result.update({
                            "status":"PROVEN_INFEASIBLE",
                            "global_valid":True,
                            "proven_vehicle_lower_bound":k+1,
                            "proof_source":"EXTERNAL_EXACT_K_PROVER",
                            "exact_proof":proof,
                            "message":f"Prova exata global excluiu todas as soluções com <= K={k}.",
                            "elapsed_seconds":time.monotonic()-t0,
                        })
                        return self._persist(result)
                    if ps=="FEASIBLE":
                        pc=proof.get("candidate")
                        pu=self._used_vehicles(pc)
                        if isinstance(pc,dict) and pu is not None and pu<=k:
                            aud=self._audit_candidate(auditor,pc)
                            if aud is not None and aud[0]:
                                result.update({
                                    "status":"FEASIBLE",
                                    "candidate":copy.deepcopy(pc),
                                    "audited":True,
                                    "audit":aud[1],
                                    "proof_source":"EXTERNAL_EXACT_PROVER_FEASIBLE_WITNESS",
                                    "exact_proof":proof,
                                    "message":f"Prover encontrou testemunho auditado <= K={k}.",
                                    "elapsed_seconds":time.monotonic()-t0,
                                })
                                return self._persist(result)

        if "proof" in locals() and isinstance(proof, dict):
            result["exact_proof"] = copy.deepcopy(proof)
            result["more_time_required"] = bool(proof.get("more_time_required", False))
            result["proven_infeasible_subsets"] = proof.get("proven_infeasible_subsets")
            result["unresolved_subsets"] = proof.get("unresolved_subsets")
            result["total_subsets"] = proof.get("total_subsets")

        result.update({
            "status":"UNKNOWN",
            "global_valid":False,
            "message":(
                f"Exact-K ainda não fechou K={k}. O histórico de tentativas foi preservado (sem árvore SCIP); "
                "é necessário mais tempo para concluir os subconjuntos pendentes."
                if result.get("more_time_required")
                else f"Não foi obtida prova para K={k}. Falha/NOT_FOUND do GA não implica inviabilidade."
            ),
            "hard_k_candidate_status": (
                cand.get("status") if isinstance(cand,dict) else None
            ),
            "hard_k_candidate_used_vehicles":cand_used,
            "elapsed_seconds":time.monotonic()-t0,
        })
        return self._persist(result)

    def attempt_elimination(self, inputs, baseline, seconds, folder, state):
        return _v512_structural_finish(_v58_eliminate, inputs, baseline, seconds, folder, state)
````

### **CONTINUITYEXACTCERTIFICATIONAGENT — CERTIFICAÇÃO EXATA NA CONTINUIDADE**

Algumas perguntas exigem uma investigação matemática maior. Este agente oferece esse trabalho exato com limite de tempo, dentro da continuidade.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class ContinuityExactCertificationAgent(OptimalityCertificationAgent):
    """Certificador exato da continuidade, com métodos comuns herdados do certificador.

    Mantém a formulação e a extração específicas desta etapa. Uma conclusão de
    ótimo exige evidência global válida e conferência independente do plano,
    respeitando as tolerâncias numéricas e os recursos disponíveis."""
    VERSION = 'SCIP_CASCADE_CERTIFIER_17_CUSTOMER_CHAIN'
    STATUS_RUNNING = 'CERTIFICATION_RUNNING'
    STATUS_OPTIMAL = 'PROVEN_OPTIMAL'
    STATUS_INFEASIBLE = 'PROVEN_INFEASIBLE'
    STATUS_TIME_LIMIT = 'TIME_LIMIT'
    STATUS_RESOURCE_LIMIT = 'RESOURCE_LIMIT'
    STATUS_ERROR = 'CERTIFICATION_ERROR'
    STATUS_NOT_PROVEN = 'NOT_PROVEN'

    def _estimate_model_complexity(self, arcs, strengthened_bounds):
        """Estima o tamanho do MILP antes de alocar variáveis SCIP."""
        p = self.problem
        n = len(p['customers'])
        nv = len(p['vehicles'])
        K = int(strengthened_bounds.get('trip_slots_safe_upper_bound', n))
        K = max(0, K)
        arc_bin = n * max(0, n - 1) * (2 if p['allow_multi_trip'] else 1)
        visit_bin = 2 * n
        assign_bin = n * nv
        vehicle_bin = nv
        total_bin = arc_bin + visit_bin + assign_bin + vehicle_bin
        seq = 2 * n * max(0, n - 1)
        reasons = []
        if arc_bin > self.max_exact_arc_binaries:
            reasons.append('ARC_BINARIES')
        if total_bin > self.max_exact_total_binary_variables:
            reasons.append('TOTAL_BINARIES')
        if seq > self.max_exact_sequence_constraints:
            reasons.append('SEQUENCE_CONSTRAINTS')
        stats = {'n_customers': n, 'n_vehicles': nv, 'trip_slots': K, 'feasible_arcs_per_slot': len(arcs), 'estimated_binary_arc_variables': arc_bin, 'estimated_binary_visit_variables': visit_bin, 'estimated_binary_assignment_variables': assign_bin, 'estimated_binary_vehicle_variables': vehicle_bin, 'estimated_total_binary_variables': total_bin, 'estimated_sequence_constraints': seq, 'guard_triggered': bool(self.enable_model_complexity_guard and reasons), 'guard_reasons': reasons, 'thresholds': {'max_exact_arc_binaries': self.max_exact_arc_binaries, 'max_exact_total_binary_variables': self.max_exact_total_binary_variables, 'max_exact_sequence_constraints': self.max_exact_sequence_constraints}}
        self.shared_memory.data['certification_model_complexity_estimate'] = copy.deepcopy(stats)
        return stats

    def _build_scip_model(self, incumbent_ub=None, feasibility_only=False, max_active_vehicles=None):
        """Exact customer-chain model: direct links or explicit depot reload links.

        A physical vehicle is one chronological chain; each depot link splits it
        into trips. No artificial trip slots or arc copies indexed by trip.
        Waiting at customers is counted; idle gaps at the depot are not counted.
        """
        Model, qs = self._import_scip()
        p = self.problem
        C = p['customers']
        V = p['vehicles']
        n = len(C)
        all_vehicles = list(V)
        dominance = bool(feasibility_only and max_active_vehicles is not None and (not p['must_use_all_vehicles']) and (0 < int(max_active_vehicles) < len(V)))
        if dominance:
            V = sorted(V, key=lambda v: (-float(v['capacity']), v['idx']))[:int(max_active_vehicles)]
        op0 = float(p['operation_start'])
        op1 = float(p['operation_end'])
        span = op1 - op0
        H = float(p['max_time'])
        svc = float(p['service_time'])
        tt = p['travel_time']
        dd = p['distance']
        Q = max((float(v['capacity']) for v in V))
        demand = {i: float(p['client_rows'][i]['demand']) for i in C}
        windows = {i: (max(op0, float(p['windows'][i][0])), min(op1, float(p['windows'][i][1]))) for i in C}
        direct = [(i, j) for i in C for j in C if i != j and windows[i][0] + svc + tt[i, j] <= windows[j][1] + 1e-09 and (demand[i] + demand[j] <= Q + 1e-09)]
        reloads = [(i, j) for i in C for j in C if i != j and windows[i][0] + svc + tt[i, 0] + tt[0, j] <= windows[j][1] + 1e-09] if p['allow_multi_trip'] else []
        pairs = sorted(set(direct) | set(reloads))
        b = self._analytical_lower_bounds(incumbent_ub=incumbent_ub)
        scf = _vrp_scf_distance_bound(p, b['trips_lb'], 30.0, self.shared_memory.data.setdefault('scf_model_cache_v52', {}))
        if scf.get('valid') and scf.get('distance_lb') is not None:
            b['distance_lb'] = max(float(b['distance_lb']), float(scf['distance_lb']) - 1e-07)
        fixed = sorted((float(v['fixed_cost']) for v in V))
        if min(fixed) >= 0 and min((float(v['cost_per_km']) for v in V)) >= 0:
            flb = sum(fixed) if p['fixed_cost_mode'] != 'WHEN_USED' or p['must_use_all_vehicles'] else sum(fixed[:int(b['vehicles_lb'])])
            b['cost_lb'] = max(float(b['cost_lb']), flb + min((float(v['cost_per_km']) for v in V)) * b['distance_lb'])
        b['scf_model_result'] = scf
        coeff = self.objective_context['coefficients']
        b['canonical_objective_lb'] = sum((float(coeff.get(k, 0)) * float(b[f]) for k, f in [('DISTANCE', 'distance_lb'), ('COST', 'cost_lb'), ('TIME', 'time_lb'), ('VEHICLES', 'vehicles_lb')]))
        model = Model('VRPTW_CUSTOMER_CHAIN_V52')
        if not self.verbose_solver:
            model.hideOutput()
        model.setParam('limits/time', float(self.time_limit_seconds))
        model.setParam('limits/gap', float(self.relative_gap_target))
        if self.scip_memory_mb > 0:
            model.setParam('limits/memory', float(self.scip_memory_mb))
        if self.scip_threads > 0:
            model.setParam('parallel/maxnthreads', int(self.scip_threads))
        from pyscipopt import SCIP_PARAMSETTING
        model.setHeuristics(SCIP_PARAMSETTING.FAST)
        for heuristic in ('adaptivediving', 'indicatordiving', 'farkasdiving', 'conflictdiving'):
            model.setParam(f'heuristics/{heuristic}/freq', -1)
        model.setParam('separating/maxroundsroot', 8)
        model.setParam('separating/maxstallroundsroot', 3)
        model.setParam('branching/relpscost/initcand', 10)
        model.setParam('branching/relpscost/inititer', 100)
        model.setParam('branching/relpscost/maxlookahead', 3)
        model.setParam('branching/relpscost/sbiterofs', 1000)
        if feasibility_only:
            model.setParam('separating/maxroundsroot', 2)
            model.setParam('separating/maxstallroundsroot', 1)
            model.setParam('separating/maxcutsroot', 100)
        x = {(i, j): model.addVar(vtype='B', name=f'x_{i}_{j}') for i, j in direct}
        r = {(i, j): model.addVar(vtype='B', name=f'reload_{i}_{j}') for i, j in reloads}
        y = {(i, v['idx']): model.addVar(vtype='B', name=f"veh_{i}_{v['idx']}") for i in C for v in V}
        used = {v['idx']: model.addVar(vtype='B', name=f"used_{v['idx']}") for v in V}
        first = {i: model.addVar(vtype='B', name=f'first_{i}') for i in C}
        last = {i: model.addVar(vtype='B', name=f'last_{i}') for i in C}
        tau = {i: model.addVar(lb=windows[i][0], ub=max(windows[i]), name=f'tau_{i}') for i in C}
        dep = {i: model.addVar(lb=op0, ub=op1, name=f'dep_{i}') for i in C}
        wait = {i: model.addVar(lb=0, ub=H, name=f'wait_{i}') for i in C}
        load = {i: model.addVar(lb=demand[i], ub=max(Q, demand[i]), name=f'load_{i}') for i in C}
        rank = {} if svc > 1e-06 else {i: model.addVar(lb=1, ub=max(1, n), name=f'rank_{i}') for i in C}
        qtime = {(i, v['idx']): model.addVar(lb=0, ub=H, name=f"work_{i}_{v['idx']}") for i in C for v in V}
        maxD = {i: max([float(dd[j, i]) for j in [0] + C if j != i]) + float(dd[i, 0]) for i in C}
        qdist = {} if feasibility_only else {(i, v['idx']): model.addVar(lb=0, ub=maxD[i], name=f"dist_{i}_{v['idx']}") for i in C for v in V}
        starts = {(i, v['idx']): model.addVar(lb=0, ub=1, name=f"start_{i}_{v['idx']}") for i in C for v in V}
        ends = {(i, v['idx']): model.addVar(lb=0, ub=1, name=f"end_{i}_{v['idx']}") for i in C for v in V}
        incoming = {i: [] for i in C}
        outgoing = {i: [] for i in C}
        for arcs in (x, r):
            for (i, j), var in arcs.items():
                outgoing[i].append(var)
                incoming[j].append(var)
        begin = {i: first[i] + qs((r[j, k] for j, k in reloads if k == i)) for i in C}
        finish = {i: last[i] + qs((r[j, k] for j, k in reloads if j == i)) for i in C}
        work = {}
        dist = {}
        labels = {i: qs(((k + 1) * y[i, v['idx']] for k, v in enumerate(V))) for i in C}
        for i in C:
            if windows[i][0] > windows[i][1]:
                model.addCons(tau[i] <= windows[i][1])
            model.addCons(first[i] + qs(incoming[i]) == 1)
            model.addCons(last[i] + qs(outgoing[i]) == 1)
            model.addCons(qs((y[i, v['idx']] for v in V)) == 1)
            model.addCons(load[i] <= qs((float(v['capacity']) * y[i, v['idx']] for v in V)))
            model.addCons(load[i] <= demand[i] + Q * (1 - begin[i]))
            M = max(1.0, abs(windows[i][0] - op1 - tt[0, i]), abs(windows[i][1] - op0 - tt[0, i])) + H
            model.addCons(wait[i] >= tau[i] - dep[i] - tt[0, i] - M * (1 - begin[i]))
            model.addCons(wait[i] <= tau[i] - dep[i] - tt[0, i] + M * (1 - begin[i]))
            model.addCons(tau[i] + svc + tt[i, 0] <= op1 + max(0.0, windows[i][1] + svc + tt[i, 0] - op1) * (1 - finish[i]))
            dist[i] = qs((float(dd[j, k]) * var for (j, k), var in x.items() if k == i)) + float(dd[0, i]) * begin[i] + float(dd[i, 0]) * finish[i]
            work[i] = svc + wait[i] + qs((float(tt[j, k]) * var for (j, k), var in x.items() if k == i)) + float(tt[0, i]) * begin[i] + float(tt[i, 0]) * finish[i]
            model.addCons(qs((qtime[i, v['idx']] for v in V)) == work[i])
            if qdist:
                model.addCons(qs((qdist[i, v['idx']] for v in V)) == dist[i])
            for v in V:
                vi = v['idx']
                a = y[i, vi]
                model.addCons(a <= used[vi])
                model.addCons(qtime[i, vi] <= H * a)
                if qdist:
                    model.addCons(qdist[i, vi] <= maxD[i] * a)
                for z, indicator in ((starts[i, vi], first[i]), (ends[i, vi], last[i])):
                    model.addCons(z <= a)
                    model.addCons(z <= indicator)
                    model.addCons(z >= a + indicator - 1)
        for i, j in pairs:
            edge = x.get((i, j), 0) + r.get((i, j), 0)
            model.addCons(labels[i] - labels[j] <= max(0, len(V) - 1) * (1 - edge))
            model.addCons(labels[j] - labels[i] <= max(0, len(V) - 1) * (1 - edge))
            if rank:
                model.addCons(rank[j] >= rank[i] + 1 - n * (1 - edge))
        for (i, j), var in x.items():
            delta = tau[j] - tau[i] - svc - float(tt[i, j])
            M = max(1.0, abs(windows[j][0] - windows[i][1] - svc - tt[i, j]), abs(windows[j][1] - windows[i][0] - svc - tt[i, j])) + H
            model.addCons(wait[j] >= delta - M * (1 - var))
            model.addCons(wait[j] <= delta + M * (1 - var))
            model.addCons(load[j] >= load[i] + demand[j] - (Q + demand[j]) * (1 - var))
        for (i, j), var in r.items():
            M = max(0.0, windows[i][1] + svc + tt[i, 0] - op0)
            model.addCons(dep[j] >= tau[i] + svc + float(tt[i, 0]) - M * (1 - var))
        for v in V:
            vi = v['idx']
            model.addCons(qs((starts[i, vi] for i in C)) == used[vi])
            model.addCons(qs((ends[i, vi] for i in C)) == used[vi])
            model.addCons(qs((qtime[i, vi] for i in C)) <= H * used[vi])
            model.addCons(qs((qtime[i, vi] for i in C)) >= float(p['minimum_usage']) * used[vi])
            if p['must_use_all_vehicles']:
                model.addCons(used[vi] == 1)
        count = qs(used.values())
        totalD = qs(dist.values())
        totalT = qs(work.values())
        fixedcost = qs((float(v['fixed_cost']) * used[v['idx']] for v in V)) if p['fixed_cost_mode'] == 'WHEN_USED' else sum(fixed)
        variablecost = 0 if feasibility_only else qs((float(v['cost_per_km']) * qdist[i, v['idx']] for i in C for v in V))
        totalCost = fixedcost + variablecost
        if max_active_vehicles is not None:
            model.addCons(count <= int(max_active_vehicles))
        model.addCons(count >= int(b['vehicles_lb']))
        model.addCons(qs(begin.values()) >= int(b['trips_lb']))
        model.addCons(totalD >= float(b['distance_lb']))
        model.addCons(totalT >= float(b['time_lb']))
        if not feasibility_only:
            model.addCons(totalCost >= float(b['cost_lb']))
        objective = sum((float(coeff.get(k, 0)) * expr for k, expr in [('DISTANCE', totalD), ('TIME', totalT), ('COST', totalCost), ('VEHICLES', count)]))
        if not feasibility_only and incumbent_ub is not None and self.use_incumbent_cutoff:
            model.addCons(objective <= float(incumbent_ub) + max(1e-08, abs(incumbent_ub) * 1e-09))
        model.setObjective(0 if feasibility_only else objective, 'minimize')
        self._model_stats = {'formulation': 'CUSTOMER_CHAIN_DEPOT_RELOAD_V52', 'feasibility_only': feasibility_only, 'max_active_vehicles': max_active_vehicles, 'binary_arc_variables': len(x) + len(r), 'binary_assignment_variables': len(y), 'binary_vehicle_variables': len(used), 'binary_endpoint_variables': len(first) + len(last), 'strengthened_analytical_bounds': b, 'capacity_dominance_applied': dominance, 'available_vehicle_ids': [v['vehicle_id'] for v in all_vehicles], 'modeled_vehicle_ids': [v['vehicle_id'] for v in V], 'dominance_scope': 'FEASIBILITY_ONLY_UNIFORM_OPERATIONAL_CONTRACT' if dominance else None, 'root_separation_round_limit': 2 if feasibility_only else 8, 'heuristics': 'FAST_ROUNDING_WITHOUT_DIVING', 'incumbent_cutoff_used': bool(not feasibility_only and incumbent_ub is not None and self.use_incumbent_cutoff)}
        self.shared_memory.data['certification_strengthened_bounds'] = copy.deepcopy(b)
        h = dict(C=C, V=V, x=x, r=r, y=y, used=used, first=first, last=last, tau=tau, dep=dep, wait=wait, load=load, rank=rank, qtime=qtime, qdist=qdist, starts=starts, ends=ends, total_distance=totalD, total_time=totalT, total_cost=totalCost, total_vehicles=count)
        return (model, h)

    def _extract_solution(self, model, h):
        sol = model.getBestSol()
        if sol is None:
            return None
        val = lambda v: float(model.getSolVal(sol, v))
        p = self.problem
        routes = []
        succ = {i: (j, False) for (i, j), v in h['x'].items() if val(v) > 0.5}
        succ.update({i: (j, True) for (i, j), v in h['r'].items() if val(v) > 0.5})
        for first in h['C']:
            if val(h['first'][first]) < 0.5:
                continue
            v = next((v for v in h['V'] if val(h['y'][first, v['idx']]) > 0.5))
            i = first
            seen = set()
            trip = 0
            while True:
                start = val(h['dep'][i])
                path = [0]
                details = []
                load = 0.0
                prev = 0
                while True:
                    if i in seen:
                        raise RuntimeError('Cycle in extracted customer chain')
                    seen.add(i)
                    path.append(i)
                    load += p['client_rows'][i]['demand']
                    st = val(h['tau'][i])
                    arrival = start + p['travel_time'][0, i] if prev == 0 else details[-1]['departure_time_minutes'] + p['travel_time'][prev, i]
                    details.append(dict(node_id=i, arrival_time_minutes=arrival, service_start_time_minutes=st, departure_time_minutes=st + p['service_time'], load=load))
                    nxt = succ.get(i)
                    if nxt is None or nxt[1]:
                        break
                    prev = i
                    i = nxt[0]
                path.append(0)
                end = val(h['tau'][i]) + p['service_time'] + p['travel_time'][i, 0]
                trip += 1
                distance = sum((p['distance'][a, b] for a, b in zip(path, path[1:])))
                routes.append(dict(vehicle_id=v['vehicle_id'], physical_vehicle_id=v['vehicle_id'], trip_number=trip, path=path, start_time_minutes=start, end_time_minutes=end, total_route_time_minutes=end - start, distance_km=distance, total_route_load=load, path_details=details))
                if nxt is None:
                    break
                i = nxt[0]
        usedids = {r['physical_vehicle_id'] for r in routes}
        vmap = {v['vehicle_id']: v for v in h['V']}
        distance = sum((r['distance_km'] for r in routes))
        duration = sum((r['total_route_time_minutes'] for r in routes))
        variable = sum((r['distance_km'] * vmap[r['physical_vehicle_id']]['cost_per_km'] for r in routes))
        fixed = sum((vmap[i]['fixed_cost'] for i in usedids)) if p['fixed_cost_mode'] == 'WHEN_USED' else sum((v['fixed_cost'] for v in p['vehicles']))
        comps = dict(DISTANCE=distance, TIME=duration, COST=variable + fixed, VEHICLES=float(len(usedids)))
        return dict(status='SUCCESS', method='SCIP_CUSTOMER_CHAIN_V52', routes=routes, summary=dict(total_distance_km=distance, total_time_minutes=duration, total_variable_cost=variable, total_fixed_cost=fixed, total_cost=variable + fixed, num_vehicles_used=len(usedids), num_routes=len(routes)), objective_components=comps, canonical_objective_score=self._canonical_score(comps), objective_context=copy.deepcopy(self.objective_context), proven_optimal=False)

    def _try_add_incumbent_mip_start(self, model, h, normalized_solution):
        report = {'submitted': False, 'validated': False, 'accepted': False, 'reason': None}
        self._mip_start_report = report
        sol = None
        try:
            p = self.problem
            sol = model.createSol()
            setv = lambda var, value: model.setSolVal(sol, var, float(value))
            for var in model.getVars():
                setv(var, 0)
            for i in h['C']:
                setv(h['dep'][i], p['operation_start'])
            grouped = defaultdict(list)
            for route in normalized_solution['routes']:
                grouped[int(route.get('physical_vehicle_id', route['vehicle_id']))].append(route)
            for vid, routes in grouped.items():
                vi = next((v['idx'] for v in h['V'] if int(v['vehicle_id']) == vid))
                routes.sort(key=lambda r: r['start_time_minutes'])
                setv(h['used'][vi], 1)
                order = 0
                for t, route in enumerate(routes):
                    path = route['path'][1:-1]
                    details = {int(d['node_id']): d for d in route['path_details']}
                    first = path[0]
                    last = path[-1]
                    start = float(route['start_time_minutes'])
                    setv(h['dep'][first], start)
                    if t == 0:
                        setv(h['first'][first], 1)
                        setv(h['starts'][first, vi], 1)
                    else:
                        setv(h['r'][routes[t - 1]['path'][-2], first], 1)
                    if t == len(routes) - 1:
                        setv(h['last'][last], 1)
                        setv(h['ends'][last, vi], 1)
                    load = 0.0
                    prev = 0
                    for i in path:
                        order += 1
                        load += p['client_rows'][i]['demand']
                        st = float(details[i]['service_start_time_minutes'])
                        arrival = start + p['travel_time'][0, i] if prev == 0 else float(details[prev]['service_start_time_minutes']) + p['service_time'] + p['travel_time'][prev, i]
                        waiting = max(0.0, st - arrival)
                        setv(h['y'][i, vi], 1)
                        setv(h['tau'][i], st)
                        setv(h['load'][i], load)
                        setv(h['wait'][i], waiting)
                        if h['rank']:
                            setv(h['rank'][i], order)
                        if prev:
                            setv(h['x'][prev, i], 1)
                        dc = p['distance'][prev, i] + (p['distance'][i, 0] if i == last else 0)
                        wc = p['service_time'] + waiting + p['travel_time'][prev, i] + (p['travel_time'][i, 0] if i == last else 0)
                        setv(h['qtime'][i, vi], wc)
                        if h['qdist']:
                            setv(h['qdist'][i, vi], dc)
                        prev = i
            report['validated'] = bool(model.checkSol(sol, printreason=True, completely=True))
            if not report['validated']:
                report['reason'] = 'MODEL_REJECTED_COMPLETE_START'
                return False
            report['submitted'] = True
            report['accepted'] = bool(model.addSol(sol, free=False))
            report['reason'] = 'ACCEPTED' if report['accepted'] else 'NOT_STORED'
            return report['accepted']
        except Exception as exc:
            report['reason'] = 'START_EXCEPTION: ' + str(exc)
            return False
        finally:
            if sol is not None:
                model.freeSol(sol)
````

## **O COORDENADOR E SUAS FERRAMENTAS DE APOIO**


### **ORCHESTRATORAGENT — COORDENAÇÃO DO SISTEMA**

Ele é o ponto central da organização. É como a pessoa que recebe o pedido do cliente, chama os especialistas, acompanha o trabalho e confere se as respostas podem ser reunidas.





Tabela com células delimitadas: Responsabilidade / Explicação

````python
class OrchestratorAgent:
    """
    Agente Orquestrador do sistema multiagente de otimização.

    Fluxo principal:
        Interface -> Validação -> Tradução -> Classificação -> MathAgent
        -> OR-Tools/GA -> Auditoria de factibilidade
        -> OptimalityCertificationAgent
        -> Auditoria final -> DiagnosticAgent -> ExplainerAgent.

    Princípios:
    - O Orchestrator não resolve o problema matemático diretamente.
    - OR-Tools e GA produzem candidatos; o melhor incumbent auditado é preservado globalmente.
    - O Certificador tenta provar ótimo global via SCIP/MILP integrado.
    - Níveis A/B/C podem seguir ao relatório somente após auditoria independente final.
    - Nível D não libera solução operacional definitiva.
    - objective_context é a única função objetivo usada para comparar
      candidatos do portfólio e incumbents entre tentativas.
    """

    VERSION = "ORCHESTRATOR_V10_14_PARALLEL_CONTINUITY_KCERT_V4"

    DEFAULT_CERTIFICATION_TIME_LIMIT_SECONDS = 120

    def __init__(self):
        self.version = self.VERSION
        self.memory = SharedMemory()

        self.tentativas = 0
        self.max_tentativas = 3
        self.force_math_change = False
        self._quality_intensification_mode = False
        self._quality_intensification_runs = 0
        self._approved_incumbents = []

        self.role = (
            "Cérebro Central e Maestro do Sistema: coordena o fluxo de "
            "otimização, certificação, auditoria, diagnóstico e explicação."
        )
        self.goal = (
            "Entregar uma solução operacional factível e, quando exigido, "
            "matematicamente certificada como ótima global."
        )
        self.backstory = (
            "Gestor de otimização logística responsável por coordenar "
            "especialistas independentes e preservar rastreabilidade."
        )

        self.tools = [
            "SharedMemory",
            "Portfólio OR-Tools/GA",
            "AVE — NeighborhoodOptimizationAgent",
            "BoundStrengtheningAgent",
            "KVehicleFeasibilityCertifier",
            "OptimalityCertificationAgent",
            "AuditorAgent",
            "DiagnosticAgent",
            "ExplainerAgent",
        ]

        self.constraints = [
            "Não executar cálculos de otimização diretamente.",
            "Toda solução deve passar por auditoria de factibilidade.",
            "Toda comparação entre solvers deve usar objective_context.",
            "Não declarar ótimo global sem proven_optimal=True.",
            "Erros de visualização não invalidam uma solução matemática.",
        ]

        self.agents = {
            "validator": ValidatorAgent(self.memory),
            "translator": TranslatorAgent(self.memory),
            "classifier": ClassifierAgent(self.memory),
            "math": MathAgent(self.memory),
            "ortools_solver": ORToolsSolverAgent(self.memory),
            "ga_solver": GeneticAlgorithmSolverAgent(self.memory),
            "certifier": OptimalityCertificationAgent(self.memory),
            "auditor": AuditorAgent(self.memory),
            "diagnostic": DiagnosticAgent(self.memory),
            "explainer": ExplainerAgent(self.memory),
            "continuity": ContinuityOptimizationAgent(self.memory),
            "ave": NeighborhoodOptimizationAgent(self.memory),
            "bound_strengthener": BoundStrengtheningAgent(self.memory),
            "k_vehicle_certifier": KVehicleFeasibilityCertifier(self.memory),
            "vehicle_program_columns": VehicleProgramColumnAgent(self.memory),
            "initial_refiner": InitialSolutionRefinerAgent(self.memory),
        }

        # MULTI13 — fail-fast único e não redundante.
        # Impede que células antigas do Colab sobrescrevam silenciosamente
        # qualquer componente crítico da arquitetura K-Proof.
        active_versions = {
            "AVE": str(getattr(self.agents["ave"], "VERSION", "UNKNOWN")),
            "Bound": str(getattr(self.agents["bound_strengthener"], "VERSION", "UNKNOWN")),
            "KCert": str(getattr(self.agents["k_vehicle_certifier"], "VERSION", "UNKNOWN")),
            "Continuity": str(getattr(self.agents["continuity"], "VERSION", "UNKNOWN")),
            "Certifier": str(getattr(self.agents["certifier"], "VERSION", "UNKNOWN")),
        }
        expected_versions = {
            "KCert": "K_VEHICLE_FEASIBILITY_CERTIFIER_V5_FAIR_DEADLINES_SCOPE",
            "Continuity": "CONTINUITY_OPTIMIZATION_AGENT_V10_PROCESS_PARALLEL_ABC",
            "Bound": "BOUND_STRENGTHENING_AGENT_V6_SCF_LP",
            "Certifier": "SCIP_CASCADE_CERTIFIER_16_K_FEASIBILITY_SCF",
        }

        # AVE  é preservado; aceita apenas a família  para não acoplar
        # desnecessariamente o Orchestrator a uma string cosmética específica.
        if not active_versions["AVE"].startswith("AVE_V2"):
            raise RuntimeError(
                "Integração inválida do AVE. "
                f"Versão ativa={active_versions['AVE']!r}; esperada família AVE_V2."
            )

        for component, expected in expected_versions.items():
            actual = active_versions[component]
            if actual != expected:
                raise RuntimeError(
                    f"Integração inválida de {component}. "
                    f"Versão esperada={expected!r} | versão ativa={actual!r}. "
                    "Reinicie a sessão do Colab e execute as células em ordem."
                )

        print(
            "🔌 Integração MULTI13 confirmada | "
            f"AVE={active_versions['AVE']} | "
            f"Bound={active_versions['Bound']} | "
            f"KCert={active_versions['KCert']} | "
            f"Continuity={active_versions['Continuity']} | "
            f"Certifier={active_versions['Certifier']}"
        )

    def _executar_fluxo_pre_solver(self, ans):


        # CLIENTES


        if self.memory.data["df_clientes"] is None:

            fname = ans.get("df_clientes_file")

            if not fname:
                raise ValueError(
                    "Arquivo de clientes não informado."
                )

            ext = (
                "excel"
                if fname.lower().endswith(("xlsx", "xls"))
                else "csv"
            )

            self.memory.data["df_clientes"] = (
                load_clientes_data(
                    ext,
                    filename_base=os.path.splitext(fname)[0]
                )
            )


        # CAPACIDADE DOS VEÍCULOS


        if self.memory.data["df_veiculos_capacidade"] is None:

            fname = ans.get(
                "df_veiculos_capacidade_file"
            )

            if not fname:
                raise ValueError(
                    "Arquivo de capacidade dos veículos não informado."
                )

            ext = (
                "excel"
                if fname.lower().endswith(("xlsx", "xls"))
                else "csv"
            )

            self.memory.data["df_veiculos_capacidade"] = (
                load_veiculos_capacidade_data(
                    ext,
                    filename_base=os.path.splitext(fname)[0]
                )
            )


        # CUSTOS


        if ans.get("df_veiculos_custo_km_file"):

            if self.memory.data["df_veiculos_custo_km"] is None:

                fname = ans.get(
                    "df_veiculos_custo_km_file"
                )

                ext = (
                    "excel"
                    if fname.lower().endswith(("xlsx", "xls"))
                    else "csv"
                )

                self.memory.data["df_veiculos_custo_km"] = (
                    load_veiculos_custo_km_data(
                        ext,
                        filename_base=os.path.splitext(fname)[0]
                    )
                )


        # JANELAS DE TEMPO


        if ans.get("df_janela_tempo_cliente_file"):

            if self.memory.data["df_janela_tempo_cliente"] is None:

                fname = ans.get(
                    "df_janela_tempo_cliente_file"
                )

                ext = (
                    "excel"
                    if fname.lower().endswith(("xlsx", "xls"))
                    else "csv"
                )

                self.memory.data["df_janela_tempo_cliente"] = (
                    load_janela_tempo_cliente_data(
                        ext,
                        filename_base=os.path.splitext(fname)[0]
                    )
                )


        # VALIDAÇÃO


        val_msg = self.agents["validator"].validate(
            ans,
            self.memory.data["df_clientes"],
            self.memory.data["df_veiculos_capacidade"],
            df_veiculos_custo_km=self.memory.data.get("df_veiculos_custo_km"),
            df_janela_tempo_cliente=self.memory.data.get("df_janela_tempo_cliente"),
        )
        print(f"[STATUS]: {val_msg}")
        validation = self.memory.data.get("validation_result") or {}
        if not validation.get("valid", False):
            errors = validation.get("errors", [])
            raise ValueError(
                "ORCHESTRATOR V10: dados de entrada reprovados pelo ValidatorAgent. "
                + " | ".join(str(e) for e in errors)
            )


        # TRADUÇÃO


        self.memory.data["translated_json"] = (
            self.agents["translator"].translate_to_json(
                ans,
                self.memory.data["df_clientes"],
                self.memory.data["df_veiculos_capacidade"],
                self.memory.data.get(
                    "df_veiculos_custo_km"
                ),
                self.memory.data.get(
                    "df_janela_tempo_cliente"
                )
            )
        )


        # CLASSIFICAÇÃO


        self.memory.data["problem_classification"] = (
            self.agents["classifier"].classify_problem(
                self.memory.data["translated_json"]
            )
        )


        # ESTRATÉGIA MATEMÁTICA INICIAL


        self.memory.data["optimization_approach"] = (
            self.agents["math"].decide_optimization_approach(
                self.memory.data["problem_classification"],
                self.memory.data["translated_json"]
            )
        )


    # FUNÇÃO OBJETIVO CANÔNICA / SELEÇÃO DO INCUMBENT


    def _extract_candidate_metrics(self, result):
        result = result or {}
        summary = result.get("summary", {}) or {}
        metrics = result.get("metrics", {}) or {}

        return {
            "DISTANCE": float(
                summary.get(
                    "total_distance_km",
                    metrics.get("total_distance", 0.0)
                ) or 0.0
            ),
            "TIME": float(
                summary.get(
                    "total_time_minutes",
                    metrics.get("total_time", 0.0)
                ) or 0.0
            ),
            "COST": float(
                summary.get(
                    "total_cost",
                    result.get(
                        "total_cost",
                        metrics.get("real_cost", 0.0)
                    )
                ) or 0.0
            ),
            "VEHICLES": float(
                summary.get(
                    "num_vehicles_used",
                    metrics.get("used_vehicles", 0.0)
                ) or 0.0
            ),
        }

    def _get_objective_context(self):
        context = self.memory.data.get("objective_context")

        if not isinstance(context, dict):
            raise ValueError(
                "ORCHESTRATOR: objective_context ausente. "
                "O MathAgent deve ser executado antes dos solvers."
            )

        if not context.get("frozen", False):
            raise ValueError(
                "ORCHESTRATOR: objective_context deve estar congelado."
            )

        return context

    def _canonical_score(self, metrics):
        context = self._get_objective_context()
        coefficients = context.get("coefficients", {}) or {}

        return float(
            sum(
                float(coefficients.get(obj, 0.0))
                * float(metrics.get(obj, 0.0))
                for obj in context.get("active_objectives", [])
            )
        )

    def _normalized_metrics(self, metrics):
        context = self._get_objective_context()
        scales = context.get("scales", {}) or {}

        return {
            obj: float(metrics.get(obj, 0.0))
            / max(1e-12, float(scales.get(obj, 1.0)))
            for obj in ("DISTANCE", "TIME", "COST", "VEHICLES")
        }

    @staticmethod
    def _dominates(a, b, objectives, tol=1e-9):
        return (
            all(a[o] <= b[o] + tol for o in objectives)
            and any(a[o] < b[o] - tol for o in objectives)
        )

    def _select_incumbent(self, candidates):
        """
        Seleciona o melhor candidato pela função canônica congelada.

        A normalização dinâmica min/max entre candidatos foi removida:
        ela alterava a função de comparação conforme o portfólio mudava.
        """
        successful = [
            c for c in candidates
            if c.get("result", {}).get("status") == "SUCCESS"
        ]

        if not successful:
            return None, candidates

        context = self._get_objective_context()
        objectives = list(context.get("active_objectives", []))

        for candidate in successful:
            metrics = self._extract_candidate_metrics(
                candidate["result"]
            )
            candidate["canonical_metrics"] = metrics
            candidate["portfolio_score"] = self._canonical_score(metrics)
            candidate["portfolio_normalized"] = (
                self._normalized_metrics(metrics)
            )

        nondominated = []

        for candidate in successful:
            dominated = any(
                self._dominates(
                    other["canonical_metrics"],
                    candidate["canonical_metrics"],
                    objectives,
                )
                for other in successful
                if other is not candidate
            )

            candidate["dominated"] = dominated

            if not dominated:
                nondominated.append(candidate)

        pool = nondominated or successful

        incumbent = min(
            pool,
            key=lambda c: (
                c["portfolio_score"],
                c["canonical_metrics"].get("COST", 0.0),
                c["canonical_metrics"].get("DISTANCE", 0.0),
                c["canonical_metrics"].get("TIME", 0.0),
                c["canonical_metrics"].get("VEHICLES", 0.0),
            ),
        )

        return incumbent, candidates

    def _register_single_solver_incumbent(self, result, solver_name):
        """
        Registra candidato de solver sem sobrescrever um incumbent já melhor.

        A decisão usa exclusivamente objective_context congelado. Esta rotina
        NÃO significa aprovação operacional; a aprovação continua sendo papel
        exclusivo do AuditorAgent.
        """
        if not isinstance(result, dict) or result.get("status") != "SUCCESS":
            return

        metrics = self._extract_candidate_metrics(result)
        result["portfolio_winner"] = solver_name
        result["portfolio_score"] = self._canonical_score(metrics)
        result["portfolio_normalized"] = self._normalized_metrics(metrics)
        result["canonical_metrics"] = metrics

        current = self.memory.data.get("best_incumbent")
        if not isinstance(current, dict) or current.get("status") != "SUCCESS":
            self.memory.data["best_incumbent"] = copy.deepcopy(result)
            return

        current_metrics = self._extract_candidate_metrics(current)
        current_score = self._canonical_score(current_metrics)
        candidate_score = float(result["portfolio_score"])

        if candidate_score < current_score - 1e-12:
            self.memory.data["best_incumbent"] = copy.deepcopy(result)

    def _audited_result_key(self, result):
        """Chave determinística de comparação entre incumbents auditados."""
        metrics = self._extract_candidate_metrics(result)
        score = self._canonical_score(metrics)
        return (
            float(score),
            float(metrics.get("COST", 0.0)),
            float(metrics.get("DISTANCE", 0.0)),
            float(metrics.get("TIME", 0.0)),
            float(metrics.get("VEHICLES", 0.0)),
        )

    def _register_audited_incumbent(self, result, recommendation, attempt, approach):
        """
        Guarda somente soluções que passaram pela auditoria independente.

        A solução normalizada pelo Auditor é preferida porque contém tempos,
        distâncias, custos e objetivo reconstruídos independentemente.
        """
        if not (
            isinstance(result, dict)
            and result.get("status") == "SUCCESS"
            and isinstance(recommendation, dict)
            and recommendation.get("solucao_aprovada", False)
        ):
            return None

        report = (
            self.memory.data.get("independent_feasibility_report_auditor")
            or self.memory.data.get("last_independent_feasibility_report")
            or {}
        )
        normalized = report.get("normalized_solution") if isinstance(report, dict) else None

        if isinstance(normalized, dict) and normalized.get("status") == "SUCCESS":
            candidate = copy.deepcopy(normalized)
        else:
            candidate = copy.deepcopy(result)

        metrics = self._extract_candidate_metrics(candidate)
        score = self._canonical_score(metrics)

        record = {
            "attempt": int(attempt),
            "approach": str(approach),
            "score": float(score),
            "metrics": copy.deepcopy(metrics),
            "result": candidate,
        }

        self._approved_incumbents.append(record)
        self._approved_incumbents.sort(
            key=lambda item: self._audited_result_key(item["result"])
        )

        best = self._approved_incumbents[0]
        self.memory.data["best_incumbent"] = copy.deepcopy(best["result"])
        self.memory.data["best_audited_incumbent"] = copy.deepcopy(best["result"])
        self.memory.data["audited_incumbent_history"] = [
            {
                "attempt": item["attempt"],
                "approach": item["approach"],
                "score": item["score"],
                "metrics": copy.deepcopy(item["metrics"]),
            }
            for item in self._approved_incumbents
        ]

        return best

    def _best_audited_incumbent(self):
        if not self._approved_incumbents:
            return None
        return min(
            self._approved_incumbents,
            key=lambda item: self._audited_result_key(item["result"])
        )

    def _should_continue_quality_search(self):
        """
        Decide se ainda vale usar o orçamento de otimização para procurar
        um incumbent melhor depois de já existir solução factível aprovada.

        Aprovação do Auditor não encerra mais a busca automaticamente.
        """
        if self.tentativas >= self.max_tentativas:
            return False

        profile = getattr(self, "_active_budget_profile", None) or {}
        max_intensifications = int(profile.get("max_quality_intensifications", 0) or 0)
        if self._quality_intensification_runs >= max_intensifications:
            return False

        remaining = self._remaining_optimization_seconds()
        ga_seconds = float(profile.get("ga_seconds", 120) or 120)

        # Só inicia nova intensificação se houver orçamento suficiente para uma
        # execução útil e pequena reserva de encerramento.
        minimum_useful = max(45.0, min(75.0, 0.45 * ga_seconds))
        return remaining > minimum_useful + 10.0


    # ORÇAMENTO COMPUTACIONAL DOS SOLVERS


    def _problem_size(self):
        try:
            raw = json.loads(self.memory.data["translated_json"])
            return int(
                (raw.get("problem_description", {}) or {})
                .get("number_of_clients", 0) or 0
            )
        except Exception:
            df = self.memory.data.get("df_clientes")
            return max(0, len(df) - 1) if df is not None else 0

    def _optimization_budget_profile(self):
        """
        Orçamento adaptado ao tamanho do problema.

        O objetivo é evitar reservar 900 s indiscriminadamente. Instâncias pequenas
        recebem menos seeds e limites menores; instâncias médias/grandes preservam
        orçamento crescente. O melhor incumbent auditado continua sendo preservado.
        """
        n = self._problem_size()

        if n <= 20:
            return {
                "profile": "SMALL_QUALITY",
                "global_seconds": 240,
                "ortools_seconds": 30,
                "ga_population": 100,
                "ga_generations": 140,
                "ga_seconds": 75,
                "ga_seeds": [101],
                "max_quality_intensifications": 1,
            }

        if n <= 50:
            return {
                "profile": "BALANCED",
                "global_seconds": 600,
                "ortools_seconds": 120,
                "ga_population": 140,
                "ga_generations": 180,
                "ga_seconds": 130,
                "ga_seeds": [101, 307],
                "max_quality_intensifications": 1,
            }

        return {
            "profile": "LARGE_SCALE",
            "global_seconds": 720,
            "ortools_seconds": 150,
            "ga_population": 60,
            "ga_generations": 120,
            "ga_seconds": 180,
            "ga_seeds": [101],
            "max_quality_intensifications": 1,
            "feasibility_recovery_seconds": 170,
        }

    def _remaining_optimization_seconds(self):
        profile = (
            getattr(self, "_active_budget_profile", None)
            or self._optimization_budget_profile()
        )

        started = getattr(
            self,
            "_global_optimization_started",
            None
        )

        if started is None:
            return float(profile["global_seconds"])

        return max(
            0.0,
            float(profile["global_seconds"])
            - (_time_module.monotonic() - started)
        )


    # PORTFÓLIO OR-TOOLS + GA


    def _run_solver_portfolio(self):
        candidates = []

        classification = self.memory.data["problem_classification"]
        translated = self.memory.data["translated_json"]

        profile = (
            getattr(self, "_active_budget_profile", None)
            or self._optimization_budget_profile()
        )

        print(
            f"🧭 Portfólio {profile['profile']} | "
            f"clientes={self._problem_size()} | "
            f"orçamento global={profile['global_seconds']}s"
        )

        # OR-Tools
        remaining = self._remaining_optimization_seconds()

        if remaining <= 5:
            return {
                "status": "TIME_LIMIT",
                "error": "Orçamento global esgotado antes do Solver."
            }

        ort_budget = min(
            profile["ortools_seconds"],
            int(remaining)
        )

        print(
            f"🔵 Etapa 1: OR-Tools | orçamento até {ort_budget}s"
        )

        self.memory.data[
            "current_solver_time_limit_seconds"
        ] = ort_budget

        ort = self.agents["ortools_solver"].solve_problem(
            classification,
            translated
        )

        candidates.append({
            "solver": "OR-Tools",
            "result": ort
        })

        incumbent, candidates = self._select_incumbent(
            candidates
        )

        if incumbent is not None:
            self.memory.data["best_incumbent"] = copy.deepcopy(
                incumbent["result"]
            )

            m = incumbent["canonical_metrics"]

            print(
                f"✅ Incumbent inicial | "
                f"custo={m['COST']:.2f} | "
                f"distância={m['DISTANCE']:.2f} km | "
                f"tempo={m['TIME']:.2f} min | "
                f"veículos={m['VEHICLES']:.0f}"
            )

        # GA
        for seed in profile["ga_seeds"]:
            remaining = self._remaining_optimization_seconds()

            if remaining <= 20:
                print(
                    "⏹️ Portfólio: orçamento insuficiente "
                    "para nova execução do GA."
                )
                break

            ga_budget = min(
                profile["ga_seconds"],
                max(20, int(remaining - 10))
            )

            print(
                f"🧬 Etapa GA seed={seed} | "
                f"população={profile['ga_population']} | "
                f"gerações={profile['ga_generations']} | "
                f"limite={ga_budget}s"
            )

            ga = self.agents["ga_solver"].solve_problem(
                classification,
                translated,
                population_size=profile["ga_population"],
                num_generations=profile["ga_generations"],
                time_limit_seconds=ga_budget,
                seed=seed,
                feasibility_first=(profile["profile"] == "LARGE_SCALE"),
                continuity_mode=False,
            )

            candidates.append({
                "solver": f"GA seed={seed}",
                "result": ga
            })

            incumbent, candidates = self._select_incumbent(
                candidates
            )

            if incumbent is not None:
                self.memory.data["best_incumbent"] = copy.deepcopy(
                    incumbent["result"]
                )

                m = incumbent["canonical_metrics"]

                print(
                    f"🏆 Melhor incumbent atual | "
                    f"{incumbent['solver']} | "
                    f"custo={m['COST']:.2f} | "
                    f"distância={m['DISTANCE']:.2f} km | "
                    f"tempo={m['TIME']:.2f} min"
                )

            if profile["profile"] == "LARGE_SCALE":
                break

        incumbent, candidates = self._select_incumbent(
            candidates
        )

        self.memory.data[
            "solver_portfolio_results"
        ] = candidates

        if incumbent is None:
            return (
                candidates[0]["result"]
                if candidates
                else {
                    "status": "ERROR",
                    "error": "Nenhum solver produziu resultado."
                }
            )

        result = copy.deepcopy(incumbent["result"])

        result["portfolio_complete"] = True
        result["portfolio_winner"] = incumbent["solver"]
        result["portfolio_score"] = incumbent["portfolio_score"]
        result["portfolio_normalized"] = (
            incumbent["portfolio_normalized"]
        )
        result["optimization_profile"] = profile["profile"]
        result["optimization_budget_seconds"] = (
            profile["global_seconds"]
        )
        result["optimization_remaining_seconds"] = round(
            self._remaining_optimization_seconds(),
            2
        )

        result["portfolio_candidates"] = [
            {
                "solver": c["solver"],
                "status": c["result"].get("status"),
                "canonical_metrics": c.get("canonical_metrics"),
                "dominated": c.get("dominated"),
                "portfolio_score": c.get("portfolio_score"),
                "runtime_seconds": c["result"].get(
                    "runtime_seconds"
                ),
            }
            for c in candidates
        ]

        self.memory.data["best_incumbent"] = copy.deepcopy(
            result
        )

        return result


    # AUDITORIA


    def _build_audit_context(
        self,
        ans,
        result,
        approach,
        tentativa
    ):
        return AuditContext(
            user_answers=ans,
            translated_json=self.memory.data["translated_json"],
            problem_classification=(
                self.memory.data["problem_classification"]
            ),
            optimization_approach=approach,
            solver_result=result,
            df_clientes=self.memory.data["df_clientes"],
            df_veiculos_capacidade=(
                self.memory.data["df_veiculos_capacidade"]
            ),
            df_veiculos_custo_km=self.memory.data.get(
                "df_veiculos_custo_km"
            ),
            df_janela_tempo_cliente=self.memory.data.get(
                "df_janela_tempo_cliente"
            ),
            teve_solucao=(
                isinstance(result, dict)
                and result.get("status") == "SUCCESS"
            ),
            metodo_utilizado=(
                result.get("method")
                if isinstance(result, dict)
                else None
            ),
            tentativas=tentativa,
        )

    def _audit_result(
        self,
        ans,
        result,
        approach,
        tentativa
    ):
        ctx = self._build_audit_context(
            ans,
            result,
            approach,
            tentativa
        )

        diag, rec = self.agents[
            "auditor"
        ].gerar_diagnostico_completo(ctx)

        self.memory.data["audit_result"] = diag
        self.memory.data["audit_recommendation"] = rec

        return diag, rec

    def _print_audit_diagnostic(self, diag, recommendation, stage="PRE_CERTIFICATION"):
        """
        Resumo legível da auditoria para o usuário.

        IMPORTANTE:
        - não altera ``diag``, ``recommendation`` nem os relatórios armazenados;
        - os dicionários técnicos completos continuam disponíveis na SharedMemory;
        - evita despejar ``checks``, ``independent_verification``,
          ``certification_assessment`` e ``multiobjective_audit`` no console.
        """
        print("\n" + "=" * 78)
        print(f"🔍 RESUMO DA AUDITORIA | etapa={stage}")
        print("=" * 78)


        # 1. Parecer principal do AuditorAgent

        if isinstance(diag, dict):
            status = diag.get("status_final", diag.get("status"))
            if status is not None:
                print(f"📋 Parecer do Auditor: {status}")

            dominated = diag.get("dominated_by_incumbent")
            if dominated is not None:
                print(f"   Dominada por incumbent melhor: {'SIM' if dominated else 'NÃO'}")
        else:
            print(f"📋 Parecer do Auditor: {diag}")


        # 2. Verificação independente — resumo, sem dicionários brutos

        report = (
            self.memory.data.get("independent_feasibility_report_auditor")
            or self.memory.data.get("last_independent_feasibility_report")
        )

        print("\n🧪 Verificação independente:")
        if not isinstance(report, dict):
            print("   ⚠️ Nenhum relatório independente foi encontrado na SharedMemory.")
        else:
            feasible = bool(report.get("feasible", False))
            print(f"   Factibilidade: {'APROVADA' if feasible else 'REPROVADA'}")

            checks = report.get("checks", {}) or {}
            friendly_checks = (
                ("all_customers_served_once", "Clientes atendidos exatamente uma vez"),
                ("capacity_per_trip", "Capacidade por viagem"),
                ("time_windows", "Janelas de tempo"),
                ("route_operation_horizon", "Horizonte operacional das viagens"),
                ("multi_trip_sequence_nonoverlap", "Sequenciamento Multi-Trip"),
                ("daily_work_time_per_physical_vehicle", "Jornada diária por veículo"),
                ("minimum_usage_per_used_vehicle", "Tempo mínimo de utilização"),
                ("must_use_all_vehicles", "Regra de utilização obrigatória da frota"),
                ("distance_recomputed", "Distância recalculada"),
                ("variable_cost_recomputed", "Custo variável recalculado"),
                ("fixed_cost_recomputed", "Custo fixo recalculado"),
                ("canonical_objective_recomputed", "Objetivo canônico recalculado"),
            )
            for key, label in friendly_checks:
                if key in checks:
                    print(f"   {'✅' if checks.get(key) else '❌'} {label}")

            violations = report.get("violations") or report.get("violacoes") or report.get("errors") or []
            if violations:
                print("   ⚠️ Violações:")
                if isinstance(violations, (list, tuple)):
                    for item in violations:
                        print(f"      - {item}")
                else:
                    print(f"      - {violations}")
            else:
                print("   Violações: nenhuma")

            recomputed = report.get("recomputed", {}) or {}
            if recomputed:
                print("\n📊 Totais reconstruídos independentemente:")
                if recomputed.get("num_vehicles_used") is not None:
                    print(f"   Veículos utilizados: {recomputed.get('num_vehicles_used')}")
                if recomputed.get("num_routes") is not None:
                    print(f"   Viagens: {recomputed.get('num_routes')}")
                if recomputed.get("total_distance_km") is not None:
                    print(f"   Distância: {float(recomputed['total_distance_km']):.3f} km")
                if recomputed.get("total_time_minutes") is not None:
                    print(f"   Tempo: {float(recomputed['total_time_minutes']):.3f} min")
                if recomputed.get("total_cost") is not None:
                    print(f"   Custo total: R$ {float(recomputed['total_cost']):.2f}")
                if recomputed.get("canonical_objective_score") is not None:
                    print(f"   Score canônico: {float(recomputed['canonical_objective_score']):.12f}")

            normalized = report.get("normalized_solution")
            if isinstance(normalized, dict):
                routes = normalized.get("routes") or []
                if routes:
                    print(f"   Rotas/viagens normalizadas: {len(routes)}")


        # 3. Recomendação — somente campos de decisão para o usuário

        print("\n🧭 Recomendação do Auditor:")
        if isinstance(recommendation, dict):
            approved = recommendation.get("solucao_aprovada")
            feasible = recommendation.get("factibilidade_aprovada")
            quality = recommendation.get("qualidade_aprovada")
            action = recommendation.get("proxima_acao_orquestrador")
            justification = recommendation.get("justificativa_tecnica_da_recomendacao")

            if approved is not None:
                print(f"   Solução aprovada: {'SIM' if approved else 'NÃO'}")
            if feasible is not None:
                print(f"   Factibilidade aprovada: {'SIM' if feasible else 'NÃO'}")
            if quality is not None:
                print(f"   Qualidade aprovada: {'SIM' if quality else 'NÃO'}")
            if action:
                print(f"   Próxima ação: {action}")
            if justification:
                print(f"   Justificativa: {justification}")

            # Resumo da certificação sem imprimir o dicionário completo.
            cert = recommendation.get("certification_assessment", {}) or {}
            if isinstance(cert, dict) and any(v is not None for v in cert.values()):
                level = cert.get("certification_level")
                gap = cert.get("gap_pct")
                proven = bool(cert.get("proven_optimal", False))
                if level is not None:
                    print(f"   Certificação: Nível {level}")
                if gap is not None:
                    try:
                        print(f"   Gap: {float(gap):.6f}%")
                    except Exception:
                        print(f"   Gap: {gap}")
                print(f"   Ótimo global provado: {'SIM' if proven else 'NÃO'}")
        else:
            print(f"   {recommendation}")

        print("=" * 78 + "\n")

    def _deve_reexecutar(self, result, recommendation):
        if self._remaining_optimization_seconds() <= 30:
            return False

        status = result.get("status")

        if status in (
            "ERROR",
            "INFEASIBLE",
            "NO_FEASIBLE_SOLUTION",
            "TIME_LIMIT",
        ):
            return True

        profile = (
            getattr(self, "_active_budget_profile", None)
            or {}
        ).get("profile")

        if recommendation.get("solucao_aprovada", False):
            return False

        if profile == "LARGE_SCALE":
            # Sem incumbent auditado, LARGE_SCALE deve usar as tentativas restantes
            # em recuperação de factibilidade, desde que ainda exista orçamento.
            return not bool(self._approved_incumbents)

        return (
            recommendation.get("proxima_acao_orquestrador")
            in (
                "TROCAR_ESTRATEGIA_MATEMATICA",
                "REEXECUTAR_MESMO_SOLVER",
            )
        )


    # CERTIFICAÇÃO DE OTIMALIDADE


    def _configure_certifier(self, ans):
        """Configura certificação SCIP adaptativa com teto rígido de tempo."""
        certifier = self.agents["certifier"]

        certifier.time_limit_seconds = max(
            1,
            int(
                ans.get(
                    "certification_time_limit_seconds",
                    self.DEFAULT_CERTIFICATION_TIME_LIMIT_SECONDS,
                )
                or self.DEFAULT_CERTIFICATION_TIME_LIMIT_SECONDS
            ),
        )
        certifier.scip_threads = max(
            0, int(ans.get("certification_threads", 0) or 0)
        )
        certifier.scip_memory_mb = max(
            0, int(ans.get("certification_memory_mb", 0) or 0)
        )
        certifier.use_incumbent_cutoff = True

        # Tolerâncias matemáticas preservadas.
        certifier.relative_gap_tolerance = float(
            ans.get("relative_gap_tolerance", 0.0001) or 0.0001
        )
        certifier.business_gap_tolerance = float(
            ans.get("business_gap_tolerance", 0.01) or 0.01
        )

        # Parada adaptativa por estagnação do lower bound. O teto padrão cai de
        # 300 s para 120 s, mas o SCIP pode parar antes quando o dual não progride.
        raw_adaptive = ans.get("adaptive_certification", True)
        if isinstance(raw_adaptive, str):
            raw_adaptive = raw_adaptive.strip().lower() in {
                "1", "true", "sim", "s", "yes", "y"
            }
        certifier.adaptive_stop_enabled = bool(raw_adaptive)
        certifier.adaptive_min_runtime_seconds = max(
            1.0, float(ans.get("adaptive_min_runtime_seconds", 30) or 30)
        )
        certifier.adaptive_patience_seconds = max(
            1.0, float(ans.get("adaptive_patience_seconds", 20) or 20)
        )
        certifier.adaptive_check_interval_seconds = max(
            0.5, float(ans.get("adaptive_check_interval_seconds", 5) or 5)
        )
        certifier.adaptive_min_relative_lb_improvement = max(
            0.0,
            float(
                ans.get(
                    "adaptive_min_relative_lb_improvement",
                    0.0001,
                )
                or 0.0001
            ),
        )

        # Camada intensiva permanece disponível somente quando explicitamente pedida.
        raw_intensive = ans.get("intensive_certification", False)
        if isinstance(raw_intensive, str):
            raw_intensive = raw_intensive.strip().lower() in {
                "1", "true", "sim", "s", "yes", "y"
            }
        certifier.intensive_certification = bool(raw_intensive)
        certifier.intensive_time_limit_seconds = max(
            1,
            int(
                ans.get(
                    "intensive_certification_time_limit_seconds",
                    1800,
                )
                or 1800
            ),
        )
        return certifier

    def _run_optimality_certification(self, ans):
        incumbent = self.memory.data.get("best_incumbent")

        if not isinstance(incumbent, dict):
            incumbent = self.memory.data.get("solver_result")

        if (
            not isinstance(incumbent, dict)
            or incumbent.get("status") != "SUCCESS"
        ):
            raise ValueError(
                "ORCHESTRATOR: não existe incumbent factível "
                "para iniciar a certificação."
            )

        # Preserva o resultado dos solvers para rastreabilidade.
        self.memory.data[
            "solver_result_pre_certification"
        ] = copy.deepcopy(incumbent)

        self.memory.data["best_incumbent"] = copy.deepcopy(
            incumbent
        )

        print(
            "\n"
            + "=" * 70
        )
        print(
            "🏆 CERTIFICAÇÃO DE OTIMALIDADE GLOBAL"
        )
        print(
            "=" * 70
        )
        print(
            "Método: Certificação adaptativa em cascata (Auditor → SCIP → Auditor)"
        )

        certifier = self._configure_certifier(ans)
        result = certifier.certify_optimality()

        self.memory.data["certification_result"] = result

        return result


    # DIAGNÓSTICO / EXPLICAÇÃO


    def _run_diagnostic_and_report(
        self,
        final_solution,
        ans,
        audit_recommendation
    ):
        diagnostico_completo = None

        try:
            print("\n" + "=" * 70)
            print("🔎 DIAGNÓSTICO PÓS-OTIMIZAÇÃO")
            print("=" * 70)

            diagnostico_completo = self.agents[
                "diagnostic"
            ].executar_diagnostico(
                gerar_painel=False,
                gerar_mapa=False
            )

            self.memory.data[
                "diagnostic_result"
            ] = diagnostico_completo

            print("✅ Diagnóstico concluído.")

        except Exception as diagnostic_error:
            print(
                "⚠️ Erro durante o diagnóstico: "
                f"{type(diagnostic_error).__name__}: "
                f"{diagnostic_error}"
            )
            print(
                "ℹ️ A solução matemática foi preservada."
            )

            diagnostico_completo = self.memory.data.get(
                "diagnostic_analysis"
            )

        lista_recom = []

        if isinstance(diagnostico_completo, dict):
            lista_recom = (
                diagnostico_completo.get("recomendacoes", [])
                or diagnostico_completo.get(
                    "recommendations",
                    []
                )
            )

        if not lista_recom:
            try:
                lista_recom = self.agents[
                    "diagnostic"
                ].analisar_performance_e_gerar_recomendacoes()
            except Exception as exc:
                print(
                    "⚠️ Não foi possível recuperar "
                    f"recomendações: {exc}"
                )
                lista_recom = []

        self.memory.data[
            "diagnostic_recommendations"
        ] = lista_recom

        relatorio = self.agents[
            "explainer"
        ].explain_solution(
            final_solution,
            self.memory.data["problem_classification"],
            ans,
            audit_recommendation=audit_recommendation,
            tentativa_atual=self.tentativas,
            recomendacoes_diag=lista_recom,
        )

        # Relatório preparado em memória; apresentação na célula de resultados.

        self.memory.data["final_report"] = relatorio

        return diagnostico_completo, lista_recom, relatorio


    # CONTINUIDADE OPCIONAL PÓS-RELATÓRIO 1

    @staticmethod
    def _continuity_yes(value):
        return str(value or "").strip().lower() in {"1","s","sim","y","yes"}

    def _ask_continuity_budget(self):
        return UserInterface.ask_continuity_budget()

    def _continuity_snapshot(self):
        keys = ["solver_result","best_incumbent","final_solution","certification_result","optimality_certificate","certified_solution","diagnostic_result","diagnostic_analysis","final_report","audit_recommendation","independent_feasibility_report_auditor","last_independent_feasibility_report","independent_feasibility_report_pre_certification","independent_feasibility_report_post_certification"]
        return {k: copy.deepcopy(self.memory.data.get(k)) for k in keys}

    def _restore_primary_snapshot(self, snap):
        for k,v in snap.items():
            self.memory.data[k] = copy.deepcopy(v)

    def _continuity_comparison(self, baseline_solution, baseline_cert, after_solution, after_cert):
        bm=self._extract_candidate_metrics(baseline_solution); am=self._extract_candidate_metrics(after_solution)
        bobj=self._canonical_score(bm); aobj=self._canonical_score(am)
        bl=str((baseline_cert or {}).get("certification_level","D") or "D").upper()
        al=str((after_cert or {}).get("certification_level","D") or "D").upper()
        rank={"A":0,"B+":1,"B":2,"C":3,"D":4}
        bg=(baseline_cert or {}).get("gap_pct"); ag=(after_cert or {}).get("gap_pct")
        cert_improved = rank.get(al,9) < rank.get(bl,9)
        if not cert_improved and al==bl and bg is not None and ag is not None:
            cert_improved = float(ag) < float(bg)-1e-9
        return {
            "solution_improved": aobj < bobj-1e-10,
            "certification_improved": cert_improved,
            "before":{"objective":bobj,"distance":bm["DISTANCE"],"time":bm["TIME"],"cost":bm["COST"],"vehicles":bm["VEHICLES"],"level":bl,"gap_pct":bg},
            "after":{"objective":aobj,"distance":am["DISTANCE"],"time":am["TIME"],"cost":am["COST"],"vehicles":am["VEHICLES"],"level":al,"gap_pct":ag},
        }

    def _offer_and_run_continuity(self, ans, baseline_solution, baseline_certificate, baseline_audit_recommendation):
        """
        Continuidade opcional pós-Relatório 1 com orçamento por DEADLINE GLOBAL.
        Regras de execução:
        - EARLY_STOPPING encerra apenas uma rodada do GA, não a continuidade.
        - enquanto houver orçamento útil, novas rodadas são iniciadas com novas seeds;
        - o melhor incumbent auditado é preservado e vira o warm start da rodada seguinte;
        - em LARGE_SCALE, reserva-se apenas uma pequena janela final para certificação,
          pois o complexity guard pode impedir o SCIP completo;
        - o orçamento efetivamente usado e o número de restarts ficam registrados.
        """
        level = str((baseline_certificate or {}).get("certification_level", "D") or "D").upper()
        if level not in {"B+", "B", "C", "D"}:
            return None

        print("\n" + "=" * 70)
        print("🔄 CONTINUIDADE OPCIONAL DA OTIMIZAÇÃO")
        print("=" * 70)
        print(f"A solução do Relatório 1 está preservada. Certificação atual: Nível {level}.")
        if level == "B":
            print("A garantia matemática já é forte; a continuidade tentará reduzir ainda mais o gap ou melhorar a solução.")
        else:
            print("É possível usar tempo adicional para tentar melhorar a solução e/ou fortalecer a certificação.")

        AvisoExecucaoV534.aguardando_continuidade()
        if not self._continuity_yes(input("Deseja continuar? [s/N]: ")):
            self.memory.data["continuity_state"] = {"status": "DECLINED"}
            return None

        budget = int(self._ask_continuity_budget())
        primary = self._continuity_snapshot()
        continuity = self.agents["continuity"]
        # referências técnicas; não fazem parte do relatório do usuário.
        self.memory.data["_ave_agent"] = self.agents["ave"]
        self.memory.data["_bound_strengthening_agent"] = self.agents["bound_strengthener"]
        self.memory.data["_k_vehicle_feasibility_certifier"] = self.agents["k_vehicle_certifier"]
        self.memory.data["_optimality_certifier_agent"] = self.agents["certifier"]
        continuity_ctx = continuity.build_context(
            baseline_solution,
            baseline_certificate,
            budget,
        )

        # validação explícita da ponte antes de gastar orçamento.
        initial_plan = (
            continuity_ctx.get("improvement_plan", {})
            if isinstance(continuity_ctx, dict)
            else {}
        )
        initial_strategies = initial_plan.get("strategies", []) or []
        print("\n🔗 PLANO RECEBIDO PELO ORQUESTRADOR:")
        for st in initial_strategies:
            if not isinstance(st, dict):
                continue
            extra = ""
            if st.get("type") == "FLEET_REDUCTION":
                extra = (
                    f" | {st.get('current_used_vehicles')}→"
                    f"{st.get('target_max_vehicles')} veículos"
                )
            print(
                f"   - {st.get('type')} | prioridade={st.get('priority')}{extra}"
            )

        started = _time_module.monotonic()
        deadline = started + float(budget)
        n_clients = int(self._problem_size())

        # LARGE_SCALE: o Certifier já possui complexity guard. Não bloqueamos a
        # certificação, mas evitamos reservar grande parte do orçamento para uma
        # etapa que pode ser encerrada imediatamente pelo guard.
        if n_clients > 50:
            cert_reserve = min(45.0, max(20.0, 0.025 * float(budget)))
            restart_slice = min(360.0, max(90.0, 0.20 * float(budget)))
        elif n_clients > 20:
            cert_reserve = min(180.0, max(60.0, 0.15 * float(budget)))
            restart_slice = min(300.0, max(75.0, 0.25 * float(budget)))
        else:
            cert_reserve = min(240.0, max(60.0, 0.20 * float(budget)))
            restart_slice = min(240.0, max(60.0, 0.25 * float(budget)))

        search_deadline = max(started, deadline - cert_reserve)
        # MULTI13.1: orçamento efetivo da fase de busca, sem a reserva de certificação.
        search_budget_seconds = max(0.0, search_deadline - started)
        self.memory.data["continuity_state"] = {
            "status": "RUNNING",
            "budget_seconds": budget,
            "deadline_monotonic": deadline,
            "search_deadline_monotonic": search_deadline,
            "certification_reserve_seconds": cert_reserve,
            "restart_slice_seconds": restart_slice,
            "restarts": 0,
            "audited_improvements": 0,
        }

        baseline_metrics = self._extract_candidate_metrics(baseline_solution)
        baseline_score = self._canonical_score(baseline_metrics)
        best_candidate = copy.deepcopy(baseline_solution)
        best_score = float(baseline_score)
        restart_history = []
        audited_improvements = 0
        restart_idx = 0
        # evita gastar o orçamento repetindo uma bacia já esgotada.
        # Só interrompe após duas rodadas consecutivas sem UB, sem LB e sem nova região GA.
        stagnant_rounds = 0

        def audit_cb(candidate, label):
            self.memory.data["solver_result"] = copy.deepcopy(candidate)
            d, r = self._audit_result(ans, candidate, label, self.tentativas)
            self.memory.data["continuity_candidate_audit"] = {
                "diagnostic": copy.deepcopy(d),
                "recommendation": copy.deepcopy(r),
                "independent_report": copy.deepcopy(
                    self.memory.data.get("independent_feasibility_report_auditor")
                ),
            }
            return d, r

        print(
            f"\n⏱️ Continuidade: orçamento global={budget}s | "
            f"busca até ~{max(0, int(search_deadline-started))}s | "
            f"reserva certificação≈{int(cert_reserve)}s"
        )
        print(
            f"🌱 Incumbent preservado | score={best_score:.12f} | "
            f"clientes={n_clients}"
        )


        # BUSCA DE MELHORIA — RESTARTS ATÉ O DEADLINE

        # EARLY_STOPPING pertence a uma execução do GA. Se ela terminar cedo,
        # voltamos ao while e iniciamos outra rodada enquanto houver tempo.
        while True:
            now = _time_module.monotonic()
            remaining_search = search_deadline - now
            if remaining_search < 35.0:
                break

            restart_idx += 1
            round_budget = int(max(30.0, min(restart_slice, remaining_search)))
            round_started = _time_module.monotonic()

            # O GA lê continuity_seed_solution como warm start.
            # Sempre apontamos essa chave para o MELHOR incumbent auditado atual.
            self.memory.data["continuity_seed_solution"] = copy.deepcopy(best_candidate)
            self.memory.data["continuity_candidate"] = copy.deepcopy(best_candidate)


            # DIVERSIFICAÇÃO DETERMINÍSTICA ENTRE RESTARTS ()

            # O ContinuityOptimizationAgent pode possuir uma seed padrão própria.
            # Para garantir que cada restart explore uma trajetória diferente SEM
            # alterar o ContinuityAgent nem o GA, o Orchestrator passa um proxy do
            # GA que força a seed desta rodada em toda chamada solve_problem().

            # 7919 é primo: gera uma sequência simples, determinística e reproduzível.
            continuity_base_seed = int(ans.get("continuity_base_seed", 20260917) or 20260917)
            restart_seed = (continuity_base_seed + (restart_idx - 1) * 7919) % 2147483647
            if restart_seed <= 0:
                restart_seed = restart_idx

            self.memory.data["continuity_restart_seed"] = int(restart_seed)

            class _SeedForcedGASolverProxy:
                """
                - NÃO destrói a phase_seed criada pelo ContinuityAgent para cada estratégia;
                - garante que a diretiva chegue ao GA mesmo quando translated_json é string;
                - mantém compatibilidade com chamadas antigas.
                """
                def __init__(self, wrapped_solver, forced_seed, shared_memory):
                    self._wrapped_solver = wrapped_solver
                    self._forced_seed = int(forced_seed)
                    self._shared_memory = shared_memory

                def solve_problem(self, *args, **kwargs):
                    import copy as _copy
                    import json as _json

                    # Só usa a seed do restart se o ContinuityAgent não tiver
                    # fornecido uma phase_seed específica.
                    kwargs.setdefault("seed", self._forced_seed)
                    kwargs["continuity_mode"] = True

                    args_list = list(args)

                    # CONTRATO LOCAL PRIMEIRO.
                    # A diretiva pertencente a ESTA chamada nunca pode ser
                    # substituída por estado global/stale da SharedMemory.
                    directive = None
                    seed_solution = None
                    if len(args_list) >= 2:
                        translated_arg0 = args_list[1]
                        try:
                            payload0 = (_json.loads(translated_arg0)
                                        if isinstance(translated_arg0, str)
                                        else translated_arg0)
                        except Exception:
                            payload0 = None
                        if isinstance(payload0, dict):
                            local_d = payload0.get("_continuity_directive")
                            if isinstance(local_d, dict):
                                directive = _copy.deepcopy(local_d)
                            local_seed = payload0.get("_continuity_seed_solution")
                            if isinstance(local_seed, dict):
                                seed_solution = _copy.deepcopy(local_seed)

                    if not isinstance(directive, dict):
                        directive = _copy.deepcopy(
                            self._shared_memory.data.get("continuity_search_directive")
                        )
                    if not isinstance(seed_solution, dict):
                        seed_solution = _copy.deepcopy(
                            self._shared_memory.data.get("continuity_seed_solution")
                        )

                    # Fail-fast: uma busca declarada FLEET_REDUCTION sem alvo é
                    # erro de integração, nunca uma busca GENERIC silenciosa.
                    if isinstance(directive, dict) and str(directive.get("type", "")).upper() == "FLEET_REDUCTION":
                        if directive.get("target_max_vehicles") is None:
                            raise RuntimeError(
                                "CONTRACT_ERROR: FLEET_REDUCTION sem target_max_vehicles na ponte Orchestrator→GA"
                            )

                    # Segundo canal de segurança: injeta a diretiva no argumento
                    # translated_json (posição 1) preservando dict/string JSON.
                    if len(args_list) >= 2 and isinstance(directive, dict):
                        translated_arg = args_list[1]
                        original_is_string = isinstance(translated_arg, str)

                        if original_is_string:
                            try:
                                payload = _json.loads(translated_arg)
                            except Exception:
                                payload = {}
                        elif isinstance(translated_arg, dict):
                            payload = _copy.deepcopy(translated_arg)
                        else:
                            payload = {}

                        if isinstance(payload, dict):
                            payload["_continuity_directive"] = _copy.deepcopy(directive)
                            payload["_continuity_seed_solution"] = _copy.deepcopy(seed_solution)
                            payload["_continuity_bridge_version"] = "ORCHESTRATOR_V10_12_EXPLICIT_K_PROOF_MODE"
                            payload["continuity_strategy"] = str(
                                directive.get("type", "") or ""
                            )
                            if directive.get("target_max_vehicles") is not None:
                                payload["continuity_target_max_vehicles"] = int(
                                    directive["target_max_vehicles"]
                                )
                            payload["continuity_mode"] = True

                            args_list[1] = (
                                _json.dumps(payload, ensure_ascii=False)
                                if original_is_string
                                else payload
                            )

                    print(
                        "      🔗 Ponte Orquestrador→GA | "
                        f"estratégia={directive.get('type') if isinstance(directive, dict) else 'N/A'} | "
                        f"target_max_vehicles="
                        f"{directive.get('target_max_vehicles', 'N/A') if isinstance(directive, dict) else 'N/A'} | "
                        f"seed={kwargs.get('seed')}"
                    )

                    return self._wrapped_solver.solve_problem(*tuple(args_list), **kwargs)

                def __getattr__(self, name):
                    return getattr(self._wrapped_solver, name)

            ga_solver_for_restart = _SeedForcedGASolverProxy(
                self.agents["ga_solver"],
                restart_seed,
                self.memory,
            )

            print("\n" + "-" * 70)
            print(
                f"🔁 CONTINUIDADE — restart {restart_idx} | "
                f"seed={restart_seed} | "
                f"limite da rodada={round_budget}s | "
                f"tempo restante de busca≈{int(remaining_search)}s"
            )
            print(f"   warm start auditado: score={best_score:.12f}")

            try:
                search = continuity.search_better_incumbent(
                    ga_solver_for_restart,
                    audit_cb,
                    self.memory.data["problem_classification"],
                    self.memory.data["translated_json"],
                    best_candidate,
                    round_budget,
                    ave_agent=self.agents["ave"],
                    bound_agent=self.agents["bound_strengthener"],
                    k_vehicle_certifier=self.agents["k_vehicle_certifier"],
                    optimality_certifier=self.agents["certifier"],
                )
            except Exception as exc:
                elapsed_round = _time_module.monotonic() - round_started
                restart_history.append({
                    "restart": restart_idx,
                    "seed": int(restart_seed),
                    "budget_seconds": round_budget,
                    "elapsed_seconds": elapsed_round,
                    "status": "ERROR",
                    "error": f"{type(exc).__name__}: {exc}",
                    "best_score_after": best_score,
                })
                print(f"⚠️ Restart {restart_idx} falhou: {type(exc).__name__}: {exc}")
                # Uma falha isolada não destrói o baseline nem encerra a continuidade.
                if elapsed_round < 1.0:
                    _time_module.sleep(0.05)
                continue

            elapsed_round = _time_module.monotonic() - round_started
            proposed = search.get("best_candidate") if isinstance(search, dict) else None
            improved_this_round = False
            proposed_score = None

            if isinstance(proposed, dict) and proposed.get("status") == "SUCCESS":
                proposed_metrics = self._extract_candidate_metrics(proposed)
                proposed_score = float(self._canonical_score(proposed_metrics))

                # O ContinuityAgent já audita seus candidatos durante a busca, mas
                # fazemos um gate final explícito antes de promover o warm start.
                if proposed_score < best_score - 1e-10:
                    _, proposed_rec = audit_cb(
                        proposed,
                        f"CONTINUITY_RESTART_{restart_idx}_IMPROVEMENT_GATE",
                    )
                    proposed_report = copy.deepcopy(
                        self.memory.data.get("independent_feasibility_report_auditor")
                        or self.memory.data.get("last_independent_feasibility_report")
                    )
                    if (
                        proposed_rec.get("solucao_aprovada", False)
                        and isinstance(proposed_report, dict)
                        and proposed_report.get("feasible", False)
                    ):
                        normalized = proposed_report.get("normalized_solution")
                        promoted = (
                            copy.deepcopy(normalized)
                            if isinstance(normalized, dict)
                            and normalized.get("status") == "SUCCESS"
                            else copy.deepcopy(proposed)
                        )
                        promoted_score = float(
                            self._canonical_score(self._extract_candidate_metrics(promoted))
                        )
                        if promoted_score < best_score - 1e-10:
                            best_candidate = promoted
                            best_score = promoted_score
                            audited_improvements += 1
                            improved_this_round = True
                            self.memory.data["continuity_seed_solution"] = copy.deepcopy(best_candidate)
                            print(
                                f"🏆 Melhoria auditada no restart {restart_idx}: "
                                f"score={best_score:.12f}"
                            )

            track_a_info = (
                search.get("track_a", {})
                if isinstance(search, dict)
                else {}
            )
            strategy_attempts = (
                track_a_info.get("strategy_attempts", [])
                if isinstance(track_a_info, dict)
                else []
            )

            restart_history.append({
                "restart": restart_idx,
                "seed": int(restart_seed),
                "budget_seconds": round_budget,
                "elapsed_seconds": elapsed_round,
                "status": (search.get("status") if isinstance(search, dict) else "UNKNOWN"),
                "proposed_score": proposed_score,
                "improved": improved_this_round,
                "best_score_after": best_score,
                "strategy_attempts": copy.deepcopy(strategy_attempts),
            })

            self.memory.data["continuity_state"].update({
                "restarts": restart_idx,
                "last_restart_seed": int(restart_seed),
                "audited_improvements": audited_improvements,
                "best_score": best_score,
                "elapsed_seconds": _time_module.monotonic() - started,
                "remaining_seconds": max(0.0, deadline - _time_module.monotonic()),
            })
            self.memory.data["continuity_history"] = copy.deepcopy(restart_history)

            # parada adaptativa por ausência REAL de informação nova.
            # Não confundir uma rodada sem melhora com prova de ótimo/inviabilidade.
            track_b_info = search.get("track_b", {}) if isinstance(search, dict) else {}
            ga_info = track_a_info.get("ga_diversification", {}) if isinstance(track_a_info, dict) else {}
            lb_changed = bool(track_b_info.get("strengthened", False)) if isinstance(track_b_info, dict) else False
            ga_new_region = bool(ga_info.get("structurally_different", False)) if isinstance(ga_info, dict) else False
            round_has_new_information = bool(improved_this_round or lb_changed or ga_new_region)

            if round_has_new_information:
                stagnant_rounds = 0
            else:
                stagnant_rounds += 1

            self.memory.data["continuity_state"]["stagnant_rounds"] = int(stagnant_rounds)

            # só encerra por estagnação se o AVE  realmente concluiu a
            # escalada disponível para redução de frota no incumbent corrente.
            ave_info = track_a_info.get("ave", {}) if isinstance(track_a_info, dict) else {}
            ave_version = str(ave_info.get("version", "")) if isinstance(ave_info, dict) else ""
            fleet_diag = ave_info.get("fleet_reduction_diagnostics", []) if isinstance(ave_info, dict) else []
            fleet_levels_complete = True
            if isinstance(fleet_diag, list) and fleet_diag:
                fleet_levels_complete = all(
                    int(d.get("max_level_attempted", 0) or 0) >= 5
                    for d in fleet_diag if isinstance(d, dict)
                )
            elif ave_version.startswith("AVE_V2"):
                # Sem diagnóstico de frota não presumimos que a escalada 3→2 foi esgotada.
                fleet_levels_complete = False

            remaining_search_now = max(
                0.0, search_deadline - _time_module.monotonic()
            )
            # MULTI13.1: a parada por estagnação considera apenas a fase de busca.
            # A reserva final para certificação permanece protegida.
            can_stop_for_stagnation = bool(
                stagnant_rounds >= 3
                and remaining_search_now
                    < max(180.0, 0.20 * search_budget_seconds)
                and (
                    not ave_version.startswith("AVE_V2")
                    or fleet_levels_complete
                )
            )
            if can_stop_for_stagnation:
                print(
                    "⏹️ Continuidade: 3 rodadas consecutivas sem novo UB, novo LB "
                    "ou nova região estrutural, e a escalada AVE disponível foi esgotada. "
                    "Encerrando busca repetitiva e preservando orçamento para certificação."
                )
                break
            elif stagnant_rounds >= 2 and ave_version.startswith("AVE_V2") and not fleet_levels_complete:
                print(
                    '🔁 Estagnação detectada, mas AVE  ainda possui níveis de FleetReduction não esgotados; continuidade será mantida sem interpretar a falha do GA como fim da busca.'
                )

        search_elapsed = _time_module.monotonic() - started
        candidate = copy.deepcopy(best_candidate)


        # AUDITORIA PRÉ-CERTIFICAÇÃO

        _, cand_rec = audit_cb(candidate, "CONTINUITY_PRE_CERTIFICATION")
        pre = copy.deepcopy(
            self.memory.data.get("independent_feasibility_report_auditor")
            or self.memory.data.get("last_independent_feasibility_report")
        )
        if (
            not cand_rec.get("solucao_aprovada", False)
            or not isinstance(pre, dict)
            or not pre.get("feasible", False)
        ):
            candidate = copy.deepcopy(baseline_solution)
            _, cand_rec = audit_cb(candidate, "CONTINUITY_BASELINE_FALLBACK")
            pre = copy.deepcopy(
                self.memory.data.get("independent_feasibility_report_auditor")
                or self.memory.data.get("last_independent_feasibility_report")
            )

        self.memory.data["independent_feasibility_report_pre_certification"] = copy.deepcopy(pre)
        self.memory.data["best_incumbent"] = copy.deepcopy(
            pre.get("normalized_solution", candidate) if isinstance(pre, dict) else candidate
        )
        # ponte segura do Track B da continuidade para o Certifier.
        # O Relatório 1 permanece imutável. Apenas carregamos para a segunda
        # certificação o maior LB globalmente válido já conhecido.
        baseline_valid_lb = (
            float(baseline_certificate["lower_bound"])
            if baseline_certificate.get("lower_bound_certified")
            and baseline_certificate.get("lower_bound") is not None
            else None
        )
        continuity_valid_lb = self.memory.data.get("continuity_best_valid_lower_bound")
        try:
            continuity_valid_lb = (
                float(continuity_valid_lb) if continuity_valid_lb is not None else None
            )
        except Exception:
            continuity_valid_lb = None

        carried_valid_lbs = [
            x for x in (baseline_valid_lb, continuity_valid_lb)
            if x is not None
        ]
        self.memory.data["continuity_baseline_valid_lower_bound"] = (
            max(carried_valid_lbs) if carried_valid_lbs else None
        )
        self.memory.data["continuity_carried_valid_lower_bound"] = (
            self.memory.data["continuity_baseline_valid_lower_bound"]
        )


        # CERTIFICAÇÃO COM O TEMPO REALMENTE RESTANTE

        certifier = self.agents["certifier"]
        remaining_total = max(0.0, deadline - _time_module.monotonic())
        cert_budget = max(1, int(remaining_total))

        # Se o orçamento já foi praticamente consumido, ainda permitimos uma
        # passagem curta: bounds analíticos/complexity guard costumam ser rápidos.
        certifier.time_limit_seconds = max(1, min(cert_budget, 900))
        certifier.intensive_certification = bool(cert_budget > 900)
        certifier.intensive_time_limit_seconds = (
            max(1, int(cert_budget - 900)) if cert_budget > 900 else 1
        )

        print(
            f"\n📐 Continuidade → certificação | "
            f"tempo restante={remaining_total:.1f}s | "
            f"limite base={certifier.time_limit_seconds}s"
        )

        provisional = certifier.certify_optimality()
        self.memory.data["continuity_certification_provisional"] = copy.deepcopy(provisional)
        post_candidate = provisional.get("certified_solution") if isinstance(provisional, dict) else None
        if not isinstance(post_candidate, dict):
            post_candidate = copy.deepcopy(candidate)
        post_candidate["status"] = "SUCCESS"

        _, post_rec = audit_cb(post_candidate, "CONTINUITY_POST_CERTIFICATION")
        post_report = copy.deepcopy(
            self.memory.data.get("independent_feasibility_report_auditor")
            or self.memory.data.get("last_independent_feasibility_report")
        )
        finalized = certifier.finalize_after_independent_audit(provisional, post_report)
        after_cert = copy.deepcopy(finalized.get("certificate", {}) or {})
        after_solution = finalized.get("certified_solution")

        if not isinstance(after_solution, dict) or not post_rec.get("factibilidade_aprovada", False):
            after_solution = copy.deepcopy(candidate)
            after_cert = copy.deepcopy(baseline_certificate)

        after_solution["status"] = "SUCCESS"
        after_solution["certificate"] = copy.deepcopy(after_cert)
        comparison = self._continuity_comparison(
            baseline_solution,
            baseline_certificate,
            after_solution,
            after_cert,
        )

        total_elapsed = _time_module.monotonic() - started
        budget_utilization_pct = min(100.0, 100.0 * total_elapsed / max(1.0, float(budget)))
        stats = {
            "budget_seconds": budget,
            "elapsed_seconds": total_elapsed,
            "budget_utilization_pct": budget_utilization_pct,
            "search_elapsed_seconds": search_elapsed,
            "certification_elapsed_seconds": max(0.0, total_elapsed - search_elapsed),
            "restarts": restart_idx,
            "audited_improvements": audited_improvements,
            "baseline_score": baseline_score,
            "final_score": self._canonical_score(self._extract_candidate_metrics(after_solution)),
            "restart_history": copy.deepcopy(restart_history),
        }

        self.memory.data["continuity_final_solution"] = copy.deepcopy(after_solution)
        self.memory.data["continuity_certificate"] = copy.deepcopy(after_cert)
        self.memory.data["continuity_final_audit"] = copy.deepcopy(post_rec)
        self.memory.data["continuity_comparison"] = copy.deepcopy(comparison)
        self.memory.data["continuity_statistics"] = copy.deepcopy(stats)

        print("\n" + "=" * 70)
        print("⏱️ RESUMO DO ORÇAMENTO DA CONTINUIDADE")
        print("=" * 70)
        print(f"Solicitado: {budget:.0f}s ({budget/60.0:.1f} min)")
        print(f"Utilizado:  {total_elapsed:.1f}s ({total_elapsed/60.0:.1f} min)")
        print(f"Utilização: {budget_utilization_pct:.1f}%")
        print(f"Restarts de busca: {restart_idx}")
        print(f"Melhorias auditadas: {audited_improvements}")
        print(f"Score baseline: {baseline_score:.12f}")
        print(f"Score final:    {stats['final_score']:.12f}")

        # Relatório 2 usa namespace próprio; o Relatório 1 é restaurado em seguida.
        self.memory.data["solver_result"] = copy.deepcopy(after_solution)
        diag2 = self.agents["diagnostic"].analisar_continuidade(
            baseline_solution, after_solution, comparison
        )
        report2 = self.agents["explainer"].explain_continuity_result(comparison, diag2)
        print("\n" + report2)

        self.memory.data["continuity_state"] = {
            "status": "REPORT_2_READY",
            "budget_seconds": budget,
            "elapsed_seconds": total_elapsed,
            "budget_utilization_pct": budget_utilization_pct,
            "restarts": restart_idx,
            "audited_improvements": audited_improvements,
        }

        saved_cont = {
            k: copy.deepcopy(self.memory.data.get(k))
            for k in [
                "continuity_context",
                "continuity_state",
                "continuity_baseline",
                "continuity_seed_solution",
                "continuity_candidate",
                "continuity_candidate_audit",
                "continuity_certification_provisional",
                "continuity_certificate",
                "continuity_final_solution",
                "continuity_final_audit",
                "continuity_comparison",
                "continuity_diagnostic",
                "continuity_report",
                "continuity_history",
                "continuity_statistics",
                "continuity_baseline_valid_lower_bound",
                "continuity_best_valid_lower_bound",
                "continuity_bound_candidates",
                "continuity_bound_certificate",
                "continuity_carried_valid_lower_bound",
            ]
        }
        self._restore_primary_snapshot(primary)
        for k, v in saved_cont.items():
            self.memory.data[k] = v

        return {
            "status": "COMPLETED",
            "final_solution": after_solution,
            "certificate": after_cert,
            "comparison": comparison,
            "report": report2,
            "statistics": stats,
        }


    # FLUXO PRINCIPAL


    def orchestrate_solution_process(self):
        print(
            "\n--- Iniciando Sistema Multiagente "
            "com Certificação de Otimalidade ---"
        )
        print(f"[Orchestrator {self.VERSION}]")

        try:
            self.tentativas = 0
            self.force_math_change = False
            self._quality_intensification_mode = False
            self._quality_intensification_runs = 0
            self._approved_incumbents = []
            self.memory.data["audited_incumbent_history"] = []
            self.memory.data["best_audited_incumbent"] = None

            user_interface = UserInterface(self.memory)
            ans = user_interface.iniciar()

            self.memory.data["user_answers"] = ans

            self._executar_fluxo_pre_solver(ans)
            AvisoExecucaoV534.iniciar_processamento()

            self._global_optimization_started = _time_module.monotonic()
            self._active_budget_profile = (
                self._optimization_budget_profile()
            )

            print(
                f"⏱️ Perfil de otimização: "
                f"{self._active_budget_profile['profile']} | "
                f"limite dos solvers: "
                f"{self._active_budget_profile['global_seconds']}s"
            )

            recommendation = {
                "solucao_aprovada": False
            }

            solver_result = None
            approach = None


            # OR-TOOLS / GA + AUDITORIA DE FACTIBILIDADE

            while self.tentativas < self.max_tentativas:
                self.tentativas += 1

                print(
                    f"\n--- Tentativa {self.tentativas} "
                    f"de {self.max_tentativas} ---"
                )

                # Depois que já existe um incumbent auditado, novas tentativas
                # são usadas como intensificação de qualidade. Evitamos repetir
                # o mesmo portfólio determinístico com as mesmas seeds.
                if self._quality_intensification_mode:
                    approach = "GA_QUALITY_INTENSIFICATION"
                elif (
                    self._active_budget_profile.get("profile") == "LARGE_SCALE"
                    and self.tentativas > 1
                    and not self._approved_incumbents
                ):
                    # Em instâncias grandes, depois de uma tentativa sem incumbent
                    # auditado, não repetimos OR-Tools deterministicamente. Mudamos
                    # explicitamente para GA orientado primeiro à factibilidade.
                    approach = "GA_LARGE_SCALE_FEASIBILITY_RECOVERY"
                    print("🛟 Recuperação LARGE_SCALE: prioridade = encontrar solução factível.")
                else:
                    approach = self.agents[
                        "math"
                    ].decide_optimization_approach(
                        self.memory.data["problem_classification"],
                        self.memory.data["translated_json"],
                        force_change=self.force_math_change,
                    )

                self.memory.data["optimization_approach"] = approach

                if "PORTFOLIO" in approach:
                    solver_result = self._run_solver_portfolio()

                elif "Solver" in approach:
                    solver_result = self.agents[
                        "ortools_solver"
                    ].solve_problem(
                        self.memory.data["problem_classification"],
                        self.memory.data["translated_json"],
                    )
                    solver_result["portfolio_complete"] = False
                    self._register_single_solver_incumbent(
                        solver_result,
                        "OR-Tools",
                    )

                else:
                    remaining_now = self._remaining_optimization_seconds()
                    if remaining_now <= 20:
                        print(
                            "⏹️ Orçamento de otimização insuficiente para "
                            "nova execução do GA."
                        )
                        break

                    requested_ga_seconds = self._active_budget_profile["ga_seconds"]
                    if approach == "GA_LARGE_SCALE_FEASIBILITY_RECOVERY":
                        requested_ga_seconds = self._active_budget_profile.get(
                            "feasibility_recovery_seconds", requested_ga_seconds
                        )
                    ga_budget = min(
                        requested_ga_seconds,
                        max(20, int(remaining_now - 5)),
                    )
                    ga_seed = random.randint(1, 100000)

                    if self._quality_intensification_mode:
                        self._quality_intensification_runs += 1
                        print(
                            "🔎 Intensificação de qualidade: procurando "
                            "incumbent auditado melhor antes da certificação."
                        )

                    solver_result = self.agents[
                        "ga_solver"
                    ].solve_problem(
                        self.memory.data["problem_classification"],
                        self.memory.data["translated_json"],
                        population_size=self._active_budget_profile["ga_population"],
                        num_generations=self._active_budget_profile["ga_generations"],
                        time_limit_seconds=ga_budget,
                        seed=ga_seed,
                        feasibility_first=(
                            approach == "GA_LARGE_SCALE_FEASIBILITY_RECOVERY"
                            or self._active_budget_profile.get("profile") == "LARGE_SCALE"
                        ),
                    )
                    solver_result["portfolio_complete"] = False
                    solver_result["search_seed"] = ga_seed
                    self._register_single_solver_incumbent(
                        solver_result,
                        ("GA feasibility recovery" if approach == "GA_LARGE_SCALE_FEASIBILITY_RECOVERY"
                         else ("GA quality" if self._quality_intensification_mode else "GA")),
                    )

                if not isinstance(solver_result, dict):
                    solver_result = {
                        "status": "ERROR",
                        "error": "Solver retornou resultado inválido.",
                    }

                self.memory.data["solver_result"] = solver_result

                audit_diag, recommendation = self._audit_result(
                    ans,
                    solver_result,
                    approach,
                    self.tentativas,
                )

                self._print_audit_diagnostic(
                    audit_diag,
                    recommendation,
                    stage=f"PRE_CERTIFICATION_ATTEMPT_{self.tentativas}",
                )

                approved_now = bool(
                    recommendation.get("solucao_aprovada", False)
                )

                if approved_now:
                    best_record = self._register_audited_incumbent(
                        solver_result,
                        recommendation,
                        self.tentativas,
                        approach,
                    )

                    print(
                        "✅ Incumbent aprovado pela auditoria "
                        f"na tentativa {self.tentativas}."
                    )

                    if best_record is not None:
                        bm = best_record["metrics"]
                        print(
                            "🏆 Melhor incumbent AUDITADO até agora | "
                            f"score={best_record['score']:.12f} | "
                            f"tempo={bm['TIME']:.2f} min | "
                            f"distância={bm['DISTANCE']:.2f} km | "
                            f"custo={bm['COST']:.2f} | "
                            f"veículos={bm['VEHICLES']:.0f}"
                        )

                    if self._should_continue_quality_search():
                        self._quality_intensification_mode = True
                        self.force_math_change = False
                        print(
                            "🔎 A solução é factível, mas ainda há orçamento. "
                            "A busca continuará para tentar melhorar o incumbent "
                            "antes da certificação."
                        )
                        continue

                    print(
                        "⏹️ Busca de incumbents encerrada: orçamento/limite de "
                        "tentativas atingido. O melhor incumbent auditado será "
                        "enviado ao Certificador."
                    )
                    break

                # A tentativa atual falhou, mas um incumbent anterior aprovado
                # nunca deve ser perdido por causa disso.
                if self._approved_incumbents:
                    if self._should_continue_quality_search():
                        self._quality_intensification_mode = True
                        self.force_math_change = False
                        print(
                            "⚠️ A tentativa atual não foi aprovada, mas existe "
                            "incumbent auditado anterior. Ele foi preservado e "
                            "a busca de qualidade continuará."
                        )
                        continue

                    print(
                        "ℹ️ A tentativa atual foi reprovada, porém existe um "
                        "incumbent auditado anterior. O melhor deles será usado."
                    )
                    break

                # Sem nenhum incumbent aprovado, mantém-se a lógica de
                # recuperação orientada pelo Auditor.
                if (
                    self.tentativas < self.max_tentativas
                    and self._deve_reexecutar(
                        solver_result,
                        recommendation,
                    )
                ):
                    next_action = recommendation.get(
                        "proxima_acao_orquestrador"
                    )

                    if next_action == "TROCAR_ESTRATEGIA_MATEMATICA":
                        print(
                            "🔄 Auditor recomendou troca "
                            "de estratégia matemática."
                        )
                        self.force_math_change = True
                    else:
                        print("🔁 Nova tentativa de otimização.")

                    continue

                next_action = recommendation.get(
                    "proxima_acao_orquestrador"
                )

                if next_action == "TROCAR_ESTRATEGIA_MATEMATICA":
                    self.force_math_change = True
                    print("🔄 Estratégia matemática será alterada.")
                elif next_action == "ABORTAR_INVIAVEL":
                    print(
                        "🛑 Problema considerado inviável pela auditoria "
                        "e nenhum incumbent aprovado foi preservado."
                    )
                    break


            # SELEÇÃO GLOBAL DO MELHOR INCUMBENT AUDITADO

            best_record = self._best_audited_incumbent()

            if best_record is None:
                print(
                    "⚠️ Nenhuma solução factível aprovada pelo Auditor foi "
                    "produzida. A certificação não será iniciada."
                )
                return None

            solver_result = copy.deepcopy(best_record["result"])
            self.memory.data["solver_result"] = copy.deepcopy(solver_result)
            self.memory.data["best_incumbent"] = copy.deepcopy(solver_result)
            self.memory.data["best_audited_incumbent"] = copy.deepcopy(
                solver_result
            )

            bm = best_record["metrics"]
            print(
                "\n🏆 INCUMBENT GLOBAL AUDITADO SELECIONADO PARA CERTIFICAÇÃO"
            )
            print(
                f"   tentativa={best_record['attempt']} | "
                f"abordagem={best_record['approach']} | "
                f"score={best_record['score']:.12f}"
            )
            print(
                f"   tempo={bm['TIME']:.6f} min | "
                f"distância={bm['DISTANCE']:.6f} km | "
                f"custo={bm['COST']:.6f} | "
                f"veículos={bm['VEHICLES']:.0f}"
            )

            # Reaudita exatamente o incumbent vencedor. Isso garante que o
            # relatório pré-certificação corresponde à solução que entrará
            # no SCIP, mesmo que a última tentativa executada tenha sido outra.
            audit_diag, recommendation = self._audit_result(
                ans,
                solver_result,
                "BEST_AUDITED_INCUMBENT_PRE_CERTIFICATION",
                self.tentativas,
            )

            self._print_audit_diagnostic(
                audit_diag,
                recommendation,
                stage="PRE_CERTIFICATION_BEST_INCUMBENT",
            )

            if not recommendation.get("solucao_aprovada", False):
                print(
                    "⚠️ Inconsistência: o incumbent global previamente aprovado "
                    "não passou na reauditoria final. Certificação cancelada."
                )
                return None


            # GATE OBRIGATÓRIO DO AUDITOR ANTES DA CERTIFICAÇÃO

            pre_report = copy.deepcopy(
                self.memory.data.get("independent_feasibility_report_auditor")
                or self.memory.data.get("last_independent_feasibility_report")
            )
            if not isinstance(pre_report, dict) or not pre_report.get("feasible", False):
                print("⚠️ O Auditor não produziu uma verificação independente válida. A certificação não será iniciada.")
                return None

            # One bounded initial-core refinement, before freezing the certification gate.
            solver_result, pre_report, refinement = self.agents['initial_refiner'].refine(
                pre_report['normalized_solution'], INITIAL_REFINEMENT_BUDGET_SECONDS)
            self.memory.data['solver_result'] = copy.deepcopy(solver_result)
            self.memory.data['independent_feasibility_report_auditor'] = copy.deepcopy(pre_report)
            self.memory.data['last_independent_feasibility_report'] = copy.deepcopy(pre_report)

            # Congela exatamente o relatório que autorizou a entrada no Certificador.
            self.memory.data["independent_feasibility_report_pre_certification"] = copy.deepcopy(pre_report)
            normalized_incumbent = pre_report.get("normalized_solution")
            if not isinstance(normalized_incumbent, dict) or normalized_incumbent.get("status") != "SUCCESS":
                raise ValueError("ORCHESTRATOR V10: Auditor aprovou a solução sem normalized_solution válida.")
            self.memory.data["best_incumbent"] = copy.deepcopy(normalized_incumbent)
            self.memory.data["solver_result_pre_certification"] = copy.deepcopy(solver_result)


            # CERTIFICAÇÃO MATEMÁTICA PROVISÓRIA (SCIP)

            provisional_result = self._run_optimality_certification(ans)
            provisional_certificate = provisional_result.get("certificate", {}) or {}
            candidate_solution = provisional_result.get("certified_solution")

            # Nível D não apaga o incumbent já auditado: ele pode ser entregue como
            # solução factível SEM certificação matemática conclusiva e, depois,
            # oferecido ao núcleo opcional de continuidade.
            if not isinstance(candidate_solution, dict):
                baseline_d = copy.deepcopy(self.memory.data.get("best_incumbent") or solver_result)
                baseline_d["status"] = "SUCCESS"
                d_cert = copy.deepcopy(provisional_certificate)
                d_cert["certification_level"] = "D"
                d_cert["proven_optimal"] = False
                d_cert["certification_declaration"] = d_cert.get("certification_declaration") or "Solução factível sem certificação matemática conclusiva"
                baseline_d["certificate"] = copy.deepcopy(d_cert)
                self.memory.data["solver_result"] = copy.deepcopy(baseline_d)
                self.memory.data["final_solution"] = copy.deepcopy(baseline_d)
                diagnostico, recomendacoes, relatorio = self._run_diagnostic_and_report(baseline_d, ans, recommendation)
                print("\n🕐 Certificação Nível D: o plano auditado foi entregue sem prova matemática conclusiva.")
                continuity_result = self._offer_and_run_continuity(ans, baseline_d, d_cert, recommendation)
                return {"status":"FEASIBLE_UNCERTIFIED","certification_level":"D","proven_optimal":False,"final_solution":baseline_d,"certificate":d_cert,"report":relatorio,"continuity_result":continuity_result}


            # AUDITORIA INDEPENDENTE OBRIGATÓRIA APÓS O SCIP

            candidate_solution = copy.deepcopy(candidate_solution)
            candidate_solution["status"] = "SUCCESS"
            self.memory.data["solver_result"] = copy.deepcopy(candidate_solution)

            final_audit_diag, final_recommendation = self._audit_result(
                ans,
                candidate_solution,
                "Auditoria final / limites analíticos iniciais",
                self.tentativas,
            )

            # Mesmo diagnóstico após o SCIP, para comparar a solução
            # certificada com o parecer independente final.
            self._print_audit_diagnostic(
                final_audit_diag,
                final_recommendation,
                stage="POST_INITIAL_ANALYTICAL_BOUNDS",
            )

            final_report = copy.deepcopy(
                self.memory.data.get("independent_feasibility_report_auditor")
                or self.memory.data.get("last_independent_feasibility_report")
            )
            self.memory.data["independent_feasibility_report_post_certification"] = copy.deepcopy(final_report)

            # O Certificador apenas FINALIZA A/B/C/D depois que o Orchestrator
            # recebe o parecer independente do Auditor.
            certification_result = self.agents["certifier"].finalize_after_independent_audit(
                provisional_result,
                final_report,
            )
            certificate = certification_result.get("certificate", {}) or {}
            level = certification_result.get("certification_level", certificate.get("certification_level", "D"))
            report_allowed = bool(certification_result.get("report_allowed", certificate.get("report_allowed", False)))
            certified_solution = certification_result.get("certified_solution")

            if level == "D" or not report_allowed or not isinstance(certified_solution, dict):
                baseline_d = copy.deepcopy(candidate_solution if isinstance(candidate_solution, dict) else self.memory.data.get("best_incumbent") or solver_result)
                baseline_d["status"] = "SUCCESS"
                d_cert = copy.deepcopy(certificate)
                d_cert["certification_level"] = "D"; d_cert["proven_optimal"] = False
                baseline_d["certificate"] = copy.deepcopy(d_cert)
                self.memory.data["solver_result"] = copy.deepcopy(baseline_d)
                self.memory.data["final_solution"] = copy.deepcopy(baseline_d)
                diagnostico, recomendacoes, relatorio = self._run_diagnostic_and_report(baseline_d, ans, final_recommendation)
                print("\n🕐 Certificação Nível D: solução factível entregue sem certificação matemática conclusiva.")
                continuity_result = self._offer_and_run_continuity(ans, baseline_d, d_cert, final_recommendation)
                return {"status":"FEASIBLE_UNCERTIFIED","certification_level":"D","proven_optimal":False,"final_solution":baseline_d,"certificate":d_cert,"report":relatorio,"continuity_result":continuity_result}

            # A/B/C somente chegam aqui se a auditoria independente pós-SCIP passou.
            if not final_recommendation.get("factibilidade_aprovada", False):
                self.memory.data["final_solution"] = None
                raise ValueError("ORCHESTRATOR V10: inconsistência — nível A/B/C sem aprovação final do Auditor.")

            final_solution = copy.deepcopy(certified_solution)
            final_solution["status"] = "SUCCESS"
            final_solution["certificate"] = copy.deepcopy(certificate)
            self.memory.data["certified_solution"] = copy.deepcopy(final_solution)
            self.memory.data["solver_result"] = copy.deepcopy(final_solution)
            self.memory.data["final_solution"] = copy.deepcopy(final_solution)

            diagnostico, recomendacoes, relatorio = self._run_diagnostic_and_report(
                final_solution, ans, final_recommendation
            )

            print("\n" + "=" * 70)
            print("🏁 PROCESSAMENTO CONCLUÍDO")
            print("=" * 70)
            print(f"📐 Nível de certificação: {level}")
            print(f"📌 {certificate.get('certification_declaration')}")
            if certificate.get("gap_pct") is not None:
                print(f"📐 Gap certificado: {float(certificate['gap_pct']):.6f}%")
            print(f"🏆 Prova pelo solver global exato: {'SIM' if certificate.get('proven_optimal') else 'NÃO'}")
            print("🛡️ Auditoria independente pré-SCIP: APROVADA")
            print("🛡️ Auditoria independente pós-SCIP: APROVADA")
            print(f"📊 Diagnóstico/infográfico executado: {'SIM' if diagnostico is not None else 'NÃO'}")
            print("✅ Sistema finalizado preservando a distinção entre A, B e C.")

            continuity_result = self._offer_and_run_continuity(
                ans, final_solution, certificate, final_recommendation
            )

            return {
                "status": certificate.get("status"),
                "certification_level": level,
                "certification_declaration": certificate.get("certification_declaration"),
                "proven_optimal": bool(certificate.get("proven_optimal", False)),
                "final_solution": final_solution,
                "certificate": certificate,
                "report": relatorio,
                "continuity_result": continuity_result,
            }

        except Exception as exc:
            print("\n" + "=" * 70)
            print("❌ ERRO CRÍTICO NA ORQUESTRAÇÃO")
            print("=" * 70)
            print(f"{type(exc).__name__}: {exc}")

            try:
                self.memory.data[
                    "orchestration_error"
                ] = {
                    "type": type(exc).__name__,
                    "message": str(exc),
                    "tentativa": self.tentativas,
                }
            except Exception:
                pass

            print(
                "ℹ️ Os dados armazenados na SharedMemory "
                "foram preservados."
            )

            return None
````

### **FERRAMENTAS INTERNAS DO ORQUESTRADOR E DOS AGENTES**




#### Recursos, decisões do coordenador e orientação

Controla quanto tempo e memória cada tarefa pode usar e orienta o coordenador sobre a próxima tentativa. Assim, o sistema aproveita os recursos disponíveis e preserva o melhor plano já aprovado quando uma busca termina.


````python
def _v534_memory_budget(inputs):
    """Budget per solver, bounded by actual free RAM; user settings remain maxima."""
    import math
    requested = float(inputs.get('complete_model_memory_mb', 512.0))
    if not math.isfinite(requested) or requested <= 0:
        raise ValueError('Orçamento de memória inválido.')
    try:
        import psutil
        available = psutil.virtual_memory().available / 1024 ** 2
    except ImportError:
        available = None
    limit = min(requested, max(64.0, available * 0.2)) if available is not None else requested
    return dict(solver_mb=limit, available_mb=available, rss_mb=min(limit * 2 + 256, max(128.0, available * 0.25)) if available is not None else limit * 2 + 256)

# Ferramentas do coordenador: diagnóstico e escolha da próxima rodada.
# Não alteram objetivo, restrições, auditoria ou regras de aceitação de LB.
def _ct_guidance(result):
    import math
    if not result:
        return dict(kind='FIRST_ROUND', message='Há um plano inicial. Ainda não existe uma rodada de continuidade para diagnosticar.', timed_stages=[], recommended='automatic')
    stats=result.get('statistics',{});stages=stats.get('stages',{})
    timed=[];failures=[]
    time_markers={'CONTINUITY_STAGE_WALL_DEADLINE','BUILD_WALL_DEADLINE','SOLVER_WALL_DEADLINE','LP_WALL_DEADLINE','TIME_LIMIT','TIMELIMIT','WALL_DEADLINE','REGION_BUDGET_EXHAUSTED','NO_REMAINING_BUDGET','DEADLINE','TIMELIMIT_REACHED','DIVERSIFIED_TIME_LIMIT'}
    def scan(row):
        if isinstance(row,dict):
            for k,v in row.items():
                if k in ('status','termination_reason','parent_stop_reason','stop_reason','reason') and isinstance(v,str) and v.upper() in time_markers:
                    return True
                if isinstance(v,(dict,list)) and scan(v):return True
        elif isinstance(row,list):
            return any(scan(x) for x in row)
        return False
    for name,row in stages.items():
        if not isinstance(row,dict):continue
        base='EXATO_LOCAL' if name.startswith('EXATO_RETORNO') else name
        if scan(row) and base not in timed:timed.append(base)
        if row.get('error') or row.get('termination_reason') in {'WORKER_EXCEPTION','WORKER_EXIT_WITHOUT_RESULT','STAGE_LAUNCH_ERROR'}:
            failures.append(name)
    stop=stats.get('coordinator_stop_v524',{}).get('reason')
    if result.get('certificate',{}).get('proven_optimal'):
        kind='OPTIMAL';message='A prova de ótimo foi concluída dentro da tolerância adotada.';recommended='stop'
    elif failures or result.get('issues'):
        kind='EXECUTION_ISSUE';message='Há uma falha ou inconsistência registrada. Mais tempo, sozinho, não corrige um problema de execução; confira o diagnóstico antes de repetir.';recommended='review'
    elif result.get('status')=='INTERRUPTED' or stop=='USER_INTERRUPTED':
        kind='INTERRUPTED';message='A rodada foi interrompida. Isso não demonstra estagnação matemática; os resultados recuperados precisam estar auditados.';recommended='automatic'
    elif stop in {'STAGNATION_TWO_LOCAL_BATCHES_WITHOUT_GAIN','NO_FRESH_ELIGIBLE_REGION'}:
        kind='STAGNATION';message='A seleção atual terminou sem outra região útil ou após lotes sem ganho. Isso não prova ótimo. Uma alternativa é priorizar grupos maiores de viagens.';recommended='alternative'
        if timed:message+=' Também houve tarefas limitadas por prazo; o tempo não foi o único motivo registrado.'
    elif timed or stop=='INSUFFICIENT_TIME_FOR_USEFUL_REGION':
        kind='TIME_LIMIT';message='Há registro de limite de tempo. Uma rodada com orçamento maior pode dar mais espaço às tarefas interrompidas, sem garantir melhoria.';recommended='more_time' if timed else 'automatic'
    else:
        kind='UNDETERMINED';message='Os registros não permitem atribuir o encerramento somente a tempo ou estagnação. Você pode manter a seleção automática ou escolher uma alternativa.';recommended='automatic'
    def valid_number(value):
        return float(value) if isinstance(value,(int,float)) and math.isfinite(value) and value>=0 else None
    return dict(kind=kind,message=message,timed_stages=timed,recommended=recommended,
        requested_seconds=valid_number(stats.get('requested_seconds')),elapsed_seconds=valid_number(stats.get('elapsed_seconds')))


def _ct_guidance_text(result):
    info=_ct_guidance(result);text=info['message']
    if info.get('elapsed_seconds') is not None and info.get('requested_seconds') is not None:
        text+=f" Tempo utilizado: {info['elapsed_seconds']/60:.1f} de até {info['requested_seconds']/60:.1f} minutos."
    return text





def _ct_guided_regions(inputs,p,best,tried,stalled):
    if inputs.get('user_continuity_mode')!='alternative':
        return _v59_neighborhood_choices(p,best,tried,stalled)
    # Usa as mesmas regras de viabilidade e tamanho; muda a prioridade, não as restrições.
    collected={}
    for phase in (2,3,0):
        for row in _v59_neighborhood_choices(p,best,tried,phase):
            collected[row[3]]=row
    order={4:0,3:1,2:2}
    return sorted((order.get(len(r[2]),3),r[1],r[2],r[3]) for r in collected.values())


def _ct_guided_stage_budget(stage,proposed,total,previous,inputs):
    if stage=='EXATO_COMPLETO_GLOBAL':
        decision=inputs['complete_model_plan'];cap=min(decision['cap_seconds']+10.,total-2.);return max(0.,min(proposed,cap)),dict(decision,cap_seconds=cap,reason=decision['reason'])
    seconds,policy=_v511_stage_budget(stage,proposed,total,previous)
    if inputs.get('user_continuity_mode')!='more_time' or stage not in inputs.get('user_timed_stages',[]):
        return seconds,policy
    # O teto global e o orçamento restante continuam sendo respeitados.
    caps={'MELHORIA':180.,'LB_COMPACTO':300.,'EXATO_LOCAL':360.,'ELIMINAR_VEICULO':180.,'GERACAO_COLUNAS':90.}
    cap=min(.65*total,caps.get(stage,policy['cap_seconds']))
    cap=max(policy['cap_seconds'],cap)
    seconds=max(0.,min(float(proposed),float(total),cap))
    return seconds,dict(policy,cap_seconds=cap,reason='USER_MORE_TIME_FOR_TIMED_STAGE',previous_cap_seconds=policy['cap_seconds'])


def _ct_guided_region_seconds(inputs,base,remaining):
    if inputs.get('user_continuity_mode')=='more_time' and 'EXATO_LOCAL' in inputs.get('user_timed_stages',[]):
        base=min(180.,base*3.)
    return max(0.,min(base,remaining))


def _ct_guided_shares(shares,inputs):
    mode=inputs.get('user_continuity_mode')
    if mode=='complete_global':return {'EXATO_COMPLETO_GLOBAL':1.}
    if mode=='more_time':
        updated=dict(shares)
        for key in inputs.get('user_timed_stages',[]):
            if key in updated:updated[key]*=2
            elif key=='EXATO_LOCAL':updated[key]=.5
        return updated
    if mode!='alternative':return shares
    # Prioriza reorganização local, preservando tarefas elegíveis de LB e outras operações.
    values={'MELHORIA':.15,'LB_COMPACTO':.20,'EXATO_LOCAL':.55,'ELIMINAR_VEICULO':.10,'GERACAO_COLUNAS':.05}
    updated={key:values.get(key,value) for key,value in shares.items()}
    updated['EXATO_LOCAL']=values['EXATO_LOCAL']
    return updated


def _ct_guided_report_section(result):
    advice=result.get('statistics',{}).get('next_action_advice',{})
    text='## TEMPO UTILIZADO E MOTIVO DO ENCERRAMENTO\n\n'+_ct_budget_explanation(result)+'\n\n## ORIENTAÇÃO PARA CONTINUAR\n\n'+advice.get('reason',_ct_guidance_text(result))+'\n\n'
    rec=advice.get('recommended')
    if rec:text+='Recomendação: **opção '+rec+'** — '+advice['options'][rec]+'\n\n'
    for key,label in advice.get('options',{}).items():text+='- **'+key+'**: '+label+'\n'
    if advice.get('options'):text+='\nVocê pode encerrar com 0. Somente opções aplicáveis são oferecidas; o tempo será confirmado depois da escolha.\n'
    hist=result.get('round_history',[])
    if hist:text+='\nBenefício acumulado das rodadas registradas: redução de UB '+format(sum(r.get('ub_gain',0) for r in hist),'.8f')+'; aumento de LB '+format(sum(r.get('lb_gain',0) for r in hist),'.8f')+'.\n'
    text+='\nO melhor plano auditado e o maior LB válido são preservados. Esgotar uma estratégia não demonstra ótimo; nenhuma opção garante novo ganho.\n'
    return text



# VERIFICAÇÃO DE INTEGRAÇÃO — MULTI14
# Esta célula DEVE ficar depois da definição do OrchestratorAgent
# e antes das células de monitoramento/execução.

_expected_exact = {
    "KVehicleFeasibilityCertifier": "K_VEHICLE_FEASIBILITY_CERTIFIER_V5_FAIR_DEADLINES_SCOPE",
    "ContinuityOptimizationAgent": "CONTINUITY_OPTIMIZATION_AGENT_V10_PROCESS_PARALLEL_ABC",
    "BoundStrengtheningAgent": "BOUND_STRENGTHENING_AGENT_V6_SCF_LP",
    "OptimalityCertificationAgent": "SCIP_CASCADE_CERTIFIER_16_K_FEASIBILITY_SCF",
    "OrchestratorAgent": "ORCHESTRATOR_V10_14_PARALLEL_CONTINUITY_KCERT_V4",
}

_missing = [name for name in _expected_exact if globals().get(name) is None]
if globals().get("NeighborhoodOptimizationAgent") is None:
    _missing.append("NeighborhoodOptimizationAgent")

if _missing:
    raise RuntimeError(
        "Classes ainda não definidas antes da verificação: "
        + ", ".join(_missing)
        + ". Execute o notebook em ordem desde o início."
    )

_ave_version = str(getattr(NeighborhoodOptimizationAgent, "VERSION", "UNKNOWN"))
if not _ave_version.startswith("AVE_V2"):
    raise RuntimeError(
        f"NeighborhoodOptimizationAgent incompatível: {_ave_version!r}; esperada família AVE_V2."
    )

for _name, _version in _expected_exact.items():
    _actual = getattr(globals()[_name], "VERSION", None)
    if _actual != _version:
        raise RuntimeError(
            f"{_name}: versão ativa={_actual!r}; versão esperada={_version!r}. "
            "Reinicie a sessão e execute as células em ordem."
        )

print("✅ MULTI13: integração crítica confirmada.")
print(f"   NeighborhoodOptimizationAgent: {_ave_version}")
for _name in _expected_exact:
    print(f"   {_name}: {globals()[_name].VERSION}")


def exibir_estado_orquestrador(orchestrator_instance):
    """
    Exibe o estado atual da memória do OrchestratorAgent,
    indicando quais etapas foram concluídas ao verificar se os atributos da memória estão preenchidos.
    """
    print("\n--- Estado Atual do Orquestrador ---")
    memory = orchestrator_instance.memory.data

    # Define a ordem das chaves para uma exibição mais clara
    ordered_keys = [
        "user_answers",
        "df_clientes",
        "df_veiculos_capacidade",
        "df_veiculos_custo_km",
        "df_janela_tempo_cliente",
        "validation_result",
        "translated_json",
        "problem_classification",
        "optimization_approach",
        "consultant_recommendation",
        "solver_result",
        "explanation"
    ]

    for key in ordered_keys:
        status = "✅ Concluído" if memory.get(key) is not None else "⚪ Pendente"
        if key == "user_answers" and memory.get(key):
            status = f"✅ Concluído ({len(memory[key])} respostas)"
        elif key.startswith("df_") and memory.get(key) is not None:
            df = memory[key]
            status = f"✅ Carregado ({len(df)} linhas, {len(df.columns)} colunas)"
        elif key == "validation_result" and memory.get(key):
            status = f"✅ Concluído: {memory[key][:50]}..."
        elif key == "translated_json" and memory.get(key):
            status = "✅ Concluído (JSON gerado)"
        elif key == "solver_result" and memory.get(key):
            status = f"✅ Concluído: {memory[key].get('status', 'N/A')} (Custo: {memory[key].get('total_cost', 'N/A')})"
        elif key == "explanation" and memory.get(key):
            status = "✅ Concluído (Explicação gerada)"

        print(f"{status} - {key.replace('_', ' ').capitalize()}")

    print("-----------------------------------")






import copy as _pd_copy


import json as _pd_json


import math as _pd_math


import os as _pd_os


import platform as _pd_platform


import sys as _pd_sys


import time as _pd_time


import traceback as _pd_traceback


import threading as _pd_threading


from pathlib import Path as _PDPath


from types import SimpleNamespace as _PDSimpleNamespace


````

#### Persistência, auditoria e processos

Organiza a execução das tarefas, acompanha seu progresso e registra os resultados em arquivos. Também confere os planos de forma independente: uma alternativa só pode ser adotada depois de respeitar as regras do problema.


````python
def _pd_clean(value):
    if hasattr(value, "to_dict") and hasattr(value, "columns"):
        return {"columns": list(map(str, value.columns)), "records": _pd_clean(value.to_dict("records"))}
    if isinstance(value, dict):
        return {str(k): _pd_clean(v) for k, v in value.items()}
    if isinstance(value, (list, tuple, set)):
        return [_pd_clean(v) for v in value]
    if hasattr(value, "item"):
        return _pd_clean(value.item())
    if isinstance(value, float) and not _pd_math.isfinite(value):
        return None
    if value is None or isinstance(value, (str, bool, int, float)):
        return value
    return str(value)


def _pd_write(path, data):
    path = _PDPath(path)
    import threading
    temporary = path.with_name(path.name + f".{_pd_os.getpid()}.{threading.get_ident()}.tmp")
    temporary.write_text(_pd_json.dumps(_pd_clean(data), ensure_ascii=False, indent=2, allow_nan=False), encoding="utf-8")
    # No Windows, um leitor pode manter o destino aberto durante milissegundos.
    for attempt in range(20):
        try:
            _pd_os.replace(temporary,path)
            break
        except PermissionError:
            if attempt == 19:
                raise
            _pd_time.sleep(0.01)



def _pd_read(path):
    try:
        return _pd_json.loads(_PDPath(path).read_text(encoding="utf-8"))
    except (OSError, ValueError):
        return None


def _pd_fmt(value, digits=3):
    if value is None:
        return "não disponível"
    if isinstance(value, float):
        return f"{value:.{digits}f}"
    return str(value)


def _pd_memory_inputs(memory):
    # Copia apenas dados/contrato. Provas antigas não entram silenciosamente no teste.
    keys = ("df_clientes", "df_veiculos_capacidade", "df_veiculos_custo_km",
            "df_janela_tempo_cliente", "translated_json", "user_answers", "objective_context")
    return {k: _pd_copy.deepcopy(memory.data.get(k)) for k in keys}




class _PDTelemetry:
    def __init__(self, folder, stage, interval=2.0):
        self.folder = _PDPath(folder)
        self.start = _pd_time.monotonic()
        self.lock = _pd_threading.RLock()
        self.interval = interval
        self.last_flush = 0.0
        self.data = {"task": stage, "pid": _pd_os.getpid(), "phase": "STARTING",
                     "build_seconds": None, "solve_seconds": None,
                     "nodes": None, "variables": None, "constraints": None,
                     "dual_bound": None, "primal_bound": None, "root_lp_objective": None,
                     "scip_status": None, "error": None}
        self.publish(force=True)

    def publish(self, force=False, **updates):
        with self.lock:
            self.data.update(updates)
            now = _pd_time.monotonic()
            if not force and now-self.last_flush < self.interval:
                return
            self.data.update(elapsed_seconds=now-self.start, updated_at=_pd_time.time())
            _pd_write(self.folder/"progress.json", self.data)
            with (self.folder/"events.jsonl").open("a", encoding="utf-8") as f:
                f.write(_pd_json.dumps(_pd_clean(self.data), ensure_ascii=False, allow_nan=False)+"\n")
            self.last_flush = now




def _pd_problem_fingerprint(problem, objective):
    # Inclui dados operacionais e a função canônica congelada.
    return _vrp_fingerprint({"problem":problem, "objective":objective})


def _pd_numeric_bound(model, value):
    try:
        number = float(value)
        if _pd_math.isfinite(number) and abs(number) < 0.99*float(model.infinity()):
            return number
    except Exception:
        pass
    return None


def _ct_math_worker(stage, inputs, baseline, cfg, folder):
    """Executado num único processo por etapa; nunca escreve na memória oficial."""
    from pyscipopt import Model, quicksum, Eventhdlr, SCIP_EVENTTYPE, SCIP_STAGE
    folder = _PDPath(folder)
    tel = _PDTelemetry(folder, stage)
    model = None
    result = {"task":stage, "completed":False, "scip_status":None,
              "candidate":None, "error":None, "termination_reason":None}
    try:
        mem = _PDSimpleNamespace(data=_pd_copy.deepcopy(inputs))
        provider = ContinuityExactCertificationAgent(mem, verbose_solver=False,
                    time_limit_seconds=int(cfg["TEMPO_SOLVER_OBJETIVO_S"]),
                    scip_threads=int(cfg["THREADS_SCIP"]),
                    scip_memory_mb=int(cfg["MEMORIA_SCIP_MB"]),
                    relative_gap_target=0.0, use_incumbent_cutoff=False,
                    adaptive_stop_enabled=False)
        provider._load_objective_context()
        tel.publish(force=True, phase="READING_OPERATIONAL_CONTRACT")
        problem = provider._build_problem()
        fingerprint = _ct_scope(inputs)
        result["fingerprint"] = fingerprint
        tel.publish(force=True, fingerprint=fingerprint)
        if stage == "LP":
            tel.publish(force=True, phase="LP_SCF_AND_ANALYTICAL_BOUNDS")
            agent = BoundStrengtheningAgent(mem)
            bounds = agent.strengthen(_pd_copy.deepcopy(baseline), {}, budget_seconds=cfg["TEMPO_LP_S"])
            scf = mem.data.get("scf_lp_last_result_v5") or {}
            result.update(completed=True, termination_reason="LP_STAGE_RETURNED",
                          bounds=bounds, scf=scf,
                          canonical_lower_bound=bounds.get("best_lower_bound"))
            tel.publish(force=True, phase="FINISHED", lp_status=scf.get("status"),
                        lp_distance_lower_bound=scf.get("distance_lb"),
                        canonical_lower_bound=result["canonical_lower_bound"])
            _pd_write(folder/"result.json",result)
            return

        build_start = _pd_time.monotonic()
        tel.publish(force=True, phase="PREPROCESSING_MODEL", build_started_at=_pd_time.time())
        original_bounds = provider._analytical_lower_bounds
        original_arcs = provider._build_feasible_arcs
        def measured_bounds(*args, **kwargs):
            tel.publish(force=True, phase="ANALYTICAL_BOUNDS_PRE_BUILD")
            value = original_bounds(*args, **kwargs)
            tel.publish(force=True, phase="BUILDING_MODEL",
                        analytical_lower_bound=value.get("canonical_objective_lb"),
                        trip_slots_estimated=value.get("trip_slots_safe_upper_bound"))
            return value
        def measured_arcs(*args, **kwargs):
            tel.publish(force=True, phase="PREPROCESSING_ARCS")
            value = original_arcs(*args, **kwargs)
            tel.publish(force=True, phase="BUILDING_MODEL", arcs=len(value[0]))
            return value
        provider._analytical_lower_bounds = measured_bounds
        provider._build_feasible_arcs = measured_arcs

        class CountingModel(Model):
            def __init__(self, *args, **kwargs):
                super().__init__(*args, **kwargs)
                self.pd_vars = 0
                self.pd_cons = 0
                self.setLogfile(str(folder/"scip.log"))
                tel.publish(force=True, phase="BUILDING_VARIABLES", variables=0, constraints=0)
            def addVar(self, *args, **kwargs):
                value = super().addVar(*args, **kwargs)
                self.pd_vars += 1
                if self.pd_vars % 1000 == 0:
                    tel.publish(phase="BUILDING_VARIABLES", variables=self.pd_vars,
                                build_seconds=_pd_time.monotonic()-build_start)
                return value
            def addCons(self, *args, **kwargs):
                value = super().addCons(*args, **kwargs)
                self.pd_cons += 1
                if self.pd_cons % 500 == 0:
                    tel.publish(phase="BUILDING_CONSTRAINTS", variables=self.pd_vars,
                                constraints=self.pd_cons, build_seconds=_pd_time.monotonic()-build_start)
                return value

        provider._import_scip = lambda: (CountingModel, quicksum)
        # Chamada DIRETA: não passa pelo complexity guard, nem pela cascata curta.
        # Não usa cutoff econômico. O modelo K inclui TODOS os veículos disponíveis.
        feasibility = stage == "K_EXATO"
        model, handles = provider._build_scip_model(
            incumbent_ub=None, feasibility_only=feasibility,
            max_active_vehicles=int(cfg["K"]) if feasibility else None)
        build_seconds = _pd_time.monotonic()-build_start
        result.update(build_seconds=build_seconds, model_stats=_pd_copy.deepcopy(provider._model_stats),
                      variables=int(model.getNVars()), constraints=int(model.getNConss()))
        tel.publish(force=True, phase="MODEL_READY", build_seconds=build_seconds,
                    variables=result["variables"], constraints=result["constraints"])
        if cfg["EXPORTAR_MODELO_CIP"]:
            tel.publish(force=True, phase="EXPORTING_MODEL")
            model.writeProblem(str(folder/"model.cip"))

        limit = float(cfg["TEMPO_SOLVER_K_S"] if feasibility else cfg["TEMPO_SOLVER_OBJETIVO_S"])
        # O clock de solução SCIP é separado da montagem Python.
        model.setParam("limits/time",limit)
        model.setParam("limits/gap",0.0)
        model.setParam("limits/absgap",0.0)
        if feasibility:
            model.setParam("limits/solutions",1)
        else:
            result["mip_start_accepted"] = provider._try_add_incumbent_mip_start(model,handles,baseline)
            result["mip_start_report"] = dict(getattr(provider,"_mip_start_report",{"submitted":False,"validated":False,"accepted":False,"reason":"NOT_ATTEMPTED"}))
            result["mip_start_submitted"] = result["mip_start_report"]["submitted"]
        model.hideOutput(True)  # SCIP continua escrevendo no arquivo de log.
        model.setParam("display/verblevel",4)
        solve_start = None
        last_checkpoint = 0.0
        latest_checkpoint = False

        def save_candidate():
            nonlocal last_checkpoint, latest_checkpoint
            if model.getBestSol() is None:
                return
            candidate = provider._extract_solution(model,handles)
            if candidate:
                _pd_write(folder/"candidate.json",candidate)
                last_checkpoint = _pd_time.monotonic()
                latest_checkpoint = True

        def snapshot(event_type=None, force=False):
            try:
                stage_code = model.getStage()
                updates = {"scip_stage":str(model.getStageName()), "scip_status":str(model.getStatus())}
                if stage_code in (SCIP_STAGE.INITPRESOLVE,SCIP_STAGE.PRESOLVING,SCIP_STAGE.EXITPRESOLVE):
                    updates["phase"] = "SCIP_PRESOLVING"
                elif stage_code in (SCIP_STAGE.INITSOLVE,SCIP_STAGE.SOLVING,SCIP_STAGE.SOLVED):
                    nodes = int(model.getNNodes())
                    updates.update(nodes=nodes,
                        phase="SCIP_SOLVED" if stage_code == SCIP_STAGE.SOLVED else (
                            "SCIP_ROOT_LP_AND_CUTS" if nodes <= 1 else "SCIP_BRANCH_AND_BOUND"),
                        dual_bound=_pd_numeric_bound(model,model.getDualbound()),
                        primal_bound=_pd_numeric_bound(model,model.getPrimalbound()))
                    if event_type in (SCIP_EVENTTYPE.LPSOLVED,SCIP_EVENTTYPE.FIRSTLPSOLVED) and nodes <= 1:
                        updates["root_lp_objective"] = _pd_numeric_bound(model,model.getLPObjVal())
                        from pyscipopt import SCIP_LPSOLSTAT
                        if model.getLPSolstat() == SCIP_LPSOLSTAT.OPTIMAL:
                            updates["root_lp_completed"] = True
                            updates["root_lp_first_completed_seconds"] = tel.data.get("root_lp_first_completed_seconds", _pd_time.monotonic()-solve_start)
                if solve_start is not None:
                    updates["solve_seconds"] = _pd_time.monotonic()-solve_start
                tel.publish(force=force, **updates)
            except Exception as exc:
                tel.publish(force=True, telemetry_warning=str(exc))

        class ProgressEvents(Eventhdlr):
            def eventinit(self):
                self.mask = 0
                for name in ("PRESOLVEROUND","NODEFOCUSED","FIRSTLPSOLVED","LPSOLVED",
                             "NODESOLVED","BESTSOLFOUND","DUALBOUNDIMPROVED"):
                    self.mask |= getattr(SCIP_EVENTTYPE,name)
                self.model.catchEvent(self.mask,self)
            def eventexit(self):
                self.model.dropEvent(self.mask,self)
            def eventexec(self,event):
                event_type = event.getType()
                now = _pd_time.monotonic()
                important = (event_type == SCIP_EVENTTYPE.BESTSOLFOUND or
                    event_type in (SCIP_EVENTTYPE.FIRSTLPSOLVED,SCIP_EVENTTYPE.LPSOLVED)
                    and not tel.data.get("root_lp_completed",False))
                phase_change = (
                    event_type == SCIP_EVENTTYPE.PRESOLVEROUND and tel.data.get("phase") != "SCIP_PRESOLVING"
                    or event_type == SCIP_EVENTTYPE.NODEFOCUSED and tel.data.get("phase") not in (
                        "SCIP_ROOT_LP_AND_CUTS","SCIP_BRANCH_AND_BOUND")
                    or event_type == SCIP_EVENTTYPE.FIRSTLPSOLVED and tel.data.get("root_lp_objective") is None)
                if important or phase_change or now-tel.last_flush >= tel.interval:
                    snapshot(event_type, force=important or phase_change)
                if important and (not latest_checkpoint or now-last_checkpoint >= 10):
                    try:
                        save_candidate()
                    except Exception as exc:
                        tel.publish(force=True, checkpoint_warning=str(exc))

        model.includeEventhdlr(ProgressEvents(),"dedicated_progress","Progresso e checkpoints sem parada por estagnação")
        solve_start = _pd_time.monotonic()
        tel.publish(force=True, phase="SCIP_STARTING_TRANSFORM_AND_PRESOLVE",
                    optimize_started_at=_pd_time.time(), solve_seconds=0.0,
                    solver_time_limit_seconds=limit)
        model.optimize()
        snapshot(force=True)
        save_candidate()
        result.update(completed=True, scip_status=str(model.getStatus()),
                      termination_reason="SCIP_RETURNED_"+str(model.getStatus()).upper(),
                      nodes=int(model.getNNodes()),
                      solve_seconds=_pd_time.monotonic()-solve_start,
                      dual_bound=_pd_numeric_bound(model,model.getDualbound()),
                      primal_bound=_pd_numeric_bound(model,model.getPrimalbound()),
                      has_model_witness=model.getBestSol() is not None,
                      root_lp_completed=bool(tel.data.get("root_lp_completed",False)),
                      root_lp_first_completed_seconds=tel.data.get("root_lp_first_completed_seconds"),
                      lp_iterations=int(model.getNLPIterations()))
        # Resultado bruto é persistido antes da auditoria no pai.
        _pd_write(folder/"result.json",result)
        tel.publish(force=True, phase="FINISHED", termination_reason=result["termination_reason"])
    except BaseException as exc:
        result.update(error=f"{type(exc).__name__}: {exc}", termination_reason="WORKER_EXCEPTION",
                      traceback=_pd_traceback.format_exc())
        _pd_write(folder/"result.json",result)
        tel.publish(force=True, phase="FAILED", error=result["error"], termination_reason="WORKER_EXCEPTION")
    finally:
        # O término do processo libera o modelo; não serializa objetos SCIP.
        if model is not None:
            try:
                model.setLogfile(None)
                model.freeProb()
            except Exception:
                pass




def _ct_run_process(stage, inputs, baseline, cfg, folder):
    import subprocess, cloudpickle
    start = _pd_time.monotonic()
    run_started_wall = _pd_time.time()
    folder = _PDPath(folder)
    folder.mkdir(parents=True,exist_ok=True)
    # Python novo: não herda locks/threads nativos da sessão Colab (sem fork).
    job = folder/"_internal_job.pkl"
    runner = folder/"_internal_worker.py"
    with job.open("wb") as f:
        cloudpickle.dump((_ct_worker,(stage,inputs,baseline,cfg,str(folder))),f)
    runner.write_text("import sys, cloudpickle\nwith open(sys.argv[1], 'rb') as f:\n    fn, args = cloudpickle.load(f)\nfn(*args)\n",encoding="utf-8")
    stdout = (folder/"worker_stdout.log").open("w",encoding="utf-8")
    stderr = (folder/"worker_stderr.log").open("w",encoding="utf-8")
    env = _pd_os.environ.copy()
    env["PYTHONIOENCODING"] = "utf-8"
    env["PYTHONUTF8"] = "1"
    # Também limita pools BLAS/HiGHS: a prova não compete com outros agentes.
    for key in ("OMP_NUM_THREADS","OPENBLAS_NUM_THREADS","MKL_NUM_THREADS","NUMEXPR_NUM_THREADS"):
        env[key] = str(int(cfg["THREADS_SCIP"]))
    process = subprocess.Popen([_pd_sys.executable,"-u",str(runner),str(job)],
        stdout=stdout,stderr=stderr,env=env,
        creationflags=getattr(subprocess,"CREATE_NO_WINDOW",0) if _pd_os.name=="nt" else 0)
    resource_budget=_v534_memory_budget(inputs)
    peak_rss_mb=0.
    try:
        import psutil
        memory_process=psutil.Process(process.pid)
    except ImportError:memory_process=None
    last_print = -float("inf")
    last_phase = None
    last_user_update = 0.
    reason = None
    try:
        while process.poll() is None:
            if memory_process is not None:
                try:
                    rss=memory_process.memory_info().rss/(1024**2)
                    peak_rss_mb=max(peak_rss_mb,rss)
                    if rss>resource_budget["rss_mb"]:
                        reason="WORKER_RSS_MEMORY_LIMIT"
                        break
                except psutil.NoSuchProcess:pass
            progress = _pd_read(folder/"progress.json") or {}
            if float(progress.get("updated_at",0))<run_started_wall:progress={}
            if stage=="LB_COMPACTO":
                live=_pd_read(folder/"lb_live.json") or {}
                if live.get("fingerprint")==_ct_scope(inputs) and float(live.get("updated_at",0))>=run_started_wall:progress.update(live)
            elapsed = _pd_time.monotonic()-start
            now = _pd_time.time()
            optimize_at = progress.get("optimize_started_at")
            if stage == "LP":
                expired = elapsed > cfg["TEMPO_LP_S"]+cfg["MARGEM_FINALIZACAO_S"]
                timeout_reason = "LP_WALL_DEADLINE"
            elif optimize_at is None:
                expired = elapsed > cfg["TEMPO_CONSTRUCAO_S"]
                timeout_reason = "BUILD_WALL_DEADLINE"
            else:
                solver_limit = cfg["TEMPO_SOLVER_K_S"] if stage == "K_EXATO" else cfg["TEMPO_SOLVER_OBJETIVO_S"]
                expired = now-float(optimize_at) > solver_limit+cfg["MARGEM_FINALIZACAO_S"]
                timeout_reason = "SOLVER_WALL_DEADLINE"
            if elapsed >= float(cfg["WALL_STAGE_SECONDS"]):
                expired=True
                timeout_reason="CONTINUITY_STAGE_WALL_DEADLINE"
            phase = progress.get("phase","STARTING")
            if elapsed-last_user_update>=60 and not globals().get('MOSTRAR_DIAGNOSTICO_TECNICO',False):
                _ct_technical_print(folder, _ct_stage_name(stage)+': em andamento há '+_ct_time_words(elapsed)+'. O resultado será conferido ao terminar.')
                last_user_update=elapsed
            if elapsed-last_print >= cfg["INTERVALO_IMPRESSAO_S"] or phase != last_phase:
                observation_age = max(0.,now-progress["updated_at"]) if progress.get("updated_at") else None
                _ct_technical_print(folder, f"[{stage}] {elapsed:.0f}s | última fase={phase} | atualização há {_pd_fmt(observation_age,1)}s")
                _ct_technical_print(folder, f"    construção={_pd_fmt(progress.get('build_seconds'),1)}s | nós={_pd_fmt(progress.get('nodes'))}"
                      f" | vars={_pd_fmt(progress.get('variables'))} | restrições={_pd_fmt(progress.get('constraints'))}")
                if observation_age is not None and observation_age>20:
                    _ct_technical_print(folder, "    Processo ainda ativo; telemetria antiga não comprova estagnação do solver.")
                if stage != "K_EXATO":
                    _ct_technical_print(folder, f"    LB SCIP={_pd_fmt(progress.get('dual_bound'),9)}"
                          f" | UB SCIP={_pd_fmt(progress.get('primal_bound'),9)}")
                if progress.get("error"):
                    _ct_technical_print(folder, "    erro:",progress["error"])
                last_print,last_phase = elapsed,phase
            if expired:
                reason = timeout_reason
                break
            _pd_time.sleep(0.2)
    except KeyboardInterrupt:
        reason = "USER_INTERRUPTED"
    finally:
        if process.poll() is None:
            process.terminate()
        try:
            process.wait(timeout=1)
        except subprocess.TimeoutExpired:
            process.kill()
            process.wait(timeout=1)
        stdout.close()
        stderr.close()
    progress = _pd_read(folder/"progress.json") or {}
    result = _pd_read(folder/"result.json")
    if result is None:
        result = {"task":stage,"completed":False,"scip_status":None,
                  "termination_reason":reason or "WORKER_EXIT_WITHOUT_RESULT",
                  "error":None,"nodes":progress.get("nodes"),
                  "dual_bound":progress.get("dual_bound"),
                  "primal_bound":progress.get("primal_bound"),
                  "fingerprint":progress.get("fingerprint")}
    if stage=='LB_COMPACTO' and not result.get('global_bound_usable'):
        available=[]
        for checkpoint in (folder/'bound_checkpoint.json',folder/'integer_completion'/'bound_checkpoint.json'):
            item=_pd_read(checkpoint) or {}
            if item.get('fingerprint')==_ct_scope(inputs) and item.get('global_bound_usable') and checkpoint.stat().st_mtime>=run_started_wall:
                value=item.get('canonical_lower_bound')
                if isinstance(value,(int,float)) and _pd_math.isfinite(value):available.append(item)
        saved_bound=max(available,key=lambda r:r['canonical_lower_bound']) if available else {}
        if saved_bound:
            result.update(saved_bound);result['bound_checkpoint_recovered']=True
    result=_v511_process_outcome(result,process.returncode,reason)
    result.update(peak_worker_rss_mb=peak_rss_mb,worker_memory_budget=resource_budget)
    if result.get('bound_checkpoint_recovered'):
        result['process_outcome']='LIMITE_VALIDO_RECUPERADO_SEM_CONCLUSAO_DO_PROCESSO'
    result.update(worker_exitcode=process.returncode, parent_wall_seconds=_pd_time.monotonic()-start,
                  last_progress=progress, parent_stop_reason=reason)
    _pd_write(folder/"collected.json",result)
    _ct_technical_print(folder, f"[{stage}] {result['process_outcome']} | resultado={result.get('worker_result_reason')} | parada={reason or 'nenhuma'} | exitcode={process.returncode}")
    return result





def _pd_audit(inputs, candidate):
    mem = _PDSimpleNamespace(data=_pd_copy.deepcopy(inputs))
    return IndependentFeasibilityVerifier(mem).verify(_pd_copy.deepcopy(candidate),store_key="dedicated_audit")


# Continuity support only. No automatic standalone proof is started here.
def _ct_scope(inputs):
    keys=('df_clientes','df_veiculos_capacidade','df_veiculos_custo_km',
          'df_janela_tempo_cliente','translated_json','user_answers','objective_context')
    return _vrp_fingerprint({k:inputs.get(k) for k in keys})


def _base_ct_worker_1(stage,inputs,baseline,cfg,folder):
    if stage!='MELHORIA':
        return _ct_math_worker(stage,inputs,baseline,cfg,folder)
    folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    mem=_PDSimpleNamespace(data=_pd_copy.deepcopy(inputs))
    fingerprint=_ct_scope(inputs);started=_pd_time.monotonic()
    result=dict(task=stage,completed=False,fingerprint=fingerprint)
    best=_pd_copy.deepcopy(baseline);best_score=float(best['canonical_objective_score'])
    _pd_write(folder/'progress.json',dict(phase='CONTINUITY_AVE',
        updated_at=_pd_time.time(),fingerprint=fingerprint,optimize_started_at=_pd_time.time()))
    try:
        agent=ContinuityOptimizationAgent(mem)
        ave=NeighborhoodOptimizationAgent(mem)
        plan=agent._objective_improvement_plan(best)
        verifier=IndependentFeasibilityVerifier(mem)
        def audit(candidate,label=None):
            nonlocal best,best_score
            report=verifier.verify(candidate,store_key='independent_feasibility_report_auditor')
            if report.get('feasible'):
                normalized=report['normalized_solution'];score=float(report['recomputed']['canonical_objective_score'])
                if score<best_score-1e-10:
                    best=_pd_copy.deepcopy(normalized);best_score=score
                    _pd_write(folder/'candidate.json',best)
                    _pd_write(folder/'progress.json',dict(phase='CONTINUITY_IMPROVED',
                        updated_at=_pd_time.time(),fingerprint=fingerprint,primal_bound=best_score,
                        optimize_started_at=_pd_time.time()-(_pd_time.monotonic()-started)))
            return dict(feasible=bool(report.get('feasible')),solucao_aprovada=bool(report.get('feasible')))
        quota=max(0.,float(cfg['WALL_STAGE_SECONDS'])-3)
        ave.optimize(best,plan,agent._score,audit,budget_seconds=quota*.65,seed=int(cfg['CONTINUITY_SEED']))
        remaining=quota-(_pd_time.monotonic()-started)
        if remaining>=20 and inputs.get('problem_classification'):
            ga=GeneticAlgorithmSolverAgent(mem)
            agent._run_primal_track(ga,audit,inputs['problem_classification'],inputs['translated_json'],
                best,remaining,int(cfg['CONTINUITY_SEED'])+1,plan)
        _pd_write(folder/'candidate.json',best)
        _pd_write(folder/'ave_state.json',mem.data.get('ave_v2_persistent_state',{}))
        result.update(completed=True,termination_reason='PRIMAL_FINISHED',best_ub=best_score)
    except BaseException as exc:
        result.update(termination_reason='PRIMAL_EXCEPTION',error=str(exc),traceback=_pd_traceback.format_exc())
    finally:
        result['elapsed_seconds']=_pd_time.monotonic()-started
        _pd_write(folder/'result.json',result)
_ct_worker = _base_ct_worker_1


def executar_continuidade_integrada(orchestrator,budget_seconds=300,baseline_solution=None,
                                     baseline_certificate=None,download=False,runner=None):
    with AvisoExecucaoV534("Buscando melhorias e conferindo os resultados"):
        import uuid
        budget_seconds=float(budget_seconds)
        if not _pd_math.isfinite(budget_seconds) or budget_seconds<30:raise ValueError('Use pelo menos 30 segundos de continuidade.')
        started=_pd_time.monotonic();deadline=started+budget_seconds
        mem=orchestrator.memory.data;inputs=_pd_memory_inputs(orchestrator.memory);scope=_ct_scope(inputs)
        previous=mem.get('continuity_support_state_v53') or {}
        if previous.get('fingerprint')!=scope:previous={}
        reports=[]
        for candidate in (baseline_solution,previous.get('best_solution'),mem.get('best_incumbent'),mem.get('solver_result')):
            if isinstance(candidate,dict):
                audit=_pd_audit(inputs,candidate)
                if audit.get('feasible'):reports.append(audit)
        if not reports:raise ValueError('A continuidade exige uma solução completa e auditada.')
        audit=min(reports,key=lambda r:r['recomputed']['canonical_objective_score']);best=audit['normalized_solution'];ub=float(audit['recomputed']['canonical_objective_score'])
        baseline_summary=_pd_copy.deepcopy(best['summary']);beforeub=ub;lb=0.;warnings=[];sources=[]
        cert=baseline_certificate or mem.get('continuity_certificate') or {}
        candidates=[]
        if previous:candidates.append(('PREVIOUS_SCOPED_BOUND',previous.get('lower_bound')))
        if cert.get('problem_fingerprint_v53')==scope and cert.get('lower_bound_certified'):candidates.append(('INITIAL_SCOPED_BOUND',cert.get('lower_bound')))
        for source,value in candidates:
            if value is not None and _pd_math.isfinite(float(value)) and 0<=float(value)<=ub+1e-7:
                lb=max(lb,float(value));sources.append(dict(source=source,value=float(value)))
        beforelb=lb;state=_pd_copy.deepcopy(previous);search=state.setdefault('search_state_v58',{})
        search.setdefault('feasible_pool',[])
        search['feasible_pool'].append(_pd_copy.deepcopy(best))
        inputs['portfolio_mode_v512']=True
        inputs.update(mem.get('continuity_options_v514',{}))
        user_request=mem.pop('continuity_user_request',None) or dict(mode='automatic',timed_stages=[],fingerprint=scope)
        if user_request.get('fingerprint')!=scope:raise ValueError('A orientação pertence a outro problema.')
        if user_request.get('mode') not in ('automatic','more_time','alternative','complete_global'):raise ValueError('Orientação de continuidade inválida.')
        inputs['user_continuity_mode']=user_request['mode']
        inputs['user_timed_stages']=list(user_request.get('timed_stages',[]))
        if user_request['mode']=='complete_global':
            decision=CompleteModelPolicy().plan(_v58_provider(inputs).problem,search,inputs,budget_seconds)
            if not decision['run']:raise ValueError(CompleteModelPolicy().explain(decision))
            inputs['complete_model_plan']=decision
            print('REVISÃO DO PLANO INTEIRO: até '+_ct_time_words(decision['cap_seconds'])+' de busca dentro de '+_ct_time_words(budget_seconds)+' autorizados. '+('Existe um teto de tempo explícito na configuração.' if decision['configured_solver_ceiling'] is not None else 'Não há corte automático de 33 segundos nem de dez minutos.'))


        inputs.pop('exclude_region_keys_v524',None)
        inputs['lb_policy_enabled_v523']=True
        lb_policy=ContinuityBoundPolicyV523()
        _v512_portfolio_add(inputs,search,[best,previous.get('k_witness')]+search['feasible_pool'])
        stages={};shares=_v58_allocation(previous);target=max(0,int(best['summary']['num_vehicles_used'])-1);kstatus='UNKNOWN'
        if previous.get('target_k')==target and previous.get('k_status')=='PROVEN_INFEASIBLE':kstatus='PROVEN_INFEASIBLE'
        base=_PDPath('/content') if _PDPath('/content').is_dir() else _PDPath.cwd()
        folder=base/('continuidade_v524_'+_pd_time.strftime('%Y%m%d_%H%M%S')+'_'+uuid.uuid4().hex[:6]);folder.mkdir(parents=True)
        _pd_write(folder/'manifest.json',dict(fingerprint=scope,input_data=inputs,budget_seconds=budget_seconds,shares=shares,policy='V510_BOUNDED_INITIAL_REFINEMENT; ADAPTIVE_CONTINUITY; GLOBAL_LB'))
        _pd_write(folder/'baseline_audit.json',audit)
        cfg=dict(TEMPO_LP_S=1,TEMPO_CONSTRUCAO_S=1,TEMPO_SOLVER_OBJETIVO_S=1,TEMPO_SOLVER_K_S=1,
            MARGEM_FINALIZACAO_S=1,INTERVALO_IMPRESSAO_S=15,EXPORTAR_MODELO_CIP=False,THREADS_SCIP=1)
        run=runner or _ct_run_process;fraction=0.;interrupted=False
        # Short rounds execute useful batches rather than fragmenting seconds across five workers.
        if budget_seconds<120:
            shares=({'MELHORIA':.1,'EXATO_LOCAL':.5,'ELIMINAR_VEICULO':.4} if int(previous.get('round',0))%2==0 else {'LB_COMPACTO':.5,'GERACAO_COLUNAS':.5})
        if not inputs.get('enable_columns_v514',False):
            shares.pop('GERACAO_COLUNAS',None)
            stages['GERACAO_COLUNAS']=dict(termination_reason='DISABLED_BY_DEFAULT',process_outcome='DESATIVADO_POR_POLITICA',parent_wall_seconds=0.)
        if inputs['lb_policy_enabled_v523'] and budget_seconds<120:
            shares={'LB_COMPACTO':.35,'MELHORIA':.2,'EXATO_LOCAL':.3,'ELIMINAR_VEICULO':.15}
        lb_decision=(lb_policy.plan(inputs,search,budget_seconds) if inputs['lb_policy_enabled_v523'] else dict(run=True,reason='LEGACY_FIRST_BOUND_PASS'))
        if not lb_decision['run']:
            shares.pop('LB_COMPACTO',None)
            stages['LB_COMPACTO']=dict(termination_reason=lb_decision['reason'],process_outcome='DISPENSADO_POR_MEMORIA',parent_wall_seconds=0.,requested_wall_seconds_v523=0.,budget_policy=lb_decision)
            if not shares:shares={'MELHORIA':.2,'EXATO_LOCAL':.6,'ELIMINAR_VEICULO':.2}
        elif inputs['lb_policy_enabled_v523'] and 'LB_COMPACTO' not in shares:
            # Short rounds must not postpone every new LP pilot because of round parity.
            shares={**shares,'LB_COMPACTO':.35}
        if lb_decision['run'] and inputs['lb_policy_enabled_v523']:
            shares={'LB_COMPACTO':shares['LB_COMPACTO'],**{k:v for k,v in shares.items() if k!='LB_COMPACTO'}}
        if int(best['summary']['num_vehicles_used'])<=1 or _v58_provider(inputs).problem['must_use_all_vehicles']:
            shares.pop('ELIMINAR_VEICULO',None)
        if kstatus=='PROVEN_INFEASIBLE':
            shares={name:value for name,value in shares.items() if name!='ELIMINAR_VEICULO'}
            stages['ELIMINAR_VEICULO']=dict(termination_reason='PREVIOUS_SCOPED_PROOF_REUSED',process_outcome='DISPENSADO_POR_PROVA',parent_wall_seconds=0.)
        shares=_ct_guided_shares(shares,inputs)
        if user_request.get('mode')!='complete_global' and 'eligible_stages' in user_request:
            shares={k:v for k,v in shares.items() if k in user_request['eligible_stages']}
        if int(best['summary']['num_vehicles_used'])<=1:kstatus='NOT_APPLICABLE'
        pending_share=sum(shares.values())
        task_policy=ContinuityTaskPolicyV524();stop_decision={'reason':'INITIAL_TASKS_COMPLETED'}
        def tasks():
            nonlocal stop_decision
            yield from list(shares.items())
            if user_request['mode']=='complete_global':return
            index=0
            while True:
                stop_decision=task_policy.next_task(inputs,best,search,stages,deadline-_pd_time.monotonic())
                if not stop_decision['run']:
                    if user_request['mode']=='automatic' and 'EXATO_COMPLETO_GLOBAL' not in stages and stop_decision.get('reason')!='LOCAL_PROCESS_INCOMPLETE_REVIEW_EXECUTION':
                        left=max(0.,deadline-_pd_time.monotonic())
                        follow=CompleteModelPolicy().plan(_v58_provider(inputs).problem,search,inputs,left)
                        if follow['run']:
                            inputs['complete_model_plan']=follow
                            print('As buscas parciais terminaram. Ainda restam '+_ct_time_words(left)+' do tempo autorizado; vamos usar esse saldo para revisar o plano inteiro, preservando o que já foi encontrado.')
                            yield ('EXATO_COMPLETO_GLOBAL',0.)
                            stop_decision=dict(run=False,reason='COMPLETE_FALLBACK_FINISHED')
                        else:
                            print('Não iniciaremos a revisão completa com o saldo: '+CompleteModelPolicy().explain(follow))
                    return
                inputs['exclude_region_keys_v524']=stop_decision['exclude_keys']
                index+=1
                yield (f'EXATO_RETORNO_{index}',0.)
        plan=tasks()
        print('Continuidade iniciada. Tempo máximo autorizado: '+_ct_time_words(budget_seconds)+'.')
        try:
            for stage,share in plan:
                if stage.startswith('EXATO_RETORNO'):
                    prior=[v for k,v in stages.items() if k=='EXATO_LOCAL' or k.startswith('EXATO_RETORNO')]
                    if deadline-_pd_time.monotonic()<20:break
                dispatch='EXATO_LOCAL' if stage.startswith('EXATO_RETORNO') else stage
                remaining=max(0.,deadline-_pd_time.monotonic()-2.)*(share/pending_share if share else 1.)
                pending_share-=share
                remaining,policy=_ct_guided_stage_budget(dispatch,remaining,budget_seconds,previous,inputs)
                if stage=='LB_COMPACTO' and inputs['lb_policy_enabled_v523']:
                    decision=lb_policy.plan(inputs,search,max(0.,deadline-_pd_time.monotonic()))
                    if not decision['run']:
                        stages[stage]=dict(termination_reason=decision['reason'],process_outcome='DISPENSADO_POR_MEMORIA',parent_wall_seconds=0.,requested_wall_seconds_v523=0.);continue
                    remaining=min(decision['cap_seconds'],max(0.,deadline-_pd_time.monotonic()-2.));policy=decision
                _ct_technical_print(folder, f"Política {stage}: {policy['reason']}; teto {policy['cap_seconds']:.1f}s.")
                if remaining<(8 if stage.startswith('EXATO_RETORNO') else 3):stages[stage]=dict(termination_reason='NO_REMAINING_BUDGET');continue
                if stage=='ELIMINAR_VEICULO' and int(best['summary']['num_vehicles_used'])<=1:continue
                if stage=='ELIMINAR_VEICULO':
                    newtarget=max(0,int(best['summary']['num_vehicles_used'])-1)
                    if target!=newtarget:kstatus='UNKNOWN'
                    target=newtarget
                    if kstatus=='PROVEN_INFEASIBLE':
                        stages[stage]=dict(termination_reason='PREVIOUS_SCOPED_PROOF_REUSED');continue
                if stage=='LB_COMPACTO' and inputs['lb_policy_enabled_v523']:
                    launch_token=uuid.uuid4().hex
                    lb_policy.claim(inputs,search,launch_token,lb)
                    inputs.update(lb_launch_token_v523=launch_token,lb_audited_ub_v523=ub)
                    state.update(search_state_v58=_pd_copy.deepcopy(search),fingerprint=scope,best_solution=_pd_copy.deepcopy(best),lower_bound=lb)
                    mem['continuity_support_state_v53']=_pd_copy.deepcopy(state)
                    _pd_write(folder/'checkpoint.json',state)
                search_before=_pd_copy.deepcopy(search)
                inputs['search_state_v58']=_pd_copy.deepcopy(search)
                inputs['lb_start_v515']=dict(fingerprint=scope,value=lb,certified=True)
                cfg.update(WALL_STAGE_SECONDS=remaining,TEMPO_CONSTRUCAO_S=remaining,K=target)
                oldub,oldlb=ub,lb;t=_pd_time.monotonic()
                _ct_technical_print(folder, f'{stage}: até {remaining:.1f}s incluindo preparação.')
                _ct_technical_print(folder, _ct_stage_name(stage)+': iniciando.')
                try:r=run(dispatch,inputs,_pd_copy.deepcopy(best),dict(cfg),folder/stage)
                except Exception as exc:r=dict(termination_reason='STAGE_LAUNCH_ERROR',error=str(exc),completed=False)
                stages[stage]=r;r['budget_policy']=policy;r['requested_wall_seconds_v523']=remaining
                scoped=r.get('fingerprint')==scope
                # Recover independently auditable artifacts even if the worker hit its deadline.
                if scoped:
                    newstate=_pd_read(folder/stage/'search_state.json')
                    if isinstance(newstate,dict):search=newstate;state['search_state_v58']=search
                    candidate=_v512_recover_candidates(inputs,folder/stage,search)
                    if candidate:
                        checked=_pd_audit(inputs,candidate)
                        if checked.get('feasible'):
                            score=float(checked['recomputed']['canonical_objective_score'])
                            search.setdefault('feasible_pool',[]).append(checked['normalized_solution'])
                            pool={_v58_signature(_v58_routes(s)):s for s in search['feasible_pool']}
                            search['feasible_pool']=sorted(pool.values(),key=lambda s:s['canonical_objective_score'])[:5]
                            if score<ub:best=checked['normalized_solution'];ub=score;audit=checked
                        else:warnings.append(stage+': candidato rejeitado; melhor solução preservada.')
                    if stage=='ELIMINAR_VEICULO':
                        witness=_pd_read(folder/stage/'k_witness.json')
                        checked=_pd_audit(inputs,witness) if witness else {}
                        if checked.get('feasible') and checked['normalized_solution']['summary']['num_vehicles_used']<=target:
                            kstatus='FEASIBLE_AUDITED';state['k_witness']=checked['normalized_solution']
                            _v512_portfolio_add(inputs,search,[checked['normalized_solution']])
                        elif r.get('k_status')=='NOT_APPLICABLE':kstatus='NOT_APPLICABLE'
                    # Explicit whitelist: local exact dual bounds must never reach the certificate.
                    if stage in ('LB_COMPACTO','GERACAO_COLUNAS','EXATO_COMPLETO_GLOBAL') and r.get('global_bound_usable') and (stage in ('LB_COMPACTO','EXATO_COMPLETO_GLOBAL') or r.get('bound_method')=='DUAL_CORRECTION_ALL_VEHICLES'):
                        value=r.get('canonical_lower_bound')
                        if value is not None and _pd_math.isfinite(float(value)):
                            if 0<=float(value)<=ub+1e-7:
                                lb=max(lb,float(value));sources.append(dict(source=stage,value=float(value)))
                            else:warnings.append('LB compacto rejeitado: valor incompatível com o UB auditado.')
                elapsed=_pd_time.monotonic()-t;r.update(parent_wall_seconds=elapsed,ub_gain=oldub-ub,lb_gain=lb-oldlb)
                if stage=='LB_COMPACTO' and inputs['lb_policy_enabled_v523']:
                    lb_policy.finalize_parent(inputs,search,launch_token,r,oldlb,lb)
                if stage=='EXATO_COMPLETO_GLOBAL':
                    record=search.setdefault('complete_model_v526',{})
                    record.update(status=r.get('status','PROCESS_INTERRUPTED') if scoped and not r.get('error') else 'ERROR',
                        assigned_seconds=record.get('assigned_seconds',remaining),fingerprint=scope,
                        nodes=r.get('nodes') if scoped else None,ub_gain=oldub-ub,lb_gain=lb-oldlb,
                        wall_seconds=elapsed,termination_reason=r.get('termination_reason'),
                        policy=_pd_copy.deepcopy(inputs['complete_model_plan']))
                    search['complete_model_history_v526']=(search.get('complete_model_history_v526',[])+[_pd_copy.deepcopy(record)])[-8:]
                if r.get('error'):warnings.append(stage+': '+r['error'])
                if not scoped:r['activity_unavailable_reason']='SCOPE_NOT_CONFIRMED'
                else:_v511_check_activity(dispatch,r,search_before,search,folder/stage)
                performance=state.setdefault('stage_performance',{}).setdefault(stage,dict(runs=0,recent_gain=0.,total_gain=0.,seconds=0.))
                gain=(lb-oldlb)+(oldub-ub)
                performance.update(runs=performance['runs']+1,recent_gain=gain/max(elapsed,1e-6),total_gain=performance['total_gain']+gain,seconds=performance['seconds']+elapsed)
                samples=performance.setdefault('samples_v511',[])
                samples.append(dict(seconds=elapsed,ub_gain=oldub-ub,lb_gain=lb-oldlb,process_outcome=r.get('process_outcome'),activity=r.get('activity')))
                performance['samples_v511']=samples[-8:]
                state.update(fingerprint=scope,best_solution=_pd_copy.deepcopy(best),lower_bound=lb)
                mem['continuity_support_state_v53']=_pd_copy.deepcopy(state)
                _pd_write(folder/'checkpoint.json',state)
                _ct_technical_print(folder, f'{stage}: ganho UB={oldub-ub:.8f}; ganho LB={lb-oldlb:.8f}; {elapsed:.1f}s.')
                _ct_technical_print(folder, _ct_stage_finished(stage,r))
                if r.get('error') or (r.get('worker_exitcode') not in (None,0) and not r.get('parent_stop_reason')):
                    print('Uma tarefa encontrou uma falha. O melhor plano aprovado foi preservado; consulte o diagnóstico técnico.')
                if r.get('parent_stop_reason')=='USER_INTERRUPTED':interrupted=True;break
        except KeyboardInterrupt:
            interrupted=True
        finally:
            # Reaudit the actually selected solution; report and all memory consumers agree.
            audit=_pd_audit(inputs,best)
            if not audit.get('feasible'):raise RuntimeError('Falha na auditoria final da continuidade.')
            best=audit['normalized_solution'];ub=float(audit['recomputed']['canonical_objective_score'])
            if lb>ub+max(1e-7,abs(ub)*1e-6):raise RuntimeError('LB incompatível com UB; certificação bloqueada.')
            gap=max(0.,(ub-lb)/max(abs(ub),1e-12));proven=abs(ub-lb)<=max(1e-7,abs(ub)*1e-6)
            level='A' if proven else ('B+' if gap<=.01 else ('B' if gap<=.35 else 'C'))
            cert=dict(version='CONTINUITY_V513',method='AUDITED_LOCAL_UB_COMPACT_GLOBAL_LB',problem_fingerprint_v53=scope,
                certification_level=level,upper_bound=ub,lower_bound=lb,lower_bound_certified=True,gap_pct=gap*100,gap_relative=gap,
                proven_optimal=proven,report_allowed=True,certification_finalized=True,post_certification_independent_audit_passed=True,
                k_status=kstatus,target_k=target,bound_sources=sources)
            stats=dict(baseline_summary=baseline_summary,before_ub=beforeub,after_ub=ub,before_lb=beforelb,after_lb=lb,ub_improved=ub<beforeub-1e-10,lb_improved=lb>beforelb+1e-10,
                requested_seconds=budget_seconds,elapsed_seconds=_pd_time.monotonic()-started,stages=stages,shares=shares)
            stats['coordinator_stop_v524']=stop_decision
            stats['user_guidance']=_pd_copy.deepcopy(user_request)
            stats['budget_ledger_v523']=dict(
                requested_seconds=budget_seconds,
                lb_seconds=sum(float(r.get('parent_wall_seconds',0.)) for k,r in stages.items() if k=='LB_COMPACTO'),
                other_agent_seconds=sum(float(r.get('parent_wall_seconds',0.)) for k,r in stages.items() if k!='LB_COMPACTO'),
                unused_seconds=max(0.,budget_seconds-stats['elapsed_seconds']),
                lb_reason=stages.get('LB_COMPACTO',{}).get('termination_reason'))
            result=dict(status='INTERRUPTED' if interrupted else 'COMPLETED',final_solution=best,certificate=cert,statistics=stats,issues=[],warnings=warnings,folder=str(folder))
            state.update(fingerprint=scope,best_solution=_pd_copy.deepcopy(best),lower_bound=lb,round=int(previous.get('round',0))+1,k_status=kstatus,target_k=target,search_state_v58=search)
            state['user_guidance_history']=(state.get('user_guidance_history',[])+[_pd_copy.deepcopy(user_request)])[-20:]
            mem.update(continuity_support_state_v53=state,continuity_certificate=_pd_copy.deepcopy(cert),best_incumbent=_pd_copy.deepcopy(best),
                continuity_final_audit=audit,continuity_statistics=stats,continuity_best_valid_lower_bound=lb,
                continuity_state=dict(status=result['status'],budget_seconds=budget_seconds,elapsed_seconds=stats['elapsed_seconds']))
            stats['next_action_advice']=_ct_available_actions(orchestrator,result)
            _ct_publish_report(result,mem,folder)
            mem['continuity_integrated_result_v53']=_pd_copy.deepcopy(result)
            _pd_write(folder/'summary.json',result);_pd_write(folder/'stages.json',stages);_pd_write(folder/'best_audited_solution.json',best)
            if CONTINUITY_GENERATE_ZIP:
                result['zip_path']=exportar_diagnostico_continuidade(result,download=download)
            _ct_technical_print(folder, f'Continuidade concluída: UB {ub:.10f}; LB {lb:.10f}; GAP {gap*100:.2f}%; certificação {level}.')
            print('Continuidade encerrada. O plano e a garantia foram conferidos; veja o resultado abaixo.')
            print('Relatório completo:',folder/'relatorio_final.html')
        return result





_ct_initial_certify_v51=OrchestratorAgent._run_optimality_certification
def _ct_initial_certify_tagged(self,ans):
    result=_ct_initial_certify_v51(self,ans)
    if isinstance(result,dict) and isinstance(result.get('certificate'),dict):
        result['certificate']['problem_fingerprint_v53']=_ct_scope(_pd_memory_inputs(self.memory))
    return result
OrchestratorAgent._run_optimality_certification=_ct_initial_certify_tagged

def _ct_offer(self,ans,baseline_solution,baseline_certificate,baseline_audit_recommendation):
    if (baseline_certificate or {}).get('certification_level')=='A':return None
    scope=_ct_scope(_pd_memory_inputs(self.memory))
    last=self.memory.data.get('continuity_integrated_result_v53')
    if (last or {}).get('certificate',{}).get('problem_fingerprint_v53')!=scope:last=None
    if not last and not _v591_yes('Deseja dedicar tempo adicional à solução e à certificação? [s/N]: '):return None
    while True:
        request=_ct_choose_next_round(self,last)
        if request is None:return last
        budget=_v591_budget(self)
        if budget is None:return last
        if request['mode']=='complete_global' and budget<request['minimum_full_seconds']:
            print('Orçamento insuficiente para uma tentativa diferente. Escolha mais tempo ou encerre.');continue
        if request['mode']=='complete_global':
            print('Você autorizou até '+_ct_time_words(budget)+'. Vamos revisar o plano inteiro, reservando cerca de 12 segundos para preparar e salvar a resposta. Se houver um teto explícito na configuração, ele será informado antes de começar. Não há garantia de melhoria.')
        else:
            print('Você autorizou até '+_ct_time_words(budget)+'. Esta opção distribui o tempo entre buscas parciais; se elas terminarem antes, explicaremos o motivo e o tempo não usado.')
        request.update(fingerprint=scope,budget_seconds=budget)
        self.memory.data['continuity_user_request']=request
        last=executar_continuidade_integrada(self,budget,baseline_solution if last is None else None,baseline_certificate if last is None else None)
        mostrar_relatorio_continuidade(last)
        if last.get('issues') or not last.get('certificate',{}).get('report_allowed'):return last
        if last['certificate'].get('proven_optimal'):
            print('Ótimo comprovado dentro da tolerância. Mantemos o plano e o certificado obtidos.')
            return last
OrchestratorAgent._offer_and_run_continuity=_ct_offer

# Preserve Report 1 but return the consolidated solution to API consumers too.
_ct_orchestrate_v51=OrchestratorAgent.orchestrate_solution_process
def _ct_orchestrate_consolidated(self):
    result=_ct_orchestrate_v51(self)
    continuation=(result or {}).get('continuity_result') if isinstance(result,dict) else None
    if isinstance(continuation,dict) and not continuation.get('issues') and continuation.get('certificate',{}).get('report_allowed'):
        result['initial_solution']=_pd_copy.deepcopy(result.get('final_solution'))
        result['initial_certificate']=_pd_copy.deepcopy(result.get('certificate'))
        result['initial_report']=result.get('report')
        result['final_solution']=_pd_copy.deepcopy(continuation['final_solution'])
        result['certificate']=_pd_copy.deepcopy(continuation['certificate'])
        result['certification_level']=continuation['certificate']['certification_level']
        result['proven_optimal']=continuation['certificate']['proven_optimal']
        result['report']=continuation['report']
        self.memory.data['initial_final_solution_v53']=result['initial_solution']
        self.memory.data['initial_certificate_v53']=result['initial_certificate']
        self.memory.data['initial_report_v53']=result['initial_report']
        self.memory.data['final_solution']=_pd_copy.deepcopy(result['final_solution'])
        self.memory.data['optimality_certificate']=_pd_copy.deepcopy(result['certificate'])
        self.memory.data['final_report']=result['report']
    return result
OrchestratorAgent.orchestrate_solution_process=_ct_orchestrate_consolidated


def _ct_k2_ga_worker(inputs,baseline,cfg,folder):
    """Restricted two-vehicle primal search; never produces a lower bound."""
    folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    scope=_ct_scope(inputs);started=_pd_time.monotonic()
    result=dict(task='K_AG',fingerprint=scope,completed=False,k_status='UNKNOWN')
    try:
        restricted=_pd_copy.deepcopy(inputs)
        for key in ('df_veiculos_capacidade','df_veiculos_custo_km','df_janela_tempo_cliente'):
            frame=restricted.get(key)
            if frame is not None:frame.columns=[str(c).strip().lower() for c in frame.columns]
        caps=restricted['df_veiculos_capacidade']
        capacity_column=next(c for c in caps.columns if str(c).strip().lower()=='capacidade')
        pair=sorted(range(len(caps)),key=lambda i:(-float(caps.iloc[i][capacity_column]),i))[:int(cfg['K'])]
        if len(pair)!=int(cfg['K']):raise ValueError('K2_AG exige veículos suficientes para o alvo K.')
        restricted['df_veiculos_capacidade']=caps.iloc[pair].reset_index(drop=True).copy()
        costs=restricted.get('df_veiculos_custo_km')
        if costs is not None:
            if len(costs)!=len(caps):raise ValueError('Capacidades e custos precisam ter a mesma correspondência de linhas.')
            restricted['df_veiculos_custo_km']=costs.iloc[pair].reset_index(drop=True).copy()
        result['selected_original_vehicle_ids']=[i+1 for i in pair]
        # Seed only the customer order. Never cache the three-vehicle schedule as feasible.
        seed_order=[]
        for route in baseline.get('routes',[]):
            for client in route.get('path',[])[1:-1]:
                if client not in seed_order:seed_order.append(client)
        mem=_PDSimpleNamespace(data=restricted)
        class RestrictedFleetGA(GeneticAlgorithmSolverAgent):
            def _initialize_population(self,data,population_size,continuity_mode=False):
                population=super()._initialize_population(data,population_size,continuity_mode)
                expected=set(data['client_lookup'])-{data['depot_id']}
                if set(seed_order)==expected and len(seed_order)==len(expected):
                    population[0]=list(seed_order)
                return population
        directive=dict(type='FLEET_REDUCTION',target_max_vehicles=int(cfg['K']))
        mem.data['continuity_search_directive']=directive
        tj=restricted['translated_json']
        if isinstance(tj,str):tj=json.loads(tj)
        tj=_pd_copy.deepcopy(tj);tj['_continuity_directive']=directive
        _pd_write(folder/'progress.json',dict(phase='GA_RESTRICTED_FLEET',fingerprint=scope,
            updated_at=_pd_time.time(),selected_original_vehicle_ids=[i+1 for i in pair]))
        ga=RestrictedFleetGA(mem)
        candidate=ga.solve_problem(restricted.get('problem_classification') or {},tj,
            population_size=48,num_generations=10000,
            time_limit_seconds=max(.1,float(cfg['WALL_STAGE_SECONDS'])-5),
            seed=int(cfg['CONTINUITY_SEED']),feasibility_first=True,continuity_mode=True)
        _pd_write(folder/'ga_result.json',candidate)
        result['ga_status']=candidate.get('status') if isinstance(candidate,dict) else None
        if result['ga_status']=='ERROR':
            result['ga_error']=candidate.get('error') or candidate.get('message')
            raise RuntimeError(result['ga_error'] or 'Falha interna no AG restrito.')
        if isinstance(candidate,dict) and candidate.get('routes'):
            candidate=_pd_copy.deepcopy(candidate)
            for route in candidate['routes']:
                for key in ('vehicle_id','physical_vehicle_id'):
                    if key in route:
                        local_id=int(route[key])
                        if not 1<=local_id<=len(pair):raise ValueError('GA retornou veículo fora da frota restrita.')
                        route[key]=pair[local_id-1]+1
            audit=_pd_audit(inputs,candidate)
            _pd_write(folder/'restricted_candidate_audit.json',audit)
            if audit.get('feasible') and int(audit['normalized_solution']['summary']['num_vehicles_used'])<=int(cfg['K']):
                _pd_write(folder/'candidate.json',audit['normalized_solution'])
                result['k_status']='FEASIBLE_AUDITED'
        result.update(completed=True,termination_reason='K_GA_RETURNED')
    except BaseException as exc:
        result.update(termination_reason='K_GA_EXCEPTION',error=str(exc),traceback=_pd_traceback.format_exc())
    finally:
        result['elapsed_seconds']=_pd_time.monotonic()-started
        _pd_write(folder/'result.json',result)

_ct_worker_v53=_ct_worker
def _base_ct_worker_2(stage,inputs,baseline,cfg,folder):
    if stage=='K_AG':return _ct_k2_ga_worker(inputs,baseline,cfg,folder)
    return _ct_worker_v53(stage,inputs,baseline,cfg,folder)
_ct_worker = _base_ct_worker_2


# Initial certification: audited incumbent + analytical bounds only.
# No MIP, exact-K or LP solver is reachable through this worker.
INITIAL_BOUND_BUDGET_SECONDS = 10.0

def _initial_bounds_worker(inputs,baseline,cfg,folder):
    folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    scope=_ct_scope(inputs);started=_pd_time.monotonic()
    result=dict(task='INITIAL_BOUNDS',fingerprint=scope,completed=False)
    _pd_write(folder/'progress.json',dict(phase='INITIAL_ANALYTICAL_BOUNDS',
        fingerprint=scope,updated_at=_pd_time.time()))
    try:
        memory=_PDSimpleNamespace(data=_pd_copy.deepcopy(inputs))
        provider=OptimalityCertificationAgent(memory,verbose_solver=False)
        provider._load_objective_context();provider._build_problem()
        provider.bound_preprocessing_time_budget_seconds=max(.01,float(cfg['WALL_STAGE_SECONDS'])-2)
        bounds=provider._analytical_lower_bounds()
        result.update(completed=True,termination_reason='ANALYTICAL_BOUNDS_RETURNED',bounds=bounds)
    except BaseException as exc:
        result.update(termination_reason='INITIAL_BOUND_ERROR',error=str(exc))
    finally:
        result['elapsed_seconds']=_pd_time.monotonic()-started
        _pd_write(folder/'result.json',result)

_ct_worker_v54=_ct_worker
def _base_ct_worker_3(stage,inputs,baseline,cfg,folder):
    if stage=='INITIAL_BOUNDS':return _initial_bounds_worker(inputs,baseline,cfg,folder)
    return _ct_worker_v54(stage,inputs,baseline,cfg,folder)
_ct_worker = _base_ct_worker_3

def _initial_audit_and_bounds(self,ans):
    import tempfile
    started=_pd_time.monotonic();mem=self.memory.data
    certifier=self._configure_certifier(ans)
    certifier.intensive_certification=False
    certifier.global_start_time=_pd_time.time()
    certifier._load_objective_context();certifier._build_problem()
    pre=certifier._get_pre_audit_report() # mandatory existing independent audit gate
    incumbent=_pd_copy.deepcopy(pre['normalized_solution'])
    ub=float(pre['recomputed']['canonical_objective_score'])
    inputs=_pd_memory_inputs(self.memory);scope=_ct_scope(inputs)
    budget=float(INITIAL_BOUND_BUDGET_SECONDS)
    if not _pd_math.isfinite(budget) or budget<=0:raise ValueError('Prazo de limites iniciais deve ser positivo.')
    cfg=dict(WALL_STAGE_SECONDS=budget,TEMPO_CONSTRUCAO_S=budget,TEMPO_LP_S=budget,
        TEMPO_SOLVER_OBJETIVO_S=budget,TEMPO_SOLVER_K_S=budget,MARGEM_FINALIZACAO_S=1,
        THREADS_SCIP=1,INTERVALO_IMPRESSAO_S=15)
    print(f'\nCertificação inicial: auditoria + limites analíticos, até {budget:g}s para os limites. Modelo global exato reservado à continuidade; refinamento local inicial tem orçamento separado.')
    with tempfile.TemporaryDirectory(prefix='initial_bounds_') as td:
        r=_ct_run_process('INITIAL_BOUNDS',inputs,incumbent,cfg,_PDPath(td))
    bounds={};lb=None
    if r.get('completed') and r.get('fingerprint')==scope:
        bounds=r.get('bounds') or {}
        value=bounds.get('canonical_objective_lb')
        if bounds.get('canonical_objective_lb_valid') and value is not None and _pd_math.isfinite(float(value)):
            if float(value)<=ub+max(1e-7,abs(ub)*1e-6):lb=min(ub,float(value))
            else:r['bound_rejected']='LB_EXCEEDS_AUDITED_UB'
    # No valid completed bound: keep audited solution, without inventing an LB.
    cls=certifier._classify_certification_level(True,'not_run_initial_policy',ub,lb)
    chosen=dict(layer='INITIAL_AUDIT_AND_ANALYTICAL_BOUNDS',scip_status='not_run_initial_policy',
        upper_bound=ub,lower_bound=lb,solution=incumbent,classification=cls)
    mem['certification_strengthened_bounds']=_pd_copy.deepcopy(bounds)
    mem['certification_bound_catalog']=[]
    result=certifier._provisional_result(chosen,chosen,None,pre,bounds,_pd_time.monotonic()-started)
    certificate=result['certificate']
    certificate.update(method='INITIAL_AUDIT_AND_ANALYTICAL_BOUNDS',formulation='NO_GLOBAL_EXACT_MODEL_IN_INITIAL_CERTIFICATION',
        problem_fingerprint_v53=scope,initial_bound_budget_seconds=budget,
        initial_bound_result=_pd_copy.deepcopy(r),exact_solver_executed=False)
    mem['initial_bound_result_v55']=_pd_copy.deepcopy(r)
    mem['solver_result_pre_certification']=_pd_copy.deepcopy(incumbent)
    mem['certification_result_provisional']=_pd_copy.deepcopy(result)
    mem['certification_result']=_pd_copy.deepcopy(result)
    return result



OrchestratorAgent._run_optimality_certification=_initial_audit_and_bounds

# Legacy finalization can merge unscoped historical bounds. Disable that merge
# for this initial certificate; its bound has already been checked above.
_finalize_v51=OptimalityCertificationAgent.finalize_after_independent_audit
def _finalize_initial_scoped(self,provisional,report):
    if provisional.get('certificate',{}).get('method')!='INITIAL_AUDIT_AND_ANALYTICAL_BOUNDS':
        return _finalize_v51(self,provisional,report)
    mem=self.shared_memory.data
    keys=('continuity_baseline_valid_lower_bound','bound_strengthening_last_result')
    saved={key:mem.pop(key) for key in keys if key in mem}
    try:return _finalize_v51(self,provisional,report)
    finally:mem.update(saved)
OptimalityCertificationAgent.finalize_after_independent_audit=_finalize_initial_scoped


def _v57_schedule(p,routes,seconds=2):
    """Exact LP scheduling for fixed trip orders; infeasibility is local only."""
    import numpy as np
    from scipy.optimize import linprog
    routes=[r for r in routes if len(r['path'])>2]
    V={v['vehicle_id']:v for v in p['vehicles']};bounds=[];slots=[];seen=[]
    used=[r['vehicle_id'] for r in routes]
    if not p['allow_multi_trip'] and len(set(used))!=len(used):return None,'MULTITRIP_FORBIDDEN'
    if p['must_use_all_vehicles'] and set(used)!=set(V):return None,'REQUIRED_FLEET'
    for r in routes:
        customers=r['path'][1:-1];seen+=customers
        if sum(p['client_rows'][i]['demand'] for i in customers)>V[r['vehicle_id']]['capacity']+1e-7:
            return None,'CAPACITY'
        start=len(bounds);bounds.append((p['operation_start'],p['operation_end']))
        times=[]
        for i in customers:
            lo=max(p['operation_start'],p['windows'][i][0]);hi=min(p['operation_end'],p['windows'][i][1])
            if lo>hi:return None,'WINDOW'
            times.append(len(bounds));bounds.append((lo,hi))
        slots.append((start,times))
    if len(set(seen))!=len(seen):return None,'DUPLICATE'
    n=len(bounds);A=[];b=[];objective=np.zeros(n);previous={};work={};constant={}
    def add(terms,rhs):
        row=np.zeros(n)
        for j,v in terms.items():row[j]+=v
        A.append(row);b.append(rhs)
    for r,(start,times) in zip(routes,slots):
        path=r['path'];vid=r['vehicle_id'];svc=p['service_time'];tt=p['travel_time']
        add({start:1,times[0]:-1},-tt[0,path[1]])
        for pos in range(1,len(times)):
            add({times[pos-1]:1,times[pos]:-1},-svc-tt[path[pos],path[pos+1]])
        endconstant=svc+tt[path[-2],0]
        add({times[-1]:1},p['operation_end']-endconstant)
        if vid in previous:
            old,ret=previous[vid];add({old:1,start:-1},-ret)
        previous[vid]=(times[-1],endconstant)
        work.setdefault(vid,np.zeros(n));work[vid][times[-1]]+=1;work[vid][start]-=1
        constant[vid]=constant.get(vid,0)+endconstant
        objective[times[-1]]+=1;objective[start]-=1
    for vid,row in work.items():
        A.append(row);b.append(p['max_time']-constant[vid])
        A.append(-row);b.append(constant[vid]-p['minimum_usage'])
    solved=linprog(objective,A_ub=np.array(A),b_ub=np.array(b),bounds=bounds,method='highs',options={'time_limit':max(.01,seconds)})
    if not solved.success:return None,'LP_'+str(solved.status)
    result=[]
    for r,(start,times) in zip(routes,slots):
        path=r['path'];details=[];departure=float(solved.x[start]);clock=departure;load=0.
        for i,var,prev in zip(path[1:-1],times,path):
            arrival=clock+p['travel_time'][prev,i];service=float(solved.x[var]);clock=service+p['service_time']
            load+=p['client_rows'][i]['demand']
            details.append(dict(node_id=i,arrival_time_minutes=arrival,service_start_time_minutes=service,departure_time_minutes=clock,cumulative_load=load))
        end=clock+p['travel_time'][path[-2],0]
        result.append(dict(vehicle_id=r['vehicle_id'],path=path,trip_id=len([x for x in result if x['vehicle_id']==r['vehicle_id']])+1,
            start_time_minutes=departure,end_time_minutes=end,total_route_time_minutes=end-departure,
            distance_km=sum(p['distance'][a,b] for a,b in zip(path,path[1:])),total_route_load=load,path_details=details))
    return dict(status='SUCCESS',method='DIRECTED_REPAIR_FIXED_TRIP_LP',routes=result),'FEASIBLE_FIXED_ORDERS'

def _v57_repair(p,partial,vehicle_ids,budget=45):
    import time,copy,itertools
    started=time.monotonic();deadline=started+budget;attempts=0;reasons={}
    routes=copy.deepcopy(partial['routes'])
    for r in routes:r['vehicle_id']=vehicle_ids[int(r['vehicle_id'])-1]
    present={i for r in routes for i in r['path'][1:-1]};missing=[i for i in p['customers'] if i not in present]
    info=dict(missing_clients=missing,attempts=0,rejections=reasons,restricted_proof=False)
    if len(missing)!=1:return None,dict(info,reason='REQUIRES_ONE_MISSING_CLIENT')
    client=missing[0]
    def check(candidate):
        nonlocal attempts
        if time.monotonic()>=deadline:return None
        attempts+=1
        answer,reason=_v57_schedule(p,candidate,min(2,max(.01,deadline-time.monotonic())))
        reasons[reason]=reasons.get(reason,0)+1
        return answer
    # Insert at every existing trip position, then reorder affected small trips.
    for t,route in enumerate(routes):
        for pos in range(1,len(route['path'])):
            candidate=copy.deepcopy(routes);candidate[t]['path'].insert(pos,client)
            answer=check(candidate)
            if answer:return answer,dict(info,attempts=attempts,mechanism='INSERT_AND_RESCHEDULE',changed_trip=t)
    # Add a depot trip at each position in the vehicle's chronological chain.
    for vid in vehicle_ids:
        positions=[i for i,r in enumerate(routes) if r['vehicle_id']==vid]
        for at in positions+[positions[-1]+1 if positions else len(routes)]:
            candidate=copy.deepcopy(routes);candidate.insert(at,dict(vehicle_id=vid,path=[0,client,0]))
            answer=check(candidate)
            if answer:return answer,dict(info,attempts=attempts,mechanism='NEW_TRIP_AND_RESCHEDULE')
    order=sorted(range(len(routes)),key=lambda t:min(p['distance'][client,i] for i in routes[t]['path'][1:-1]))
    for t in order:
        seq=routes[t]['path'][1:-1]+[client]
        if len(seq)>7:continue
        for perm in itertools.permutations(seq):
            if time.monotonic()>=deadline:break
            candidate=copy.deepcopy(routes);candidate[t]['path']=[0]+list(perm)+[0]
            answer=check(candidate)
            if answer:return answer,dict(info,attempts=attempts,mechanism='TRIP_REORDER_AND_RESCHEDULE',changed_trip=t)
    return None,dict(info,attempts=attempts,reason='NO_REPAIR_IN_NEIGHBORHOOD',elapsed_seconds=time.monotonic()-started)


"""Continuity primal tasks. All exported candidates require a full independent audit."""
def _v58_provider(inputs):
    agent=ContinuityExactCertificationAgent(_PDSimpleNamespace(data=_pd_copy.deepcopy(inputs)),verbose_solver=False)
    agent._load_objective_context();agent._build_problem()
    return agent

def _v58_routes(solution):
    return [dict(vehicle_id=r['vehicle_id'],path=list(r['path'])) for r in
            sorted(solution['routes'],key=lambda r:(r['vehicle_id'],r.get('start_time_minutes',0))) if len(r['path'])>2]

def _v58_signature(routes):
    return _vrp_fingerprint([(r['vehicle_id'],r['path']) for r in routes if len(r['path'])>2])

def _v58_surrogate(p,coeff,routes):
    V={v['vehicle_id']:v for v in p['vehicles']};used=set();score=0.
    for r in routes:
        if len(r['path'])<3:continue
        vid=r['vehicle_id'];used.add(vid)
        distance=sum(p['distance'][i,j] for i,j in zip(r['path'],r['path'][1:]))
        score+=distance*(coeff.get('DISTANCE',0)+coeff.get('COST',0)*V[vid]['cost_per_km'])
    fixed=sum(v['fixed_cost'] for v in V.values()) if p['fixed_cost_mode']!='WHEN_USED' else sum(V[v]['fixed_cost'] for v in used)
    return score+coeff.get('COST',0)*fixed+coeff.get('VEHICLES',0)*len(used)

def _v58_moves(p,routes,kind):
    import copy
    if kind=='REORDER':
        for a,r in enumerate(routes):
            for i in range(1,len(r['path'])-2):
                for j in range(i+1,len(r['path'])-1):
                    trial=copy.deepcopy(routes);trial[a]['path'][i:j+1]=reversed(trial[a]['path'][i:j+1]);yield trial
    elif kind=='RELOCATE':
        for a,r in enumerate(routes):
            for i in range(1,len(r['path'])-1):
                for b,s in enumerate(routes):
                    if a==b:continue
                    for j in range(1,len(s['path'])):
                        trial=copy.deepcopy(routes);client=trial[a]['path'].pop(i);trial[b]['path'].insert(j,client)
                        yield [x for x in trial if len(x['path'])>2]
    elif kind=='SWAP':
        for a,r in enumerate(routes):
            for b in range(a+1,len(routes)):
                for i in range(1,len(r['path'])-1):
                    for j in range(1,len(routes[b]['path'])-1):
                        trial=copy.deepcopy(routes);trial[a]['path'][i],trial[b]['path'][j]=trial[b]['path'][j],trial[a]['path'][i];yield trial
    elif kind=='MERGE':
        for a,r in enumerate(routes):
            for b,s in enumerate(routes):
                if a==b:continue
                for reverse in (False,True):
                    trial=copy.deepcopy(routes);seq=s['path'][1:-1]
                    trial[a]['path']=r['path'][:-1]+(list(reversed(seq)) if reverse else seq)+[0]
                    trial.pop(b);yield trial
    elif kind=='VEHICLE':
        used=sorted({r['vehicle_id'] for r in routes})
        for old in used:
            for v in p['vehicles']:
                new=v['vehicle_id']
                if old==new:continue
                trial=copy.deepcopy(routes)
                # Swap entire vehicle schedules if both are active.
                for r in trial:
                    if r['vehicle_id']==old:r['vehicle_id']=new
                    elif r['vehicle_id']==new:r['vehicle_id']=old
                yield trial

def _v58_local_search(inputs,best,seconds,folder,state):
    import heapq,time
    started=time.monotonic();deadline=started+seconds;provider=_v58_provider(inputs)
    p=provider.problem;coeff=provider.objective_context['coefficients'];ub=best['canonical_objective_score']
    seen=set(state.get('tested_moves',[]));history=state.setdefault('operators',{})
    kinds=['REORDER','RELOCATE','SWAP','MERGE','VEHICLE'];round_index=int(state.get('scan_round',0))
    kinds=kinds[round_index%5:]+kinds[:round_index%5];attempts=0;improvements=0
    while time.monotonic()<deadline-.5:
        improved=False
        for kind in kinds:
            opstart=time.monotonic();record=history.setdefault(kind,dict(attempts=0,gain=0.,seconds=0.))
            routes=_v58_routes(best);base=_v58_surrogate(p,coeff,routes);batch=[];counter=0
            for trial in _v58_moves(p,routes,kind):
                if time.monotonic()>deadline-.5:break
                sig=_v58_signature(trial)
                if sig in seen:continue
                value=_v58_surrogate(p,coeff,trial)
                if not coeff.get('TIME',0) and value>=base-1e-10:continue
                # Keep a bounded set of promising changes, without exhausting memory.
                counter+=1;item=(-value,counter,sig,trial)
                if len(batch)<250:heapq.heappush(batch,item)
                elif value < -batch[0][0]:heapq.heapreplace(batch,item)
            for neg,_,sig,trial in sorted(batch,reverse=True):
                if time.monotonic()>deadline-.3:break
                candidate,reason=_v57_schedule(p,trial,min(.7,deadline-time.monotonic()))
                if reason!='LP_1':seen.add(sig)
                attempts+=1;record['attempts']+=1
                if candidate:
                    audit=_pd_audit(inputs,candidate)
                    if audit.get('feasible'):
                        score=audit['recomputed']['canonical_objective_score']
                        if score<ub-1e-9:
                            record['gain']+=ub-score;ub=score;best=audit['normalized_solution'];improved=True;improvements+=1
                            _pd_write(folder/'candidate.json',best);break
            record['seconds']+=time.monotonic()-opstart
            state.update(tested_moves=sorted(seen)[-12000:],scan_round=round_index+1)
            _pd_write(folder/'search_state.json',state)
            if improved:break
        if not improved:break
    return best,dict(attempts=attempts,improvements=improvements,operators=history,termination_reason='LOCAL_SEARCH_FINISHED')



def _v58_insertions(p,routes,client,vehicle_ids):
    import copy
    for index,r in enumerate(routes):
        for at in range(1,len(r['path'])):
            trial=copy.deepcopy(routes);trial[index]['path'].insert(at,client);yield trial
    if p['allow_multi_trip']:
        for vid in vehicle_ids:
            positions=[i for i,r in enumerate(routes) if r['vehicle_id']==vid]
            for at in positions+([positions[-1]+1] if positions else [len(routes)]):
                trial=copy.deepcopy(routes);trial.insert(at,dict(vehicle_id=vid,path=[0,client,0]));yield trial

def _v58_eliminate(inputs,best,seconds,folder,state):
    import copy,time,random
    deadline=time.monotonic()+seconds;provider=_v58_provider(inputs);p=provider.problem
    if p['must_use_all_vehicles']:return best,dict(termination_reason='REQUIRED_FLEET',k_status='NOT_APPLICABLE')
    coeff=provider.objective_context['coefficients'];routes=_v58_routes(best);vids=sorted({r['vehicle_id'] for r in routes})
    if len(vids)<2:return best,dict(termination_reason='ONE_VEHICLE',k_status='NOT_APPLICABLE')
    partial_p=copy.deepcopy(p);partial_p['minimum_usage']=0.;partial_p['must_use_all_vehicles']=False
    variants_examined=[];rejection_reasons={}
    tried=set(state.get('elimination_attempts',[]));partials=state.setdefault('elimination_partials',[])
    seed=int(state.get('elimination_round',0));state['elimination_round']=seed+1;rng=random.Random(8131+seed);attempts=0;maxserved=max([item.get('served',0) for item in partials if item.get('base_signature')==_v58_signature(routes)]+[0])
    ordered=sorted(vids,key=lambda v:sum(len(r['path'])-2 for r in routes if r['vehicle_id']==v))
    ordered=ordered[seed%len(ordered):]+ordered[:seed%len(ordered)]
    def difficulty(i):
        lo=max(p['operation_start'],p['windows'][i][0]);hi=min(p['windows'][i][1],p['operation_end']-p['service_time']-p['travel_time'][i,0])
        return hi-lo,-p['client_rows'][i]['demand'],i
    for vid in ordered:
        for variant in range(5):
            if time.monotonic()>deadline-1:break
            key=_v58_signature(routes)+f':{vid}:{variant}:{seed}'
            if key in tried:continue
            variants_examined.append(dict(vehicle=vid,variant=variant,seed=seed))
            remaining=[copy.deepcopy(r) for r in routes if r['vehicle_id']!=vid];allowed=[v for v in vids if v!=vid]
            pending=[i for r in routes if r['vehicle_id']==vid for i in r['path'][1:-1]]
            if variant==0 or variant>=3:
                saved=[item for item in partials if item.get('base_signature')==_v58_signature(routes) and item.get('removed_vehicle')==vid]
                if saved:
                    remaining=copy.deepcopy(max(saved,key=lambda item:item['served'])['routes'])
                    present={i for r in remaining for i in r['path'][1:-1]}
                    pending=[i for i in p['customers'] if i not in present]
            if variant:
                # Eject flexible customers too, so the receiver's current arrangement is not immutable.
                receivers=sorted([i for r in remaining for i in r['path'][1:-1]],key=difficulty,reverse=True)
                if variant>=3 and pending:
                    receivers.sort(key=lambda i:min(p['distance'][i,j] for j in pending))
                ejected=set(receivers[:min(6,2*variant)]);pending+=sorted(ejected)
                pending=list(dict.fromkeys(pending))
                for r in remaining:r['path']=[i for i in r['path'] if i not in ejected]
                remaining=[r for r in remaining if len(r['path'])>2]
            pending.sort(key=difficulty)
            if seed%3==2 and len(pending)>3:
                tail=pending[3:];rng.shuffle(tail);pending=pending[:3]+tail
            beam=[remaining];complete=True
            for client in pending:
                nextbeam=[];local_seen=set()
                for current in beam:
                    ranked=sorted(_v58_insertions(p,current,client,allowed),key=lambda r:_v58_surrogate(p,coeff,r))
                    accepted=0
                    for trial in ranked:
                        if time.monotonic()>deadline-.4:break
                        sig=_v58_signature(trial)
                        if sig in local_seen:continue
                        local_seen.add(sig);attempts+=1
                        answer,reason=_v57_schedule(partial_p,trial,min(.5,deadline-time.monotonic()))
                        if not answer:
                            rejection_reasons[str(reason)]=rejection_reasons.get(str(reason),0)+1
                            blocked=state.setdefault('elimination_blockers_v514',{}).setdefault(str(client),{})
                            blocked[str(reason)]=blocked.get(str(reason),0)+1
                        if answer:
                            nextbeam.append(trial);accepted+=1
                            if accepted>=4:break
                if not nextbeam:complete=False;break
                beam=sorted(nextbeam,key=lambda r:_v58_surrogate(p,coeff,r))[:6]
                served=sum(len(r['path'])-2 for r in beam[0]);maxserved=max(maxserved,served)
                partials.append(dict(base_signature=_v58_signature(routes),removed_vehicle=vid,served=served,routes=copy.deepcopy(beam[0]),missing=[i for i in p['customers'] if i not in {j for r in beam[0] for j in r['path'][1:-1]}]))
                partials.sort(key=lambda x:-x['served']);partials[:]=partials[:8]
            tried.add(key)
            if complete:
                for trial in beam:
                    if time.monotonic()>deadline-.1:break
                    candidate,_=_v57_schedule(p,trial,min(.7,deadline-time.monotonic()))
                    audit=_pd_audit(inputs,candidate) if candidate else {}
                    if audit.get('feasible'):
                        witness=audit['normalized_solution'];_pd_write(folder/'k_witness.json',witness)
                        if witness['canonical_objective_score']<best['canonical_objective_score']:
                            best=witness;_pd_write(folder/'candidate.json',best)
                        state.update(elimination_attempts=sorted(tried)[-300:],elimination_round=seed+1)
                        _pd_write(folder/'search_state.json',state)
                        return best,dict(termination_reason='ELIMINATION_WITNESS',k_status='FEASIBLE_AUDITED',attempts=attempts,max_served=len(p['customers']),variants_examined=variants_examined,rejection_reasons=rejection_reasons)
            state.update(elimination_attempts=sorted(tried)[-300:],elimination_round=seed+1)
            _pd_write(folder/'search_state.json',state)
    return best,dict(termination_reason='ELIMINATION_SEARCH_FINISHED' if attempts else 'NO_ELIMINATION_ATTEMPT',k_status='UNKNOWN',attempts=attempts,max_served=maxserved,variants_examined=variants_examined,rejection_reasons=rejection_reasons)





def _v58_cheap_bounds(p,coeff):
    import math
    C=p['customers'];V=p['vehicles'];nodes=[0]+C;Q=max(v['capacity'] for v in V)
    distance=max(sum(min(p['distance'][j,i] for j in nodes if j!=i) for i in C),
                 sum(min(p['distance'][i,j] for j in nodes if j!=i) for i in C))
    duration=len(C)*p['service_time']+distance*60/p['speed']
    trips=max(1,math.ceil(sum(p['client_rows'][i]['demand'] for i in C)/Q-1e-10))
    count=max(1,math.ceil(duration/p['max_time']-1e-10),trips if not p['allow_multi_trip'] else 1)
    if p['must_use_all_vehicles']:count=len(V)
    duration=max(duration,count*p['minimum_usage'])
    fixed=sorted(v['fixed_cost'] for v in V)
    cost=(sum(fixed) if p['fixed_cost_mode']!='WHEN_USED' else sum(fixed[:count]))+min(v['cost_per_km'] for v in V)*distance
    b=dict(distance_lb=distance,time_lb=duration,cost_lb=cost,vehicles_lb=count,trips_lb=trips)
    b['canonical_objective_lb']=sum(coeff.get(k,0)*b[f] for k,f in [('DISTANCE','distance_lb'),('TIME','time_lb'),('COST','cost_lb'),('VEHICLES','vehicles_lb')])
    b['canonical_objective_lb_valid']=True
    return b



def _v58_exact_search(inputs,best,seconds,folder,state):
    import time,itertools
    deadline=time.monotonic()+seconds;tried=set(state.get('exact_neighborhoods',[]));runs=[]
    while time.monotonic()<deadline-2:
        ordered=sorted(best['routes'],key=lambda r:(r['vehicle_id'],r['start_time_minutes']))
        signature=_v58_signature(_v58_routes(best));choices=[]
        for a,b in itertools.combinations(range(len(ordered)),2):
            if ordered[a]['vehicle_id']==ordered[b]['vehicle_id'] and b!=a+1:continue
            count=sum(len(ordered[i]['path'])-2 for i in (a,b))
            key=signature+f':{a}:{b}'
            if 2<=count<=15 and key not in tried:
                priority=sum(ordered[i]['distance_km'] for i in (a,b))/count
                choices.append((-priority,a,b,key))
        if not choices:break
        _,a,b,key=min(choices);tried.add(key)
        candidate,info=_v58_exact_neighborhood(inputs,best,{a,b},min(15.,deadline-time.monotonic()-.5))
        runs.append(info)
        if candidate:
            audit=_pd_audit(inputs,candidate)
            if audit.get('feasible') and audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:
                best=audit['normalized_solution'];_pd_write(folder/'candidate.json',best)
        state['exact_neighborhoods']=sorted(tried)[-2000:];_pd_write(folder/'search_state.json',state)
    return best,dict(termination_reason='SMALL_EXACT_FINISHED',neighborhoods=runs,nodes=sum(r['nodes'] or 0 for r in runs),global_bound_usable=False)






CONTINUITY_GENERATE_ZIP = False
CONTINUITY_DOWNLOAD_ZIP = False

_v58_legacy_worker=_ct_worker













def _v534_validate_result(result):
    import math
    cert=result.get('certificate') or {}
    if result.get('issues') or not cert.get('report_allowed') or not cert.get('post_certification_independent_audit_passed'):
        raise ValueError('Resultado sem auditoria final aprovada.')
    ub=float(cert['upper_bound']);lb=float(cert['lower_bound'])
    if not math.isfinite(ub) or not math.isfinite(lb) or lb<0 or lb>ub+1e-7:
        raise ValueError('Intervalo de certificação inconsistente.')
    gap=100*max(0.,ub-lb)/max(abs(ub),1e-12)
    if not math.isclose(gap,float(cert['gap_pct']),abs_tol=1e-5):raise ValueError('GAP inconsistente.')
    score=(result.get('final_solution') or {}).get('canonical_objective_score')
    if score is not None and not math.isclose(float(score),ub,abs_tol=1e-7):raise ValueError('Plano e certificado divergentes.')
    return cert

def relatorio_tecnico_continuidade(result):
    cert=_v534_validate_result(result)
    text='| Etapa | Ganho no objetivo | Ganho no limite | Pico de memória do processo (MiB) | Estado |\n|---|---:|---:|---:|---|\n'
    for name,row in result.get('statistics',{}).get('stages',{}).items():
        text+=f"| {name} | {row.get('ub_gain',0)} | {row.get('lb_gain',0)} | {row.get('peak_worker_rss_mb','não medido')} | {row.get('termination_reason','não registrado')} |\n"
    return text

def relatorio_final_continuidade(result):
    return _ct_customer_report(result)



def _ct_publish_report(result, mem, folder):
    if result.get('issues') or not result.get('certificate', {}).get('report_allowed'):
        return
    result['customer_context_v534'] = {'user_answers': _pd_copy.deepcopy(mem.get('user_answers') or {}), 'df_veiculos_capacidade': _pd_clean(mem.get('df_veiculos_capacidade'))}
    cert = result['certificate']
    cert['certification_scale'] = CERTIFICATION_SCALE_V591
    cert['certification_scale_note'] = 'B intermediário até 35%; antigo B até 1% agora B+. Alteração de rótulo não é ganho matemático.'
    stats = result['statistics']
    state = mem['continuity_support_state_v53']
    history = list(state.get('round_history_v591', []))
    beforegap = 100 * max(0.0, (stats['before_ub'] - stats['before_lb']) / max(abs(stats['before_ub']), 1e-12))
    quality = _v511_round_quality(stats, cert)
    relevant = quality['relevant']
    result['productivity_v511'] = quality
    row = dict(round=state.get('round', len(history) + 1), folder=str(folder), seconds=stats['elapsed_seconds'], ub_gain=stats['before_ub'] - stats['after_ub'], lb_gain=stats['after_lb'] - stats['before_lb'], gap_pct=cert['gap_pct'], level=cert['certification_level'], relevant_gain=bool(relevant), productivity_v511=quality)
    row.update(before_ub=stats['before_ub'], after_ub=stats['after_ub'], before_lb=stats['before_lb'], after_lb=stats['after_lb'], baseline_summary=_pd_copy.deepcopy(stats.get('baseline_summary')), final_summary=_pd_copy.deepcopy(result['final_solution'].get('summary')))
    if not history or history[-1]['folder'] != str(folder):
        history.append(row)
    stagnant = 0
    for r in reversed(history):
        if r['relevant_gain']:
            break
        stagnant += 1
    state['round_history_v591'] = history
    result.update(round_history=history, stagnant_rounds=stagnant)
    mem['continuity_certificate'] = _pd_copy.deepcopy(cert)
    _pd_write(_PDPath(folder) / 'checkpoint.json', state)
    result['final_solution'] = _v59_number_routes(result['final_solution'])
    report = salvar_relatorio_continuidade(result, folder)
    result['report'] = report
    cert = _pd_copy.deepcopy(result['certificate'])
    sol = _pd_copy.deepcopy(result['final_solution'])
    sol.update(certificate=cert, certification_level=cert['certification_level'], proven_optimal=cert['proven_optimal'])
    sol['optimality'] = _pd_copy.deepcopy(cert)
    result['final_solution'] = sol
    mem.setdefault('initial_report_v53', mem.get('final_report'))
    mem.setdefault('initial_final_solution_v53', _pd_copy.deepcopy(mem.get('solver_result')))
    mem['solver_result'] = _pd_copy.deepcopy(sol)
    mem['final_solution'] = _pd_copy.deepcopy(sol)
    mem['continuity_final_solution'] = _pd_copy.deepcopy(sol)
    mem['optimality_certificate'] = cert
    mem['certification_result'] = _pd_copy.deepcopy(result)
    mem['final_report'] = report
    mem.update(certification_level=cert['certification_level'], certification_scale=CERTIFICATION_SCALE_V591)

_ct_orchestrate_v55_report=OrchestratorAgent.orchestrate_solution_process
def _ct_orchestrate_with_full_report(self):
    result=_ct_orchestrate_v55_report(self)
    continuation=result.get('continuity_result') if isinstance(result,dict) else None
    if isinstance(continuation,dict) and not continuation.get('issues') and continuation.get('certificate',{}).get('report_allowed'):
        # All legacy visualization consumers must see the same final solution.
        self.memory.data['solver_result']=_pd_copy.deepcopy(result['final_solution'])
        try:
            if hasattr(self,'_run_diagnostic_and_report'):
                self._run_diagnostic_and_report(result['final_solution'],self.memory.data.get('user_answers') or {},{'solucao_aprovada':True})
        except Exception as exc:
            print('Diagnóstico visual indisponível; relatório completo preservado:',exc)
        self.memory.data['final_report']=result['report']
        mostrar_relatorio_continuidade(continuation)
    return result
OrchestratorAgent.orchestrate_solution_process=_ct_orchestrate_with_full_report


def importar_continuidade_zip(caminho,orchestrator):
    """Import JSON only; never execute serialized code from an archive."""
    import zipfile
    mem=orchestrator.memory.data;inputs=_pd_memory_inputs(orchestrator.memory);scope=_ct_scope(inputs)
    with zipfile.ZipFile(caminho) as z:
        result=_pd_json.loads(z.read('summary.json'));manifest=_pd_json.loads(z.read('manifest.json'))
        certificate_scope=result.get('certificate',{}).get('problem_fingerprint_v53')
        if certificate_scope!=manifest.get('fingerprint'):raise ValueError('Certificado e manifesto pertencem a escopos distintos.')
        archived=_pd_copy.deepcopy(manifest['input_data'])
        import pandas as pd
        for key,value in list(archived.items()):
            if isinstance(value,dict) and 'records' in value:archived[key]=pd.DataFrame(value['records'],columns=value['columns'])
        def math_scope(data):
            provider=ContinuityExactCertificationAgent(_PDSimpleNamespace(data=_pd_copy.deepcopy(data)),verbose_solver=False)
            provider._load_objective_context();problem=provider._build_problem()
            return _vrp_fingerprint(dict(problem=problem,objective=provider.objective_context['coefficients']))
        if math_scope(archived)!=math_scope(inputs):raise ValueError('ZIP de outra instância/objetivo: importação recusada.')
        report=_pd_audit(inputs,result['final_solution'])
        if not report.get('feasible'):raise ValueError('Solução do ZIP não passou na auditoria.')
        cert=result['certificate'];ub=float(report['recomputed']['canonical_objective_score'])
        lb=cert.get('lower_bound') if cert.get('lower_bound_certified') else None
        if lb is None or not _pd_math.isfinite(float(lb)) or not 0<=float(lb)<=ub:raise ValueError('Limite do ZIP inválido.')
        prior=mem.get('continuity_support_state_v53') or {}
        if prior.get('fingerprint')!=scope:prior={}
        prior.update(fingerprint=scope,lower_bound=max(float(lb),float(prior.get('lower_bound',0))),best_solution=report['normalized_solution'],
            k_status=cert.get('k_status','UNKNOWN'),target_k=cert.get('target_k'),round=int(prior.get('round',0)))
        if 'K_AG/partial.json' in z.namelist():prior['repair_partial_v57']=_pd_json.loads(z.read('K_AG/partial.json'))
        elif 'K_AG/ga_result.json' in z.namelist() and 'K_AG/result.json' in z.namelist():
            ga=_pd_json.loads(z.read('K_AG/ga_result.json'));meta=_pd_json.loads(z.read('K_AG/result.json'))
            if ga.get('routes') and meta.get('selected_original_vehicle_ids'):
                prior['repair_partial_v57']=dict(candidate=ga,vehicle_ids=meta['selected_original_vehicle_ids'],round=0)
        if 'checkpoint.json' in z.namelist():
            checkpoint=_pd_json.loads(z.read('checkpoint.json'))
            if checkpoint.get('fingerprint')==manifest.get('fingerprint'):
                prior['search_state_v58']=checkpoint.get('search_state_v58',{})
                prior['stage_performance']=checkpoint.get('stage_performance',{})
        mem['continuity_support_state_v53']=prior
        print('Checkpoint importado e solução reauditada. Esta versão preserva soluções, limites e histórico; não depende de uma árvore SCIP persistente.')
    return prior


def retomar_do_zip(caminho,minutos=5):
    import zipfile,pandas as pd
    with zipfile.ZipFile(caminho) as z:inputs=_pd_json.loads(z.read('manifest.json'))['input_data']
    for key,value in list(inputs.items()):
        if isinstance(value,dict) and 'records' in value:inputs[key]=pd.DataFrame(value['records'],columns=value['columns'])
    obj=OrchestratorAgent();obj.memory.data.update(inputs)
    importar_continuidade_zip(caminho,obj)
    result=continuar_rodada(obj,minutos)
    return obj,result







def baixar_relatorio_continuidade(resultado):
    from google.colab import files
    files.download(str(_PDPath(resultado['folder'])/'relatorio_final.html'))




````

#### Modelos locais e limites

Revisa partes do plano, como um grupo de viagens ou a distribuição entre veículos, e calcula limites que ajudam a avaliar sua qualidade.


````python
def _v59_temporal_data(p):
    """Necessary earliest/latest service starts, using metric depot travel."""
    C=p['customers'];s=p['service_time'];tt=p['travel_time']
    E={i:max(p['windows'][i][0],p['operation_start']+tt[0,i]) for i in C}
    L={i:min(p['windows'][i][1],p['operation_end']-s-tt[i,0]) for i in C}
    conflicts=set()
    for at,i in enumerate(C):
        for j in C[at+1:]:
            if E[i]+s+tt[i,j]>L[j]+1e-8 and E[j]+s+tt[j,i]>L[i]+1e-8:
                conflicts.add(tuple(sorted((i,j))))
    # Energy lower bound: minimum overlap of a length-s service with [a,b],
    # when its start can be anywhere in [E,L]. No observed incumbent times used.
    endpoints=sorted(set([p['operation_start'],p['operation_end']]+list(E.values())+[L[i]+s for i in C]))
    intervals=[]
    for ai,a in enumerate(endpoints):
        for b in endpoints[ai+1:]:
            if b<=a+1e-8:continue
            energy={i:max(0.,min(s,b-a,E[i]+s-a,b-L[i])) for i in C}
            total=sum(energy.values())
            if total>1e-8:intervals.append((total/(b-a),a,b,energy))
    # Bounded, instance-derived set; skipping a valid cut only weakens the LB.
    intervals=sorted(intervals,key=lambda r:(-r[0],r[2]-r[1]))[:100]
    return E,L,conflicts,intervals

def _v59_temporal_cuts(model,p,x,z,assignment,tripv,work):
    import math
    from pyscipopt import quicksum as qs
    C=p['customers'];V=p['vehicles'];s=p['service_time'];tt=p['travel_time'];Q=max(v['capacity'] for v in V)
    E,L,conflicts,intervals=_v59_temporal_data(p);count=0
    tau={i:model.addVar(lb=E[i],ub=max(E[i],L[i]),name=f'temporal_start_{i}') for i in C}
    for i in C:
        if E[i]>L[i]+1e-8:model.addCons(tau[i]<=L[i]);count+=1
    for (i,j),edge in x.items():
        if i and j:
            M=max(0.,L[i]+s+tt[i,j]-E[j])
            model.addCons(tau[j]>=tau[i]+s+tt[i,j]-M*(1-edge));count+=1
    for i,j in sorted(conflicts):
        for v in V:
            vi=v['idx'];model.addCons(assignment[i,vi]+assignment[j,vi]<=z[vi]);count+=1
    for ratio,a,b,energy in intervals:
        required=math.ceil(sum(energy.values())/(b-a)-1e-9)
        model.addCons(qs(z.values())>=required);count+=1
        for v in V:
            vi=v['idx'];model.addCons(qs(e*assignment[i,vi] for i,e in energy.items() if e>0)<=(b-a)*z[vi]);count+=1
    for v in V:
        vi=v['idx'];eligible=[i for i in C if p['client_rows'][i]['demand']<=v['capacity']+1e-8]
        if not eligible:continue
        travel=min(tt[0,i]+tt[i,0] for i in eligible)
        horizon=min(p['max_time'],p['operation_end']-p['operation_start'])
        if s+travel>1e-8:
            model.addCons(tripv[vi]<=math.floor(horizon/(s+travel)+1e-9)*z[vi]);count+=1
        model.addCons(tripv[vi]<=qs(assignment[i,vi] for i in C));count+=1
        model.addCons(work[vi]>=s*qs(assignment[i,vi] for i in C)+travel*tripv[vi]);count+=1
    # Cliques whose clients cannot share any single trip: capacity OR temporal conflict.
    incompatible=set(conflicts)
    for at,i in enumerate(C):
        for j in C[at+1:]:
            if p['client_rows'][i]['demand']+p['client_rows'][j]['demand']>Q+1e-8:incompatible.add(tuple(sorted((i,j))))
    cliques=set()
    for anchor in C:
        group=[anchor]
        for i in sorted(C,key=lambda j:(L[j]-E[j],j)):
            if i not in group and all(tuple(sorted((i,j))) in incompatible for j in group):group.append(i)
        if len(group)>1:cliques.add(tuple(sorted(group)))
    for clique in sorted(cliques):
        S=set(clique)
        model.addCons(qs(edge for (i,j),edge in x.items() if i not in S and j in S)>=len(S));count+=1
    return dict(temporal_constraints=count,intervals=len(intervals),incompatible_pairs=len(conflicts),trip_cliques=len(cliques))

def _v59_neighborhood_choices(p,best,tried,stalled):
    import itertools
    routes=sorted(best['routes'],key=lambda r:(r['vehicle_id'],r['start_time_minutes']))
    signature=_vrp_fingerprint([(r['vehicle_id'],r['path'],r['start_time_minutes'],r['end_time_minutes']) for r in routes]);V={v['vehicle_id']:v for v in p['vehicles']}
    def priority(r):
        distance=r['distance_km'];duration=r['total_route_time_minutes'];n=len(r['path'])-2
        waiting=max(0.,duration-n*p['service_time']-distance*60/p['speed'])
        fill=r.get('total_route_load',0)/V[r['vehicle_id']]['capacity']
        return distance/max(n,1)+waiting/20+max(0.,1-fill)*3
    choices=[]
    # Try pairs first; after stagnation interleave 3/4-trip regions rather than
    # exhausting every possible pair before expansion.
    sizes=((4,3,2) if stalled%2==0 else (3,4,2)) if stalled>=2 else (2,3,4)
    for rank,size in enumerate(sizes):
        for indices in itertools.combinations(range(len(routes)),size):
            customers=sum(len(routes[i]['path'])-2 for i in indices)
            if customers>24:continue
            contiguous=True
            for vid in {routes[i]['vehicle_id'] for i in indices}:
                seq=[i for i,r in enumerate(routes) if r['vehicle_id']==vid]
                pos=[seq.index(i) for i in indices if routes[i]['vehicle_id']==vid]
                if max(pos)-min(pos)+1!=len(pos):contiguous=False;break
            key=signature+':'+','.join(map(str,indices))
            if contiguous and key not in tried:
                score=sum(priority(routes[i]) for i in indices)/size
                choices.append((rank,-score,indices,key))
        if choices:break
    return sorted(choices)



def _base_v59_exact_search_1(inputs,best,seconds,folder,state):
    import time
    deadline=time.monotonic()+seconds;provider=_v58_provider(inputs);p=provider.problem
    round_id=int(state.get('region_round_v511',0))+1;state['region_round_v511']=round_id
    ledger=state.setdefault('region_ledger_v511',{});cooldown_skips=0
    tried=set(state.get('closed_neighborhoods_v511',[]));attempted=set();pending=state.setdefault('pending_neighborhoods_v592',{});stalled=int(state.get('local_stalled',0));runs=[];stop_reason="REGION_BUDGET_EXHAUSTED"
    while time.monotonic()<deadline-2:
        choices=_v59_neighborhood_choices(p,best,tried|attempted,stalled)
        if not choices:
            stop_reason="NO_UNEXAMINED_REGIONS";break
        eligible=[c for c in choices if int(ledger.get(c[3],{}).get('retry_after_round',0))<=round_id]
        blocked={c[3] for c in choices if c not in eligible};cooldown_skips+=len(blocked);attempted.update(blocked)
        if not eligible:continue
        _,_,indices,key=min(eligible,key=lambda c:(int(ledger.get(c[3],{}).get('attempts',0)),c[0],c[1]))
        attempted.add(key);entry=ledger.setdefault(key,dict(attempts=0));retry=entry['attempts']>0
        entry.update(attempts=entry['attempts']+1,status='RUNNING',round=round_id,released_trips=len(indices))
        if entry['attempts']%3==0:entry['retry_after_round']=round_id+3
        _pd_write(folder/'search_state.json',state)
        candidate,info=_v58_exact_neighborhood(inputs,best,set(indices),min((12 if len(indices)==2 else 18)*(1+min(2,pending.get(key,0))),deadline-time.monotonic()-.5))
        entry['status']=info.get('status');info['retry']=retry
        if entry['attempts']%3==0 and info.get('status') not in ('optimal','infeasible'):entry['retry_after_round']=round_id+3
        info['released_trips']=len(indices);info['neighborhood_key']=key;runs.append(info);improved=False
        if info.get('status') in ('optimal','infeasible'):tried.add(key);pending.pop(key,None)
        else:pending[key]=pending.get(key,0)+1
        info['before_ub']=best['canonical_objective_score']
        if candidate:
            audit=_pd_audit(inputs,candidate)
            info['candidate_audit_passed']=bool(audit.get('feasible'))
            if audit.get('feasible'):
                # Keep complete alternatives as sources of reusable vehicle programs.
                state.setdefault('feasible_pool',[]).append(audit['normalized_solution'])
                pool={_v58_signature(_v58_routes(s)):s for s in state['feasible_pool']}
                state['feasible_pool']=sorted(pool.values(),key=lambda s:s['canonical_objective_score'])[:8]
                if audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:
                    best=audit['normalized_solution'];_pd_write(folder/'candidate.json',best);improved=True
        stalled=0 if improved else stalled+1
        state.update(closed_neighborhoods_v511=sorted(tried)[-3000:],local_stalled=stalled)
        if len(ledger)>2000:
            keep=sorted(ledger,key=lambda k:ledger[k].get('round',0),reverse=True)[:2000]
            ledger={k:ledger[k] for k in keep};state['region_ledger_v511']=ledger
        _pd_write(folder/'search_state.json',state)
        info['after_ub']=best['canonical_objective_score']
        if improved and inputs.get('portfolio_mode_v512'):
            stop_reason='NEW_CANDIDATE_FOR_POLISH';break
    return best,dict(termination_reason=stop_reason,assigned_seconds=seconds,neighborhoods=runs,cooldown_skips=cooldown_skips,nodes=sum(r['nodes'] or 0 for r in runs),global_bound_usable=False)
_v59_exact_search = _base_v59_exact_search_1





def _v58_exact_neighborhood(inputs,best,indices,seconds):
    """Small exact subproblem; frozen outside trips define per-vehicle time intervals.
    The subproblem's dual and infeasibility are NEVER global certificates.
    """
    import copy,time
    from pyscipopt import quicksum as qs
    started=time.monotonic();full=_v58_provider(inputs);p=full.problem
    ordered=sorted(best['routes'],key=lambda r:(r['vehicle_id'],r['start_time_minutes']))
    selected=[ordered[i] for i in indices];outside=[r for i,r in enumerate(ordered) if i not in indices]
    customers=sorted({i for r in selected for i in r['path'][1:-1]})
    vids={r['vehicle_id'] for r in selected};windows={};outsidework={}
    for vid in vids:
        seq=[(i,r) for i,r in enumerate(ordered) if r['vehicle_id']==vid];positions=[j for j,(i,r) in enumerate(seq) if i in indices]
        if positions!=list(range(min(positions),max(positions)+1)):raise ValueError('Selected trips must be contiguous per vehicle')
        lo=seq[min(positions)-1][1]['end_time_minutes'] if min(positions)>0 else p['operation_start']
        hi=seq[max(positions)+1][1]['start_time_minutes'] if max(positions)+1<len(seq) else p['operation_end']
        windows[vid]=(lo,hi)
        outsidework[vid]=sum(r['total_route_time_minutes'] for r in outside if r['vehicle_id']==vid)
    localp=copy.deepcopy(p);localp.update(customers=customers,vehicles=[copy.deepcopy(v) for v in p['vehicles'] if v['vehicle_id'] in vids],minimum_usage=0.,must_use_all_vehicles=False)
    localp['client_rows']={i:p['client_rows'][i] for i in [0]+customers if i in p['client_rows']}
    localp['windows']={i:p['windows'][i] for i in customers}
    for v in localp['vehicles']:
        if outsidework[v['vehicle_id']]>0 or p['fixed_cost_mode']!='WHEN_USED':v['fixed_cost']=0.
    class LocalProvider(ContinuityExactCertificationAgent):
        def _analytical_lower_bounds(self,incumbent_ub=None):return _v58_cheap_bounds(self.problem,self.objective_context['coefficients'])
    provider=LocalProvider(_PDSimpleNamespace(data={}),verbose_solver=False,time_limit_seconds=max(.1,seconds))
    provider.problem=localp;provider.objective_context=copy.deepcopy(full.objective_context)
    bounds=provider._analytical_lower_bounds()
    # Avoid the global preparatory SCF solve inside a small local task.
    provider.shared_memory.data['scf_model_cache_v52']={_vrp_fingerprint((localp,int(bounds['trips_lb']))):dict(valid=False,status='LOCAL_SKIP')}
    model,h=provider._build_scip_model();nodes=None
    try:
        for v in h['V']:
            vi=v['idx'];vid=v['vehicle_id'];lo,hi=windows[vid]
            work=qs(h['qtime'][i,vi] for i in customers)
            model.addCons(work<=max(0.,p['max_time']-outsidework[vid])*h['used'][vi])
            minimum=max(0.,p['minimum_usage']-outsidework[vid])
            model.addCons(work>=minimum if outsidework[vid]>0 else work>=minimum*h['used'][vi])
            if p['must_use_all_vehicles'] and outsidework[vid]<=1e-9:model.addCons(h['used'][vi]==1)
            # These bounds protect the fixed trips before and after the released block.
            for i in customers:
                M=max(1.,p['operation_end']-p['operation_start']+p['max_time']+p['travel_time'][i,0]+p['travel_time'][0,i])
                model.addCons(h['tau'][i]>=lo+p['travel_time'][0,i]-M*(1-h['y'][i,vi]))
                model.addCons(h['tau'][i]+p['service_time']+p['travel_time'][i,0]<=hi+M*(1-h['y'][i,vi]))
                model.addCons(h['dep'][i]>=lo-M*(2-h['y'][i,vi]-h['first'][i]))
        # The incumbent is a feasible starting point for this neighborhood.
        provider._try_add_incumbent_mip_start(model,h,dict(routes=selected))
        model.setParam('limits/time',max(.05,seconds-(time.monotonic()-started)))
        model.optimize();status=str(model.getStatus());nodes=model.getNNodes()
        candidate=provider._extract_solution(model,h)
        if candidate:candidate=dict(status='SUCCESS',routes=outside+candidate['routes'])
        return candidate,dict(status=status,nodes=nodes,clients=len(customers),restricted_proof=True,global_bound_usable=False,elapsed_seconds=time.monotonic()-started)
    finally:model.freeProb()








"""Bounded column generation over complete single-vehicle day programs.
The restricted-master LP objective is NEVER used directly as a global LB.
Dual correction uses a valid reduced-cost lower bound for EVERY vehicle.
"""
def _v59_canonical_program(p,vid,routes,earliest=False):
    """Replay the auditor's service rule; voluntary padding is not workload."""
    v=next(v for v in p['vehicles'] if v['vehicle_id']==vid)
    clock=p['operation_start'];work=0.;answer=[];seen=set()
    if not p['allow_multi_trip'] and len(routes)>1:return None
    for route in routes:
        path=list(route['path'])
        if len(path)<3 or path[0]!=0 or path[-1]!=0:return None
        start=clock if earliest else max(clock,float(route.get('start_time_minutes',clock)))
        now=start;details=[];load=0.;distance=0.
        for previous,i in zip(path,path[1:]):
            distance+=p['distance'][previous,i];arrival=now+p['travel_time'][previous,i]
            if i==0:now=arrival;continue
            if i in seen or i not in p['customers']:return None
            seen.add(i);service=max(arrival,p['windows'][i][0])
            load+=p['client_rows'][i]['demand']
            if service>p['windows'][i][1]+1e-6 or load>v['capacity']+1e-6:return None
            now=service+p['service_time']
            details.append(dict(node_id=i,arrival_time_minutes=arrival,service_start_time_minutes=service,departure_time_minutes=now,cumulative_load=load))
        if now>p['operation_end']+1e-6:return None
        duration=now-start;work+=duration;clock=now
        answer.append(dict(vehicle_id=vid,path=path,trip_id=len(answer)+1,start_time_minutes=start,end_time_minutes=now,total_route_time_minutes=duration,distance_km=distance,total_route_load=load,path_details=details))
    if work<p['minimum_usage']-1e-6 or work>p['max_time']+1e-6:return None
    return answer

def _v59_program(p,coeff,vid,routes,seconds=.5):
    import copy
    if not routes:return None
    # Audited solutions can store frozen trips first and newly optimized trips last.
    # Recover physical chronological order before rescheduling a whole day.
    if all('start_time_minutes' in route for route in routes):
        routes=sorted(routes,key=lambda route:route['start_time_minutes'])
    local=copy.deepcopy(p);local['must_use_all_vehicles']=False
    raw=[dict(vehicle_id=vid,path=list(r['path'])) for r in routes if len(r['path'])>2]
    if not raw:return None
    solution,reason=_v57_schedule(local,raw,seconds)
    canonical=_v59_canonical_program(p,vid,solution['routes']) if solution else None
    if canonical is None and all('start_time_minutes' in r for r in routes):
        canonical=_v59_canonical_program(p,vid,routes)
    if canonical is None:canonical=_v59_canonical_program(p,vid,raw,earliest=True)
    if canonical is None:return None
    return _v59_program_from_routes(p,coeff,vid,canonical)

def _v59_program_from_routes(p,coeff,vid,routes):
    routes=_pd_copy.deepcopy(routes);clients=[i for r in routes for i in r['path'][1:-1]]
    if not clients or len(set(clients))!=len(clients):return None
    v=next(v for v in p['vehicles'] if v['vehicle_id']==vid)
    distance=sum(r['distance_km'] for r in routes);duration=sum(r['total_route_time_minutes'] for r in routes)
    fixed=v['fixed_cost'] if p['fixed_cost_mode']=='WHEN_USED' else 0.
    cost=coeff.get('DISTANCE',0)*distance+coeff.get('COST',0)*(fixed+v['cost_per_km']*distance)+coeff.get('TIME',0)*duration+coeff.get('VEHICLES',0)
    return dict(vehicle_id=vid,customers=sorted(clients),routes=routes,cost=float(cost),signature=_v58_signature(routes))

def _v59_master(p,coeff,columns,seconds,integer=False):
    import numpy as np
    C=p['customers'];V=p['vehicles'];constant=coeff.get('COST',0)*sum(v['fixed_cost'] for v in V) if p['fixed_cost_mode']!='WHEN_USED' else 0.
    if not integer:
        from scipy.optimize import linprog
        from scipy.sparse import csc_matrix
        A=np.array([[float(i in col['customers']) for col in columns] for i in C])
        B=np.array([[float(v['vehicle_id']==col['vehicle_id']) for col in columns] for v in V])
        mandatory=p['must_use_all_vehicles']
        solved=linprog([col['cost'] for col in columns],A_eq=csc_matrix(np.vstack([A,B]) if mandatory else A),
            b_eq=np.ones(len(C)+len(V) if mandatory else len(C)),A_ub=None if mandatory else csc_matrix(B),b_ub=None if mandatory else np.ones(len(V)),
            bounds=(0,None),method='highs',options={'time_limit':max(.05,seconds)})
        if not solved.success:return dict(status=str(solved.message),completed=False)
        duals=solved.eqlin.marginals
        return dict(completed=True,status='OPTIMAL_RESTRICTED_LP',objective=float(solved.fun)+constant,
            pi={i:float(duals[j]) for j,i in enumerate(C)},
            sigma={v['vehicle_id']:float(duals[len(C)+j] if mandatory else solved.ineqlin.marginals[j]) for j,v in enumerate(V)},
            weights=list(map(float,solved.x)),global_bound_usable=False)
    from pyscipopt import Model,quicksum as qs
    model=Model('PROGRAM_SELECTION_RESTRICTED');model.hideOutput();model.setParam('limits/time',max(.05,seconds))
    try:
        chosen=[model.addVar(vtype='B') for col in columns]
        for i in C:model.addCons(qs(var for var,col in zip(chosen,columns) if i in col['customers'])==1)
        for v in V:
            expr=qs(var for var,col in zip(chosen,columns) if col['vehicle_id']==v['vehicle_id'])
            model.addCons(expr==1 if p['must_use_all_vehicles'] else expr<=1)
        model.setObjective(qs(col['cost']*var for col,var in zip(columns,chosen))+constant)
        model.optimize();result=dict(status=str(model.getStatus()),completed=True,global_bound_usable=False)
        if model.getNSols():
            sol=model.getBestSol();selected=[col for col,var in zip(columns,chosen) if model.getSolVal(sol,var)>.5]
            result['candidate']=dict(status='SUCCESS',routes=[r for col in selected for r in col['routes']])
        return result
    finally:model.freeProb()



def _v59_pricing_exact(p,coeff,vid,pi,sigma,seconds):
    """Exact optional-customer single-vehicle, multi-trip pricing formulation.
    A time limit still permits use of a finite SCIP DUAL bound, never its primal.
    """
    import time,math
    from pyscipopt import Model,quicksum as qs
    started=time.monotonic();v=next(v for v in p['vehicles'] if v['vehicle_id']==vid);Q=v['capacity'];svc=p['service_time']
    E={i:max(p['windows'][i][0],p['operation_start']+p['travel_time'][0,i]) for i in p['customers']}
    L={i:min(p['windows'][i][1],p['operation_end']-svc-p['travel_time'][i,0]) for i in p['customers']}
    C=[i for i in p['customers'] if p['client_rows'][i]['demand']<=Q+1e-8 and E[i]<=L[i]+1e-8]
    fixed=coeff.get('COST',0)*(v['fixed_cost'] if p['fixed_cost_mode']=='WHEN_USED' else 0.)+coeff.get('VEHICLES',0)
    analytic=fixed-sum(max(0.,pi[i]) for i in C)-sigma
    if not C:return None,dict(status='NO_ELIGIBLE_CUSTOMERS',reduced_cost_lb=0.,completed=True)
    dd=p['distance'];tt=p['travel_time'];op0=p['operation_start'];op1=p['operation_end'];H=min(p['max_time'],op1-op0)
    direct=[(i,j) for i in C for j in C if i!=j and E[i]+svc+tt[i,j]<=L[j]+1e-8 and p['client_rows'][i]['demand']+p['client_rows'][j]['demand']<=Q+1e-8]
    reloads=[(i,j) for i in C for j in C if i!=j and E[i]+svc+tt[i,0]+tt[0,j]<=L[j]+1e-8] if p['allow_multi_trip'] else []
    model=Model('EXACT_VEHICLE_PROGRAM_PRICING');model.hideOutput();model.setParam('limits/memory',1024.)
    model.setParam('separating/maxroundsroot',2)
    try:
        sel={i:model.addVar(vtype='B') for i in C};first={i:model.addVar(vtype='B') for i in C};last={i:model.addVar(vtype='B') for i in C}
        x={a:model.addVar(vtype='B') for a in direct};r={a:model.addVar(vtype='B') for a in reloads}
        tau={i:model.addVar(lb=E[i],ub=L[i]) for i in C};dep={i:model.addVar(lb=op0,ub=op1) for i in C}
        wait={i:model.addVar(lb=0,ub=H) for i in C};load={i:model.addVar(lb=0,ub=Q) for i in C}
        rank={i:model.addVar(lb=0,ub=len(C)) for i in C}
        incoming={i:[] for i in C};outgoing={i:[] for i in C}
        for arcs in (x,r):
            for (i,j),edge in arcs.items():incoming[j].append(edge);outgoing[i].append(edge)
        begin={i:first[i]+qs(edge for (j,k),edge in r.items() if k==i) for i in C}
        finish={i:last[i]+qs(edge for (j,k),edge in r.items() if j==i) for i in C}
        model.addCons(qs(first.values())==1);model.addCons(qs(last.values())==1)
        for i in C:
            demand=p['client_rows'][i]['demand']
            model.addCons(first[i]+qs(incoming[i])==sel[i]);model.addCons(last[i]+qs(outgoing[i])==sel[i])
            model.addCons(wait[i]<=H*sel[i]);model.addCons(load[i]<=Q*sel[i]);model.addCons(load[i]>=demand*sel[i])
            model.addCons(load[i]<=demand+Q*(1-begin[i]))
            M=max(abs(E[i]-op1-tt[0,i]),abs(L[i]-op0-tt[0,i]))+H+1
            delta=tau[i]-dep[i]-tt[0,i]
            model.addCons(wait[i]>=delta-M*(1-begin[i]));model.addCons(wait[i]<=delta+M*(1-begin[i]))
            # E/L already enforce necessary ability to depart and return to depot.
        for (i,j),edge in x.items():
            delta=tau[j]-tau[i]-svc-tt[i,j]
            M=max(abs(E[j]-L[i]-svc-tt[i,j]),abs(L[j]-E[i]-svc-tt[i,j]))+H+1
            model.addCons(wait[j]>=delta-M*(1-edge));model.addCons(wait[j]<=delta+M*(1-edge))
            model.addCons(load[j]>=load[i]+p['client_rows'][j]['demand']-(Q+p['client_rows'][j]['demand'])*(1-edge))
        for (i,j),edge in r.items():
            M=max(0.,L[i]+svc+tt[i,0]-op0)
            model.addCons(dep[j]>=tau[i]+svc+tt[i,0]-M*(1-edge))
        for arcs in (x,r):
            for (i,j),edge in arcs.items():model.addCons(rank[j]>=rank[i]+1-(len(C)+1)*(1-edge))
        D=qs(dd[a]*edge for a,edge in x.items())+qs(dd[0,i]*begin[i]+dd[i,0]*finish[i] for i in C)
        T=svc*qs(sel.values())+qs(wait.values())+60/p['speed']*D
        model.addCons(T<=H);model.addCons(T>=p['minimum_usage'])
        cost=(coeff.get('DISTANCE',0)+coeff.get('COST',0)*v['cost_per_km'])*D+coeff.get('TIME',0)*T+fixed
        model.setObjective(cost-qs(pi[i]*sel[i] for i in C)-sigma)
        remaining=seconds-(time.monotonic()-started)
        if remaining<=.05:return None,dict(status='BUILD_LIMIT',reduced_cost_lb=analytic,completed=False)
        model.setParam('limits/time',remaining);model.optimize();status=str(model.getStatus());dual=float(model.getDualbound())
        rc=analytic
        if math.isfinite(dual) and abs(dual)<1e19:rc=max(rc,dual-1e-7*max(1,abs(dual)))
        result=dict(status=status,reduced_cost_lb=rc,nodes=model.getNNodes(),completed=True,build_seconds=seconds-remaining)
        program=None
        if model.getNSols():
            sol=model.getBestSol();val=lambda a:float(model.getSolVal(sol,a))
            starts=[i for i in C if val(first[i])>.5]
            successor={i:(j,False) for (i,j),a in x.items() if val(a)>.5}
            successor.update({i:(j,True) for (i,j),a in r.items() if val(a)>.5})
            if starts:
                i=starts[0];path=[0];routes=[];seen=set()
                while i not in seen:
                    seen.add(i);path.append(i);nxt=successor.get(i)
                    if nxt is None or nxt[1]:routes.append(dict(vehicle_id=vid,path=path+[0]));path=[0]
                    if nxt is None:break
                    i=nxt[0]
                # Reschedule independently before using as a new column.
                program=_v59_program(p,coeff,vid,routes,min(.5,max(.05,seconds-(time.monotonic()-started))))
        return program,result
    finally:model.freeProb()

def _v59_heuristic_price(p,coeff,vid,pi,sigma,columns,seconds):
    import time,copy
    deadline=time.monotonic()+seconds
    def reduced(col):return col['cost']-sum(pi[i] for i in col['customers'])-sigma
    sources=sorted([col for col in columns if col['vehicle_id']==vid],key=reduced)[:2]
    eligible=[i for i in p['customers'] if p['client_rows'][i]['demand']<=next(v['capacity'] for v in p['vehicles'] if v['vehicle_id']==vid)]
    if eligible:
        seed=max(eligible,key=lambda i:pi[i]);singleton=_v59_program(p,coeff,vid,[dict(path=[0,seed,0])],.3)
        if singleton:sources.append(singleton)
    generated=[]
    for base in sources:
        current=base
        for _ in range(12):
            if time.monotonic()>deadline-.1:break
            present=set(current['customers']);routes=_v58_routes(current);proposals=[]
            # Existing good trips can be trimmed, allowing a different master combination.
            for client in sorted(present,key=lambda i:pi[i])[:5]:
                if len(present)<=1:break
                trial=copy.deepcopy(routes)
                for route in trial:route['path']=[i for i in route['path'] if i!=client]
                proposals.append([r for r in trial if len(r['path'])>2])
            for client in sorted(set(eligible)-present,key=lambda i:-pi[i])[:5]:
                if pi[client]<=0:continue
                trials=sorted(_v58_insertions(p,routes,client,[vid]),key=lambda r:_v58_surrogate(p,coeff,r))[:6]
                proposals.extend(trials)
            winner=current
            for trial in proposals:
                if time.monotonic()>deadline-.1:break
                program=_v59_program(p,coeff,vid,trial,min(.25,deadline-time.monotonic()))
                if program and reduced(program)<reduced(winner)-1e-8:winner=program
            if winner is current:break
            current=winner;generated.append(current)
    return [col for col in generated if reduced(col)<-1e-7]

def _v59_corrected_dual(p,coeff,lp,reduced_cost_lower_bounds):
    """At most one program per vehicle => add min(0,rc_LB_v) for each vehicle.
    Valid even if pricing is interrupted, provided every rc lower bound is valid.
    """
    constant=coeff.get('COST',0)*sum(v['fixed_cost'] for v in p['vehicles']) if p['fixed_cost_mode']!='WHEN_USED' else 0.
    value=constant+sum(lp['pi'].values())+sum(lp['sigma'].values())
    value+=sum(min(0.,reduced_cost_lower_bounds[v['vehicle_id']]) for v in p['vehicles'])
    return value-1e-7*max(1.,abs(value))

def _v59_columns(inputs,best,seconds,folder,state):
    import time,math
    started=time.monotonic();deadline=started+seconds;provider=_v58_provider(inputs);p=provider.problem;coeff=provider.objective_context['coefficients']
    if any(float(c)<0 for c in coeff.values()) or any(v['cost_per_km']<0 or v['fixed_cost']<0 for v in p['vehicles']):
        return best,dict(termination_reason='COLUMN_NEGATIVE_COST_UNSUPPORTED',global_bound_usable=False)
    bank={};rounds=[];global_lb=0.;bestub=best['canonical_objective_score'];generated=0
    # Bank entries are independently rescheduled, including imports from earlier rounds.
    for position,solution in enumerate([best]+state.get('feasible_pool',[])):
        for vid in {r['vehicle_id'] for r in solution['routes']}:
            routes=[r for r in solution['routes'] if r['vehicle_id']==vid]
            col=_v59_program(p,coeff,vid,routes,.5)
            if col is None and position==0:
                col=_v59_program(p,coeff,vid,routes,1.5)
                if col is None:raise ValueError(f'Incumbent program could not be seeded for vehicle {vid}; restricted master not executed.')
            if col:bank[col['signature']]=col
    for item in state.get('program_bank_v59',[])[:160]:
        if time.monotonic()>deadline-3:break
        col=_v59_program(p,coeff,item['vehicle_id'],item['routes'],.25)
        if col:bank[col['signature']]=col
    # Reuse good trip sequences on other compatible vehicles; charge fixed cost once.
    seeds=list(bank.values())
    for col in seeds[:12]:
        for vehicle in p['vehicles']:
            if time.monotonic()>started+seconds*.12:break
            if vehicle['vehicle_id']==col['vehicle_id']:continue
            mapped=_v59_program(p,coeff,vehicle['vehicle_id'],col['routes'],.2)
            if mapped:bank[mapped['signature']]=mapped
    for iteration in range(12):
        if time.monotonic()>deadline-4:break
        lp=_v59_master(p,coeff,list(bank.values()),min(3.,deadline-time.monotonic()))
        if not lp.get('completed'):
            if iteration==0:raise ValueError('Restricted master failed despite seeded incumbent: '+str(lp.get('status')))
            rounds.append(lp);break
        pi,sigma=lp['pi'],lp['sigma'];rc_bounds={};price_logs=[];added=0
        for v in p['vehicles']:
            vid=v['vehicle_id'];fixed=coeff.get('COST',0)*(v['fixed_cost'] if p['fixed_cost_mode']=='WHEN_USED' else 0.)+coeff.get('VEHICLES',0)
            rc_bounds[vid]=fixed-sum(max(0.,pi[i]) for i in p['customers'] if p['client_rows'][i]['demand']<=v['capacity']+1e-8)-sigma[vid]
        # Price all vehicles heuristically, then selected ones exactly within a short cap.
        for v in p['vehicles']:
            if time.monotonic()>deadline-5:break
            vid=v['vehicle_id']
            for col in _v59_heuristic_price(p,coeff,vid,pi,sigma[vid],list(bank.values()),min(.7,max(.1,(deadline-time.monotonic()-5)/len(p['vehicles'])))):
                if col['signature'] not in bank:bank[col['signature']]=col;added+=1
        offset=int(state.get('pricing_offset_v59',0))
        eligible=sorted(p['vehicles'],key=lambda v:v['vehicle_id']);eligible=eligible[offset%len(eligible):]+eligible[:offset%len(eligible)]
        # Small instances price every vehicle; larger ones rotate a bounded subset.
        cap=len(eligible);processed=0
        for vehicle in eligible[:cap]:
            if time.monotonic()>deadline-5:break
            vid=vehicle['vehicle_id'];col,info=_v59_pricing_exact(p,coeff,vid,pi,sigma[vid],min(4.,max(.1,(deadline-time.monotonic()-4)/max(1,cap-processed))))
            processed+=1;price_logs.append(dict(vehicle_id=vid,**info));rc_bounds[vid]=max(rc_bounds[vid],info['reduced_cost_lb'])
            if col and col['cost']-sum(pi[i] for i in col['customers'])-sigma[vid]<-1e-7 and col['signature'] not in bank:
                bank[col['signature']]=col;added+=1
        corrected=_v59_corrected_dual(p,coeff,lp,rc_bounds)
        if math.isfinite(corrected):global_lb=max(global_lb,corrected)
        info=dict(iteration=iteration,restricted_lp=lp['objective'],corrected_global_lb=corrected,added_columns=added,
                  pricing=price_logs,pricing_vehicle_count=processed,pricing_cycle_complete=processed==len(eligible),all_pricing_closed=all(value>=-1e-7 for value in rc_bounds.values()))
        rounds.append(info);generated+=added;state['pricing_offset_v59']=offset+processed
        state['program_bank_v59']=sorted(bank.values(),key=lambda c:c['cost']/max(1,len(c['customers'])))[:250]
        _pd_write(folder/'search_state.json',state)
        _pd_write(folder/'columns_progress.json',dict(rounds=rounds,global_lower_bound=global_lb,restricted_lp_is_global_bound=False))
        if not added and (processed==len(eligible) or time.monotonic()>deadline-5):break
    left=deadline-time.monotonic()
    integer={}
    if left>.1:
        integer=_v59_master(p,coeff,list(bank.values()),max(.05,left-.1),True)
        candidate=integer.pop('candidate',None)
        if candidate:
            audit=_pd_audit(inputs,candidate)
            if audit.get('feasible') and audit['recomputed']['canonical_objective_score']<bestub-1e-9:
                best=audit['normalized_solution'];_pd_write(folder/'candidate.json',best)
    state['program_bank_v59']=sorted(bank.values(),key=lambda c:c['cost']/max(1,len(c['customers'])))[:250]
    _pd_write(folder/'search_state.json',state)
    return best,dict(termination_reason='COLUMN_GENERATION_FINISHED',rounds=rounds,columns=len(bank),new_columns=generated,
        restricted_master=integer,global_bound_usable=True,canonical_lower_bound=global_lb,
        bound_method='DUAL_CORRECTION_ALL_VEHICLES',restricted_lp_is_global_bound=False)






CONTINUITY_GENERATE_ZIP = False
CONTINUITY_DOWNLOAD_ZIP = False

def _ct_worker(stage,inputs,baseline,cfg,folder):
    if stage not in ('MELHORIA','EXATO_LOCAL','LB_COMPACTO','GERACAO_COLUNAS','ELIMINAR_VEICULO','EXATO_COMPLETO_GLOBAL'):
        return _v58_legacy_worker(stage,inputs,baseline,cfg,folder)
    import threading
    folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True);started=_pd_time.monotonic()
    state=_pd_copy.deepcopy(inputs.get('search_state_v58',{}));fingerprint=_ct_scope(inputs)
    result=dict(task=stage,fingerprint=fingerprint,completed=False,global_bound_usable=False)
    stop=threading.Event()
    def progress():
        while not stop.is_set():
            live=(_pd_read(folder/'lb_live.json') or {}) if stage=='LB_COMPACTO' else {}
            progress_row=dict(phase=stage,fingerprint=fingerprint,updated_at=_pd_time.time(),elapsed_seconds=_pd_time.monotonic()-started)
            progress_row.update(live)
            _pd_write(folder/'progress.json',progress_row)
            stop.wait(2)
    heartbeat=threading.Thread(target=progress,daemon=True);heartbeat.start()
    try:
        seconds=max(.1,float(cfg['WALL_STAGE_SECONDS'])-max(3.,min(8.,.08*float(cfg['WALL_STAGE_SECONDS']))))
        if stage=='LB_COMPACTO':
            bound_agent=BoundStrengtheningAgent(_PDSimpleNamespace(data=_pd_copy.deepcopy(inputs)))
            result.update(bound_agent.strengthen_continuity(inputs,seconds,folder,state))
        else:
            memory=_PDSimpleNamespace(data=_pd_copy.deepcopy(inputs))
            owners={'EXATO_COMPLETO_GLOBAL':(ContinuityExactCertificationAgent,'solve_complete_global'),'MELHORIA':(NeighborhoodOptimizationAgent,'improve_incumbent'),
                'EXATO_LOCAL':(NeighborhoodOptimizationAgent,'optimize_region'),
                'GERACAO_COLUNAS':(VehicleProgramColumnAgent,'optimize'),
                'ELIMINAR_VEICULO':(KVehicleFeasibilityCertifier,'attempt_elimination')}
            cls,method=owners[stage];agent=cls(memory)
            best,stats=getattr(agent,method)(inputs,baseline,seconds,folder,state);result.update(stats)
            result['agent']=cls.__name__
            _pd_write(folder/'candidate.json',best)
        result['completed']=True
    except BaseException as exc:
        result.update(error=f'{type(exc).__name__}: {exc}',traceback=_pd_traceback.format_exc(),termination_reason='WORKER_EXCEPTION')
    finally:
        stop.set();heartbeat.join(timeout=3)
        result['elapsed_seconds']=_pd_time.monotonic()-started
        _pd_write(folder/'search_state.json',state);_pd_write(folder/'result.json',result)




def _v58_allocation(previous):
    # Reserve 25% for independent global bounds and 30% for incumbent improvement.
    floors={'MELHORIA':.15,'LB_COMPACTO':.25,'EXATO_LOCAL':.15,'GERACAO_COLUNAS':.05,'ELIMINAR_VEICULO':.05}
    history=previous.get('stage_performance',{})
    prior={'MELHORIA':3.,'LB_COMPACTO':2.,'EXATO_LOCAL':3.,'GERACAO_COLUNAS':1.,'ELIMINAR_VEICULO':1.}
    rates={k:max(0.,float(history.get(k,{}).get('recent_gain',0.))) for k in floors}
    peak=max(rates.values(),default=0.)
    weights={}
    for k in floors:
        row=history.get(k,{})
        weights[k]=prior[k] if not row.get('runs') else (.15+4*rates[k]/peak if peak>0 else prior[k]*.25)
    total=sum(weights.values());free=1.-sum(floors.values())
    return {k:floors[k]+free*weights[k]/total for k in floors}













def exportar_diagnostico_continuidade(result,download=False):
    import zipfile
    folder=_PDPath(result['folder']);archive=str(folder)+'.zip'
    with zipfile.ZipFile(archive,'w',zipfile.ZIP_DEFLATED) as z:
        for path in folder.rglob('*'):
            if path.is_file() and not path.name.startswith('_internal_'):z.write(path,path.relative_to(folder))
    if download:
        from google.colab import files
        files.download(archive)
    return archive

def continuar_rodada(orchestrator,minutos=5):
    result=executar_continuidade_integrada(orchestrator,float(minutos)*60)
    mostrar_relatorio_continuidade(result)
    return result


def _v59_number_routes(solution):
    import copy
    result=copy.deepcopy(solution);counts={}
    result['routes']=sorted(result['routes'],key=lambda r:(r['vehicle_id'],r['start_time_minutes']))
    for route in result['routes']:
        vid=route['vehicle_id'];counts[vid]=counts.get(vid,0)+1;route['trip_id']=counts[vid]
    return result



def _v59_md_html(text):
    import html,re
    def inline(s):return re.sub(r'\*\*(.*?)\*\*',r'<strong>\1</strong>',html.escape(s))
    result=[];table=False
    for line in text.splitlines():
        if line.startswith('|'):
            if set(line.replace('|','').replace(':','').replace(' ',''))<=set('-'):continue
            if not table:result.append('<div class="table"><table>');table=True
            result.append('<tr>'+''.join('<td>'+inline(c.strip())+'</td>' for c in line.strip('|').split('|'))+'</tr>')
        else:
            if table:result.append('</table></div>');table=False
            if line.startswith('#'):
                level=min(6,len(line)-len(line.lstrip('#')));result.append(f'<h{level}>'+inline(line.lstrip('#').strip())+f'</h{level}>')
            elif line:result.append('<p>'+inline(line)+'</p>')
    if table:result.append('</table></div>')
    return ''.join(result)

def salvar_relatorio_continuidade(result,folder):
    import html
    folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    report=relatorio_final_continuidade(result);sol=_v59_number_routes(result['final_solution'])
    plan=['## Plano final de viagens','','| Veículo | Viagem | Saída (min) | Percurso | Distância (km) | Carga |','|---|---|---:|---|---:|---:|']
    visits=['## Atendimento por cliente','','| Veículo | Viagem | Cliente | Chegada | Serviço | Saída |','|---|---|---|---:|---:|---:|']
    for r in sol['routes']:
        plan.append(f"| {r['vehicle_id']} | {r['trip_id']} | {r['start_time_minutes']:.2f} | {' → '.join(map(str,r['path']))} | {r['distance_km']:.3f} | {r['total_route_load']:.2f} |")
        for v in r.get('path_details',[]):visits.append(f"| {r['vehicle_id']} | {r['trip_id']} | {v['node_id']} | {v['arrival_time_minutes']:.2f} | {v['service_start_time_minutes']:.2f} | {v['departure_time_minutes']:.2f} |")
    technical=['## Diagnóstico técnico','','| Etapa | Tempo (s) | Ganho UB | Ganho LB | Encerramento |','|---|---:|---:|---:|---|']
    for stage,row in result['statistics'].get('stages',{}).items():technical.append(f"| {stage} | {row.get('parent_wall_seconds',0):.2f} | {row.get('ub_gain',0):.8f} | {row.get('lb_gain',0):.8f} | {row.get('termination_reason','não disponível')} |")
    technical+=['','A LP do catálogo restrito de programações não é um LB global. Quando disponível, o limite das colunas é corrigido por limites válidos de pricing para todos os veículos. Ótimos de problemas locais não são certificados globais.']
    technical.extend(result.get('warnings',[]))
    technical+=['','## Registro completo para análise técnica','',_v529_detailed_report(result)]
    body=_v59_md_html(report)
    for title,lines in [('Ver plano final e atendimentos',plan+['']+visits),('Ver diagnóstico técnico',technical)]:
        body+='<details><summary>'+html.escape(title)+'</summary>'+_v59_md_html('\n'.join(lines))+'</details>'
    page='<!doctype html><html lang="pt-BR"><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><title>Resultado da continuidade</title><style>body{font:16px Arial;max-width:1050px;margin:30px auto;padding:0 20px;color:#172d3c;line-height:1.55}h1,h2{color:#14546b}table{border-collapse:collapse;width:100%;font-size:14px}td{padding:10px;border:1px solid #d4dfe3}tr:first-child{font-weight:bold;background:#edf5f6}.table{overflow:auto}details{border:1px solid #ccdce0;padding:16px;margin-top:24px}summary{cursor:pointer;font-weight:bold}strong{color:#12485b}</style><body>'+body+'</body></html>'
    (folder/'relatorio_final.md').write_text(report,encoding='utf-8')
    (folder/'relatorio_final.html').write_text(page,encoding='utf-8')
    (folder/'plano_detalhado.md').write_text('\n'.join(plan+['']+visits),encoding='utf-8')
    (folder/'diagnostico_tecnico.md').write_text('\n'.join(technical),encoding='utf-8')
    return report






_v532_round_feedback=set()

CERTIFICATION_SCALE_V591='SISTEMA_V5.9.1_A_BPLUS_B_C'














def _v591_yes(prompt):
    AvisoExecucaoV534.aguardando_continuidade()
    while True:
        try:
            answer = input(prompt).strip().lower()
        except (EOFError, KeyboardInterrupt):
            return False
        if answer in {'s', 'sim', 'y', 'yes', '1'}:
            return True
        if answer in {'', 'n', 'não', 'nao', 'no', '2'}:
            return False
        print('Resposta inválida. Digite S para sim ou N para não.')


def _v591_budget(self):
    return UserInterface.ask_continuity_budget()




OrchestratorAgent._offer_and_run_continuity=_ct_offer
OrchestratorAgent._ask_continuity_budget=_v591_budget


"""Additional necessary conditions, owned by BoundStrengtheningAgent.
All cuts admit every original feasible route solution; no incumbent restrictions.
"""
def _v593_vehicle_links(self,model,p,x,flow,assignment,tripv):
    from pyscipopt import quicksum as qs
    C=p['customers'];V=p['vehicles'];created=0;starts={};ends={}
    # Remaining load on an incoming edge cannot exceed the capacity of its owner.
    cap={i:qs(v['capacity']*assignment[i,v['idx']] for v in V) for i in C}
    for (i,j),f in flow.items():
        if j:model.addCons(f<=cap[j]);created+=1
    # Depot departures/returns and the trip counts refer to the same vehicle.
    for i in C:
        for v in V:
            vi=v['idx'];a=assignment[i,vi]
            starts[i,vi]=model.addVar(lb=0,ub=1,name=f'owned_departure_{i}_{vi}')
            ends[i,vi]=model.addVar(lb=0,ub=1,name=f'owned_return_{i}_{vi}')
            model.addCons(starts[i,vi]<=a);model.addCons(ends[i,vi]<=a);created+=2
        model.addCons(qs(starts[i,v['idx']] for v in V)==x[0,i]);created+=1
        model.addCons(qs(ends[i,v['idx']] for v in V)==x[i,0]);created+=1
    for v in V:
        vi=v['idx'];model.addCons(qs(starts[i,vi] for i in C)==tripv[vi]);created+=1
        model.addCons(qs(ends[i,vi] for i in C)==tripv[vi]);created+=1
    return dict(vehicle_link_constraints=created,owned_depot_variables=len(starts)+len(ends))

def _v593_weighted_cut(self,model,p,x,assignment,subset):
    from pyscipopt import quicksum as qs
    S=set(subset)
    incoming=qs(edge for (i,j),edge in x.items() if i not in S and j in S)
    load=qs(p['client_rows'][i]['demand']/v['capacity']*assignment[i,v['idx']] for i in S for v in p['vehicles'])
    model.addCons(incoming>=load)

def _v593_separate(self,model,p,x,assignment,existing,link_existing):
    """Return violated valid cuts from the relaxation; no model mutations here."""
    if not model.getNSols():return [],[]
    sol=model.getBestSol();value=lambda e:float(model.getSolVal(sol,e))
    xv={a:value(v) for a,v in x.items()};av={k:value(v) for k,v in assignment.items()}
    C=p['customers'];V=p['vehicles'];candidates=set()
    for anchor in C:
        S={anchor}
        while len(S)<min(24,len(C)):
            nxt=max((i for i in C if i not in S),key=lambda i:sum(xv.get((i,j),0)+xv.get((j,i),0) for j in S))
            S.add(nxt)
            if len(S) in (2,3,4,6,8,12,16,20,24):candidates.add(tuple(sorted(S)))
    violated=[]
    for subset in candidates-set(existing):
        S=set(subset);incoming=sum(v for (i,j),v in xv.items() if i not in S and j in S)
        rhs=sum(p['client_rows'][i]['demand']/v['capacity']*av[i,v['idx']] for i in S for v in V)
        if rhs-incoming>1e-6:violated.append((rhs-incoming,subset))
    # If customers are adjacent in a trip they have the same owner.
    links=[]
    for (i,j),edge in xv.items():
        if not i or not j or i>=j:continue
        together=edge+xv.get((j,i),0.)
        if together<=1e-7:continue
        violation=max(abs(av[i,v['idx']]-av[j,v['idx']])+together-1 for v in V)
        if violation>1e-6 and (i,j) not in link_existing:links.append((violation,(i,j)))
    return [s for _,s in sorted(violated,reverse=True)[:60]],[pair for _,pair in sorted(links,reverse=True)[:100]]

def _v593_add_links(self,model,p,x,assignment,pairs):
    count=0
    for i,j in pairs:
        together=x.get((i,j),0)+x.get((j,i),0)
        for v in p['vehicles']:
            delta=assignment[i,v['idx']]-assignment[j,v['idx']]
            model.addCons(delta<=1-together);model.addCons(-delta<=1-together);count+=2
    return count

BoundStrengtheningAgent._add_vehicle_links=_v593_vehicle_links
BoundStrengtheningAgent._add_weighted_capacity_cut=_v593_weighted_cut
BoundStrengtheningAgent._separate_global_cuts=_v593_separate
BoundStrengtheningAgent._add_owner_consistency=_v593_add_links
BoundStrengtheningAgent.CONTINUITY_BOUND_VERSION='V593_OWNED_CAPACITY_AND_TRIPS'


def _v593_strengthen_continuity(self,inputs,seconds,folder,state):
    return self.strengthen_lp_continuity_v515(inputs,seconds,folder,state)



BoundStrengtheningAgent.strengthen_continuity=_v593_strengthen_continuity

def _v58_compact_lb(inputs,seconds,folder,state):
    agent=BoundStrengtheningAgent(_PDSimpleNamespace(data=_pd_copy.deepcopy(inputs)))
    return agent.strengthen_continuity(inputs,seconds,folder,state)


# Continuity policy only: objective, feasibility and initial core are unchanged.
def _v511_stage_budget(stage, proposed, total, previous):
    row=previous.get('stage_performance',{}).get(stage,{})
    recent=max(0.,float(row.get('recent_gain',0.)))
    average=max(0.,float(row.get('total_gain',0.)))/max(1.,float(row.get('seconds',0.)))
    low=int(row.get('runs',0))>=2 and (recent<=1e-10 or (average>0 and recent<.15*average))
    caps={'MELHORIA':min(60.,.25*total),'LB_COMPACTO':min(150.,.50*total),
          'EXATO_LOCAL':min(120.,.45*total),'GERACAO_COLUNAS':min(45.,.15*total),
          'ELIMINAR_VEICULO':min(45.,.15*total)}
    cap=caps[stage];reason='TETO_POR_ETAPA'
    if low:
        cap=min(cap,30. if stage=='LB_COMPACTO' else 25.)
        reason='SONDAGEM_APOS_BAIXA_PRODUTIVIDADE'
        if stage=='LB_COMPACTO' and int(row.get('runs',0))%4==1:
            cap=min(caps[stage],90.);reason='REAVALIACAO_PERIODICA_DO_LB'
    return max(0.,min(float(proposed),cap)),dict(cap_seconds=cap,reason=reason,low_productivity=low)



def _v511_round_quality(stats, cert):
    seconds=max(1.,float(stats['elapsed_seconds']))
    ub_gain=max(0.,stats['before_ub']-stats['after_ub'])/max(abs(stats['before_ub']),1e-12)
    beforegap=100*max(0.,1-stats['before_lb']/max(abs(stats['before_ub']),1e-12))
    gap_gain=max(0.,beforegap-cert['gap_pct'])
    # Both an absolute materiality floor and a productivity floor must be met.
    ub_ok=ub_gain>=.001 and ub_gain/(seconds/60)>=.0002
    gap_ok=gap_gain>=.25 and gap_gain/(seconds/60)>=.05
    return dict(relevant=bool(ub_ok or gap_ok),ub_relative_gain=ub_gain,gap_gain_pp=gap_gain,
                gap_gain_pp_per_minute=gap_gain/(seconds/60),policy='V511_MATERIALITY_AND_RATE')




def _v511_process_outcome(result, exitcode, stop_reason):
    result=dict(result)
    saved=bool(result.get('completed'))
    if stop_reason or exitcode not in (0,None):
        status='RESULTADO_RECUPERADO_APOS_INTERRUPCAO' if saved else 'PROCESSO_INTERROMPIDO_SEM_RESULTADO_COMPLETO'
    elif exitcode==0 and saved:status='ENCERRAMENTO_NORMAL'
    else:status='ENCERRAMENTO_SEM_RESULTADO_COMPLETO'
    result.update(process_outcome=status,worker_exitcode=exitcode,parent_stop_reason=stop_reason,
                  worker_result_reason=result.get('termination_reason'),result_recovered=bool(saved and status!='ENCERRAMENTO_NORMAL'))
    return result


def _v511_activity(stage, result, before, after):
    if stage=='EXATO_COMPLETO_GLOBAL':return dict(nodes=result.get('nodes'),variables=result.get('variables'),constraints=result.get('constraints'),bound=result.get('canonical_lower_bound'),termination=result.get('termination_reason'))
    if stage=='EXATO_LOCAL':
        rows=result.get('neighborhoods')
        if rows is None:return dict(regions=None,sizes=None,retries=None,cooldown_skips=None,statuses=None)
        return dict(regions=len(rows),sizes={str(n):sum(r.get('released_trips')==n for r in rows) for n in (2,3,4)},
                    retries=sum(bool(r.get('retry')) for r in rows),cooldown_skips=result.get('cooldown_skips',0),
                    statuses={str(s):sum(r.get('status')==s for r in rows) for s in {r.get('status') for r in rows}})
    if stage=='GERACAO_COLUNAS':
        rounds=result.get('rounds')
        if rounds is None:return dict(cycles=None,columns=None,new_columns=None,complete_pricing_cycles=None,priced_vehicles=None)
        return dict(cycles=len(rounds),columns=result.get('columns'),new_columns=result.get('new_columns'),
                    complete_pricing_cycles=sum(bool(r.get('pricing_cycle_complete')) for r in rounds),
                    priced_vehicles=sum(int(r.get('pricing_vehicle_count',0)) for r in rounds))
    if stage=='ELIMINAR_VEICULO':
        return dict(attempts=result.get('attempts'),max_served=result.get('max_served'),
                    variants=result.get('variants_examined'),rejections=result.get('rejection_reasons'),k_status=result.get('k_status'))
    if stage=='MELHORIA' and 'portfolio_runs' in result:
        return dict(portfolio_runs=[{k:v for k,v in row.items() if k!='stats'} for row in result['portfolio_runs']],attempts=result.get('attempts'))
    if stage=='MELHORIA':
        return dict(operators={k:dict(attempts=int(v.get('attempts',0))-int(before.get('operators',{}).get(k,{}).get('attempts',0)),
                                     gain=float(v.get('gain',0))-float(before.get('operators',{}).get(k,{}).get('gain',0)))
                    for k,v in after.get('operators',{}).items()})
    return dict(phases=result.get('passes'),bound=result.get('canonical_lower_bound'),
                termination=result.get('termination_reason'))



def _v511_check_activity(stage, result, before, after, folder):
    activity=_v511_activity(stage,result,before,after)
    if result.get('structural_polish'):activity['structural_polish']={k:v for k,v in result['structural_polish'].items() if k!='stats'}
    result['activity']=activity
    _pd_write(_PDPath(folder)/'activity.json',dict(stage=stage,activity=activity,process_outcome=result.get('process_outcome')))
    _ct_technical_print(folder, 'Atividade examinada:',_v511_activity_text(stage,activity))
    return activity



def _v511_activity_text(stage, activity):
    activity={k:('não disponível' if v is None else v) for k,v in activity.items()}
    if stage.startswith('EXATO_RETORNO'):stage='EXATO_LOCAL'
    if stage=='EXATO_COMPLETO_GLOBAL':return f"Modelo completo: nós={activity.get('nodes')}; variáveis={activity.get('variables')}; restrições={activity.get('constraints')}; término={activity.get('termination')}."
    if stage=='EXATO_LOCAL':
        return f"Regiões: {activity.get('regions','não disponível')}; tamanhos: {activity.get('sizes',{})}; repetições: {activity.get('retries','não disponível')}; adiadas: {activity.get('cooldown_skips','não disponível')}."
    if stage=='GERACAO_COLUNAS':
        return f"Ciclos: {activity.get('cycles','não disponível')}; novas colunas: {activity.get('new_columns','não disponível')}; ciclos completos de precificação: {activity.get('complete_pricing_cycles','não disponível')}."
    if stage=='MELHORIA' and 'portfolio_runs' in activity:
        return 'Alternativas examinadas: '+str(len(activity['portfolio_runs']))+'; tentativas: '+str(activity.get('attempts'))+'.'
    if stage=='MELHORIA':
        return 'Tentativas por operação: '+str({k:v['attempts'] for k,v in activity.get('operators',{}).items()})+'.'
    if stage=='ELIMINAR_VEICULO':
        return f"Inserções examinadas: {activity.get('attempts','não disponível')}; máximo de clientes atendidos na reparação: {activity.get('max_served','não disponível')}."
    return f"Fases com diagnóstico: {len(activity['phases']) if isinstance(activity.get('phases'),list) else 'não disponível'}. Detalhes preservados no arquivo activity.json."




# Additional continuity operators. Initial construction and global LB formulation stay unchanged.
def _base_v512_portfolio_add_1(inputs, state, candidates, limit=6):
    scope=_ct_scope(inputs)
    if state.get('portfolio_scope_v512')!=scope:
        state['portfolio_v512']=[];state['portfolio_visits_v512']={};state['portfolio_local_v512']={}
    state['portfolio_scope_v512']=scope
    unique={}
    for candidate in list(state.get('portfolio_v512',[]))+list(candidates):
        if not isinstance(candidate,dict):continue
        audit=_pd_audit(inputs,candidate)
        if not audit.get('feasible'):continue
        sol=audit['normalized_solution'];sig=_v58_signature(_v58_routes(sol))
        if sig not in unique or sol['canonical_objective_score']<unique[sig]['canonical_objective_score']:unique[sig]=sol
    ordered=sorted(unique.values(),key=lambda s:s['canonical_objective_score'])
    chosen=[];ks=set()
    # First protect a champion for each available fleet size, including a worse K-1 witness.
    for sol in ordered:
        k=sol['summary']['num_vehicles_used']
        if k not in ks:chosen.append(sol);ks.add(k)
        if len(chosen)>=limit:break
    chosen_sigs={_v58_signature(_v58_routes(s)) for s in chosen}
    # Then retain cost/distance extremes and remaining objective elites.
    extremes=[]
    if ordered:
        for metric in ('total_cost','total_distance_km'):
            extremes.append(min(ordered,key=lambda s:s['summary'][metric]))
    for sol in extremes+ordered:
        sig=_v58_signature(_v58_routes(sol))
        if sig not in chosen_sigs and len(chosen)<limit:chosen.append(sol);chosen_sigs.add(sig)
    state['portfolio_v512']=_pd_copy.deepcopy(chosen)
    # Bound state growth; per-seed local search memory remains isolated.
    for field in ('portfolio_local_v512','portfolio_visits_v512'):
        state[field]={k:v for k,v in state.get(field,{}).items() if k in chosen_sigs}
    return chosen
_v512_portfolio_add = _base_v512_portfolio_add_1


def _v512_local_portfolio(inputs, best, seconds, folder, state):
    if not inputs.get('portfolio_mode_v512'):
        return _v58_local_search(inputs,best,seconds,folder,state)
    deadline=_pd_time.monotonic()+seconds
    pool=_v512_portfolio_add(inputs,state,[best]+state.get('feasible_pool',[]))
    visits=state.setdefault('portfolio_visits_v512',{});local=state.setdefault('portfolio_local_v512',{})
    seeds=sorted(pool,key=lambda s:(visits.get(_v58_signature(_v58_routes(s)),0),s['canonical_objective_score']))[:3]
    runs=[];winner=_pd_copy.deepcopy(best)
    for i,seed in enumerate(seeds):
        left=deadline-_pd_time.monotonic()
        if left<1:break
        sig=_v58_signature(_v58_routes(seed));budget=left/max(1,len(seeds)-i)
        branch=folder/('portfolio_'+str(i));branch.mkdir(parents=True,exist_ok=True)
        private=local.setdefault(sig,{})
        private['route_archive_scope_v536']=state.get('route_archive_scope_v536')
        private['route_archive_v536']=_pd_copy.deepcopy(state.get('route_archive_v536',[]))
        start=_pd_time.monotonic()
        candidate,stats=_v58_local_search(inputs,seed,budget,branch,private)
        state['route_archive_scope_v536']=private.get('route_archive_scope_v536')
        state['route_archive_v536']=_pd_copy.deepcopy(private.get('route_archive_v536',[]))
        check=_pd_audit(inputs,candidate)
        visits[sig]=visits.get(sig,0)+1
        row=dict(seed_signature=sig,vehicles=seed['summary']['num_vehicles_used'],before_ub=seed['canonical_objective_score'],seconds=_pd_time.monotonic()-start,stats=stats)
        if check.get('feasible'):
            candidate=check['normalized_solution'];row['after_ub']=candidate['canonical_objective_score']
            if candidate['canonical_objective_score']<winner['canonical_objective_score']-1e-9:
                winner=candidate;_pd_write(folder/'candidate.json',winner)
            new_sig=_v58_signature(_v58_routes(candidate));local[new_sig]=private;visits[new_sig]=visits[sig]
            state['portfolio_v512'].append(candidate)
        runs.append(row)
        _pd_write(folder/'search_state.json',state)
    _v512_portfolio_add(inputs,state,[winner]);_pd_write(folder/'search_state.json',state)
    return winner,dict(termination_reason='PORTFOLIO_LOCAL_FINISHED',portfolio_runs=runs,global_bound_usable=False,
        attempts=sum(r['stats'].get('attempts',0) for r in runs),improvements=sum(r['stats'].get('improvements',0) for r in runs))


def _v512_structural_finish(task,inputs,best,seconds,folder,state):
    if not inputs.get('portfolio_mode_v512'):return task(inputs,best,seconds,folder,state)
    deadline=_pd_time.monotonic()+seconds
    reserve=0. if task is _v59_exact_search else (min(15.,max(0.,seconds*.25)) if seconds>=8 else 0.)
    candidate,stats=task(inputs,best,max(.1,seconds-reserve),folder,state)
    audit=_pd_audit(inputs,candidate)
    winner=best
    if audit.get('feasible') and audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']:
        winner=audit['normalized_solution']
    witness=_pd_read(folder/'k_witness.json')
    sources=_v512_portfolio_add(inputs,state,[candidate,witness,best]+state.get('feasible_pool',[]))
    # A newly found K-1 witness gets polish even if its raw objective is worse.
    target=None
    if witness:
        checked=_pd_audit(inputs,witness)
        if checked.get('feasible'):target=checked['normalized_solution']
    if target is None and audit.get('feasible') and _v58_signature(_v58_routes(audit['normalized_solution']))!=_v58_signature(_v58_routes(best)):
        target=audit['normalized_solution']
    if target is not None and deadline-_pd_time.monotonic()>1:
        branch=folder/'structural_polish';branch.mkdir(parents=True,exist_ok=True)
        before=target['canonical_objective_score'];start=_pd_time.monotonic()
        polished,detail=_v58_local_search(inputs,target,min(15.,deadline-start),branch,{})
        checked=_pd_audit(inputs,polished)
        if checked.get('feasible'):
            polished=checked['normalized_solution'];_v512_portfolio_add(inputs,state,[polished])
            if polished['canonical_objective_score']<winner['canonical_objective_score']-1e-9:winner=polished
            stats['structural_polish']=dict(before_ub=before,after_ub=polished['canonical_objective_score'],seconds=_pd_time.monotonic()-start,stats=detail)
    _pd_write(folder/'candidate.json',winner);_pd_write(folder/'search_state.json',state)
    # Keep valid global-bound evidence from columns; polishing itself adds no LB.
    return winner,stats



````

````python
def carregar_reauditoria_v512(orchestrator, caminho, alternativas=()):
    document=_pd_read(_PDPath(caminho))
    if not isinstance(document,dict) or not isinstance(document.get('input_data'),dict):raise ValueError('Arquivo de reauditoria inválido.')
    import pandas as pd
    inputs={k:(pd.DataFrame(v['records'],columns=v['columns']) if isinstance(v,dict) and 'records' in v and 'columns' in v else v) for k,v in document['input_data'].items()}
    scope=_ct_scope(inputs)
    if document.get('problem_fingerprint')!=scope:raise ValueError('Identificação dos dados não confere; importação bloqueada.')
    candidate=(document.get('audit') or {}).get('normalized_solution') or document.get('historical_solution')
    audit=_pd_audit(inputs,candidate)
    if not audit.get('feasible'):raise ValueError('Plano importado reprovado na nova auditoria.')
    mem=orchestrator.memory.data
    existing=mem.get('best_incumbent') or mem.get('continuity_support_state_v53')
    if existing and _ct_scope(_pd_memory_inputs(orchestrator.memory))!=scope:raise ValueError('A sessão contém outro problema; use um orquestrador novo.')
    mem.update(_pd_copy.deepcopy(inputs))
    state=mem.get('continuity_support_state_v53') or {}
    if state.get('fingerprint')!=scope:state={}
    search=state.setdefault('search_state_v58',{})
    supplied=[audit['normalized_solution'],mem.get('best_incumbent'),state.get('best_solution')]+list(alternativas)
    pool=_v512_portfolio_add(inputs,search,supplied)
    best=min(pool,key=lambda s:s['canonical_objective_score'])
    # A reauditoria file carries no proof of a global LB. Never import one from its text.
    state.update(fingerprint=scope,best_solution=_pd_copy.deepcopy(best),lower_bound=state.get('lower_bound',0.))
    provider=_v58_provider(inputs);problem=provider.problem
    required_service=len(problem['customers'])*problem['service_time'];available=problem['max_time']
    k=int(best['summary']['num_vehicles_used']);target=max(0,k-1)
    if available>0 and required_service>target*available+1e-7:
        state.update(target_k=target,k_status='PROVEN_INFEASIBLE',vehicle_service_proof_v512=dict(fingerprint=scope,target_k=target,required_service_minutes=required_service,available_service_minutes=target*available))
        print(f'Contagem mínima de veículos comprovada: {k}; o serviço obrigatório excede a jornada de {target} veículo(s).')
    mem.update(best_incumbent=_pd_copy.deepcopy(best),solver_result=_pd_copy.deepcopy(best),continuity_support_state_v53=state)
    print(f"Reauditoria carregada: {len(pool)} alternativa(s), UB {best['canonical_objective_score']:.10f}. Nenhum LB importado do arquivo.")
    return audit




def _v512_portfolio_report(result):
    stages=result.get('statistics',{}).get('stages',{})
    runs=stages.get('MELHORIA',{}).get('portfolio_runs',[])
    text=''
    if runs:
        text+='\n## Soluções alternativas examinadas\n\n| Veículos | UB de partida | UB após acabamento | Tempo (s) |\n|---|---:|---:|---:|\n'
        for row in runs:
            after=row.get('after_ub')
            value=f"{after:.8f}" if after is not None else 'não aprovado'
            text+=f"| {row['vehicles']} | {row['before_ub']:.8f} | {value} | {row['seconds']:.1f} |\n"
        text+='\nO plano entregue é o de menor objetivo auditado. Melhorar uma alternativa não significa superar o incumbent.\n'
    for name,stage in stages.items():
        row=stage.get('structural_polish')
        if row:text+=f"\nAcabamento após {name}: UB da candidata {row['before_ub']:.8f} → {row['after_ub']:.8f}, em {row['seconds']:.1f}s, dentro do orçamento da etapa.\n"
    return text


def _base_v512_recover_candidates_1(inputs, folder, state):
    paths=[folder/'candidate.json']+list(folder.glob('portfolio_*/candidate.json'))+list(folder.glob('structural_polish/candidate.json'))
    candidates=[]
    for path in paths:
        data=_pd_read(path)
        if data:
            audit=_pd_audit(inputs,data)
            if audit.get('feasible'):candidates.append(audit['normalized_solution'])
    _v512_portfolio_add(inputs,state,candidates)
    return min(candidates,key=lambda s:s['canonical_objective_score']) if candidates else None
_v512_recover_candidates = _base_v512_recover_candidates_1


_V513_BINDINGS = (OrchestratorAgent, _initial_audit_and_bounds, _ct_offer, _ct_orchestrate_with_full_report, executar_continuidade_integrada, _ct_worker, _v58_compact_lb, _v59_exact_search, _v59_columns, relatorio_final_continuidade, BoundStrengtheningAgent.strengthen_continuity, InitialSolutionRefinerAgent.refine, VehicleProgramColumnAgent.optimize, NeighborhoodOptimizationAgent.improve_incumbent, NeighborhoodOptimizationAgent.optimize_region, KVehicleFeasibilityCertifier.attempt_elimination, _v58_allocation, _v511_stage_budget, _v511_round_quality, _ct_run_process, _v59_neighborhood_choices, _v512_local_portfolio, _v512_structural_finish, _v512_portfolio_add)
def validar_integracao_v513(instance):
    expected=_V513_BINDINGS
    actual=(type(instance),getattr(instance._run_optimality_certification,'__func__',None),getattr(instance._offer_and_run_continuity,'__func__',None),getattr(instance.orchestrate_solution_process,'__func__',None),executar_continuidade_integrada,_ct_worker,_v58_compact_lb,_v59_exact_search,_v59_columns,relatorio_final_continuidade,BoundStrengtheningAgent.strengthen_continuity, InitialSolutionRefinerAgent.refine, VehicleProgramColumnAgent.optimize, NeighborhoodOptimizationAgent.improve_incumbent, NeighborhoodOptimizationAgent.optimize_region, KVehicleFeasibilityCertifier.attempt_elimination, _v58_allocation, _v511_stage_budget, _v511_round_quality, _ct_run_process, _v59_neighborhood_choices, _v512_local_portfolio, _v512_structural_finish, _v512_portfolio_add)
    if OrchestratorAgent is not expected[0] or any(a is not b for a,b in zip(actual,expected)):
        raise RuntimeError('EXECUCAO BLOQUEADA: integracao V5.13 sobrescrita. Execute o arquivo completo em ambiente novo e na ordem original.')
    if not isinstance(instance.agents.get('initial_refiner'),InitialSolutionRefinerAgent) or not isinstance(instance.agents.get('vehicle_program_columns'),VehicleProgramColumnAgent):
        raise RuntimeError('Novos agentes ausentes no orquestrador.')
    print('INTEGRACAO CONFIRMADA — melhoria, LP primeiro, exato local, colunas opcionais e eliminacao de veiculo. ZIP automatico desativado.')
    return True

def carregar_estado_continuidade_v513(orchestrator,caminho):
    import pandas as pd
    document=_pd_read(_PDPath(caminho))
    if not isinstance(document,dict):raise ValueError('Arquivo inválido.')
    raw=document.get('input_data')
    if not isinstance(raw,dict):raise ValueError('Dados de entrada ausentes.')
    inputs={k:(pd.DataFrame(v['records'],columns=v['columns']) if isinstance(v,dict) and 'records' in v and 'columns' in v else v) for k,v in raw.items()}
    scope=_ct_scope(inputs)
    if document.get('fingerprint')!=scope:raise ValueError('Identificação dos dados não confere.')
    state=_pd_copy.deepcopy(document.get('state') or {})
    result=document.get('result') or {};cert=result.get('certificate') or {}
    if state.get('fingerprint')!=scope:raise ValueError('Estado pertence a outro problema.')
    check=_pd_audit(inputs,state.get('best_solution'))
    if not check.get('feasible'):raise ValueError('Plano salvo reprovado na auditoria.')
    solution=check['normalized_solution'];ub=solution['canonical_objective_score']
    lb=float(state.get('lower_bound',0.))
    if not _pd_math.isfinite(lb) or lb<0 or lb>ub+1e-7:raise ValueError('LB incompatível com plano auditado.')
    if lb>0 and (cert.get('problem_fingerprint_v53')!=scope or not cert.get('lower_bound_certified') or abs(float(cert.get('lower_bound',-1))-lb)>1e-8):
        raise ValueError('Certificado do LB salvo ausente ou inconsistente.')
    mem=orchestrator.memory.data
    if mem.get('best_incumbent') or mem.get('continuity_support_state_v53'):
        raise ValueError('Use um orquestrador novo para restaurar o checkpoint sem substituir uma sessão ativa.')
    state['best_solution']=solution
    mem.update(_pd_copy.deepcopy(inputs))
    mem['continuity_integrated_result_v53']=_pd_copy.deepcopy(result)
    mem['continuity_options_v514']=_pd_copy.deepcopy(document.get('options') or {})
    mem.update(best_incumbent=solution,solver_result=solution,continuity_support_state_v53=state,continuity_certificate=cert)
    print(f'Checkpoint restaurado e plano reauditado: UB {ub:.10f}; LB registrado {lb:.10f}. Somente o histórico presente no arquivo foi restaurado.')
    print('O LB é retomado do certificado salvo do sistema; esta importação não resolve novamente o modelo que o produziu.')
    return check




def _v514_analytical_floor(self,p,coeff):
    """Global floors from a spanning tree and metric capacitated depot travel."""
    import math
    base=_v58_cheap_bounds(p,coeff);C=p['customers'];V=p['vehicles']
    if not C:return base
    remaining=set(C);best={i:min(p['distance'][0,i],p['distance'][i,0]) for i in C};tree=0.
    while remaining:
        i=min(remaining,key=lambda j:best[j]);tree+=best[i];remaining.remove(i)
        for j in remaining:best[j]=min(best[j],p['distance'][i,j],p['distance'][j,i])
    Q=max(v['capacity'] for v in V)
    radial=sum(p['client_rows'][i]['demand']*(p['distance'][0,i]+p['distance'][i,0]) for i in C)/Q
    distance=max(base['distance_lb'],tree,radial,max(p['distance'][0,i]+p['distance'][i,0] for i in C))
    distance=max(0.,distance-1e-7*max(1.,distance))
    duration=len(C)*p['service_time']+distance*60/p['speed']
    count=max(base['vehicles_lb'],math.ceil(duration/p['max_time']-1e-10))
    if not p['allow_multi_trip']:
        cumulative=0.;needed=0;demand=sum(p['client_rows'][i]['demand'] for i in C)
        for capacity in sorted((v['capacity'] for v in V),reverse=True):
            cumulative+=capacity;needed+=1
            if cumulative>=demand-1e-8:break
        count=max(count,needed)
    duration=max(duration,count*p['minimum_usage'])
    fixed=sorted(v['fixed_cost'] for v in V)
    cost=(sum(fixed) if p['fixed_cost_mode']!='WHEN_USED' else sum(fixed[:count]))+min(v['cost_per_km'] for v in V)*distance
    base.update(distance_lb=distance,time_lb=duration,vehicles_lb=count,cost_lb=cost)
    base['canonical_objective_lb']=sum(coeff.get(k,0)*base[f] for k,f in [('DISTANCE','distance_lb'),('TIME','time_lb'),('COST','cost_lb'),('VEHICLES','vehicles_lb')])
    base['floor_diagnostics_v514']=dict(spanning_tree=tree,radial_capacity=radial)
    return base

BoundStrengtheningAgent._analytical_floor_v514=_v514_analytical_floor

def _v514_temporal_relaxation(self,model,p,x,z,assignment,tripv,work):
    # With full-horizon windows, test a smaller valid relaxation; no feasible
    # solution is removed. This is a performance choice, not an equivalence claim.
    broad=all(p['windows'][i][0]<=p['operation_start'] and p['windows'][i][1]>=p['operation_end'] for i in p['customers'])
    if not broad:return _v59_temporal_cuts(model,p,x,z,assignment,tripv,work)
    from pyscipopt import quicksum as qs
    import math
    count=0
    for v in p['vehicles']:
        vi=v['idx'];eligible=[i for i in p['customers'] if p['client_rows'][i]['demand']<=v['capacity']+1e-8]
        if not eligible:continue
        travel=min(p['travel_time'][0,i]+p['travel_time'][i,0] for i in eligible)
        horizon=min(p['max_time'],p['operation_end']-p['operation_start']);s=p['service_time']
        if s+travel>1e-8:model.addCons(tripv[vi]<=math.floor(horizon/(s+travel)+1e-9)*z[vi]);count+=1
        model.addCons(tripv[vi]<=qs(assignment[i,vi] for i in p['customers']));count+=1
        model.addCons(work[vi]>=s*qs(assignment[i,vi] for i in p['customers'])+travel*tripv[vi]);count+=1
    return dict(temporal_constraints=count,intervals=0,incompatible_pairs=0,trip_cliques=0,mode='FULL_HORIZON_COMPACT_RELAXATION')

BoundStrengtheningAgent._temporal_relaxation_v514=_v514_temporal_relaxation

def _v514_vehicle_trip_cliques(self,model,p,assignment,tripv,inputs):
    """A capacity-incompatible clique needs distinct trips, not distinct vehicles."""
    if not inputs.get('enable_trip_cliques_v514',False) or not model.getNSols():return 0
    from pyscipopt import quicksum
    sol=model.getBestSol();cuts=[]
    for v in p['vehicles']:
        vi=v['idx'];eligible=[i for i in p['customers'] if p['client_rows'][i]['demand']<=v['capacity']]
        group=[]
        for i in sorted(eligible,key=lambda i:-p['client_rows'][i]['demand']):
            if all(p['client_rows'][i]['demand']+p['client_rows'][j]['demand']>v['capacity']+1e-8 for j in group):group.append(i)
        if len(group)>1 and sum(model.getSolVal(sol,assignment[i,vi]) for i in group)>model.getSolVal(sol,tripv[vi])+1e-6:
            cuts.append((vi,group))
    if cuts:
        model.freeTransform()
        for vi,group in cuts:model.addCons(quicksum(assignment[i,vi] for i in group)<=tripv[vi])
    return len(cuts)

BoundStrengtheningAgent._add_vehicle_trip_cliques_v514=_v514_vehicle_trip_cliques
BoundStrengtheningAgent.CONTINUITY_BOUND_VERSION='V514_LP_FIRST_SELECTIVE'

def configurar_continuidade_v514(orchestrator,colunas=False,cortes_viagens=False,fracao_lp=.70):
    if not .2<=float(fracao_lp)<=1:raise ValueError('A fração da LP deve estar entre 0,2 e 1.')
    orchestrator.memory.data['continuity_options_v514']=dict(enable_columns_v514=bool(colunas),enable_trip_cliques_v514=bool(cortes_viagens),lb_lp_fraction_v514=float(fracao_lp))
    return dict(orchestrator.memory.data['continuity_options_v514'])

# The previous integrity guard holds identities for the newly loaded definitions.
# No core method is replaced here.
_V514_BOUND=BoundStrengtheningAgent.strengthen_continuity
def validar_integracao_v514(orchestrator):
    validar_integracao_v513(orchestrator)
    if BoundStrengtheningAgent.strengthen_continuity is not _V514_BOUND:raise RuntimeError('BoundStrengtheningAgent V5.14 substituído.')
    print('núcleo preservado; continuidade com memória de LB e seleção de tarefas por resultado.')
    return True

def salvar_checkpoint_v514(orchestrator,caminho='checkpoint_continuidade_v534.json',download=True):
    inputs=_pd_memory_inputs(orchestrator.memory);mem=orchestrator.memory.data
    scope=_ct_scope(inputs)
    state=_pd_copy.deepcopy(mem.get('continuity_support_state_v53'))
    result=_pd_copy.deepcopy(mem.get('continuity_integrated_result_v53'))
    if not state:
        solution=mem.get('final_solution') or mem.get('solver_result') or mem.get('best_incumbent')
        if not isinstance(solution,dict):raise ValueError('Não há plano para guardar.')
        report=_pd_audit(inputs,solution)
        if not report.get('feasible'):raise ValueError('Plano inicial não passou na auditoria.')
        cert=_pd_copy.deepcopy(solution.get('certificate') or mem.get('optimality_certificate') or mem.get('certification_result',{}).get('certificate') or {})
        if not cert.get('certification_finalized') or not cert.get('post_certification_independent_audit_passed'):
            raise ValueError('Aguarde a auditoria e a finalização do certificado antes de guardar.')
        ub=float(report['recomputed']['canonical_objective_score']);lb=cert.get('lower_bound')
        if cert.get('problem_fingerprint_v53')!=scope:
            raise ValueError('Certificado pertence a outro problema.')
        if not _pd_math.isclose(float(cert.get('upper_bound',-1)),ub,abs_tol=1e-7):
            raise ValueError('Certificado não corresponde ao plano auditado.')
        if lb is None or not cert.get('lower_bound_certified'):lb=0.
        lb=float(lb)
        if not _pd_math.isfinite(lb) or lb<0 or lb>ub+1e-7:raise ValueError('Limite inferior inválido.')
        state=dict(fingerprint=scope,best_solution=report['normalized_solution'],lower_bound=lb,
            round=0,search_state_v58={},stage_performance={},k_status='UNKNOWN',
            target_k=max(0,report['recomputed']['num_vehicles_used']-1),round_history_v591=[])
        result=dict(status='INITIAL_CHECKPOINT',final_solution=_pd_copy.deepcopy(solution),certificate=cert,
            statistics={},issues=[],warnings=[],round_history=[],checkpoint_origin='INITIAL_CORE')
    report=_pd_audit(inputs,state.get('best_solution'))
    if not report.get('feasible'):raise ValueError('Plano salvo não passou na auditoria.')
    document=dict(version='V5.24',revision='V5.34',fingerprint=scope,input_data=inputs,state=state,
        result=result,options=mem.get('continuity_options_v514',{}))
    path=_PDPath(caminho);_pd_write(path,document)
    if download:
        try:
            from google.colab import files
            files.download(str(path))
        except ImportError:print('Arquivo salvo localmente; download pelo Colab não disponível neste ambiente.')
    print('Checkpoint salvo:',path)
    return str(path)


def _v514_aggregate_bound(self,p,coeff,seconds):
    """Independent global relaxation: shared maximum capacity, cheapest km rate.
    Actual routes embed in the arc/flow model; these simplifications lower costs.
    The fleet activation variables retain real fixed costs. No incumbent fixes.
    """
    import time,math
    from pyscipopt import Model,quicksum as qs
    started=time.monotonic();b=self._analytical_floor_v514(p,coeff)
    C=p['customers'];N=[0]+C;V=p['vehicles'];Q=max(v['capacity'] for v in V)
    A=[(i,j) for i in N for j in N if i!=j and (not i or not j or p['client_rows'][i]['demand']+p['client_rows'][j]['demand']<=Q+1e-8)]
    m=Model('V514_AGGREGATE_GLOBAL_FLOOR');m.hideOutput();m.setParam('limits/memory',1024.)
    x={a:m.addVar(lb=0,ub=1) for a in A};f={a:m.addVar(lb=0,ub=Q if a[1] else 0) for a in A}
    incoming={i:[a for a in A if a[1]==i] for i in N};outgoing={i:[a for a in A if a[0]==i] for i in N}
    for a in A:m.addCons(f[a]<=Q*x[a])
    for i in C:
        m.addCons(qs(x[a] for a in incoming[i])==1);m.addCons(qs(x[a] for a in outgoing[i])==1)
        m.addCons(qs(f[a] for a in incoming[i])-qs(f[a] for a in outgoing[i])==p['client_rows'][i]['demand'])
    z={v['idx']:m.addVar(lb=1 if p['must_use_all_vehicles'] else 0,ub=1) for v in V}
    D=qs(p['distance'][a]*x[a] for a in A);K=qs(z.values());trips=qs(x[a] for a in outgoing[0])
    m.addCons(trips==qs(x[a] for a in incoming[0]));m.addCons(trips>=K);m.addCons(trips>=b['trips_lb'])
    m.addCons(K>=b['vehicles_lb']);m.addCons(D>=b['distance_lb'])
    T=m.addVar(lb=0);m.addCons(T>=len(C)*p['service_time']+D*60/p['speed'])
    m.addCons(T>=p['minimum_usage']*K);m.addCons(T<=p['max_time']*K)
    if not p['allow_multi_trip']:
        m.addCons(trips==K);m.addCons(qs(v['capacity']*z[v['idx']] for v in V)>=sum(p['client_rows'][i]['demand'] for i in C))
    sets=set()
    for anchor in C:
        order=sorted(C,key=lambda j:p['distance'][anchor,j])
        for size in (4,8,12,20):
            if size<len(C):sets.add(tuple(sorted(order[:size])))
    for subset in sets:
        S=set(subset);rhs=math.ceil(sum(p['client_rows'][i]['demand'] for i in S)/Q-1e-10)
        m.addCons(qs(x[a] for a in A if a[0] not in S and a[1] in S)>=rhs)
    fixed=qs(v['fixed_cost']*z[v['idx']] for v in V) if p['fixed_cost_mode']=='WHEN_USED' else sum(v['fixed_cost'] for v in V)
    cost=fixed+min(v['cost_per_km'] for v in V)*D
    expr={'DISTANCE':D,'COST':cost,'TIME':T,'VEHICLES':K}
    m.setObjective(qs(coeff.get(k,0)*e for k,e in expr.items()),'minimize')
    best=b['canonical_objective_lb'];passes=[]
    try:
        for phase in ('AGGREGATE_LP','AGGREGATE_ACTIVATION'):
            remaining=seconds-(time.monotonic()-started)
            if remaining<=.2:break
            if phase=='AGGREGATE_ACTIVATION':
                if passes[-1]['status']!='optimal':break
                if m.getNSols() and all(abs(m.getSolVal(m.getBestSol(),v)-round(m.getSolVal(m.getBestSol(),v)))<1e-7 for v in z.values()):break
                m.freeTransform()
                for v in z.values():m.chgVarType(v,'B')
            m.setParam('limits/time',m.getSolvingTime()+remaining*(.75 if phase=='AGGREGATE_LP' else 1.))
            m.optimize();d=float(m.getDualbound());valid=math.isfinite(d) and abs(d)<1e19 and str(m.getStatus())!='infeasible'
            if valid:best=max(best,d-1e-7*max(1.,abs(d)))
            passes.append(dict(phase=phase,status=str(m.getStatus()),dual_bound=d if valid else None,seconds=time.monotonic()-started))
        return dict(canonical_lower_bound=best,passes=passes,seconds=time.monotonic()-started,method='AGGREGATE_GLOBAL_RELAXATION')
    finally:m.freeProb()

BoundStrengtheningAgent._aggregate_bound_v514=_v514_aggregate_bound


def _v5141_build_reference(self,inputs):
    import time,math
    from pyscipopt import Model,quicksum as qs
    started=time.monotonic();provider=_v58_provider(inputs)
    p=provider.problem;coeff=provider.objective_context['coefficients'];C=p['customers'];V=p['vehicles']
    if any(float(c)<0 for c in coeff.values()) or any(v[k]<0 for v in V for k in ('cost_per_km','fixed_cost')):
        raise ValueError('Coeficientes negativos não suportados nesta prova.')
    b=self._analytical_floor_v514(p,coeff);Q=max(v['capacity'] for v in V);H=p['max_time'];svc=p['service_time'];speed=p['speed'];N=[0]+C
    # Fixed reference: no persistent cuts and no incumbent route restrictions.
    persisted_subsets=set();persisted_links=set()
    lower={i:max(p['operation_start'],p['windows'][i][0]) for i in C}
    upper={i:min(p['operation_end'],p['windows'][i][1]) for i in C}
    arcs=[(i,j) for i in N for j in N if i!=j and (i==0 or j==0 or
          (lower[i]+svc+p['travel_time'][i,j]<=upper[j]+1e-8 and p['client_rows'][i]['demand']+p['client_rows'][j]['demand']<=Q+1e-8))]
    model=Model('COMPACT_GLOBAL_RELAXATION_V58');model.hideOutput();model.setParam('limits/gap',0.)
    model.setParam('limits/memory',_v534_memory_budget(inputs)['solver_mb'])
    model.setParam('randomization/randomseedshift',int(inputs.get('continuity_seed_v514',0)))
    x={a:model.addVar(lb=0,ub=1,name=f'arc_{a}') for a in arcs}
    flow={a:model.addVar(lb=0,ub=Q if a[1]!=0 else 0,name=f'flow_{a}') for a in arcs}
    z={v['idx']:model.addVar(lb=0,ub=1,name=f'active_{v["idx"]}') for v in V}
    incoming={i:[a for a in arcs if a[1]==i] for i in N};outgoing={i:[a for a in arcs if a[0]==i] for i in N}
    for a in arcs:model.addCons(flow[a]<=Q*x[a])
    D=qs(p['distance'][a]*x[a] for a in arcs);work={};distv={};assignment={};split={};tripv={}
    for i in C:
        model.addCons(qs(x[a] for a in incoming[i])==1);model.addCons(qs(x[a] for a in outgoing[i])==1)
        model.addCons(qs(flow[a] for a in incoming[i])-qs(flow[a] for a in outgoing[i])==p['client_rows'][i]['demand'])
        di=qs(p['distance'][a]*x[a] for a in incoming[i])+p['distance'][i,0]*x[i,0]
        Mi=max(p['distance'][a] for a in incoming[i])+p['distance'][i,0]
        for v in V:
            vi=v['idx'];possible=p['client_rows'][i]['demand']<=v['capacity']+1e-8
            assignment[i,vi]=model.addVar(lb=0,ub=1 if possible else 0)
            split[i,vi]=model.addVar(lb=0,ub=Mi)
            model.addCons(assignment[i,vi]<=z[vi]);model.addCons(split[i,vi]<=Mi*assignment[i,vi])
        model.addCons(qs(assignment[i,v['idx']] for v in V)==1)
        model.addCons(qs(split[i,v['idx']] for v in V)==di)
    for v in V:
        vi=v['idx'];distv[vi]=qs(split[i,vi] for i in C);work[vi]=model.addVar(lb=0,ub=H)
        tripv[vi]=model.addVar(lb=0,ub=len(C))
        model.addCons(tripv[vi]<=len(C)*z[vi]);model.addCons(tripv[vi]>=z[vi])
        model.addCons(qs(p['client_rows'][i]['demand']*assignment[i,vi] for i in C)<=v['capacity']*tripv[vi])
        # Metric radial capacity bound, valid separately for each physical vehicle's trips.
        model.addCons(distv[vi]>=qs((p['distance'][0,i]+p['distance'][i,0])*p['client_rows'][i]['demand']/v['capacity']*assignment[i,vi] for i in C))
        minimum_trip_distance=min(p['distance'][0,i]+p['distance'][i,0] for i in C)
        model.addCons(distv[vi]>=minimum_trip_distance*tripv[vi])
        model.addCons(work[vi]>=60/speed*distv[vi]+svc*qs(assignment[i,vi] for i in C))
        model.addCons(work[vi]<=min(H,p['operation_end']-p['operation_start'])*z[vi]);model.addCons(work[vi]>=p['minimum_usage']*z[vi])
        if not p['allow_multi_trip']:model.addCons(qs(p['client_rows'][i]['demand']*assignment[i,vi] for i in C)<=v['capacity']*z[vi])
        if p['must_use_all_vehicles']:model.addCons(z[vi]==1)
    linked=self._add_vehicle_links(model,p,x,flow,assignment,tripv)
    count=qs(z.values());T=qs(work.values());trips=qs(x[a] for a in outgoing[0])
    model.addCons(trips==qs(x[a] for a in incoming[0]));model.addCons(trips>=b['trips_lb']);model.addCons(trips>=count)
    model.addCons(trips==qs(tripv.values()))
    if not p['allow_multi_trip']:model.addCons(trips==count)
    model.addCons(count>=b['vehicles_lb']);model.addCons(D>=b['distance_lb']);model.addCons(T>=b['time_lb'])
    fixed=qs(v['fixed_cost']*z[v['idx']] for v in V) if p['fixed_cost_mode']=='WHEN_USED' else sum(v['fixed_cost'] for v in V)
    variable=qs(v['cost_per_km']*distv[v['idx']] for v in V);cost=fixed+variable
    model.addCons(cost>=b['cost_lb'])
    expr={'DISTANCE':D,'TIME':T,'COST':cost,'VEHICLES':count}
    model.setObjective(qs(coeff.get(k,0)*e for k,e in expr.items()),'minimize')
    try:
        # Capacity cuts use all possible edges of the relaxation, never incumbent routes.
        subsets=set()
        for anchor in C:
            ordered=sorted(C,key=lambda i:p['distance'][anchor,i])
            for size in (4,8,12,20,30):
                if size<len(C):subsets.add(tuple(sorted(ordered[:size])))
        ordered=sorted(C,key=lambda i:lower[i])
        for size in (5,10,15,20):
            for start in range(0,len(C)-size+1,max(1,size//2)):subsets.add(tuple(sorted(ordered[start:start+size])))
        subsets.update(persisted_subsets)
        for subset in sorted(subsets):
            S=set(subset);needed=math.ceil(sum(p['client_rows'][i]['demand'] for i in S)/Q-1e-10)
            model.addCons(qs(x[a] for a in arcs if a[0] not in S and a[1] in S)>=needed)
            self._add_weighted_capacity_cut(model,p,x,assignment,subset)
        link_count=self._add_owner_consistency(model,p,x,assignment,persisted_links)
        temporal=self._temporal_relaxation_v514(model,p,x,z,assignment,tripv,work)
        build=time.monotonic()-started
        return dict(model=model,p=p,x=x,assignment=assignment,z=z,tripv=tripv,subsets=subsets,links=persisted_links,
                    analytical=b,build_seconds=time.monotonic()-started,temporal=temporal)
    except BaseException:
        model.freeProb();raise

BoundStrengtheningAgent._build_reference_lp_v5141=_v5141_build_reference

def _v5141_matrix(self,model):
    """Export original linear rows, before any SCIP transformation or solve."""
    import numpy as np,hashlib
    from collections import Counter
    from scipy.sparse import coo_matrix
    vs=model.getVars(transformed=False);names=[v.name for v in vs];index={n:i for i,n in enumerate(names)}
    types=dict(Counter(str(v.vtype()) for v in vs))
    if any(str(v.vtype())!='CONTINUOUS' for v in vs):raise ValueError('A referência contém variáveis não contínuas.')
    if model.getObjectiveSense()!='minimize':raise ValueError('Somente minimização suportada.')
    lb=np.array([v.getLbOriginal() for v in vs]);ub=np.array([v.getUbOriginal() for v in vs])
    # All variables of this formulation have finite bounds. Fail closed otherwise.
    if not np.all(np.isfinite(lb)) or not np.all(np.isfinite(ub)) or np.max(np.abs(np.r_[lb,ub]))>=1e19:
        raise ValueError('Limites finitos necessários para a verificação dual conservadora.')
    cc=np.array([v.getObj() for v in vs]);offset=float(model.getObjoffset(original=True))
    rows=[[],[]];cols=[[],[]];vals=[[],[]];rhs=[[],[]]
    for cons in model.getConss(transformed=False):
        if cons.getConshdlrName()!='linear':raise ValueError('Restrição não linear inesperada.')
        coefficients=model.getValsLinear(cons);left=float(model.getLhs(cons));right=float(model.getRhs(cons))
        targets=[(1,right,1.)] if left==right else ([(0,right,1.)] if right<1e19 else [])+([(0,-left,-1.)] if left>-1e19 else [])
        for kind,b,sign in targets:
            row=len(rhs[kind]);rhs[kind].append(b)
            for name,value in coefficients.items():rows[kind].append(row);cols[kind].append(index[name]);vals[kind].append(sign*value)
    matrices=[coo_matrix((vals[k],(rows[k],cols[k])),shape=(len(rhs[k]),len(vs))).tocsr() for k in (0,1)]
    h=hashlib.sha256()
    for a in [cc,lb,ub,np.array([offset])]+[np.asarray(v,dtype=float) for v in rhs]:h.update(a.tobytes())
    for m in matrices:
        for a in (m.data,m.indices,m.indptr):h.update(a.tobytes())
    return dict(c=cc,A_ub=matrices[0],b_ub=np.asarray(rhs[0]),A_eq=matrices[1],b_eq=np.asarray(rhs[1]),
                lb=lb,ub=ub,offset=offset,names=names,index=index,variable_types=types,matrix_hash=h.hexdigest())


def _v5141_solve(self,data,seconds,label,folder,baseline_lb):
    import time,math,numpy as np,pyscipopt
    from pyscipopt import LP,SCIP_LPPARAM
    from types import SimpleNamespace
    t=time.monotonic();n=len(data['c'])
    row=dict(phase=label,engine='PySCIPOpt.LP/direct_LPI_dual_simplex',pyscipopt_version=pyscipopt.__version__,model_kind='CONTINUOUS_LP',
             original_variable_types=data['variable_types'],submitted_variable_types={'CONTINUOUS':n},
             integer_variables=0,binary_variables=0,matrix_hash=data['matrix_hash'],variables=n,
             equalities=len(data['b_eq']),inequalities=len(data['b_ub']),time_budget_seconds=seconds,
             nodes=None,nodes_note='Não aplicável: resolução LP, sem busca inteira.',bound_before=baseline_lb)
    _pd_write(folder/'lp_phase.json',dict(row,status='RUNNING',updated_at=time.time()))
    print(f"{label}: {n} variáveis contínuas, 0 inteiras; até {seconds:.1f}s.",flush=True)
    lp=LP(name=label,sense='minimize')
    lp.setIntParam(SCIP_LPPARAM.LPINFO,0)
    lp.setRealParam(SCIP_LPPARAM.FEASTOL,1e-7);lp.setRealParam(SCIP_LPPARAM.DUALFEASTOL,1e-7)
    lp.addCols([[] for _ in range(n)],objs=data['c'].tolist(),lbs=data['lb'].tolist(),ubs=data['ub'].tolist())
    for a,b,equality in ((data['A_ub'],data['b_ub'],False),(data['A_eq'],data['b_eq'],True)):
        entries=[list(zip(a.indices[a.indptr[i]:a.indptr[i+1]].tolist(),a.data[a.indptr[i]:a.indptr[i+1]].tolist())) for i in range(a.shape[0])]
        lp.addRows(entries,lhss=b.tolist() if equality else [-lp.infinity()]*len(b),rhss=b.tolist())
    warm=data.get('_warmstart_basis')
    if warm is not None:lp.setBase(*warm)
    row.update(warm_start_from_reference=warm is not None,solver_load_seconds=time.monotonic()-t)
    solve_start=time.monotonic();lp.setRealParam(SCIP_LPPARAM.LPTILIM,max(.01,float(seconds)));lp.solve(dual=True)
    optimal=lp.isOptimal();status=0 if optimal else 1
    if optimal:data['_completed_basis']=lp.getBase()
    duals=np.asarray(lp.getDual()) if optimal else None
    answer=SimpleNamespace(success=optimal,status=status,message='OPTIMAL' if optimal else 'NOT_OPTIMAL_WITHIN_BUDGET',
        nit=lp.getNIterations(),x=np.asarray(lp.getPrimal()) if optimal else None,
        ineqlin=SimpleNamespace(marginals=duals[:len(data['b_ub'])] if optimal else None),
        eqlin=SimpleNamespace(marginals=duals[len(data['b_ub']):] if optimal else None))
    row['solve_seconds']=time.monotonic()-solve_start
    row.update(seconds=time.monotonic()-t,status={0:'OPTIMAL',1:'NOT_OPTIMAL',2:'INFEASIBLE',3:'UNBOUNDED',4:'ERROR'}.get(answer.status,'UNKNOWN'),
               message=str(answer.message),iterations=int(answer.nit or 0),certified_lb=None,usable=False)
    vector=None
    if answer.success and answer.status==0:
        vector=np.asarray(answer.x);yu=np.minimum(np.asarray(answer.ineqlin.marginals),0.);ye=np.asarray(answer.eqlin.marginals)
        residual=data['c']-data['A_ub'].T@yu-data['A_eq'].T@ye
        support=np.where(residual>=0,residual*data['lb'],residual*data['ub'])
        dual=float(data['b_ub']@yu+data['b_eq']@ye+np.sum(support)+data['offset'])
        primal=float(data['c']@vector+data['offset'])
        primal_error=max(float(np.max(data['A_ub']@vector-data['b_ub'],initial=0)),
                         float(np.max(np.abs(data['A_eq']@vector-data['b_eq']),initial=0)),
                         float(np.max(data['lb']-vector,initial=0)),float(np.max(vector-data['ub'],initial=0)))
        # Dual certificate: b*y + min_{lb<=x<=ub} (c-A'y)*x. No primal objective used as LB.
        magnitude=float(np.abs(data['b_ub'])@np.abs(yu)+np.abs(data['b_eq'])@np.abs(ye)+np.sum(np.abs(support))+abs(data['offset']))
        margin=1e-7*max(1.,abs(dual))+1e-12*max(1.,magnitude)
        safe=dual-margin
        row.update(primal_objective=primal,dual_objective=dual,dual_margin=margin,primal_max_violation=primal_error,
                   duality_gap=primal-dual,dual_verification='BOX_CORRECTED_DUAL',fractional_variables=int(np.sum(np.abs(vector-np.round(vector))>1e-6)))
        if math.isfinite(safe) and primal_error<=1e-5 and dual<=primal+1e-7:
            row.update(certified_lb=max(0.,safe),usable=True)
        else:row['status']='NUMERICAL_CHECK_FAILED';vector=None
    row['bound_after']=max(baseline_lb,row['certified_lb']) if row['usable'] else baseline_lb
    row['bound_gain']=row['bound_after']-baseline_lb;row['gain_per_second']=row['bound_gain']/max(row['seconds'],1e-9)
    row['meaningful_gain']=row['bound_gain']>1e-7
    _pd_write(folder/'lp_phase.json',row)
    return row,vector



BoundStrengtheningAgent._matrix_lp_v5141=_v5141_matrix
BoundStrengtheningAgent._solve_lp_v5141=_v5141_solve

def _v515_strengthen(self,inputs,seconds,folder,state):
    """Continuity only. Pure LP families, bounded optional legacy completion."""
    import time,math,copy,numpy as np
    from pyscipopt import SCIP_BASESTAT
    started=time.monotonic();deadline=started+max(0.,float(seconds));folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    scope=_ct_scope(inputs);model=None;passes=[]
    settings=dict(version='V515_LPI_FAMILIES_2',matrix='V5141_REFERENCE',engine='LPI',tolerance=1e-7,
                  fallback=bool(inputs.get('lb_integer_completion_v515',True)))
    key=_vrp_fingerprint((scope,settings));old=state.get('lp_continuity_v515',{})
    cache=copy.deepcopy(old) if old.get('key')==key and old.get('scope')==scope else dict(key=key,scope=scope,attempts={})
    provider=_v58_provider(inputs);p=provider.problem;coeff=provider.objective_context['coefficients']
    if any(float(c)<0 for c in coeff.values()) or any(v[k]<0 for v in p['vehicles'] for k in ('fixed_cost','cost_per_km')):
        return dict(completed=True,global_bound_usable=False,termination_reason='UNSUPPORTED_NEGATIVE_COEFFICIENT')
    best=max(float(self._analytical_floor_v514(p,coeff)['canonical_objective_lb']),float(cache.get('lower_bound',0.)))
    # Only a coordinator-validated bound for the same problem may enter here.
    imported=inputs.get('lb_start_v515',{})
    if imported.get('fingerprint')==scope and imported.get('certified') and math.isfinite(float(imported.get('value',0))):
        best=max(best,float(imported['value']))
    initial=best
    result=dict(agent='BoundStrengtheningAgent',bound_version='V515_LPI_FAMILIES_2',fingerprint=scope,
                completed=True,global_bound_usable=True,canonical_lower_bound=best,passes=passes,
                root_lp_completed=False,integer_phase_executed=False,binary_variables=0,variables=None,
                model_kind='CONTINUOUS_LP',bound_method='BOX_CORRECTED_DUAL',bound_before=initial)
    def save(reason=None):
        if reason:result['termination_reason']=reason
        cache['lower_bound']=best;state['lp_continuity_v515']=cache
        result.update(canonical_lower_bound=best,elapsed_seconds=time.monotonic()-started,cache_summary=dict(
            reference_cached=bool(cache.get('reference')),families=list(cache.get('attempts',{})),exhausted=bool(cache.get('exhausted'))))
        _pd_write(folder/'search_state.json',state)
        _pd_write(folder/'bound_checkpoint.json',dict(result,completed=False,termination_reason='V515_VALID_BOUND_CHECKPOINT'))
        _pd_write(folder/'lb_progress.json',result)
    def progress(phase,**kw):
        _pd_write(folder/'lb_live.json',dict(phase=phase,updated_at=time.time(),fingerprint=scope,
            dual_bound=best,nodes=None,variables=result.get('variables'),build_seconds=result.get('build_seconds'),**kw))
    try:
        save('PREPARING_LP')
        previous_budget=float(cache.get('cycle_budget',seconds))
        previous_integer=cache.get('integer_attempt',{})
        if settings['fallback'] and cache.get('exhausted') and seconds>=1.5*previous_budget and previous_integer.get('solver_status') not in ('optimal','infeasible'):
            cache['exhausted']=False;cache.pop('integer_attempt',None)
        if cache.get('exhausted') and not inputs.get('lb_retry_v515',False):
            result['root_lp_completed']=bool(cache.get('reference'));result['cache_hit']=True
            return dict(result,termination_reason='SCOPED_LP_WORK_REUSED',canonical_lower_bound=best,model_kind='CACHED_VALID_BOUND',binary_variables=None)
        failed=cache.get('reference_failure',{})
        if failed and seconds<=1.25*failed.get('budget',0) and not inputs.get('lb_retry_v515',False):
            return dict(result,termination_reason='PREVIOUS_REFERENCE_BUDGET_INSUFFICIENT',canonical_lower_bound=best,cache_hit=True)
        if seconds<5:return dict(result,termination_reason='INSUFFICIENT_LP_BUDGET')
        broad=all(p['windows'][i][0]<=p['operation_start'] and p['windows'][i][1]>=p['operation_end'] for i in p['customers'])
        if not p['allow_multi_trip'] and broad and len(p['customers'])>=50 and seconds>=30 and not cache.get('aggregate_attempt'):
            progress('AGGREGATE_SINGLE_TRIP_FLOOR')
            aggregate=self._aggregate_bound_v514(p,coeff,min(15.,.25*seconds))
            cache['aggregate_attempt']=aggregate;best=max(best,float(aggregate['canonical_lower_bound']))
            result['aggregate']=aggregate;save('AGGREGATE_FLOOR_COMPLETED')
        progress('BUILD_PURE_LP')
        built=self._build_reference_lp_v5141(inputs);model=built['model'];data=self._matrix_lp_v5141(model)
        result.update(build_seconds=time.monotonic()-started,variables=len(data['c']),variable_types=data['variable_types'],
                      reference_matrix_hash=data['matrix_hash'])
        reference=cache.get('reference')
        if reference and reference.get('matrix_hash')==data['matrix_hash']:
            vector=np.asarray(reference['values']);data['_completed_basis']=reference['basis']
            refrow=copy.deepcopy(reference['row']);refrow.update(phase='REFERENCE_REUSED',seconds=0.,cache_hit=True)
            passes.append(refrow)
        else:
            cache.pop('reference',None);cache.pop('candidates',None);cache.pop('integer_attempt',None);cache['attempts']={}
            remaining=deadline-time.monotonic()-1.
            if remaining<1:return dict(result,termination_reason='BUDGET_USED_IN_MODEL_BUILD')
            allowance=min(90.,max(.1,remaining*.75))
            progress('REFERENCE_LP')
            refrow,vector=self._solve_lp_v5141(data,allowance,'REFERENCE',folder,best);passes.append(refrow)
            if vector is None:
                cache['reference_failure']=dict(budget=seconds,status=refrow['status']);save('REFERENCE_LP_NOT_COMPLETED')
                return result
            cache.pop('reference_failure',None)
            cache['reference']=dict(matrix_hash=data['matrix_hash'],values=vector.tolist(),basis=data['_completed_basis'],row=refrow)
        best=max(best,refrow['certified_lb']);result['root_lp_completed']=True
        save('REFERENCE_LP_COMPLETED')
        class FrozenSolution:
            def getNSols(self):return 1
            def getBestSol(self):return vector
            def getSolVal(self,sol,var):return float(sol[data['index'][var.name]])
        if 'candidates' not in cache:
            progress('SEPARATE_FAMILIES')
            subsets,pairs=self._separate_global_cuts(FrozenSolution(),built['p'],built['x'],built['assignment'],built['subsets'],set())
            cache['candidates']=dict(CAPACITY_ONLY=subsets[:20],VEHICLE_CONSISTENCY_ONLY=pairs[:30])
        original_cons={c.name for c in model.getConss(transformed=False)}
        for label in ('VEHICLE_CONSISTENCY_ONLY','CAPACITY_ONLY'):
            candidates=cache['candidates'][label];prior=cache['attempts'].get(label)
            if prior and not inputs.get('lb_retry_v515',False):
                passes.append(dict(phase=label,status='SCOPED_ATTEMPT_REUSED',seconds=0.,previous_status=prior['status']))
                if prior.get('certified_lb') is not None:best=max(best,prior['certified_lb'])
                continue
            if not candidates:
                cache['attempts'][label]=dict(status='NO_VIOLATED_CUTS',certified_lb=None)
                passes.append(dict(phase=label,status='NO_VIOLATED_CUTS',seconds=0.));continue
            if deadline-time.monotonic()<3:break
            setup=time.monotonic();progress(label)
            if label=='CAPACITY_ONLY':
                for subset in candidates:self._add_weighted_capacity_cut(model,built['p'],built['x'],built['assignment'],subset)
            else:self._add_owner_consistency(model,built['p'],built['x'],built['assignment'],candidates)
            added=[c for c in model.getConss(transformed=False) if c.name not in original_cons]
            variant=self._matrix_lp_v5141(model);nbase=len(data['b_ub']);extra=len(variant['b_ub'])-nbase
            if variant['names']!=data['names'] or (variant['A_ub'][:nbase]-data['A_ub']).nnz or (variant['A_eq']-data['A_eq']).nnz or not np.array_equal(variant['b_ub'][:nbase],data['b_ub']) or not np.array_equal(variant['b_eq'],data['b_eq']):
                raise RuntimeError('A família alterou a referência.')
            cs,rs=data['_completed_basis'];variant['_warmstart_basis']=(list(cs),list(rs[:nbase])+[SCIP_BASESTAT.BASIC]*extra+list(rs[nbase:]))
            remaining=deadline-time.monotonic()-1.
            if remaining<=.1:
                for c in added:model.delCons(c)
                break
            limit=min(30. if label=='VEHICLE_CONSISTENCY_ONLY' else 10.,remaining)
            setup_seconds=time.monotonic()-setup
            row,_=self._solve_lp_v5141(variant,limit,label,folder,best)
            row.update(added_constraints=len(added),setup_seconds=setup_seconds,reference_matrix_hash=data['matrix_hash'])
            passes.append(row);best=max(best,row['bound_after'])
            cache['attempts'][label]=dict(status=row['status'],certified_lb=row['certified_lb'],seconds=row['seconds'],gain=row['bound_gain'],repeat_automatically=False)
            for c in added:model.delCons(c)
            if self._matrix_lp_v5141(model)['matrix_hash']!=data['matrix_hash']:raise RuntimeError('Falha ao restaurar a referência LP.')
            row['reference_restored_exactly']=True;save('FAMILY_COMPLETED')
        # Keep the existing optional integer strengthening for inexpensive references.
        # It has a separate budget/status; it is never mislabeled as an LP.
        left=deadline-time.monotonic()-1.
        fast=float(refrow.get('seconds',1e9))<=.35*seconds
        if settings['fallback'] and fast and left>=10 and not cache.get('integer_attempt'):
            progress('INTEGER_COMPLETION_SAME_MODEL')
            separate=folder/'integer_completion';separate.mkdir(exist_ok=True)
            try:legacy=self._integer_completion_v515(built,cache,scope,best,max(.1,left-.5),separate)
            except Exception as exc:
                legacy=dict(termination_reason='INTEGER_COMPLETION_FAILED',global_bound_usable=False,error=f'{type(exc).__name__}: {exc}')
            if legacy.get('global_bound_usable'):best=max(best,float(legacy['canonical_lower_bound']))
            cache['integer_attempt']=dict(status=legacy.get('termination_reason'),lower_bound=legacy.get('canonical_lower_bound'),
                solver_status=(legacy.get('passes') or [{}])[-1].get('status'),budget=seconds)
            result['integer_completion']=legacy;result['integer_phase_executed']=legacy.get('integer_phase_executed',False)
            if result['integer_phase_executed']:
                result.update(model_kind='LP_FAMILIES_WITH_SEPARATE_INTEGER_COMPLETION',
                    binary_variables=legacy.get('binary_variables'),variables=legacy.get('variables'),
                    variable_types=legacy.get('passes',[{}])[-1].get('original_variable_types'),counts_refer_to='SEPARATE_INTEGER_COMPLETION')
            passes.extend(dict(row,phase='SEPARATE_'+row['phase']) for row in legacy.get('passes',[]))
        cache['cycle_budget']=seconds
        cache['exhausted']=all(k in cache['attempts'] for k in ('CAPACITY_ONLY','VEHICLE_CONSISTENCY_ONLY')) and (not settings['fallback'] or bool(cache.get('integer_attempt')))
        save('LP_CONTINUITY_FINISHED');return result
    finally:
        save()
        if model is not None:model.freeProb()



BoundStrengtheningAgent.strengthen_lp_continuity_v515=_v515_strengthen
BoundStrengtheningAgent.CONTINUITY_BOUND_VERSION='V515_LPI_FAMILIES_2'


def _v515_integer_completion(self,built,cache,scope,best,seconds,folder):
    """Same global relaxation, strongest tested family, explicit integrality."""
    import time,math
    from collections import Counter
    from pyscipopt import Eventhdlr,SCIP_EVENTTYPE
    started=time.monotonic();model=built['model'];p=built['p']
    usable=[(v['certified_lb'],k) for k,v in cache['attempts'].items() if v.get('certified_lb') is not None]
    family=max(usable)[1] if usable else None
    if family=='VEHICLE_CONSISTENCY_ONLY':self._add_owner_consistency(model,p,built['x'],built['assignment'],cache['candidates'][family])
    elif family=='CAPACITY_ONLY':
        for subset in cache['candidates'][family]:self._add_weighted_capacity_cut(model,p,built['x'],built['assignment'],subset)
    for var in list(built['z'].values())+list(built['x'].values())+list(built['assignment'].values()):model.chgVarType(var,'B')
    for var in built['tripv'].values():model.chgVarType(var,'I')
    types=dict(Counter(str(v.vtype()) for v in model.getVars(transformed=False)));last=[-10.]
    def save(value,phase):
        _pd_write(folder/'bound_checkpoint.json',dict(fingerprint=scope,completed=False,global_bound_usable=True,
            canonical_lower_bound=value,termination_reason='V515_INTEGER_BOUND_CHECKPOINT',bound_version='V515_INTEGER_RELAXATION'))
        _pd_write(folder.parent/'lb_live.json',dict(fingerprint=scope,phase=phase,updated_at=time.time(),dual_bound=value,
            nodes=model.getNNodes(),variables=len(model.getVars(transformed=False))))
    class Events(Eventhdlr):
        def eventinit(self):self.model.catchEvent(SCIP_EVENTTYPE.LPSOLVED|SCIP_EVENTTYPE.NODESOLVED,self)
        def eventexit(self):self.model.dropEvent(SCIP_EVENTTYPE.LPSOLVED|SCIP_EVENTTYPE.NODESOLVED,self)
        def eventexec(self,event):
            now=time.monotonic()-started
            if now-last[0]<2:return
            last[0]=now;dual=float(self.model.getDualbound())
            if math.isfinite(dual) and abs(dual)<1e19:save(max(best,dual-1e-7*max(1.,abs(dual))),'INTEGER_RELAXATION_SEARCH')
    model.includeEventhdlr(Events(),'integer_bound_v515','Progress of separate integer phase')
    model.setParam('limits/time',max(.05,seconds-(time.monotonic()-started)))
    model.optimize();status=str(model.getStatus());dual=float(model.getDualbound())
    valid=math.isfinite(dual) and abs(dual)<1e19 and status!='infeasible'
    value=max(best,dual-1e-7*max(1.,abs(dual))) if valid else best
    rec=dict(phase='INTEGER_RELAXATION',status=status,seconds=time.monotonic()-started,dual_bound=dual if valid else None,
        bound_before=best,bound_after=value,bound_gain=value-best,nodes=model.getNNodes(),
        original_variable_types=types,transformed_variable_types=dict(Counter(str(v.vtype()) for v in model.getVars(transformed=True))))
    save(value,'INTEGER_RELAXATION_FINISHED')
    return dict(termination_reason='INTEGER_RELAXATION_FINISHED',global_bound_usable=True,canonical_lower_bound=value,
        integer_phase_executed=True,variables=sum(types.values()),binary_variables=types.get('BINARY',0),passes=[rec],selected_family=family)

BoundStrengtheningAgent._integer_completion_v515=_v515_integer_completion


"""Continuation-only revision: bounded, fresh fractional separation on resume."""
_v516_previous_bound = BoundStrengtheningAgent.strengthen_lp_continuity_v515

def _v516_strengthen(self, inputs, seconds, folder, state):
    import copy, time, math, numpy as np
    from pyscipopt import SCIP_BASESTAT
    start=time.monotonic(); deadline=start+max(0.,float(seconds)); folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    scope=_ct_scope(inputs); cache=state.get('lp_continuity_v515',{})
    if inputs.get('lb_reference_policy_v515',False):return _v516_previous_bound(self,inputs,seconds,folder,state)
    settings=dict(version='V515_LPI_FAMILIES_2',matrix='V5141_REFERENCE',engine='LPI',tolerance=1e-7,
                  fallback=bool(inputs.get('lb_integer_completion_v515',True)))
    expected=_vrp_fingerprint((scope,settings))
    # Keep the proven first pass. A retry never silently reruns its integer tree.
    if cache.get('scope')!=scope or cache.get('key')!=expected or not cache.get('reference') or not cache.get('exhausted'):
        return _v516_previous_bound(self,inputs,seconds,folder,state)
    key=_vrp_fingerprint((scope,'V516_FRESH_SEPARATION_1',cache['reference']['matrix_hash']))
    old=state.get('lb_reseparation_v516',{})
    memo=copy.deepcopy(old) if old.get('key')==key else dict(key=key,scope=scope,tracks={})
    best=max(float(cache.get('lower_bound',0.)),float(memo.get('lower_bound',0.)))
    imported=inputs.get('lb_start_v515',{})
    if imported.get('fingerprint')==scope and imported.get('certified') and math.isfinite(float(imported.get('value',0.))):best=max(best,float(imported.get('value',0.)))
    initial=best; rows=[]; model=None
    result=dict(agent='BoundStrengtheningAgent',fingerprint=scope,bound_version='V516_FRESH_SEPARATION_1',completed=True,
                global_bound_usable=True,canonical_lower_bound=best,bound_before=best,passes=rows,
                model_kind='CONTINUOUS_LP',binary_variables=0,integer_phase_executed=False,root_lp_completed=True)
    def save(reason):
        memo['lower_bound']=best;state['lb_reseparation_v516']=memo
        cache['lower_bound']=max(float(cache.get('lower_bound',0.)),best)
        result.update(canonical_lower_bound=best,termination_reason=reason,elapsed_seconds=time.monotonic()-start,
                      bound_gain=best-initial,tracks={k:{q:v for q,v in t.items() if q not in ('values','basis')} for k,t in memo['tracks'].items()})
        _pd_write(folder/'search_state.json',state)
        _pd_write(folder/'bound_checkpoint.json',dict(result,completed=False))
        _pd_write(folder/'lb_progress.json',result)
    labels=('VEHICLE_CONSISTENCY_ONLY','CAPACITY_ONLY','COMBINED_INCREMENTAL')
    if not inputs.get('lb_experiment_v516',False):
        save('VALID_BOUND_REUSED_NEW_STRATEGY_REQUIRES_SHORT_TEST');return result
    if memo.get('global_stalled'):
        save('FRESH_SEPARATION_STOPPED_WITHOUT_GLOBAL_GAIN');return result
    if all(memo['tracks'].get(k,{}).get('closed') for k in labels):
        save('FRESH_SEPARATION_EXHAUSTED_NO_NEW_STRATEGY');return result
    if seconds<5:save('INSUFFICIENT_FRESH_SEPARATION_BUDGET');return result
    try:
        # Rotate tracks between calls so a slow first family cannot starve the other.
        turn=int(memo.get('turn',0));memo['turn']=turn+1
        for label in labels[turn%len(labels):]+labels[:turn%len(labels)]:
            track=memo['tracks'].setdefault(label,dict(cuts=[],cycles=0,closed=False))
            if track['closed'] or deadline-time.monotonic()<3:continue
            build_start=time.monotonic();built=self._build_reference_lp_v5141(inputs);model=built['model'];base=self._matrix_lp_v5141(model)
            result.update(variables=len(base['c']),variable_types=base['variable_types'],build_seconds=time.monotonic()-build_start)
            reference=memo.get('reference',cache['reference'])
            if base['matrix_hash']!=reference['matrix_hash']:
                allowance=min(30.,max(.1,deadline-time.monotonic()-2.))
                row,rebuilt_vector=self._solve_lp_v5141(base,allowance,'REFERENCE_REBUILT_FOR_CURRENT_RUNTIME',folder,best)
                rows.append(row)
                if rebuilt_vector is None:
                    save('REFERENCE_REBUILD_NOT_COMPLETED');model.freeProb();model=None;break
                reference=dict(matrix_hash=base['matrix_hash'],values=rebuilt_vector.tolist(),basis=base['_completed_basis'],row=row)
                memo['reference']=reference
                for item in memo['tracks'].values():item.update(cuts=[],cycles=0,closed=False)
                best=max(best,row['bound_after'])
                save('REFERENCE_REBUILT')
            data=base;vector=np.asarray(reference['values']);basis=reference['basis']
            def add(cuts):
                if label=='CAPACITY_ONLY':
                    for subset in cuts:self._add_weighted_capacity_cut(model,built['p'],built['x'],built['assignment'],subset)
                elif label=='VEHICLE_CONSISTENCY_ONLY':self._add_owner_consistency(model,built['p'],built['x'],built['assignment'],cuts)
                else:
                    for cut in cuts:
                        if cut[0]=='C':self._add_weighted_capacity_cut(model,built['p'],built['x'],built['assignment'],cut[1:])
                        else:self._add_owner_consistency(model,built['p'],built['x'],built['assignment'],[cut[1:]])
            if track['cuts']:
                add(track['cuts']);data=self._matrix_lp_v5141(model)
                if data['matrix_hash']!=track.get('matrix_hash'):raise RuntimeError('Cached cut matrix mismatch.')
                vector=np.asarray(track['values']);basis=track['basis']
            previous_lp=float(track.get('lp_bound',reference['row']['certified_lb']))
            for local_cycle in range(3):
                if deadline-time.monotonic()<3:break
                setup=time.monotonic()
                class Fractional:
                    def getNSols(self):return 1
                    def getBestSol(self):return vector
                    def getSolVal(self,sol,var):return float(sol[data['index'][var.name]])
                existing=set(map(tuple,built['subsets']))
                capacity_existing=existing|set(map(tuple,track['cuts'])) if label=='CAPACITY_ONLY' else existing
                link_existing=set(map(tuple,track['cuts'])) if label=='VEHICLE_CONSISTENCY_ONLY' else set()
                if label=='COMBINED_INCREMENTAL':
                    capacity_existing=existing|{tuple(c[1:]) for c in track['cuts'] if c[0]=='C'}
                    link_existing={tuple(c[1:]) for c in track['cuts'] if c[0]=='L'}
                subsets,pairs=self._separate_global_cuts(Fractional(),built['p'],built['x'],built['assignment'],capacity_existing,link_existing)
                cuts=(subsets[:20] if label=='CAPACITY_ONLY' else pairs[:30])
                if label=='COMBINED_INCREMENTAL':cuts=[['C']+list(c) for c in subsets[:20]]+[['L']+list(c) for c in pairs[:30]]
                if not cuts:
                    track.update(closed=True,stop_reason='NO_NEW_VIOLATED_CUTS');save('TRACK_CLOSED');break
                old_data=data;add(cuts);variant=self._matrix_lp_v5141(model);nbase=len(old_data['b_ub']);extra=len(variant['b_ub'])-nbase
                if variant['names']!=old_data['names'] or (variant['A_ub'][:nbase]-old_data['A_ub']).nnz or (variant['A_eq']-old_data['A_eq']).nnz or not np.array_equal(variant['b_ub'][:nbase],old_data['b_ub']) or not np.array_equal(variant['b_eq'],old_data['b_eq']):raise RuntimeError('Cut insertion changed existing matrix.')
                cs,rs=basis;variant['_warmstart_basis']=(list(cs),list(rs[:nbase])+[SCIP_BASESTAT.BASIC]*extra+list(rs[nbase:]))
                allowance=min(20.,max(0.,deadline-time.monotonic()-1.))
                if allowance<.1:break
                phase='FRESH_'+label+'_'+str(track['cycles']+1)
                _pd_write(folder/'lb_live.json',dict(phase=phase,updated_at=time.time(),fingerprint=scope,dual_bound=best,nodes=None,variables=result.get('variables'),build_seconds=result.get('build_seconds')))
                row,new_vector=self._solve_lp_v5141(variant,allowance,phase,folder,best)
                row.update(new_cuts=len(cuts),source='LATEST_COMPLETED_FRACTIONAL_LP',separation_setup_seconds=time.monotonic()-setup-row['seconds'])
                rows.append(row)
                if new_vector is None:
                    track.update(closed=True,stop_reason='LP_NOT_COMPLETED_PRESERVED_PREVIOUS_BOUND');save('TRACK_CLOSED');break
                lp_bound=float(row['certified_lb']);gain=max(0.,lp_bound-previous_lp)
                best=max(best,lp_bound);row.update(lp_gain=gain,global_gain=row['bound_gain'],lp_gain_per_second=gain/max(row['seconds'],1e-9))
                track['cuts'].extend([list(c) for c in cuts]);track.update(cycles=track['cycles']+1,values=new_vector.tolist(),basis=variant['_completed_basis'],matrix_hash=variant['matrix_hash'],lp_bound=lp_bound)
                data=variant;vector=new_vector;basis=variant['_completed_basis'];previous_lp=lp_bound
                # Stop on the LP's own marginal gain; a stronger historical integer LB is preserved separately.
                if gain<=1e-7 or track['cycles']>=8:
                    track.update(closed=True,stop_reason='LP_MARGINAL_GAIN_EXHAUSTED' if gain<=1e-7 else 'CUT_CYCLE_LIMIT')
                save('FRESH_LP_COMPLETED')
                if track['closed']:break
            model.freeProb();model=None
        memo['global_stalled']=best<=initial+1e-7
        save('FRESH_SEPARATION_FINISHED');return result
    finally:
        save(result.get('termination_reason','FRESH_SEPARATION_INTERRUPTED'))
        if model is not None:model.freeProb()

BoundStrengtheningAgent.strengthen_lp_continuity_v515=_v516_strengthen
BoundStrengtheningAgent.CONTINUITY_BOUND_VERSION='V516_FRESH_SEPARATION_1'


# Diagnostics are methods of the existing continuity agents, coordinated as before.
_v516_previous_elimination=KVehicleFeasibilityCertifier.attempt_elimination
_v516_previous_region=NeighborhoodOptimizationAgent.optimize_region

def _v516_partial_diagnosis(self, inputs, best, state):
    p=_v58_provider(inputs).problem;signature=_v58_signature(_v58_routes(best));partials=[]
    for item in state.get('elimination_partials',[]):
        present={i for r in item.get('routes',[]) for i in r['path'][1:-1]}
        missing=[i for i in p['customers'] if i not in present]
        partials.append(dict(matches_current_incumbent=item.get('base_signature')==signature,removed_vehicle=item.get('removed_vehicle'),served=len(present),missing=missing,
            missing_details=[dict(client=i,demand=p['client_rows'][i]['demand'],window=list(p['windows'][i]),
                                 rejection_counts=state.get('elimination_blockers_v514',{}).get(str(i),{})) for i in missing]))
    return dict(status='PARTIALS_AVAILABLE' if partials else 'NO_MATCHING_PARTIAL_CHECKPOINT',
                incumbent_signature=signature,partials=sorted(partials,key=lambda r:-r['served']),
                note='Rejection counts are search observations, not an infeasibility proof.')

def _v516_region_inventory(self, inputs, best):
    import itertools,time
    start=time.monotonic();p=_v58_provider(inputs).problem
    routes=sorted(best['routes'],key=lambda r:(r['vehicle_id'],r['start_time_minutes']))
    counts={};complete=True
    for size in (2,3,4):
        row=dict(examined=0,eligible=0,above_24_customers=0,noncontiguous=0,min_customers=None);counts[str(size)]=row
        for indices in itertools.combinations(range(len(routes)),size):
            if time.monotonic()-start>1.0:complete=False;break
            row['examined']+=1;n=sum(len(routes[i]['path'])-2 for i in indices)
            row['min_customers']=n if row['min_customers'] is None else min(n,row['min_customers'])
            if n>24:row['above_24_customers']+=1;continue
            contiguous=True
            for vid in {routes[i]['vehicle_id'] for i in indices}:
                seq=[i for i,r in enumerate(routes) if r['vehicle_id']==vid];pos=[seq.index(i) for i in indices if routes[i]['vehicle_id']==vid]
                if max(pos)-min(pos)+1!=len(pos):contiguous=False;break
            row['eligible' if contiguous else 'noncontiguous']+=1
    return dict(counts=counts,complete=complete,seconds=time.monotonic()-start,customer_cap=24,
                note='Eligible means geometrically admissible for selection, not necessarily solved or feasible.')

def _v516_elimination(self,inputs,best,seconds,folder,state):
    answer,stats=_v516_previous_elimination(self,inputs,best,seconds,folder,state)
    diagnosis=self.diagnose_partials_v516(inputs,best,state);stats['partial_diagnosis']=diagnosis
    _pd_write(_PDPath(folder)/'partial_diagnosis.json',diagnosis)
    return answer,stats

def _v516_region(self,inputs,best,seconds,folder,state):
    if inputs.get('lb_start_v515',{}).get('fingerprint')!=_ct_scope(inputs):
        return _v516_previous_region(self,inputs,best,seconds,folder,state)
    import time
    t=time.monotonic();inventory=self.region_inventory_v516(inputs,best)
    answer,stats=_v516_previous_region(self,inputs,best,max(.1,seconds-(time.monotonic()-t)),folder,state)
    stats['region_inventory']=inventory
    _pd_write(_PDPath(folder)/'region_inventory.json',inventory)
    return answer,stats

KVehicleFeasibilityCertifier.diagnose_partials_v516=_v516_partial_diagnosis
KVehicleFeasibilityCertifier.attempt_elimination=_v516_elimination
NeighborhoodOptimizationAgent.region_inventory_v516=_v516_region_inventory
NeighborhoodOptimizationAgent.optimize_region=_v516_region

_v516_previous_activity=_v511_activity
def _v516_activity(stage,result,before,after):
    activity=_v516_previous_activity(stage,result,before,after)
    for key in ('partial_diagnosis','region_inventory'):
        if key in result:activity[key]=result[key]
    if stage=='LB_COMPACTO':
        phases=activity.get('phases') or []
        activity['phase_benefits']=[dict(phase=r.get('phase'),status=r.get('status'),seconds=r.get('seconds'),
            produced_lb=r.get('certified_lb',r.get('dual_bound')),preserved_global_lb=r.get('bound_after'),
            global_gain=r.get('bound_gain'),lp_gain=r.get('lp_gain'),gain_per_second=r.get('gain_per_second'),
            model_kind=r.get('model_kind'),integer_variables=r.get('integer_variables'),nodes=r.get('nodes')) for r in phases]
    return activity
_v511_activity=_v516_activity
print('Agentes de base carregados. A integração final será verificada antes do acionamento.')


````

#### Fortalecimento da prova global

#### Política e execução da continuidade

Escolhe e executa as próximas tarefas conforme o tempo restante e o progresso observado. Alterna a busca de planos melhores com o fortalecimento da prova, evitando repetir tentativas sem informação nova e preservando os ganhos já confirmados.


````python
class ContinuityTaskPolicyV524:
    """Select fresh local work; stop on measured stagnation, not a fixed return count."""
    def next_task(self, inputs, best, search, stages, remaining):
        exact=[r for k,r in stages.items() if k=='EXATO_LOCAL' or k.startswith('EXATO_RETORNO')]
        if remaining<22:return dict(run=False,reason='INSUFFICIENT_TIME_FOR_USEFUL_REGION')
        if exact and exact[-1].get('parent_stop_reason')=='USER_INTERRUPTED':return dict(run=False,reason='USER_INTERRUPTED')
        if exact and (exact[-1].get('error') or exact[-1].get('completed') is False):
            return dict(run=False,reason='LOCAL_PROCESS_INCOMPLETE_REVIEW_EXECUTION')
        if len(exact)>=2 and all(float(r.get('ub_gain',0.))<=1e-9 for r in exact[-2:]):
            return dict(run=False,reason='STAGNATION_TWO_LOCAL_BATCHES_WITHOUT_GAIN')
        keys={r['neighborhood_key'] for batch in exact for r in batch.get('neighborhoods',[]) if r.get('neighborhood_key')}
        closed=set(search.get('closed_neighborhoods_v511',[]))|keys
        p=_v58_provider(inputs).problem
        choices=_ct_guided_regions(inputs,p,best,closed,int(search.get('local_stalled',0)))
        if not choices:return dict(run=False,reason='NO_FRESH_ELIGIBLE_REGION')
        return dict(run=True,reason='FRESH_REGION_AVAILABLE',exclude_keys=sorted(keys),candidate_key=choices[0][3])


def _v59_exact_search(inputs,best,seconds,folder,state):
    import time
    deadline=time.monotonic()+seconds;provider=_v58_provider(inputs);p=provider.problem
    round_id=int(state.get('region_round_v511',0))+1;state['region_round_v511']=round_id
    ledger=state.setdefault('region_ledger_v511',{});cooldown_skips=0
    tried=set(state.get('closed_neighborhoods_v511',[]));attempted=set(inputs.get('exclude_region_keys_v524',[]));pending=state.setdefault('pending_neighborhoods_v592',{});stalled=int(state.get('local_stalled',0));runs=[];stop_reason="REGION_BUDGET_EXHAUSTED"
    while time.monotonic()<deadline-2:
        choices=_ct_guided_regions(inputs,p,best,tried|attempted,stalled)
        if not choices:
            stop_reason="NO_UNEXAMINED_REGIONS";break
        eligible=[c for c in choices if int(ledger.get(c[3],{}).get('retry_after_round',0))<=round_id]
        blocked={c[3] for c in choices if c not in eligible};cooldown_skips+=len(blocked);attempted.update(blocked)
        if not eligible:continue
        _,_,indices,key=min(eligible,key=lambda c:(int(ledger.get(c[3],{}).get('attempts',0)),c[0],c[1]))
        attempted.add(key);entry=ledger.setdefault(key,dict(attempts=0));retry=entry['attempts']>0
        entry.update(attempts=entry['attempts']+1,status='RUNNING',round=round_id,released_trips=len(indices))
        if entry['attempts']%3==0:entry['retry_after_round']=round_id+3
        _pd_write(folder/'search_state.json',state)
        candidate,info=_v58_exact_neighborhood(inputs,best,set(indices),_ct_guided_region_seconds(inputs,(12 if len(indices)==2 else 18)*(1+min(2,pending.get(key,0))),deadline-time.monotonic()-.5))
        entry['status']=info.get('status');info['retry']=retry
        if entry['attempts']%3==0 and info.get('status') not in ('optimal','infeasible'):entry['retry_after_round']=round_id+3
        info['released_trips']=len(indices);info['neighborhood_key']=key;runs.append(info);improved=False
        if info.get('status') in ('optimal','infeasible'):tried.add(key);pending.pop(key,None)
        else:pending[key]=pending.get(key,0)+1
        info['before_ub']=best['canonical_objective_score']
        if candidate:
            audit=_pd_audit(inputs,candidate)
            info['candidate_audit_passed']=bool(audit.get('feasible'))
            if audit.get('feasible'):
                # Keep complete alternatives as sources of reusable vehicle programs.
                state.setdefault('feasible_pool',[]).append(audit['normalized_solution'])
                pool={_v58_signature(_v58_routes(s)):s for s in state['feasible_pool']}
                state['feasible_pool']=sorted(pool.values(),key=lambda s:s['canonical_objective_score'])[:8]
                if audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:
                    best=audit['normalized_solution'];_pd_write(folder/'candidate.json',best);improved=True
        stalled=0 if improved else stalled+1
        state.update(closed_neighborhoods_v511=sorted(tried)[-3000:],local_stalled=stalled)
        if len(ledger)>2000:
            keep=sorted(ledger,key=lambda k:ledger[k].get('round',0),reverse=True)[:2000]
            ledger={k:ledger[k] for k in keep};state['region_ledger_v511']=ledger
        _pd_write(folder/'search_state.json',state)
        info['after_ub']=best['canonical_objective_score']
        if improved and inputs.get('portfolio_mode_v512'):
            stop_reason='NEW_CANDIDATE_FOR_POLISH';break
    return best,dict(termination_reason=stop_reason,assigned_seconds=seconds,neighborhoods=runs,cooldown_skips=cooldown_skips,nodes=sum(r['nodes'] or 0 for r in runs),global_bound_usable=False)





_v524_search_bindings=list(_V513_BINDINGS);_v524_search_bindings[7]=_v59_exact_search;_V513_BINDINGS=tuple(_v524_search_bindings)




_v523_bindings=list(_V513_BINDINGS)
_v523_bindings[4]=executar_continuidade_integrada
_v523_bindings[14]=NeighborhoodOptimizationAgent.optimize_region
_v523_bindings[15]=KVehicleFeasibilityCertifier.attempt_elimination
_V513_BINDINGS=tuple(_v523_bindings)




_v529_detailed_report=relatorio_tecnico_continuidade

# Comunicação da continuidade: dados técnicos são preservados nos arquivos.
MOSTRAR_DIAGNOSTICO_TECNICO = False

def _ct_stage_name(stage):
    if stage.startswith('EXATO_RETORNO'): return 'Reorganização de viagens'
    return {'MELHORIA':'Melhoria do plano','LB_COMPACTO':'Verificação da garantia',
        'EXATO_LOCAL':'Reorganização de viagens','ELIMINAR_VEICULO':'Tentativa de reduzir a frota',
        'EXATO_COMPLETO_GLOBAL':'Revisão do plano inteiro','GERACAO_COLUNAS':'Combinação de alternativas'}.get(stage,'Busca de melhorias')

````

#### Comunicação e revisão completa

Separa os registros técnicos da explicação ao usuário e administra a tentativa de revisar o plano inteiro. Informa o que foi concluído ou interrompido, sem apresentar falta de tempo, memória ou melhoria como prova de impossibilidade.


````python
def _ct_technical_print(folder, *args, **kwargs):
    import builtins
    from pathlib import Path
    path=Path(folder);path.mkdir(parents=True,exist_ok=True)
    options={k:v for k,v in kwargs.items() if k not in ('file','flush')}
    try:
        with (path/'progresso_tecnico.log').open('a',encoding='utf-8') as stream:
            builtins.print(*args,file=stream,**options)
    except OSError:
        # A display-log failure must not discard an audited solution.
        builtins.print('Não foi possível gravar um trecho do diagnóstico de acompanhamento.')
    if globals().get('MOSTRAR_DIAGNOSTICO_TECNICO',False):builtins.print(*args,**kwargs)

def _ct_stage_finished(stage, row):
    name=_ct_stage_name(stage)
    if row.get('error') or (row.get('worker_exitcode') not in (None,0) and not row.get('parent_stop_reason')) or row.get('process_outcome')=='FALHA_DE_PROCESSO':
        return name+': houve uma falha; o plano aprovado foi preservado. Consulte o diagnóstico.'
    if row.get('parent_stop_reason')=='USER_INTERRUPTED':return name+': interrompida a seu pedido; os resultados recuperáveis foram conferidos.'
    gains=[]
    if row.get('ub_gain',0)>1e-9:gains.append('o plano melhorou')
    if row.get('lb_gain',0)>1e-9:gains.append('a garantia avançou')
    text=name+': '+(' e '.join(gains) if gains else 'sem ganho adicional nesta tentativa')+'.'
    if row.get('parent_stop_reason') and row.get('parent_stop_reason')!='USER_INTERRUPTED':
        text+=' A tentativa chegou ao limite destinado a ela; os resultados recuperados foram conferidos.'
    elif not row.get('completed',True):text+=' A tarefa não concluiu; consulte o diagnóstico antes de repetir.'
    return text

def _ct_friendly_label(text):
    for key in ('EXATO_COMPLETO_GLOBAL','ELIMINAR_VEICULO','GERACAO_COLUNAS','LB_COMPACTO','EXATO_LOCAL','MELHORIA'):
        text=text.replace(key,_ct_stage_name(key).lower())
    return text

def _ct_short_budget(result):
    stats=result.get('statistics',{});seconds=stats.get('elapsed_seconds');requested=stats.get('requested_seconds')
    if seconds is None or requested is None:return 'O tempo utilizado não está disponível neste registro.'
    text='Tempo usado nesta rodada: **'+_ct_time_words(seconds)+'**, de até **'+_ct_time_words(requested)+'** autorizados. '
    reason=stats.get('coordinator_stop_v524',{}).get('reason')
    stages=stats.get('stages',{});full=stages.get('EXATO_COMPLETO_GLOBAL',{})
    if result.get('certificate',{}).get('proven_optimal'):text+='Encerramos porque a prova de ótimo foi concluída.'
    elif result.get('status')=='INTERRUPTED':text+='A rodada foi interrompida a seu pedido; os resultados aprovados foram preservados.'
    elif any(r.get('error') or (r.get('worker_exitcode') not in (None,0) and not r.get('parent_stop_reason')) for r in stages.values()):text+='Uma tarefa teve uma falha; o diagnóstico explica o ocorrido. O melhor plano aprovado foi preservado.'
    elif full.get('status')=='memlimit':text+='A revisão atingiu o limite de memória, não o tempo total autorizado.'
    elif full.get('termination_reason')=='TIME_LIMIT':text+='A revisão do plano inteiro chegou ao limite destinado a ela, sem concluir a prova.'
    elif seconds>=requested-2:text+='A rodada chegou ao prazo autorizado.'
    elif reason=='STAGNATION_TWO_LOCAL_BATCHES_WITHOUT_GAIN':text+='Paramos antes do prazo porque as últimas tentativas de reorganização não trouxeram ganho. Isso não prova que o plano seja ótimo.'
    elif reason=='LOCAL_PROCESS_INCOMPLETE_REVIEW_EXECUTION':text+='Uma busca não terminou normalmente; confira o diagnóstico antes de repetir.'
    elif reason=='INSUFFICIENT_TIME_FOR_USEFUL_REGION':text+='O saldo não permitia iniciar outra tarefa com tempo útil.'
    else:text+='As tarefas selecionadas para esta rodada terminaram antes do prazo. Isso não prova que o plano seja ótimo.'
    return text

def _ct_time_breakdown(solution):
    import math
    rows={}
    for route in solution.get('routes',[]):
        vid=route.get('physical_vehicle_id',route.get('vehicle_id'))
        duration=route.get('total_route_time_minutes')
        if duration is None or not math.isfinite(float(duration)):return ''
        row=rows.setdefault(vid,dict(trips=0,time=0.))
        row['trips']+=1;row['time']+=float(duration)
    if not rows:return ''
    def fmt(value):return f'{value:,.2f}'.replace(',','@').replace('.',',').replace('@','.')
    text='\n## Como ler o tempo das viagens\n\nÉ a soma do tempo de cada viagem: deslocamento, atendimento e espera durante o percurso. Como os veículos podem trabalhar ao mesmo tempo, essa soma não é o tempo de relógio até terminar todas as entregas. Não inclui intervalos no depósito entre viagens nem o tempo que o programa levou para calcular o plano.\n\n'
    text+='| Veículo | Viagens | Tempo somado das viagens (min) |\n|---|---:|---:|\n'
    def vehicle_key(item):
        try:return (0,float(item[0]))
        except (TypeError,ValueError):return (1,str(item[0]))
    for vid,row in sorted(rows.items(),key=vehicle_key):text+=f"| {vid} | {row['trips']} | {fmt(row['time'])} |\n"
    total=sum(r['time'] for r in rows.values())
    text+=f"| Total da frota | {sum(r['trips'] for r in rows.values())} | {fmt(total)} |\n"
    return text

def _ct_customer_report(result, incluir_plano=True):
    _v534_validate_result(result)  # Preserve the audit and certificate validation gate.
    return _v534_customer_narrative(result['final_solution'],result['certificate'],mem=result.get('customer_context_v534'),result=result,incluir_plano=incluir_plano)




_v524_report_bindings=list(_V513_BINDINGS);_v524_report_bindings[9]=relatorio_final_continuidade;_V513_BINDINGS=tuple(_v524_report_bindings)


def _ct_continue_guided(self):
    mem=self.memory.data
    last=mem.get('continuity_integrated_result_v53') or {}
    best=last.get('final_solution') or mem.get('best_incumbent') or mem.get('solver_result')
    cert=last.get('certificate') or mem.get('continuity_certificate') or mem.get('certification_result') or {}
    return self._offer_and_run_continuity(mem.get('user_answers',{}),best,cert,mem.get('audit_recommendation'))

OrchestratorAgent.continuar_com_orientacao=_ct_continue_guided

# Copy of complete routing model for diagnostics only. Production model is unchanged.
def _v525_complete_builder(self, incumbent_ub=None, feasibility_only=False, max_active_vehicles=None):
    """Exact customer-chain model: direct links or explicit depot reload links.

    A physical vehicle is one chronological chain; each depot link splits it
    into trips. No artificial trip slots or arc copies indexed by trip.
    Waiting at customers is counted; idle gaps at the depot are not counted.
    """
    Model, qs = self._import_scip()
    p=self.problem; C=p['customers']; V=p['vehicles']; n=len(C)
    all_vehicles=list(V)
    # In this operational contract every vehicle shares speed, work horizon,
    # windows and minimum use. Only capacity and costs differ. Costs are
    # irrelevant in feasibility mode. Any <=K feasible fleet can therefore
    # be injected into the K largest capacities, in capacity order.
    dominance=bool(feasibility_only and max_active_vehicles is not None and
                   not p['must_use_all_vehicles'] and 0<int(max_active_vehicles)<len(V))
    if dominance:V=sorted(V,key=lambda v:(-float(v['capacity']),v['idx']))[:int(max_active_vehicles)]
    op0=float(p['operation_start']); op1=float(p['operation_end']); span=op1-op0
    H=float(p['max_time']); svc=float(p['service_time']); tt=p['travel_time']; dd=p['distance']
    Q=max(float(v['capacity']) for v in V)
    demand={i:float(p['client_rows'][i]['demand']) for i in C}
    windows={i:(max(op0,float(p['windows'][i][0])),min(op1,float(p['windows'][i][1]))) for i in C}
    direct=[(i,j) for i in C for j in C if i!=j and
            windows[i][0]+svc+tt[i,j]<=windows[j][1]+1e-9 and demand[i]+demand[j]<=Q+1e-9]
    reloads=[(i,j) for i in C for j in C if i!=j and
             windows[i][0]+svc+tt[i,0]+tt[0,j]<=windows[j][1]+1e-9] if p['allow_multi_trip'] else []
    pairs=sorted(set(direct)|set(reloads))
    b=self._analytical_lower_bounds(incumbent_ub=incumbent_ub)
    scf={'valid':False,'diagnostic_note':'Derived bound cuts disabled for independent model comparison'}
    if scf.get('valid') and scf.get('distance_lb') is not None:
        b['distance_lb']=max(float(b['distance_lb']),float(scf['distance_lb'])-1e-7)
    fixed=sorted(float(v['fixed_cost']) for v in V)
    if min(fixed)>=0 and min(float(v['cost_per_km']) for v in V)>=0:
        flb=sum(fixed) if p['fixed_cost_mode']!='WHEN_USED' or p['must_use_all_vehicles'] else sum(fixed[:int(b['vehicles_lb'])])
        b['cost_lb']=max(float(b['cost_lb']),flb+min(float(v['cost_per_km']) for v in V)*b['distance_lb'])
    b['scf_model_result']=scf
    coeff=self.objective_context['coefficients']
    b['canonical_objective_lb']=sum(float(coeff.get(k,0))*float(b[f]) for k,f in
        [('DISTANCE','distance_lb'),('COST','cost_lb'),('TIME','time_lb'),('VEHICLES','vehicles_lb')])
    model=Model('VRPTW_CUSTOMER_CHAIN_V52')
    if not self.verbose_solver:model.hideOutput()
    model.setParam('limits/time',float(self.time_limit_seconds))
    model.setParam('limits/gap',float(self.relative_gap_target))
    if self.scip_memory_mb>0:model.setParam('limits/memory',float(self.scip_memory_mb))
    if self.scip_threads>0:model.setParam('parallel/maxnthreads',int(self.scip_threads))
    from pyscipopt import SCIP_PARAMSETTING
    model.setHeuristics(SCIP_PARAMSETTING.FAST)
    for heuristic in ('adaptivediving','indicatordiving','farkasdiving','conflictdiving'):
        model.setParam(f'heuristics/{heuristic}/freq',-1)
    # Limit root separation work, not the validity of the formulation.
    model.setParam('separating/maxroundsroot',8)
    model.setParam('separating/maxstallroundsroot',3)
    model.setParam('branching/relpscost/initcand',10)
    model.setParam('branching/relpscost/inititer',100)
    model.setParam('branching/relpscost/maxlookahead',3)
    model.setParam('branching/relpscost/sbiterofs',1000)
    if feasibility_only:
        model.setParam('separating/maxroundsroot',2)
        model.setParam('separating/maxstallroundsroot',1)
        model.setParam('separating/maxcutsroot',100)
    x={(i,j):model.addVar(vtype='B',name=f'x_{i}_{j}') for i,j in direct}
    r={(i,j):model.addVar(vtype='B',name=f'reload_{i}_{j}') for i,j in reloads}
    y={(i,v['idx']):model.addVar(vtype='B',name=f'veh_{i}_{v["idx"]}') for i in C for v in V}
    used={v['idx']:model.addVar(vtype='B',name=f'used_{v["idx"]}') for v in V}
    first={i:model.addVar(vtype='B',name=f'first_{i}') for i in C}
    last={i:model.addVar(vtype='B',name=f'last_{i}') for i in C}
    tau={i:model.addVar(lb=windows[i][0],ub=max(windows[i]),name=f'tau_{i}') for i in C}
    dep={i:model.addVar(lb=op0,ub=op1,name=f'dep_{i}') for i in C}
    wait={i:model.addVar(lb=0,ub=H,name=f'wait_{i}') for i in C}
    load={i:model.addVar(lb=demand[i],ub=max(Q,demand[i]),name=f'load_{i}') for i in C}
    rank={} if svc>1e-6 else {i:model.addVar(lb=1,ub=max(1,n),name=f'rank_{i}') for i in C}
    qtime={(i,v['idx']):model.addVar(lb=0,ub=H,name=f'work_{i}_{v["idx"]}') for i in C for v in V}
    maxD={i:max([float(dd[j,i]) for j in [0]+C if j!=i])+float(dd[i,0]) for i in C}
    qdist={} if feasibility_only else {(i,v['idx']):model.addVar(lb=0,ub=maxD[i],name=f'dist_{i}_{v["idx"]}') for i in C for v in V}
    starts={(i,v['idx']):model.addVar(lb=0,ub=1,name=f'start_{i}_{v["idx"]}') for i in C for v in V}
    ends={(i,v['idx']):model.addVar(lb=0,ub=1,name=f'end_{i}_{v["idx"]}') for i in C for v in V}
    incoming={i:[] for i in C};outgoing={i:[] for i in C}
    for arcs in (x,r):
        for (i,j),var in arcs.items():outgoing[i].append(var);incoming[j].append(var)
    begin={i:first[i]+qs(r[j,k] for j,k in reloads if k==i) for i in C}
    finish={i:last[i]+qs(r[j,k] for j,k in reloads if j==i) for i in C}
    work={};dist={}
    labels={i:qs((k+1)*y[i,v['idx']] for k,v in enumerate(V)) for i in C}
    for i in C:
        if windows[i][0]>windows[i][1]:model.addCons(tau[i]<=windows[i][1])
        model.addCons(first[i]+qs(incoming[i])==1)
        model.addCons(last[i]+qs(outgoing[i])==1)
        model.addCons(qs(y[i,v['idx']] for v in V)==1)
        model.addCons(load[i]<=qs(float(v['capacity'])*y[i,v['idx']] for v in V))
        model.addCons(load[i]<=demand[i]+Q*(1-begin[i]))
        # Start-of-trip arrival and waiting; dep can be chosen within depot horizon.
        M=max(1.,abs(windows[i][0]-op1-tt[0,i]),abs(windows[i][1]-op0-tt[0,i]))+H
        model.addCons(wait[i]>=tau[i]-dep[i]-tt[0,i]-M*(1-begin[i]))
        model.addCons(wait[i]<=tau[i]-dep[i]-tt[0,i]+M*(1-begin[i]))
        model.addCons(tau[i]+svc+tt[i,0]<=op1+max(0.,windows[i][1]+svc+tt[i,0]-op1)*(1-finish[i]))
        dist[i]=qs(float(dd[j,k])*var for (j,k),var in x.items() if k==i)+float(dd[0,i])*begin[i]+float(dd[i,0])*finish[i]
        work[i]=svc+wait[i]+qs(float(tt[j,k])*var for (j,k),var in x.items() if k==i)+float(tt[0,i])*begin[i]+float(tt[i,0])*finish[i]
        model.addCons(qs(qtime[i,v['idx']] for v in V)==work[i])
        if qdist:model.addCons(qs(qdist[i,v['idx']] for v in V)==dist[i])
        for v in V:
            vi=v['idx'];a=y[i,vi]
            model.addCons(a<=used[vi])
            model.addCons(qtime[i,vi]<=H*a)
            if qdist:model.addCons(qdist[i,vi]<=maxD[i]*a)
            for z,indicator in ((starts[i,vi],first[i]),(ends[i,vi],last[i])):
                model.addCons(z<=a);model.addCons(z<=indicator);model.addCons(z>=a+indicator-1)
    for i,j in pairs:
        edge=x.get((i,j),0)+r.get((i,j),0)
        model.addCons(labels[i]-labels[j]<=max(0,len(V)-1)*(1-edge))
        model.addCons(labels[j]-labels[i]<=max(0,len(V)-1)*(1-edge))
        if rank:model.addCons(rank[j]>=rank[i]+1-n*(1-edge))
    for (i,j),var in x.items():
        # Exact customer waiting on a selected direct edge.
        delta=tau[j]-tau[i]-svc-float(tt[i,j])
        M=max(1.,abs(windows[j][0]-windows[i][1]-svc-tt[i,j]),abs(windows[j][1]-windows[i][0]-svc-tt[i,j]))+H
        model.addCons(wait[j]>=delta-M*(1-var));model.addCons(wait[j]<=delta+M*(1-var))
        model.addCons(load[j]>=load[i]+demand[j]-(Q+demand[j])*(1-var))
    for (i,j),var in r.items():
        M=max(0.,windows[i][1]+svc+tt[i,0]-op0)
        model.addCons(dep[j]>=tau[i]+svc+float(tt[i,0])-M*(1-var))
    for v in V:
        vi=v['idx']
        model.addCons(qs(starts[i,vi] for i in C)==used[vi])
        model.addCons(qs(ends[i,vi] for i in C)==used[vi])
        model.addCons(qs(qtime[i,vi] for i in C)<=H*used[vi])
        model.addCons(qs(qtime[i,vi] for i in C)>=float(p['minimum_usage'])*used[vi])
        if p['must_use_all_vehicles']:model.addCons(used[vi]==1)
    count=qs(used.values());totalD=qs(dist.values());totalT=qs(work.values())
    fixedcost=qs(float(v['fixed_cost'])*used[v['idx']] for v in V) if p['fixed_cost_mode']=='WHEN_USED' else sum(fixed)
    variablecost=0 if feasibility_only else qs(float(v['cost_per_km'])*qdist[i,v['idx']] for i in C for v in V)
    totalCost=fixedcost+variablecost
    if max_active_vehicles is not None:model.addCons(count<=int(max_active_vehicles))
    model.addCons(count>=int(b['vehicles_lb']))
    model.addCons(qs(begin.values())>=int(b['trips_lb']))
    model.addCons(totalD>=float(b['distance_lb']))
    model.addCons(totalT>=float(b['time_lb']))
    if not feasibility_only:model.addCons(totalCost>=float(b['cost_lb']))
    objective=sum(float(coeff.get(k,0))*expr for k,expr in [('DISTANCE',totalD),('TIME',totalT),('COST',totalCost),('VEHICLES',count)])
    if not feasibility_only and incumbent_ub is not None and self.use_incumbent_cutoff:
        model.addCons(objective<=float(incumbent_ub)+max(1e-8,abs(incumbent_ub)*1e-9))
    model.setObjective(0 if feasibility_only else objective,'minimize')
    self._model_stats={'formulation':'CUSTOMER_CHAIN_DEPOT_RELOAD_V52','feasibility_only':feasibility_only,
        'max_active_vehicles':max_active_vehicles,'binary_arc_variables':len(x)+len(r),
        'binary_assignment_variables':len(y),'binary_vehicle_variables':len(used),
        'binary_endpoint_variables':len(first)+len(last),'strengthened_analytical_bounds':b,
        'capacity_dominance_applied':dominance,'available_vehicle_ids':[v['vehicle_id'] for v in all_vehicles],
        'modeled_vehicle_ids':[v['vehicle_id'] for v in V],
        'dominance_scope':'FEASIBILITY_ONLY_UNIFORM_OPERATIONAL_CONTRACT' if dominance else None,
        'root_separation_round_limit':2 if feasibility_only else 8,
        'heuristics':'FAST_ROUNDING_WITHOUT_DIVING',
        'incumbent_cutoff_used':bool(not feasibility_only and incumbent_ub is not None and self.use_incumbent_cutoff)}
    self.shared_memory.data['certification_strengthened_bounds']=copy.deepcopy(b)
    h=dict(C=C,V=V,x=x,r=r,y=y,used=used,first=first,last=last,tau=tau,dep=dep,wait=wait,
           load=load,rank=rank,qtime=qtime,qdist=qdist,starts=starts,ends=ends,
           total_distance=totalD,total_time=totalT,total_cost=totalCost,total_vehicles=count)
    return model,h

def _ct_available_actions(orchestrator,last=None):
    """Check actual scoped work before offering another budget."""
    if (last or {}).get('certificate',{}).get('proven_optimal'):
        return dict(options={},reason='Ótimo comprovado dentro da tolerância: não é necessário repetir a continuidade.')
    mem=orchestrator.memory.data;inputs=_pd_memory_inputs(orchestrator.memory);scope=_ct_scope(inputs)
    inputs.update(mem.get('continuity_options_v514',{}))
    state=mem.get('continuity_support_state_v53') or {}
    if state.get('fingerprint')!=scope:state={}
    search=_pd_copy.deepcopy(state.get('search_state_v58',{}))
    best=state.get('best_solution') or (last or {}).get('final_solution') or mem.get('best_incumbent') or mem.get('solver_result')
    if not best:return dict(options={},reason='Não há plano auditado disponível para continuar.')
    p=_v58_provider(inputs).problem;closed=set(search.get('closed_neighborhoods_v511',[]))
    regions=_ct_guided_regions(dict(inputs,user_continuity_mode='alternative'),p,best,closed,int(search.get('local_stalled',0)))
    lp=search.get('lp_continuity_v515',{})
    lb=not lp.get('exhausted') or ContinuityBoundPolicyV523().plan(inputs,search,1800)['run']
    k=int(best['summary']['num_vehicles_used'])>1 and not p['must_use_all_vehicles'] and state.get('k_status')!='PROVEN_INFEASIBLE'
    # An unchanged, exhausted portfolio is not another useful automatic round.
    improve=True  # Diversificação possui novas sementes mesmo após esgotar regiões exatas.
    stages=[]
    if lb:stages.append('LB_COMPACTO')
    if improve:stages.append('MELHORIA')
    if regions:stages.append('EXATO_LOCAL')
    if k:stages.append('ELIMINAR_VEICULO')
    options={}
    if stages:options['1']='Deixar o sistema escolher entre as buscas ainda disponíveis para melhorar o plano e sua garantia.'
    timed=[s for s in _ct_guidance(last)['timed_stages'] if s in stages]
    if timed:options['2']='Dar mais tempo às buscas que pararam no limite destinado a elas: '+', '.join(timed)+'.'
    if any(len(r[2])>=3 for r in regions):options['3']='Reorganizar grupos maiores: ainda existem grupos de 3/4 viagens não concluídos.'
    policy=CompleteModelPolicy()
    full_decision=policy.plan(p,search,inputs)
    minimum=full_decision['minimum_budget_seconds']
    if full_decision['run']:
        options['4']='Revisar o plano inteiro: procurar uma solução melhor e uma garantia mais forte, permitindo reorganizar todas as viagens. Tempo mínimo desta tentativa: '+_ct_time_words(minimum)+'.'
    recommendation='4' if '4' in options and not stages else '2' if '2' in options else '1' if '1' in options else '3' if '3' in options else None
    reason=('Já examinamos as pequenas reorganizações disponíveis sem encontrar outro ganho. O plano já utiliza apenas um veículo. Para continuar de uma forma diferente, podemos revisar o plano inteiro, se essa opção estiver disponível abaixo.' if not stages and not k and int(best['summary']['num_vehicles_used'])==1 else 'As opções abaixo foram verificadas contra a memória do problema atual.')
    reason+=' Revisão completa: '+policy.explain(full_decision)
    if not options:reason+=' Não há outra tarefa implementada elegível neste estado. A continuidade será encerrada; isso não prova ótimo.'
    return dict(options=options,recommended=recommendation,reason=reason,stages=stages,timed_stages=timed,minimum_full_seconds=minimum,complete_model_policy=full_decision)


def _ct_choose_next_round(orchestrator,last):
    if last:mostrar_relatorio_continuidade(last)
    info=_ct_guidance(last)
    if info['kind'] in ('OPTIMAL','EXECUTION_ISSUE'):
        print(_ct_guidance_text(last));return None
    available=_ct_available_actions(orchestrator,last)
    available=dict(available,options={k:_ct_friendly_label(v) for k,v in available['options'].items()})
    print('\nORIENTAÇÃO PARA A PRÓXIMA RODADA\n'+available['reason'])
    if not available['options']:return None
    choice=available['recommended']
    if choice:print('Recomendação: opção '+choice+'. '+available['options'][choice])
    for key,label in available['options'].items():print('['+key+'] '+label)
    print('[0] Encerrar e manter o plano aprovado. Nenhuma opção garante melhoria.')
    try:choice=UserInterface._ask_choice('Escolha: ',tuple(['0']+list(available['options'])))
    except (EOFError,KeyboardInterrupt):return None
    if choice=='0':return None
    return dict(mode={'1':'automatic','2':'more_time','3':'alternative','4':'complete_global'}[choice],
                timed_stages=available['timed_stages'],eligible_stages=available['stages'],
                minimum_full_seconds=available['minimum_full_seconds'],diagnosis=info['kind'])


def _ct_complete_global(self,inputs,baseline,seconds,folder,state):
    """Complete model, all available vehicles; no local dual bound is promoted."""
    import time,math,copy
    seconds=min(float(seconds),float(inputs.get('complete_model_plan',{}).get('cap_seconds',seconds)))
    started=time.monotonic();model=None;scope=_ct_scope(inputs)
    record=dict(status='RUNNING',assigned_seconds=float(seconds),fingerprint=scope)
    state['complete_model_v526']=record;_pd_write(folder/'search_state.json',state)
    class Complete(ContinuityExactCertificationAgent):
        def _analytical_lower_bounds(self,incumbent_ub=None):
            return dict(vehicles_lb=0,trips_lb=0,distance_lb=0.,time_lb=0.,cost_lb=0.,canonical_objective_lb=0.)
    Complete._build_scip_model=_v525_complete_builder
    agent=Complete(_PDSimpleNamespace(data=copy.deepcopy(inputs)),verbose_solver=False,time_limit_seconds=seconds)
    result=dict(solver_budget_seconds=seconds,global_bound_usable=False,termination_reason='NOT_RUN',bound_method='COMPLETE_CUSTOMER_CHAIN')
    try:
        agent._load_objective_context();agent._build_problem()
        decision=CompleteModelPolicy().plan(agent.problem,state,inputs)
        # The parent already authorizes the attempt; this check only guards resources.
        if decision['estimated_memory_mb']>decision['memory_limit_mb']:raise ValueError(CompleteModelPolicy().explain(decision))
        agent.relative_gap_target=0.;agent.use_incumbent_cutoff=False
        model,h=agent._build_scip_model()
        record.update(build_seconds=time.monotonic()-started,variables=model.getNVars(),constraints=model.getNConss(),memory_limit_mb=decision['memory_limit_mb'])
        model.setParam('limits/memory',decision['memory_limit_mb'])
        # SCIP exposes the variable part and the constant offset separately.
        floor=inputs.get('lb_start_v515',{})
        checked=_pd_audit(inputs,baseline)
        if not checked.get('feasible'):raise ValueError('Plano inicial reprovado pela auditoria.')
        ub=float(checked['recomputed']['canonical_objective_score'])
        if floor.get('certified') and floor.get('fingerprint')==scope:
            lb=float(floor['value'])
            if not math.isfinite(lb) or not 0<=lb<=ub+1e-7:raise ValueError('Piso incompatível com o plano.')
            model.addCons(model.getObjective()+model.getObjoffset()>=max(0.,lb-1e-7))
        agent._try_add_incumbent_mip_start(model,h,baseline)
        model.setParam('limits/time',max(.01,seconds-(time.monotonic()-started)))
        model.setParam('limits/gap',0.);model.setParam('parallel/maxnthreads',1)
        model.optimize();status=str(model.getStatus());record['status']=status
        result.update(status=status,termination_reason='TIME_LIMIT' if status=='timelimit' else status.upper(),nodes=model.getNNodes(),variables=model.getNVars(),constraints=model.getNConss())
        if status in ('infeasible','unbounded','inforunbd'):raise ValueError('Modelo completo contradiz o plano auditado ou está sem limite.')
        bound=float(model.getDualbound());winner=checked['normalized_solution']
        if model.getNSols():
            raw=agent._extract_solution(model,h);audit=_pd_audit(inputs,raw)
            if not audit.get('feasible'):raise ValueError('Candidato do modelo completo reprovado.')
            score=float(audit['recomputed']['canonical_objective_score'])
            if abs(score-float(model.getObjVal()))>1e-6*max(1.,abs(score)):raise ValueError('Objetivo reconstruído diverge do modelo.')
            if score<ub:winner=audit['normalized_solution'];ub=score
        if math.isfinite(bound) and abs(bound)<1e19:
            if bound>ub+1e-7:raise ValueError('Dual excede o plano auditado; limite rejeitado.')
            result.update(global_bound_usable=True,canonical_lower_bound=max(0.,bound-1e-7),objective_offset=model.getObjoffset())
        record.update(lower_bound=result.get('canonical_lower_bound'),upper_bound=ub,nodes=result['nodes'])
        return winner,result
    except Exception:
        record['status']='ERROR';raise
    finally:
        record['elapsed_seconds']=time.monotonic()-started
        _pd_write(folder/'search_state.json',state)
        if model is not None:model.freeProb()

ContinuityExactCertificationAgent.solve_complete_global=_ct_complete_global

class CompleteModelPolicy:
    """Count the actual sparse customer-chain formulation, not a dense per-vehicle arc model."""
    VERSION = 'V534_SPARSE_COUNTS_MEASURED_RSS'

    def plan(self, p, search, inputs, available=None):
        import math
        C = p['customers']
        V = p['vehicles']
        n = len(C)
        m = len(V)
        svc = float(p['service_time'])
        tt = p['travel_time']
        op0 = float(p['operation_start'])
        op1 = float(p['operation_end'])
        Q = max((float(v['capacity']) for v in V))
        windows = {i: (max(op0, float(p['windows'][i][0])), min(op1, float(p['windows'][i][1]))) for i in C}
        demand = {i: float(p['client_rows'][i]['demand']) for i in C}
        direct = {(i, j) for i in C for j in C if i != j and windows[i][0] + svc + tt[i, j] <= windows[j][1] + 1e-09 and (demand[i] + demand[j] <= Q + 1e-09)}
        reloads = {(i, j) for i in C for j in C if i != j and windows[i][0] + svc + tt[i, 0] + tt[0, j] <= windows[j][1] + 1e-09} if p['allow_multi_trip'] else set()
        variables = len(direct) + len(reloads) + 5 * m * n + 6 * n + m + (n if svc <= 1e-06 else 0)
        constraints = 10 * n + 9 * m * n + 2 * len(direct | reloads) + 3 * len(direct) + len(reloads) + 5 * m + 6
        if svc <= 1e-06:
            constraints += len(direct | reloads)
        if p['must_use_all_vehicles']:
            constraints += m
        memory = (4096 * variables + 2048 * constraints) / 1024 ** 2
        resource = _v534_memory_budget(inputs)
        limit = resource['solver_mb']
        ceiling = inputs.get('complete_model_max_seconds')
        if ceiling is not None:
            ceiling = float(ceiling)
            if not math.isfinite(ceiling) or ceiling <= 0:
                raise ValueError('Limite de tempo inválido.')
        old = search.get('complete_model_v526') or search.get('small_complete_v525') or {}
        previous = float(old.get('assigned_seconds', 0.0))
        minimum = max(30.0, 1.25 * previous + 12.0) if old else 30.0
        decision = dict(run=True, reason='AUTHORIZED_FULL_BUDGET', policy=self.VERSION, estimated_variables=variables, estimated_constraints=constraints, estimated_memory_mb=memory, memory_limit_mb=limit, estimate_is_conservative=True, minimum_budget_seconds=minimum, cap_seconds=None, configured_solver_ceiling=ceiling, finalization_reserve_seconds=12.0, resource_budget=resource, direct_arcs=len(direct), reload_arcs=len(reloads), estimator='SPARSE_CUSTOMER_CHAIN')
        if memory > limit:
            decision.update(run=False, reason='MEMORY_GUARD')
        elif old and str(old.get('status')).lower() == 'optimal':
            decision.update(run=False, reason='ALREADY_OPTIMAL')
        elif old and str(old.get('status')).lower() not in ('timelimit', 'memlimit', 'memory_guard', 'memory_limit'):
            decision.update(run=False, reason='REVIEW_EXECUTION')
        elif old and str(old.get('status')).lower() in ('memlimit', 'memory_limit') and (limit <= float(old.get('memory_limit_mb', limit))):
            decision.update(run=False, reason='MEMORY_GUARD')
        elif ceiling is not None and ceiling < minimum - 12.0:
            decision.update(run=False, reason='CONFIGURED_CEILING_PREVENTS_RETRY')
        elif old:
            decision['reason'] = 'LARGER_AUTHORIZED_ATTEMPT'
        if available is not None:
            available = float(available)
            if not math.isfinite(available) or available <= 0:
                raise ValueError('Orçamento de tempo inválido.')
            decision['cap_seconds'] = min(max(0.0, available - 12.0), ceiling) if ceiling is not None else max(0.0, available - 12.0)
            if decision['run'] and available < minimum:
                decision.update(run=False, reason='INSUFFICIENT_AUTHORIZED_BUDGET')
        return decision

    def explain(self, d):
        return {'AUTHORIZED_FULL_BUDGET': 'Podemos revisar o plano inteiro dentro do tempo autorizado, com memória monitorada e reserva para conferir e salvar.', 'LARGER_AUTHORIZED_ATTEMPT': 'A revisão anterior não concluiu a prova. Uma tentativa com recursos adicionais pode avançar, sem garantia de melhoria.', 'MEMORY_GUARD': 'A estimativa do modelo ou uma tentativa anterior exige mais memória que o orçamento atual. Outras buscas continuam disponíveis.', 'ALREADY_OPTIMAL': 'A melhor qualidade já foi comprovada dentro da tolerância.', 'REVIEW_EXECUTION': 'A revisão anterior teve uma interrupção ou falha que precisa ser conferida antes de repetir.', 'CONFIGURED_CEILING_PREVENTS_RETRY': 'O teto configurado de tempo impede uma tentativa maior.', 'INSUFFICIENT_AUTHORIZED_BUDGET': 'O tempo autorizado é menor que o mínimo para uma tentativa diferente.'}[d['reason']]

def _ct_time_words(seconds):
    seconds=max(0.,float(seconds));minutes=int(seconds//60);rest=seconds-minutes*60
    return (str(minutes)+' min e ' if minutes else '')+format(rest,'.1f').replace('.',',')+' s'


def _ct_budget_explanation(result):
    stats=result.get('statistics',{});requested=stats.get('requested_seconds');elapsed=stats.get('elapsed_seconds')
    if requested is None or elapsed is None:return 'O registro não contém os tempos necessários para comparar o orçamento com o uso real.'
    remaining=max(0.,float(requested)-float(elapsed))
    text='Você autorizou até **'+_ct_time_words(requested)+'**. Esta rodada utilizou **'+_ct_time_words(elapsed)+'**; ficaram **'+_ct_time_words(remaining)+'** sem uso. '
    stages=stats.get('stages',{});full=stages.get('EXATO_COMPLETO_GLOBAL',{})
    failures=[r for r in stages.values() if r.get('error')]
    if result.get('certificate',{}).get('proven_optimal'):
        text+='Encerramos porque a prova de ótimo foi concluída; gastar o restante do tempo não é necessário.'
    elif result.get('status')=='INTERRUPTED':
        text+='A execução foi interrompida. O tempo total não foi necessariamente consumido; preservamos os resultados que puderam ser verificados.'
    elif failures:
        text+='Uma tarefa encontrou um problema de execução. Não foi uma conclusão de que o plano não pode melhorar. Consulte o diagnóstico antes de repetir.'
    elif full.get('termination_reason')=='TIME_LIMIT':
        assigned=full.get('solver_budget_seconds')
        text+='A revisão do plano inteiro atingiu o limite destinado à busca'+(' (**'+_ct_time_words(assigned)+'**)' if assigned is not None else '')+'. '
        configured=full.get('budget_policy',{}).get('configured_solver_ceiling')
        if configured is not None:
            text+='Há um teto explícito de '+_ct_time_words(configured)+' na configuração; ele pode explicar o encerramento antes do orçamento total. '
        else:text+='Uma pequena parte do orçamento é reservada para preparar, conferir e salvar o resultado. '
        text+='A busca ficou sem concluir a prova, não sem alternativas matematicamente possíveis. Mais tempo pode ajudar, mas não garante avanço.'
    elif full.get('status')=='memlimit':
        text+='A revisão atingiu o limite de memória. O motivo não foi falta de tempo; aumentar apenas os minutos não remove esse limite.'
    elif full:
        text+='A tarefa de revisão encerrou antes do orçamento total. Seu motivo técnico foi '+str(full.get('termination_reason','não registrado'))+'. Isso não deve ser interpretado, isoladamente, como prova de ótimo.'
    elif remaining>2:
        text+='Encerramos porque a seleção atual de tarefas terminou ou não havia outra busca elegível para este estado. Não foi o seu orçamento total que acabou. A orientação abaixo informa se há uma forma diferente de continuar.'
    else:text+='A rodada chegou ao limite do orçamento autorizado; os resultados aprovados foram preservados.'
    text+=' O tempo autorizado é um máximo: a execução também pode parar antes por prova concluída, interrupção, falha ou falta de uma tarefa elegível.'
    return text


#  single presentation entry point; solver and proof rules are unchanged.
def apresentar_resultados_cliente(orchestrator):
    from IPython.display import display, Markdown
    mem=orchestrator.memory.data
    result=mem.get('continuity_integrated_result_v53')
    sol=(result or {}).get('final_solution') or mem.get('final_solution') or mem.get('solver_result')
    if not isinstance(sol,dict):
        print('Nenhum plano disponível. Execute o acionamento ou restaure um checkpoint.');return
    audit=_pd_audit(_pd_memory_inputs(orchestrator.memory),sol)
    if not audit.get('feasible'):
        raise ValueError('Apresentação bloqueada: o plano não passou na reauditoria.')
    if result and (result.get('issues') or not result.get('certificate',{}).get('report_allowed')):
        raise ValueError('Resultado da continuidade não liberado.')
    if result and result.get('checkpoint_origin')!='INITIAL_CORE':
        _v534_validate_result(result)  # Validate the certificate before rendering any visual.
    # All visual consumers use this same independently reconstructed plan.
    visual_solution=_pd_copy.deepcopy(sol)
    visual_solution.update(_pd_copy.deepcopy(audit['normalized_solution']))
    if result:
        visual_solution['certificate']=_pd_copy.deepcopy(result['certificate'])
        visual_solution['certification_level']=result['certificate']['certification_level']
        if not _pd_math.isclose(float(visual_solution['canonical_objective_score']),float(result['certificate']['upper_bound']),abs_tol=1e-7):
            raise ValueError('Plano e certificado têm objetivos diferentes.')
    mem['solver_result']=visual_solution
    diagnostic=orchestrator.agents['diagnostic'].executar_diagnostico(gerar_painel=True,gerar_mapa=True)
    for warning in (diagnostic or {}).get('visual_warnings',[]):
        print('Visualização indisponível:',warning)
    if result and result.get('checkpoint_origin')!='INITIAL_CORE':
        display(Markdown(_v534_customer_narrative(visual_solution,result['certificate'],mem=mem,result=result,incluir_plano=False)))
    else:
        cert=sol.get('certificate') or mem.get('optimality_certificate') or mem.get('certification_result',{}).get('certificate',{})
        text=_v534_customer_narrative(visual_solution,cert,mem=mem,incluir_plano=False)
        display(Markdown(text))
    if result and result.get('checkpoint_origin')!='INITIAL_CORE':
        display(Markdown(resumo_continuidade_cliente(result)))
    exibir_opcao_checkpoint(orchestrator)

def baixar_checkpoint_continuidade(orchestrator):
    """Download opcional para preservar a retomada após desligar o Colab."""
    return salvar_checkpoint_v514(orchestrator,'checkpoint_continuidade_v534.json',download=True)


def _v533_otimo_certificado(cert):
    return bool(cert.get('proven_optimal') or
        (cert.get('certification_level')=='A' and cert.get('optimal_within_tolerance')))

def _v533_explicar_certificado(cert):
    if cert.get('proven_optimal'):
        return 'Ótimo global certificado pelo modelo exato, dentro da tolerância numérica, para os dados, restrições e objetivo escolhidos.'
    if cert.get('certification_level')=='A' and cert.get('optimal_within_tolerance'):
        return ('Ótimo global certificado dentro da tolerância numérica (nível A): o limite inferior válido e o valor do plano auditado coincidem dentro dessa tolerância. '
                'Esta garantia pode vir dos limites matemáticos, mesmo sem executar o solver global exato. Não é necessário ampliar a busca para o mesmo problema e objetivo.')
    if cert.get('certification_level')=='A':
        return 'Certificação A registrada, mas a confirmação de ótimo dentro da tolerância está ausente. Consulte o certificado técnico antes de interpretar este resultado como ótimo.'
    return 'O ótimo global ainda não foi comprovado. O GAP expressa a incerteza da garantia matemática; não é uma promessa de economia.'

def exibir_opcao_checkpoint(orchestrator):
    from IPython.display import display,Markdown
    display(Markdown('**Guardar esta sessão:** o checkpoint permite recuperar os dados, plano e certificado em outra sessão. O download é opcional.'))
    try:
        import ipywidgets as widgets
    except ImportError:
        print('Para baixar, execute: baixar_checkpoint_continuidade(orchestrator)');return
    button=widgets.Button(description='Baixar checkpoint',icon='download',button_style='info')
    output=widgets.Output()
    def download_clicked(_):
        with output:
            output.clear_output(wait=True)
            try:baixar_checkpoint_continuidade(orchestrator)
            except Exception as exc:print('Não foi possível baixar o checkpoint:',exc)
    button.on_click(download_clicked)
    display(button,output)
    # Keep widgets reachable for interactive callbacks.
    orchestrator._checkpoint_download_widgets=(button,output)


````

#### Reconstrução de viagens

Retira e reposiciona grupos de clientes para experimentar novas combinações de viagens. Também revisa a ordem dos atendimentos e reutiliza alternativas úteis. Um novo plano só é adotado quando melhora o objetivo escolhido e passa pela auditoria.


````python
def _v535_lns(inputs,best,seconds,folder,seed,remember_routes=True):
    import time,random,copy,math
    route_memory={}
    """Destroy/repair with related or expensive clients and controlled uphill acceptance.
    No local objective or failed reconstruction is ever accepted as a global certificate.
    """
    deadline=time.monotonic()+seconds;rng=random.Random(seed)
    provider=_v58_provider(inputs);p=provider.problem;coeff=provider.objective_context['coefficients']
    capacities={v['vehicle_id']:float(v['capacity']) for v in p['vehicles']}
    vids=list(capacities);demand={i:p['client_rows'][i]['demand'] for i in p['customers']}
    current=copy.deepcopy(best);initial=float(best['canonical_objective_score']);attempts=accepted=improved=0
    rejected={};trace=[];last=0;used={r['vehicle_id'] for r in best['routes']}
    vehicle_candidates=sorted(p['vehicles'],key=lambda v:coeff.get('COST',0)*v['fixed_cost']+coeff.get('VEHICLES',0))
    while time.monotonic()<deadline-.2:
        attempts+=1;routes=_v58_routes(current)
        clients=[i for r in routes for i in r['path'][1:-1]]
        size=min(len(clients),rng.choice([3,4,6,8,10,12]))
        mode=attempts%4
        if mode==0:
            anchor=rng.choice(clients);removed=sorted(clients,key=lambda i:p['distance'][anchor,i]*(.8+rng.random()*.4))[:size]
        elif mode==1:
            scores={i:p['distance'][a,i]+p['distance'][i,b]-p['distance'][a,b] for r in routes for a,i,b in zip(r['path'],r['path'][1:],r['path'][2:])}
            removed=sorted(clients,key=lambda i:scores[i]*rng.uniform(.7,1.3),reverse=True)[:size]
        elif mode==2:
            r=rng.choice(routes);removed=r['path'][1:-1]
        else:removed=rng.sample(clients,size)
        removed=set(removed);partial=[dict(vehicle_id=r['vehicle_id'],path=[i for i in r['path'] if i not in removed]) for r in routes]
        partial=[r for r in partial if len(r['path'])>2]
        remaining=list(removed);rng.shuffle(remaining)
        # Prioritize large demands to avoid infeasible late insertions.
        remaining.sort(key=lambda i:-demand[i]+rng.random()*2)
        allowed=sorted({r['vehicle_id'] for r in partial}|used)
        if attempts%6==0:allowed=sorted(set(allowed)|{v['vehicle_id'] for v in vehicle_candidates[:2]})
        failed=False
        for client in remaining:
            if time.monotonic()>deadline-.1:failed=True;break
            options=[];base=_v58_surrogate(p,coeff,partial)
            for index,r in enumerate(partial):
                load=sum(demand[i] for i in r['path'][1:-1])
                if load+demand[client]>capacities[r['vehicle_id']]+1e-7:continue
                for pos in range(1,len(r['path'])):
                    path=r['path'];a,b=path[pos-1],path[pos]
                    # Necessary pairwise time compatibility; final audit checks the full schedule.
                    if a and p['windows'][a][0]+p['service_time']+p['travel_time'][a,client]>p['windows'][client][1]+1e-7:continue
                    if b and p['windows'][client][0]+p['service_time']+p['travel_time'][client,b]>p['windows'][b][1]+1e-7:continue
                    trial=copy.deepcopy(partial);trial[index]['path'].insert(pos,client)
                    score=_v58_surrogate(p,coeff,trial)
                    options.append((score-base,index,pos,False))
            for vid in allowed:
                if demand[client]>capacities[vid]+1e-7:continue
                if not p['allow_multi_trip'] and any(r['vehicle_id']==vid for r in partial):continue
                positions=[j for j,r in enumerate(partial) if r['vehicle_id']==vid]
                at=(positions[-1]+1) if positions else len(partial)
                trial=copy.deepcopy(partial);trial.insert(at,dict(vehicle_id=vid,path=[0,client,0]))
                score=_v58_surrogate(p,coeff,trial);options.append((score-base,at,vid,True))
            if not options:failed=True;break
            options.sort(key=lambda x:x[0]);chosen=options[0] if rng.random()<.75 else rng.choice(options[:min(3,len(options))])
            _,index,pos,new=chosen
            if new:partial.insert(index,dict(vehicle_id=pos,path=[0,client,0]))
            else:partial[index]['path'].insert(pos,client)
        if failed:continue
        candidate,reason=_v57_schedule(p,partial,min(.35,max(.01,deadline-time.monotonic())))
        if not candidate:rejected[reason]=rejected.get(reason,0)+1;continue
        audit=_pd_audit(inputs,candidate)
        if not audit['feasible']:rejected['AUDIT']=rejected.get('AUDIT',0)+1;continue
        _v536_remember_valid_paths(route_memory,audit['normalized_solution']) if remember_routes else None
        candidate=audit['normalized_solution'];value=float(candidate['canonical_objective_score']);old=float(current['canonical_objective_score'])
        temperature=max(.00001,initial*.015*(max(0.,deadline-time.monotonic())/seconds))
        if value<old or rng.random()<math.exp(min(0.,(old-value)/temperature)):
            current=candidate;accepted+=1
        if value<float(best['canonical_objective_score'])-1e-9:
            best=copy.deepcopy(candidate);improved+=1;_pd_write(folder/'candidate.json',best)
            trace.append(dict(seconds=seconds-(deadline-time.monotonic()),ub=value));current=copy.deepcopy(best)
        if attempts%80==0:current=copy.deepcopy(best)
        if time.monotonic()-last>5:
            _pd_write(folder/'progress.json',dict(attempts=attempts,improvements=improved,ub=best['canonical_objective_score']));_v534_store_route_memory(inputs,folder,route_memory,remember_routes);last=time.monotonic()
    _v534_store_route_memory(inputs,folder,route_memory,remember_routes)
    return best,dict(attempts=attempts,accepted=accepted,improvements=improved,rejections=rejected,trace=trace,seed=seed)
def _v535_regret(inputs,best,seconds,folder,seed,remember_routes=True):
    import time,random,copy,math
    route_memory={}
    start=time.monotonic();deadline=start+seconds;rng=random.Random(seed);p=_v58_provider(inputs).problem
    coeff=inputs['objective_context']['coefficients'];caps={v['vehicle_id']:v['capacity'] for v in p['vehicles']};dem={i:p['client_rows'][i]['demand'] for i in p['customers']}
    current=copy.deepcopy(best);trace=[];attempts=accepted=0;rejects={}
    def options(partial,i,noise):
        candidates=[];base=_v58_surrogate(p,coeff,partial)
        for j,r in enumerate(partial):
            if sum(dem[k] for k in r['path'][1:-1])+dem[i]>caps[r['vehicle_id']]+1e-8:continue
            for pos in range(1,len(r['path'])):
                a,b=r['path'][pos-1:pos+1]
                if a and p['windows'][a][0]+p['service_time']+p['travel_time'][a,i]>p['windows'][i][1]+1e-8:continue
                if b and p['windows'][i][0]+p['service_time']+p['travel_time'][i,b]>p['windows'][b][1]+1e-8:continue
                trial=copy.deepcopy(partial);trial[j]['path'].insert(pos,i)
                delta=_v58_surrogate(p,coeff,trial)-base;candidates.append((delta*(1+noise*rng.uniform(-1,1)),j,pos,False))
        # Include every actual vehicle: alternative fleet configurations are permitted.
        for vid in caps:
            if dem[i]>caps[vid]+1e-8 or (not p['allow_multi_trip'] and any(r['vehicle_id']==vid for r in partial)):continue
            positions=[j for j,r in enumerate(partial) if r['vehicle_id']==vid];at=positions[-1]+1 if positions else len(partial)
            trial=copy.deepcopy(partial);trial.insert(at,dict(vehicle_id=vid,path=[0,i,0]));delta=_v58_surrogate(p,coeff,trial)-base
            candidates.append((delta*(1+noise*rng.uniform(-1,1)),at,vid,True))
        return sorted(candidates)
    while time.monotonic()<deadline-.5:
        attempts+=1;routes=_v58_routes(current);clients=[i for r in routes for i in r['path'][1:-1]]
        size=min(len(clients),rng.choice([6,10,14,18,22]));mode=attempts%4
        if mode==0:
            anchor=rng.choice(clients);removed=set(sorted(clients,key=lambda i:p['distance'][anchor,i]*rng.uniform(.8,1.2))[:size])
        elif mode==1:
            selected=rng.sample(routes,min(2,len(routes)));removed={i for r in selected for i in r['path'][1:-1]}
        elif mode==2:
            score={i:p['distance'][a,i]+p['distance'][i,b]-p['distance'][a,b] for r in routes for a,i,b in zip(r['path'],r['path'][1:],r['path'][2:])}
            removed=set(sorted(clients,key=lambda i:score[i]*rng.uniform(.7,1.3),reverse=True)[:size])
        else:removed=set(rng.sample(clients,size))
        partial=[dict(vehicle_id=r['vehicle_id'],path=[i for i in r['path'] if i not in removed]) for r in routes];partial=[r for r in partial if len(r['path'])>2]
        noise=rng.choice([0.,.1,.25]);failed=False
        while removed:
            if time.monotonic()>deadline-.35:failed=True;break
            choices=[]
            for i in sorted(removed):
                opts=options(partial,i,noise)
                if not opts:failed=True;break
                # Regret-3: insert first the client whose alternatives are most costly.
                regret=sum(opts[min(k,len(opts)-1)][0]-opts[0][0] for k in (1,2))
                choices.append((regret,-opts[0][0],i,opts[0]))
                if time.monotonic()>deadline-.35:failed=True;break
            if failed:break
            _,_,i,(_,j,pos,new)=max(choices,key=lambda r:r[:3]);removed.remove(i)
            if new:partial.insert(j,dict(vehicle_id=pos,path=[0,i,0]))
            else:partial[j]['path'].insert(pos,i)
        if failed:continue
        candidate,reason=_v57_schedule(p,partial,min(.5,max(.01,deadline-time.monotonic())))
        if not candidate:rejects[reason]=rejects.get(reason,0)+1;continue
        audit=_pd_audit(inputs,candidate)
        if not audit['feasible']:rejects['AUDIT']=rejects.get('AUDIT',0)+1;continue
        _v536_remember_valid_paths(route_memory,audit['normalized_solution']) if remember_routes else None
        candidate=audit['normalized_solution'];value=candidate['canonical_objective_score'];old=current['canonical_objective_score']
        temperature=max(1e-6,best['canonical_objective_score']*.012*max(0.,(deadline-time.monotonic())/seconds))
        if value<old or rng.random()<math.exp(min(0.,(old-value)/temperature)):current=candidate;accepted+=1
        if value<best['canonical_objective_score']-1e-9:
            best=copy.deepcopy(candidate);current=copy.deepcopy(best);trace.append(dict(seconds=time.monotonic()-start,ub=value));_pd_write(folder/'candidate.json',best)
        if attempts%25==0:current=copy.deepcopy(best)
        _pd_write(folder/'progress.json',dict(attempts=attempts,accepted=accepted,ub=best['canonical_objective_score']))
    _v534_store_route_memory(inputs,folder,route_memory,remember_routes)
    return best,dict(attempts=attempts,accepted=accepted,trace=trace,rejections=rejects,global_bound_usable=False)
def _v535_trip_orders(inputs,best,seconds):
    import time,random,copy,math
    """Held-Karp exact travel order for fixed trip memberships, with final full-plan audit.
    It proves only the shortest order for those customers, never global routing optimality.
    """
    deadline=time.monotonic()+seconds;p=_v58_provider(inputs).problem;routes=_v58_routes(best);proofs=[]
    for route in routes:
        clients=route['path'][1:-1];n=len(clients)
        if n>14 or time.monotonic()>deadline-.2:continue
        dp={(1<<j,j):(p['distance'][0,i],None) for j,i in enumerate(clients)};complete=True
        for mask in range(1,1<<n):
            if time.monotonic()>deadline-.2:complete=False;break
            for j,i in enumerate(clients):
                if (mask,j) not in dp:continue
                value=dp[mask,j][0]
                for k,h in enumerate(clients):
                    if mask&(1<<k):continue
                    key=(mask|(1<<k),k);nextvalue=value+p['distance'][i,h]
                    if nextvalue<dp.get(key,(math.inf,None))[0]:dp[key]=(nextvalue,j)
        if not complete:continue
        mask=(1<<n)-1;last=min(range(n),key=lambda j:dp[mask,j][0]+p['distance'][clients[j],0]);distance=dp[mask,last][0]+p['distance'][clients[last],0];path=[]
        while last is not None:
            path.append(clients[last]);previous=dp[mask,last][1];mask^=1<<last;last=previous
        path.reverse();old=sum(p['distance'][a,b] for a,b in zip(route['path'],route['path'][1:]))
        route['path']=[0]+path+[0];proofs.append(dict(vehicle=route['vehicle_id'],customers=clients,old_distance=old,optimal_order_distance=distance,restricted_distance_order_optimal=True))
    candidate,reason=_v57_schedule(p,routes,max(.01,min(2.,deadline-time.monotonic())))
    if candidate:
        audit=_pd_audit(inputs,candidate)
        if audit['feasible'] and audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:best=audit['normalized_solution']
    return best,dict(trip_order_proofs=proofs,schedule_status=reason,global_bound_usable=False)
# Executed at the end of the existing definitions cell, before orchestration.
_v535_previous_local=_v58_local_search
_v535_previous_exact=_v59_exact_search
_v535_previous_budget=_v511_stage_budget
_v535_previous_policy=ContinuityBoundPolicyV523

def _v535_local(inputs,best,seconds,folder,state):
    import time,copy
    deadline=time.monotonic()+seconds;folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    checked=_pd_audit(inputs,best)
    if not checked.get('feasible'):raise ValueError('A diversificação exige um plano auditado.')
    best=checked['normalized_solution'];before=best['canonical_objective_score'];runs=[]
    if seconds<6 or before<=1e-12:return _v535_previous_local(inputs,best,seconds,folder,state)
    seed=int(state.get('diversification_round_v535',0));state['diversification_round_v535']=seed+1
    tasks=[('LOCAL',_v535_previous_local,min(5.,seconds*.1)),('LNS',_v535_lns,None),('REGRET3',_v535_regret,None)]
    for name,task,quota in tasks:
        left=deadline-time.monotonic()-.3
        if left<.5:break
        budget=min(left,quota if quota is not None else left*.70 if name=='LNS' else left)
        branch=folder/name;branch.mkdir(exist_ok=True);start=time.monotonic();old=best['canonical_objective_score']
        candidate,stats=task(inputs,copy.deepcopy(best),budget,branch,state) if name=='LOCAL' else task(inputs,copy.deepcopy(best),budget,branch,8131+1009*seed)
        audit=_pd_audit(inputs,candidate)
        if audit.get('feasible') and audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:
            best=audit['normalized_solution'];_pd_write(folder/'candidate.json',best)
        record=state.setdefault('operators',{}).setdefault(name,dict(attempts=0,gain=0.,seconds=0.))
        record['attempts']+=stats.get('attempts',0);record['gain']+=old-best['canonical_objective_score'];record['seconds']+=time.monotonic()-start
        runs.append(dict(operator=name,budget_seconds=budget,stats=stats,before_ub=old,after_ub=best['canonical_objective_score']))
        _pd_write(folder/'search_state.json',state)
    return best,dict(termination_reason='DIVERSIFIED_TIME_LIMIT',attempts=sum(r['stats'].get('attempts',0) for r in runs),
        improvements=int(best['canonical_objective_score']<before-1e-9),operators=state.get('operators',{}),diversification_runs=runs,global_bound_usable=False)

def _v535_exact(inputs,best,seconds,folder,state):
    import time
    start=time.monotonic();folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    # Exact fixed-membership distance orders supplement, rather than replace, integer block optimization.
    original=best['canonical_objective_score'];candidate,order_stats=_v535_trip_orders(inputs,best,min(8.,max(.1,seconds*.10)))
    if candidate['canonical_objective_score']<original-1e-9:_pd_write(folder/'candidate.json',candidate)
    _pd_write(folder/'trip_order_proofs.json',order_stats)
    best,stats=_v535_previous_exact(inputs,candidate,max(.1,seconds-(time.monotonic()-start)),folder,state)
    stats.update(order_stats);stats['global_bound_usable']=False
    return best,stats

def _v535_highs(self,data,seconds,label,folder,baseline_lb):
    import time,math,numpy as np,warnings
    from scipy.optimize import linprog,OptimizeWarning
    start=time.monotonic()
    with warnings.catch_warnings():
        warnings.simplefilter('ignore',OptimizeWarning)
        answer=linprog(data['c'],A_ub=data['A_ub'],b_ub=data['b_ub'],A_eq=data['A_eq'],b_eq=data['b_eq'],
            bounds=list(zip(data['lb'],data['ub'])),method='highs',options={'time_limit':max(.01,seconds-.1),'threads':1,'dual_feasibility_tolerance':1e-8,'primal_feasibility_tolerance':1e-8})
    row=dict(phase=label,engine='SciPy/HiGHS',matrix_hash=data['matrix_hash'],variables=len(data['c']),nodes=None,
        model_kind='CONTINUOUS_LP',status='OPTIMAL' if answer.success else 'NOT_OPTIMAL_WITHIN_BUDGET',usable=False,certified_lb=None)
    vector=None
    if answer.success:
        yu=np.minimum(np.asarray(answer.ineqlin.marginals),0.);ye=np.asarray(answer.eqlin.marginals)
        residual=data['c']-data['A_ub'].T@yu-data['A_eq'].T@ye
        support=np.where(residual>=0,residual*data['lb'],residual*data['ub'])
        dual=float(data['b_ub']@yu+data['b_eq']@ye+np.sum(support)+data['offset']);primal=float(data['c']@answer.x+data['offset'])
        magnitude=float(np.abs(data['b_ub'])@np.abs(yu)+np.abs(data['b_eq'])@np.abs(ye)+np.sum(np.abs(support))+abs(data['offset']))
        margin=1e-7*max(1.,abs(dual))+1e-12*max(1.,magnitude)
        error=max(float(np.max(data['A_ub']@answer.x-data['b_ub'],initial=0)),float(np.max(np.abs(data['A_eq']@answer.x-data['b_eq']),initial=0)),float(np.max(data['lb']-answer.x,initial=0)),float(np.max(answer.x-data['ub'],initial=0)))
        if math.isfinite(dual) and dual<=primal+1e-7 and error<=1e-5:
            row.update(usable=True,certified_lb=max(0.,dual-margin),dual_objective=dual,primal_objective=primal,dual_margin=margin,primal_max_violation=error,dual_verification='BOX_CORRECTED_DUAL')
            vector=answer.x
            np.savez_compressed(_PDPath(folder)/'highs_dual_v535.npz',yu=yu,ye=ye,residual=residual,support=support,primal=vector)
        else:row['status']='NUMERICAL_CHECK_FAILED'
    row.update(seconds=time.monotonic()-start,bound_before=baseline_lb,bound_after=max(baseline_lb,row['certified_lb'] or 0))
    _pd_write(_PDPath(folder)/'highs_phase_v535.json',row)
    return row,vector

class ContinuityBoundPolicyV523(_v535_previous_policy):
    VERSION='V535_SCOPED_GLOBAL_CUTS_AND_HIGHS'
    STRATEGY={'implementation':'basis_and_highs_v535','family':'ALL_CHEAP','round_limit':30}
    def memory(self,inputs,search):
        self._tier=int(inputs.get('lb_tier_v535',30));return super().memory(inputs,search)
    def key(self):return self.VERSION+'_'+str(getattr(self,'_tier',30))
    def plan(self,inputs,search,available):
        # A larger authorized budget is a different attempt, reusing accepted scoped cuts.
        target=min(600. if inputs.get('user_continuity_mode')=='more_time' else 300.,max(0.,available*.5))
        tiers=[n for n in (30,60,120,180,300,600) if n<=target and n<available-2]
        if not tiers:return dict(run=False,cap_seconds=0.,reason='DEFERRED_INSUFFICIENT_BOUND_BUDGET')
        inputs['lb_tier_v535']=max(tiers);memo=self.memory(inputs,search);entry=memo['attempts'].get(self.key())
        if entry:return dict(run=False,cap_seconds=0.,reason='SAME_BOUND_STRATEGY_AND_BUDGET_ALREADY_ATTEMPTED')
        return dict(run=True,cap_seconds=float(self._tier),reason='V535_EXTENDED_SCOPED_BOUND')

def _v535_bound(self,inputs,seconds,folder,state):
    import time,math,json,numpy as np
    from pyscipopt import SCIP_BASESTAT
    folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True);start=time.monotonic();deadline=start+seconds;scope=_ct_scope(inputs)
    imported=inputs.get('lb_start_v515',{})
    if imported.get('fingerprint')!=scope or not imported.get('certified'):raise ValueError('LB de partida sem escopo validado.')
    before=float(imported['value']);ub=float(inputs['lb_audited_ub_v523'])
    if not math.isfinite(before) or not math.isfinite(ub) or before<0 or before>ub+1e-7:raise ValueError('Intervalo inicial inválido.')
    policy=ContinuityBoundPolicyV523();memo=policy.memory(inputs,state);entry=memo['attempts'].get(policy.key())
    if not entry or entry.get('token')!=inputs.get('lb_launch_token_v523'):raise ValueError('Tentativa de prova não autorizada pelo coordenador.')
    saved=state.setdefault('global_cuts_v535',dict(fingerprint=scope,cuts=[]))
    if saved.get('fingerprint')!=scope:raise ValueError('Cortes salvos pertencem a outro problema.')
    built=self._build_reference_lp_v5141(inputs);model=built['model'];adder=OwnerConsistencyCutExperimentAgent(globals())
    cuts=list(saved['cuts']);keys={json.dumps({k:v for k,v in c.items() if k!='violation'},sort_keys=True) for c in cuts}
    passes=[];best=before;lp_best=None;previous=None;termination='BUDGET';data=None
    def snapshot(status):
        entry.update(status=status,after_lb=best,effective_gain=max(0.,best-before),lp_lb=lp_best,phases=passes,termination=termination)
        saved['cuts']=cuts
        if status=='COMPLETED':
            legacy=state.setdefault('lp_continuity_v515',{})
            legacy.update(scope=scope,exhausted=True,lower_bound=best,completed_strategy_v535=True)
        answer=dict(agent='BoundStrengtheningAgent',fingerprint=scope,bound_version=policy.VERSION,global_bound_usable=True,
            canonical_lower_bound=best,bound_before=before,bound_gain=best-before,passes=passes,model_kind='CONTINUOUS_LP',
            integer_phase_executed=False,root_lp_completed=any(r.get('usable') for r in passes),termination_reason=termination,
            elapsed_seconds=time.monotonic()-start,completed=status=='COMPLETED')
        _pd_write(folder/'search_state.json',state);_pd_write(folder/'bound_checkpoint.json',answer)
        return answer
    def accept(row):
        nonlocal best,lp_best
        if row.get('usable'):
            value=float(row['certified_lb'])
            if not math.isfinite(value) or value>ub+1e-7:raise ValueError('LB rejeitado: incompatível com o plano auditado.')
            lp_best=max(lp_best or 0.,value);best=max(best,value)
        passes.append(row);snapshot('RUNNING')
        _pd_write(folder/'lb_live.json',dict(phase=row['phase'],fingerprint=scope,updated_at=time.time(),dual_bound=best))
    try:
        for c in cuts:adder.add(built,c)
        for iteration in range(30):
            left=deadline-time.monotonic()
            if left<.3:break
            data=self._matrix_lp_v5141(model)
            if previous is not None and '_completed_basis' in previous:
                n=len(previous['b_ub']);extra=len(data['b_ub'])-n
                compatible=extra>=0 and previous['names']==data['names'] and previous['offset']==data['offset']
                compatible=compatible and all(np.array_equal(previous[k],data[k]) for k in ('c','lb','ub','b_eq'))
                compatible=compatible and np.array_equal(previous['b_ub'],data['b_ub'][:n]) and (previous['A_eq']!=data['A_eq']).nnz==0 and (previous['A_ub']!=data['A_ub'][:n]).nnz==0
                if compatible:
                    cs,rs=previous['_completed_basis'];data['_warmstart_basis']=(list(cs),list(rs[:n])+[SCIP_BASESTAT.BASIC]*extra+list(rs[n:]))
            row,vector=self._solve_lp_v5141(data,max(.01,left-.2),f'V535_CUTS_{iteration}',folder,best);accept(row)
            if not row.get('usable') or vector is None:termination='LP_NOT_CONCLUDED';break
            previous=data
            if iteration==29:termination='ROUND_LIMIT';break
            fresh=[]
            for cut in adder.candidates(built,dict(zip(data['names'],vector)),'ALL_CHEAP',min(deadline-.2,time.monotonic()+4)):
                key=json.dumps({k:v for k,v in cut.items() if k!='violation'},sort_keys=True)
                if key not in keys:keys.add(key);fresh.append(cut)
            if not fresh:termination='NO_NEW_VIOLATED_CUT';break
            for cut in fresh:adder.add(built,cut);cuts.append(cut)
            snapshot('RUNNING')
        # An independent solver may remove more conservative residual correction in the first LP engine.
        left=deadline-time.monotonic()
        if data is not None and left>=8 and lp_best is not None:
            data=self._matrix_lp_v5141(model)
            row,vector=_v535_highs(self,data,left,'V535_HIGHS_RECHECK',folder,best);accept(row)
        _pd_write(folder/'global_cut_evidence_v535.json',dict(fingerprint=scope,cuts=cuts,passes=passes,lp_lb=lp_best))
        return snapshot('COMPLETED')
    except BaseException:
        snapshot('INTERRUPTED_OR_FAILED');raise
    finally:model.freeProb()

def _v535_budget(stage,proposed,total,previous):
    seconds,policy=_v535_previous_budget(stage,proposed,total,previous)
    if stage in ('MELHORIA','EXATO_LOCAL'):
        cap=min(180.,total*.40);return max(0.,min(proposed,cap)),dict(policy,cap_seconds=cap,reason='V535_DIVERSIFICATION_WITHIN_TOTAL_BUDGET')
    return seconds,policy

_v58_local_search=_v535_local
_v59_exact_search=_v535_exact
_v511_stage_budget=_v535_budget
BoundStrengtheningAgent.strengthen_lp_continuity_v515=_v535_bound
BoundStrengtheningAgent.CONTINUITY_BOUND_VERSION=ContinuityBoundPolicyV523.VERSION
_v535_binding_list=[_v511_stage_budget if value is _v535_previous_budget else value for value in _V513_BINDINGS]
_v535_binding_list[7]=_v59_exact_search;_V513_BINDINGS=tuple(_v535_binding_list)


````

#### Cooperação e redistribuição de frota

Combina alternativas de diferentes buscas e testa como distribuir as viagens entre os veículos disponíveis. Procura uma organização melhor dentro dos recursos autorizados.

````python
def _v536_fleet(inputs, best, seconds, folder, seed):
    """Any number of trips: restricted assignment MILP plus scheduling and full audit."""
    import time, random, copy
    from pyscipopt import Model, quicksum as qs
    start = time.monotonic()
    deadline = start + float(seconds)
    rng = random.Random(seed)
    folder = _PDPath(folder)
    folder.mkdir(parents=True, exist_ok=True)
    p = _v58_provider(inputs).problem
    coeff = inputs['objective_context']['coefficients']
    routes = _v58_routes(best)
    V = p['vehicles']
    R = range(len(routes))
    model = None
    detail = dict(method='FIXED_TRIP_ALL_VEHICLE_ASSIGNMENT', trips=len(routes), assignments=0, schedules=0, improvements=[], global_bound_usable=False)
    loads = [sum((p['client_rows'][i]['demand'] for i in r['path'][1:-1])) for r in routes]
    distances = [sum((p['distance'][a, b] for a, b in zip(r['path'], r['path'][1:]))) for r in routes]
    durations = [d * 60 / float(p['speed']) + (len(r['path']) - 2) * float(p['service_time']) for d, r in zip(distances, routes)]
    try:
        model = Model('FIXED_TRIP_FLEET_V534')
        model.hideOutput()
        model.setParam('parallel/maxnthreads', 1)
        model.setParam('limits/gap', 0.0)
        model.setParam('limits/memory', _v534_memory_budget(inputs)['solver_mb'])
        x = {(j, k): model.addVar(vtype='B', name=f'a_{j}_{k}') for j in R for k, v in enumerate(V) if loads[j] <= float(v['capacity']) + 1e-08}
        y = {k: model.addVar(vtype='B', name=f'use_{k}') for k in range(len(V))}
        for j in R:
            model.addCons(qs((x[j, k] for k in y if (j, k) in x)) == 1)
        for k in y:
            choices = [x[j, k] for j in R if (j, k) in x]
            for z in choices:
                model.addCons(z <= y[k])
            model.addCons(qs(choices) >= y[k])
            model.addCons(qs((durations[j] * x[j, k] for j in R if (j, k) in x)) <= float(p['max_time']) * y[k])
            if p['must_use_all_vehicles']:
                model.addCons(y[k] == 1)
        fixed = qs((float(v['fixed_cost']) * y[k] for k, v in enumerate(V))) if p['fixed_cost_mode'] == 'WHEN_USED' else sum((float(v['fixed_cost']) for v in V))
        cost = fixed + qs((distances[j] * float(V[k]['cost_per_km']) * z for (j, k), z in x.items()))
        surrogate = coeff.get('COST', 0) * cost + coeff.get('DISTANCE', 0) * sum(distances) + coeff.get('VEHICLES', 0) * qs(y.values()) + coeff.get('TIME', 0) * sum(durations)
        model.setObjective(surrogate, 'minimize')
        model.addCons(surrogate <= float(best['canonical_objective_score']) - 1e-09)
        detail.update(variables=model.getNVars(), constraints=model.getNConss(), build_seconds=time.monotonic() - start)
        while time.monotonic() < deadline - 0.12 and detail['assignments'] < 24:
            model.setParam('limits/time', max(0.01, (deadline - time.monotonic()) * 0.55))
            model.optimize()
            status = str(model.getStatus())
            detail['solver_status'] = status
            if not model.getNSols():
                break
            selected = [next((k for k in y if (j, k) in x and model.getVal(x[j, k]) > 0.5)) for j in R]
            detail['assignments'] += 1
            for mode in range(4):
                if time.monotonic() > deadline - 0.05:
                    break
                trial = [dict(vehicle_id=V[k]['vehicle_id'], path=list(r['path'])) for k, r in zip(selected, routes)]
                if mode == 1:
                    trial.sort(key=lambda r: min((float(p['windows'][i][1]) for i in r['path'][1:-1])))
                elif mode >= 2:
                    rng.shuffle(trial)
                detail['schedules'] += 1
                candidate, reason = _v57_schedule(p, trial, min(0.2, max(0.01, deadline - time.monotonic())))
                if candidate:
                    audit = _pd_audit(inputs, candidate)
                    if audit.get('feasible') and audit['recomputed']['canonical_objective_score'] < best['canonical_objective_score'] - 1e-09:
                        best = audit['normalized_solution']
                        detail['improvements'].append(dict(ub=best['canonical_objective_score'], vehicles=best['summary']['num_vehicles_used']))
                        _pd_write(folder / 'candidate.json', best)
            if status in ('memlimit', 'timelimit'):
                break
            model.freeTransform()
            model.addCons(qs((x[j, k] for j, k in enumerate(selected))) <= len(routes) - 1)
        detail['termination_reason'] = 'ASSIGNMENT_SEARCH_FINISHED'
    finally:
        detail['elapsed_seconds'] = time.monotonic() - start
        _pd_write(folder / 'fleet_progress.json', detail)
        if model is not None:
            model.freeProb()
    return (best, detail)
def _v536_pair(inputs,best,seconds,folder,seed):
    import time,random,copy,itertools,numpy as np
    deadline=time.monotonic()+seconds; rng=random.Random(seed); p=_v58_provider(inputs).problem
    coeff=inputs['objective_context']['coefficients']; V={v['vehicle_id']:v for v in p['vehicles']}
    detail=dict(pairs=0,partitions_screened=0,schedules=0,improvements=[],global_bound_usable=False)
    cache={}; seen=set(); pool_paths={}
    while time.monotonic()<deadline-2:
        routes=_v58_routes(best); pairs=list(itertools.combinations(range(len(routes)),2));rng.shuffle(pairs)
        pairs.sort(key=lambda ij:min(p['distance'][a,b] for a in routes[ij[0]]['path'][1:-1] for b in routes[ij[1]]['path'][1:-1]))
        advanced=False
        for ia,ib in pairs:
            if time.monotonic()>deadline-2:break
            clients=sorted(routes[ia]['path'][1:-1]+routes[ib]['path'][1:-1]); n=len(clients)
            key=(tuple(clients),routes[ia]['vehicle_id'],routes[ib]['vehicle_id'],tuple(routes[ia]['path']),tuple(routes[ib]['path']))
            if n>18 or key in seen:continue
            seen.add(key);advanced=True;detail['pairs']+=1
            size=1<<n; full=size-1; bits=np.arange(size,dtype=np.int32); cardinal=np.array([i.bit_count() for i in range(size)],dtype=np.int8)
            dp=np.full((size,n),np.inf); parent=np.full((size,n),-1,dtype=np.int8)
            D=np.array([[p['distance'][a,b] for b in clients] for a in clients]); depot=np.array([p['distance'][0,a] for a in clients]);back=np.array([p['distance'][a,0] for a in clients])
            for j in range(n):dp[1<<j,j]=depot[j]
            complete=True
            for card in range(2,n+1):
                for j in range(n):
                    if time.monotonic()>deadline-1:complete=False;break
                    masks=bits[(cardinal==card)&((bits&(1<<j))!=0)];prev=masks^(1<<j)
                    values=dp[prev,:]+D[:,j]; arg=np.argmin(values,axis=1)
                    dp[masks,j]=values[np.arange(len(masks)),arg];parent[masks,j]=arg
                if not complete:break
            if not complete:break
            values=dp+back;last=np.argmin(values,axis=1);distance=values[bits,last];distance[0]=0
            loads=np.zeros(size)
            demands=[p['client_rows'][i]['demand'] for i in clients]
            for mask in range(1,size):bit=mask&-mask;loads[mask]=loads[mask^bit]+demands[bit.bit_length()-1]
            va=routes[ia]['vehicle_id'];vb=routes[ib]['vehicle_id']
            wa=coeff.get('DISTANCE',0)+coeff.get('COST',0)*V[va]['cost_per_km'];wb=coeff.get('DISTANCE',0)+coeff.get('COST',0)*V[vb]['cost_per_km']
            score=distance*wa+distance[full^bits]*wb
            valid=(bits>0)&(bits<full)&(loads<=V[va]['capacity']+1e-8)&(loads[full^bits]<=V[vb]['capacity']+1e-8)
            old=_v58_surrogate(p,coeff,routes);oldpair=sum(p['distance'][a,b] for a,b in zip(routes[ia]['path'],routes[ia]['path'][1:]))*wa+sum(p['distance'][a,b] for a,b in zip(routes[ib]['path'],routes[ib]['path'][1:]))*wb
            masks=bits[valid & (score<oldpair-1e-10)];masks=masks[np.argsort(score[masks])];detail['partitions_screened']+=int(np.count_nonzero(valid))
            def path(mask):
                j=int(last[mask]);seq=[]
                while mask:
                    seq.append(clients[j]);k=int(parent[mask,j]);mask^=1<<j;j=k
                return [0]+seq[::-1]+[0]
            diverse=bits[valid];diverse=diverse[np.argsort(score[diverse])][:30]
            for mask in diverse:
                for subset in [int(mask),full^int(mask)]:
                    seq=path(subset);pool_paths[tuple(seq)]=seq
            _pd_write(folder/'route_pool_paths.json',list(pool_paths.values()))
            for mask in masks[:1500]:
                if time.monotonic()>deadline-.5:break
                trial=copy.deepcopy(routes);trial[ia]['path']=path(int(mask));trial[ib]['path']=path(full^int(mask))
                detail['schedules']+=1;candidate,reason=_v57_schedule(p,trial,min(.3,max(.01,deadline-time.monotonic())))
                if not candidate:continue
                audit=_pd_audit(inputs,candidate)
                if audit['feasible'] and audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:
                    best=audit['normalized_solution'];detail['improvements'].append(dict(pair=[ia,ib],ub=best['canonical_objective_score'],members=[trial[ia]['path'],trial[ib]['path']]))
                    _pd_write(folder/'candidate.json',best);print('PAIR_GAIN',best['canonical_objective_score'],flush=True);break
            _pd_write(folder/'progress.json',detail)
            if detail['improvements'] and _v58_signature(_v58_routes(best))!=_v58_signature(routes):break
        if not advanced:break
    return best,detail
# Shared, scoped route memory; restricted models never supply global bounds.
def _v536_collect(inputs,state,solutions):
    scope=_ct_scope(inputs)
    if state.get('route_archive_scope_v536')!=scope:
        state['route_archive_v536']=[];state['route_archive_scope_v536']=scope
    paths={tuple(p):list(p) for p in state.get('route_archive_v536',[])}
    for solution in solutions:
        audit=_pd_audit(inputs,solution) if isinstance(solution,dict) else {}
        if audit.get('feasible'):
            for r in audit['normalized_solution']['routes']:
                key=tuple(r['path']);paths.pop(key,None);paths[key]=list(r['path'])
    state['route_archive_v536']=list(paths.values())[-1600:]

_v536_portfolio_previous=_v512_portfolio_add
def _v512_portfolio_add(inputs,state,candidates,limit=6):
    candidates=list(candidates);_v536_collect(inputs,state,candidates)
    chosen=_v536_portfolio_previous(inputs,state,candidates,limit);_v536_collect(inputs,state,chosen)
    return chosen

def _v536_remember_valid_paths(memory,solution):
    for r in solution['routes']:
        key=tuple(r['path']);memory.pop(key,None);memory[key]=list(r['path'])
    while len(memory)>1000:memory.pop(next(iter(memory)))

def _v536_harvest(inputs,state,folder):
    scope=_ct_scope(inputs)
    for file in folder.rglob('audited_route_memory_v536.json'):
        saved=_pd_read(file) or {}
        if saved.get('fingerprint')==scope:state['route_archive_v536']=(state.get('route_archive_v536',[])+saved.get('paths',[]))[-1600:]

def _v536_pool(inputs,best,seconds,folder,state,allowed=None):
    import time,copy,random,math
    from pyscipopt import Model,quicksum
    deadline=time.monotonic()+seconds;p=_v58_provider(inputs).problem;c=inputs['objective_context']['coefficients']
    stats=dict(global_bound_usable=False,attempts=0,improvements=0)
    if c.get('TIME',0):return best,dict(stats,termination_reason='TIME_OBJECTIVE_NOT_SUPPORTED_BY_ROUTE_POOL')
    _v536_collect(inputs,state,[best]);paths={tuple(x) for x in state['route_archive_v536']}
    paths.update(tuple(r['path']) for r in best['routes']);paths.update(tuple(reversed(x)) for x in list(paths))
    vehicles=p['vehicles'];columns=[]
    for path in sorted(paths):
        if time.monotonic()>deadline-.5:break
        load=sum(p['client_rows'][i]['demand'] for i in path[1:-1]);distance=sum(p['distance'][a,b] for a,b in zip(path,path[1:]))
        for v in vehicles:
            if allowed is not None and v['vehicle_id'] not in allowed:continue
            if load>v['capacity']+1e-8:continue
            columns.append(dict(path=list(path),vehicle_id=v['vehicle_id'],customers=set(path[1:-1]),duration=distance/p['speed']*60+len(path[1:-1])*p['service_time'],score=distance*(c.get('DISTANCE',0)+c.get('COST',0)*v['cost_per_km'])))
    stats['columns']=len(columns)
    if time.monotonic()>deadline-.5 or not columns:return best,dict(stats,termination_reason='POOL_PREPARATION_BUDGET')
    model=Model('SHARED_ROUTE_POOL_V534');model.hideOutput();model.setParam('limits/memory',_v534_memory_budget(inputs)['solver_mb']);model.setParam('parallel/maxnthreads',1);model.setParam('limits/gap',0.);model.setParam('numerics/feastol',1e-9)
    try:
        x=[model.addVar(vtype='B') for _ in columns];u={v['vehicle_id']:model.addVar(vtype='B') for v in vehicles}
        for i in p['customers']:model.addCons(quicksum(x[j] for j,r in enumerate(columns) if i in r['customers'])==1)
        for v in vehicles:
            vid=v['vehicle_id'];ids=[j for j,r in enumerate(columns) if r['vehicle_id']==vid]
            for j in ids:model.addCons(x[j]<=u[vid])
            model.addCons(u[vid]<=quicksum(x[j] for j in ids));model.addCons(quicksum(columns[j]['duration']*x[j] for j in ids)<=p['max_time']*u[vid])
            if p['must_use_all_vehicles']:model.addCons(u[vid]==1)
            if not p['allow_multi_trip']:model.addCons(quicksum(x[j] for j in ids)<=1)
        fixed=quicksum(v['fixed_cost']*u[v['vehicle_id']] for v in vehicles) if p['fixed_cost_mode']=='WHEN_USED' else sum(v['fixed_cost'] for v in vehicles)
        objective=quicksum(r['score']*x[j] for j,r in enumerate(columns))+c.get('COST',0)*fixed+c.get('VEHICLES',0)*quicksum(u.values())
        model.setObjective(objective,'minimize');model.addCons(objective<=best['canonical_objective_score']-1e-8)
        rng=random.Random(int(state.get('cooperation_round_v536',0))+171)
        for attempt in range(25):
            if time.monotonic()>deadline-.5:break
            model.setParam('limits/time',max(.01,deadline-time.monotonic()-.4));model.optimize();status=str(model.getStatus());stats['solver_status']=status;sol=model.getBestSol()
            if sol is None:break
            ids=[j for j,var in enumerate(x) if model.getSolVal(sol,var)>.5]
            value=model.getSolObjVal(sol)
            if value>=best['canonical_objective_score']-1e-9:break
            routes=[dict(vehicle_id=columns[j]['vehicle_id'],path=columns[j]['path']) for j in ids]
            routes.sort(key=lambda r:min(p['windows'][i][1] for i in r['path'][1:-1]));valid=False
            for order in range(12):
                if time.monotonic()>deadline-.2:break
                if order:rng.shuffle(routes)
                candidate,reason=_v57_schedule(p,routes,min(.3,max(.01,deadline-time.monotonic())))
                stats['attempts']+=1
                audit=_pd_audit(inputs,candidate) if candidate else {}
                if audit.get('feasible'):
                    valid=True
                    if audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:
                        best=audit['normalized_solution'];stats['improvements']+=1;_pd_write(folder/'candidate.json',best);_v536_collect(inputs,state,[best])
                    break
            model.freeTransform()
            if valid:model.addCons(objective<=best['canonical_objective_score']-1e-8)
            else:model.addCons(quicksum(x[j] for j in ids)<=len(ids)-1)
            _pd_write(folder/'search_state.json',state)
            if status in ('memlimit','timelimit','userinterrupt'):break
            if valid and status=='optimal':break
    finally:model.freeProb()
    return best,dict(stats,termination_reason='RESTRICTED_POOL_FINISHED')

_v536_previous_local=_v58_local_search
def _v536_local(inputs,best,seconds,folder,state):
    import time,copy
    # Do not fragment a first short round; retain the proven baseline method and build shared memory.
    if seconds<30 and not state.get('cooperation_round_v536') and not state.get('stalled_rounds_v536'):
        folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True);before=best['canonical_objective_score']
        candidate,stats=_v536_short_classic(inputs,best,seconds,folder,state)
        _v536_collect(inputs,state,[best,candidate]);_v536_harvest(inputs,state,folder);_v536_collect(inputs,state,[candidate])
        state['cooperation_round_v536']=1;state['stalled_rounds_v536']=int(candidate['canonical_objective_score']>=before-1e-9)
        stats['cooperation_runs']=[dict(operator='DIVERSIFY',before_ub=before,after_ub=candidate['canonical_objective_score'],stats=dict(stats))]
        stats['dispatch_policy_v536']='CLASSIC_FIRST_FOR_SHORT_BUDGET';_pd_write(folder/'search_state.json',state)
        return candidate,stats
    folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True);deadline=time.monotonic()+seconds
    audit=_pd_audit(inputs,best)
    if not audit.get('feasible'):raise ValueError('Plano de partida rejeitado pela auditoria.')
    best=audit['normalized_solution'];before=best['canonical_objective_score'];_v536_collect(inputs,state,[best])
    round_id=int(state.get('cooperation_round_v536',0));state['cooperation_round_v536']=round_id+1
    stalled=int(state.get('stalled_rounds_v536',0));seed=7001+round_id*1009
    families=['FLEET','PAIR','POOL','DIVERSIFY'] if stalled or round_id%2 else ['DIVERSIFY','FLEET','PAIR','POOL']
    if inputs['objective_context']['coefficients'].get('TIME',0):families=[x for x in families if x not in ('PAIR','POOL')]
    if seconds<15:families=['FLEET','DIVERSIFY']
    runs=[]
    for j,family in enumerate(families):
        remaining=deadline-time.monotonic()-.3
        if remaining<.4:break
        quota=remaining/max(1,len(families)-j)
        if family=='DIVERSIFY' and j==0:quota=remaining*.70
        if family=='FLEET':quota=min(quota,1.5)
        branch=folder/family;branch.mkdir(exist_ok=True);old=best['canonical_objective_score'];started=time.monotonic()
        try:
            if family=='FLEET':candidate,detail=_v536_fleet(inputs,copy.deepcopy(best),quota,branch,seed)
            elif family=='PAIR':
                candidate,detail=_v536_pair(inputs,copy.deepcopy(best),quota,branch,seed)
                paths=_pd_read(branch/'route_pool_paths.json') or [];state['route_archive_v536']=(state['route_archive_v536']+paths)[-1600:]
            elif family=='POOL':candidate,detail=_v536_pool(inputs,copy.deepcopy(best),quota,branch,state, {r['vehicle_id'] for r in best['routes']} if round_id%2==0 else None)
            else:candidate,detail=_v536_previous_local(inputs,copy.deepcopy(best),quota,branch,state)
            if family=='DIVERSIFY':
                _v536_harvest(inputs,state,branch)
            checked=_pd_audit(inputs,candidate)
            if checked.get('feasible'):
                _v536_collect(inputs,state,[checked['normalized_solution']])
                if checked['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:best=checked['normalized_solution'];_pd_write(folder/'candidate.json',best)
            runs.append(dict(operator=family,before_ub=old,after_ub=best['canonical_objective_score'],stats=detail))
        except Exception as exc:runs.append(dict(operator=family,error=str(exc),before_ub=old,after_ub=best['canonical_objective_score']))
        performance=state.setdefault('cooperation_performance_v536',{}).setdefault(family,dict(runs=0,gain=0.,seconds=0.,stalled=0))
        performance['runs']+=1;performance['gain']+=old-best['canonical_objective_score'];performance['seconds']+=time.monotonic()-started
        performance['stalled']=0 if best['canonical_objective_score']<old-1e-9 else performance['stalled']+1
        _pd_write(folder/'search_state.json',state)
    state['stalled_rounds_v536']=0 if best['canonical_objective_score']<before-1e-9 else stalled+1
    state['cooperation_history_v536']=(state.get('cooperation_history_v536',[])+[dict(round=round_id,order=families,before_ub=before,after_ub=best['canonical_objective_score'])])[-20:]
    _pd_write(folder/'search_state.json',state)
    return best,dict(termination_reason='COOPERATIVE_SEARCH_FINISHED',cooperation_runs=runs,improvements=int(best['canonical_objective_score']<before-1e-9),global_bound_usable=False)

_v58_local_search=_v536_local
_V513_BINDINGS=tuple(_v512_portfolio_add if item is _v536_portfolio_previous else item for item in _V513_BINDINGS)

_v536_recover_previous=_v512_recover_candidates
def _v512_recover_candidates(inputs,folder,state):
    scope=_ct_scope(inputs)
    if state.get('route_archive_scope_v536')!=scope:state['route_archive_scope_v536']=scope;state['route_archive_v536']=[]
    for file in folder.rglob('audited_route_memory_v536.json'):
        saved=_pd_read(file) or {}
        if saved.get('fingerprint')==scope:state['route_archive_v536']=(state['route_archive_v536']+saved.get('paths',[]))[-1600:]
    return _v536_recover_previous(inputs,folder,state)

def _v536_lns_without_memory(inputs,best,seconds,folder,seed):
    return _v535_lns(inputs,best,seconds,folder,seed,remember_routes=False)

def _v536_regret_without_memory(inputs,best,seconds,folder,seed):
    return _v535_regret(inputs,best,seconds,folder,seed,remember_routes=False)

def _v536_short_classic(inputs,best,seconds,folder,state):
    import time,copy
    deadline=time.monotonic()+seconds;folder=_PDPath(folder);folder.mkdir(parents=True,exist_ok=True)
    checked=_pd_audit(inputs,best)
    if not checked.get('feasible'):raise ValueError('A diversificação exige um plano auditado.')
    best=checked['normalized_solution'];before=best['canonical_objective_score'];runs=[]
    if seconds<6 or before<=1e-12:return _v535_previous_local(inputs,best,seconds,folder,state)
    seed=int(state.get('diversification_round_v535',0));state['diversification_round_v535']=seed+1
    tasks=[('LOCAL',_v535_previous_local,min(5.,seconds*.1)),('LNS',_v536_lns_without_memory,None),('REGRET3',_v536_regret_without_memory,None)]
    for name,task,quota in tasks:
        left=deadline-time.monotonic()-.3
        if left<.5:break
        budget=min(left,quota if quota is not None else left*.70 if name=='LNS' else left)
        branch=folder/name;branch.mkdir(exist_ok=True);start=time.monotonic();old=best['canonical_objective_score']
        candidate,stats=task(inputs,copy.deepcopy(best),budget,branch,state) if name=='LOCAL' else task(inputs,copy.deepcopy(best),budget,branch,8131+1009*seed)
        audit=_pd_audit(inputs,candidate)
        if audit.get('feasible') and audit['recomputed']['canonical_objective_score']<best['canonical_objective_score']-1e-9:
            best=audit['normalized_solution'];_pd_write(folder/'candidate.json',best)
        record=state.setdefault('operators',{}).setdefault(name,dict(attempts=0,gain=0.,seconds=0.))
        record['attempts']+=stats.get('attempts',0);record['gain']+=old-best['canonical_objective_score'];record['seconds']+=time.monotonic()-start
        runs.append(dict(operator=name,budget_seconds=budget,stats=stats,before_ub=old,after_ub=best['canonical_objective_score']))
        _pd_write(folder/'search_state.json',state)
    return best,dict(termination_reason='DIVERSIFIED_TIME_LIMIT',attempts=sum(r['stats'].get('attempts',0) for r in runs),
        improvements=int(best['canonical_objective_score']<before-1e-9),operators=state.get('operators',{}),diversification_runs=runs,global_bound_usable=False)

_v536_narrative_previous=_v534_customer_narrative
def _v534_customer_narrative(solution,cert,mem=None,result=None,incluir_plano=False):
    import re
    text=_v536_narrative_previous(solution,cert,mem,result,incluir_plano)
    text=text.replace('a melhor referência matemática que conseguimos demonstrar','um patamar de comparação que os cálculos permitem garantir')
    if _v533_otimo_certificado(cert):
        message='**O que significa a certificação A:** para os dados e as regras cadastradas, os cálculos confirmaram que este plano é ótimo dentro da pequena tolerância usada para arredondamentos. Não se trata apenas de não termos encontrado algo melhor: a qualidade foi comprovada. Se os dados mudarem, essa confirmação precisa ser refeita.'
        text=re.sub(r'\*\*Como interpretar a garantia:\*\*[^\n]+',lambda m:message,text,count=1)
    elif cert.get('lower_bound_certified') and cert.get('gap_pct') is not None:
        value=f"{float(cert['gap_pct']):.2f}".replace('.',',')
        message='**O que significa o GAP de '+value+'%:** essa é a margem que ainda está em aberto na avaliação do plano. Pode existir uma alternativa melhor; também pode ser que este plano esteja muito bom e falte demonstrar isso. Essa margem diminui quando encontramos um plano melhor ou quando os cálculos conseguem confirmar melhor sua qualidade. Ela se refere ao conjunto de prioridades que você escolheu, não é uma promessa de economia e não é uma probabilidade de erro.'
        text=re.sub(r'\*\*Como interpretar a margem de melhoria:\*\*[^\n]+',lambda m:message,text,count=1)
    stats=(result or {}).get('statistics') or {}
    no_gain=stats.get('before_ub') is not None and stats.get('after_ub') is not None and stats.get('after_ub')>=stats.get('before_ub')-1e-9 and stats.get('after_lb',0)<=stats.get('before_lb',0)+1e-9
    if not _v533_otimo_certificado(cert) and no_gain and not (result or {}).get('issues'):
        text+='\n\n**Por que a busca pode parar de avançar:** a busca pode deixar de encontrar ganhos enquanto o sistema continua funcionando normalmente. Cada ferramenta explora possibilidades de uma maneira diferente. Mais tempo permite novas tentativas, mas também pode levar a repetir caminhos que já não trazem ganho. Os agentes compartilham viagens e resultados, mudam a composição das viagens e alternam estratégias. Não encontrar uma alternativa melhor ainda não prova que ela não exista. Não conseguimos prometer outro avanço ou que ele aparecerá com mais tempo. Você pode manter este plano auditado e retomar a busca depois pelo checkpoint.\n'
    return text


"""Apresentação e evidências; não altera a busca nem a certificação."""
````

Mostra se a continuidade melhorou o plano, fortaleceu a prova ou avançou nas duas frentes, ajudando o usuário a decidir se continua. Também apresenta o aviso de execução em andamento e de conclusão, sem repetir o relatório completo.


````python
def _v534_fmt(value, digits=2):
    if value is None: return 'não registrado'
    return f'{float(value):,.{digits}f}'.replace(',', '@').replace('.', ',').replace('@', '.')

def dados_evolucao_continuidade(result):
    """Reconstrói somente métricas documentadas, sem inventar custos antigos."""
    import copy
    stats=result.get('statistics') or {}; history=copy.deepcopy(result.get('round_history') or [])
    if not history and stats.get('before_ub') is not None:
        history=[dict(round=1, ub_gain=stats['before_ub']-stats['after_ub'],
                      lb_gain=stats['after_lb']-stats['before_lb'], gap_pct=result['certificate'].get('gap_pct'))]
    if not history: return []
    ub=(result.get('certificate') or {}).get('upper_bound'); lb=(result.get('certificate') or {}).get('lower_bound')
    rows=[]
    for row in reversed(history):
        after_ub=row.get('after_ub',ub); after_lb=row.get('after_lb',lb)
        before_ub=row.get('before_ub', after_ub+float(row.get('ub_gain',0)) if after_ub is not None else None)
        before_lb=row.get('before_lb', after_lb-float(row.get('lb_gain',0)) if after_lb is not None else None)
        def gap(u,l): return 100*max(0.,u-l)/max(abs(u),1e-12) if u is not None and l is not None else None
        rows.append(dict(round=row.get('round'), before_ub=before_ub, after_ub=after_ub,
                         before_lb=before_lb, after_lb=after_lb, before_gap=gap(before_ub,before_lb),
                         after_gap=row.get('gap_pct',gap(after_ub,after_lb)),
                         plan_gain_pct=100*(before_ub-after_ub)/max(abs(before_ub),1e-12) if before_ub is not None and after_ub is not None else None,
                         baseline_summary=row.get('baseline_summary'), final_summary=row.get('final_summary')))
        ub,lb=before_ub,before_lb
    return list(reversed(rows))

def resumo_continuidade_cliente(result):
    stats=result.get('statistics') or {};cert=result.get('certificate') or {}
    rows=dados_evolucao_continuidade(result)
    if not rows:return 'Ainda não há uma continuidade registrada para comparar.'
    latest=rows[-1]; fmt=_v534_fmt
    text='### O que esta continuidade acrescentou\n\n'
    text+='| Indicador | Antes desta rodada | Agora |\n|---|---:|---:|\n'
    before=stats.get('baseline_summary') or {};after=(result.get('final_solution') or {}).get('summary') or {}
    currency=(result.get('customer_context_v534') or {}).get('user_answers',{}).get('unidade_monetaria','R$')
    for key,label,unit,digits in [('total_cost','Custo calculado',currency+' ',2),('total_distance_km','Percurso','',2),('num_vehicles_used','Veículos','',0),('num_routes','Viagens','',0)]:
        if key in before and key in after:
            suffix=' km' if key=='total_distance_km' else ''
            text+=f'| {label} | {unit}{fmt(before[key],digits)}{suffix} | {unit}{fmt(after[key],digits)}{suffix} |\n'
    text+=f"| Margem em aberto (GAP) | {fmt(latest['before_gap'])}% | {fmt(latest['after_gap'])}% |\n"
    text+='\n| Continuidade | Melhoria no objetivo escolhido | GAP antes → depois | Garantia avançou? |\n|---|---:|---:|---|\n'
    for row in rows:
        stronger=row['before_lb'] is not None and row['after_lb']>row['before_lb']+1e-9
        text+=f"| {row['round']} | {fmt(row['plan_gain_pct'])}% | {fmt(row['before_gap'])}% → {fmt(row['after_gap'])}% | {'Sim' if stronger else 'Sem avanço registrado'} |\n"
    text+='\nA melhoria no objetivo considera suas prioridades em conjunto; não é uma porcentagem de economia.\n\n'
    issues=result.get('issues') or any(r.get('error') for r in stats.get('stages',{}).values())
    if issues:text+='**Antes de continuar:** houve um problema de execução. Confira o diagnóstico; isso não demonstra que o plano deixou de poder melhorar.'
    elif _v533_otimo_certificado(cert):text+='**Nossa orientação:** a melhor qualidade já foi comprovada dentro da tolerância. Você pode encerrar para estes mesmos dados e prioridades.'
    elif (latest['plan_gain_pct'] or 0)>1e-7 or latest['after_lb']>latest['before_lb']+1e-9:
        text+='**Nossa orientação:** esta rodada trouxe avanço. Você pode usar o plano aprovado ou autorizar outra tentativa, se uma melhoria adicional justificar o esforço. Os ganhos podem diminuir; continuar não garante outro avanço.'
    else:text+='**Nossa orientação:** esta rodada não trouxe ganho no plano nem na comprovação de sua qualidade. Você pode encerrar com o plano aprovado ou escolher uma estratégia diferente entre as opções disponíveis. Mais tempo não garante melhoria.'
    return text

def mostrar_relatorio_continuidade(result):
    if result.get('issues') or not result.get('certificate',{}).get('report_allowed'):
        print('Rodada não liberada para apresentação:',result.get('issues',[]));return
    key=(result.get('folder'),result['certificate'].get('upper_bound'),result['certificate'].get('lower_bound'))
    if key in _v532_round_feedback:return
    _v534_validate_result(result)
    from IPython.display import display,Markdown
    display(Markdown(resumo_continuidade_cliente(result)))
    _v532_round_feedback.add(key)

class AvisoExecucaoV534:
    """Mantém um único aviso durante toda a execução e suas etapas internas."""
    _handle = None
    _active = []

    def __init__(self, label, iniciar_depois_dos_dados=False):
        self.label = label
        self.iniciar_depois_dos_dados = iniciar_depois_dos_dados

    @classmethod
    def _mostrar(cls, title, detail, color):
        from IPython.display import display, HTML
        import html
        content = HTML('<div role="status" aria-live="polite" style="position:fixed;bottom:20px;right:20px;z-index:9999;max-width:min(520px,calc(100vw - 40px));box-sizing:border-box;box-shadow:0 6px 24px #0003;pointer-events:none;padding:20px;background:'
                       + color + ';color:white;border-radius:12px;font-size:26px;font-weight:bold">'
                       + html.escape(title) + '<br><span style="font-size:17px">'
                       + html.escape(detail) + '</span></div>')
        if cls._handle is None:
            cls._handle = display(content, display_id=True)
        else:
            cls._handle.update(content)

    @classmethod
    def aguardando_continuidade(cls):
        if cls._active:
            cls._mostrar('⏸ AGUARDANDO SUA DECISÃO',
                         'Esta etapa terminou. Leia os resultados e escolha se deseja continuar.', '#725017')

    @classmethod
    def iniciar_processamento(cls, label=None):
        if cls._active:
            cls._mostrar('⏳ SISTEMA EM EXECUÇÃO',
                         (label or cls._active[-1].label)
                         + '. Aguarde: o processamento ainda não terminou.', '#163b60')

    def __enter__(self):
        cls = type(self)
        if not cls._active:
            cls._handle = None
        cls._active.append(self)
        if not self.iniciar_depois_dos_dados:
            cls.iniciar_processamento(self.label)
        return self

    def __exit__(self, kind, value, tb):
        cls = type(self)
        cls._active.remove(self)
        if kind is not None:
            cls._mostrar('⚠ EXECUÇÃO INTERROMPIDA',
                         'Confira a mensagem apresentada abaixo antes de prosseguir.', '#9b392e')
        elif cls._active:
            cls._mostrar('⏳ SISTEMA EM EXECUÇÃO',
                         cls._active[-1].label + '. Aguarde: o processamento ainda não terminou.', '#163b60')
        else:
            cls._mostrar('✅ PROCESSAMENTO TERMINADO',
                         'A execução terminou. Os resultados estão disponíveis para consulta.', '#26835c')
        return False


def _v534_com_aviso(method,label):
    import functools
    @functools.wraps(method)
    def decorated(*args,**kwargs):
        with AvisoExecucaoV534(label):return method(*args,**kwargs)
    return decorated

OrchestratorAgent._run_solver_portfolio=_v534_com_aviso(OrchestratorAgent._run_solver_portfolio,'Procurando o plano inicial')
_v534_certifier_original=OrchestratorAgent._run_optimality_certification
OrchestratorAgent._run_optimality_certification=_v534_com_aviso(OrchestratorAgent._run_optimality_certification,'Conferindo a qualidade do plano inicial')
# Preserve the strict identity guard for the explicitly installed display wrapper.
_V513_BINDINGS=tuple(OrchestratorAgent._run_optimality_certification if value is _v534_certifier_original else value for value in _V513_BINDINGS)











def _v534_store_route_memory(inputs,folder,memory,enabled=True):
    if enabled:_pd_write(folder/"audited_route_memory_v536.json",dict(fingerprint=_ct_scope(inputs),paths=list(memory.values())))
````

## **INTERPRETAÇÃO DOS RESULTADOS**

A análise dos resultados considera três aspectos complementares: **a solução encontrada, sua viabilidade operacional e sua qualidade matemática**.

| Análise | O que permite entender |
|---|---|
| **Plano de viagens** | Como a operação foi organizada: veículos, viagens, clientes, horários, distância, tempo e custo |

**Solução encontrada → Auditoria → Avaliação da qualidade**

Infográfico real: 12 clientes, um veículo, seis viagens, certificação A

## **FUNCIONAMENTO DO SISTEMA**

### **NÚCLEO CENTRAL DE SOLUÇÃO**

Depois que o problema é recebido, validado e traduzido para uma estrutura matemática, começa a etapa de **busca da solução**.

O núcleo central de solução dispõe de dois agentes especializados:

| Agente | Abordagem | Papel no sistema |
|---|---|---|
| **ORToolsSolverAgent** | Otimização combinatória | Explora o problema por meio dos recursos de roteirização e busca do **Google OR-Tools** |
| **GeneticAlgorithmSolverAgent** | Metaheurística evolutiva | Explora diferentes combinações de rotas por meio de **Algoritmos Genéticos** |

Os dois agentes recebem a mesma representação do problema e trabalham considerando os **objetivos e restrições definidos pelo usuário**, como capacidade, tempo disponível, janelas de atendimento, custos, utilização da frota e múltiplas viagens.

### **COMO O NÚCLEO FUNCIONA**

```text
                    PROBLEMA VALIDADO
                           │
                           ↓
                 REPRESENTAÇÃO MATEMÁTICA
                           │
                           ↓
                  NÚCLEO DE SOLUÇÃO
                           │
                 ┌─────────┴─────────┐
                 │                   │
                 ↓                   ↓
        ORToolsSolverAgent   GeneticAlgorithmSolverAgent
            OR-TOOLS          ALGORITMOS GENÉTICOS
                 │                   │
                 └─────────┬─────────┘
                           ↓
                 SOLUÇÃO CANDIDATA
                           │
                           ↓
                       AUDITORIA
```

O **OR-Tools** oferece uma abordagem estruturada para problemas de roteirização e utiliza métodos especializados de otimização combinatória.

O **Algoritmo Genético**, por sua vez, amplia a busca por meio de uma abordagem evolutiva, criando e aprimorando diferentes combinações de rotas ao longo das gerações. Essa característica é especialmente útil para explorar problemas complexos e configurações em que a busca precisa de maior flexibilidade.

O sistema pode **acionar o agente mais adequado ou utilizar os resultados dos dois agentes**, de acordo com a estratégia definida pelo Orquestrador. Quando existem diferentes soluções candidatas, elas são comparadas utilizando a mesma função objetivo.

A solução selecionada, porém, **ainda não é aceita automaticamente como resultado final**. Ela segue para o **AuditorAgent**, que verifica de forma independente se todas as restrições foram realmente respeitadas.

> **O núcleo de solução procura o melhor plano; a auditoria confirma se esse plano é realmente válido.**

### **AUDITORIA**

Encontrar uma solução com bom desempenho não é suficiente. Antes de aceitá-la, o sistema verifica de forma independente se o plano realmente **pode ser executado nas condições definidas pelo usuário**. **Nos diz se a solução é Factível**

Essa verificação é realizada pelo **AuditorAgent**, que reconstrói os principais elementos da solução e confere o cumprimento das restrições.

| O que é verificado | Na prática |
|---|---|
| **Atendimento** | todos os clientes foram atendidos corretamente |
| **Capacidade** | nenhuma viagem excede a capacidade do veículo |
| **Horários** | as janelas de atendimento são respeitadas |
| **Operação** | depósito, retorno e regras de Multi-Trip estão corretos |
| **Jornada** | o tempo disponível de cada veículo é respeitado |
| **Utilização mínima** | quando aplicável, a exigência mínima de uso é atendida |
| **Custos** | custos fixos e variáveis são recalculados |
| **Objetivo** | o valor da função objetivo é reconstruído de forma independente |

Ao final, a solução pode ser **aprovada** ou **reprovada**.

Uma solução aprovada é considerada **operacionalmente factível**. Isso, porém, ainda não significa que seja a melhor solução possível. Essa segunda questão pertence à etapa de certificação.

### **CERTIFICAÇÃO**

Depois de verificar que a solução é válida, surge uma segunda questão: **até onde podemos afirmar que ela é boa do ponto de vista matemático?**

Em problemas de minimização, essa análise utiliza dois limites:

**UB — Upper Bound:** corresponde ao valor da **melhor solução válida conhecida**.

**LB — Lower Bound:** corresponde a um **piso matemático válido**, isto é, um valor abaixo do qual sabemos que nenhuma solução factível pode existir.

Por isso, o valor ótimo está necessariamente entre esses dois limites:

```text id="d6j59h"
Melhor direção
      ←

───────────────●────────────────── ? ─────────────────●──────────────
              LB                                    UB
      piso matemático                     melhor solução encontrada

                    LB ≤ ÓTIMO ≤ UB
```

Durante a otimização, esse intervalo pode diminuir de duas maneiras: encontrando uma solução melhor, o que reduz o **UB**, ou fortalecendo a demonstração matemática, o que aumenta o **LB**.

Quanto mais próximos estiverem esses limites, menor é a incerteza sobre a qualidade da solução encontrada.

**A Otimalidade** nos diz se existe alguma outra solução válida que seja melhor?

**uma solução pode ser válida e apresentar excelente desempenho, mesmo que sua otimalidade ainda não tenha sido comprovada**.

### **COMO INTERPRETAR O GAP**

O **GAP** mede a separação relativa entre a melhor solução conhecida (**UB**) e o piso matemático demonstrado (**LB**).

### **EXEMPLO**

Se:

**UB = 100**  
**LB = 80**

então: GAP=20\%

Visualmente:

```text
LB = 80                                      UB = 100
   │████████████ região não fechada ███████████│
   └──────────────── GAP = 20% ────────────────┘
```

### **COMO INTERPRETAR**

**GAP = 0%** → UB e LB coincidem; a prova foi fechada.  
**GAP pequeno** → existe forte evidência matemática sobre a qualidade da solução.  
**GAP grande** → ainda existe distância entre o que foi encontrado e o que conseguimos demonstrar.

Um GAP de 20% **não significa que existe uma melhoria de 20% disponível**. O ótimo verdadeiro pode estar em qualquer ponto válido entre LB e UB.

### **EVOLUÇÃO DA CERTIFICAÇÃO**

Uma forma simples de visualizar o processo é acompanhar a aproximação entre **LB e UB**:

```text id="w4o3p4"
INÍCIO

LB ●────────────────────────────────────────● UB
          maior região ainda aberta


DURANTE A OTIMIZAÇÃO

       LB ●────────────────────────● UB
              intervalo menor


PROVA FECHADA

                     ●
                   LB = UB
```

O intervalo pode diminuir porque o sistema encontrou uma **solução melhor**, porque conseguiu um **piso matemático mais forte**, ou pelas duas razões simultaneamente.

Assim, otimização e certificação desempenham papéis complementares: **uma procura melhorar a solução; a outra procura demonstrar matematicamente a qualidade alcançada**.

**NÍVEL DE CERTIFICAÇÃO**

Para tornar a interpretação mais simples, o sistema traduz a informação matemática disponível em um **nível de certificação**.

| Nível | Interpretação |
|---|---|
| **A** | Evidência matemática muito forte |
| **B** | Forte evidência sobre a qualidade da solução |
| **C** | Solução válida, mas ainda existe incerteza matemática relevante |
| **D** | Solução disponível, porém com certificação ainda limitada |

Essa classificação deve ser interpretada como uma medida da **força da evidência matemática**, e não simplesmente da qualidade operacional da rota.

Por exemplo, uma solução pode apresentar bom custo, distância e utilização da frota e ainda permanecer no **nível C**. Isso pode acontecer porque o problema é complexo e, dentro do tempo disponível, o sistema ainda não conseguiu fortalecer suficientemente o **Lower Bound**.

Portanto, **nível C não significa necessariamente uma solução ruim**. Significa que ainda existe espaço entre aquilo que foi encontrado e aquilo que conseguimos demonstrar matematicamente.

### **MULTIOBJETIVO**

Em alguns problemas, uma única métrica não representa adequadamente a decisão. O usuário pode, por exemplo, desejar reduzir simultaneamente **distância e custo**, atribuindo pesos diferentes a cada objetivo.

Considere:

### **Distância 40% + Custo 60%**

Nesse caso, a solução é avaliada pelo **resultado combinado**, e não por cada indicador isoladamente.

```text id="dclcgo"
             DISTÂNCIA 40%
                   \
                    \
                     >  RESULTADO COMBINADO
                    /
                   /
                CUSTO 60%
```

Isso significa que uma nova solução pode percorrer um pouco mais de distância e, ainda assim, ser melhor se a redução de custo compensar essa diferença segundo os pesos definidos.

O mesmo raciocínio vale para outras combinações.

> Em problemas multiobjetivo, a comparação entre soluções deve respeitar a **função objetivo completa definida pelo usuário**.

### **USAR MENOS VEÍCULOS É SEMPRE MELHOR?**

Não necessariamente.

Utilizar menos veículos pode reduzir custos fixos, especialmente quando o aluguel é cobrado apenas para os veículos efetivamente utilizados. Porém, concentrar a operação em uma frota menor também pode exigir **mais viagens por veículo**, aumentando distância, tempo ou custo variável.

```text id="9zh6do"
MENOS VEÍCULOS
      │
      ├── pode reduzir custos fixos
      │
      └── pode exigir mais viagens
                    ↓
           mais tempo / distância
                    ↓
          possível aumento de custo
```

Por isso, o sistema não considera automaticamente que **menos veículos significa uma solução melhor**.

A decisão depende do objetivo escolhido. Uma redução de frota somente representa uma melhoria quando produz um resultado melhor para a **função objetivo completa** e continua respeitando todas as restrições operacionais.

### **O QUE ACONTECE DEPOIS QUE UMA SOLUÇÃO É ENCONTRADA?**

A primeira solução válida funciona como um **ponto de partida**, e não necessariamente como o encerramento do processo.

A partir dela, o sistema pode continuar trabalhando em duas frentes:

```text id="gnwvgk"
                   SOLUÇÃO ATUAL
                        │
             ┌──────────┴──────────┐
             │                     │
             ↓                     ↓
      MELHORAR O PLANO      FORTALECER A PROVA
             │                     │
        reduzir o UB           aumentar o LB
             │                     │
             └──────────┬──────────┘
                        ↓
                  REDUZIR O GAP
```

Na primeira frente, são procuradas alternativas que melhorem o resultado de acordo com os objetivos definidos pelo usuário.

Na segunda, o esforço é matemático: fortalecer os limites que permitem avaliar até onde a qualidade da solução pode ser comprovada.

São avanços diferentes, mas complementares: **um melhora o plano; o outro aumenta o conhecimento que temos sobre sua qualidade**.

### **CONTINUIDADE DA OTIMIZAÇÃO**

Quando ainda existe espaço para melhorar a solução ou fortalecer sua certificação, o usuário pode disponibilizar **tempo computacional adicional**.

Esse processo não começa novamente do zero. A melhor solução já aprovada é preservada e utilizada como referência para as novas tentativas.

```text id="c04pyx"
MELHOR SOLUÇÃO APROVADA
          ↓
     permanece salva
          ↓
   NOVAS TENTATIVAS
      ↙         ↘
 melhorar     fortalecer
 o plano       a prova
      ↘         ↙
    COMPARAR E AUDITAR
          ↓
   MANTER O MELHOR RESULTADO
```

Uma nova solução somente substitui a anterior quando **melhora o objetivo definido e passa novamente pela auditoria**.

Se isso não acontecer, a melhor solução anterior continua preservada.

Assim, conceder mais tempo significa ampliar a oportunidade de **melhoria e certificação**, sem perder o resultado válido já alcançado.

### **E SE NÃO FOR ENCONTRADA UMA SOLUÇÃO MELHOR?**

Nem toda nova tentativa produz uma melhoria. Isso pode acontecer porque a solução atual já é difícil de superar, porque a região explorada não continha uma alternativa melhor ou simplesmente porque o método não conseguiu encontrá-la dentro do tempo disponível.

Nesse ponto, é importante separar **resultado da busca** de **prova matemática**.

> **Não encontrar uma solução melhor não significa provar que ela não existe.**

Por exemplo, se uma tentativa não consegue construir um plano válido utilizando apenas **2 veículos**, isso não permite concluir automaticamente que operar com 2 veículos seja impossível.

Essa conclusão somente pode ser feita quando existe uma **prova matemática válida de inviabilidade**.

Essa distinção é especialmente importante porque o sistema utiliza métodos heurísticos: eles podem encontrar excelentes soluções, mas uma tentativa sem sucesso, por si só, **não constitui uma demonstração matemática**.

### **COMO SABER SE A CONTINUIDADE FOI ÚTIL?**

A avaliação deve comparar o estado do sistema **antes e depois** do tempo adicional.

Há duas dimensões a observar:

| Dimensão | O que pode melhorar |
|---|---|
| **Plano operacional** | função objetivo e indicadores associados à operação |
| **Certificação** | UB, LB, GAP e nível de certificação |

A continuidade pode, portanto, produzir resultados diferentes.

**Melhoria da solução:** o sistema encontra um plano com melhor função objetivo.

**Melhoria da certificação:** a solução pode permanecer igual, mas o sistema consegue fortalecer o LB ou reduzir o GAP.

**Melhoria em ambas:** além de encontrar um plano melhor, também aumenta a evidência matemática sobre sua qualidade.

Isso significa que uma rodada adicional pode ser útil **mesmo quando distância, custo ou número de veículos não mudam**, desde que ela consiga fortalecer de forma válida a certificação do resultado.

### **COMO INTERPRETAR O RESULTADO FINAL**

Ao final da execução, a análise pode seguir uma sequência simples:

**1. Plano operacional**  
Observe como as entregas foram organizadas: veículos utilizados, viagens, clientes atendidos, horários, distância, tempo e custo.

**2. Auditoria**  
Confirme se a solução foi aprovada e se todas as restrições foram respeitadas.

**3. Desempenho**  
Analise o resultado de acordo com os objetivos e pesos escolhidos pelo usuário.

**4. Certificação**  
Observe UB, LB, GAP e o nível alcançado para entender a força da evidência matemática disponível.

**5. Continuidade**  
Quando ainda existe um GAP relevante, avalie se faz sentido disponibilizar mais tempo para buscar uma solução melhor ou fortalecer a certificação.

---

> **O resultado final combina três elementos: um plano operacional, uma verificação independente de sua viabilidade e uma medida da evidência matemática disponível sobre sua qualidade.**

Os resultados não devem ser interpretados como uma relação linear entre “menos minutos” e “menor GAP”. Cada configuração produz um problema de otimização diferente.

> O ponto relevante é que o sistema mantém a mesma lógica de **busca, auditoria e certificação**, mesmo quando as condições operacionais mudam.

Esses testes verificam um comportamento importante: **o sistema não deve transformar uma solução inválida em recomendação operacional**.

Existe, porém, uma distinção metodológica:

> **Não encontrar um plano válido não é, por si só, uma prova matemática de inviabilidade.**

### Por isso, o sistema preserva a diferença entre **“não foi encontrada solução”** e **“foi demonstrado que não existe solução factível”**.

### **O QUE A BATERIA DE TESTES MOSTROU?**

Os diferentes cenários permitem observar quatro comportamentos relevantes do sistema:

| Evidência observada | Interpretação |
|---|---|
| **Planos passam por auditoria antes de serem aceitos** | A qualidade numérica não substitui a verificação de factibilidade |
| **Problemas menores podem alcançar prova completa** | Quando UB e LB convergem, a incerteza matemática é eliminada |
| **Problemas maiores ainda podem apresentar melhorias relevantes** | A ausência de prova completa não impede o avanço operacional |
| **A continuidade atua sobre solução e certificação** | O sistema pode reduzir o UB, elevar o LB ou atuar nas duas frentes |

A bateria evidencia que o sistema não trata a otimização como uma única resposta do solver. O resultado passa por **busca, validação, melhoria e avaliação matemática**, e o nível de conclusão apresentado depende da evidência efetivamente obtida em cada execução.

## **LIÇÕES APRENDIDAS**

O desenvolvimento e os testes do sistema permitiram consolidar alguns aprendizados importantes:

1. **Memória compartilhada evita perda de conhecimento.**  
   Centralizar dados, soluções e histórico mantém os agentes trabalhando com informações consistentes e permite continuar a otimização sem recomeçar do zero.

2. **Uma boa orquestração é mais importante do que simplesmente ter muitos agentes.**  
   O ganho da arquitetura multiagente aparece quando existem responsabilidades claras, coordenação e troca organizada de informações.

3. **Tarefas independentes podem avançar em paralelo.**  
   Buscar uma solução melhor, fortalecer limites matemáticos e investigar factibilidade simultaneamente permite aproveitar melhor o tempo computacional.

4. **Encontrar uma boa solução e provar que ela é ótima são desafios diferentes.**  
   O sistema pode encontrar rapidamente um plano válido, enquanto sua certificação matemática pode exigir muito mais esforço.

5. **Não encontrar uma solução não significa provar que ela não existe.**  
   Uma busca sem sucesso deve ser tratada como resultado inconclusivo, e não como prova de inviabilidade.

6. **A auditoria independente é essencial.**  
   Toda solução candidata precisa ser verificada antes de ser aceita, reduzindo o risco de considerar como boa uma solução que viole alguma restrição.

7. **O melhor resultado conhecido deve ser preservado.**  
   Novas soluções só devem substituir o plano atual quando forem válidas e realmente melhorarem o objetivo definido.

8. **Mais tempo de processamento não garante melhoria contínua.**  
   Quando uma estratégia deixa de produzir avanços, é mais eficiente diversificar a busca do que simplesmente repetir o mesmo processo.

9. **Melhorar a solução e melhorar sua certificação são avanços diferentes.**  
   O sistema pode encontrar um plano melhor, fortalecer o limite matemático e reduzir o GAP, ou realizar esses avanços de forma independente.

10. **Um sistema confiável também precisa reconhecer seus limites.**  
    Resultados inconclusivos, como `UNKNOWN`, devem ser apresentados como tal. Explicar o que foi encontrado, o que foi comprovado e o que permanece em aberto torna a resposta mais transparente.

> **Síntese:** o principal aprendizado foi que a qualidade do sistema não depende apenas de encontrar boas rotas, mas de combinar **especialização dos agentes, coordenação, auditoria, continuidade e rigor na interpretação das evidências matemáticas**.
